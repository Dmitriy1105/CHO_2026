# Задачи этапов

Этап 1 — загрузить структуру белка и подготовить данные для будущей модели.
В результате получаем не просто файл PDB, а подготовленное числовое представление белка.

Структура ноутбука и назначение ячеек
Ячейки	Раздел	Результат
1–3	Загрузка структуры	Загрузка структуры белка из базы данных PDB в формате PDB/mmCIF и подготовка исходных данных к анализу.
4	Чтение структуры	Получение информации о моделях, белковых цепях и общей организации структуры.
5	Аминокислотные последовательности	Извлечение аминокислотных последовательностей белковых цепей и сохранение их в формате FASTA.
6	Анализ аминокислотных остатков	Формирование таблицы аминокислотных остатков с указанием их типов, идентификаторов и характеристик.
7	Координаты атомов	Формирование таблицы атомов с их пространственными координатами X, Y, Z и основными характеристиками.
8	Анализ комплекса	Выявление белковых остатков, небелковых компонентов и молекул воды, входящих в состав исследуемой структуры.
9	Геометрия структуры	Расчёт пространственного центра, габаритных размеров и других геометрических характеристик белка.
10	Метаданные	Получение сведений о методе экспериментального определения структуры, разрешении и других структурных характеристиках.
11	Распределение аминокислот	Расчёт частоты встречаемости аминокислот и построение графиков их распределения в белке.
12	Сохранение результатов	Сохранение промежуточных результатов структурного анализа в файлы JSON и CSV.
13	Проверка последовательности	Сравнение экспериментальных аминокислотных последовательностей с референсными последовательностями с помощью выравнивания.
14	Полнота структуры	Проверка отсутствующих остатков и атомов, альтернативных положений атомов, значений occupancy и полноты структурных данных.
15	Разделение компонентов	Разделение структуры на белковые цепи, небелковые компоненты и молекулы воды с сохранением отдельных результатов.
16	3D-визуализация	Построение интерактивной трёхмерной модели белка и его компонентов с использованием NGLView.
17	Обнаружение и идентификация небелковых компонентов	Определение кодов небелковых компонентов, цепей, номеров остатков, атомного и элементного состава, а также пространственных координат. Выбор компонентов для дальнейшего анализа.
18	Анализ области связывания	Определение аминокислотных остатков белка, расположенных вблизи выбранных небелковых компонентов, и расчёт минимальных расстояний между ними.
19	Универсальный анализ взаимодействий белка с компонентами	Определение атомных контактов, потенциальных полярных взаимодействий, возможных водородных связей, гидрофобных контактов и координации различных металлов. Дополнительно анализируются дисульфидные мостики и другие структурные взаимодействия.
20–21	Итоговые структурные и функциональные характеристики	Сбор и систематизация основных характеристик белка, его цепей, пространственной организации и структурных взаимодействий. Формирование итоговых метрик и результатов анализа.
22	Извлечение структурных признаков (Feature Extraction)	Преобразование структурной информации в числовые признаки для машинного обучения. Формирование характеристик атомов, аминокислотных остатков, глобальных параметров белка и информации об области связывания.
23	Построение пространственных графов (Protein Graph Construction)	Представление белковой структуры в виде атомного и остаточного графов. Формирование узлов, пространственных рёбер, координат и характеристик связей между пространственно близкими элементами структуры.
24	Формирование единого машинного представления (Unified Protein Representation)	Объединение атомных и остаточных графов, числовых признаков, координат, глобальных характеристик и масок области связывания в единую структуру данных. Проверка корректности представления и сохранение результатов в форматах NPZ, CSV и JSON.

1. Какую задачу решает текущий ноутбук
Текущий ноутбук 01_PDB_Protein_Analysis.ipynb решает задачу структурного анализа белка и его окружения.

Он отвечает на вопросы:
- Как устроен белок?
- Из каких аминокислот он состоит?
- Где в пространстве находятся его атомы?
- Какие дополнительные молекулы присутствуют в структуре?
- Какие небелковые компоненты находятся рядом с белком?
- Какие аминокислотные остатки образуют окружение выбранного компонента?
- Какие возможные взаимодействия возникают между ними?
То есть ноутбук переводит исходный файл PDB/mmCIF в более удобный для дальнейшей обработки набор структурных данных.
2. Для чего нужны полученные CSV-файлы
Разные таблицы могут использоваться для разных задач.
Данные	Для чего пригодятся
Аминокислотные последовательности	Представление белка в виде последовательности аминокислот
Координаты атомов	Восстановление пространственной структуры белка
Таблицы аминокислотных остатков	Анализ состава и свойств белка
Небелковые компоненты	Получение информации о молекулах, находящихся в структуре
Координаты выбранного компонента	Определение его положения относительно белка
Аминокислоты в области связывания	Описание окружения потенциального участка связывания
Таблицы атомных контактов	Изучение возможных взаимодействий между белком и компонентом
Геометрические характеристики	Дополнительные признаки для анализа структуры

ДАЛЕЕ БУДУТ
Модуль	Назначение
02_PDB_Ligand_Analysis.ipynb	Изучение связанных с белком малых молекул и их аннотаций
03_Protein_Feature_Extraction.ipynb	Подготовка признаков белка для нейросети
04_Protein_Encoder.ipynb	Разработка и проверка Protein Encoder

Этап 2. Создание Protein Encoder
После изучения структур разрабатываем отдельный модуль:

ProteinEncoder

Его задача — преобразовать структуру белка в векторное представление, которое сможет использовать нейросеть.

3. Этап 3. Интеграция с текущей Conditional Transformer VAE
Сейчас твоя архитектура получает три источника информации:
- SELFIES Encoder;
- Property Encoder;
- Fingerprint Encoder.
В перспективе добавится четвёртый источник — Protein Encoder.

Это схема возможного расширения, а не готовое решение: для генерации под белок особенно важно передавать декодеру информацию о структуре белка. Поэтому в дальнейшем стоит рассмотреть передачу последовательности признаков белка через cross-attention, а не только одного общего вектора.

4. Этап 4. Подготовка обучающей выборки
Это важный момент: для обучения генерации под белок недостаточно взять один PDB-файл.

Нужны данные, связывающие структуры белков с соответствующими молекулами. Например, несколько белок-лигандных комплексов для выбранной мишени. Затем формируется набор пар:

(Protein Structure, SELFIES, Molecular Properties)

На нём уже можно будет обучать расширенную модель. Текущая VAE останется базовой моделью для сравнения.

```python

# ============================================================
# CELL 2. PROJECT CONFIGURATION AND BATCH INPUT
# ============================================================
# Назначение:
# 1. Настройка путей проекта.
# 2. Загрузка списка 100 отобранных PDB.
# 3. Выбор конкретной структуры для анализа.
# 4. Подготовка индивидуальных каталогов результатов.
#
# Режим работы:
# - по умолчанию берётся первая структура из CSV;
# - для обработки конкретной структуры задаётся PDB_ID_OVERRIDE;
# - управляющий ноутбук сможет передавать PDB_ID извне.
# ============================================================

import os
import warnings
import requests
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path

from Bio.PDB import MMCIFParser, is_aa
from Bio.PDB.MMCIF2Dict import MMCIF2Dict
from Bio.SeqUtils import seq1

# Не отключаем все предупреждения:
# сообщения BioPython могут быть важны для контроля качества PDB.
# warnings.filterwarnings("ignore")


# ============================================================
# 1. PROJECT PATHS
# ============================================================

# Корневая директория проекта.
PROJECT_ROOT = Path(
    r"C:\apex\APEX\МИФИ\СНО\medic"
)

# Корневая директория датасета.
DATASET_ROOT = (
    PROJECT_ROOT / "Protein_Pocket_Dataset"
)

# Входной список отобранных структур.
SELECTION_CSV = (
    DATASET_ROOT
    / "05_final_dataset"
    / "protein_final_100.csv"
)

# Рабочая директория анализа структур.
STRUCTURE_ANALYSIS_DIR = (
    DATASET_ROOT / "07_structure_analysis"
)

# Каталог исходных структур.
STRUCTURES_DIR = (
    STRUCTURE_ANALYSIS_DIR / "structures"
)

# Общий каталог результатов.
RESULTS_ROOT = (
    STRUCTURE_ANALYSIS_DIR / "results"
)


# ============================================================
# 2. PDB SELECTION
# ============================================================

# Если None:
# автоматически выбирается первая структура из CSV.
#
# Для ручного тестирования можно указать:
# PDB_ID_OVERRIDE = "1OCC"
#
# Также поддерживается передача через переменную окружения:
# PDB_ID=1OCC

PDB_ID_OVERRIDE = None

ENV_PDB_ID = os.environ.get(
    "PDB_ID", ""
).strip().upper()

if ENV_PDB_ID:
    PDB_ID_OVERRIDE = ENV_PDB_ID


# ============================================================
# 3. LOAD SELECTION REGISTRY
# ============================================================

if not SELECTION_CSV.exists():
    raise FileNotFoundError(
        f"Selection CSV not found:\n{SELECTION_CSV}\n"
        "Проверьте расположение protein_final_100.csv."
    )

selection_df = pd.read_csv(
    SELECTION_CSV,
    dtype={"pdb_id": "string"}
)

required_columns = {"pdb_id"}

missing_columns = (
    required_columns - set(selection_df.columns)
)

if missing_columns:
    raise ValueError(
        "В selection CSV отсутствуют обязательные "
        f"столбцы: {sorted(missing_columns)}"
    )

# Нормализация PDB ID.
selection_df["pdb_id"] = (
    selection_df["pdb_id"]
    .astype("string")
    .str.strip()
    .str.upper()
)

selection_df = selection_df.dropna(
    subset=["pdb_id"]
).copy()

selection_df = selection_df[
    selection_df["pdb_id"] != ""
].copy()

# Проверка формата идентификаторов.
invalid_pdb_ids = selection_df.loc[
    ~selection_df["pdb_id"].str.fullmatch(
        r"[A-Z0-9]{4}"
    ),
    "pdb_id"
].tolist()

if invalid_pdb_ids:
    raise ValueError(
        "Обнаружены некорректные PDB ID: "
        f"{invalid_pdb_ids[:10]}"
    )

# Проверка уникальности.
duplicate_pdb_ids = (
    selection_df.loc[
        selection_df["pdb_id"].duplicated(keep=False),
        "pdb_id"
    ]
    .unique()
    .tolist()
)

if duplicate_pdb_ids:
    raise ValueError(
        "В selection CSV найдены повторяющиеся PDB ID: "
        f"{duplicate_pdb_ids[:10]}"
    )

# Сортировка по исходному порядку отбора.
if "selection_order" in selection_df.columns:
    selection_df["selection_order"] = pd.to_numeric(
        selection_df["selection_order"],
        errors="coerce"
    )

    selection_df = selection_df.sort_values(
        "selection_order",
        na_position="last"
    ).reset_index(drop=True)
else:
    selection_df = selection_df.reset_index(drop=True)

if selection_df.empty:
    raise ValueError(
        "В selection CSV отсутствуют записи PDB."
    )

# Проверка ожидаемого размера выборки.
if len(selection_df) != 100:
    warnings.warn(
        f"Ожидалось 100 структур, "
        f"но в CSV найдено {len(selection_df)}."
    )


# ============================================================
# 4. SELECT CURRENT PDB
# ============================================================

if PDB_ID_OVERRIDE is not None:

    PDB_ID = str(
        PDB_ID_OVERRIDE
    ).strip().upper()

    if not PDB_ID:
        raise ValueError(
            "PDB_ID_OVERRIDE не может быть пустым."
        )

else:
    PDB_ID = str(
        selection_df.iloc[0]["pdb_id"]
    )

# Метаданные выбранной структуры.
matching_rows = selection_df.loc[
    selection_df["pdb_id"] == PDB_ID
]

if not matching_rows.empty:
    PDB_METADATA = (
        matching_rows.iloc[0].to_dict()
    )
    PDB_IS_SELECTED = True
else:
    # Разрешаем отдельные тестовые структуры,
    # отсутствующие в финальном списке 100 PDB.
    PDB_METADATA = {}
    PDB_IS_SELECTED = False

    warnings.warn(
        f"PDB {PDB_ID} отсутствует в "
        "protein_final_100.csv. "
        "Структура будет обработана как тестовая."
    )


# ============================================================
# 5. DIRECTORY CONFIGURATION
# ============================================================

# PROJECT_DIR сохраняется для совместимости
# с последующими ячейками исходного ноутбука.
PROJECT_DIR = STRUCTURE_ANALYSIS_DIR

# DATA_DIR содержит исходные PDB/mmCIF-файлы.
DATA_DIR = STRUCTURES_DIR

# RESULTS_DIR индивидуален для каждой структуры.
RESULTS_DIR = RESULTS_ROOT / PDB_ID

# Путь к текущему mmCIF-файлу.
CIF_PATH = DATA_DIR / f"{PDB_ID}.cif"

# Создание необходимых каталогов.
for directory in [
    PROJECT_DIR,
    DATA_DIR,
    RESULTS_ROOT,
    RESULTS_DIR
]:
    directory.mkdir(
        parents=True,
        exist_ok=True
    )


# ============================================================
# 6. CONFIGURATION SUMMARY
# ============================================================

print("=" * 70)
print("PROTEIN STRUCTURE ANALYSIS")
print("=" * 70)

print(f"Project root:       {PROJECT_ROOT}")
print(f"Dataset root:       {DATASET_ROOT}")
print(f"Selection CSV:      {SELECTION_CSV}")

print("-" * 70)

print(f"Selected PDB ID:    {PDB_ID}")
print(f"Is selected PDB:    {PDB_IS_SELECTED}")
print(f"Selection size:     {len(selection_df)}")

if "selection_order" in PDB_METADATA:
    print(
        "Selection order:   ",
        PDB_METADATA["selection_order"]
    )

if "dominant_category" in PDB_METADATA:
    print(
        "Functional category:",
        PDB_METADATA["dominant_category"]
    )

print("-" * 70)

print(f"Project directory:  {PROJECT_DIR}")
print(f"Data directory:     {DATA_DIR}")
print(f"Results directory:  {RESULTS_DIR}")
print(f"CIF file:           {CIF_PATH}")

print("=" * 70)
```

```python
# ============================================================
# START TEXT REPORT
# ============================================================

import sys
from pathlib import Path

class TeeOutput:

    """Display output in Jupyter and simultaneously save it to a file."""

    def __init__(self, original_stream, log_file):
        self.original_stream = original_stream
        self.log_file = log_file

    def write(self, text):
        self.original_stream.write(text)
        self.log_file.write(text)
        self.log_file.flush()

    def flush(self):
        self.original_stream.flush()
        self.log_file.flush()

    def isatty(self):
        return self.original_stream.isatty()

    @property
    def encoding(self):
        return self.original_stream.encoding


# Create report file
REPORT_PATH = RESULTS_DIR / f"{PDB_ID}_analysis_report.txt"

report_file = open(
    REPORT_PATH,
    "w",
    encoding="utf-8"
)

# Save original output streams
original_stdout = sys.stdout
original_stderr = sys.stderr

# Redirect output to screen + file
sys.stdout = TeeOutput(original_stdout, report_file)
sys.stderr = TeeOutput(original_stderr, report_file)

print("=" * 70)
print("PROTEIN STRUCTURE ANALYSIS REPORT")
print("=" * 70)

print("PDB ID:", PDB_ID)
print("Report file:", REPORT_PATH)
print(" ")
```

```python

# ============================================================
# CELL 3. DOWNLOAD PDB STRUCTURE (mmCIF)
# ============================================================
# Назначение:
# Загрузка mmCIF-файла для текущей структуры PDB.
#
# Поддерживает:
# 1. Пакетную обработку разных PDB_ID.
# 2. Повторное использование ранее загруженных файлов.
# 3. Проверку HTTP-ответа и содержимого файла.
# 4. Защиту от сохранения пустых и некорректных файлов.
# 5. Повторную загрузку повреждённого файла.
# ============================================================

PDB_ID = str(PDB_ID).upper().strip()

# ------------------------------------------------------------
# 1. Проверка PDB ID
# ------------------------------------------------------------

if len(PDB_ID) != 4 or not PDB_ID.isalnum():
    raise ValueError(
        f"Некорректный PDB ID: {PDB_ID}. "
        "Ожидается 4 буквенно-цифровых символа."
    )

# ------------------------------------------------------------
# 2. Подготовка пути
# ------------------------------------------------------------

CIF_PATH = DATA_DIR / f"{PDB_ID}.cif"

url = f"https://files.rcsb.org/download/{PDB_ID}.cif"

# ------------------------------------------------------------
# 3. Проверка существующего файла
# ------------------------------------------------------------

def validate_cif_file(path):
    """
    Минимальная проверка наличия и содержимого mmCIF-файла.
    Полноценный разбор структуры выполняется в следующих ячейках.
    """

    if not path.exists():
        return False

    if path.stat().st_size < 100:
        return False

    try:
        with open(path, "r", encoding="utf-8", errors="replace") as f:
            content = f.read(20000)

        if "data_" not in content:
            return False

        if "_atom_site." not in content:
            return False

        return True

    except OSError:
        return False


# ------------------------------------------------------------
# 4. Загрузка при отсутствии или повреждении файла
# ------------------------------------------------------------

if validate_cif_file(CIF_PATH):

    print(f"Используется существующий файл: {PDB_ID}")

else:

    if CIF_PATH.exists():
        print("Обнаружен некорректный файл. Выполняется повторная загрузка.")
        CIF_PATH.unlink()

    print(f"Загрузка структуры {PDB_ID} из RCSB PDB...")

    response = requests.get(
        url,
        timeout=(15, 120),
        headers={"User-Agent": "ProteinStructureAnalyzer/1.0"}
    )

    response.raise_for_status()

    if not response.content:
        raise RuntimeError(
            f"RCSB вернул пустой файл для {PDB_ID}."
        )

    # Проверка содержимого до сохранения
    content = response.content.decode(
        "utf-8",
        errors="replace"
    )

    if "data_" not in content[:20000]:
        raise RuntimeError(
            f"Полученный файл {PDB_ID} не похож на mmCIF."
        )

    if "_atom_site." not in content:
        raise RuntimeError(
            f"В файле {PDB_ID} отсутствует секция atom_site."
        )

    # Сначала временный файл, затем атомарная замена
    temp_path = CIF_PATH.with_suffix(".cif.tmp")

    try:
        temp_path.write_bytes(response.content)
        temp_path.replace(CIF_PATH)

    finally:
        if temp_path.exists():
            temp_path.unlink()

# ------------------------------------------------------------
# 5. Финальная проверка
# ------------------------------------------------------------

if not validate_cif_file(CIF_PATH):
    raise RuntimeError(
        f"Файл структуры {PDB_ID} не прошёл проверку: {CIF_PATH}"
    )

file_size_kb = CIF_PATH.stat().st_size / 1024

print("\nSTRUCTURE DOWNLOAD")
print("-" * 60)
print("PDB ID:", PDB_ID)
print("File:", CIF_PATH.resolve())
print("Size:", f"{file_size_kb:.2f} KB")
print("Validation: PASSED")
```

```python

# ============================================================
# CELL 4. ЧТЕНИЕ СТРУКТУРЫ БЕЛКА
# ============================================================
# Назначение:
# 1. Чтение mmCIF-файла.
# 2. Получение моделей и цепей.
# 3. Подсчёт белковых, небелковых остатков и воды.
# 4. Подсчёт атомов по категориям.
# 5. Диагностика альтернативных положений атомов.
# 6. Сохранение статистики структуры и отдельных цепей.
#
# ВАЖНО:
# Для единообразия последующего анализа используется первая
# модель структуры. Остальные модели учитываются в статистике,
# но не включаются в подсчёты.
# ============================================================

import json
from pathlib import Path

from Bio.PDB.Polypeptide import is_aa


print("=" * 70)
print("CELL 4. ЧТЕНИЕ СТРУКТУРЫ БЕЛКА")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка исходных данных
# ------------------------------------------------------------

if "PDB_ID" not in globals():
    raise NameError("PDB_ID не определён.")

if "CIF_PATH" not in globals():
    raise NameError("CIF_PATH не определён.")

if "RESULTS_DIR" not in globals():
    raise NameError("RESULTS_DIR не определён.")

PDB_ID = str(PDB_ID).upper().strip()
cif_path = Path(CIF_PATH)

if not cif_path.is_file():
    raise FileNotFoundError(
        f"Файл структуры не найден: {cif_path}"
    )

if cif_path.stat().st_size == 0:
    raise ValueError("Файл структуры пустой.")

# Отдельный каталог для результатов конкретного PDB.
PDB_RESULTS_DIR = Path(RESULTS_DIR) / PDB_ID
PDB_RESULTS_DIR.mkdir(parents=True, exist_ok=True)

print(f"PDB ID: {PDB_ID}")
print(f"Файл структуры: {cif_path.resolve()}")
print(f"Каталог результатов: {PDB_RESULTS_DIR.resolve()}")
print(f"Размер файла: {cif_path.stat().st_size / 1024:.2f} KB")


# ------------------------------------------------------------
# 2. Чтение mmCIF
# ------------------------------------------------------------

parser = MMCIFParser(QUIET=True)

try:
    structure = parser.get_structure(
        PDB_ID,
        str(cif_path)
    )
except Exception as e:
    raise RuntimeError(
        f"Ошибка чтения mmCIF для {PDB_ID}: {e}"
    ) from e

models = list(structure.get_models())

if not models:
    raise ValueError(
        f"В структуре {PDB_ID} не обнаружено ни одной модели."
    )

# Для последующего анализа используем первую модель.
model = models[0]

chains = list(model.get_chains())

if not chains:
    raise ValueError(
        f"В первой модели {PDB_ID} не обнаружено цепей."
    )


# ------------------------------------------------------------
# 3. Настройки классификации остатков
# ------------------------------------------------------------

WATER_NAMES = {"HOH", "WAT", "DOD"}

chain_statistics = []

# Общие диагностические счётчики.
total_altloc_atom_sites = 0
total_altloc_variants = 0
total_disordered_residues = 0

all_atom_count = 0


# ------------------------------------------------------------
# 4. Подсчёт характеристик каждой цепи
# ------------------------------------------------------------

for chain in chains:

    residues = list(chain.get_residues())

    protein_residues = [
        residue
        for residue in residues
        if is_aa(residue, standard=False)
    ]

    water_residues = [
        residue
        for residue in residues
        if residue.resname.strip().upper() in WATER_NAMES
    ]

    protein_residue_ids = {
        id(residue)
        for residue in protein_residues
    }

    water_residue_ids = {
        id(residue)
        for residue in water_residues
    }

    nonprotein_residues = [
        residue
        for residue in residues
        if (
            id(residue) not in protein_residue_ids
            and id(residue) not in water_residue_ids
        )
    ]

    # Атомные счётчики цепи.
    protein_atoms = 0
    nonprotein_atoms = 0
    water_atoms = 0

    # Диагностика альтернативных положений.
    chain_altloc_atom_sites = 0
    chain_altloc_variants = 0

    for residue in residues:

        atoms = list(residue.get_atoms())

        if id(residue) in protein_residue_ids:
            protein_atoms += len(atoms)

        elif id(residue) in water_residue_ids:
            water_atoms += len(atoms)

        else:
            nonprotein_atoms += len(atoms)

        # DisorderedAtom представляет один атомный сайт,
        # который может иметь несколько альтернативных вариантов.
        for atom in atoms:

            if atom.is_disordered() == 2:

                variants = len(atom.child_dict)

                if variants > 1:
                    chain_altloc_atom_sites += 1
                    chain_altloc_variants += variants

    total_atoms = (
        protein_atoms
        + nonprotein_atoms
        + water_atoms
    )

    chain_disordered_residues = sum(
        1
        for residue in residues
        if residue.is_disordered()
    )

    chain_id = str(chain.id)

    # Пустой идентификатор цепи сохраняем явно.
    if chain_id == " ":
        chain_id = "<blank>"

    chain_statistics.append({

        "pdb_id": PDB_ID,
        "chain_id": chain_id,

        "total_residues": len(residues),
        "protein_residues": len(protein_residues),
        "nonprotein_residues": len(nonprotein_residues),
        "water_residues": len(water_residues),

        "protein_atoms": protein_atoms,
        "nonprotein_atoms": nonprotein_atoms,
        "water_atoms": water_atoms,
        "total_atoms": total_atoms,

        "altloc_atom_sites": chain_altloc_atom_sites,
        "altloc_variants": chain_altloc_variants,
        "disordered_residues": chain_disordered_residues
    })

    total_altloc_atom_sites += chain_altloc_atom_sites
    total_altloc_variants += chain_altloc_variants
    total_disordered_residues += chain_disordered_residues

    all_atom_count += total_atoms


# ------------------------------------------------------------
# 5. Общая статистика первой модели
# ------------------------------------------------------------

total_residues = sum(
    item["total_residues"]
    for item in chain_statistics
)

protein_residue_count = sum(
    item["protein_residues"]
    for item in chain_statistics
)

nonprotein_residue_count = sum(
    item["nonprotein_residues"]
    for item in chain_statistics
)

water_residue_count = sum(
    item["water_residues"]
    for item in chain_statistics
)

protein_atom_count = sum(
    item["protein_atoms"]
    for item in chain_statistics
)

nonprotein_atom_count = sum(
    item["nonprotein_atoms"]
    for item in chain_statistics
)

water_atom_count = sum(
    item["water_atoms"]
    for item in chain_statistics
)

total_atom_count = sum(
    item["total_atoms"]
    for item in chain_statistics
)


# ------------------------------------------------------------
# 6. Проверка корректности подсчётов
# ------------------------------------------------------------

calculated_atom_sum = (
    protein_atom_count
    + nonprotein_atom_count
    + water_atom_count
)

atom_count_check = (
    calculated_atom_sum == total_atom_count
)

if total_atom_count == 0:
    raise ValueError(
        f"В первой модели {PDB_ID} не обнаружено атомов."
    )

if protein_residue_count == 0:
    raise ValueError(
        f"В первой модели {PDB_ID} не обнаружены белковые остатки. "
        "Проверьте структуру и её соответствие входному набору."
    )

if not atom_count_check:
    raise ValueError(
        "Сумма категорий атомов не совпадает "
        "с общим количеством атомов."
    )


# ------------------------------------------------------------
# 7. Формирование общей статистики
# ------------------------------------------------------------

structure_statistics = {

    "pdb_id": PDB_ID,
    "source_file": str(cif_path.resolve()),

    "model_count": len(models),
    "analyzed_model": str(model.id),
    "analysis_uses_first_model_only": True,

    "chain_count": len(chains),

    "total_residues": total_residues,
    "protein_residues": protein_residue_count,
    "nonprotein_residues": nonprotein_residue_count,
    "water_residues": water_residue_count,

    "protein_atoms": protein_atom_count,
    "nonprotein_atoms": nonprotein_atom_count,
    "water_atoms": water_atom_count,
    "total_atoms": total_atom_count,

    "altloc_atom_sites": total_altloc_atom_sites,
    "altloc_variants": total_altloc_variants,
    "disordered_residues": total_disordered_residues,

    "atom_count_check": atom_count_check,
    "status": "OK"
}


# ------------------------------------------------------------
# 8. Сохранение результатов
# ------------------------------------------------------------

chain_statistics_df = pd.DataFrame(chain_statistics)

chain_statistics_path = (
    PDB_RESULTS_DIR / f"{PDB_ID}_chain_statistics.csv"
)

structure_statistics_path = (
    PDB_RESULTS_DIR / f"{PDB_ID}_structure_statistics.json"
)

chain_statistics_df.to_csv(
    chain_statistics_path,
    index=False,
    encoding="utf-8-sig"
)

with open(
    structure_statistics_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        structure_statistics,
        f,
        ensure_ascii=False,
        indent=4
    )


# ------------------------------------------------------------
# 9. Вывод общей информации
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ОБЩАЯ ИНФОРМАЦИЯ О СТРУКТУРЕ")
print("-" * 70)

print(f"Количество моделей:              {len(models)}")
print(f"Анализируемая модель:             {model.id}")
print(f"Количество цепей:                 {len(chains)}")

print(f"\nВсего остатков:                   {total_residues}")
print(f"Белковых остатков:                {protein_residue_count}")
print(f"Небелковых остатков:              {nonprotein_residue_count}")
print(f"Остатков воды:                    {water_residue_count}")

print(f"\nВсего атомов:                     {total_atom_count}")
print(f"Белковых атомов:                  {protein_atom_count}")
print(f"Небелковых атомов:                {nonprotein_atom_count}")
print(f"Атомов воды:                      {water_atom_count}")

print(f"\nАтомных сайтов с altloc:          {total_altloc_atom_sites}")
print(f"Альтернативных вариантов атомов:  {total_altloc_variants}")
print(f"Дискретных остатков:              {total_disordered_residues}")


# ------------------------------------------------------------
# 10. Статистика по отдельным цепям
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("СТАТИСТИКА ПО ЦЕПЯМ")
print("-" * 70)

for item in chain_statistics:

    print(f"\nChain {item['chain_id']}")

    print(
        f"  Остатки: всего={item['total_residues']}, "
        f"белковые={item['protein_residues']}, "
        f"небелковые={item['nonprotein_residues']}, "
        f"вода={item['water_residues']}"
    )

    print(
        f"  Атомы: всего={item['total_atoms']}, "
        f"белковые={item['protein_atoms']}, "
        f"небелковые={item['nonprotein_atoms']}, "
        f"вода={item['water_atoms']}"
    )

    print(
        f"  Altloc: сайтов={item['altloc_atom_sites']}, "
        f"вариантов={item['altloc_variants']}"
    )


# ------------------------------------------------------------
# 11. Контроль и сохранённые файлы
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("КОНТРОЛЬ РЕЗУЛЬТАТОВ")
print("-" * 70)

print(
    "Проверка суммы категорий атомов:",
    "OK" if atom_count_check else "ERROR"
)

print(
    "Наличие белковых остатков:",
    "OK" if protein_residue_count > 0 else "ERROR"
)

print("\nСохранённые файлы:")
print(chain_statistics_path)
print(structure_statistics_path)

print("\nСтатус: OK")
print("Структура прочитана.")
print("Статистика подготовлена для последующих ячеек.")

print("=" * 70)
print("CELL 4 COMPLETED")
print("=" * 70)
```

```python

# ============================================================
# CELL 5. АНАЛИЗ АМИНОКИСЛОТНЫХ ПОСЛЕДОВАТЕЛЬНОСТЕЙ
# ============================================================
# Назначение:
# 1. Извлечение последовательностей белковых цепей.
# 2. Преобразование трехбуквенных обозначений в однобуквенный формат.
# 3. Обработка распространённых нестандартных аминокислот.
# 4. Контроль неизвестных и нестандартных остатков.
# 5. Сохранение FASTA и статистики для каждой структуры.
#
# ВАЖНО:
# Каждая цепь обрабатывается отдельно.
# Последовательности разных цепей не объединяются.
# ============================================================

import json
from pathlib import Path

from Bio.SeqUtils import seq1
from Bio.PDB.Polypeptide import is_aa


print("=" * 70)
print("CELL 5. АНАЛИЗ АМИНОКИСЛОТНЫХ ПОСЛЕДОВАТЕЛЬНОСТЕЙ")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка входных данных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "chains",
    "PDB_RESULTS_DIR"
]

for variable in required_variables:

    if variable not in globals():
        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )

PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)
PDB_RESULTS_DIR.mkdir(parents=True, exist_ok=True)

print(f"PDB ID: {PDB_ID}")
print(f"Количество цепей для анализа: {len(chains)}")
print(f"Каталог результатов: {PDB_RESULTS_DIR.resolve()}")


# ------------------------------------------------------------
# 2. Карта нестандартных аминокислот
# ------------------------------------------------------------

CUSTOM_AMINO_ACIDS = {

    "MSE": "M",  # Селено-метионин
    "SEC": "U",  # Селеноцистеин
    "PYL": "O",  # Пирролизин

    "HYP": "P",  # Гидроксипролин
    "CSO": "C",  # Окисленный цистеин
    "CME": "C",  # Модифицированный цистеин
    "MLY": "K",  # Метиллизин
    "M3L": "K",  # Триметиллизин
    "KCX": "K",  # Карбоксилированный лизин
    "FME": "M",  # Формилметионин
    "SEP": "S",  # Фосфосерин
    "TPO": "T",  # Фосфотреонин
    "PTR": "Y",  # Фосфотирозин
    "HID": "H",
    "HIE": "H",
    "HIP": "H"
}


# ------------------------------------------------------------
# 3. Извлечение последовательностей
# ------------------------------------------------------------

chain_sequences = {}
sequence_statistics = []

for chain in chains:

    sequence = []

    unknown_residues = []
    nonstandard_residues = []

    protein_residues = [
        residue
        for residue in chain.get_residues()
        if is_aa(residue, standard=False)
    ]

    for residue in protein_residues:

        residue_name = residue.resname.strip().upper()

        amino_acid = seq1(
            residue_name,
            custom_map=CUSTOM_AMINO_ACIDS,
            undef_code="X"
        )

        sequence.append(amino_acid)

        if residue_name in CUSTOM_AMINO_ACIDS:

            nonstandard_residues.append(residue_name)

        elif amino_acid == "X":

            unknown_residues.append(residue_name)

    sequence = "".join(sequence)

    # Сохраняем фактический ID цепи.
    chain_sequences[chain.id] = sequence

    sequence_statistics.append({

        "pdb_id": PDB_ID,

        "chain_id": str(chain.id),

        "residue_count": len(protein_residues),

        "sequence_length": len(sequence),

        "unknown_residue_count": sequence.count("X"),

        "unknown_residue_names": sorted(
            set(unknown_residues)
        ),

        "nonstandard_residue_count": len(nonstandard_residues),

        "nonstandard_residue_names": sorted(
            set(nonstandard_residues)
        ),

        "sequence_empty": len(sequence) == 0
    })


# ------------------------------------------------------------
# 4. Контроль извлечённых последовательностей
# ------------------------------------------------------------

total_sequence_length = sum(
    item["sequence_length"]
    for item in sequence_statistics
)

total_unknown_count = sum(
    item["unknown_residue_count"]
    for item in sequence_statistics
)

total_nonstandard_count = sum(
    item["nonstandard_residue_count"]
    for item in sequence_statistics
)

empty_sequence_count = sum(
    item["sequence_empty"]
    for item in sequence_statistics
)

sequence_length_check = all(
    item["residue_count"] == item["sequence_length"]
    for item in sequence_statistics
)

if not sequence_length_check:

    raise ValueError(
        "Обнаружено несоответствие между количеством "
        "белковых остатков и длиной последовательности."
    )

if total_sequence_length == 0:

    raise ValueError(
        f"Для {PDB_ID} не удалось извлечь ни одной "
        "аминокислотной последовательности."
    )


# ------------------------------------------------------------
# 5. Сохранение последовательностей FASTA
# ------------------------------------------------------------

FASTA_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_sequences.fasta"
)

with open(FASTA_PATH, "w", encoding="utf-8") as file:

    for chain_id, sequence in chain_sequences.items():

        if not sequence:
            continue

        fasta_chain_id = (
            str(chain_id)
            if str(chain_id).strip()
            else "<blank>"
        )

        file.write(
            f">{PDB_ID}|chain={fasta_chain_id}\n"
        )

        for i in range(0, len(sequence), 80):

            file.write(
                sequence[i:i + 80] + "\n"
            )


# Совместимость с предыдущими ячейками.
fasta_path = FASTA_PATH


# ------------------------------------------------------------
# 6. Формирование статистики
# ------------------------------------------------------------

sequence_summary = {

    "pdb_id": PDB_ID,

    "chain_count": len(chain_sequences),

    "nonempty_chain_count": (
        len(chain_sequences) - empty_sequence_count
    ),

    "empty_chain_count": empty_sequence_count,

    "total_sequence_length": total_sequence_length,

    "unknown_residue_count": total_unknown_count,

    "nonstandard_residue_count": total_nonstandard_count,

    "sequence_length_check": sequence_length_check,

    "fasta_path": str(FASTA_PATH.resolve()),

    "chain_statistics": sequence_statistics
}


# ------------------------------------------------------------
# 7. Сохранение статистики в файлы
# ------------------------------------------------------------

sequence_statistics_df = pd.DataFrame(
    sequence_statistics
)

sequence_statistics_path = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_sequence_statistics.csv"
)

sequence_summary_path = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_sequence_summary.json"
)

sequence_statistics_df.to_csv(
    sequence_statistics_path,
    index=False,
    encoding="utf-8-sig"
)

with open(
    sequence_summary_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        sequence_summary,
        file,
        ensure_ascii=False,
        indent=4
    )


# ------------------------------------------------------------
# 8. Вывод последовательностей
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ПОСЛЕДОВАТЕЛЬНОСТИ БЕЛКОВЫХ ЦЕПЕЙ")
print("-" * 70)

for chain_id, sequence in chain_sequences.items():

    display_chain_id = (
        str(chain_id)
        if str(chain_id).strip()
        else "<blank>"
    )

    print(f"\nChain {display_chain_id}")

    print(f"Длина последовательности: {len(sequence)}")

    print("Последовательность:")

    print(
        sequence
        if sequence
        else "Пустая последовательность"
    )

    print("-" * 70)


# ------------------------------------------------------------
# 9. Вывод технической статистики
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ТЕХНИЧЕСКАЯ СТАТИСТИКА")
print("-" * 70)

for item in sequence_statistics:

    print(f"\nChain {item['chain_id']}")

    print(
        f"  Белковых остатков: "
        f"{item['residue_count']}"
    )

    print(
        f"  Длина последовательности: "
        f"{item['sequence_length']}"
    )

    print(
        f"  Неизвестных остатков (X): "
        f"{item['unknown_residue_count']}"
    )

    print(
        f"  Нестандартных остатков: "
        f"{item['nonstandard_residue_count']}"
    )

    if item["unknown_residue_names"]:

        print(
            "  Неизвестные обозначения:",
            item["unknown_residue_names"]
        )

    if item["nonstandard_residue_names"]:

        print(
            "  Нестандартные обозначения:",
            item["nonstandard_residue_names"]
        )


# ------------------------------------------------------------
# 10. Итоговый контроль
# ------------------------------------------------------------

if not FASTA_PATH.is_file():

    raise FileNotFoundError(
        f"FASTA-файл не был создан: {FASTA_PATH}"
    )

if FASTA_PATH.stat().st_size == 0:

    raise ValueError(
        f"FASTA-файл пустой: {FASTA_PATH}"
    )

print("\n" + "-" * 70)
print("ИТОГИ CELL 5")
print("-" * 70)

print(f"Количество цепей: {len(chain_sequences)}")

print(f"Общая длина последовательностей: {total_sequence_length}")

print(f"Неизвестные аминокислоты (X): {total_unknown_count}")

print(f"Нестандартные аминокислоты: {total_nonstandard_count}")

print(f"Цепей без последовательности: {empty_sequence_count}")

print(
    "Проверка соответствия остатков и длины:",
    "OK" if sequence_length_check else "ERROR"
)

print(f"\nFASTA: {FASTA_PATH}")
print(f"Размер FASTA: {FASTA_PATH.stat().st_size} bytes")

print(f"CSV статистики: {sequence_statistics_path}")
print(f"JSON статистики: {sequence_summary_path}")

print("\nСтатус: OK")
print("Последовательности подготовлены для дальнейшего анализа.")

print("=" * 70)
print("CELL 5 COMPLETED")
print("=" * 70)
```

```python

# ============================================================
# CELL 6. АНАЛИЗ АМИНОКИСЛОТНЫХ ОСТАТКОВ
# ============================================================
# Назначение:
# 1. Формирование таблицы белковых остатков.
# 2. Контроль основных атомов пептидного остова.
# 3. Проверка соответствия последовательностям CELL 5.
# 4. Сверка количества остатков с CELL 4.
# 5. Контроль повторяющихся идентификаторов.
# 6. Сохранение результатов отдельно для каждого PDB.
#
# ВАЖНО:
# Сохраняется исходная идентификация цепей и остатков.
# Проверка последовательностей выполняется по символам,
# а не только по длине.
# ============================================================

import json
import pandas as pd
from pathlib import Path
from Bio.SeqUtils import seq1
from Bio.PDB.Polypeptide import is_aa


print("=" * 70)
print("CELL 6. АНАЛИЗ АМИНОКИСЛОТНЫХ ОСТАТКОВ")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка входных данных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "chains",
    "chain_sequences",
    "PDB_RESULTS_DIR",
    "CUSTOM_AMINO_ACIDS"
]

for variable in required_variables:

    if variable not in globals():
        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )

PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)
PDB_RESULTS_DIR.mkdir(parents=True, exist_ok=True)

print(f"PDB ID: {PDB_ID}")
print(f"Количество цепей: {len(chains)}")
print(f"Каталог результатов: {PDB_RESULTS_DIR.resolve()}")


# ------------------------------------------------------------
# 2. Настройки анализа
# ------------------------------------------------------------

# Используем карту нестандартных аминокислот из CELL 5.
# Не переопределяем её, чтобы избежать расхождения
# между последовательностями CELL 5 и CELL 6.

BACKBONE_ATOMS = ["N", "CA", "C", "O"]


def normalize_chain_id(chain_id):
    """
    Нормализация идентификатора только для сравнения
    и отображения.

    Исходный идентификатор цепи при этом сохраняется.
    """

    chain_id = str(chain_id)

    if not chain_id.strip():
        return "<blank>"

    return chain_id


# ------------------------------------------------------------
# 3. Формирование записей по каждому остатку
# ------------------------------------------------------------

residue_records = []

for chain in chains:

    for residue in chain.get_residues():

        if not is_aa(residue, standard=False):
            continue

        residue_name = residue.resname.strip().upper()

        amino_acid = seq1(
            residue_name,
            custom_map=CUSTOM_AMINO_ACIDS,
            undef_code="X"
        )

        present_backbone_atoms = [
            atom_name
            for atom_name in BACKBONE_ATOMS
            if residue.has_id(atom_name)
        ]

        missing_backbone_atoms = [
            atom_name
            for atom_name in BACKBONE_ATOMS
            if not residue.has_id(atom_name)
        ]

        # Нестандартность определяем относительно
        # стандартного набора аминокислот Bio.PDB.
        is_nonstandard = not is_aa(
            residue,
            standard=True
        )

        residue_records.append({

            "pdb_id": PDB_ID,

            "chain_id": chain.id,

            "chain_id_display": normalize_chain_id(
                chain.id
            ),

            "hetero_flag": residue.id[0],

            "residue_id": residue.id[1],

            "insertion_code": residue.id[2].strip(),

            "residue_name": residue_name,

            "amino_acid": amino_acid,

            "atom_count": len(list(residue.get_atoms())),

            "has_N": residue.has_id("N"),

            "has_CA": residue.has_id("CA"),

            "has_C": residue.has_id("C"),

            "has_O": residue.has_id("O"),

            "present_backbone_atoms": ",".join(
                present_backbone_atoms
            ),

            "missing_backbone_atoms": ",".join(
                missing_backbone_atoms
            ),

            "missing_backbone_atom_count": len(
                missing_backbone_atoms
            ),

            "is_nonstandard": is_nonstandard,

            "is_custom_mapped": (
                residue_name in CUSTOM_AMINO_ACIDS
            ),

            "is_unknown": amino_acid == "X"

        })


# ------------------------------------------------------------
# 4. Создание DataFrame
# ------------------------------------------------------------

RESIDUE_COLUMNS = [

    "pdb_id",
    "chain_id",
    "chain_id_display",
    "hetero_flag",
    "residue_id",
    "insertion_code",
    "residue_name",
    "amino_acid",
    "atom_count",
    "has_N",
    "has_CA",
    "has_C",
    "has_O",
    "present_backbone_atoms",
    "missing_backbone_atoms",
    "missing_backbone_atom_count",
    "is_nonstandard",
    "is_custom_mapped",
    "is_unknown"

]

df_residues = pd.DataFrame(
    residue_records,
    columns=RESIDUE_COLUMNS
)

if df_residues.empty:

    raise ValueError(
        "Белковые остатки не обнаружены. "
        "Проверьте структуру и результат CELL 4."
    )


# ------------------------------------------------------------
# 5. Контроль основных атомов
# ------------------------------------------------------------

missing_backbone_df = df_residues.loc[
    df_residues["missing_backbone_atom_count"] > 0
].copy()

complete_backbone_count = int(
    (df_residues["missing_backbone_atom_count"] == 0).sum()
)

incomplete_backbone_count = len(missing_backbone_df)

missing_atom_counts = {

    atom_name: int(
        (~df_residues[f"has_{atom_name}"]).sum()
    )

    for atom_name in BACKBONE_ATOMS
}


# ------------------------------------------------------------
# 6. Статистика по цепям
# ------------------------------------------------------------

residue_chain_statistics = []

for chain_id, group in df_residues.groupby(
    "chain_id",
    sort=False,
    dropna=False
):

    residue_chain_statistics.append({

        "pdb_id": PDB_ID,

        "chain_id": chain_id,

        "chain_id_display": normalize_chain_id(chain_id),

        "residue_count": len(group),

        "atom_count": int(group["atom_count"].sum()),

        "complete_backbone_count": int(
            (group["missing_backbone_atom_count"] == 0).sum()
        ),

        "incomplete_backbone_count": int(
            (group["missing_backbone_atom_count"] > 0).sum()
        ),

        "unknown_residue_count": int(
            group["is_unknown"].sum()
        ),

        "nonstandard_residue_count": int(
            group["is_nonstandard"].sum()
        )

    })

df_residue_chain_statistics = pd.DataFrame(
    residue_chain_statistics
)


# ------------------------------------------------------------
# 7. Проверка соответствия CELL 5
# ------------------------------------------------------------

sequence_comparison = []

for chain_id, expected_sequence in chain_sequences.items():

    chain_residues = df_residues.loc[
        df_residues["chain_id"] == chain_id
    ]

    observed_sequence = "".join(
        chain_residues["amino_acid"].tolist()
    )

    observed_count = len(chain_residues)

    sequence_length = len(expected_sequence)

    length_matches = (
        observed_count == sequence_length
    )

    sequence_matches = (
        observed_sequence == expected_sequence
    )

    sequence_comparison.append({

        "pdb_id": PDB_ID,

        "chain_id": chain_id,

        "chain_id_display": normalize_chain_id(chain_id),

        "residue_count": observed_count,

        "sequence_length": sequence_length,

        "observed_sequence": observed_sequence,

        "expected_sequence": expected_sequence,

        "length_matches": length_matches,

        "sequence_matches": sequence_matches,

        "matches": (
            length_matches and sequence_matches
        )

    })

df_sequence_comparison = pd.DataFrame(
    sequence_comparison
)

sequence_match_check = (
    not df_sequence_comparison.empty
    and bool(df_sequence_comparison["matches"].all())
)


# ------------------------------------------------------------
# 8. Проверка соответствия CELL 4
# ------------------------------------------------------------

cell4_comparison = []

if "chain_statistics" in globals():

    for item in chain_statistics:

        chain_id = item["chain_id"]

        normalized_chain_id = normalize_chain_id(
            chain_id
        )

        observed_count = int(
            (
                df_residues["chain_id_display"]
                == normalized_chain_id
            ).sum()
        )

        expected_count = item["protein_residues"]

        cell4_comparison.append({

            "pdb_id": PDB_ID,

            "chain_id": chain_id,

            "chain_id_display": normalized_chain_id,

            "cell4_protein_residues": expected_count,

            "cell6_protein_residues": observed_count,

            "matches": expected_count == observed_count

        })

df_cell4_comparison = pd.DataFrame(
    cell4_comparison
)

if not df_cell4_comparison.empty:

    cell4_match_check = bool(
        df_cell4_comparison["matches"].all()
    )

else:

    cell4_match_check = None


# ------------------------------------------------------------
# 9. Проверка повторяющихся идентификаторов
# ------------------------------------------------------------

residue_key_columns = [

    "chain_id",
    "hetero_flag",
    "residue_id",
    "insertion_code"

]

duplicate_residues_df = df_residues.loc[
    df_residues.duplicated(
        subset=residue_key_columns,
        keep=False
    )
].copy()

duplicate_residue_count = len(
    duplicate_residues_df
)


# ------------------------------------------------------------
# 10. Общая статистика
# ------------------------------------------------------------

total_residue_count = len(df_residues)

total_residue_atoms = int(
    df_residues["atom_count"].sum()
)

unknown_residue_count = int(
    df_residues["is_unknown"].sum()
)

nonstandard_residue_count = int(
    df_residues["is_nonstandard"].sum()
)

custom_mapped_residue_count = int(
    df_residues["is_custom_mapped"].sum()
)

residue_analysis_summary = {

    "pdb_id": PDB_ID,

    "residue_count": total_residue_count,

    "chain_count": int(
        df_residues["chain_id"].nunique()
    ),

    "atom_count": total_residue_atoms,

    "complete_backbone_count": complete_backbone_count,

    "incomplete_backbone_count": incomplete_backbone_count,

    "missing_backbone_atom_counts": missing_atom_counts,

    "unknown_residue_count": unknown_residue_count,

    "nonstandard_residue_count": nonstandard_residue_count,

    "custom_mapped_residue_count": custom_mapped_residue_count,

    "duplicate_residue_count": duplicate_residue_count,

    "sequence_match_check": sequence_match_check,

    "cell4_match_check": cell4_match_check

}


# ------------------------------------------------------------
# 11. Сохранение результатов
# ------------------------------------------------------------

RESIDUE_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_residue_analysis.csv"
)

RESIDUE_CHAIN_CSV_PATH = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_residue_chain_statistics.csv"
)

MISSING_BACKBONE_CSV_PATH = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_missing_backbone_atoms.csv"
)

SEQUENCE_CHECK_CSV_PATH = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_sequence_verification.csv"
)

CELL4_CHECK_CSV_PATH = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_cell4_verification.csv"
)

DUPLICATE_RESIDUES_CSV_PATH = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_duplicate_residues.csv"
)

RESIDUE_SUMMARY_JSON_PATH = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_residue_analysis_summary.json"
)

df_residues.to_csv(
    RESIDUE_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

df_residue_chain_statistics.to_csv(
    RESIDUE_CHAIN_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

missing_backbone_df.to_csv(
    MISSING_BACKBONE_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

df_sequence_comparison.to_csv(
    SEQUENCE_CHECK_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

df_cell4_comparison.to_csv(
    CELL4_CHECK_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

duplicate_residues_df.to_csv(
    DUPLICATE_RESIDUES_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

with open(
    RESIDUE_SUMMARY_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        residue_analysis_summary,
        file,
        ensure_ascii=False,
        indent=4
    )


# ------------------------------------------------------------
# 12. Вывод таблиц
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ТАБЛИЦА БЕЛКОВЫХ ОСТАТКОВ")
print("-" * 70)

display(df_residues.head(10))

print("\n" + "-" * 70)
print("СТАТИСТИКА ПО ЦЕПЯМ")
print("-" * 70)

display(df_residue_chain_statistics)

print("\n" + "-" * 70)
print("ПРОВЕРКА ПОСЛЕДОВАТЕЛЬНОСТЕЙ CELL 5")
print("-" * 70)

display(
    df_sequence_comparison.drop(
        columns=["observed_sequence", "expected_sequence"],
        errors="ignore"
    )
)

print("\n" + "-" * 70)
print("ПРОВЕРКА ДАННЫХ CELL 4")
print("-" * 70)

if not df_cell4_comparison.empty:

    display(df_cell4_comparison)

else:

    print("Сравнение не выполнено: статистика CELL 4 недоступна.")


# ------------------------------------------------------------
# 13. Итоговая диагностика
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ИТОГОВАЯ СТАТИСТИКА")
print("-" * 70)

print(f"Всего белковых остатков: {total_residue_count}")

print(f"Всего атомов белковых остатков: {total_residue_atoms}")

print(f"Цепей: {df_residues['chain_id'].nunique()}")

print(f"Полностью представленных остатков: {complete_backbone_count}")

print(
    f"Остатков с отсутствующими атомами остова: "
    f"{incomplete_backbone_count}"
)

print("\nОтсутствующие основные атомы:")

for atom_name, count in missing_atom_counts.items():

    print(f"  {atom_name}: {count}")

print(f"\nНеизвестные аминокислоты (X): {unknown_residue_count}")

print(f"Нестандартные аминокислоты: {nonstandard_residue_count}")

print(f"Нестандартные остатки с заданным отображением: {custom_mapped_residue_count}")

print(f"Повторяющиеся идентификаторы: {duplicate_residue_count}")

print(
    "\nСоответствие последовательностям CELL 5:",
    "OK" if sequence_match_check else "ERROR"
)

if cell4_match_check is not None:

    print(
        "Соответствие подсчётам CELL 4:",
        "OK" if cell4_match_check else "ERROR"
    )

else:

    print("Соответствие CELL 4: не проверено")


# ------------------------------------------------------------
# 14. Проверка сохранённых файлов
# ------------------------------------------------------------

output_paths = [

    RESIDUE_CSV_PATH,
    RESIDUE_CHAIN_CSV_PATH,
    MISSING_BACKBONE_CSV_PATH,
    SEQUENCE_CHECK_CSV_PATH,
    CELL4_CHECK_CSV_PATH,
    DUPLICATE_RESIDUES_CSV_PATH,
    RESIDUE_SUMMARY_JSON_PATH

]

print("\n" + "-" * 70)
print("СОХРАНЁННЫЕ ФАЙЛЫ")
print("-" * 70)

for path in output_paths:

    file_ok = path.is_file() and path.stat().st_size > 0

    print(
        f"{path.name}: "
        f"{'OK' if file_ok else 'ERROR'}"
    )

    if not file_ok:

        raise RuntimeError(
            f"Файл не сохранён или пустой: {path}"
        )


# ------------------------------------------------------------
# 15. Итоговый статус
# ------------------------------------------------------------

if not sequence_match_check:

    raise ValueError(
        "Последовательности CELL 5 и CELL 6 не совпадают. "
        "Проверьте идентификаторы цепей, порядок остатков "
        "и карту нестандартных аминокислот."
    )

if cell4_match_check is False:

    raise ValueError(
        "Количество остатков не соответствует статистике CELL 4."
    )

print("\nСтатус: OK")
print("Таблица остатков подготовлена для дальнейшего анализа.")

print("=" * 70)
print("CELL 6 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# CELL 7. АНАЛИЗ АТОМОВ И ИХ КООРДИНАТ
# ============================================================
# Назначение:
# 1. Извлечение координат белковых атомов.
# 2. Сохранение химических и структурных характеристик атомов.
# 3. Контроль координат, occupancy, B-factor и altloc.
# 4. Формирование статистики по атомам, элементам и цепям.
# 5. Сверка с результатами CELL 4 и CELL 6.
# 6. Сохранение таблиц и итогового JSON-отчёта.
#
# ВАЖНО:
# Используется тот же способ обхода атомов, что и в CELL 4
# и CELL 6. Это необходимо для сопоставимости результатов.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.PDB import is_aa
from Bio.SeqUtils import seq1
from IPython.display import display


print("=" * 70)
print("CELL 7. АНАЛИЗ АТОМОВ И ИХ КООРДИНАТ")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка входных данных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "chains",
    "PDB_RESULTS_DIR",
    "df_residues",
    "CUSTOM_AMINO_ACIDS"
]

for variable in required_variables:
    if variable not in globals():
        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )

PDB_ID = str(PDB_ID).strip().upper()

PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)
PDB_RESULTS_DIR.mkdir(parents=True, exist_ok=True)

if not isinstance(CUSTOM_AMINO_ACIDS, dict):
    raise TypeError(
        "CUSTOM_AMINO_ACIDS должна быть словарём."
    )

if df_residues.empty:
    raise ValueError(
        "Таблица df_residues пуста. "
        "Проверьте результаты CELL 6."
    )

required_residue_columns = [
    "chain_id",
    "hetero_flag",
    "residue_id",
    "insertion_code",
    "residue_name",
    "atom_count"
]

missing_columns = [
    column
    for column in required_residue_columns
    if column not in df_residues.columns
]

if missing_columns:
    raise ValueError(
        f"В df_residues отсутствуют столбцы: {missing_columns}"
    )

print(f"PDB ID: {PDB_ID}")
print(f"Количество цепей: {len(chains)}")
print(f"Каталог результатов: {PDB_RESULTS_DIR}")


# ------------------------------------------------------------
# 2. Извлечение координат белковых атомов
# ------------------------------------------------------------

atom_records = []

for chain in chains:

    for residue in chain.get_residues():

        if not is_aa(residue, standard=False):
            continue

        residue_name = residue.resname.strip().upper()

        amino_acid = seq1(
            residue_name,
            custom_map=CUSTOM_AMINO_ACIDS,
            undef_code="X"
        )

        residue_number = int(residue.id[1])
        insertion_code = residue.id[2].strip()
        hetero_flag = residue.id[0]

        # Уникальный идентификатор остатка внутри структуры.
        residue_key = (
            f"{chain.id}:{hetero_flag}:"
            f"{residue_number}:{insertion_code}"
        )

        # Используем get_atoms(), как в CELL 4 и CELL 6.
        # Для DisorderedAtom учитывается выбранный вариант.

        for atom in residue.get_atoms():

            coord = atom.coord

            element = (
                atom.element.strip().upper()
                if atom.element
                else "UNKNOWN"
            )

            atom_name = atom.get_name().strip()

            occupancy = atom.get_occupancy()
            bfactor = atom.get_bfactor()

            altloc = atom.get_altloc()

            atom_records.append({

                "pdb_id": PDB_ID,

                "chain_id": chain.id,

                "residue_key": residue_key,

                "hetero_flag": hetero_flag,

                "residue_id": residue_number,

                "insertion_code": insertion_code,

                "residue_name": residue_name,

                "amino_acid": amino_acid,

                "atom_serial": atom.get_serial_number(),

                "atom_name": atom_name,

                "atom_fullname": atom.get_fullname().strip(),

                "element": element,

                "x": float(coord[0]),

                "y": float(coord[1]),

                "z": float(coord[2]),

                "occupancy": occupancy,

                "bfactor": bfactor,

                "altloc": altloc.strip() if altloc else "",

                "is_backbone": atom_name in {
                    "N", "CA", "C", "O", "OXT"
                },

                "is_hydrogen": element in {"H", "D"}

            })


# ------------------------------------------------------------
# 3. Создание таблицы атомов
# ------------------------------------------------------------

ATOM_COLUMNS = [

    "pdb_id",
    "chain_id",
    "residue_key",
    "hetero_flag",
    "residue_id",
    "insertion_code",
    "residue_name",
    "amino_acid",
    "atom_serial",
    "atom_name",
    "atom_fullname",
    "element",
    "x",
    "y",
    "z",
    "occupancy",
    "bfactor",
    "altloc",
    "is_backbone",
    "is_hydrogen"

]

df_atoms = pd.DataFrame(
    atom_records,
    columns=ATOM_COLUMNS
)

if df_atoms.empty:
    raise ValueError(
        "Белковые атомы не обнаружены. "
        "Проверьте результаты CELL 4 и CELL 6."
    )

print(f"\nИзвлечено белковых атомов: {len(df_atoms)}")


# ------------------------------------------------------------
# 4. Проверка координат
# ------------------------------------------------------------

coordinate_columns = ["x", "y", "z"]

coordinate_array = df_atoms[
    coordinate_columns
].to_numpy(dtype=float)

finite_coordinate_mask = np.isfinite(
    coordinate_array
).all(axis=1)

invalid_coordinates_df = df_atoms.loc[
    ~finite_coordinate_mask
].copy()

invalid_coordinate_count = len(
    invalid_coordinates_df
)


# ------------------------------------------------------------
# 5. Проверка occupancy и B-factor
# ------------------------------------------------------------

missing_occupancy_count = int(
    df_atoms["occupancy"].isna().sum()
)

missing_bfactor_count = int(
    df_atoms["bfactor"].isna().sum()
)

occupancy_values = pd.to_numeric(
    df_atoms["occupancy"],
    errors="coerce"
)

bfactor_values = pd.to_numeric(
    df_atoms["bfactor"],
    errors="coerce"
)

invalid_occupancy_mask = (
    occupancy_values.notna()
    & (
        ~np.isfinite(occupancy_values)
        | (occupancy_values < 0)
        | (occupancy_values > 1)
    )
)

invalid_bfactor_mask = (
    bfactor_values.notna()
    & (
        ~np.isfinite(bfactor_values)
        | (bfactor_values < 0)
    )
)

invalid_occupancy_df = df_atoms.loc[
    invalid_occupancy_mask
].copy()

invalid_bfactor_df = df_atoms.loc[
    invalid_bfactor_mask
].copy()


# ------------------------------------------------------------
# 6. Статистика по химическим элементам
# ------------------------------------------------------------

element_statistics = (
    df_atoms
    .groupby("element", dropna=False)
    .agg(
        atom_count=("atom_name", "size")
    )
    .reset_index()
    .sort_values(
        "atom_count",
        ascending=False
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 7. Статистика по цепям
# ------------------------------------------------------------

chain_atom_statistics = (
    df_atoms
    .groupby("chain_id", sort=False, dropna=False)
    .agg(
        atom_count=("atom_name", "size"),
        residue_count=("residue_key", "nunique"),
        backbone_atom_count=("is_backbone", "sum"),
        hydrogen_atom_count=("is_hydrogen", "sum"),
        altloc_atom_count=(
            "altloc",
            lambda x: int((x != "").sum())
        ),
        mean_occupancy=("occupancy", "mean"),
        mean_bfactor=("bfactor", "mean")
    )
    .reset_index()
)


# ------------------------------------------------------------
# 8. Общая статистика
# ------------------------------------------------------------

total_protein_atoms = len(df_atoms)

total_protein_residues = int(
    df_atoms["residue_key"].nunique()
)

unique_elements = sorted(
    df_atoms["element"].dropna().unique().tolist()
)

backbone_atom_count = int(
    df_atoms["is_backbone"].sum()
)

hydrogen_atom_count = int(
    df_atoms["is_hydrogen"].sum()
)

altloc_atom_count = int(
    (df_atoms["altloc"] != "").sum()
)

mean_occupancy = df_atoms["occupancy"].mean()

mean_bfactor = df_atoms["bfactor"].mean()


# ------------------------------------------------------------
# 9. Сверка с CELL 4
# ------------------------------------------------------------

cell4_atom_check = None
cell4_expected_atoms = None

if "protein_atom_count" in globals():

    cell4_expected_atoms = int(protein_atom_count)

    cell4_atom_check = (
        total_protein_atoms == cell4_expected_atoms
    )


# ------------------------------------------------------------
# 10. Сверка с CELL 6: общее количество атомов и остатков
# ------------------------------------------------------------

cell6_expected_atoms = int(
    df_residues["atom_count"].sum()
)

cell6_expected_residues = len(df_residues)

cell6_atom_check = (
    total_protein_atoms == cell6_expected_atoms
)

cell6_residue_check = (
    total_protein_residues == cell6_expected_residues
)


# ------------------------------------------------------------
# 11. Сверка количества атомов для каждого остатка
# ------------------------------------------------------------

residue_key_columns = [
    "chain_id",
    "hetero_flag",
    "residue_id",
    "insertion_code",
    "residue_name"
]

# Фактическое количество атомов по CELL 7.

atoms_per_residue = (
    df_atoms
    .groupby(
        residue_key_columns,
        dropna=False
    )
    .agg(
        cell7_atom_count=("atom_name", "size")
    )
    .reset_index()
)

# Количество атомов по CELL 6.

residues_cell6 = df_residues[
    residue_key_columns + ["atom_count"]
].copy()

residues_cell6 = (
    residues_cell6
    .groupby(
        residue_key_columns,
        dropna=False
    )
    .agg(
        cell6_atom_count=("atom_count", "sum")
    )
    .reset_index()
)

# Сопоставление двух таблиц.

residue_atom_comparison = residues_cell6.merge(
    atoms_per_residue,
    on=residue_key_columns,
    how="outer",
    indicator=True
)

residue_atom_comparison["cell6_atom_count"] = (
    residue_atom_comparison["cell6_atom_count"]
    .fillna(0)
    .astype(int)
)

residue_atom_comparison["cell7_atom_count"] = (
    residue_atom_comparison["cell7_atom_count"]
    .fillna(0)
    .astype(int)
)

residue_atom_comparison["atom_count_match"] = (
    residue_atom_comparison["cell6_atom_count"]
    == residue_atom_comparison["cell7_atom_count"]
)

residue_atom_comparison["residue_present_in_both"] = (
    residue_atom_comparison["_merge"] == "both"
)

residue_atom_mismatch_df = residue_atom_comparison.loc[
    ~residue_atom_comparison["atom_count_match"]
    | ~residue_atom_comparison["residue_present_in_both"]
].copy()

residue_atom_check = residue_atom_mismatch_df.empty


# ------------------------------------------------------------
# 12. Формирование итогового отчёта
# ------------------------------------------------------------

atom_analysis_summary = {

    "pdb_id": PDB_ID,

    "protein_atom_count": int(total_protein_atoms),

    "protein_residue_count": int(total_protein_residues),

    "element_count": int(len(unique_elements)),

    "elements": unique_elements,

    "backbone_atom_count": int(backbone_atom_count),

    "hydrogen_atom_count": int(hydrogen_atom_count),

    "altloc_atom_count": int(altloc_atom_count),

    "invalid_coordinate_count": int(invalid_coordinate_count),

    "missing_occupancy_count": int(missing_occupancy_count),

    "missing_bfactor_count": int(missing_bfactor_count),

    "invalid_occupancy_count": int(len(invalid_occupancy_df)),

    "invalid_bfactor_count": int(len(invalid_bfactor_df)),

    "mean_occupancy": (
        float(mean_occupancy)
        if pd.notna(mean_occupancy)
        else None
    ),

    "mean_bfactor": (
        float(mean_bfactor)
        if pd.notna(mean_bfactor)
        else None
    ),

    "cell4_atom_check": cell4_atom_check,

    "cell4_expected_atoms": cell4_expected_atoms,

    "cell6_atom_check": bool(cell6_atom_check),

    "cell6_expected_atoms": int(cell6_expected_atoms),

    "cell6_residue_check": bool(cell6_residue_check),

    "cell6_expected_residues": int(cell6_expected_residues),

    "residue_atom_check": bool(residue_atom_check),

    "residue_atom_mismatch_count": int(
        len(residue_atom_mismatch_df)
    )

}


# ------------------------------------------------------------
# 13. Сохранение результатов
# ------------------------------------------------------------

ATOM_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_protein_atoms.csv"
)

ELEMENT_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_element_statistics.csv"
)

CHAIN_ATOM_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_chain_atom_statistics.csv"
)

INVALID_COORDINATES_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_invalid_coordinates.csv"
)

INVALID_OCCUPANCY_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_invalid_occupancy.csv"
)

INVALID_BFACTOR_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_invalid_bfactor.csv"
)

RESIDUE_COMPARISON_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_residue_atom_comparison.csv"
)

RESIDUE_MISMATCH_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_residue_atom_mismatches.csv"
)

ATOM_SUMMARY_JSON_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_atom_analysis_summary.json"
)


df_atoms.to_csv(
    ATOM_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

element_statistics.to_csv(
    ELEMENT_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

chain_atom_statistics.to_csv(
    CHAIN_ATOM_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

invalid_coordinates_df.to_csv(
    INVALID_COORDINATES_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

invalid_occupancy_df.to_csv(
    INVALID_OCCUPANCY_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

invalid_bfactor_df.to_csv(
    INVALID_BFACTOR_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

residue_atom_comparison.to_csv(
    RESIDUE_COMPARISON_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

residue_atom_mismatch_df.to_csv(
    RESIDUE_MISMATCH_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

with open(
    ATOM_SUMMARY_JSON_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        atom_analysis_summary,
        f,
        ensure_ascii=False,
        indent=4
    )


# ------------------------------------------------------------
# 14. Вывод таблиц
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ПРИМЕР ТАБЛИЦЫ АТОМОВ")
print("-" * 70)

display(df_atoms.head(10))

print("\n" + "-" * 70)
print("РАСПРЕДЕЛЕНИЕ АТОМОВ ПО ЭЛЕМЕНТАМ")
print("-" * 70)

display(element_statistics)

print("\n" + "-" * 70)
print("СТАТИСТИКА ПО ЦЕПЯМ")
print("-" * 70)

display(chain_atom_statistics)

print("\n" + "-" * 70)
print("СВЕРКА КОЛИЧЕСТВА АТОМОВ ПО ОСТАТКАМ")
print("-" * 70)

display(residue_atom_comparison.head(10))


# ------------------------------------------------------------
# 15. Техническая статистика
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ТЕХНИЧЕСКАЯ СТАТИСТИКА")
print("-" * 70)

print(f"Всего белковых атомов: {total_protein_atoms}")

print(f"Белковых остатков: {total_protein_residues}")

print(f"Количество химических элементов: {len(unique_elements)}")

print(f"Элементы: {unique_elements}")

print(f"Атомов основного остова: {backbone_atom_count}")

print(f"Атомов водорода и дейтерия: {hydrogen_atom_count}")

print(f"Атомов с altloc: {altloc_atom_count}")

print(
    "Среднее occupancy:",
    f"{mean_occupancy:.4f}" if pd.notna(mean_occupancy) else "N/A"
)

print(
    "Средний B-factor:",
    f"{mean_bfactor:.4f}" if pd.notna(mean_bfactor) else "N/A"
)

print(f"Атомов с отсутствующим occupancy: {missing_occupancy_count}")

print(f"Атомов с отсутствующим B-factor: {missing_bfactor_count}")

print(f"Атомов с некорректными координатами: {invalid_coordinate_count}")

print(f"Атомов с некорректным occupancy: {len(invalid_occupancy_df)}")

print(f"Атомов с некорректным B-factor: {len(invalid_bfactor_df)}")

print(
    "Остатков с расхождением количества атомов:",
    len(residue_atom_mismatch_df)
)


# ------------------------------------------------------------
# 16. Контроль согласованности
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("СВЕРКА С ПРЕДЫДУЩИМИ ЯЧЕЙКАМИ")
print("-" * 70)

print("Количество атомов CELL 4:", cell4_expected_atoms)

print("Количество атомов CELL 6:", cell6_expected_atoms)

print("Количество атомов CELL 7:", total_protein_atoms)

if cell4_atom_check is not None:

    print(
        "Сверка с CELL 4:",
        "OK" if cell4_atom_check else "ERROR"
    )

print(
    "Сверка общего количества атомов с CELL 6:",
    "OK" if cell6_atom_check else "ERROR"
)

print(
    "Сверка количества остатков с CELL 6:",
    "OK" if cell6_residue_check else "ERROR"
)

print(
    "Сверка количества атомов по каждому остатку:",
    "OK" if residue_atom_check else "ERROR"
)


# ------------------------------------------------------------
# 17. Проверка сохранённых файлов
# ------------------------------------------------------------

output_paths = [

    ATOM_CSV_PATH,
    ELEMENT_CSV_PATH,
    CHAIN_ATOM_CSV_PATH,
    INVALID_COORDINATES_CSV_PATH,
    INVALID_OCCUPANCY_CSV_PATH,
    INVALID_BFACTOR_CSV_PATH,
    RESIDUE_COMPARISON_CSV_PATH,
    RESIDUE_MISMATCH_CSV_PATH,
    ATOM_SUMMARY_JSON_PATH

]

print("\n" + "-" * 70)
print("СОХРАНЁННЫЕ ФАЙЛЫ")
print("-" * 70)

for path in output_paths:

    status = (
        "OK"
        if path.is_file() and path.stat().st_size > 0
        else "ERROR"
    )

    print(f"{path.name}: {status}")


# ------------------------------------------------------------
# 18. Финальная проверка
# ------------------------------------------------------------

# Совместимость со старым именем переменной.

atom_csv_path = ATOM_CSV_PATH

# Сохраняем дополнительную таблицу остатков,
# чтобы не нарушать совместимость с исходной CELL 7.

residue_csv_path = (
    PDB_RESULTS_DIR / f"{PDB_ID}_residues.csv"
)

df_residues.to_csv(
    residue_csv_path,
    index=False,
    encoding="utf-8-sig"
)

print(f"\nТаблица остатков: {residue_csv_path}")

# Проверяем, что все результаты действительно сохранены.

missing_output_files = [
    path
    for path in output_paths
    if not path.is_file() or path.stat().st_size == 0
]

if missing_output_files:

    raise IOError(
        "Не удалось сохранить некоторые результаты: "
        + ", ".join(path.name for path in missing_output_files)
    )

# Проверки согласованности.

if invalid_coordinate_count > 0:

    raise ValueError(
        f"Обнаружено атомов с некорректными координатами: "
        f"{invalid_coordinate_count}"
    )

if cell6_atom_check is False:

    raise ValueError(
        "Общее количество атомов CELL 7 не совпадает с CELL 6."
    )

if cell6_residue_check is False:

    raise ValueError(
        "Количество остатков CELL 7 не совпадает с CELL 6."
    )

if not residue_atom_check:

    raise ValueError(
        "Обнаружены расхождения в количестве атомов "
        "для отдельных остатков между CELL 6 и CELL 7."
    )

if cell4_atom_check is False:

    raise ValueError(
        "Количество атомов CELL 7 не совпадает с CELL 4."
    )

print("\nСтатус: OK")
print("Координаты белковых атомов подготовлены.")

print("=" * 70)
print("CELL 7 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# CELL 8. АНАЛИЗ МАЛЫХ МОЛЕКУЛ И ДРУГИХ КОМПОНЕНТОВ
# ============================================================
# Назначение:
# 1. Выделение небелковых компонентов и молекул воды.
# 2. Предварительная классификация компонентов.
# 3. Извлечение координат атомов.
# 4. Сохранение химических и структурных характеристик.
# 5. Сверка результатов с CELL 4.
# 6. Подготовка данных для последующего анализа компонентов,
#    лигандов и взаимодействий с белковыми карманами.
#
# ВАЖНО:
# Классификация компонентов предварительная.
# Окончательная идентификация лигандов и коферментов
# будет выполняться на следующих этапах.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.PDB.Polypeptide import is_aa
from IPython.display import display


print("=" * 70)
print("CELL 8. АНАЛИЗ МАЛЫХ МОЛЕКУЛ И ДРУГИХ КОМПОНЕНТОВ")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка входных данных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "chains",
    "PDB_RESULTS_DIR",
    "structure_statistics"
]

for variable in required_variables:

    if variable not in globals():

        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )

PDB_ID = str(PDB_ID).strip().upper()

PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)
PDB_RESULTS_DIR.mkdir(parents=True, exist_ok=True)

if not isinstance(structure_statistics, dict):

    raise TypeError(
        "structure_statistics должен быть словарём."
    )

required_statistics = [
    "nonprotein_residues",
    "water_residues",
    "nonprotein_atoms",
    "water_atoms"
]

missing_statistics = [
    key
    for key in required_statistics
    if key not in structure_statistics
]

if missing_statistics:

    raise KeyError(
        "В structure_statistics отсутствуют поля: "
        f"{missing_statistics}"
    )

print(f"PDB ID: {PDB_ID}")
print(f"Количество анализируемых цепей: {len(chains)}")
print(f"Каталог результатов: {PDB_RESULTS_DIR}")


# ------------------------------------------------------------
# 2. Настройки классификации компонентов
# ------------------------------------------------------------

WATER_CODES = {
    "HOH",
    "WAT",
    "DOD"
}

METAL_ELEMENTS = {
    "FE", "ZN", "MG", "MN", "CU",
    "CO", "NI", "CA", "NA", "K",
    "CD", "HG"
}


def classify_component(residue_name, elements, atom_count):
    """
    Предварительная классификация небелкового компонента.

    Классификация выполняется по элементному составу
    и количеству атомов.

    Она не заменяет идентификацию химического соединения
    по справочникам компонентов PDB.
    """

    elements = set(elements)

    if not elements:
        return "Unknown component"

    # Органический компонент, содержащий металл.
    if (
        elements.intersection(METAL_ELEMENTS)
        and "C" in elements
    ):
        return "Metal-containing organic component"

    # Компонент, состоящий только из металлов.
    elif elements.issubset(METAL_ELEMENTS):

        return "Metal-containing component"

    # Одноатомный компонент.
    elif atom_count == 1:

        return "Single atom / ion"

    # Органический небелковый компонент.
    elif "C" in elements:

        return "Organic non-protein component"

    # Остальные неорганические компоненты.
    else:

        return "Other non-protein component"


# ------------------------------------------------------------
# 3. Подготовка списков записей
# ------------------------------------------------------------

hetero_records = []
water_records = []

hetero_atom_records = []
water_atom_records = []


# ------------------------------------------------------------
# 4. Обход цепей и остатков
# ------------------------------------------------------------

for chain in chains:

    for residue in chain.get_residues():

        residue_name = residue.resname.strip().upper()

        hetero_flag = residue.id[0]

        residue_number = int(residue.id[1])

        insertion_code = residue.id[2].strip()

        # ----------------------------------------------------
        # 4.1. Пропускаем белковые остатки
        # ----------------------------------------------------
        # Используем ту же проверку, что в CELL 4 и CELL 6.
        # Это необходимо для согласованности подсчётов.

        if is_aa(residue, standard=False):
            continue

        atoms = list(residue.get_atoms())

        if not atoms:
            continue

        # ----------------------------------------------------
        # 4.2. Определение элементного состава
        # ----------------------------------------------------

        elements = sorted({

            atom.element.strip().upper()

            for atom in atoms

            if atom.element and atom.element.strip()

        })

        atom_count = len(atoms)

        residue_key = (
            f"{chain.id}:{hetero_flag}:"
            f"{residue_number}:{insertion_code}"
        )

        # Идентификатор основан на полном ключе остатка.
        # Это позволяет различать одинаковые компоненты
        # в разных цепях и позициях структуры.

        component_id = f"{PDB_ID}|{residue_key}"

        residue_info = {

            "pdb_id": PDB_ID,

            "component_id": component_id,

            "residue_key": residue_key,

            "chain_id": chain.id,

            "residue_name": residue_name,

            "residue_id": residue_number,

            "insertion_code": insertion_code,

            "hetero_flag": hetero_flag,

            "atom_count": atom_count,

            "elements": ", ".join(elements),

            "element_count": len(elements)

        }

        # ----------------------------------------------------
        # 4.3. Обработка воды
        # ----------------------------------------------------

        if residue_name in WATER_CODES:

            residue_info["component_type"] = "Water"

            water_records.append(residue_info)

            for atom in atoms:

                coord = atom.get_coord()

                water_atom_records.append({

                    **residue_info,

                    "atom_serial": atom.get_serial_number(),

                    "atom_name": atom.get_name().strip(),

                    "element": (
                        atom.element.strip().upper()
                        if atom.element
                        else "UNKNOWN"
                    ),

                    "x": float(coord[0]),

                    "y": float(coord[1]),

                    "z": float(coord[2]),

                    "occupancy": atom.get_occupancy(),

                    "bfactor": atom.get_bfactor(),

                    "altloc": atom.get_altloc().strip()

                })

            continue

        # ----------------------------------------------------
        # 4.4. Обработка остальных небелковых компонентов
        # ----------------------------------------------------

        component_type = classify_component(
            residue_name,
            elements,
            atom_count
        )

        residue_info["component_type"] = component_type

        hetero_records.append(residue_info)

        # ----------------------------------------------------
        # 4.5. Извлечение координат атомов компонента
        # ----------------------------------------------------

        for atom in atoms:

            coord = atom.get_coord()

            hetero_atom_records.append({

                **residue_info,

                "atom_serial": atom.get_serial_number(),

                "atom_name": atom.get_name().strip(),

                "element": (
                    atom.element.strip().upper()
                    if atom.element
                    else "UNKNOWN"
                ),

                "x": float(coord[0]),

                "y": float(coord[1]),

                "z": float(coord[2]),

                "occupancy": atom.get_occupancy(),

                "bfactor": atom.get_bfactor(),

                "altloc": atom.get_altloc().strip()

            })


# ------------------------------------------------------------
# 5. Формирование таблиц
# ------------------------------------------------------------

HETERO_COLUMNS = [

    "pdb_id",
    "component_id",
    "residue_key",
    "chain_id",
    "residue_name",
    "residue_id",
    "insertion_code",
    "hetero_flag",
    "atom_count",
    "elements",
    "element_count",
    "component_type"

]

ATOM_COLUMNS = HETERO_COLUMNS + [

    "atom_serial",
    "atom_name",
    "element",
    "x",
    "y",
    "z",
    "occupancy",
    "bfactor",
    "altloc"

]

df_hetero = pd.DataFrame(
    hetero_records,
    columns=HETERO_COLUMNS
)

df_water = pd.DataFrame(
    water_records,
    columns=HETERO_COLUMNS
)

df_hetero_atoms = pd.DataFrame(
    hetero_atom_records,
    columns=ATOM_COLUMNS
)

df_water_atoms = pd.DataFrame(
    water_atom_records,
    columns=ATOM_COLUMNS
)


# ------------------------------------------------------------
# 6. Проверка координат
# ------------------------------------------------------------

coordinate_columns = ["x", "y", "z"]


def get_invalid_coordinate_mask(df):
    """
    Возвращает маску атомов с некорректными координатами.
    """

    if df.empty:
        return np.zeros(0, dtype=bool)

    coordinates = df[
        coordinate_columns
    ].to_numpy(dtype=float)

    return ~np.isfinite(coordinates).all(axis=1)


invalid_hetero_coordinates = get_invalid_coordinate_mask(
    df_hetero_atoms
)

invalid_water_coordinates = get_invalid_coordinate_mask(
    df_water_atoms
)

invalid_hetero_coordinates_df = df_hetero_atoms.loc[
    invalid_hetero_coordinates
].copy()

invalid_water_coordinates_df = df_water_atoms.loc[
    invalid_water_coordinates
].copy()

invalid_hetero_coordinate_count = len(
    invalid_hetero_coordinates_df
)

invalid_water_coordinate_count = len(
    invalid_water_coordinates_df
)


# ------------------------------------------------------------
# 7. Сводная статистика
# ------------------------------------------------------------

hetero_residue_count = len(df_hetero)

water_residue_count = len(df_water)

hetero_atom_count = len(df_hetero_atoms)

water_atom_count = len(df_water_atoms)

unique_components = (
    sorted(df_hetero["residue_name"].unique().tolist())
    if not df_hetero.empty
    else []
)


# ------------------------------------------------------------
# 8. Статистика по типам компонентов
# ------------------------------------------------------------

if not df_hetero.empty:

    component_occurrence_statistics = (

        df_hetero
        .groupby(
            ["residue_name", "component_type"],
            dropna=False
        )
        .agg(
            occurrence_count=("component_id", "count"),
            total_atom_count=("atom_count", "sum"),
            chains=(
                "chain_id",
                lambda x: ", ".join(
                    sorted(set(str(v) for v in x))
                )
            )
        )
        .reset_index()
        .sort_values(
            ["occurrence_count", "residue_name"],
            ascending=[False, True]
        )
        .reset_index(drop=True)

    )

else:

    component_occurrence_statistics = pd.DataFrame(
        columns=[
            "residue_name",
            "component_type",
            "occurrence_count",
            "total_atom_count",
            "chains"
        ]
    )


# ------------------------------------------------------------
# 9. Сверка с CELL 4
# ------------------------------------------------------------

cell4_nonprotein_residues = int(
    structure_statistics["nonprotein_residues"]
)

cell4_water_residues = int(
    structure_statistics["water_residues"]
)

cell4_nonprotein_atoms = int(
    structure_statistics["nonprotein_atoms"]
)

cell4_water_atoms = int(
    structure_statistics["water_atoms"]
)

nonprotein_residue_check = (
    hetero_residue_count == cell4_nonprotein_residues
)

water_residue_check = (
    water_residue_count == cell4_water_residues
)

nonprotein_atom_check = (
    hetero_atom_count == cell4_nonprotein_atoms
)

water_atom_check = (
    water_atom_count == cell4_water_atoms
)

cell4_component_check = all([

    nonprotein_residue_check,

    water_residue_check,

    nonprotein_atom_check,

    water_atom_check

])


# ------------------------------------------------------------
# 10. Итоговая информация
# ------------------------------------------------------------

component_analysis_summary = {

    "pdb_id": PDB_ID,

    "nonprotein_residue_count": int(hetero_residue_count),

    "nonprotein_atom_count": int(hetero_atom_count),

    "water_residue_count": int(water_residue_count),

    "water_atom_count": int(water_atom_count),

    "component_names": unique_components,

    "component_occurrence_statistics": (
        component_occurrence_statistics.to_dict("records")
    ),

    "cell4_nonprotein_residue_check": bool(
        nonprotein_residue_check
    ),

    "cell4_water_residue_check": bool(
        water_residue_check
    ),

    "cell4_nonprotein_atom_check": bool(
        nonprotein_atom_check
    ),

    "cell4_water_atom_check": bool(
        water_atom_check
    ),

    "cell4_consistency_check": bool(
        cell4_component_check
    ),

    "invalid_hetero_coordinate_count": int(
        invalid_hetero_coordinate_count
    ),

    "invalid_water_coordinate_count": int(
        invalid_water_coordinate_count
    )

}


# ------------------------------------------------------------
# 11. Сохранение результатов
# ------------------------------------------------------------

HETERO_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_hetero_components.csv"
)

WATER_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_water_components.csv"
)

HETERO_ATOMS_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_hetero_atoms.csv"
)

WATER_ATOMS_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_water_atoms.csv"
)

COMPONENT_SUMMARY_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_component_summary.csv"
)

INVALID_HETERO_COORDINATES_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_invalid_hetero_coordinates.csv"
)

INVALID_WATER_COORDINATES_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_invalid_water_coordinates.csv"
)

COMPONENT_SUMMARY_JSON_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_component_analysis_summary.json"
)


df_hetero.to_csv(
    HETERO_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

df_water.to_csv(
    WATER_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

df_hetero_atoms.to_csv(
    HETERO_ATOMS_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

df_water_atoms.to_csv(
    WATER_ATOMS_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

component_occurrence_statistics.to_csv(
    COMPONENT_SUMMARY_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

invalid_hetero_coordinates_df.to_csv(
    INVALID_HETERO_COORDINATES_PATH,
    index=False,
    encoding="utf-8-sig"
)

invalid_water_coordinates_df.to_csv(
    INVALID_WATER_COORDINATES_PATH,
    index=False,
    encoding="utf-8-sig"
)

with open(
    COMPONENT_SUMMARY_JSON_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        component_analysis_summary,
        f,
        ensure_ascii=False,
        indent=4
    )


# ------------------------------------------------------------
# 12. Вывод общей статистики
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ОБЩАЯ ИНФОРМАЦИЯ")
print("-" * 70)

print(f"Небелковых остатков: {hetero_residue_count}")

print(f"Атомов небелковых компонентов: {hetero_atom_count}")

print(f"Остатков воды: {water_residue_count}")

print(f"Атомов воды: {water_atom_count}")

print(f"Уникальных типов компонентов: {len(unique_components)}")

print(f"Компоненты: {unique_components}")


# ------------------------------------------------------------
# 13. Распределение компонентов
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("КЛАССИФИКАЦИЯ НЕБЕЛКОВЫХ КОМПОНЕНТОВ")
print("-" * 70)

if not df_hetero.empty:

    display(component_occurrence_statistics)

    display(
        df_hetero.sort_values(
            ["residue_name", "chain_id", "residue_id"]
        )
    )

else:

    print("Небелковые компоненты не обнаружены.")


# ------------------------------------------------------------
# 14. Информация о воде
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("АНАЛИЗ МОЛЕКУЛ ВОДЫ")
print("-" * 70)

if not df_water.empty:

    display(df_water.head(20))

else:

    print("Молекулы воды не обнаружены.")


# ------------------------------------------------------------
# 15. Атомы небелковых компонентов
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("КООРДИНАТЫ НЕБЕЛКОВЫХ КОМПОНЕНТОВ")
print("-" * 70)

if not df_hetero_atoms.empty:

    display(df_hetero_atoms.head(10))

    print(
        "Всего атомов в таблице:",
        len(df_hetero_atoms)
    )

else:

    print("Таблица атомов небелковых компонентов пуста.")


# ------------------------------------------------------------
# 16. Контроль соответствия CELL 4
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("СВЕРКА С CELL 4")
print("-" * 70)

print(
    f"Небелковые остатки: "
    f"CELL 4 = {cell4_nonprotein_residues}, "
    f"CELL 8 = {hetero_residue_count}"
)

print(
    f"Молекулы воды: "
    f"CELL 4 = {cell4_water_residues}, "
    f"CELL 8 = {water_residue_count}"
)

print(
    f"Атомы небелковых компонентов: "
    f"CELL 4 = {cell4_nonprotein_atoms}, "
    f"CELL 8 = {hetero_atom_count}"
)

print(
    f"Атомы воды: "
    f"CELL 4 = {cell4_water_atoms}, "
    f"CELL 8 = {water_atom_count}"
)

print(
    "\nОбщая проверка:",
    "OK" if cell4_component_check else "ERROR"
)


# ------------------------------------------------------------
# 17. Проверка сохранённых файлов
# ------------------------------------------------------------

output_paths = [

    HETERO_CSV_PATH,
    WATER_CSV_PATH,
    HETERO_ATOMS_CSV_PATH,
    WATER_ATOMS_CSV_PATH,
    COMPONENT_SUMMARY_CSV_PATH,
    INVALID_HETERO_COORDINATES_PATH,
    INVALID_WATER_COORDINATES_PATH,
    COMPONENT_SUMMARY_JSON_PATH

]

print("\n" + "-" * 70)
print("СОХРАНЁННЫЕ ФАЙЛЫ")
print("-" * 70)

missing_output_files = []

for path in output_paths:

    file_ok = (
        path.is_file()
        and path.stat().st_size > 0
    )

    print(
        f"{path.name}: "
        f"{'OK' if file_ok else 'ERROR'}"
    )

    if not file_ok:
        missing_output_files.append(path)


# ------------------------------------------------------------
# 18. Финальные проверки
# ------------------------------------------------------------

if missing_output_files:

    raise IOError(
        "Не удалось сохранить файлы: "
        + ", ".join(
            path.name for path in missing_output_files
        )
    )

if (
    invalid_hetero_coordinate_count > 0
    or invalid_water_coordinate_count > 0
):

    raise ValueError(
        "Обнаружены атомы с некорректными координатами. "
        "Подробности сохранены в диагностических CSV."
    )

if not cell4_component_check:

    raise ValueError(
        "Количество небелковых компонентов или воды "
        "не совпадает с результатами CELL 4. "
        "Проверьте классификацию остатков и подсчёт атомов."
    )

print("\nСтатус: OK")
print("Небелковые компоненты и вода выделены.")
print("Координаты компонентов подготовлены.")
print("Результаты готовы для последующего анализа.")

print("=" * 70)
print("CELL 8 COMPLETED")
print("=" * 70)
```

Что изменилось по сравнению со старой версией
Возможность	Старая версия	Новая версия
Поиск небелковых компонентов	Да	Да
Поиск воды	Да	Да
Определение цепи и номера остатка	Да	Да
Количество атомов	Да	Да
Элементный состав	Нет	Да
Классификация металлов и ионов	Нет	Да
Выделение органических компонентов	Нет	Да
Распознавание нестандартных аминокислот	Нет	Да
Группировка по типам компонентов	Нет	Да

Важное уточнение: классификация здесь основана на элементном составе и типе остатка. Она не определяет достоверно биологическую функцию молекулы. Например, органический небелковый компонент может быть природным кофактором, субстратом или лекарственным лигандом.

```python

# ============================================================
# CELL 9. АНАЛИЗ ГЕОМЕТРИИ БЕЛКОВОЙ СТРУКТУРЫ
# ============================================================
# Назначение:
# Расчёт геометрических характеристик белковой структуры
# на основе координат белковых атомов из CELL 7.
#
# Рассчитываются:
# 1. Геометрический центр.
# 2. Минимальные и максимальные координаты.
# 3. Размеры структуры по осям X, Y, Z.
# 4. Объём ограничивающего параллелепипеда.
# 5. Расстояния атомов от геометрического центра.
# 6. Геометрический радиус инерции.
# 7. Отношение максимального размера к минимальному.
#
# Вода и небелковые компоненты не учитываются.
# Массы атомов не учитываются.
#
# ВАЖНО:
# Объём ограничивающего параллелепипеда не является
# физическим объёмом белковой молекулы.
# ============================================================

import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from IPython.display import display


print("=" * 70)
print("CELL 9. АНАЛИЗ ГЕОМЕТРИИ БЕЛКОВОЙ СТРУКТУРЫ")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка входных данных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "df_atoms",
    "PDB_RESULTS_DIR"
]

for variable in required_variables:

    if variable not in globals():
        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()

PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)
PDB_RESULTS_DIR.mkdir(parents=True, exist_ok=True)


if not isinstance(df_atoms, pd.DataFrame):

    raise TypeError(
        "df_atoms должен быть объектом pandas.DataFrame."
    )


if df_atoms.empty:

    raise ValueError(
        "Таблица df_atoms пуста. "
        "Невозможно рассчитать геометрию структуры."
    )


required_columns = [
    "pdb_id",
    "chain_id",
    "residue_key",
    "x",
    "y",
    "z"
]

missing_columns = [
    column
    for column in required_columns
    if column not in df_atoms.columns
]

if missing_columns:

    raise ValueError(
        "В df_atoms отсутствуют обязательные столбцы: "
        f"{missing_columns}"
    )


# Проверяем, что таблица относится к текущей структуре.

source_pdb_ids = (
    df_atoms["pdb_id"]
    .dropna()
    .astype(str)
    .str.strip()
    .str.upper()
    .unique()
)

if len(source_pdb_ids) != 1 or source_pdb_ids[0] != PDB_ID:

    raise ValueError(
        "Несоответствие PDB_ID: "
        f"текущая структура = {PDB_ID}, "
        f"структуры в df_atoms = {source_pdb_ids.tolist()}"
    )


print(f"PDB ID: {PDB_ID}")
print("Источник координат: df_atoms (CELL 7)")
print("Тип анализируемых атомов: только белковые атомы")
print("Единицы измерения: Å")
print(f"Каталог результатов: {PDB_RESULTS_DIR}")


# ------------------------------------------------------------
# 2. Проверка координат
# ------------------------------------------------------------

coordinate_columns = ["x", "y", "z"]

coordinates = df_atoms[
    coordinate_columns
].to_numpy(dtype=np.float64)


finite_coordinate_mask = np.isfinite(
    coordinates
).all(axis=1)


invalid_coordinate_count = int(
    (~finite_coordinate_mask).sum()
)


# Некорректные координаты не удаляем автоматически.

if invalid_coordinate_count > 0:

    raise ValueError(
        f"Обнаружено {invalid_coordinate_count} атомов "
        "с некорректными координатами. "
        "Проверьте CELL 7."
    )


atom_count = len(coordinates)

if atom_count < 2:

    raise ValueError(
        "Недостаточно атомов для геометрического анализа."
    )


# ------------------------------------------------------------
# 3. Геометрический центр
# ------------------------------------------------------------
# Центр рассчитывается как среднее арифметическое
# координат всех белковых атомов.
#
# Каждый атом имеет одинаковый вес.

center = coordinates.mean(axis=0)

center_x = float(center[0])
center_y = float(center[1])
center_z = float(center[2])


# ------------------------------------------------------------
# 4. Границы структуры
# ------------------------------------------------------------

coord_min = coordinates.min(axis=0)

coord_max = coordinates.max(axis=0)

dimensions = coord_max - coord_min


# Объём ограничивающего параллелепипеда.

bounding_box_volume = float(
    np.prod(dimensions)
)


# ------------------------------------------------------------
# 5. Расстояния от геометрического центра
# ------------------------------------------------------------

distances_from_center = np.linalg.norm(
    coordinates - center,
    axis=1
)


max_distance = float(
    distances_from_center.max()
)

mean_distance = float(
    distances_from_center.mean()
)

median_distance = float(
    np.median(distances_from_center)
)


# ------------------------------------------------------------
# 6. Геометрический радиус инерции
# ------------------------------------------------------------
# Формула:
#
# Rg = sqrt(mean(r_i^2))
#
# где r_i — расстояние каждого атома
# от геометрического центра.
#
# Массы атомов не учитываются.

radius_of_gyration = float(
    np.sqrt(
        np.mean(distances_from_center ** 2)
    )
)


# ------------------------------------------------------------
# 7. Дополнительные характеристики
# ------------------------------------------------------------

residue_count = int(
    df_atoms["residue_key"].nunique()
)

chain_count = int(
    df_atoms["chain_id"].nunique()
)


min_dimension = float(
    dimensions.min()
)

max_dimension = float(
    dimensions.max()
)


# Если одна из размерностей равна нулю,
# отношение размеров математически не определяется.

aspect_ratio = (
    float(max_dimension / min_dimension)
    if min_dimension > 0
    else None
)


# ------------------------------------------------------------
# 8. Сводная таблица геометрии
# ------------------------------------------------------------

geometry_records = [

    ("Number of protein atoms", atom_count, "atoms"),

    ("Number of protein residues", residue_count, "residues"),

    ("Number of chains", chain_count, "chains"),

    ("Center X", center_x, "Å"),

    ("Center Y", center_y, "Å"),

    ("Center Z", center_z, "Å"),

    ("Minimum X", float(coord_min[0]), "Å"),

    ("Minimum Y", float(coord_min[1]), "Å"),

    ("Minimum Z", float(coord_min[2]), "Å"),

    ("Maximum X", float(coord_max[0]), "Å"),

    ("Maximum Y", float(coord_max[1]), "Å"),

    ("Maximum Z", float(coord_max[2]), "Å"),

    ("Size X", float(dimensions[0]), "Å"),

    ("Size Y", float(dimensions[1]), "Å"),

    ("Size Z", float(dimensions[2]), "Å"),

    ("Bounding box volume", bounding_box_volume, "Å³"),

    ("Maximum distance from center", max_distance, "Å"),

    ("Mean distance from center", mean_distance, "Å"),

    ("Median distance from center", median_distance, "Å"),

    ("Geometric radius of gyration", radius_of_gyration, "Å"),

    ("Bounding box aspect ratio", aspect_ratio, "ratio")

]


df_geometry = pd.DataFrame(
    geometry_records,
    columns=[
        "Parameter",
        "Value",
        "Unit"
    ]
)


# ------------------------------------------------------------
# 9. Сверка с CELL 7
# ------------------------------------------------------------

cell7_atom_check = None

cell7_atom_count = None


if "total_protein_atoms" in globals():

    cell7_atom_count = int(
        total_protein_atoms
    )

    cell7_atom_check = (
        atom_count == cell7_atom_count
    )

    print("\n" + "-" * 70)
    print("КОНТРОЛЬ СООТВЕТСТВИЯ CELL 7")
    print("-" * 70)

    print(
        f"Атомы CELL 7: {cell7_atom_count}"
    )

    print(
        f"Атомы CELL 9: {atom_count}"
    )

    print(
        "Проверка:",
        "OK" if cell7_atom_check else "ERROR"
    )

    if not cell7_atom_check:

        raise ValueError(
            "Количество атомов в геометрическом анализе "
            "не совпадает с CELL 7."
        )

else:

    print("\nСверка с CELL 7 не выполнена:")
    print("Переменная total_protein_atoms отсутствует.")


# ------------------------------------------------------------
# 10. Сводный словарь результатов
# ------------------------------------------------------------

geometry_summary = {

    "pdb_id": PDB_ID,

    "coordinate_source": "df_atoms",

    "atom_selection": "protein atoms only",

    "water_included": False,

    "hetero_components_included": False,

    "atom_count": atom_count,

    "residue_count": residue_count,

    "chain_count": chain_count,

    "geometric_center": {
        "x": center_x,
        "y": center_y,
        "z": center_z
    },

    "minimum_coordinates": {
        "x": float(coord_min[0]),
        "y": float(coord_min[1]),
        "z": float(coord_min[2])
    },

    "maximum_coordinates": {
        "x": float(coord_max[0]),
        "y": float(coord_max[1]),
        "z": float(coord_max[2])
    },

    "dimensions": {
        "x": float(dimensions[0]),
        "y": float(dimensions[1]),
        "z": float(dimensions[2])
    },

    "bounding_box_volume": bounding_box_volume,

    "maximum_distance_from_center": max_distance,

    "mean_distance_from_center": mean_distance,

    "median_distance_from_center": median_distance,

    "geometric_radius_of_gyration": radius_of_gyration,

    "aspect_ratio": aspect_ratio,

    "coordinate_unit": "Angstrom",

    "radius_of_gyration_method": "unweighted_atom_coordinates",

    "cell7_atom_count": cell7_atom_count,

    "cell7_atom_check": cell7_atom_check,

    "invalid_coordinate_count": invalid_coordinate_count

}


# ------------------------------------------------------------
# 11. Сохранение результатов
# ------------------------------------------------------------

GEOMETRY_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_geometry.csv"
)

GEOMETRY_JSON_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_geometry.json"
)


df_geometry.to_csv(
    GEOMETRY_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


with open(
    GEOMETRY_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        geometry_summary,
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


# ------------------------------------------------------------
# 12. Вывод результатов
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ГЕОМЕТРИЧЕСКИЕ ХАРАКТЕРИСТИКИ")
print("-" * 70)

print(f"Количество белковых атомов: {atom_count}")

print(f"Количество белковых остатков: {residue_count}")

print(f"Количество цепей: {chain_count}")


print("\nГеометрический центр (Å):")

print(f"X: {center_x:.4f}")
print(f"Y: {center_y:.4f}")
print(f"Z: {center_z:.4f}")


print("\nМинимальные координаты (Å):")

print(coord_min)


print("\nМаксимальные координаты (Å):")

print(coord_max)


print("\nРазмеры структуры (Å):")

print(dimensions)


print(
    f"\nОбъём ограничивающего параллелепипеда: "
    f"{bounding_box_volume:.4f} Å³"
)


print(
    f"Максимальное расстояние от центра: "
    f"{max_distance:.4f} Å"
)


print(
    f"Среднее расстояние от центра: "
    f"{mean_distance:.4f} Å"
)


print(
    f"Медианное расстояние от центра: "
    f"{median_distance:.4f} Å"
)


print(
    f"Геометрический радиус инерции: "
    f"{radius_of_gyration:.4f} Å"
)


print(
    f"Отношение максимального размера "
    f"к минимальному: {aspect_ratio}"
)


print("\nСводная таблица:")

display(df_geometry)


# ------------------------------------------------------------
# 13. 3D-визуализация распределения атомов
# ------------------------------------------------------------

GEOMETRY_PLOT_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_protein_geometry.png"
)


fig = plt.figure(figsize=(10, 8))

ax = fig.add_subplot(
    111,
    projection="3d"
)


ax.scatter(
    coordinates[:, 0],
    coordinates[:, 1],
    coordinates[:, 2],
    s=3,
    alpha=0.45,
    label="Protein atoms"
)


ax.scatter(
    center_x,
    center_y,
    center_z,
    color="red",
    marker="X",
    s=100,
    label="Geometric center"
)


ax.set_xlabel("X (Å)")
ax.set_ylabel("Y (Å)")
ax.set_zlabel("Z (Å)")


ax.set_title(
    f"Protein geometry: {PDB_ID}"
)


# Защита от нулевой размерности при отображении.

plot_dimensions = np.maximum(
    dimensions,
    1e-6
)

ax.set_box_aspect(
    plot_dimensions
)


ax.legend()

plt.tight_layout()


plt.savefig(
    GEOMETRY_PLOT_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close(fig)


# ------------------------------------------------------------
# 14. Проверка сохранённых файлов
# ------------------------------------------------------------

output_paths = [

    GEOMETRY_CSV_PATH,

    GEOMETRY_JSON_PATH,

    GEOMETRY_PLOT_PATH

]


print("\n" + "-" * 70)
print("ПРОВЕРКА СОХРАНЁННЫХ ФАЙЛОВ")
print("-" * 70)


output_checks = {}

for path in output_paths:

    file_exists = (
        path.is_file()
        and path.stat().st_size > 0
    )

    output_checks[path.name] = bool(
        file_exists
    )

    print(
        f"{path.name}: "
        f"{'OK' if file_exists else 'ERROR'}"
    )


if not all(output_checks.values()):

    raise RuntimeError(
        "Не все файлы геометрического анализа "
        "успешно сохранены."
    )


# ------------------------------------------------------------
# 15. Итоговый статус
# ------------------------------------------------------------

geometry_summary["output_checks"] = output_checks

geometry_summary["status"] = "OK"


# Обновляем JSON с итоговым статусом.

with open(
    GEOMETRY_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        geometry_summary,
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


print("\nМетод расчёта:")
print("Геометрия рассчитана по координатам белковых атомов.")
print("Массы атомов не учитывались.")
print("Вода и небелковые компоненты исключены.")
print("Объём ограничивающего параллелепипеда не является")
print("физическим объёмом белковой молекулы.")

print("\nСтатус: OK")
print("Геометрические характеристики рассчитаны и сохранены.")

print("=" * 70)
print("CELL 9 COMPLETED")
print("=" * 70)
```

Это облако координат атомов, а не полноценная молекулярная визуализация. Для изучения пространственной структуры, вторичной структуры и взаимодействий удобнее использовать специализированные визуализаторы, например PyMOL.

Что добавили
Показатель	Что означает
Геометрический центр	Среднее положение всех атомов
Bounding box	Габариты структуры по X, Y и Z
Объём bounding box	Объём параллелепипеда, охватывающего все атомы
Максимальное расстояние от центра	Расстояние до наиболее удалённого атома
Радиус инерции	Характеристика пространственной компактности структуры

Особенно полезен радиус инерции (Radius of gyration). Он позволяет количественно описать, насколько компактно расположены атомы белка. Это может быть полезным структурным признаком при дальнейшем машинном обучении.

```python

# ============================================================
# CELL 10. МЕТАДАННЫЕ И КАЧЕСТВО СТРУКТУРЫ
# ============================================================
# Назначение:
# Извлечение метаданных и сведений о качестве структуры
# непосредственно из исходного файла mmCIF.
#
# Извлекаются:
# 1. Название структуры.
# 2. Экспериментальный метод.
# 3. Разрешение и источник его значения.
# 4. Количество моделей, цепей, остатков и атомов.
# 5. Сведения о неразрешённых остатках.
# 6. Сведения о неразрешённых и zero-occupancy атомах.
# 7. Количество атомов с occupancy = 0
#    в фактически извлечённых координатных таблицах.
#
# ВАЖНО:
# CELL 4–8 анализируют первый модельный экземпляр структуры.
# Поэтому количество цепей, остатков и атомов в этих таблицах
# относится к обработанной модели, а не ко всем моделям mmCIF.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.PDB.MMCIF2Dict import MMCIF2Dict
from IPython.display import display


print("=" * 70)
print("CELL 10. МЕТАДАННЫЕ И КАЧЕСТВО СТРУКТУРЫ")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка входных данных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "CIF_PATH",
    "PDB_RESULTS_DIR",
    "models",
    "chains",
    "df_residues",
    "df_atoms",
    "df_hetero",
    "df_water"
]

for variable in required_variables:

    if variable not in globals():

        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()

CIF_PATH = Path(CIF_PATH)

PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


if not CIF_PATH.is_file():

    raise FileNotFoundError(
        f"Файл mmCIF не найден: {CIF_PATH}"
    )


dataframe_variables = [
    "df_residues",
    "df_atoms",
    "df_hetero",
    "df_water"
]

for variable in dataframe_variables:

    if not isinstance(globals()[variable], pd.DataFrame):

        raise TypeError(
            f"{variable} должен быть объектом pandas.DataFrame."
        )


if df_atoms.empty:

    raise ValueError(
        "df_atoms пуста. "
        "Невозможно проверить качество координатной модели."
    )


if df_residues.empty:

    raise ValueError(
        "df_residues пуста. "
        "Невозможно проверить белковые остатки."
    )


if not isinstance(models, (list, tuple)):

    try:
        models = list(models)

    except TypeError:

        raise TypeError(
            "Переменная models должна содержать "
            "список моделей структуры."
        )


if not isinstance(chains, (list, tuple)):

    try:
        chains = list(chains)

    except TypeError:

        raise TypeError(
            "Переменная chains должна содержать "
            "список цепей структуры."
        )


print(f"PDB ID: {PDB_ID}")
print(f"mmCIF file: {CIF_PATH.name}")
print(f"Каталог результатов: {PDB_RESULTS_DIR}")


# ------------------------------------------------------------
# 2. Чтение mmCIF
# ------------------------------------------------------------

cif_dict = MMCIF2Dict(str(CIF_PATH))

print(
    f"Количество полей mmCIF: {len(cif_dict)}"
)


# ------------------------------------------------------------
# 3. Функции обработки значений mmCIF
# ------------------------------------------------------------

def normalize_cif_value(value):
    """
    Преобразует значение mmCIF в удобный вид.

    Символы '.' и '?' обозначают отсутствующие
    или неизвестные значения.
    """

    if value is None:
        return None

    value = str(value).strip()

    if value in {"", ".", "?"}:
        return None

    return value


def get_cif_values(data, key):
    """
    Возвращает список доступных значений поля mmCIF.
    """

    value = data.get(key)

    if value is None:
        return []

    if not isinstance(value, (list, tuple)):
        value = [value]

    result = []

    for item in value:

        normalized = normalize_cif_value(item)

        if normalized is not None:
            result.append(normalized)

    return result


def get_cif_value(data, key):
    """
    Возвращает первое доступное значение поля mmCIF.
    """

    values = get_cif_values(data, key)

    return values[0] if values else None


def get_cif_category(data, category_prefix):
    """
    Извлекает все поля категории mmCIF
    и формирует DataFrame.

    Различает:
    - категорию отсутствует;
    - категория присутствует без записей;
    - категория содержит записи.
    """

    category_fields = {

        key: value

        for key, value in data.items()

        if key.startswith(category_prefix)

    }

    if not category_fields:

        return (
            None,
            "Категория отсутствует в mmCIF"
        )


    normalized_fields = {}

    for key, values in category_fields.items():

        if not isinstance(values, (list, tuple)):
            values = [values]

        normalized_fields[key] = [
            normalize_cif_value(value)
            for value in values
        ]


    record_lengths = {

        len(values)

        for values in normalized_fields.values()

    }


    if len(record_lengths) != 1:

        raise ValueError(
            f"Несогласованное число записей "
            f"в категории {category_prefix}: "
            f"{record_lengths}"
        )


    record_count = next(
        iter(record_lengths)
    )


    if record_count == 0:

        return (
            pd.DataFrame(
                columns=list(normalized_fields.keys())
            ),
            "Категория присутствует, записи отсутствуют"
        )


    df_category = pd.DataFrame(
        normalized_fields
    )


    return (
        df_category,
        "Категория присутствует, имеются записи"
    )


def parse_numeric_value(value):
    """
    Преобразует значение в число.
    При отсутствии или некорректном значении возвращает None.
    """

    if value is None:
        return None

    try:

        result = float(value)

        if np.isfinite(result):
            return result

    except (TypeError, ValueError):
        pass

    return None


# ------------------------------------------------------------
# 4. Извлечение основных метаданных
# ------------------------------------------------------------

structure_title = get_cif_value(
    cif_dict,
    "_struct.title"
)


experimental_method_values = get_cif_values(
    cif_dict,
    "_exptl.method"
)


experimental_method = (
    "; ".join(experimental_method_values)
    if experimental_method_values
    else None
)


method_lower = (
    experimental_method.lower()
    if experimental_method
    else ""
)


# Рентгеновская кристаллография.

resolution_xray = get_cif_value(
    cif_dict,
    "_refine.ls_d_res_high"
)

resolution_xray_fallback = get_cif_value(
    cif_dict,
    "_reflns.d_resolution_high"
)

if resolution_xray is None:

    resolution_xray = resolution_xray_fallback


# Электронная микроскопия.

resolution_em = get_cif_value(
    cif_dict,
    "_em_3d_reconstruction.resolution"
)


# Выбор разрешения с учётом экспериментального метода.

resolution = None

resolution_source = None


if "electron microscopy" in method_lower:

    if resolution_em is not None:

        resolution = resolution_em

        resolution_source = "Electron microscopy"

    elif resolution_xray is not None:

        resolution = resolution_xray

        resolution_source = "X-ray fallback"


elif (
    "x-ray" in method_lower
    or "x ray" in method_lower
):

    if resolution_xray is not None:

        resolution = resolution_xray

        resolution_source = "X-ray refinement"

    elif resolution_em is not None:

        resolution = resolution_em

        resolution_source = "Electron microscopy fallback"


else:

    if resolution_xray is not None:

        resolution = resolution_xray

        resolution_source = "X-ray refinement"

    elif resolution_em is not None:

        resolution = resolution_em

        resolution_source = "Electron microscopy"


if resolution_source is None:

    resolution_source = "Not available in mmCIF"


resolution_angstrom = parse_numeric_value(
    resolution
)


# Дополнительные сведения о структуре.

deposition_date = get_cif_value(
    cif_dict,
    "_pdbx_database_status.recvd_initial_deposition_date"
)

release_date = get_cif_value(
    cif_dict,
    "_pdbx_database_status.release_date"
)

structure_method_details = get_cif_value(
    cif_dict,
    "_struct_keywords.pdbx_keywords"
)


# ------------------------------------------------------------
# 5. Подсчёт структурных характеристик
# ------------------------------------------------------------
# Эти показатели получены из объектов Bio.PDB
# и таблиц предыдущих ячеек.
#
# CELL 4–8 обрабатывают первый модельный экземпляр.

number_of_models = len(models)

number_of_chains = len(chains)

number_of_protein_residues = len(df_residues)

number_of_protein_atoms = len(df_atoms)

number_of_nonwater_hetero_residues = len(df_hetero)

number_of_water_residues = len(df_water)


# Дополнительная проверка числа цепей по атомной таблице.

number_of_atom_chains = int(
    df_atoms["chain_id"].nunique()
) if "chain_id" in df_atoms.columns else None


# ------------------------------------------------------------
# 6. Проверка occupancy в координатных таблицах
# ------------------------------------------------------------
# occupancy = 0 означает нулевую занятость атомной позиции.
#
# Проверяем белковые, небелковые и водные атомы отдельно.
# Таблицы df_hetero_atoms и df_water_atoms создаются CELL 8.

def count_zero_occupancy(dataframe, table_name):
    """
    Подсчитывает атомы с occupancy = 0.
    Если столбца occupancy нет, возвращает None.
    """

    if dataframe is None or dataframe.empty:

        return {
            "table": table_name,
            "atom_count": 0,
            "zero_occupancy_count": 0,
            "occupancy_column_available": (
                dataframe is not None
                and "occupancy" in dataframe.columns
            )
        }


    if "occupancy" not in dataframe.columns:

        return {
            "table": table_name,
            "atom_count": len(dataframe),
            "zero_occupancy_count": None,
            "occupancy_column_available": False
        }


    occupancy = pd.to_numeric(
        dataframe["occupancy"],
        errors="coerce"
    )


    zero_count = int(
        (occupancy == 0).sum()
    )


    return {
        "table": table_name,
        "atom_count": len(dataframe),
        "zero_occupancy_count": zero_count,
        "occupancy_column_available": True
    }


occupancy_records = [

    count_zero_occupancy(
        df_atoms,
        "Protein atoms"
    ),

    count_zero_occupancy(
        globals().get("df_hetero_atoms"),
        "Hetero atoms"
    ),

    count_zero_occupancy(
        globals().get("df_water_atoms"),
        "Water atoms"
    )

]


df_occupancy_summary = pd.DataFrame(
    occupancy_records
)


# ------------------------------------------------------------
# 7. Формирование словаря основных метаданных
# ------------------------------------------------------------

metadata = {

    "PDB_ID": PDB_ID,

    "Title": structure_title,

    "Experimental_method": experimental_method,

    "Structure_keywords": structure_method_details,

    "Resolution": resolution,

    "Resolution_Angstrom": resolution_angstrom,

    "Resolution_source": resolution_source,

    "Resolution_X_ray": resolution_xray,

    "Resolution_EM": resolution_em,

    "Deposition_date": deposition_date,

    "Release_date": release_date,

    "Number_of_models_in_parsed_structure": number_of_models,

    "Number_of_chains_in_analyzed_model": number_of_chains,

    "Number_of_chains_in_protein_atom_table": number_of_atom_chains,

    "Number_of_protein_residues": number_of_protein_residues,

    "Number_of_protein_atoms": number_of_protein_atoms,

    "Number_of_nonwater_hetero_residues":
        number_of_nonwater_hetero_residues,

    "Number_of_water_residues": number_of_water_residues,

    "Analyzed_model_scope": "First model",

    "Coordinate_unit": "Angstrom"

}


# ------------------------------------------------------------
# 8. Извлечение сведений о неразрешённых остатках
# ------------------------------------------------------------

(
    df_unobserved_residues,
    residues_category_status
) = get_cif_category(
    cif_dict,
    "_pdbx_unobs_or_zero_occ_residues."
)


if df_unobserved_residues is None:

    unobserved_residue_count = None

else:

    unobserved_residue_count = len(
        df_unobserved_residues
    )


# ------------------------------------------------------------
# 9. Извлечение сведений о неразрешённых атомах
# ------------------------------------------------------------

(
    df_unobserved_atoms,
    atoms_category_status
) = get_cif_category(
    cif_dict,
    "_pdbx_unobs_or_zero_occ_atoms."
)


if df_unobserved_atoms is None:

    unobserved_atom_count = None

else:

    unobserved_atom_count = len(
        df_unobserved_atoms
    )


# ------------------------------------------------------------
# 10. Сводка сведений о неразрешённых остатках и атомах
# ------------------------------------------------------------

df_unobserved_summary = pd.DataFrame(

    [

        {
            "Parameter": "Unobserved / zero-occupancy residue records",

            "Count": unobserved_residue_count,

            "Status": residues_category_status

        },

        {
            "Parameter": "Unobserved / zero-occupancy atom records",

            "Count": unobserved_atom_count,

            "Status": atoms_category_status

        }

    ]

)


# ------------------------------------------------------------
# 11. Сводка основных метаданных
# ------------------------------------------------------------

df_metadata = pd.DataFrame(

    [

        {
            "Parameter": key,
            "Value": value
        }

        for key, value in metadata.items()

    ]

)


# ------------------------------------------------------------
# 12. Вывод результатов
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ОСНОВНЫЕ МЕТАДАННЫЕ")
print("-" * 70)

for key, value in metadata.items():

    print(f"{key}: {value}")


print("\n" + "-" * 70)
print("НЕРАЗРЕШЁННЫЕ ОСТАТКИ И АТОМЫ")
print("-" * 70)

display(df_unobserved_summary)


print("\n" + "-" * 70)
print("АТОМЫ С НУЛЕВОЙ ЗАНЯТОСТЬЮ")
print("-" * 70)

display(df_occupancy_summary)


# ------------------------------------------------------------
# 13. Просмотр содержимого категорий mmCIF
# ------------------------------------------------------------

if df_unobserved_residues is not None:

    print("\nЗаписи о неразрешённых остатках:")

    if not df_unobserved_residues.empty:

        display(
            df_unobserved_residues.head(20)
        )

    else:

        print(
            "Категория присутствует, "
            "но записей в ней нет."
        )

else:

    print(
        "\nКатегория неразрешённых остатков "
        "отсутствует в mmCIF."
    )


if df_unobserved_atoms is not None:

    print("\nЗаписи о неразрешённых атомах:")

    if not df_unobserved_atoms.empty:

        display(
            df_unobserved_atoms.head(20)
        )

    else:

        print(
            "Категория присутствует, "
            "но записей в ней нет."
        )

else:

    print(
        "\nКатегория неразрешённых атомов "
        "отсутствует в mmCIF."
    )


# ------------------------------------------------------------
# 14. Сохранение результатов
# ------------------------------------------------------------

METADATA_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_metadata.csv"
)

METADATA_JSON_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_metadata.json"
)

UNOBSERVED_RESIDUES_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_unobserved_residues.csv"
)

UNOBSERVED_ATOMS_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_unobserved_atoms.csv"
)

UNOBSERVED_SUMMARY_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_unobserved_summary.csv"
)

OCCUPANCY_SUMMARY_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_occupancy_summary.csv"
)


df_metadata.to_csv(
    METADATA_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


if df_unobserved_residues is not None:

    df_unobserved_residues.to_csv(
        UNOBSERVED_RESIDUES_CSV_PATH,
        index=False,
        encoding="utf-8-sig"
    )

else:

    pd.DataFrame().to_csv(
        UNOBSERVED_RESIDUES_CSV_PATH,
        index=False,
        encoding="utf-8-sig"
    )


if df_unobserved_atoms is not None:

    df_unobserved_atoms.to_csv(
        UNOBSERVED_ATOMS_CSV_PATH,
        index=False,
        encoding="utf-8-sig"
    )

else:

    pd.DataFrame().to_csv(
        UNOBSERVED_ATOMS_CSV_PATH,
        index=False,
        encoding="utf-8-sig"
    )


df_unobserved_summary.to_csv(
    UNOBSERVED_SUMMARY_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


df_occupancy_summary.to_csv(
    OCCUPANCY_SUMMARY_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 15. Контроль согласованности данных
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("КОНТРОЛЬ СОГЛАСОВАННОСТИ ДАННЫХ")
print("-" * 70)


checks = {

    "At least one parsed model":
        number_of_models > 0,

    "At least one chain":
        number_of_chains > 0,

    "Protein residues available":
        number_of_protein_residues > 0,

    "Protein atoms available":
        number_of_protein_atoms > 0,

    "PDB ID matches coordinate table":
        (
            "pdb_id" in df_atoms.columns
            and df_atoms["pdb_id"]
                .astype(str)
                .str.strip()
                .str.upper()
                .eq(PDB_ID)
                .all()
        )

}


for check_name, check_result in checks.items():

    print(
        f"{check_name}: "
        f"{'OK' if check_result else 'ERROR'}"
    )


if not all(checks.values()):

    raise ValueError(
        "Обнаружена проблема при проверке метаданных "
        "или структурных таблиц."
    )


# ------------------------------------------------------------
# 16. Сохранение JSON с полной сводкой качества
# ------------------------------------------------------------

quality_summary = {

    "pdb_id": PDB_ID,

    "unobserved_residues": {

        "record_count": unobserved_residue_count,

        "category_status": residues_category_status

    },

    "unobserved_atoms": {

        "record_count": unobserved_atom_count,

        "category_status": atoms_category_status

    },

    "zero_occupancy": occupancy_records,

    "checks": checks,

    "analysis_scope": "First model",

    "interpretation": {

        "unobserved_categories":
            "mmCIF records describing unobserved or zero-occupancy positions",

        "zero_occupancy_tables":
            "Atoms present in extracted coordinate tables with occupancy equal to zero"

    }

}


METADATA_JSON_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_metadata.json"
)

metadata_complete = {

    "metadata": metadata,

    "quality": quality_summary

}


with open(
    METADATA_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata_complete,
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


# ------------------------------------------------------------
# 17. Проверка сохранённых файлов
# ------------------------------------------------------------

output_paths = [

    METADATA_CSV_PATH,

    METADATA_JSON_PATH,

    UNOBSERVED_RESIDUES_CSV_PATH,

    UNOBSERVED_ATOMS_CSV_PATH,

    UNOBSERVED_SUMMARY_CSV_PATH,

    OCCUPANCY_SUMMARY_CSV_PATH

]


print("\n" + "-" * 70)
print("ПРОВЕРКА СОХРАНЁННЫХ ФАЙЛОВ")
print("-" * 70)


output_checks = {}

for path in output_paths:

    file_exists = (
        path.is_file()
        and path.stat().st_size > 0
    )

    output_checks[path.name] = bool(
        file_exists
    )

    print(
        f"{path.name}: "
        f"{'OK' if file_exists else 'ERROR'}"
    )


if not all(output_checks.values()):

    raise RuntimeError(
        "Не все файлы метаданных и контроля качества "
        "успешно сохранены."
    )


print("\nИнтерпретация:")

print(
    "- Количество моделей относится к моделям, "
    "разобранным Bio.PDB."
)

print(
    "- Количество цепей, остатков и атомов относится "
    "к анализируемой модели."
)

print(
    "- Записи unobserved/zero occupancy из mmCIF "
    "не следует напрямую приравнивать к отсутствующим "
    "атомам в координатной таблице."
)

print(
    "- Отсутствие категории mmCIF не доказывает "
    "полное отсутствие структурных проблем."
)


print("\nСтатус: OK")
print("Метаданные и сведения о качестве извлечены и сохранены.")

print("=" * 70)
print("CELL 10 COMPLETED")
print("=" * 70)
```

Здесь важно учитывать ограничение: отсутствие записей в этих полях не гарантирует, что структура полностью разрешена. Для полноценного анализа необходимо также учитывать пропуски в последовательности, occupancy, альтернативные положения атомов и особенности конкретной структуры.

```python

# ============================================================
# CELL 11. AMINO ACID DISTRIBUTION
# ============================================================
# Назначение:
# Анализ аминокислотного состава белковой структуры.
#
# Рассчитывается:
# 1. Количество остатков каждого типа.
# 2. Процентное содержание аминокислот.
# 3. Накопительный процент.
# 4. Распределение аминокислот по отдельным цепям.
# 5. Доля неизвестных и исключённых обозначений.
#
# Источник данных: df_residues (CELL 6).
#
# Вода и небелковые компоненты не учитываются.
# Нестандартные аминокислоты, сопоставленные с кодами
# в CELL 5, сохраняются в анализе.
# ============================================================

import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from IPython.display import display


print("=" * 70)
print("CELL 11. AMINO ACID DISTRIBUTION")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка входных данных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "df_residues",
    "PDB_RESULTS_DIR"
]

for variable in required_variables:

    if variable not in globals():

        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()

PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


if not isinstance(df_residues, pd.DataFrame):

    raise TypeError(
        "df_residues должен быть объектом pandas.DataFrame."
    )


if df_residues.empty:

    raise ValueError(
        "Таблица df_residues пуста."
    )


required_columns = [
    "pdb_id",
    "chain_id",
    "amino_acid"
]

missing_columns = [
    column
    for column in required_columns
    if column not in df_residues.columns
]

if missing_columns:

    raise ValueError(
        "В df_residues отсутствуют обязательные столбцы: "
        f"{missing_columns}"
    )


# Проверяем принадлежность таблицы текущей структуре.

source_pdb_ids = (
    df_residues["pdb_id"]
    .dropna()
    .astype(str)
    .str.strip()
    .str.upper()
    .unique()
)

if len(source_pdb_ids) != 1 or source_pdb_ids[0] != PDB_ID:

    raise ValueError(
        "Несоответствие PDB_ID: "
        f"текущая структура = {PDB_ID}, "
        f"структуры в df_residues = {source_pdb_ids.tolist()}"
    )


print(f"PDB ID: {PDB_ID}")
print(f"Количество записей в df_residues: {len(df_residues)}")
print(f"Каталог результатов: {PDB_RESULTS_DIR}")


# ------------------------------------------------------------
# 2. Подготовка аминокислотных данных
# ------------------------------------------------------------

amino_acid_series = (

    df_residues["amino_acid"]

    .fillna("")

    .astype(str)

    .str.strip()

    .str.upper()

)


# Неизвестные и неоднозначные обозначения.
#
# U и O сохраняются, поскольку могут соответствовать
# нестандартным аминокислотам:
# U — селеноцистеин;
# O — пирролизин.

invalid_amino_acid_values = {

    "",
    "X",
    "UNK",
    "NONE",
    "NAN",
    "?",
    "-",
    "B",
    "Z",
    "J"

}


valid_amino_acid_mask = (

    ~amino_acid_series.isin(
        invalid_amino_acid_values
    )

    & amino_acid_series.str.len().eq(1)

)


valid_amino_acid_series = (
    amino_acid_series[valid_amino_acid_mask]
)


excluded_amino_acid_count = int(
    (~valid_amino_acid_mask).sum()
)


if valid_amino_acid_series.empty:

    raise ValueError(
        "Не найдено пригодных для анализа "
        "аминокислотных остатков."
    )


# ------------------------------------------------------------
# 3. Подсчёт количества аминокислот
# ------------------------------------------------------------

amino_acid_counts = (

    valid_amino_acid_series

    .value_counts()

    .sort_values(
        ascending=False,
        kind="stable"
    )

)


amino_acid_distribution = pd.DataFrame({

    "amino_acid": amino_acid_counts.index,

    "count": amino_acid_counts.values

})


# ------------------------------------------------------------
# 4. Расчёт процентного содержания
# ------------------------------------------------------------

total_residues = int(
    amino_acid_distribution["count"].sum()
)


# Сначала рассчитываем точные значения,
# затем округляем только для отображения.

amino_acid_distribution["percentage_raw"] = (

    amino_acid_distribution["count"]
    / total_residues
    * 100

)


amino_acid_distribution["percentage"] = (

    amino_acid_distribution["percentage_raw"]
    .round(2)

)


# ------------------------------------------------------------
# 5. Накопительный процент
# ------------------------------------------------------------

amino_acid_distribution["cumulative_percentage"] = (

    amino_acid_distribution["percentage_raw"]
    .cumsum()
    .round(2)

)


# Сумма точных процентов используется для проверки.

percentage_sum_raw = float(
    amino_acid_distribution["percentage_raw"].sum()
)


percentage_sum = float(
    amino_acid_distribution["percentage"].sum()
)


# Удаляем вспомогательный столбец с неокруглёнными
# процентами из итоговой пользовательской таблицы.

amino_acid_distribution = (

    amino_acid_distribution

    .drop(
        columns=["percentage_raw"]
    )

)


# ------------------------------------------------------------
# 6. Дополнительная статистика
# ------------------------------------------------------------

number_of_amino_acid_types = len(
    amino_acid_distribution
)


most_common_amino_acid = str(
    amino_acid_distribution.iloc[0]["amino_acid"]
)


most_common_amino_acid_count = int(
    amino_acid_distribution.iloc[0]["count"]
)


least_common_amino_acid = str(
    amino_acid_distribution.iloc[-1]["amino_acid"]
)


least_common_amino_acid_count = int(
    amino_acid_distribution.iloc[-1]["count"]
)


total_residue_records = len(
    df_residues
)


excluded_percentage = (

    excluded_amino_acid_count
    / total_residue_records
    * 100

)


# ------------------------------------------------------------
# 7. Распределение аминокислот по цепям
# ------------------------------------------------------------

chain_data = df_residues[
    ["chain_id", "amino_acid"]
].copy()


chain_data["amino_acid"] = (

    chain_data["amino_acid"]

    .fillna("")

    .astype(str)

    .str.strip()

    .str.upper()

)


chain_valid_mask = (

    ~chain_data["amino_acid"].isin(
        invalid_amino_acid_values
    )

    & chain_data["amino_acid"].str.len().eq(1)

)


chain_data = chain_data.loc[
    chain_valid_mask
].copy()


# Сохраняем исходный chain_id для идентификации.
# Пустую цепь отображаем отдельно как <blank>.

chain_data["chain_id_display"] = (

    chain_data["chain_id"]

    .fillna("<blank>")

    .astype(str)

    .replace(
        "",
        "<blank>"
    )

    .replace(
        " ",
        "<blank>"
    )

)


chain_amino_acid_distribution = (

    chain_data

    .groupby(
        ["chain_id_display", "amino_acid"],
        dropna=False
    )

    .size()

    .reset_index(name="count")

)


# Процентное содержание аминокислот внутри каждой цепи.

chain_amino_acid_distribution["percentage"] = (

    chain_amino_acid_distribution

    .groupby("chain_id_display")["count"]

    .transform(
        lambda values:
            values / values.sum() * 100
    )

    .round(2)

)


chain_amino_acid_distribution = (

    chain_amino_acid_distribution

    .sort_values(
        ["chain_id_display", "count", "amino_acid"],
        ascending=[True, False, True]
    )

    .reset_index(drop=True)

)


# ------------------------------------------------------------
# 8. Вывод общего распределения
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ОБЩЕЕ РАСПРЕДЕЛЕНИЕ АМИНОКИСЛОТ")
print("-" * 70)

display(amino_acid_distribution)


print(f"\nTotal residue records: {total_residue_records}")

print(f"Counted amino acid residues: {total_residues}")

print(
    f"Number of amino acid types: "
    f"{number_of_amino_acid_types}"
)

print(
    f"Excluded records (missing/unknown): "
    f"{excluded_amino_acid_count}"
)

print(
    f"Excluded percentage: "
    f"{excluded_percentage:.2f}%"
)

print(
    f"Most common amino acid: "
    f"{most_common_amino_acid} "
    f"({most_common_amino_acid_count})"
)

print(
    f"Least common amino acid: "
    f"{least_common_amino_acid} "
    f"({least_common_amino_acid_count})"
)

print(
    f"Sum of exact percentages: "
    f"{percentage_sum_raw:.6f}%"
)

print(
    f"Sum of displayed percentages: "
    f"{percentage_sum:.2f}%"
)


# ------------------------------------------------------------
# 9. Вывод распределения по цепям
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("РАСПРЕДЕЛЕНИЕ АМИНОКИСЛОТ ПО ЦЕПЯМ")
print("-" * 70)

display(
    chain_amino_acid_distribution
)


# ------------------------------------------------------------
# 10. Сохранение общей таблицы
# ------------------------------------------------------------

AMINO_ACID_CSV_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_amino_acid_distribution.csv"

)


amino_acid_distribution.to_csv(
    AMINO_ACID_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


print(
    "\nAmino acid distribution saved:",
    AMINO_ACID_CSV_PATH
)


# ------------------------------------------------------------
# 11. Сохранение распределения по цепям
# ------------------------------------------------------------

AMINO_ACID_CHAIN_CSV_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_amino_acid_distribution_by_chain.csv"

)


chain_amino_acid_distribution.to_csv(
    AMINO_ACID_CHAIN_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


print(
    "Distribution by chain saved:",
    AMINO_ACID_CHAIN_CSV_PATH
)


# ------------------------------------------------------------
# 12. Диаграмма количества аминокислот
# ------------------------------------------------------------

AMINO_ACID_PLOT_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_amino_acid_distribution.png"

)


fig, ax = plt.subplots(
    figsize=(12, 5)
)


ax.bar(
    amino_acid_distribution["amino_acid"],
    amino_acid_distribution["count"]
)


ax.set_title(
    f"Amino acid distribution: {PDB_ID}"
)

ax.set_xlabel("Amino acid")

ax.set_ylabel("Number of residues")


ax.tick_params(
    axis="x",
    rotation=0
)


ax.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    AMINO_ACID_PLOT_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close(fig)


# ------------------------------------------------------------
# 13. Диаграмма процентного содержания
# ------------------------------------------------------------

AMINO_ACID_PERCENTAGE_PLOT_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_amino_acid_percentage.png"

)


fig, ax = plt.subplots(
    figsize=(12, 5)
)


ax.bar(
    amino_acid_distribution["amino_acid"],
    amino_acid_distribution["percentage"]
)


ax.set_title(
    f"Amino acid percentage distribution: {PDB_ID}"
)

ax.set_xlabel("Amino acid")

ax.set_ylabel("Percentage (%)")


ax.tick_params(
    axis="x",
    rotation=0
)


ax.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    AMINO_ACID_PERCENTAGE_PLOT_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close(fig)


# ------------------------------------------------------------
# 14. Проверка результатов
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("VERIFICATION")
print("-" * 70)


count_check = (

    total_residues + excluded_amino_acid_count

    == total_residue_records

)


percentage_check = np.isclose(
    percentage_sum_raw,
    100.0,
    atol=1e-8
)


chain_count_check = (

    int(chain_amino_acid_distribution["count"].sum())

    == total_residues

)


print(
    "Rows in df_residues:",
    total_residue_records
)

print(
    "Counted amino acid residues:",
    total_residues
)

print(
    "Excluded records:",
    excluded_amino_acid_count
)

print(
    "Residue count check:",
    "OK" if count_check else "ERROR"
)

print(
    "Percentage check:",
    "OK" if percentage_check else "ERROR"
)

print(
    "Chain distribution check:",
    "OK" if chain_count_check else "ERROR"
)


if not all([
    count_check,
    percentage_check,
    chain_count_check
]):

    raise ValueError(
        "Ошибка проверки распределения аминокислот."
    )


# ------------------------------------------------------------
# 15. Формирование итоговой сводки
# ------------------------------------------------------------

amino_acid_summary = {

    "pdb_id": PDB_ID,

    "total_residue_records": total_residue_records,

    "counted_amino_acid_residues": total_residues,

    "excluded_residue_records": excluded_amino_acid_count,

    "excluded_percentage": float(excluded_percentage),

    "number_of_amino_acid_types": number_of_amino_acid_types,

    "most_common_amino_acid": most_common_amino_acid,

    "most_common_amino_acid_count": most_common_amino_acid_count,

    "least_common_amino_acid": least_common_amino_acid,

    "least_common_amino_acid_count": least_common_amino_acid_count,

    "percentage_sum": percentage_sum_raw,

    "chain_count": int(
        df_residues["chain_id"].nunique()
    ),

    "included_nonstandard_codes": ["U", "O"],

    "excluded_codes": sorted(
        invalid_amino_acid_values
    ),

    "checks": {

        "residue_count_check": bool(count_check),

        "percentage_check": bool(percentage_check),

        "chain_distribution_check": bool(chain_count_check)

    }

}


AMINO_ACID_JSON_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_amino_acid_summary.json"

)


with open(
    AMINO_ACID_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        amino_acid_summary,
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


# ------------------------------------------------------------
# 16. Проверка сохранённых файлов
# ------------------------------------------------------------

output_paths = [

    AMINO_ACID_CSV_PATH,

    AMINO_ACID_CHAIN_CSV_PATH,

    AMINO_ACID_PLOT_PATH,

    AMINO_ACID_PERCENTAGE_PLOT_PATH,

    AMINO_ACID_JSON_PATH

]


print("\n" + "-" * 70)
print("SAVED FILES")
print("-" * 70)


output_checks = {}

for path in output_paths:

    file_exists = (

        path.is_file()

        and path.stat().st_size > 0

    )

    output_checks[path.name] = bool(
        file_exists
    )

    print(
        f"{path.name}: "
        f"{'OK' if file_exists else 'ERROR'}"
    )


if not all(output_checks.values()):

    raise RuntimeError(
        "Не все файлы анализа аминокислот "
        "успешно сохранены."
    )


# ------------------------------------------------------------
# 17. Итоговый статус
# ------------------------------------------------------------

amino_acid_summary["output_checks"] = output_checks

amino_acid_summary["status"] = "OK"


with open(
    AMINO_ACID_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        amino_acid_summary,
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


print("\nИтог:")

print(
    "Распределение аминокислот рассчитано "
    "по всем белковым остаткам структуры."
)

print(
    "Дополнительно сформировано распределение "
    "по отдельным цепям."
)

print(
    "Неизвестные и отсутствующие обозначения "
    "учтены отдельно и не включены в процентный расчёт."
)

print(
    "Нестандартные аминокислоты с распознанными "
    "однобуквенными кодами сохранены."
)

print("\nСтатус: OK")

print("=" * 70)
print("CELL 11 COMPLETED")
print("=" * 70)
```

```python

# ============================================================
# CELL 12. SAVE ANALYSIS RESULTS
# ============================================================
# Назначение:
# Централизованное сохранение результатов анализа структуры.
#
# Используются результаты предыдущих ячеек:
# - CELL 4: общая статистика структуры;
# - CELL 5: последовательности белковых цепей;
# - CELL 6: таблица белковых остатков;
# - CELL 7: таблица белковых атомов;
# - CELL 8: гетерокомпоненты и вода;
# - CELL 9: геометрические характеристики;
# - CELL 10: метаданные и качество;
# - CELL 11: распределение аминокислот.
#
# Результаты сохраняются в отдельном каталоге PDB_ID.
# ============================================================

import json
import pandas as pd
import numpy as np

from pathlib import Path
from IPython.display import display


print("=" * 70)
print("CELL 12. SAVE ANALYSIS RESULTS")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка основных переменных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "PDB_RESULTS_DIR",
    "chain_sequences",
    "df_residues",
    "df_atoms",
    "df_hetero",
    "df_water"
]

for variable in required_variables:

    if variable not in globals():

        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()

if not PDB_ID:

    raise ValueError(
        "PDB_ID не может быть пустым."
    )


PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# Проверяем, что каталог соответствует текущей структуре.

if PDB_RESULTS_DIR.name.upper() != PDB_ID:

    raise ValueError(
        "Каталог результатов не соответствует текущему PDB_ID: "
        f"{PDB_RESULTS_DIR}"
    )


# Проверяем типы основных таблиц.

dataframe_variables = [
    "df_residues",
    "df_atoms",
    "df_hetero",
    "df_water"
]

for variable in dataframe_variables:

    if not isinstance(globals()[variable], pd.DataFrame):

        raise TypeError(
            f"{variable} должен быть объектом pandas.DataFrame."
        )


if not isinstance(chain_sequences, dict):

    raise TypeError(
        "chain_sequences должен быть словарём."
    )


print(f"PDB ID: {PDB_ID}")
print(f"Results directory: {PDB_RESULTS_DIR}")


# ------------------------------------------------------------
# 2. Проверка принадлежности таблиц текущей структуре
# ------------------------------------------------------------

for variable in dataframe_variables:

    df = globals()[variable]

    if "pdb_id" not in df.columns:

        raise ValueError(
            f"В {variable} отсутствует столбец pdb_id. "
            "Невозможно проверить принадлежность данных."
        )

    source_ids = (

        df["pdb_id"]

        .dropna()

        .astype(str)

        .str.strip()

        .str.upper()

        .unique()

    )

    if len(source_ids) > 0:

        if len(source_ids) != 1 or source_ids[0] != PDB_ID:

            raise ValueError(
                f"Обнаружено несоответствие PDB_ID в {variable}: "
                f"ожидалось {PDB_ID}, "
                f"получено {source_ids.tolist()}"
            )


print("DataFrame PDB_ID consistency: OK")


# ------------------------------------------------------------
# 3. Функция преобразования значений для JSON
# ------------------------------------------------------------

def make_json_serializable(value):
    """
    Рекурсивно преобразует объекты NumPy и pandas
    в типы, поддерживаемые JSON.

    NaN, Infinity, NaT и pd.NA преобразуются в None.
    """

    if value is None:

        return None


    if value is pd.NA or value is pd.NaT:

        return None


    if isinstance(value, Path):

        return str(value)


    if isinstance(value, pd.DataFrame):

        return [

            make_json_serializable(record)

            for record in value.to_dict(
                orient="records"
            )

        ]


    if isinstance(value, pd.Series):

        return make_json_serializable(
            value.to_dict()
        )


    if isinstance(value, dict):

        return {

            str(key): make_json_serializable(item)

            for key, item in value.items()

        }


    if isinstance(value, (list, tuple, set)):

        return [

            make_json_serializable(item)

            for item in value

        ]


    if isinstance(value, np.ndarray):

        return make_json_serializable(
            value.tolist()
        )


    if isinstance(value, np.generic):

        return make_json_serializable(
            value.item()
        )


    if isinstance(value, (pd.Timestamp, pd.Timedelta)):

        return value.isoformat()


    if isinstance(value, float):

        if not np.isfinite(value):

            return None

        return float(value)


    # Дополнительная проверка пропущенных значений pandas.

    try:

        missing_value = pd.isna(value)

        if isinstance(
            missing_value,
            (bool, np.bool_)
        ) and missing_value:

            return None

    except (TypeError, ValueError):

        pass


    return value


# ------------------------------------------------------------
# 4. Информация о белковых последовательностях
# ------------------------------------------------------------

chain_lengths = {

    str(chain_id): len(sequence)

    for chain_id, sequence in chain_sequences.items()

}


chain_information = {

    "sequence_chains": list(chain_lengths.keys()),

    "chain_lengths": chain_lengths,

    "number_of_sequence_chains": len(chain_sequences),

    "total_sequence_length": sum(
        chain_lengths.values()
    )

}


# ------------------------------------------------------------
# 5. Общая статистика структуры
# ------------------------------------------------------------

number_of_models = (

    len(models)

    if "models" in globals()

    else None

)


number_of_chains = (

    len(chains)

    if "chains" in globals()

    else len(chain_sequences)

)


structure_summary = {

    "PDB_ID": PDB_ID,

    "number_of_models": number_of_models,

    "number_of_chains": number_of_chains,

    "number_of_sequence_chains": len(chain_sequences),

    "protein_residues": len(df_residues),

    "protein_atoms": len(df_atoms),

    "hetero_components": len(df_hetero),

    "water_residues": len(df_water),

    "chain_information": chain_information

}


# ------------------------------------------------------------
# 6. Добавление геометрических характеристик
# ------------------------------------------------------------

if "geometry_summary" in globals():

    structure_summary["geometry"] = geometry_summary

elif "dimensions" in globals():

    structure_summary["geometry"] = {

        "coordinate_dimensions": np.asarray(
            dimensions
        ).tolist()

    }

else:

    structure_summary["geometry"] = None


# ------------------------------------------------------------
# 7. Добавление метаданных
# ------------------------------------------------------------

if "metadata" in globals():

    structure_summary["metadata"] = metadata

else:

    structure_summary["metadata"] = None


# ------------------------------------------------------------
# 8. Добавление результатов контроля качества
# ------------------------------------------------------------

quality_summary = {

    "unobserved_residue_records": (

        unobserved_residue_count

        if "unobserved_residue_count" in globals()

        else None

    ),

    "unobserved_atom_records": (

        unobserved_atom_count

        if "unobserved_atom_count" in globals()

        else None

    ),

    "residues_category_status": (

        residues_category_status

        if "residues_category_status" in globals()

        else "Not checked"

    ),

    "atoms_category_status": (

        atoms_category_status

        if "atoms_category_status" in globals()

        else "Not checked"

    )

}


structure_summary["quality_information"] = quality_summary


# ------------------------------------------------------------
# 9. Добавление результатов анализа аминокислот
# ------------------------------------------------------------

if "amino_acid_summary" in globals():

    structure_summary["amino_acid_analysis"] = (
        amino_acid_summary
    )

elif "amino_acid_distribution" in globals():

    distribution = amino_acid_distribution

    if isinstance(distribution, pd.DataFrame):

        structure_summary["amino_acid_analysis"] = {

            "number_of_amino_acid_types": len(distribution),

            "total_counted_residues": (

                int(distribution["count"].sum())

                if "count" in distribution.columns

                else None

            )

        }

    else:

        structure_summary["amino_acid_analysis"] = None

else:

    structure_summary["amino_acid_analysis"] = None


# ------------------------------------------------------------
# 10. Сохранение итоговой сводки JSON
# ------------------------------------------------------------

SUMMARY_JSON_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_summary.json"

)


summary = make_json_serializable(
    structure_summary
)


with open(
    SUMMARY_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        summary,
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


print("\nИтоговая сводка JSON сохранена:")
print(SUMMARY_JSON_PATH)


# ------------------------------------------------------------
# 11. Сохранение краткой сводной таблицы CSV
# ------------------------------------------------------------

SUMMARY_CSV_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_summary.csv"

)


summary_table = pd.DataFrame([{

    "PDB_ID": PDB_ID,

    "Number of models": number_of_models,

    "Number of chains": number_of_chains,

    "Number of sequence chains": len(chain_sequences),

    "Protein residues": len(df_residues),

    "Protein atoms": len(df_atoms),

    "Hetero components": len(df_hetero),

    "Water residues": len(df_water),

    "Total sequence length": sum(
        chain_lengths.values()
    )

}])


summary_table.to_csv(
    SUMMARY_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 12. Инвентаризация файлов текущей структуры
# ------------------------------------------------------------

# Учитываем только файлы непосредственно в каталоге PDB_ID.
# Другие структуры не могут попасть в реестр.
#
# Сам реестр исключается из собственного содержимого.

OUTPUT_FILES_CSV_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_output_files.csv"

)


file_records = []


for path in sorted(PDB_RESULTS_DIR.iterdir()):

    if not path.is_file():

        continue


    if path == OUTPUT_FILES_CSV_PATH:

        continue


    file_size = path.stat().st_size


    file_records.append({

        "file_name": path.name,

        "extension": path.suffix.lower(),

        "size_bytes": file_size,

        "size_kb": round(
            file_size / 1024,
            2
        ),

        "exists": True

    })


df_output_files = pd.DataFrame(

    file_records,

    columns=[
        "file_name",
        "extension",
        "size_bytes",
        "size_kb",
        "exists"
    ]

)


# ------------------------------------------------------------
# 13. Сохранение реестра файлов
# ------------------------------------------------------------

df_output_files.to_csv(
    OUTPUT_FILES_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


print("\nРеестр файлов сохранён:")
print(OUTPUT_FILES_CSV_PATH)

print(
    "Количество найденных файлов "
    "(без самого реестра):",
    len(df_output_files)
)


# ------------------------------------------------------------
# 14. Вывод общей сводки
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("STRUCTURE SUMMARY")
print("-" * 70)


print(f"PDB ID: {PDB_ID}")

print(
    f"Number of models: "
    f"{number_of_models}"
)

print(
    f"Number of chains: "
    f"{number_of_chains}"
)

print(
    f"Number of chains with sequences: "
    f"{len(chain_sequences)}"
)

print(
    f"Protein residues: "
    f"{len(df_residues)}"
)

print(
    f"Protein atoms: "
    f"{len(df_atoms)}"
)

print(
    f"Hetero components: "
    f"{len(df_hetero)}"
)

print(
    f"Water residues: "
    f"{len(df_water)}"
)


print("\nChain sequence lengths:")

for chain_id, length in chain_lengths.items():

    print(
        f"Chain {chain_id}: {length} residues"
    )


# ------------------------------------------------------------
# 15. Вывод реестра файлов
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("OUTPUT FILES")
print("-" * 70)

display(df_output_files)


# ------------------------------------------------------------
# 16. Проверка сохранённых файлов
# ------------------------------------------------------------

important_files = [

    SUMMARY_JSON_PATH,

    SUMMARY_CSV_PATH,

    OUTPUT_FILES_CSV_PATH

]


print("\nFile verification:")


output_checks = {}


for path in important_files:

    file_exists = (

        path.is_file()

        and path.stat().st_size > 0

    )


    output_checks[path.name] = bool(
        file_exists
    )


    print(

        f"{path.name}: "

        f"{'OK' if file_exists else 'ERROR'}"

    )


if not all(output_checks.values()):

    raise RuntimeError(
        "Не все итоговые файлы были успешно сохранены."
    )


# ------------------------------------------------------------
# 17. Итоговый статус
# ------------------------------------------------------------

# Реестр не включает сам себя.
# При общем подсчёте добавляем его отдельно.

total_pdb_files = len(df_output_files) + 1


print("\n" + "-" * 70)

print("ANALYSIS RESULTS SAVED")

print(
    f"Results directory: {PDB_RESULTS_DIR}"
)

print(
    f"Number of output files for {PDB_ID}: "
    f"{total_pdb_files}"
)

print(
    "Files from other PDB structures: excluded"
)

print("\nСтатус: OK")

print("=" * 70)
print("CELL 12 COMPLETED")
print("=" * 70)
```

```python

# ============================================================
# CELL 13. PROTEIN SEQUENCE VALIDATION
# ============================================================
# Назначение:
# Сравнить последовательности белковых цепей, извлечённые
# из координатной модели, с референсными последовательностями
# полимерных сущностей из mmCIF.
#
# Анализ:
# 1. Загрузка референсных последовательностей из mmCIF.
# 2. Определение соответствия chain_id и entity_id.
# 3. Глобальное выравнивание последовательностей.
# 4. Расчёт идентичности и покрытия.
# 5. Подсчёт несовпадений и gap-позиций.
# 6. Проверка полноты сравнения цепей.
#
# ВАЖНО:
# - Сравниваются последовательности, а не координаты атомов.
# - Gap в выравнивании не доказывает отсутствие остатка
#   в экспериментальной координатной модели.
# - Сравниваются цепи первого структурного контекста,
#   сформированного в предыдущих ячейках.
# - При невозможности установить entity_id используется
#   резервный поиск по всем референсным последовательностям.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.Align import PairwiseAligner
from Bio.PDB.MMCIF2Dict import MMCIF2Dict


print("=" * 70)
print("CELL 13. PROTEIN SEQUENCE VALIDATION")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка входных данных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "CIF_PATH",
    "chain_sequences"
]

for variable in required_variables:

    if variable not in globals():

        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()

CIF_PATH = Path(CIF_PATH)

if not PDB_ID:

    raise ValueError("PDB_ID не может быть пустым.")

if not CIF_PATH.is_file():

    raise FileNotFoundError(
        f"Файл mmCIF не найден: {CIF_PATH}"
    )

if not chain_sequences:

    raise ValueError(
        "Не найдены экспериментальные последовательности цепей."
    )


# Используем индивидуальную директорию текущей структуры.
if "PDB_RESULTS_DIR" in globals():

    PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

else:

    if "RESULTS_DIR" not in globals():

        raise NameError(
            "Не определена PDB_RESULTS_DIR или RESULTS_DIR."
        )

    PDB_RESULTS_DIR = Path(RESULTS_DIR) / PDB_ID


PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print(f"PDB ID: {PDB_ID}")
print(f"CIF path: {CIF_PATH}")
print(f"Results directory: {PDB_RESULTS_DIR}")
print(f"Experimental chains: {len(chain_sequences)}")


# ------------------------------------------------------------
# 2. Загрузка mmCIF
# ------------------------------------------------------------

# Загружаем словарь независимо от того,
# была ли переменная cif_dict создана ранее.

cif_dict = MMCIF2Dict(str(CIF_PATH))


# ------------------------------------------------------------
# 3. Вспомогательные функции
# ------------------------------------------------------------

def cif_values(value):

    """
    Преобразует значение MMCIF2Dict в список строк.
    """

    if value is None:

        return []

    if isinstance(value, str):

        return [value]

    return list(value)


def normalize_sequence(sequence):

    """
    Нормализация последовательности.
    """

    return (
        str(sequence)
        .replace("\n", "")
        .replace("\r", "")
        .replace(" ", "")
        .replace("\t", "")
        .upper()
    )


def clean_cif_value(value):

    """
    Нормализация значений mmCIF.
    """

    value = str(value).strip()

    if value in {".", "?", ""}:

        return None

    return value


# ------------------------------------------------------------
# 4. Загрузка референсных последовательностей
# ------------------------------------------------------------

reference_sequences = cif_values(
    cif_dict.get(
        "_entity_poly.pdbx_seq_one_letter_code_can"
    )
)

reference_sequence_source = (
    "_entity_poly.pdbx_seq_one_letter_code_can"
)

# Если канонические последовательности отсутствуют,
# используем исходные последовательности полимеров.

if not reference_sequences:

    reference_sequences = cif_values(
        cif_dict.get(
            "_entity_poly.pdbx_seq_one_letter_code"
        )
    )

    reference_sequence_source = (
        "_entity_poly.pdbx_seq_one_letter_code"
    )


entity_ids = cif_values(
    cif_dict.get("_entity_poly.entity_id")
)


if not reference_sequences:

    raise ValueError(
        "В mmCIF отсутствуют референсные последовательности."
    )


if len(entity_ids) != len(reference_sequences):

    raise ValueError(
        "Количество entity_id не совпадает "
        "с количеством референсных последовательностей."
    )


reference_records = []

for entity_id, sequence in zip(
    entity_ids,
    reference_sequences
):

    entity_id = clean_cif_value(entity_id)

    sequence = normalize_sequence(sequence)

    if entity_id is None or not sequence:

        continue

    reference_records.append({

        "entity_id": entity_id,

        "sequence": sequence,

        "sequence_length": len(sequence),

        "sequence_source": reference_sequence_source

    })


df_reference = pd.DataFrame(
    reference_records
)


if df_reference.empty:

    raise ValueError(
        "Не удалось сформировать таблицу референсных последовательностей."
    )


# Контроль уникальности entity_id.

if df_reference["entity_id"].duplicated().any():

    raise ValueError(
        "В таблице референсных последовательностей "
        "обнаружены повторяющиеся entity_id."
    )


print("\n" + "-" * 70)
print("REFERENCE SEQUENCES")
print("-" * 70)

print(f"Reference source: {reference_sequence_source}")
print(f"Reference entities: {len(df_reference)}")

display(
    df_reference[
        [
            "entity_id",
            "sequence_length",
            "sequence_source"
        ]
    ]
)


# ------------------------------------------------------------
# 5. Определение соответствия chain_id и entity_id
# ------------------------------------------------------------

# В Bio.PDB обычно используются авторские идентификаторы цепей
# auth_asym_id.
#
# В mmCIF:
# - label_asym_id связан с entity_id через _struct_asym;
# - auth_asym_id может отличаться от label_asym_id.
#
# Для связи используем таблицу атомов _atom_site.

auth_chain_to_entities = {}

atom_site_auth = cif_values(
    cif_dict.get("_atom_site.auth_asym_id")
)

atom_site_label = cif_values(
    cif_dict.get("_atom_site.label_asym_id")
)

atom_site_entity = cif_values(
    cif_dict.get("_atom_site.label_entity_id")
)

atom_site_group = cif_values(
    cif_dict.get("_atom_site.group_PDB")
)

atom_site_model = cif_values(
    cif_dict.get("_atom_site.pdbx_PDB_model_num")
)


atom_site_columns = [

    atom_site_auth,

    atom_site_label,

    atom_site_entity,

    atom_site_group,

    atom_site_model

]


available_columns = [
    len(column)
    for column in atom_site_columns
    if column
]


if available_columns and len(set(available_columns)) != 1:

    raise ValueError(
        "Количество записей в колонках _atom_site "
        "не совпадает. Невозможно надёжно сопоставить цепи."
    )


if all(atom_site_columns):

    for auth_id, label_id, entity_id, group, model in zip(
        atom_site_auth,
        atom_site_label,
        atom_site_entity,
        atom_site_group,
        atom_site_model
    ):

        auth_id = clean_cif_value(auth_id)

        entity_id = clean_cif_value(entity_id)

        group = clean_cif_value(group)

        model = clean_cif_value(model)

        # Используем полимерные атомы первого модели.
        if auth_id is None or entity_id is None:

            continue

        if group != "ATOM":

            continue

        if model not in {"1", None}:

            continue

        auth_chain_to_entities.setdefault(
            auth_id,
            set()
        ).add(entity_id)


# ------------------------------------------------------------
# 6. Резервная связь через pdbx_poly_seq_scheme
# ------------------------------------------------------------

# Если atom_site не дал соответствие,
# дополнительно проверяем таблицу полимерных последовательностей.

scheme_auth = cif_values(
    cif_dict.get(
        "_pdbx_poly_seq_scheme.pdb_strand_id"
    )
)

scheme_entity = cif_values(
    cif_dict.get(
        "_pdbx_poly_seq_scheme.entity_id"
    )
)

scheme_columns = [
    scheme_auth,
    scheme_entity
]


if all(scheme_columns):

    if len(scheme_auth) != len(scheme_entity):

        raise ValueError(
            "Некорректная длина колонок "
            "_pdbx_poly_seq_scheme."
        )

    for auth_ids, entity_id in zip(
        scheme_auth,
        scheme_entity
    ):

        entity_id = clean_cif_value(entity_id)

        if entity_id is None:

            continue

        # pdb_strand_id может содержать несколько
        # авторских идентификаторов через запятую.
        chain_ids = [
            item.strip()
            for item in str(auth_ids).split(",")
            if item.strip()
        ]

        for chain_id in chain_ids:

            auth_chain_to_entities.setdefault(
                chain_id,
                set()
            ).add(entity_id)


# ------------------------------------------------------------
# 7. Проверка карты цепей
# ------------------------------------------------------------

chain_entity_mapping = {}

for chain_id in chain_sequences:

    chain_id = str(chain_id)

    candidate_entities = sorted(
        auth_chain_to_entities.get(
            chain_id,
            set()
        )
    )

    # Оставляем только entity_id, для которых
    # действительно загружена референсная последовательность.
    known_entities = set(
        df_reference["entity_id"]
    )

    candidate_entities = [
        entity_id
        for entity_id in candidate_entities
        if entity_id in known_entities
    ]

    if len(candidate_entities) == 1:

        chain_entity_mapping[chain_id] = {

            "entity_id": candidate_entities[0],

            "mapping_status": "Direct",

            "candidate_entities": candidate_entities

        }

    elif len(candidate_entities) > 1:

        chain_entity_mapping[chain_id] = {

            "entity_id": None,

            "mapping_status": "Ambiguous",

            "candidate_entities": candidate_entities

        }

    else:

        chain_entity_mapping[chain_id] = {

            "entity_id": None,

            "mapping_status": "Not found",

            "candidate_entities": []

        }


mapping_records = []

for chain_id, mapping in chain_entity_mapping.items():

    mapping_records.append({

        "chain_id": chain_id,

        "entity_id": mapping["entity_id"],

        "mapping_status": mapping["mapping_status"],

        "candidate_entities": ";".join(
            mapping["candidate_entities"]
        )

    })


df_chain_mapping = pd.DataFrame(
    mapping_records
)


print("\n" + "-" * 70)
print("CHAIN TO ENTITY MAPPING")
print("-" * 70)

display(df_chain_mapping)


# ------------------------------------------------------------
# 8. Настройка глобального выравнивания
# ------------------------------------------------------------

aligner = PairwiseAligner()

aligner.mode = "global"

aligner.match_score = 2.0

aligner.mismatch_score = -1.0

aligner.open_gap_score = -3.0

aligner.extend_gap_score = -0.5


# ------------------------------------------------------------
# 9. Функция анализа выравнивания
# ------------------------------------------------------------

def calculate_alignment_metrics(
    reference_sequence,
    experimental_sequence,
    alignment
):

    """
    Рассчитывает метрики глобального выравнивания.

    Identity:
        Доля совпадений среди позиций, где обе
        последовательности содержат остатки.

    Reference coverage:
        Доля референсных остатков, участвующих
        в сопоставлении.

    Experimental coverage:
        Доля экспериментальных остатков, участвующих
        в сопоставлении.
    """

    indices = alignment.indices

    reference_indices = indices[0]

    experimental_indices = indices[1]


    paired_positions = (

        (reference_indices >= 0)
        &
        (experimental_indices >= 0)

    )


    aligned_count = int(
        paired_positions.sum()
    )


    if aligned_count == 0:

        raise ValueError(
            "Выравнивание не содержит сопоставленных позиций."
        )


    paired_reference_indices = (
        reference_indices[paired_positions]
    )

    paired_experimental_indices = (
        experimental_indices[paired_positions]
    )


    matches = sum(

        reference_sequence[int(ref_idx)]
        == experimental_sequence[int(exp_idx)]

        for ref_idx, exp_idx in zip(
            paired_reference_indices,
            paired_experimental_indices
        )

    )


    mismatches = aligned_count - matches


    identity = (
        matches / aligned_count * 100
    )


    reference_coverage = (

        len(np.unique(paired_reference_indices))
        / len(reference_sequence)
        * 100

    )


    experimental_coverage = (

        len(np.unique(paired_experimental_indices))
        / len(experimental_sequence)
        * 100

    )


    # Gap в экспериментальной последовательности:
    # позиция референса не имеет соответствующей
    # позиции экспериментальной последовательности.

    missing_reference = int(

        np.sum(
            (reference_indices >= 0)
            &
            (experimental_indices < 0)
        )

    )


    # Gap в референсной последовательности:
    # позиция экспериментальной последовательности
    # не имеет соответствующей позиции референса.

    extra_experimental = int(

        np.sum(
            (reference_indices < 0)
            &
            (experimental_indices >= 0)
        )

    )


    reference_self_score = aligner.score(
        reference_sequence,
        reference_sequence
    )

    experimental_self_score = aligner.score(
        experimental_sequence,
        experimental_sequence
    )


    normalization_score = max(
        reference_self_score,
        experimental_self_score
    )


    normalized_score = (

        alignment.score / normalization_score

        if normalization_score > 0

        else 0

    )


    return {

        "alignment_score": float(alignment.score),

        "normalized_score": float(normalized_score),

        "matches": int(matches),

        "mismatches": int(mismatches),

        "aligned_positions": aligned_count,

        "identity_percent": float(identity),

        "reference_coverage_percent": float(
            reference_coverage
        ),

        "experimental_coverage_percent": float(
            experimental_coverage
        ),

        "missing_reference_residues": missing_reference,

        "extra_experimental_residues": extra_experimental

    }


# ------------------------------------------------------------
# 10. Сравнение экспериментальных и референсных последовательностей
# ------------------------------------------------------------

comparison_records = []

best_alignments = {}

best_alignment_details = {}


for chain_id, experimental_sequence in chain_sequences.items():

    chain_id = str(chain_id)

    experimental_sequence = normalize_sequence(
        experimental_sequence
    )

    if not experimental_sequence:

        print(
            f"WARNING: Chain {chain_id} has an empty sequence."
        )

        continue


    mapping = chain_entity_mapping[chain_id]

    mapped_entity = mapping["entity_id"]

    mapping_status = mapping["mapping_status"]


    # Если entity_id установлен однозначно,
    # сравниваем только с соответствующей сущностью.
    #
    # Если связь отсутствует или неоднозначна,
    # выполняем резервный поиск по всем сущностям.

    if mapped_entity is not None:

        candidate_references = df_reference[
            df_reference["entity_id"] == mapped_entity
        ]

        comparison_method = "Mapped entity"

    else:

        candidate_references = df_reference

        comparison_method = "Fallback search"


    best_result = None

    best_alignment = None

    best_reference_sequence = None

    best_rank = None


    print(
        f"\nProcessing chain {chain_id}: "
        f"{len(experimental_sequence)} residues"
    )

    print(
        f"Mapping status: {mapping_status}"
    )

    print(
        f"Comparison method: {comparison_method}"
    )


    for _, reference in candidate_references.iterrows():

        reference_sequence = reference["sequence"]

        if not reference_sequence:

            continue


        alignments = aligner.align(
            reference_sequence,
            experimental_sequence
        )

        alignment = alignments[0]


        metrics = calculate_alignment_metrics(
            reference_sequence,
            experimental_sequence,
            alignment
        )


        result_rank = (

            metrics["normalized_score"],

            metrics["identity_percent"],

            min(
                metrics["reference_coverage_percent"],
                metrics["experimental_coverage_percent"]
            )

        )


        if best_result is None or result_rank > best_rank:

            best_rank = result_rank

            best_alignment = alignment

            best_reference_sequence = reference_sequence

            best_result = {

                "chain_id": chain_id,

                "entity_id": reference["entity_id"],

                "reference_length": len(reference_sequence),

                "experimental_length": len(experimental_sequence),

                **metrics

            }


    if best_result is None:

        print(
            f"WARNING: No reference found for chain {chain_id}."
        )

        continue


    best_result["mapping_status"] = mapping_status

    best_result["comparison_method"] = comparison_method

    best_result["candidate_entities"] = (
        mapping["candidate_entities"]
    )


    comparison_records.append({

        "chain_id": chain_id,

        "entity_id": best_result["entity_id"],

        "mapping_status": mapping_status,

        "comparison_method": comparison_method,

        "reference_length": best_result["reference_length"],

        "experimental_length": best_result["experimental_length"],

        "alignment_score": best_result["alignment_score"],

        "normalized_score": round(
            best_result["normalized_score"],
            4
        ),

        "matches": best_result["matches"],

        "mismatches": best_result["mismatches"],

        "identity_percent": round(
            best_result["identity_percent"],
            2
        ),

        "reference_coverage_percent": round(
            best_result["reference_coverage_percent"],
            2
        ),

        "experimental_coverage_percent": round(
            best_result["experimental_coverage_percent"],
            2
        ),

        "aligned_positions": best_result["aligned_positions"],

        "missing_reference_residues": (
            best_result["missing_reference_residues"]
        ),

        "extra_experimental_residues": (
            best_result["extra_experimental_residues"]
        )

    })


    best_alignments[chain_id] = best_alignment

    best_alignment_details[chain_id] = {

        "entity_id": best_result["entity_id"],

        "reference_sequence": best_reference_sequence,

        "experimental_sequence": experimental_sequence,

        "alignment": best_alignment,

        "metrics": best_result

    }


# ------------------------------------------------------------
# 11. Итоговая таблица сравнения
# ------------------------------------------------------------

df_reference_sequence_comparison = pd.DataFrame(
    comparison_records
)


if df_reference_sequence_comparison.empty:

    raise ValueError(
        "Не удалось выполнить сравнение ни для одной цепи."
    )


print("\n" + "-" * 70)
print("REFERENCE SEQUENCE COMPARISON")
print("-" * 70)

display(
    df_reference_sequence_comparison
)


# ------------------------------------------------------------
# 12. Проверка полноты сравнения
# ------------------------------------------------------------

experimental_chain_ids = set(
    map(str, chain_sequences.keys())
)

compared_chain_ids = set(
    df_reference_sequence_comparison["chain_id"]
)

not_compared_chains = sorted(
    experimental_chain_ids - compared_chain_ids
)

print("\n" + "-" * 70)
print("CHAIN VALIDATION")
print("-" * 70)

print(
    f"Experimental chains: {len(experimental_chain_ids)}"
)

print(
    f"Compared chains: {len(compared_chain_ids)}"
)

print(
    f"Not compared chains: {len(not_compared_chains)}"
)

if not_compared_chains:

    print(
        "WARNING: Some experimental chains were not compared:",
        not_compared_chains
    )

else:

    print(
        "All experimental chains were compared."
    )


# ------------------------------------------------------------
# 13. Сохранение референсных последовательностей
# ------------------------------------------------------------

REFERENCE_SEQUENCE_CSV_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_reference_sequences.csv"

)

df_reference.to_csv(
    REFERENCE_SEQUENCE_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 14. Сохранение карты chain_id -> entity_id
# ------------------------------------------------------------

CHAIN_MAPPING_CSV_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_chain_entity_mapping.csv"

)

df_chain_mapping.to_csv(
    CHAIN_MAPPING_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 15. Сохранение результатов сравнения
# ------------------------------------------------------------

SEQUENCE_COMPARISON_CSV_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_reference_sequence_comparison.csv"

)

df_reference_sequence_comparison.to_csv(
    SEQUENCE_COMPARISON_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 16. Сохранение текстовых выравниваний
# ------------------------------------------------------------

ALIGNMENTS_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_sequence_alignments.txt"

)

with open(
    ALIGNMENTS_PATH,
    "w",
    encoding="utf-8"
) as file:

    file.write(
        f"REFERENCE SEQUENCE ALIGNMENTS: {PDB_ID}\n"
    )

    file.write("=" * 80 + "\n\n")


    for chain_id, details in best_alignment_details.items():

        metrics = details["metrics"]

        file.write(
            f"CHAIN: {chain_id}\n"
        )

        file.write(
            f"ENTITY ID: {details['entity_id']}\n"
        )

        file.write(
            f"REFERENCE LENGTH: "
            f"{len(details['reference_sequence'])}\n"
        )

        file.write(
            f"EXPERIMENTAL LENGTH: "
            f"{len(details['experimental_sequence'])}\n"
        )

        file.write(
            f"IDENTITY: "
            f"{metrics['identity_percent']:.2f}%\n"
        )

        file.write(
            f"REFERENCE COVERAGE: "
            f"{metrics['reference_coverage_percent']:.2f}%\n"
        )

        file.write(
            f"EXPERIMENTAL COVERAGE: "
            f"{metrics['experimental_coverage_percent']:.2f}%\n\n"
        )

        file.write(
            str(details["alignment"])
        )

        file.write("\n\n")

        file.write("-" * 80 + "\n\n")


# ------------------------------------------------------------
# 17. Интерпретация результатов
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("INTERPRETATION")
print("-" * 70)


for _, row in df_reference_sequence_comparison.iterrows():

    print(
        f"\nChain: {row['chain_id']}"
    )

    print(
        f"Reference entity: {row['entity_id']}"
    )

    print(
        f"Mapping method: {row['comparison_method']}"
    )

    print(
        f"Identity: {row['identity_percent']:.2f}%"
    )

    print(
        f"Reference coverage: "
        f"{row['reference_coverage_percent']:.2f}%"
    )

    print(
        f"Experimental coverage: "
        f"{row['experimental_coverage_percent']:.2f}%"
    )

    print(
        f"Aligned positions: {row['aligned_positions']}"
    )

    print(
        f"Missing reference positions in alignment: "
        f"{row['missing_reference_residues']}"
    )

    print(
        f"Extra experimental positions in alignment: "
        f"{row['extra_experimental_residues']}"
    )


    if (

        row["identity_percent"] >= 99

        and row["reference_coverage_percent"] >= 99

        and row["experimental_coverage_percent"] >= 99

    ):

        print(
            "Almost complete sequence correspondence."
        )

    elif row["identity_percent"] >= 95:

        print(
            "High sequence identity, but differences "
            "or coverage gaps are present."
        )

    else:

        print(
            "Noticeable sequence differences. "
            "The alignment should be examined."
        )


# ------------------------------------------------------------
# 18. Сохранение JSON-сводки
# ------------------------------------------------------------

SEQUENCE_VALIDATION_JSON_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_sequence_validation.json"

)


sequence_validation_summary = {

    "PDB_ID": PDB_ID,

    "reference_sequence_source": reference_sequence_source,

    "number_of_reference_sequences": int(
        len(df_reference)
    ),

    "number_of_experimental_chains": int(
        len(experimental_chain_ids)
    ),

    "number_of_compared_chains": int(
        len(compared_chain_ids)
    ),

    "number_of_uncompared_chains": int(
        len(not_compared_chains)
    ),

    "not_compared_chains": not_compared_chains,

    "chain_entity_mapping": df_chain_mapping.to_dict(
        orient="records"
    ),

    "results": df_reference_sequence_comparison.to_dict(
        orient="records"
    ),

    "interpretation": {

        "identity": (
            "Percentage of identical residues among "
            "aligned residue pairs."
        ),

        "reference_coverage": (
            "Percentage of reference residues participating "
            "in the alignment."
        ),

        "experimental_coverage": (
            "Percentage of experimental residues participating "
            "in the alignment."
        ),

        "gap_interpretation": (
            "Alignment gaps do not independently prove "
            "absence of residues from experimental coordinates."
        )

    }

}


# Преобразование NumPy и pandas в JSON-совместимые типы.

def make_json_serializable(value):

    if value is None:

        return None

    if isinstance(value, dict):

        return {
            str(key): make_json_serializable(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple, set)):

        return [
            make_json_serializable(item)
            for item in value
        ]

    if isinstance(value, np.ndarray):

        return make_json_serializable(
            value.tolist()
        )

    if isinstance(value, np.generic):

        return make_json_serializable(
            value.item()
        )

    if isinstance(value, float):

        return float(value) if np.isfinite(value) else None

    if isinstance(value, Path):

        return str(value)

    if isinstance(value, pd.Timestamp):

        return value.isoformat()

    if value is pd.NA or value is pd.NaT:

        return None

    return value


with open(
    SEQUENCE_VALIDATION_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        make_json_serializable(sequence_validation_summary),
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


# ------------------------------------------------------------
# 19. Проверка сохранённых файлов
# ------------------------------------------------------------

output_paths = [

    REFERENCE_SEQUENCE_CSV_PATH,

    CHAIN_MAPPING_CSV_PATH,

    SEQUENCE_COMPARISON_CSV_PATH,

    ALIGNMENTS_PATH,

    SEQUENCE_VALIDATION_JSON_PATH

]


print("\n" + "-" * 70)
print("SAVED FILES")
print("-" * 70)


for path in output_paths:

    print(
        f"{path.name}: "
        f"{'OK' if path.is_file() else 'ERROR'}"
    )


if not all(
    path.is_file()
    for path in output_paths
):

    raise RuntimeError(
        "Не все результаты проверки последовательностей сохранены."
    )


# ------------------------------------------------------------
# 20. Итог
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("SEQUENCE VALIDATION SUMMARY")
print("-" * 70)

print(f"PDB ID: {PDB_ID}")

print(
    f"Reference entities: {len(df_reference)}"
)

print(
    f"Experimental chains: {len(experimental_chain_ids)}"
)

print(
    f"Compared chains: {len(compared_chain_ids)}"
)

print(
    f"Uncompared chains: {len(not_compared_chains)}"
)

print(
    f"Results directory: {PDB_RESULTS_DIR}"
)

print("\nСтатус: OK")

print("=" * 70)
print("CELL 13 COMPLETED")
print("=" * 70)
```

Что анализируем: экспериментальная последовательность может быть короче референсной, потому что некоторые участки белка не удалось определить в экспериментальной структуре.

Важно: приведённое выравнивание показывает сходство и длины последовательностей, но для точного подсчёта пропущенных остатков нужно учитывать соответствие цепей и нумерацию остатков. Это сделаем в следующей ячейке.
Что теперь будет в результатах
Показатель	Значение
reference_length	Длина референсной последовательности
experimental_length	Длина экспериментальной последовательности
alignment_score	Оценка качества выравнивания
identity_percent	Процент совпадения аминокислот
reference_coverage_percent	Процент покрытия референсной последовательности
experimental_coverage_percent	Процент покрытия экспериментальной последовательности
aligned_positions	Количество сопоставленных позиций
missing_reference_residues	Количество позиций референса без соответствующего остатка в экспериментальной последовательности
extra_experimental_residues	Количество экспериментальных позиций без соответствия в референсе

Важное замечание
Здесь сравниваются две последовательности:
- референсная — полная последовательность белка из аннотации PDB;
- экспериментальная — последовательность, извлечённая из атомной структуры.
Если часть белка не видна в экспериментальной структуре, она может отсутствовать в экспериментальной последовательности. Это не обязательно означает, что белок был повреждён: такие участки могут быть просто не разрешены экспериментальным методом.

Что добавилось
Показатель	Назначение
Пропущенные остатки	Выявление неразрешённых участков белка
Пропущенные атомы	Выявление неполного атомного состава остатков
Альтернативные положения	Определение атомов с несколькими координатными вариантами
Occupancy	Оценка заполненности атомных позиций
Распределение occupancy	Визуальная оценка заполненности
Статистика по цепям	Сравнение качества представления разных цепей
Общая сводка	Подготовка результатов для итогового отчёта

Важные научные уточнения
1. Пропущенные остатки и пропущенные атомы — не одно и то же.

Пропущенный остаток означает, что целый аминокислотный остаток не представлен в координатной модели. Пропущенные атомы означают, что остаток присутствует, но часть его атомов не разрешена.

2. Occupancy меньше 1 не обязательно означает плохое качество структуры.

Это может отражать частичную заселённость позиции или несколько конформаций молекулы.

3. Альтернативные положения нельзя считать независимыми атомами.

Они представляют несколько возможных координатных вариантов одного атома или группы атомов.

4. Количество пропущенных остатков не следует автоматически делить на количество наблюдаемых остатков.

Для корректного процента полноты нужен полный референсный состав каждой цепи и корректное сопоставление идентификаторов остатков. Результаты ячейки 13 могут стать основой для такого расчёта, но простого вычитания количеств недостаточно.

5. В этой ячейке df_atoms должен содержать атомы белка.

Если таблица содержит воду и другие небелковые компоненты, статистика occupancy и altloc будет относиться ко всему содержимому таблицы, а не только к белку.

```python
# ============================================================
# CELL 15. SEPARATE PROTEIN CHAINS AND HETERO COMPONENTS
# ============================================================
# Назначение:
# Разделение результатов структурного анализа на:
# - белковые атомы;
# - белковые остатки;
# - небелковые компоненты;
# - молекулы воды;
# - отдельные PDB-файлы белковых цепей.
#
# ВАЖНО:
# 1. Используется первая модель структуры.
# 2. df_atoms и df_residues уже содержат белковые данные
#    из CELL 6 и CELL 7. Повторная фильтрация не нужна.
# 3. Все альтернативные положения атомов отдельно
#    анализируются в CELL 14.
# 4. Небелковые компоненты не считаются автоматически
#    лекарственными лигандами.
# ============================================================

from Bio.PDB import PDBIO, Select
from Bio.PDB.Polypeptide import is_aa

import pandas as pd
import numpy as np

from pathlib import Path
import json
import re


print("=" * 70)
print("CELL 15. SEPARATE PROTEIN CHAINS AND HETERO COMPONENTS")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка входных данных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "structure",
    "PDB_RESULTS_DIR",
    "df_atoms",
    "df_residues",
    "df_hetero",
    "df_water"
]

for variable in required_variables:

    if variable not in globals():

        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()

PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


if not isinstance(df_atoms, pd.DataFrame):
    raise TypeError("df_atoms должен быть pandas DataFrame.")

if not isinstance(df_residues, pd.DataFrame):
    raise TypeError("df_residues должен быть pandas DataFrame.")

if not isinstance(df_hetero, pd.DataFrame):
    raise TypeError("df_hetero должен быть pandas DataFrame.")

if not isinstance(df_water, pd.DataFrame):
    raise TypeError("df_water должен быть pandas DataFrame.")


if df_atoms.empty:
    raise ValueError("Таблица df_atoms не содержит белковых атомов.")

if df_residues.empty:
    raise ValueError("Таблица df_residues не содержит белковых остатков.")


required_atom_columns = {
    "pdb_id",
    "chain_id",
    "residue_key",
    "atom_name"
}

missing_atom_columns = (
    required_atom_columns - set(df_atoms.columns)
)

if missing_atom_columns:

    raise KeyError(
        "В df_atoms отсутствуют необходимые столбцы: "
        f"{sorted(missing_atom_columns)}"
    )


required_residue_columns = {
    "pdb_id",
    "chain_id"
}

missing_residue_columns = (
    required_residue_columns - set(df_residues.columns)
)

if missing_residue_columns:

    raise KeyError(
        "В df_residues отсутствуют необходимые столбцы: "
        f"{sorted(missing_residue_columns)}"
    )


# Проверяем принадлежность таблиц текущей структуре.

for table_name, table in [
    ("df_atoms", df_atoms),
    ("df_residues", df_residues)
]:

    pdb_ids = (
        table["pdb_id"]
        .dropna()
        .astype(str)
        .str.upper()
        .unique()
    )

    if len(pdb_ids) > 0 and set(pdb_ids) != {PDB_ID}:

        raise ValueError(
            f"{table_name} содержит данные других PDB ID: "
            f"{sorted(pdb_ids)}"
        )


print(f"PDB ID: {PDB_ID}")
print(f"Results directory: {PDB_RESULTS_DIR}")


# ------------------------------------------------------------
# 2. Выбор модели для анализа
# ------------------------------------------------------------

models = list(structure.get_models())

if not models:

    raise ValueError(
        "В структуре отсутствуют модели."
    )


first_model = models[0]

MODEL_INDEX = 1

print("\nMODEL INFORMATION")

print(
    f"Total models in structure: {len(models)}"
)

print(
    f"Selected model ID: {first_model.id}"
)

print(
    f"Selected model index: {MODEL_INDEX}"
)

print(
    "Analysis scope: first model only"
)


# ------------------------------------------------------------
# 3. Вспомогательные функции
# ------------------------------------------------------------

def normalize_chain_id(value):

    """
    Нормализует идентификатор цепи.

    Пустая цепь представляется пустой строкой.
    """

    if pd.isna(value):
        return ""

    return str(value).strip()


def display_chain_id(value):

    """
    Формирует читаемое обозначение цепи.
    """

    normalized = normalize_chain_id(value)

    return normalized if normalized else "<blank>"


def safe_filename(value):

    """
    Создаёт безопасное имя файла.
    """

    value = str(value).strip()

    if not value:
        return "blank"

    return re.sub(
        r"[^A-Za-z0-9_-]",
        "_",
        value
    )


# ------------------------------------------------------------
# 4. Определение белковых цепей первой модели
# ------------------------------------------------------------

protein_chain_ids = set()

protein_residue_records = []

structure_atom_count = 0

for chain in first_model:

    chain_id = normalize_chain_id(chain.id)

    chain_residue_count = 0

    for residue in chain:

        if not is_aa(
            residue,
            standard=False
        ):
            continue

        protein_chain_ids.add(chain_id)

        chain_residue_count += 1

        # Используем get_atoms(), а не get_unpacked_list().
        # Это соответствует представлению атомов в CELL 6–7
        # и не разворачивает альтернативные положения повторно.

        residue_atoms = list(
            residue.get_atoms()
        )

        structure_atom_count += len(
            residue_atoms
        )

        protein_residue_records.append({

            "pdb_id": PDB_ID,

            "model_index": MODEL_INDEX,

            "model_id": first_model.id,

            "chain_id": chain_id,

            "chain_id_display": display_chain_id(
                chain_id
            ),

            "hetero_flag": str(
                residue.id[0]
            ).strip(),

            "residue_id": int(
                residue.id[1]
            ),

            "insertion_code": str(
                residue.id[2]
            ).strip(),

            "residue_name": residue.resname.strip(),

            "atom_count": len(residue_atoms)

        })


df_structure_protein_residues = pd.DataFrame(
    protein_residue_records
)


if not protein_chain_ids:

    raise ValueError(
        "В первой модели не обнаружены белковые цепи."
    )


print("\n" + "-" * 70)
print("PROTEIN STRUCTURE")
print("-" * 70)

print(
    f"Protein chains: {len(protein_chain_ids)}"
)

print(
    f"Protein residues: {len(df_structure_protein_residues)}"
)

print(
    f"Protein atoms in first model: {structure_atom_count}"
)

print(
    f"Chain IDs: "
    f"{[display_chain_id(x) for x in sorted(protein_chain_ids)]}"
)


# ------------------------------------------------------------
# 5. Формирование отдельных таблиц
# ------------------------------------------------------------

# CELL 6 и CELL 7 уже выделили белковые остатки и атомы.
# Поэтому используем их напрямую, без повторной фильтрации.

df_protein_atoms = df_atoms.copy()

df_protein_residues = df_residues.copy()


# Небелковые компоненты.
# Сюда могут входить:
# HEM, NDP, ACT, ионы металлов, кофакторы и другие компоненты.

df_hetero_components = df_hetero.copy()

# Предварительные кандидаты для последующей классификации.
# Это НЕ подтверждённый набор лекарственных лигандов.

df_ligand_candidates = df_hetero_components.copy()

# Вода сохраняется отдельно.

df_water_components = df_water.copy()


# ------------------------------------------------------------
# 6. Проверка согласованности таблиц и структуры
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("DATAFRAME CONSISTENCY CHECK")
print("-" * 70)


# Сравнение числа атомов.
# В CELL 7 используется тот же принцип выбора altloc,
# что и в get_atoms().

cell7_atom_count = len(df_protein_atoms)

atom_count_match = (
    cell7_atom_count == structure_atom_count
)

print(
    f"Atoms in structure: {structure_atom_count}"
)

print(
    f"Atoms in CELL 7: {cell7_atom_count}"
)

print(
    f"Atom count matches: {atom_count_match}"
)


# Сравнение числа остатков.

structure_residue_count = len(
    df_structure_protein_residues
)

cell6_residue_count = len(
    df_protein_residues
)

residue_count_match = (
    structure_residue_count == cell6_residue_count
)

print(
    f"Residues in structure: {structure_residue_count}"
)

print(
    f"Residues in CELL 6: {cell6_residue_count}"
)

print(
    f"Residue count matches: {residue_count_match}"
)


# Проверка цепей.

structure_chain_ids = set(
    protein_chain_ids
)

cell7_chain_ids = set(
    df_protein_atoms["chain_id"]
    .dropna()
    .map(normalize_chain_id)
)

chain_ids_match = (
    structure_chain_ids == cell7_chain_ids
)

print(
    f"Chain IDs in structure: "
    f"{sorted(structure_chain_ids)}"
)

print(
    f"Chain IDs in CELL 7: "
    f"{sorted(cell7_chain_ids)}"
)

print(
    f"Chain IDs match: {chain_ids_match}"
)


if not atom_count_match:

    raise RuntimeError(
        "Количество белковых атомов в структуре "
        "не совпадает с CELL 7. "
        "Проверьте обработку альтернативных положений."
    )


if not residue_count_match:

    raise RuntimeError(
        "Количество белковых остатков в структуре "
        "не совпадает с CELL 6."
    )


if not chain_ids_match:

    raise RuntimeError(
        "Наборы белковых цепей не совпадают "
        "между структурой и CELL 7."
    )


# ------------------------------------------------------------
# 7. Сводка небелковых компонентов
# ------------------------------------------------------------

# Не предполагаем заранее, что в df_hetero обязательно
# присутствуют все столбцы классификации.

df_component_summary = pd.DataFrame(
    columns=[
        "residue_name",
        "component_type",
        "occurrence_count"
    ]
)


if not df_hetero_components.empty:

    residue_name_col = next(

        (
            column
            for column in [
                "residue_name",
                "resname",
                "component_name"
            ]
            if column in df_hetero_components.columns
        ),

        None

    )

    component_type_col = next(

        (
            column
            for column in [
                "component_type",
                "classification",
                "component_class"
            ]
            if column in df_hetero_components.columns
        ),

        None

    )

    if residue_name_col is not None:

        grouping_columns = [residue_name_col]

        if component_type_col is not None:

            grouping_columns.append(
                component_type_col
            )

        df_component_summary = (

            df_hetero_components
            .groupby(
                grouping_columns,
                dropna=False
            )
            .size()
            .reset_index(name="occurrence_count")

        )

        df_component_summary = (
            df_component_summary.rename(
                columns={
                    residue_name_col: "residue_name",
                    component_type_col: "component_type"
                    if component_type_col is not None
                    else residue_name_col
                }
            )
        )

        if component_type_col is None:

            df_component_summary["component_type"] = (
                "Not classified"
            )

        df_component_summary = (
            df_component_summary[
                [
                    "residue_name",
                    "component_type",
                    "occurrence_count"
                ]
            ]
        )


print("\n" + "-" * 70)
print("HETERO COMPONENTS")
print("-" * 70)

print(
    f"Total hetero component records: "
    f"{len(df_hetero_components)}"
)

if not df_component_summary.empty:

    display(df_component_summary)

else:

    print(
        "No component classification available "
        "or no hetero components were found."
    )


# ------------------------------------------------------------
# 8. Сохранение отдельных PDB-файлов белковых цепей
# ------------------------------------------------------------

class ProteinChainSelect(Select):

    """
    Сохраняет одну белковую цепь выбранной модели.

    Нестандартные аминокислоты сохраняются,
    если Bio.PDB распознаёт их через is_aa().
    """

    def __init__(
        self,
        chain_id,
        model_id
    ):

        self.chain_id = chain_id

        self.model_id = model_id


    def accept_model(self, model):

        return model.id == self.model_id


    def accept_chain(self, chain):

        return (
            normalize_chain_id(chain.id)
            == self.chain_id
        )


    def accept_residue(self, residue):

        return is_aa(
            residue,
            standard=False
        )


CHAIN_DIR = (
    PDB_RESULTS_DIR / "protein_chains"
)

CHAIN_DIR.mkdir(
    parents=True,
    exist_ok=True
)


io = PDBIO()

io.set_structure(structure)


saved_chain_files = []

chain_file_records = []

for chain_index, chain_id in enumerate(
    sorted(protein_chain_ids),
    start=1
):

    safe_chain = safe_filename(chain_id)

    output_name = (
        f"{PDB_ID}_chain_{chain_index:02d}_{safe_chain}.pdb"
    )

    output_path = (
        CHAIN_DIR / output_name
    )

    io.save(

        str(output_path),

        ProteinChainSelect(
            chain_id,
            first_model.id
        )

    )

    if not output_path.is_file():

        raise FileNotFoundError(
            f"Файл цепи не сохранён: {output_path}"
        )

    if output_path.stat().st_size == 0:

        raise RuntimeError(
            f"Файл цепи пустой: {output_path}"
        )

    saved_chain_files.append(
        output_path
    )

    chain_file_records.append({

        "pdb_id": PDB_ID,

        "model_index": MODEL_INDEX,

        "chain_id": chain_id,

        "chain_id_display": display_chain_id(
            chain_id
        ),

        "file_name": output_path.name,

        "file_size_bytes": output_path.stat().st_size

    })

    print(
        f"Saved chain {display_chain_id(chain_id)}: "
        f"{output_path.name}"
    )


df_chain_files = pd.DataFrame(
    chain_file_records
)


# ------------------------------------------------------------
# 9. Проверка сохранённых цепей
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("CHAIN FILE VALIDATION")
print("-" * 70)

if len(saved_chain_files) != len(protein_chain_ids):

    raise RuntimeError(
        "Количество сохранённых PDB-файлов "
        "не совпадает с количеством белковых цепей."
    )


for path in saved_chain_files:

    if not path.is_file() or path.stat().st_size == 0:

        raise RuntimeError(
            f"Некорректный файл цепи: {path}"
        )

    print(
        f"{path.name}: "
        f"{path.stat().st_size:,} bytes"
    )


# ------------------------------------------------------------
# 10. Сохранение отдельных таблиц
# ------------------------------------------------------------

output_tables = {

    "separated_protein_atoms": df_protein_atoms,

    "separated_protein_residues": df_protein_residues,

    "separated_hetero_components": df_hetero_components,

    "separated_water_components": df_water_components,

    "separated_structure_protein_residues": (
        df_structure_protein_residues
    ),

    "separated_component_summary": (
        df_component_summary
    ),

    "separated_chain_files": df_chain_files

}


saved_table_files = {}

for name, table in output_tables.items():

    output_path = (

        PDB_RESULTS_DIR
        / f"{PDB_ID}_{name}.csv"

    )

    table.to_csv(

        output_path,

        index=False,

        encoding="utf-8-sig"

    )

    saved_table_files[name] = output_path

    print(
        f"Saved: {output_path.name}"
    )


# ------------------------------------------------------------
# 11. Сохранение JSON-сводки
# ------------------------------------------------------------

SEPARATION_JSON_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_component_separation.json"

)


separation_summary = {

    "PDB_ID": PDB_ID,

    "analysis_scope": {

        "selected_model_index": MODEL_INDEX,

        "selected_model_id": first_model.id,

        "total_models_in_structure": len(models),

        "note": (
            "Protein chain separation uses the first model. "
            "This is consistent with CELL 4-11."
        )

    },

    "protein": {

        "number_of_protein_chains": len(
            protein_chain_ids
        ),

        "protein_chain_ids": sorted(
            protein_chain_ids
        ),

        "number_of_protein_residues": len(
            df_protein_residues
        ),

        "number_of_protein_atoms": len(
            df_protein_atoms
        )

    },

    "hetero_components": {

        "number_of_component_records": len(
            df_hetero_components
        ),

        "classification_available": (
            not df_component_summary.empty
        ),

        "note": (
            "Hetero components are not automatically "
            "classified as drug-like ligands."
        )

    },

    "water": {

        "number_of_water_component_records": len(
            df_water_components
        )

    },

    "validation": {

        "atom_count_matches_structure": bool(
            atom_count_match
        ),

        "residue_count_matches_structure": bool(
            residue_count_match
        ),

        "chain_ids_match_structure": bool(
            chain_ids_match
        ),

        "number_of_saved_chain_files": len(
            saved_chain_files
        )

    },

    "saved_chain_files": [

        path.name
        for path in saved_chain_files

    ],

    "component_types": (

        df_component_summary
        .to_dict(orient="records")

    )

}


with open(
    SEPARATION_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        separation_summary,
        file,
        ensure_ascii=False,
        indent=4,
        default=str,
        allow_nan=False
    )


# ------------------------------------------------------------
# 12. Итоговая статистика
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("COMPONENT SEPARATION SUMMARY")
print("=" * 70)

print(
    f"Protein chains: {len(protein_chain_ids)}"
)

print(
    f"Protein residues: {len(df_protein_residues)}"
)

print(
    f"Protein atoms: {len(df_protein_atoms)}"
)

print(
    f"Hetero components: {len(df_hetero_components)}"
)

print(
    f"Water components: {len(df_water_components)}"
)

print(
    f"Saved chain files: {len(saved_chain_files)}"
)

print(
    f"Output directory: {PDB_RESULTS_DIR}"
)


# ------------------------------------------------------------
# 13. Проверка сохранённых результатов
# ------------------------------------------------------------

all_output_files = (

    list(saved_table_files.values())

    + saved_chain_files

    + [SEPARATION_JSON_PATH]

)


print("\n" + "-" * 70)
print("OUTPUT FILE VERIFICATION")
print("-" * 70)

for path in all_output_files:

    exists = (
        path.is_file()
        and path.stat().st_size > 0
    )

    print(
        f"{path.name}: "
        f"{'OK' if exists else 'ERROR'}"
    )


if not all(

    path.is_file()
    and path.stat().st_size > 0

    for path in all_output_files

):

    raise RuntimeError(
        "Не все результаты разделения структуры сохранены."
    )


# ------------------------------------------------------------
# 14. Итог
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("CELL 15 FINAL STATUS")
print("-" * 70)

print(
    f"Atom count validation: "
    f"{'OK' if atom_count_match else 'ERROR'}"
)

print(
    f"Residue count validation: "
    f"{'OK' if residue_count_match else 'ERROR'}"
)

print(
    f"Chain validation: "
    f"{'OK' if chain_ids_match else 'ERROR'}"
)

print(
    f"Saved output files: {len(all_output_files)}"
)

print(
    f"Results directory: {PDB_RESULTS_DIR}"
)

print("\nСтатус: OK")

print("=" * 70)
print("CELL 15 COMPLETED")
print("=" * 70)
```

Что изменилось по сравнению с исходным кодом
Элемент	Исходная версия	Новая версия
Белковые цепи	Используются ключи chain_sequences	Определяются непосредственно по structure
Белковые остатки	Копируется df_residues	Фильтруются по белковым остаткам структуры
Белковые атомы	Копируется df_atoms	Фильтруются по белковым остаткам структуры
Гетерокомпоненты	Копируются из df_hetero	Без изменений
Вода	Копируется из df_water	Без изменений
PDB-файлы	Сохраняются цепи из chain_sequences	Сохраняются все обнаруженные белковые цепи

Важное замечание: код предполагает, что в df_atoms и df_residues есть колонки с идентификатором цепи и номером остатка. Если они называются иначе или residue_id имеет нестандартный формат, фильтрацию нужно будет адаптировать под реальные названия колонок.

Ещё один нюанс: формат PDB имеет ограничения на идентификаторы цепей. Если исходный файл mmCIF использует многосимвольные идентификаторы цепей, сохранение в PDB может привести к потере информации. Для таких структур предпочтительнее сохранять отдельные цепи в mmCIF.

```python
# ============================================================
# CELL 16. INTERACTIVE 3D PROTEIN VISUALIZATION
# BATCH PIPELINE: 100 PROTEINS
# NGLView + Jupyter Notebook
# ============================================================
# Назначение:
# Универсальная визуализация структур белков.
#
# BATCH MODE:
# - регистрация визуализации для каждого PDB;
# - определение небелковых компонентов;
# - сохранение JSON-метаданных;
# - отсутствие автоматического открытия 100 окон.
#
# INTERACTIVE MODE:
# - визуализация выбранной структуры;
# - Cartoon;
# - Ball and Stick;
# - Surface;
# - Chains;
# - All Components;
# - Individual Components.
#
# ВАЖНО:
# 1. Основная обработка белков не зависит от NGLView.
# 2. Ошибка визуализации не должна останавливать pipeline.
# 3. Для каждого PDB используется PDB_RESULTS_DIR.
# 4. Исходный CIF не копируется повторно.
# 5. Анализ выполняется для исходной структуры,
#    а не для отдельных PDB-файлов цепей из CELL 15.
# ============================================================


import json
import re
import importlib.util

from pathlib import Path
from collections import Counter

from IPython.display import display


print("=" * 70)
print("CELL 16. INTERACTIVE 3D PROTEIN VISUALIZATION")
print("=" * 70)


# ------------------------------------------------------------
# 1. НАСТРОЙКИ
# ------------------------------------------------------------

# False: пакетная регистрация без открытия визуализации.
# True: открыть визуализацию выбранного PDB.

ENABLE_INTERACTIVE_VISUALIZATION = False


# PDB ID для интерактивного просмотра.
# Например: "1OCC".
#
# Если None, используется текущий PDB_ID.

VISUALIZE_PDB_ID = None


# Автоматически открывать визуализацию,
# если текущий PDB совпадает с VISUALIZE_PDB_ID.

AUTO_OPEN_SELECTED_STRUCTURE = True


# Разрешить экспорт PNG через интерфейс NGLView.

ENABLE_PNG_EXPORT = True


# ------------------------------------------------------------
# 2. ПРОВЕРКА УСТАНОВКИ NGLVIEW
# ------------------------------------------------------------

NGLVIEW_AVAILABLE = (
    importlib.util.find_spec("nglview") is not None
)

IPYWIDGETS_AVAILABLE = (
    importlib.util.find_spec("ipywidgets") is not None
)


print("\nDEPENDENCIES")

print(
    f"NGLView available: {NGLVIEW_AVAILABLE}"
)

print(
    f"ipywidgets available: {IPYWIDGETS_AVAILABLE}"
)


# ------------------------------------------------------------
# 3. ФУНКЦИИ НОРМАЛИЗАЦИИ
# ------------------------------------------------------------

def normalize_pdb_id(value):

    value = str(value).strip().upper()

    if not re.fullmatch(r"[A-Z0-9]{4}", value):

        raise ValueError(
            f"Некорректный PDB ID: {value}"
        )

    return value


def normalize_component_name(value):

    if value is None:

        return None

    value = str(value).strip().upper()

    if value in {"", ".", "?", "NAN", "NONE"}:

        return None

    return value


def safe_json_value(value):

    """
    Преобразование NumPy/pandas-значений
    в JSON-совместимый формат.
    """

    if hasattr(value, "item"):

        try:
            value = value.item()
        except Exception:
            pass

    if value is None:
        return None

    if isinstance(value, (str, int, float, bool)):
        return value

    return str(value)


# ------------------------------------------------------------
# 4. ОПРЕДЕЛЕНИЕ КОМПОНЕНТОВ
# ------------------------------------------------------------

def detect_hetero_components(df_hetero):

    """
    Определяет небелковые компоненты
    по таблице CELL 8.

    Возвращает:
    - список названий компонентов;
    - количество экземпляров;
    - признак доступности классификации.
    """

    if df_hetero is None:

        return [], {}, False


    if not hasattr(df_hetero, "columns"):

        return [], {}, False


    if df_hetero.empty:

        return [], {}, True


    name_column = next(

        (
            column
            for column in [
                "residue_name",
                "resname",
                "component_name"
            ]
            if column in df_hetero.columns
        ),

        None

    )


    if name_column is None:

        return [], {}, False


    normalized_names = (

        df_hetero[name_column]
        .map(normalize_component_name)
        .dropna()

    )


    component_counts = Counter(
        normalized_names.tolist()
    )


    component_names = sorted(
        component_counts.keys()
    )


    return (

        component_names,

        dict(component_counts),

        True

    )


# ------------------------------------------------------------
# 5. СОХРАНЕНИЕ МЕТАДАННЫХ ВИЗУАЛИЗАЦИИ
# ------------------------------------------------------------

def register_structure_visualization(

    pdb_id,

    cif_path,

    pdb_results_dir,

    df_hetero=None

):

    """
    Регистрирует возможность визуализации структуры.

    Функция безопасна для пакетного pipeline:
    она не открывает NGLView и не изменяет исходные файлы.
    """

    pdb_id = normalize_pdb_id(pdb_id)

    cif_path = Path(cif_path)

    pdb_results_dir = Path(pdb_results_dir)


    pdb_results_dir.mkdir(

        parents=True,

        exist_ok=True

    )


    # Проверка исходной структуры.

    structure_exists = cif_path.is_file()

    structure_size = (

        cif_path.stat().st_size

        if structure_exists

        else 0

    )


    structure_valid = (

        structure_exists

        and structure_size > 0

    )


    # Определение компонентов.

    component_names, component_counts, component_detection_available = (

        detect_hetero_components(df_hetero)

    )


    # Информация о визуализации.

    visualization_info = {

        "PDB_ID": pdb_id,

        "structure_file": str(cif_path),

        "structure_exists": bool(structure_exists),

        "structure_valid": bool(structure_valid),

        "structure_size_bytes": int(structure_size),

        "visualization_library": "NGLView",

        "nglview_available": bool(NGLVIEW_AVAILABLE),

        "ipywidgets_available": bool(IPYWIDGETS_AVAILABLE),

        "component_detection_available": bool(
            component_detection_available
        ),

        "detected_component_count": len(
            component_names
        ),

        "detected_components": component_names,

        "component_instance_counts": component_counts,

        "available_visualization_modes": [

            "Cartoon",

            "Ball and Stick",

            "Surface",

            "Chains",

            "All Components",

            "Individual Components"

        ],

        "visualization_enabled": False,

        "png_export_enabled": bool(
            ENABLE_PNG_EXPORT
        ),

        "png_automatically_saved": False,

        "analysis_scope": (
            "Original mmCIF structure"
        )

    }


    output_path = (

        pdb_results_dir
        / f"{pdb_id}_visualization.json"

    )


    with open(

        output_path,

        "w",

        encoding="utf-8"

    ) as file:

        json.dump(

            visualization_info,

            file,

            ensure_ascii=False,

            indent=4,

            default=safe_json_value,

            allow_nan=False

        )


    return {

        "PDB_ID": pdb_id,

        "metadata_path": output_path,

        "structure_valid": structure_valid,

        "component_names": component_names,

        "component_counts": component_counts,

        "nglview_available": NGLVIEW_AVAILABLE

    }


# ------------------------------------------------------------
# 6. СОЗДАНИЕ ИНТЕРАКТИВНОЙ ВИЗУАЛИЗАЦИИ
# ------------------------------------------------------------

def create_interactive_protein_view(

    pdb_id,

    cif_path,

    df_hetero=None

):

    """
    Создаёт интерактивную NGLView-визуализацию.

    Вызывается только для выбранного PDB.
    """

    if not NGLVIEW_AVAILABLE:

        raise ImportError(
            "NGLView не установлен. "
            "Установите пакет: pip install nglview"
        )


    if not IPYWIDGETS_AVAILABLE:

        raise ImportError(
            "ipywidgets не установлен. "
            "Установите пакет: pip install ipywidgets"
        )


    import nglview as nv

    import ipywidgets as widgets


    pdb_id = normalize_pdb_id(pdb_id)

    cif_path = Path(cif_path)


    if not cif_path.is_file():

        raise FileNotFoundError(
            f"Структура не найдена: {cif_path}"
        )


    if cif_path.stat().st_size == 0:

        raise ValueError(
            f"Файл структуры пуст: {cif_path}"
        )


    component_names, component_counts, _ = (

        detect_hetero_components(df_hetero)

    )


    print("\n" + "-" * 70)

    print(f"INTERACTIVE VIEW: {pdb_id}")

    print("-" * 70)


    print(
        "Detected components:",
        component_names if component_names else "None"
    )


    # Создание представления.

    view = nv.show_file(
        str(cif_path)
    )


    view.clear_representations()


    # Начальное представление.

    view.add_cartoon(

        selection="protein",

        color="sstruc"

    )


    view.add_ball_and_stick(

        selection="organic"

    )


    view.center()


    view.layout.width = "950px"

    view.layout.height = "650px"


    # --------------------------------------------------------
    # Режимы визуализации
    # --------------------------------------------------------

    def show_cartoon():

        view.clear_representations()

        view.add_cartoon(

            selection="protein",

            color="sstruc"

        )

        view.add_ball_and_stick(
            selection="organic"
        )

        view.center()


    def show_ball_and_stick():

        view.clear_representations()

        view.add_ball_and_stick(
            selection="protein"
        )

        view.add_ball_and_stick(
            selection="not protein and not water"
        )

        view.center()


    def show_surface():

        view.clear_representations()

        view.add_surface(

            selection="protein",

            color="chainid",

            opacity=0.7

        )

        view.add_ball_and_stick(
            selection="organic"
        )

        view.center()


    def show_chains():

        view.clear_representations()

        view.add_cartoon(

            selection="protein",

            color="chainid"

        )

        view.add_ball_and_stick(
            selection="organic"
        )

        view.center()


    def show_all_components():

        view.clear_representations()

        # Белковые цепи.

        view.add_cartoon(

            selection="protein",

            color="chainid"

        )

        # Все небелковые компоненты, кроме воды.

        view.add_ball_and_stick(

            selection="not protein and not water",

            color="element"

        )

        # Вода.

        view.add_ball_and_stick(

            selection="water",

            color="element"

        )

        view.center()


    def show_individual_components():

        view.clear_representations()

        view.add_cartoon(

            selection="protein",

            color="chainid"

        )

        # Компоненты отображаются отдельно
        # по их трёхбуквенным идентификаторам.

        for component_name in component_names:

            if not re.fullmatch(
                r"[A-Z0-9]{1,8}",
                component_name
            ):

                continue


            view.add_ball_and_stick(

                selection=f"resname {component_name}",

                color="element"

            )


        view.center()


    # --------------------------------------------------------
    # Кнопки
    # --------------------------------------------------------

    button_cartoon = widgets.Button(

        description="Cartoon",

        button_style="primary",

        tooltip="Protein secondary structure"

    )


    button_atoms = widgets.Button(

        description="Ball and Stick",

        tooltip="Atomic representation"

    )


    button_surface = widgets.Button(

        description="Surface",

        tooltip="Protein molecular surface"

    )


    button_chains = widgets.Button(

        description="Chains",

        tooltip="Color chains separately"

    )


    button_all = widgets.Button(

        description="All Components",

        tooltip="Protein, hetero components and water"

    )


    button_components = widgets.Button(

        description="Individual Components",

        tooltip="Display detected components"

    )


    button_cartoon.on_click(
        lambda button: show_cartoon()
    )

    button_atoms.on_click(
        lambda button: show_ball_and_stick()
    )

    button_surface.on_click(
        lambda button: show_surface()
    )

    button_chains.on_click(
        lambda button: show_chains()
    )

    button_all.on_click(
        lambda button: show_all_components()
    )

    button_components.on_click(
        lambda button: show_individual_components()
    )


    # --------------------------------------------------------
    # Экспорт PNG
    # --------------------------------------------------------

    button_png = None


    if ENABLE_PNG_EXPORT:

        button_png = widgets.Button(

            description="Download PNG",

            button_style="success",

            tooltip="Export current NGLView image"

        )


        def download_png(button):

            try:

                view.download_image(

                    filename=f"{pdb_id}_structure.png",

                    factor=3,

                    antialias=True,

                    trim=True

                )

                print(
                    "PNG export requested. "
                    "The browser determines the download location."
                )

            except Exception as error:

                print(
                    "PNG export error:",
                    repr(error)
                )


        button_png.on_click(
            download_png
        )


    # --------------------------------------------------------
    # Отображение интерфейса
    # --------------------------------------------------------

    button_list = [

        button_cartoon,

        button_atoms,

        button_surface,

        button_chains,

        button_all,

        button_components

    ]


    if button_png is not None:

        button_list.append(
            button_png
        )


    controls = widgets.Box(

        children=button_list,

        layout=widgets.Layout(

            display="flex",

            flex_flow="row wrap",

            width="100%",

            gap="5px"

        )

    )


    display(controls)

    display(view)


    return view


# ------------------------------------------------------------
# 7. РЕГИСТРАЦИЯ ТЕКУЩЕЙ СТРУКТУРЫ
# ------------------------------------------------------------

# При пакетной обработке этот блок выполняется
# для текущего PDB внутри основного цикла.
#
# Если CELL 16 запускается отдельно,
# используются переменные текущей структуры.


current_pdb_id = None

current_cif_path = None

current_results_dir = None

current_hetero = None


if "PDB_ID" in globals():

    current_pdb_id = PDB_ID


if "CIF_PATH" in globals():

    current_cif_path = CIF_PATH


if "PDB_RESULTS_DIR" in globals():

    current_results_dir = PDB_RESULTS_DIR


if "df_hetero" in globals():

    current_hetero = df_hetero


if (

    current_pdb_id is not None

    and current_cif_path is not None

    and current_results_dir is not None

):

    registration_result = register_structure_visualization(

        pdb_id=current_pdb_id,

        cif_path=current_cif_path,

        pdb_results_dir=current_results_dir,

        df_hetero=current_hetero

    )


    print("\nCURRENT STRUCTURE REGISTRATION")

    print(
        "PDB ID:",
        registration_result["PDB_ID"]
    )

    print(
        "Structure valid:",
        registration_result["structure_valid"]
    )

    print(
        "Detected components:",
        registration_result["component_names"]
    )

    print(
        "Metadata:",
        registration_result["metadata_path"]
    )


else:

    print(
        "\nCurrent structure variables are unavailable."
    )

    print(
        "Visualization functions are ready for batch processing."
    )


# ------------------------------------------------------------
# 8. ЗАПУСК ИНТЕРАКТИВНОЙ ВИЗУАЛИЗАЦИИ
# ------------------------------------------------------------

if ENABLE_INTERACTIVE_VISUALIZATION:

    if VISUALIZE_PDB_ID is None:

        selected_pdb_id = normalize_pdb_id(
            current_pdb_id
        )

    else:

        selected_pdb_id = normalize_pdb_id(
            VISUALIZE_PDB_ID
        )


    if current_pdb_id is None:

        raise RuntimeError(
            "Для интерактивного просмотра "
            "не определён текущий PDB ID."
        )


    if selected_pdb_id != normalize_pdb_id(current_pdb_id):

        raise ValueError(

            f"Выбран PDB {selected_pdb_id}, "
            f"но текущие данные относятся к {current_pdb_id}. "
            "Загрузите выбранную структуру перед визуализацией."

        )


    try:

        interactive_view = create_interactive_protein_view(

            pdb_id=current_pdb_id,

            cif_path=current_cif_path,

            df_hetero=current_hetero

        )

    except Exception as error:

        print(
            "\nInteractive visualization unavailable:"
        )

        print(
            repr(error)
        )

        print(
            "Основная обработка структуры не затронута."
        )


else:

    print(
        "\nInteractive visualization is disabled."
    )

    print(
        "Batch pipeline will not create NGLView widgets."
    )


# ------------------------------------------------------------
# 9. ИТОГ
# ------------------------------------------------------------

print("\n" + "=" * 70)

print("CELL 16 COMPLETED")

print("=" * 70)

print(
    "Batch-compatible visualization: YES"
)

print(
    "Interactive visualization enabled:",
    ENABLE_INTERACTIVE_VISUALIZATION
)

print(
    "NGLView available:",
    NGLVIEW_AVAILABLE
)

print(
    "PNG export enabled:",
    ENABLE_PNG_EXPORT
)

print(
    "Main pipeline dependency on visualization: NO"
)

print("=" * 70)
```

```python
# ============================================================
# CELL 17. DETECTION AND IDENTIFICATION OF NON-PROTEIN COMPONENTS
# ============================================================
# Назначение:
# 1. Обнаружение небелковых компонентов первой модели.
# 2. Исключение воды и распознанных аминокислотных остатков.
# 3. Определение элементного состава.
# 4. Предварительная классификация компонентов.
# 5. Получение химической информации из mmCIF _chem_comp.
# 6. Выбор компонента для последующего анализа.
# 7. Сохранение таблиц компонентов и их атомов.
#
# ВАЖНО:
# - Анализируется первая модель, как в CELL 4–8.
# - Альтернативные положения атомов сохраняются.
# - number_of_atoms — число уникальных атомных позиций.
# - number_of_atom_conformers — число записей атомов
#   с учетом альтернативных положений.
# - Классификация по элементам предварительная.
# - Выбранный компонент не обязательно является лигандом.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.PDB import MMCIFParser, is_aa
from Bio.PDB.MMCIF2Dict import MMCIF2Dict


print("=" * 70)
print("CELL 17. NON-PROTEIN COMPONENT DETECTION")
print("=" * 70)


# ------------------------------------------------------------
# 1. ПРОВЕРКА ВХОДНЫХ ДАННЫХ
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "CIF_PATH",
    "PDB_RESULTS_DIR"
]

for variable in required_variables:
    if variable not in globals():
        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()
CIF_PATH = Path(CIF_PATH)
PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

if not CIF_PATH.is_file():
    raise FileNotFoundError(
        f"Structure file not found: {CIF_PATH}"
    )

if CIF_PATH.stat().st_size == 0:
    raise ValueError("Structure file is empty.")

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ------------------------------------------------------------
# 2. ЗАГРУЗКА СТРУКТУРЫ И mmCIF
# ------------------------------------------------------------

parser = MMCIFParser(QUIET=True)

component_structure = parser.get_structure(
    PDB_ID,
    str(CIF_PATH)
)

models = list(component_structure)

if not models:
    raise ValueError(
        f"No models found in structure {PDB_ID}."
    )

# Для согласованности с CELL 4–8 анализируем первую модель.
model = models[0]

MODEL_INDEX = 0

MODEL_SERIAL_NUMBER = getattr(
    model,
    "serial_num",
    None
)

if MODEL_SERIAL_NUMBER is None:
    MODEL_SERIAL_NUMBER = model.id


# Загружаем mmCIF отдельно, чтобы не зависеть
# от состояния cif_dict в предыдущих ячейках.
cif_dict = MMCIF2Dict(str(CIF_PATH))


print("PDB ID:", PDB_ID)
print("Structure:", CIF_PATH)
print("Total models in file:", len(models))
print("Analyzed model index:", MODEL_INDEX)
print("Model serial number:", MODEL_SERIAL_NUMBER)


# ------------------------------------------------------------
# 3. НАСТРОЙКИ
# ------------------------------------------------------------

WATER_CODES = {
    "HOH",
    "WAT",
    "DOD"
}

METAL_ELEMENTS = {
    "FE", "ZN", "MG", "MN", "CU",
    "CO", "NI", "CA", "NA", "K",
    "CD", "HG"
}

PREFERRED_COMPONENT_NAME = "HEM"
PREFERRED_COMPONENT_CHAIN = "A"

# Коды, которые могут быть полезны для последующего
# анализа белковых комплексов. Это не означает, что
# каждый такой компонент является лекарственным лигандом.
IMPORTANT_COMPONENT_CODES = [
    "HEM", "NDP", "NAD", "NAP",
    "FAD", "FMN", "ATP", "ADP",
    "GTP", "GDP", "SAM", "PLP",
    "COA", "TPP", "MG", "ZN",
    "FE", "PO4", "SO4", "ACT", "OXY"
]


# ------------------------------------------------------------
# 4. ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ
# ------------------------------------------------------------

def clean_cif_value(value):
    """Нормализация значения из mmCIF."""

    if value is None:
        return None

    value = str(value).strip()

    if value in {"", ".", "?"}:
        return None

    return value


def get_cif_column(cif_data, key):
    """Безопасное получение столбца mmCIF."""

    values = cif_data.get(key, [])

    if values is None:
        values = []

    if isinstance(values, str):
        values = [values]

    return [
        clean_cif_value(value)
        for value in values
    ]


def normalize_chain_id(chain_id):
    """Нормализация отображаемого идентификатора цепи."""

    if chain_id is None:
        return "<blank>"

    value = str(chain_id).strip()

    return value if value else "<blank>"


def get_element(atom):
    """Получение нормализованного химического элемента."""

    element = str(atom.element).strip().upper()

    if not element:
        element = "UNKNOWN"

    return element


def get_unique_atom_site_key(atom):
    """
    Идентификатор атомной позиции внутри одного остатка.

    Альтернативные конформеры одного атома имеют
    одинаковое имя атома и относятся к одной позиции.
    """

    return (
        str(atom.get_name()).strip(),
        get_element(atom)
    )


def classify_component(elements, atom_count):
    """
    Предварительная классификация по элементному составу.

    Не определяет биологическую функцию компонента.
    """

    element_set = set(elements.keys())

    has_carbon = "C" in element_set

    metal_elements = sorted(
        element_set.intersection(METAL_ELEMENTS)
    )

    if metal_elements:
        if has_carbon:
            return "Metal-containing organic component"

        if atom_count == 1:
            return "Metal ion"

        return "Metal-containing inorganic component"

    if atom_count == 1:
        return "Single-atom component / ion"

    if has_carbon:
        return "Organic non-protein component"

    return "Inorganic non-protein component"


def json_safe(value):
    """Рекурсивное преобразование значений в JSON-формат."""

    if isinstance(value, dict):
        return {
            str(key): json_safe(val)
            for key, val in value.items()
        }

    if isinstance(value, (list, tuple)):
        return [
            json_safe(val)
            for val in value
        ]

    if isinstance(value, np.ndarray):
        return json_safe(value.tolist())

    if isinstance(value, (np.integer,)):
        return int(value)

    if isinstance(value, (np.floating, float)):
        value = float(value)

        return value if np.isfinite(value) else None

    if isinstance(value, (np.bool_,)):
        return bool(value)

    if value is None:
        return None

    if pd.isna(value):
        return None

    return value


# ------------------------------------------------------------
# 5. ПОЛУЧЕНИЕ CHEMICAL COMPONENT DICTIONARY
# ------------------------------------------------------------

chem_comp_fields = {
    "component": "_chem_comp.id",
    "name": "_chem_comp.name",
    "type": "_chem_comp.type",
    "formula": "_chem_comp.formula",
    "formal_charge": "_chem_comp.pdbx_formal_charge",
    "synonyms": "_chem_comp.pdbx_synonyms",
    "mon_nstd_flag": "_chem_comp.mon_nstd_flag"
}

chem_comp_columns = {
    name: get_cif_column(cif_dict, cif_key)
    for name, cif_key in chem_comp_fields.items()
}

chem_comp_max_length = max(
    (len(values) for values in chem_comp_columns.values()),
    default=0
)

chem_comp_records = []

for i in range(chem_comp_max_length):

    record = {}

    for column, values in chem_comp_columns.items():

        record[column] = (
            values[i]
            if i < len(values)
            else None
        )

    chem_comp_records.append(record)


df_chem_comp_dictionary = pd.DataFrame(
    chem_comp_records,
    columns=list(chem_comp_fields.keys())
)


if not df_chem_comp_dictionary.empty:

    df_chem_comp_dictionary["component"] = (
        df_chem_comp_dictionary["component"]
        .fillna("")
        .astype(str)
        .str.strip()
        .str.upper()
    )

    df_chem_comp_dictionary = (
        df_chem_comp_dictionary[
            df_chem_comp_dictionary["component"] != ""
        ]
        .drop_duplicates(subset=["component"])
        .reset_index(drop=True)
    )


chem_comp_lookup = {
    str(row["component"]): row.to_dict()
    for _, row in df_chem_comp_dictionary.iterrows()
    if pd.notna(row["component"])
}


print("\nChemical component dictionary:")
print("Records:", len(df_chem_comp_dictionary))


# ------------------------------------------------------------
# 6. ОБНАРУЖЕНИЕ НЕБЕЛКОВЫХ КОМПОНЕНТОВ
# ------------------------------------------------------------

ligand_residues = []

water_count = 0

standard_amino_acid_count = 0
modified_amino_acid_count = 0

# Для контроля общего числа остатков первой модели.
total_residue_count = 0

for chain in model:

    for residue in chain:

        total_residue_count += 1

        residue_name = (
            residue.resname.strip().upper()
        )

        # get_unpacked_list() возвращает все альтернативные
        # атомные конформеры.
        all_atoms = residue.get_unpacked_list()

        if not all_atoms:
            continue

        # Вода.
        if residue_name in WATER_CODES:
            water_count += 1
            continue

        # Стандартные аминокислоты.
        if is_aa(residue, standard=True):
            standard_amino_acid_count += 1
            continue

        # Нестандартные и модифицированные аминокислоты.
        if is_aa(residue, standard=False):
            modified_amino_acid_count += 1
            continue

        # Остальные остатки рассматриваем как небелковые
        # компоненты для дальнейшей проверки.
        ligand_residues.append({
            "chain": chain,
            "residue": residue
        })


recognized_amino_acid_count = (
    standard_amino_acid_count
    + modified_amino_acid_count
)


print("\n" + "-" * 70)
print("DETECTION RESULTS")
print("-" * 70)

print("Residues in first model:", total_residue_count)
print("Water residues:", water_count)

print(
    "Standard amino acid residues excluded:",
    standard_amino_acid_count
)

print(
    "Modified / non-standard amino acid residues excluded:",
    modified_amino_acid_count
)

print(
    "Recognized amino acid residues excluded:",
    recognized_amino_acid_count
)

print(
    "Non-protein component instances:",
    len(ligand_residues)
)


# ------------------------------------------------------------
# 7. СБОР ИНФОРМАЦИИ О КАЖДОМ КОМПОНЕНТЕ
# ------------------------------------------------------------

components_info = []
component_atoms_data = []

for component_index, item in enumerate(
    ligand_residues,
    start=1
):

    chain = item["chain"]
    residue = item["residue"]

    chain_id = str(chain.id)
    chain_id_display = normalize_chain_id(chain_id)

    residue_id = residue.id

    residue_name = (
        residue.resname.strip().upper()
    )

    residue_number = residue_id[1]

    insertion_code = (
        str(residue_id[2]).strip()
    )

    hetero_flag = str(residue_id[0]).strip()

    # Все альтернативные атомные конформеры.
    all_atoms = residue.get_unpacked_list()

    # Атомы, выбранные Bio.PDB для представления позиций.
    unique_atoms = list(residue.get_atoms())

    if not all_atoms or not unique_atoms:
        continue

    # Уникальные позиции определяем по имени атома и элементу.
    unique_site_keys = {
        get_unique_atom_site_key(atom)
        for atom in all_atoms
    }

    number_of_atom_sites = len(unique_site_keys)
    number_of_atom_conformers = len(all_atoms)

    if number_of_atom_sites == 0:
        continue

    # Центр рассчитываем по одному выбранному атому
    # на атомную позицию, а не по всем altloc-конформерам.
    coordinates = np.asarray(
        [atom.coord for atom in unique_atoms],
        dtype=float
    )

    if not np.isfinite(coordinates).all():
        raise ValueError(
            f"Invalid coordinates in component {residue_name}, "
            f"chain {chain_id}, residue {residue_number}."
        )

    centroid = coordinates.mean(axis=0)

    # Элементный состав считаем по уникальным позициям.
    elements = {}

    for atom in unique_atoms:

        element = get_element(atom)

        elements[element] = (
            elements.get(element, 0) + 1
        )

    metal_elements = sorted(
        set(elements.keys()).intersection(METAL_ELEMENTS)
    )

    component_type = classify_component(
        elements,
        number_of_atom_sites
    )

    # Данные химического справочника.
    dictionary_info = chem_comp_lookup.get(
        residue_name,
        {}
    )

    component_name = dictionary_info.get("name")
    component_formula = dictionary_info.get("formula")
    component_synonyms = dictionary_info.get("synonyms")
    component_chem_type = dictionary_info.get("type")
    component_formal_charge = dictionary_info.get("formal_charge")
    component_mon_nstd_flag = dictionary_info.get("mon_nstd_flag")

    # Устойчивый идентификатор экземпляра компонента.
    component_id = (
        f"{PDB_ID}|model_{MODEL_INDEX}|"
        f"chain_{chain_id}|{hetero_flag}|"
        f"{residue_number}|{insertion_code}|{residue_name}"
    )

    info = {
        "component_index": component_index,
        "component_id": component_id,

        "component": residue_name,
        "component_name": component_name,
        "component_formula": component_formula,
        "component_synonyms": component_synonyms,
        "chemical_component_type": component_chem_type,
        "formal_charge": component_formal_charge,
        "mon_nstd_flag": component_mon_nstd_flag,

        "component_type": component_type,

        "model_index": MODEL_INDEX,
        "model_id": model.id,
        "model_serial_number": MODEL_SERIAL_NUMBER,

        "chain": chain_id,
        "chain_display": chain_id_display,

        "residue_number": residue_number,
        "insertion_code": insertion_code,
        "hetero_flag": hetero_flag,

        "number_of_atoms": number_of_atom_sites,
        "number_of_atom_conformers": number_of_atom_conformers,

        "number_of_alternative_atom_records": (
            number_of_atom_conformers - number_of_atom_sites
        ),

        "elements": elements,
        "metal_elements": metal_elements,

        "center_x": float(centroid[0]),
        "center_y": float(centroid[1]),
        "center_z": float(centroid[2])
    }

    components_info.append(info)

    # Сохраняем все атомные конформеры.
    for atom in all_atoms:

        element = get_element(atom)

        atom_data = {
            "component_index": component_index,
            "component_id": component_id,

            "component": residue_name,
            "component_name": component_name,
            "component_type": component_type,

            "model_index": MODEL_INDEX,
            "model_id": model.id,
            "model_serial_number": MODEL_SERIAL_NUMBER,

            "chain": chain_id,
            "chain_display": chain_id_display,

            "residue_number": residue_number,
            "insertion_code": insertion_code,
            "hetero_flag": hetero_flag,

            "atom_name": atom.get_name(),
            "element": element,

            "x": float(atom.coord[0]),
            "y": float(atom.coord[1]),
            "z": float(atom.coord[2]),

            "occupancy": atom.occupancy,
            "B_factor": atom.bfactor,
            "altloc": atom.get_altloc()
        }

        component_atoms_data.append(atom_data)


# ------------------------------------------------------------
# 8. СОЗДАНИЕ DATAFRAME
# ------------------------------------------------------------

COMPONENT_COLUMNS = [
    "component_index",
    "component_id",
    "component",
    "component_name",
    "component_formula",
    "component_synonyms",
    "chemical_component_type",
    "formal_charge",
    "mon_nstd_flag",
    "component_type",
    "model_index",
    "model_id",
    "model_serial_number",
    "chain",
    "chain_display",
    "residue_number",
    "insertion_code",
    "hetero_flag",
    "number_of_atoms",
    "number_of_atom_conformers",
    "number_of_alternative_atom_records",
    "elements",
    "metal_elements",
    "center_x",
    "center_y",
    "center_z"
]

ATOM_COLUMNS = [
    "component_index",
    "component_id",
    "component",
    "component_name",
    "component_type",
    "model_index",
    "model_id",
    "model_serial_number",
    "chain",
    "chain_display",
    "residue_number",
    "insertion_code",
    "hetero_flag",
    "atom_name",
    "element",
    "x",
    "y",
    "z",
    "occupancy",
    "B_factor",
    "altloc"
]

df_components = pd.DataFrame(
    components_info,
    columns=COMPONENT_COLUMNS
)

df_component_atoms = pd.DataFrame(
    component_atoms_data,
    columns=ATOM_COLUMNS
)


print("\n" + "=" * 70)
print("DETECTED NON-PROTEIN COMPONENTS")
print("=" * 70)

if df_components.empty:

    print("No non-protein components detected.")

else:

    display(df_components)

    print("\nComponent classification:")

    display(
        df_components["component_type"]
        .value_counts()
        .rename_axis("component_type")
        .reset_index(name="count")
    )


# ------------------------------------------------------------
# 9. СТАТИСТИКА ПО КОДАМ КОМПОНЕНТОВ
# ------------------------------------------------------------

if not df_components.empty:

    df_component_occurrences = (
        df_components
        .groupby("component", dropna=False)
        .agg(
            occurrence_count=("component_index", "count"),
            total_atom_count=("number_of_atoms", "sum"),
            total_atom_conformer_count=(
                "number_of_atom_conformers",
                "sum"
            ),
            chains=(
                "chain_display",
                lambda x: ", ".join(sorted(set(x.astype(str))))
            )
        )
        .reset_index()
    )

else:

    df_component_occurrences = pd.DataFrame(
        columns=[
            "component",
            "occurrence_count",
            "total_atom_count",
            "total_atom_conformer_count",
            "chains"
        ]
    )


number_of_component_instances = len(df_components)

number_of_unique_component_types = (
    df_components["component"].nunique()
    if not df_components.empty
    else 0
)


print("\nComponent occurrences:")
display(df_component_occurrences)

print(
    "\nNumber of component instances:",
    number_of_component_instances
)

print(
    "Number of unique component types:",
    number_of_unique_component_types
)


# ------------------------------------------------------------
# 10. ПРОВЕРКА ВАЖНЫХ КОМПОНЕНТОВ
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("IMPORTANT COMPONENTS")
print("=" * 70)

for component_code in IMPORTANT_COMPONENT_CODES:

    print(f"\n{component_code}:")

    if df_components.empty:
        print("Not detected.")
        continue

    selected_rows = df_components[
        df_components["component"] == component_code
    ]

    if selected_rows.empty:
        print("Not detected.")
        continue

    display(
        selected_rows[
            [
                "component",
                "component_name",
                "component_formula",
                "chain_display",
                "residue_number",
                "number_of_atoms",
                "elements",
                "metal_elements"
            ]
        ]
    )


# ------------------------------------------------------------
# 11. ВЫБОР КОМПОНЕНТА ДЛЯ ДАЛЬНЕЙШЕГО АНАЛИЗА
# ------------------------------------------------------------

selected_component = None
selected_component_index = None
selected_info = None
selected_item = None

SELECTED_COMPONENT_NAME = None
SELECTED_COMPONENT_CHAIN = None

if df_components.empty:

    SELECTION_METHOD = "No components available"

    print("\nNo component available for selection.")

else:

    print("\nAVAILABLE COMPONENTS")
    print("-" * 70)

    for _, row in df_components.iterrows():

        print(
            f"{int(row['component_index'])}. "
            f"{row['component']} | "
            f"Chain: {row['chain_display']} | "
            f"Residue: {row['residue_number']} | "
            f"Atoms: {row['number_of_atoms']}"
        )

    # Приоритет: HEM в цепи A.
    selected_rows = df_components[
        (df_components["component"] == PREFERRED_COMPONENT_NAME)
        &
        (df_components["chain"] == PREFERRED_COMPONENT_CHAIN)
    ]

    if not selected_rows.empty:

        SELECTION_METHOD = (
            "Preferred component and chain"
        )

    else:

        # Резервный вариант — первый обнаруженный компонент.
        # Выбор является техническим, а не функциональным.
        selected_rows = df_components.iloc[[0]]

        SELECTION_METHOD = (
            "First available component fallback; "
            "biological relevance not established"
        )

    selected_info = selected_rows.iloc[0].to_dict()

    selected_component_index = int(
        selected_info["component_index"]
    )

    selected_item = ligand_residues[
        selected_component_index - 1
    ]

    selected_component = selected_item["residue"]

    SELECTED_COMPONENT_NAME = selected_info["component"]
    SELECTED_COMPONENT_CHAIN = selected_info["chain"]

    print("\n" + "=" * 70)
    print("SELECTED COMPONENT")
    print("=" * 70)

    print("Selection method:", SELECTION_METHOD)
    print("Component:", selected_info["component"])
    print("Chemical name:", selected_info["component_name"])
    print("Formula:", selected_info["component_formula"])
    print("Chain:", selected_info["chain_display"])
    print("Residue number:", selected_info["residue_number"])
    print("Model:", selected_info["model_serial_number"])

    print(
        "Unique atom sites:",
        selected_info["number_of_atoms"]
    )

    print(
        "Atom conformers:",
        selected_info["number_of_atom_conformers"]
    )

    print(
        "Elemental composition:",
        selected_info["elements"]
    )

    print(
        "Metal elements:",
        selected_info["metal_elements"]
    )

    print(
        "Geometric center:",
        np.round(
            [
                selected_info["center_x"],
                selected_info["center_y"],
                selected_info["center_z"]
            ],
            3
        )
    )


# ------------------------------------------------------------
# 12. АТОМЫ ВЫБРАННОГО КОМПОНЕНТА
# ------------------------------------------------------------

if selected_component is not None:

    df_selected_atoms = df_component_atoms[
        df_component_atoms["component_index"]
        == selected_component_index
    ].copy()

else:

    df_selected_atoms = pd.DataFrame(
        columns=ATOM_COLUMNS
    )


print("\nSelected component atom records:")
print(len(df_selected_atoms))

print(
    "Selected component unique atom sites:",
    (
        len({
            (row["atom_name"], row["element"])
            for _, row in df_selected_atoms.iterrows()
        })
        if not df_selected_atoms.empty
        else 0
    )
)

print(
    "Selected component atom conformers:",
    len(df_selected_atoms)
)


# ------------------------------------------------------------
# 13. СОПОСТАВЛЕНИЕ С CELL 8
# ------------------------------------------------------------

cell8_comparison = {
    "available": False,
    "cell8_component_count": None,
    "cell17_component_count": len(df_components),
    "difference": None,
    "status": "Not checked"
}

if "df_hetero" in globals() and isinstance(df_hetero, pd.DataFrame):

    cell8_comparison["available"] = True
    cell8_comparison["cell8_component_count"] = len(df_hetero)

    cell8_comparison["difference"] = (
        len(df_components) - len(df_hetero)
    )

    if len(df_components) == len(df_hetero):

        cell8_comparison["status"] = "Counts match"

    else:

        cell8_comparison["status"] = (
            "Counts differ; inspect classification and model scope"
        )

    print("\nCELL 8 COMPARISON")
    print("-" * 70)

    print(
        "CELL 8 hetero residues:",
        len(df_hetero)
    )

    print(
        "CELL 17 component instances:",
        len(df_components)
    )

    print(
        "Difference:",
        cell8_comparison["difference"]
    )

    print(
        "Status:",
        cell8_comparison["status"]
    )

else:

    print(
        "\nCELL 8 comparison skipped: df_hetero unavailable."
    )


# ------------------------------------------------------------
# 14. СОХРАНЕНИЕ CSV
# ------------------------------------------------------------

COMPONENTS_CSV_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_components.csv"
)

COMPONENT_ATOMS_CSV_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_component_atoms.csv"
)

COMPONENT_OCCURRENCES_CSV_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_component_occurrences.csv"
)

CHEM_COMP_CSV_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_chemical_component_dictionary.csv"
)

SELECTED_COMPONENT_ATOMS_CSV_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_selected_component_atoms.csv"
)


df_components.to_csv(
    COMPONENTS_CSV_PATH,
    index=False
)

df_component_atoms.to_csv(
    COMPONENT_ATOMS_CSV_PATH,
    index=False
)

df_component_occurrences.to_csv(
    COMPONENT_OCCURRENCES_CSV_PATH,
    index=False
)

df_chem_comp_dictionary.to_csv(
    CHEM_COMP_CSV_PATH,
    index=False
)

df_selected_atoms.to_csv(
    SELECTED_COMPONENT_ATOMS_CSV_PATH,
    index=False
)


# ------------------------------------------------------------
# 15. СОХРАНЕНИЕ JSON
# ------------------------------------------------------------

selection_info = {
    "PDB_ID": PDB_ID,

    "selected_component_index": selected_component_index,
    "selected_component": selected_info,

    "preferred_component_name": PREFERRED_COMPONENT_NAME,
    "preferred_component_chain": PREFERRED_COMPONENT_CHAIN,

    "selected_component_name": SELECTED_COMPONENT_NAME,
    "selected_component_chain": SELECTED_COMPONENT_CHAIN,

    "selection_method": SELECTION_METHOD,

    "number_of_component_instances": number_of_component_instances,
    "number_of_unique_component_types": number_of_unique_component_types,

    "number_of_waters": water_count,

    "number_of_standard_amino_acid_residues_excluded":
        standard_amino_acid_count,

    "number_of_modified_amino_acid_residues_excluded":
        modified_amino_acid_count,

    "number_of_recognized_amino_acid_residues_excluded":
        recognized_amino_acid_count
}


component_info = {
    "PDB_ID": PDB_ID,

    "model_scope": {
        "total_models_in_file": len(models),
        "analyzed_model_index": MODEL_INDEX,
        "analyzed_model_id": model.id,
        "model_serial_number": MODEL_SERIAL_NUMBER
    },

    "components": components_info,

    "component_occurrences": (
        df_component_occurrences.to_dict(orient="records")
    ),

    "selection": selection_info,

    "cell8_comparison": cell8_comparison,

    "classification_method": (
        "Preliminary classification based on elemental "
        "composition and unique atom sites"
    ),

    "chemical_dictionary_source": (
        "mmCIF _chem_comp category"
    ),

    "atom_count_definition": {
        "number_of_atoms": (
            "Number of unique atom sites, identified by atom name and element"
        ),

        "number_of_atom_conformers": (
            "Number of atom records including alternate locations"
        ),

        "number_of_alternative_atom_records": (
            "Difference between conformer records and unique atom sites"
        )
    },

    "limitations": [
        "Only the first structural model is analyzed.",
        "Unrecognized modified amino acids may be classified as non-protein components.",
        "Element-based classification does not establish biological function.",
        "Fallback component selection does not establish ligand relevance.",
        "The geometric center uses one selected coordinate per atom site."
    ]
}


component_info = json_safe(component_info)

COMPONENT_INFO_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_component_info.json"
)

with open(
    COMPONENT_INFO_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        component_info,
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


# ------------------------------------------------------------
# 16. ПРОВЕРКИ
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("VALIDATION")
print("=" * 70)


# Проверка количества атомных конформеров.
if not df_component_atoms.empty:

    atom_conformer_counts = (
        df_component_atoms
        .groupby("component_index")
        .size()
        .to_dict()
    )

else:

    atom_conformer_counts = {}


for _, row in df_components.iterrows():

    index = int(row["component_index"])

    actual_count = atom_conformer_counts.get(index, 0)

    expected_count = int(
        row["number_of_atom_conformers"]
    )

    if actual_count != expected_count:

        raise ValueError(
            f"Atom conformer count mismatch for component {index}: "
            f"{actual_count} != {expected_count}"
        )


print("Atom conformer counts: OK")


# Проверка уникальных атомных позиций.
for _, row in df_components.iterrows():

    index = int(row["component_index"])

    component_atom_rows = df_component_atoms[
        df_component_atoms["component_index"] == index
    ]

    actual_unique_sites = len({
        (str(atom_name).strip(), str(element).strip().upper())
        for atom_name, element in zip(
            component_atom_rows["atom_name"],
            component_atom_rows["element"]
        )
    })

    expected_unique_sites = int(
        row["number_of_atoms"]
    )

    if actual_unique_sites != expected_unique_sites:

        raise ValueError(
            f"Unique atom site count mismatch for component {index}: "
            f"{actual_unique_sites} != {expected_unique_sites}"
        )


print("Unique atom site counts: OK")


# Проверка выбранного компонента.
if selected_component is not None:

    if selected_info["component"] != SELECTED_COMPONENT_NAME:
        raise ValueError("Selected component name mismatch.")

    if selected_info["chain"] != SELECTED_COMPONENT_CHAIN:
        raise ValueError("Selected component chain mismatch.")

    if (
        len(df_selected_atoms)
        != selected_info["number_of_atom_conformers"]
    ):
        raise ValueError("Selected component atom count mismatch.")

    print("Selected component: OK")

else:

    print("Selected component: Not available")


# Проверка координат.
if not df_component_atoms.empty:

    coordinate_columns = ["x", "y", "z"]

    coordinates = df_component_atoms[
        coordinate_columns
    ].to_numpy(dtype=float)

    if not np.isfinite(coordinates).all():

        raise ValueError(
            "Invalid component atom coordinates."
        )


print("Coordinates: OK")


# Проверка количества экземпляров.
if len(df_components) != len(ligand_residues):

    raise ValueError(
        "Component instance count mismatch."
    )


print("Component instance count: OK")


# ------------------------------------------------------------
# 17. ПРОВЕРКА СОХРАНЕННЫХ ФАЙЛОВ
# ------------------------------------------------------------

files_to_check = [
    COMPONENTS_CSV_PATH,
    COMPONENT_ATOMS_CSV_PATH,
    COMPONENT_OCCURRENCES_CSV_PATH,
    CHEM_COMP_CSV_PATH,
    SELECTED_COMPONENT_ATOMS_CSV_PATH,
    COMPONENT_INFO_PATH
]

print("\nSaved files:")

for path in files_to_check:

    print(
        f"- {path.name}: "
        f"{'OK' if path.is_file() else 'ERROR'}"
    )


if not all(path.is_file() for path in files_to_check):

    raise RuntimeError(
        "Не все файлы анализа компонентов были сохранены."
    )


# ------------------------------------------------------------
# 18. ИТОГ
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("COMPONENT ANALYSIS COMPLETED")
print("=" * 70)

print("PDB ID:", PDB_ID)

print(
    "Analyzed model:",
    MODEL_INDEX,
    "| serial:",
    MODEL_SERIAL_NUMBER
)

print(
    "Non-protein component instances:",
    number_of_component_instances
)

print(
    "Unique component types:",
    number_of_unique_component_types
)

print("Water residues:", water_count)

print(
    "Standard amino acid residues excluded:",
    standard_amino_acid_count
)

print(
    "Modified amino acid residues excluded:",
    modified_amino_acid_count
)

print(
    "Component atom records:",
    len(df_component_atoms)
)

print(
    "Selected component:",
    (
        f"{SELECTED_COMPONENT_NAME}, "
        f"chain {normalize_chain_id(SELECTED_COMPONENT_CHAIN)}"
        if selected_component is not None
        else "None"
    )
)

print("\nOutput directory:", PDB_RESULTS_DIR)

print("\nSTATUS: OK")
print("CELL 17 COMPLETED")
print("=" * 70)
```

```python

# ============================================================
# CELL 18. BINDING SITE ANALYSIS
# ============================================================
# Назначение:
# 1. Найти аминокислотные остатки рядом с выбранным компонентом.
# 2. Рассчитать минимальные расстояния между атомами.
# 3. Определить количество атомных контактов.
# 4. Сформировать таблицы области взаимодействия.
# 5. Выполнить статистику по белковым цепям.
# 6. Сохранить результаты в CSV и JSON.
#
# Пакетная обработка:
# - результаты сохраняются в PDB_RESULTS_DIR текущего PDB;
# - используется первая модель, как в CELL 4–8;
# - при отсутствии выбранного компонента анализ пропускается;
# - пустые таблицы также сохраняются.
#
# Методика:
# - используются тяжелые атомы;
# - водород и дейтерий исключаются;
# - для DisorderedAtom используется выбранный Bio.PDB конформер;
# - контакт определяется как расстояние <= BINDING_CUTOFF;
# - контакт не является доказательством химической связи.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.PDB import is_aa


print("=" * 70)
print("CELL 18. BINDING SITE ANALYSIS")
print("=" * 70)


# ------------------------------------------------------------
# 1. ПРОВЕРКА ВХОДНЫХ ДАННЫХ
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "PDB_RESULTS_DIR",
    "selected_component",
    "structure"
]

for variable in required_variables:

    if variable not in globals():
        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()

if len(PDB_ID) != 4 or not PDB_ID.isalnum():
    raise ValueError(f"Некорректный PDB_ID: {PDB_ID}")


PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# Проверяем, что текущая структура соответствует PDB_ID.
structure_id = str(
    getattr(structure, "id", "")
).strip().upper()

if structure_id and structure_id != PDB_ID:
    raise ValueError(
        f"Несоответствие структуры: "
        f"structure.id={structure_id}, PDB_ID={PDB_ID}"
    )


models = list(structure.get_models())

if not models:
    raise ValueError(
        f"В структуре {PDB_ID} отсутствуют модели."
    )


# В соответствии с CELL 4–8 используем первую модель.
first_model = models[0]


# ------------------------------------------------------------
# 2. ПАРАМЕТРЫ АНАЛИЗА
# ------------------------------------------------------------

BINDING_CUTOFF = 5.0  # Å

HYDROGEN_ELEMENTS = {
    "H",
    "D"
}

ALTLOC_POLICY = (
    "Bio.PDB selected conformer; "
    "alternative conformers are not counted simultaneously"
)


# ------------------------------------------------------------
# 3. ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ
# ------------------------------------------------------------

def get_heavy_atoms(residue):
    """
    Возвращает тяжелые атомы остатка.

    Используется get_atoms(), а не get_unpacked_list(),
    чтобы не учитывать альтернативные конформации
    одновременно.
    """

    result = []

    for atom in residue.get_atoms():

        element = str(
            atom.element
        ).strip().upper()

        if element in HYDROGEN_ELEMENTS:
            continue

        result.append(atom)

    return result


def get_atom_element(atom):

    element = str(
        atom.element
    ).strip().upper()

    return element if element else "UNKNOWN"


def get_atom_altloc(atom):

    altloc = atom.get_altloc()

    if altloc is None:
        return ""

    return str(altloc).strip()


def normalize_chain_id(chain_id):

    value = str(chain_id)

    if not value.strip():
        return "<blank>"

    return value


def get_residue_key(chain, residue):

    """
    Формирует идентификатор остатка.

    Включает:
    - цепь;
    - hetero flag;
    - номер остатка;
    - код вставки.
    """

    hetero_flag, residue_number, insertion_code = residue.id

    return (
        f"{chain.id}:"
        f"{hetero_flag}:"
        f"{residue_number}:"
        f"{str(insertion_code).strip()}"
    )


def json_safe(value):
    """Преобразует значения NumPy/Pandas в JSON-совместимый формат."""

    if isinstance(value, dict):

        return {
            str(key): json_safe(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple)):

        return [
            json_safe(item)
            for item in value
        ]

    if isinstance(value, np.ndarray):

        return json_safe(value.tolist())

    if isinstance(value, np.integer):

        return int(value)

    if isinstance(value, np.floating):

        value = float(value)

        return value if np.isfinite(value) else None

    if isinstance(value, float):

        return value if np.isfinite(value) else None

    if isinstance(value, np.bool_):

        return bool(value)

    if value is None:
        return None

    if isinstance(value, (str, int, bool)):
        return value

    try:

        if pd.isna(value):
            return None

    except (TypeError, ValueError):

        pass

    return str(value)


# ------------------------------------------------------------
# 4. ИНИЦИАЛИЗАЦИЯ
# ------------------------------------------------------------

binding_data = []
atom_contact_data = []

protein_residue_count = 0
checked_residue_count = 0
invalid_residue_count = 0
residues_without_heavy_atoms = 0

selected_component_available = (
    selected_component is not None
)

selected_component_name = None
selected_chain = None
selected_model_id = None
selected_model_serial = None
selected_component_number = None
selected_component_index_value = None
selected_component_id = None

component_atoms = []
component_coordinates = np.empty((0, 3), dtype=float)


# ------------------------------------------------------------
# 5. ПОДГОТОВКА ВЫБРАННОГО КОМПОНЕНТА
# ------------------------------------------------------------

if selected_component_available:

    # Получаем родительские объекты компонента.
    selected_component_chain = (
        selected_component.get_parent()
    )

    selected_model = (
        selected_component_chain.get_parent()
    )

    # Компонент должен принадлежать первой модели
    # именно текущей структуры.
    if selected_model is not first_model:

        raise ValueError(
            "Выбранный компонент не принадлежит первой "
            "модели текущей структуры."
        )

    # Дополнительная проверка принадлежности компонента.
    component_found = any(

        residue is selected_component

        for chain in first_model

        for residue in chain

    )

    if not component_found:

        raise ValueError(
            "Выбранный компонент не найден "
            "в текущей структуре."
        )

    selected_component_name = (
        selected_component.resname.strip().upper()
    )

    selected_chain = str(
        selected_component_chain.id
    )

    selected_model_id = selected_model.id

    selected_model_serial = getattr(
        selected_model,
        "serial_num",
        None
    )

    selected_residue_id = selected_component.id

    selected_component_number = (
        selected_residue_id[1]
    )

    selected_component_index_value = globals().get(
        "selected_component_index",
        None
    )

    if selected_component_index_value is not None:

        selected_component_index_value = int(
            selected_component_index_value
        )

    selected_component_id = (

        f"{PDB_ID}|"
        f"model_0|"
        f"{selected_chain}|"
        f"{selected_residue_id[0]}|"
        f"{selected_residue_id[1]}|"
        f"{str(selected_residue_id[2]).strip()}|"
        f"{selected_component_name}"

    )

    component_atoms = get_heavy_atoms(
        selected_component
    )

    if not component_atoms:

        raise ValueError(
            "Выбранный компонент не содержит тяжелых атомов."
        )

    component_coordinates = np.asarray(

        [
            atom.coord
            for atom in component_atoms
        ],

        dtype=float

    )

    if component_coordinates.ndim != 2:
        raise ValueError(
            "Некорректная размерность координат компонента."
        )

    if component_coordinates.shape[1] != 3:
        raise ValueError(
            "Координаты компонента должны иметь три измерения."
        )

    if not np.isfinite(component_coordinates).all():

        raise ValueError(
            "В координатах выбранного компонента "
            "обнаружены NaN или Inf."
        )

    print("PDB ID:", PDB_ID)

    print(
        "Selected component:",
        selected_component_name
    )

    print(
        "Component chain:",
        normalize_chain_id(selected_chain)
    )

    print(
        "Component residue number:",
        selected_component_number
    )

    print(
        "Model ID:",
        selected_model_id
    )

    print(
        "Component heavy atoms:",
        len(component_atoms)
    )

    print(
        "Search radius:",
        BINDING_CUTOFF,
        "Å"
    )

else:

    selected_model = None

    print("PDB ID:", PDB_ID)

    print("No component selected in CELL 17.")

    print("Binding site analysis will be skipped.")


# ------------------------------------------------------------
# 6. ПОИСК БЕЛКОВЫХ ОСТАТКОВ ВОКРУГ КОМПОНЕНТА
# ------------------------------------------------------------

if selected_component_available:

    for chain in selected_model:

        for residue in chain:

            if not is_aa(
                residue,
                standard=False
            ):
                continue

            # Считаем все распознанные белковые остатки.
            protein_residue_count += 1

            # Выбранный компонент не должен анализироваться
            # как белковый остаток.
            if residue is selected_component:
                continue

            protein_atoms = get_heavy_atoms(
                residue
            )

            if not protein_atoms:

                residues_without_heavy_atoms += 1

                continue

            protein_coordinates = np.asarray(

                [
                    atom.coord
                    for atom in protein_atoms
                ],

                dtype=float

            )

            checked_residue_count += 1

            # Проверяем координаты до расчета расстояний.
            if (
                protein_coordinates.ndim != 2
                or protein_coordinates.shape[1] != 3
                or not np.isfinite(protein_coordinates).all()
            ):

                invalid_residue_count += 1

                continue

            # ------------------------------------------------
            # Матрица расстояний:
            # строки — атомы белкового остатка;
            # столбцы — атомы выбранного компонента.
            # ------------------------------------------------

            distances = np.linalg.norm(

                protein_coordinates[:, None, :]
                -
                component_coordinates[None, :, :],

                axis=2

            )

            if not np.isfinite(distances).all():

                invalid_residue_count += 1

                continue

            # Минимальное расстояние.
            min_flat_index = int(
                np.argmin(distances)
            )

            protein_atom_index, component_atom_index = (
                np.unravel_index(
                    min_flat_index,
                    distances.shape
                )
            )

            min_distance = float(
                distances[
                    protein_atom_index,
                    component_atom_index
                ]
            )

            # Матрица контактных пар.
            contact_mask = (
                distances <= BINDING_CUTOFF
            )

            atom_contacts = int(
                contact_mask.sum()
            )

            # Число уникальных атомов белкового остатка,
            # участвующих хотя бы в одном контакте.
            protein_contact_atoms = int(
                np.any(
                    contact_mask,
                    axis=1
                ).sum()
            )

            # Число уникальных атомов компонента,
            # контактирующих с данным остатком.
            component_contact_atoms = int(
                np.any(
                    contact_mask,
                    axis=0
                ).sum()
            )

            # Остатки вне радиуса не включаются в результат.
            if min_distance > BINDING_CUTOFF:
                continue

            closest_protein_atom = protein_atoms[
                protein_atom_index
            ]

            closest_component_atom = component_atoms[
                component_atom_index
            ]

            hetero_flag, residue_number, insertion_code = (
                residue.id
            )

            insertion_code = str(
                insertion_code
            ).strip()

            residue_name = (
                residue.resname.strip().upper()
            )

            residue_key = get_residue_key(
                chain,
                residue
            )

            chain_id_display = normalize_chain_id(
                chain.id
            )

            # ------------------------------------------------
            # 6.1. Информация об остатке
            # ------------------------------------------------

            binding_data.append({

                "pdb_id": PDB_ID,

                "component_id": selected_component_id,

                "component_index": (
                    selected_component_index_value
                ),

                "component": selected_component_name,

                "component_chain": selected_chain,

                "component_residue_number": (
                    selected_component_number
                ),

                "model_id": selected_model_id,

                "protein_chain": str(chain.id),

                "protein_chain_display": chain_id_display,

                "residue_key": residue_key,

                "hetero_flag": str(hetero_flag),

                "residue_name": residue_name,

                "residue_number": residue_number,

                "insertion_code": insertion_code,

                "min_distance_A": min_distance,

                "atom_contacts": atom_contacts,

                "protein_contact_atoms": (
                    protein_contact_atoms
                ),

                "component_contact_atoms": (
                    component_contact_atoms
                ),

                "closest_protein_atom": (
                    closest_protein_atom.get_name()
                ),

                "closest_protein_element": (
                    get_atom_element(
                        closest_protein_atom
                    )
                ),

                "closest_protein_altloc": (
                    get_atom_altloc(
                        closest_protein_atom
                    )
                ),

                "closest_component_atom": (
                    closest_component_atom.get_name()
                ),

                "closest_component_element": (
                    get_atom_element(
                        closest_component_atom
                    )
                ),

                "closest_component_altloc": (
                    get_atom_altloc(
                        closest_component_atom
                    )
                )

            })

            # ------------------------------------------------
            # 6.2. Информация об атомных парах
            # ------------------------------------------------

            contact_indices = np.argwhere(
                contact_mask
            )

            for protein_index, component_index in contact_indices:

                protein_atom = protein_atoms[
                    protein_index
                ]

                component_atom = component_atoms[
                    component_index
                ]

                # Уникальный идентификатор атома компонента.
                component_atom_id = (

                    f"{selected_component_id}|"
                    f"{component_atom.get_name().strip()}|"
                    f"{get_atom_element(component_atom)}|"
                    f"{get_atom_altloc(component_atom)}"

                )

                atom_contact_data.append({

                    "pdb_id": PDB_ID,

                    "component_id": selected_component_id,

                    "component_index": (
                        selected_component_index_value
                    ),

                    "component": selected_component_name,

                    "component_chain": selected_chain,

                    "component_residue_number": (
                        selected_component_number
                    ),

                    "model_id": selected_model_id,

                    "protein_chain": str(chain.id),

                    "protein_chain_display": chain_id_display,

                    "residue_name": residue_name,

                    "residue_number": residue_number,

                    "insertion_code": insertion_code,

                    "residue_key": residue_key,

                    "protein_atom": (
                        protein_atom.get_name().strip()
                    ),

                    "protein_element": (
                        get_atom_element(protein_atom)
                    ),

                    "protein_altloc": (
                        get_atom_altloc(protein_atom)
                    ),

                    "component_atom": (
                        component_atom.get_name().strip()
                    ),

                    "component_element": (
                        get_atom_element(component_atom)
                    ),

                    "component_altloc": (
                        get_atom_altloc(component_atom)
                    ),

                    "component_atom_id": component_atom_id,

                    "distance_A": float(
                        distances[
                            protein_index,
                            component_index
                        ]
                    )

                })


# ------------------------------------------------------------
# 7. СОЗДАНИЕ ТАБЛИЦ РЕЗУЛЬТАТОВ
# ------------------------------------------------------------

binding_columns = [

    "pdb_id",
    "component_id",
    "component_index",
    "component",
    "component_chain",
    "component_residue_number",
    "model_id",
    "protein_chain",
    "protein_chain_display",
    "residue_key",
    "hetero_flag",
    "residue_name",
    "residue_number",
    "insertion_code",
    "min_distance_A",
    "atom_contacts",
    "protein_contact_atoms",
    "component_contact_atoms",
    "closest_protein_atom",
    "closest_protein_element",
    "closest_protein_altloc",
    "closest_component_atom",
    "closest_component_element",
    "closest_component_altloc"

]


atom_contact_columns = [

    "pdb_id",
    "component_id",
    "component_index",
    "component",
    "component_chain",
    "component_residue_number",
    "model_id",
    "protein_chain",
    "protein_chain_display",
    "residue_name",
    "residue_number",
    "insertion_code",
    "residue_key",
    "protein_atom",
    "protein_element",
    "protein_altloc",
    "component_atom",
    "component_element",
    "component_altloc",
    "component_atom_id",
    "distance_A"

]


df_binding = pd.DataFrame(
    binding_data,
    columns=binding_columns
)

df_binding_atom_contacts = pd.DataFrame(
    atom_contact_data,
    columns=atom_contact_columns
)


if not df_binding.empty:

    df_binding = (

        df_binding
        .sort_values(

            [
                "min_distance_A",
                "protein_chain",
                "residue_number"
            ]

        )
        .reset_index(drop=True)

    )


if not df_binding_atom_contacts.empty:

    df_binding_atom_contacts = (

        df_binding_atom_contacts
        .sort_values(

            [
                "distance_A",
                "protein_chain",
                "residue_number"
            ]

        )
        .reset_index(drop=True)

    )


print("\n" + "=" * 70)
print("BINDING SITE RESULTS")
print("=" * 70)


if not df_binding.empty:

    print(
        "Nearby amino acid residues:",
        len(df_binding)
    )

    print(
        "Atom-pair contacts:",
        len(df_binding_atom_contacts)
    )

    print("\nNearby protein residues:")

    display(df_binding)

    print("\nIndividual atom contacts:")

    display(
        df_binding_atom_contacts.head(30)
    )

else:

    print(
        "No amino acid residues found "
        "within the specified radius."
    )


# ------------------------------------------------------------
# 8. СВОДКА ПО БЕЛКОВЫМ ЦЕПЯМ
# ------------------------------------------------------------

chain_columns = [

    "protein_chain",
    "protein_chain_display",
    "number_of_residues",
    "minimum_distance_A",
    "maximum_residue_min_distance_A",
    "total_atom_contacts",
    "protein_contact_atoms",
    "unique_component_contact_atoms"

]


chain_records = []

if not df_binding.empty:

    for chain_id, chain_group in df_binding.groupby(
        "protein_chain",
        dropna=False
    ):

        chain_contacts = df_binding_atom_contacts[
            df_binding_atom_contacts["protein_chain"]
            == chain_id
        ]

        unique_component_atoms = (

            chain_contacts["component_atom_id"].nunique()

            if not chain_contacts.empty

            else 0

        )

        chain_records.append({

            "protein_chain": chain_id,

            "protein_chain_display": (
                chain_group["protein_chain_display"].iloc[0]
            ),

            "number_of_residues": len(chain_group),

            "minimum_distance_A": float(
                chain_group["min_distance_A"].min()
            ),

            # Это максимум минимальных расстояний
            # среди остатков, попавших в радиус.
            "maximum_residue_min_distance_A": float(
                chain_group["min_distance_A"].max()
            ),

            "total_atom_contacts": int(
                chain_group["atom_contacts"].sum()
            ),

            "protein_contact_atoms": int(
                chain_group["protein_contact_atoms"].sum()
            ),

            "unique_component_contact_atoms": int(
                unique_component_atoms
            )

        })


df_binding_by_chain = pd.DataFrame(
    chain_records,
    columns=chain_columns
)


print("\nBinding site summary by chain:")

if not df_binding_by_chain.empty:

    display(df_binding_by_chain)

else:

    print("No contacts to summarize.")


# ------------------------------------------------------------
# 9. ОБЩАЯ СТАТИСТИКА
# ------------------------------------------------------------

if not df_binding.empty:

    minimum_distance = float(
        df_binding["min_distance_A"].min()
    )

    maximum_residue_min_distance = float(
        df_binding["min_distance_A"].max()
    )

    total_atom_contacts = int(
        df_binding["atom_contacts"].sum()
    )

    average_contacts = float(
        df_binding["atom_contacts"].mean()
    )

    # Уникальные атомы белка.
    unique_protein_contact_atoms = int(

        df_binding_atom_contacts[

            [
                "protein_chain",
                "residue_key",
                "protein_atom",
                "protein_element",
                "protein_altloc"
            ]

        ]
        .drop_duplicates()
        .shape[0]

    )

    # Уникальные атомы выбранного компонента.
    unique_component_contact_atoms = int(
        df_binding_atom_contacts[
            "component_atom_id"
        ].nunique()
    )

else:

    minimum_distance = None

    maximum_residue_min_distance = None

    total_atom_contacts = 0

    average_contacts = None

    unique_protein_contact_atoms = 0

    unique_component_contact_atoms = 0


analysis_status = (

    "COMPLETED"
    if selected_component_available
    else "SKIPPED_NO_COMPONENT"

)


binding_summary = {

    "PDB_ID": PDB_ID,

    "analysis_status": analysis_status,

    "component": selected_component_name,

    "component_id": selected_component_id,

    "component_index": selected_component_index_value,

    "component_chain": selected_chain,

    "component_residue_number": selected_component_number,

    "model_id": selected_model_id,

    "model_serial_number": selected_model_serial,

    "binding_cutoff_A": BINDING_CUTOFF,

    "hydrogen_atoms_excluded": True,

    "altloc_policy": ALTLOC_POLICY,

    "protein_residue_count": protein_residue_count,

    "checked_residue_count": checked_residue_count,

    "residues_without_heavy_atoms": (
        residues_without_heavy_atoms
    ),

    "invalid_residue_count": invalid_residue_count,

    "nearby_residue_count": len(df_binding),

    "atom_pair_contact_count": total_atom_contacts,

    "unique_protein_contact_atom_count": (
        unique_protein_contact_atoms
    ),

    "unique_component_contact_atom_count": (
        unique_component_contact_atoms
    ),

    "minimum_distance_A": minimum_distance,

    "maximum_residue_min_distance_A": (
        maximum_residue_min_distance
    ),

    "average_contacts_per_residue": average_contacts,

    "analysis_scope": (
        "Protein amino acid residues in the first model "
        "around the selected component"
    ),

    "distance_interpretation": (
        "maximum_residue_min_distance_A is the largest "
        "of the per-residue minimum distances; it is not "
        "the maximum distance among all atom pairs"
    ),

    "interpretation": (
        "Spatial contacts within the specified cutoff; "
        "not proof of chemical bonding"
    )

}


print("\nOVERALL STATISTICS")
print("-" * 70)

print("Analysis status:", analysis_status)

print(
    "Protein residues in model:",
    protein_residue_count
)

print(
    "Checked protein residues:",
    checked_residue_count
)

print(
    "Residues without heavy atoms:",
    residues_without_heavy_atoms
)

print(
    "Residues with invalid coordinates:",
    invalid_residue_count
)

print(
    "Nearby amino acid residues:",
    len(df_binding)
)

print(
    "Total atom-pair contacts:",
    total_atom_contacts
)

print(
    "Unique protein contact atoms:",
    unique_protein_contact_atoms
)

print(
    "Unique component contact atoms:",
    unique_component_contact_atoms
)

if minimum_distance is not None:

    print(
        "Minimum distance:",
        round(minimum_distance, 3),
        "Å"
    )

    print(
        "Maximum residue minimum distance:",
        round(maximum_residue_min_distance, 3),
        "Å"
    )

    print(
        "Average contacts per residue:",
        round(average_contacts, 2)
    )


# ------------------------------------------------------------
# 10. СОХРАНЕНИЕ CSV
# ------------------------------------------------------------

BINDING_FILE = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_binding_site.csv"
)

BINDING_ATOM_CONTACTS_FILE = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_binding_atom_contacts.csv"
)

BINDING_SUMMARY_FILE = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_binding_site_summary.csv"
)


df_binding.to_csv(
    BINDING_FILE,
    index=False
)

df_binding_atom_contacts.to_csv(
    BINDING_ATOM_CONTACTS_FILE,
    index=False
)

df_binding_by_chain.to_csv(
    BINDING_SUMMARY_FILE,
    index=False
)


# ------------------------------------------------------------
# 11. СОХРАНЕНИЕ JSON
# ------------------------------------------------------------

BINDING_JSON_FILE = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_binding_site.json"
)


with open(
    BINDING_JSON_FILE,
    "w",
    encoding="utf-8"
) as file:

    json.dump(

        json_safe(binding_summary),

        file,

        ensure_ascii=False,

        indent=4,

        allow_nan=False

    )


# ------------------------------------------------------------
# 12. ПРОВЕРКИ
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("VALIDATION")
print("=" * 70)


# 12.1. Проверка расстояний.
if not df_binding_atom_contacts.empty:

    if not (
        df_binding_atom_contacts["distance_A"]
        <= BINDING_CUTOFF
    ).all():

        raise ValueError(
            "Some atom contacts exceed the specified cutoff."
        )

    if (
        df_binding_atom_contacts["distance_A"] < 0
    ).any():

        raise ValueError(
            "Negative interatomic distance detected."
        )


print("Contact distance threshold: OK")


# 12.2. Проверка количества атомных пар.
if (
    len(df_binding_atom_contacts)
    != total_atom_contacts
):

    raise ValueError(
        "Mismatch between atom contact records "
        "and calculated contact count."
    )


print("Atom-pair contact count: OK")


# 12.3. Проверка количества остатков.
if len(df_binding) != len(binding_data):

    raise ValueError(
        "Mismatch in binding residue count."
    )


print("Binding residue count: OK")


# 12.4. Проверка уникальных атомов компонента.
if not df_binding_atom_contacts.empty:

    calculated_unique_component_atoms = int(
        df_binding_atom_contacts[
            "component_atom_id"
        ].nunique()
    )

    if (
        calculated_unique_component_atoms
        != unique_component_contact_atoms
    ):

        raise ValueError(
            "Mismatch in unique component contact atom count."
        )


print("Unique component contact atoms: OK")


# 12.5. Проверка согласованности таблиц.
if not df_binding_atom_contacts.empty:

    contact_residues = set(
        df_binding_atom_contacts["residue_key"]
    )

    binding_residues = set(
        df_binding["residue_key"]
    )

    if not contact_residues.issubset(binding_residues):

        raise ValueError(
            "Some atom contacts refer to residues "
            "missing from the binding site table."
        )


print("Contact-to-residue consistency: OK")


# 12.6. Проверка выходных файлов.
saved_files = [

    BINDING_FILE,

    BINDING_ATOM_CONTACTS_FILE,

    BINDING_SUMMARY_FILE,

    BINDING_JSON_FILE

]


for path in saved_files:

    if not path.is_file():

        raise RuntimeError(
            f"Output file was not saved: {path}"
        )

    if path.stat().st_size == 0:

        raise RuntimeError(
            f"Output file is empty: {path}"
        )


print("Output files: OK")


# ------------------------------------------------------------
# 13. ИТОГ
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("BINDING SITE ANALYSIS COMPLETED")
print("=" * 70)

print("PDB ID:", PDB_ID)

print("Analysis status:", analysis_status)

print(
    "Selected component:",
    (
        selected_component_name
        if selected_component_available
        else "None"
    )
)

print(
    "Component chain:",
    (
        normalize_chain_id(selected_chain)
        if selected_component_available
        else "None"
    )
)

print(
    "Binding cutoff:",
    BINDING_CUTOFF,
    "Å"
)

print("Nearby residues:", len(df_binding))

print("Atom-pair contacts:", total_atom_contacts)

print(
    "Unique protein contact atoms:",
    unique_protein_contact_atoms
)

print(
    "Unique component contact atoms:",
    unique_component_contact_atoms
)

print("\nSaved files:")

for path in saved_files:

    print(
        f"- {path.name}: "
        f"{'OK' if path.is_file() else 'ERROR'}"
    )


print("\nOutput directory:", PDB_RESULTS_DIR)

print("\nSTATUS:", analysis_status)

print("CELL 18 COMPLETED")
print("=" * 70)
```

Результат ячейки: таблица ближайших аминокислотных остатков с указанием цепи, номера остатка, минимального расстояния до гема и числа атомных контактов в пределах 5 Å.
3. Что означают новые показатели
Показатель	Значение
min_distance_A	Минимальное расстояние между атомами остатка и компонента
atom_contacts	Количество пар атомов на расстоянии не более 5 Å
protein_contact_atoms	Количество уникальных атомов белка, участвующих в контактах
component_contact_atoms	Количество уникальных атомов компонента, участвующих в контактах
closest_protein_atom	Название ближайшего атома белка
closest_component_atom	Название ближайшего атома компонента

4. Важное методическое уточнение
Радиус 5 Å — это критерий поиска соседних остатков, а не доказательство химического взаимодействия.

Например:
- расстояние 3,0 Å может соответствовать потенциальной водородной связи, но нужно проверить типы атомов и геометрию;
- расстояние 4,5 Å указывает на пространственную близость, но само по себе не доказывает наличие конкретного взаимодействия;
- контакты с металлами требуют отдельного анализа координационного окружения.
Также новая версия анализирует только модель, в которой находится выбранный компонент. Это важно для структур с несколькими моделями.

```python

# ============================================================
# CELL 19. PROTEIN-COMPONENT INTERACTION ANALYSIS
# ============================================================
# Назначение:
# 1. Найти атомные контакты выбранного компонента с белком.
# 2. Выделить потенциальные контакты полярных атомов.
# 3. Выделить потенциальные гидрофобные контакты.
# 4. Найти геометрические контакты Fe в гемах.
# 5. Выполнить универсальный поиск контактов распознаваемых
#    металлов.
# 6. Найти контакты Fe с компонентами OXY.
# 7. Найти потенциальные дисульфидные мостики CYS-SG-SG.
# 8. Сохранить результаты в CSV и JSON.
#
# Согласованность:
# - анализируется первая модель, как в CELL 4-18;
# - используется выбранный Bio.PDB конформер;
# - водородные атомы исключаются;
# - результаты сохраняются в PDB_RESULTS_DIR.
#
# ВАЖНО:
# df_metal_coordination:
#   контакты Fe в распознанных гемовых компонентах.
#
# df_all_metal_coordination:
#   контакты всех распознаваемых металлов.
#
# Геометрические контакты не являются доказательством
# химической связи или водородной связи.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.PDB import is_aa


print("=" * 70)
print("CELL 19. PROTEIN-COMPONENT INTERACTION ANALYSIS")
print("=" * 70)


# ============================================================
# 1. PARAMETERS
# ============================================================

CONTACT_CUTOFF = 4.0
POLAR_CUTOFF = 3.5
HYDROPHOBIC_CUTOFF = 4.0

METAL_COORDINATION_CUTOFF = 2.8
DISULFIDE_CUTOFF = 2.3

METAL_ELEMENTS = {
    "FE", "ZN", "MG", "MN", "CU",
    "CO", "NI", "CA", "CD", "HG"
}

POLAR_ELEMENTS = {"N", "O", "S"}
HYDROPHOBIC_ELEMENTS = {"C", "S"}

HYDROGEN_ELEMENTS = {"H", "D"}

WATER_CODES = {"HOH", "WAT", "DOD"}

HEME_COMPONENT_CODES = {
    "HEM", "HEA", "HEB", "HEC",
    "HEO", "HEF", "HEG", "HEV"
}

OXYGEN_COMPONENT_CODES = {
    "OXY", "O2"
}


# ============================================================
# 2. INPUT VALIDATION
# ============================================================

required_variables = [
    "PDB_ID",
    "PDB_RESULTS_DIR",
    "component_structure"
]

missing_variables = [
    name
    for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise NameError(
        "Не найдены обязательные переменные: "
        + ", ".join(missing_variables)
    )


PDB_ID = str(PDB_ID).strip().upper()

if len(PDB_ID) != 4 or not PDB_ID.isalnum():
    raise ValueError(
        f"Некорректный PDB_ID: {PDB_ID}"
    )


PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


if component_structure is None:
    raise ValueError(
        "component_structure не содержит структуру."
    )


structure_id = str(
    getattr(component_structure, "id", "")
).strip().upper()

if structure_id and structure_id != PDB_ID:
    raise ValueError(
        f"Несоответствие PDB ID: "
        f"component_structure.id={structure_id}, "
        f"PDB_ID={PDB_ID}"
    )


models = list(
    component_structure.get_models()
)

if not models:
    raise ValueError(
        "В структуре не обнаружено ни одной модели."
    )


# Используем первую модель для согласованности
# с CELL 4-18.

analysis_model = models[0]

analysis_model_id = analysis_model.id

analysis_model_serial = getattr(
    analysis_model,
    "serial_num",
    None
)


# Выбранный компонент может отсутствовать.
# Это допустимо: остальные виды анализа всё равно выполняются.

selected_component = globals().get(
    "selected_component",
    None
)


# ============================================================
# 3. HELPER FUNCTIONS
# ============================================================

def get_atom_element(atom):
    """
    Возвращает химический элемент атома.

    Приоритет:
    1. atom.element из Bio.PDB.
    2. Осторожное определение по имени атома.

    Для корректно прочитанного mmCIF предпочтительно
    использовать исходное значение atom.element.
    """

    element = getattr(
        atom,
        "element",
        None
    )

    if element is not None:

        element = str(element).strip().upper()

        if element not in {"", "?", ".", "X"}:
            return element

    atom_name = str(
        atom.get_fullname()
    ).strip().upper()

    letters = "".join(
        character
        for character in atom_name
        if character.isalpha()
    )

    if not letters:
        return "UNKNOWN"

    # Не пытаемся угадывать двухбуквенные элементы
    # только по имени атома: например, CA может означать
    # как кальций, так и атом C-alpha.
    return letters[0]


def get_residue_code(residue):

    return str(
        residue.get_resname()
    ).strip().upper()


def get_residue_number(residue):

    return int(
        residue.id[1]
    )


def get_insertion_code(residue):

    code = residue.id[2]

    if code is None:
        return ""

    return str(code).strip()


def get_hetero_flag(residue):

    return str(
        residue.id[0]
    ).strip()


def get_chain_id(residue):

    return str(
        residue.get_parent().id
    )


def get_model_id(residue):

    return residue.get_parent().get_parent().id


def get_residue_component_id(residue):
    """
    Устойчивый идентификатор остатка/компонента.

    Включает:
    - PDB ID;
    - модель;
    - цепь;
    - hetero-флаг;
    - номер остатка;
    - код вставки;
    - код компонента.
    """

    return (
        f"{PDB_ID}|"
        f"model_{get_model_id(residue)}|"
        f"chain_{get_chain_id(residue)}|"
        f"{get_hetero_flag(residue)}|"
        f"{get_residue_number(residue)}|"
        f"{get_insertion_code(residue)}|"
        f"{get_residue_code(residue)}"
    )


def get_atom_altloc(atom):

    altloc = atom.get_altloc()

    if altloc is None:
        return ""

    return str(altloc).strip()


def get_valid_atoms_and_coordinates(residue):
    """
    Возвращает тяжелые атомы с корректными координатами.

    get_atoms() использует выбранный Bio.PDB конформер
    для DisorderedAtom.
    """

    atoms = []
    coordinates = []

    for atom in residue.get_atoms():

        element = get_atom_element(atom)

        if element in HYDROGEN_ELEMENTS:
            continue

        coordinate = np.asarray(
            atom.coord,
            dtype=float
        )

        if coordinate.shape != (3,):
            continue

        if not np.isfinite(coordinate).all():
            continue

        atoms.append(atom)
        coordinates.append(coordinate)

    if not coordinates:
        return (
            [],
            np.empty((0, 3), dtype=float)
        )

    return (
        atoms,
        np.asarray(coordinates, dtype=float)
    )


def get_component_type(residue):
    """
    Предварительная классификация небелкового компонента.
    """

    code = get_residue_code(residue)

    if code in HEME_COMPONENT_CODES:
        return "HEME"

    if code in OXYGEN_COMPONENT_CODES:
        return "OXY"

    if code in WATER_CODES:
        return "WATER"

    atoms = list(residue.get_atoms())

    if atoms and any(
        get_atom_element(atom) in METAL_ELEMENTS
        for atom in atoms
    ):
        return "METAL_CONTAINING_COMPONENT"

    return "OTHER"


def is_protein_residue(residue):

    try:
        return bool(
            is_aa(
                residue,
                standard=False
            )
        )

    except Exception:
        return False


def is_water_residue(residue):

    return (
        get_residue_code(residue)
        in WATER_CODES
    )


def is_heme_residue(residue):
    """
    Распознает гемовый компонент, содержащий Fe.
    """

    if get_residue_code(residue) not in HEME_COMPONENT_CODES:
        return False

    return any(
        get_atom_element(atom) == "FE"
        for atom in residue.get_atoms()
    )


def get_metal_atoms(residue):

    return [
        atom
        for atom in residue.get_atoms()
        if get_atom_element(atom) in METAL_ELEMENTS
    ]


def get_atom_distance(atom_1, atom_2):

    coord_1 = np.asarray(
        atom_1.coord,
        dtype=float
    )

    coord_2 = np.asarray(
        atom_2.coord,
        dtype=float
    )

    if (
        coord_1.shape != (3,)
        or coord_2.shape != (3,)
        or not np.isfinite(coord_1).all()
        or not np.isfinite(coord_2).all()
    ):
        return np.nan

    return float(
        np.linalg.norm(coord_1 - coord_2)
    )


def get_atom_record(atom, residue):

    return {
        "model_id": get_model_id(residue),
        "chain": get_chain_id(residue),
        "residue": get_residue_code(residue),
        "residue_number": get_residue_number(residue),
        "insertion_code": get_insertion_code(residue),
        "hetero_flag": get_hetero_flag(residue),
        "atom": atom.get_name().strip(),
        "element": get_atom_element(atom),
        "altloc": get_atom_altloc(atom),
        "component_id": get_residue_component_id(residue)
    }


def json_safe(value):

    if isinstance(value, dict):
        return {
            str(key): json_safe(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple)):
        return [
            json_safe(item)
            for item in value
        ]

    if isinstance(value, np.ndarray):
        return json_safe(value.tolist())

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):

        value = float(value)

        return (
            value
            if np.isfinite(value)
            else None
        )

    if isinstance(value, float):
        return (
            value
            if np.isfinite(value)
            else None
        )

    if isinstance(value, np.bool_):
        return bool(value)

    if value is None:
        return None

    if isinstance(value, (str, int, bool)):
        return value

    try:
        if pd.isna(value):
            return None
    except (TypeError, ValueError):
        pass

    return str(value)


# ============================================================
# 4. STRUCTURE COLLECTION
# ============================================================

all_residues = [
    residue
    for chain in analysis_model
    for residue in chain
]


protein_residues = [
    residue
    for residue in all_residues
    if is_protein_residue(residue)
]


water_residues = [
    residue
    for residue in all_residues
    if is_water_residue(residue)
]


nonprotein_residues = [
    residue
    for residue in all_residues
    if (
        not is_protein_residue(residue)
        and not is_water_residue(residue)
    )
]


heme_residues = [
    residue
    for residue in nonprotein_residues
    if is_heme_residue(residue)
]


oxy_residues = [
    residue
    for residue in nonprotein_residues
    if get_residue_code(residue) in OXYGEN_COMPONENT_CODES
]


# ============================================================
# 5. SELECTED COMPONENT VALIDATION
# ============================================================

selected_residue = selected_component

selected_atoms = []

selected_coordinates = np.empty(
    (0, 3),
    dtype=float
)

selected_component_code = None
selected_component_id = None
selected_component_chain = None
selected_component_number = None
selected_component_model_id = None

if selected_residue is not None:

    if not hasattr(selected_residue, "get_atoms"):
        raise TypeError(
            "selected_component должен быть объектом "
            "Residue из Bio.PDB."
        )

    # Проверяем принадлежность именно первой модели.
    component_found = any(
        residue is selected_residue
        for residue in all_residues
    )

    if not component_found:
        raise ValueError(
            "selected_component не принадлежит первой модели "
            "component_structure. Используйте один и тот же "
            "объект структуры в CELL 17 и CELL 19."
        )

    if is_protein_residue(selected_residue):
        raise ValueError(
            "selected_component определен как белковый остаток. "
            "Для этой ячейки требуется небелковый компонент."
        )

    if is_water_residue(selected_residue):
        raise ValueError(
            "Вода не может использоваться как выбранный "
            "лиганд/компонент в данном анализе."
        )

    selected_component_code = get_residue_code(
        selected_residue
    )

    selected_component_id = get_residue_component_id(
        selected_residue
    )

    selected_component_chain = get_chain_id(
        selected_residue
    )

    selected_component_number = get_residue_number(
        selected_residue
    )

    selected_component_model_id = get_model_id(
        selected_residue
    )

    selected_atoms, selected_coordinates = (
        get_valid_atoms_and_coordinates(
            selected_residue
        )
    )

    if not selected_atoms:
        raise ValueError(
            "Выбранный компонент не содержит тяжелых атомов "
            "с корректными координатами."
        )


# ============================================================
# 6. STRUCTURE INFORMATION
# ============================================================

print("\nSTRUCTURE INFORMATION")

print("PDB ID:", PDB_ID)
print("Model count in source:", len(models))
print("Analyzed model ID:", analysis_model_id)
print("Protein residues:", len(protein_residues))
print("Nonprotein components:", len(nonprotein_residues))
print("Water residues:", len(water_residues))
print("HEME components:", len(heme_residues))
print("OXY components:", len(oxy_residues))
print("Selected component:", selected_component_code)


# ============================================================
# 7. SELECTED COMPONENT - PROTEIN INTERACTIONS
# ============================================================

interaction_data = []

if selected_residue is not None:

    for protein_residue in protein_residues:

        protein_atoms, protein_coordinates = (
            get_valid_atoms_and_coordinates(
                protein_residue
            )
        )

        if not protein_atoms:
            continue

        # Матрица расстояний между атомами компонента
        # и атомами белкового остатка.

        distances = np.linalg.norm(
            selected_coordinates[:, None, :]
            -
            protein_coordinates[None, :, :],
            axis=2
        )

        component_indices, protein_indices = np.where(
            distances <= CONTACT_CUTOFF
        )

        for component_index, protein_index in zip(
            component_indices,
            protein_indices
        ):

            component_atom = selected_atoms[
                component_index
            ]

            protein_atom = protein_atoms[
                protein_index
            ]

            component_element = get_atom_element(
                component_atom
            )

            protein_element = get_atom_element(
                protein_atom
            )

            distance = float(
                distances[
                    component_index,
                    protein_index
                ]
            )

            # ------------------------------------------------
            # Предварительная геометрическая классификация.
            #
            # Полярная близость не является доказанной
            # водородной связью.
            #
            # Гидрофобная близость не является доказанным
            # гидрофобным взаимодействием.
            # ------------------------------------------------

            if (
                component_element in POLAR_ELEMENTS
                and protein_element in POLAR_ELEMENTS
                and distance <= POLAR_CUTOFF
            ):

                interaction_type = (
                    "Potential polar-atom proximity"
                )

            elif (
                component_element in HYDROPHOBIC_ELEMENTS
                and protein_element in HYDROPHOBIC_ELEMENTS
                and distance <= HYDROPHOBIC_CUTOFF
            ):

                interaction_type = (
                    "Potential hydrophobic proximity"
                )

            else:

                interaction_type = "Other contact"

            component_record = get_atom_record(
                component_atom,
                selected_residue
            )

            protein_record = get_atom_record(
                protein_atom,
                protein_residue
            )

            interaction_data.append({

                "pdb_id": PDB_ID,

                "model_id": analysis_model_id,

                "component": selected_component_code,

                "component_id": selected_component_id,

                "component_chain": component_record["chain"],

                "component_residue_number": (
                    component_record["residue_number"]
                ),

                "component_atom": component_record["atom"],

                "component_element": component_element,

                "component_altloc": component_record["altloc"],

                "protein_chain": protein_record["chain"],

                "protein_residue": protein_record["residue"],

                "protein_residue_number": (
                    protein_record["residue_number"]
                ),

                "protein_insertion_code": (
                    protein_record["insertion_code"]
                ),

                "protein_atom": protein_record["atom"],

                "protein_element": protein_element,

                "protein_altloc": protein_record["altloc"],

                "protein_component_id": (
                    protein_record["component_id"]
                ),

                "distance": distance,

                "interaction_type": interaction_type

            })


# ============================================================
# 8. UNIVERSAL METAL COORDINATION ANALYSIS
# ============================================================
# Анализируются металлы, представленные в небелковых
# компонентах первой модели.
#
# Соседние атомы:
# - белковые N/O/S;
# - N/O/S того же компонента;
# - N/O/S других небелковых компонентов.
#
# Вода исключена.
#
# Расстояние является геометрическим критерием поиска,
# а не доказательством координационной связи.
# ============================================================

metal_coordination_data = []

for metal_residue in nonprotein_residues:

    metal_atoms = get_metal_atoms(
        metal_residue
    )

    if not metal_atoms:
        continue

    metal_component = get_residue_code(
        metal_residue
    )

    metal_component_id = get_residue_component_id(
        metal_residue
    )

    metal_is_heme = is_heme_residue(
        metal_residue
    )

    for metal_atom in metal_atoms:

        metal_element = get_atom_element(
            metal_atom
        )

        metal_coordinate = np.asarray(
            metal_atom.coord,
            dtype=float
        )

        if not np.isfinite(metal_coordinate).all():
            continue

        metal_record = get_atom_record(
            metal_atom,
            metal_residue
        )

        # ----------------------------------------------------
        # 8.1. Атомы того же компонента
        # ----------------------------------------------------

        for neighbor_atom in metal_residue.get_atoms():

            if neighbor_atom is metal_atom:
                continue

            neighbor_element = get_atom_element(
                neighbor_atom
            )

            if neighbor_element not in POLAR_ELEMENTS:
                continue

            distance = get_atom_distance(
                metal_atom,
                neighbor_atom
            )

            if (
                not np.isfinite(distance)
                or distance > METAL_COORDINATION_CUTOFF
            ):
                continue

            neighbor_record = get_atom_record(
                neighbor_atom,
                metal_residue
            )

            neighbor_type = (
                "Same HEME component"
                if metal_is_heme
                else "Same component"
            )

            metal_coordination_data.append({

                "pdb_id": PDB_ID,

                "model_id": analysis_model_id,

                "metal_element": metal_element,

                "metal_component": metal_component,

                "component": metal_component,

                "metal_chain": metal_record["chain"],

                "metal_residue_number": (
                    metal_record["residue_number"]
                ),

                "metal_atom": metal_record["atom"],

                "metal_component_id": metal_component_id,

                "neighbor_type": neighbor_type,

                "neighbor_component": metal_component,

                "neighbor_component_id": metal_component_id,

                "neighbor_chain": neighbor_record["chain"],

                "neighbor_residue": neighbor_record["residue"],

                "neighbor_residue_number": (
                    neighbor_record["residue_number"]
                ),

                "neighbor_atom": neighbor_record["atom"],

                "neighbor_element": neighbor_element,

                "distance": distance

            })

        # ----------------------------------------------------
        # 8.2. Белковые атомы
        # ----------------------------------------------------

        for protein_residue in protein_residues:

            protein_atoms, _ = (
                get_valid_atoms_and_coordinates(
                    protein_residue
                )
            )

            for neighbor_atom in protein_atoms:

                neighbor_element = get_atom_element(
                    neighbor_atom
                )

                if neighbor_element not in POLAR_ELEMENTS:
                    continue

                distance = get_atom_distance(
                    metal_atom,
                    neighbor_atom
                )

                if (
                    not np.isfinite(distance)
                    or distance > METAL_COORDINATION_CUTOFF
                ):
                    continue

                neighbor_record = get_atom_record(
                    neighbor_atom,
                    protein_residue
                )

                metal_coordination_data.append({

                    "pdb_id": PDB_ID,

                    "model_id": analysis_model_id,

                    "metal_element": metal_element,

                    "metal_component": metal_component,

                    "component": metal_component,

                    "metal_chain": metal_record["chain"],

                    "metal_residue_number": (
                        metal_record["residue_number"]
                    ),

                    "metal_atom": metal_record["atom"],

                    "metal_component_id": metal_component_id,

                    "neighbor_type": "PROTEIN",

                    "neighbor_component": "PROTEIN",

                    "neighbor_component_id": (
                        neighbor_record["component_id"]
                    ),

                    "neighbor_chain": neighbor_record["chain"],

                    "neighbor_residue": neighbor_record["residue"],

                    "neighbor_residue_number": (
                        neighbor_record["residue_number"]
                    ),

                    "neighbor_atom": neighbor_record["atom"],

                    "neighbor_element": neighbor_element,

                    "distance": distance

                })

        # ----------------------------------------------------
        # 8.3. Другие небелковые компоненты
        # ----------------------------------------------------

        for neighbor_residue in nonprotein_residues:

            if neighbor_residue is metal_residue:
                continue

            neighbor_component = get_residue_code(
                neighbor_residue
            )

            neighbor_component_type = get_component_type(
                neighbor_residue
            )

            if neighbor_component_type == "OXY":
                neighbor_type = "OXY"
            else:
                neighbor_type = "OTHER_COMPONENT"

            neighbor_atoms, _ = (
                get_valid_atoms_and_coordinates(
                    neighbor_residue
                )
            )

            for neighbor_atom in neighbor_atoms:

                neighbor_element = get_atom_element(
                    neighbor_atom
                )

                if neighbor_element not in POLAR_ELEMENTS:
                    continue

                distance = get_atom_distance(
                    metal_atom,
                    neighbor_atom
                )

                if (
                    not np.isfinite(distance)
                    or distance > METAL_COORDINATION_CUTOFF
                ):
                    continue

                neighbor_record = get_atom_record(
                    neighbor_atom,
                    neighbor_residue
                )

                metal_coordination_data.append({

                    "pdb_id": PDB_ID,

                    "model_id": analysis_model_id,

                    "metal_element": metal_element,

                    "metal_component": metal_component,

                    "component": metal_component,

                    "metal_chain": metal_record["chain"],

                    "metal_residue_number": (
                        metal_record["residue_number"]
                    ),

                    "metal_atom": metal_record["atom"],

                    "metal_component_id": metal_component_id,

                    "neighbor_type": neighbor_type,

                    "neighbor_component": neighbor_component,

                    "neighbor_component_id": (
                        neighbor_record["component_id"]
                    ),

                    "neighbor_chain": neighbor_record["chain"],

                    "neighbor_residue": neighbor_record["residue"],

                    "neighbor_residue_number": (
                        neighbor_record["residue_number"]
                    ),

                    "neighbor_atom": neighbor_record["atom"],

                    "neighbor_element": neighbor_element,

                    "distance": distance

                })


# ============================================================
# 9. DISULFIDE BRIDGE ANALYSIS
# ============================================================

disulfide_data = []

cysteine_residues = []

for residue in protein_residues:

    if get_residue_code(residue) != "CYS":
        continue

    if "SG" not in residue:
        continue

    sg_atom = residue["SG"]

    coordinate = np.asarray(
        sg_atom.coord,
        dtype=float
    )

    if not np.isfinite(coordinate).all():
        continue

    cysteine_residues.append(
        (residue, sg_atom)
    )


for i in range(len(cysteine_residues)):

    residue_1, atom_1 = cysteine_residues[i]

    for j in range(i + 1, len(cysteine_residues)):

        residue_2, atom_2 = cysteine_residues[j]

        distance = get_atom_distance(
            atom_1,
            atom_2
        )

        if (
            not np.isfinite(distance)
            or distance > DISULFIDE_CUTOFF
        ):
            continue

        disulfide_data.append({

            "pdb_id": PDB_ID,

            "model_id": analysis_model_id,

            "chain_1": get_chain_id(residue_1),

            "residue_1": "CYS",

            "residue_number_1": (
                get_residue_number(residue_1)
            ),

            "insertion_code_1": (
                get_insertion_code(residue_1)
            ),

            "residue_component_id_1": (
                get_residue_component_id(residue_1)
            ),

            "atom_1": "SG",

            "chain_2": get_chain_id(residue_2),

            "residue_2": "CYS",

            "residue_number_2": (
                get_residue_number(residue_2)
            ),

            "insertion_code_2": (
                get_insertion_code(residue_2)
            ),

            "residue_component_id_2": (
                get_residue_component_id(residue_2)
            ),

            "atom_2": "SG",

            "distance": distance,

            "interaction_type": (
                "Potential disulfide bridge"
            )

        })


# ============================================================
# 10. DATAFRAME CREATION
# ============================================================

interaction_columns = [

    "pdb_id",
    "model_id",
    "component",
    "component_id",
    "component_chain",
    "component_residue_number",
    "component_atom",
    "component_element",
    "component_altloc",
    "protein_chain",
    "protein_residue",
    "protein_residue_number",
    "protein_insertion_code",
    "protein_atom",
    "protein_element",
    "protein_altloc",
    "protein_component_id",
    "distance",
    "interaction_type"

]


metal_columns = [

    "pdb_id",
    "model_id",
    "metal_element",
    "metal_component",
    "component",
    "metal_chain",
    "metal_residue_number",
    "metal_atom",
    "metal_component_id",
    "neighbor_type",
    "neighbor_component",
    "neighbor_component_id",
    "neighbor_chain",
    "neighbor_residue",
    "neighbor_residue_number",
    "neighbor_atom",
    "neighbor_element",
    "distance"

]


disulfide_columns = [

    "pdb_id",
    "model_id",
    "chain_1",
    "residue_1",
    "residue_number_1",
    "insertion_code_1",
    "residue_component_id_1",
    "atom_1",
    "chain_2",
    "residue_2",
    "residue_number_2",
    "insertion_code_2",
    "residue_component_id_2",
    "atom_2",
    "distance",
    "interaction_type"

]


df_interactions = pd.DataFrame(
    interaction_data,
    columns=interaction_columns
)


df_all_metal_coordination = pd.DataFrame(
    metal_coordination_data,
    columns=metal_columns
)


df_disulfide_bridges = pd.DataFrame(
    disulfide_data,
    columns=disulfide_columns
)


# ============================================================
# 11. LEGACY HEME IRON TABLE
# ============================================================
# Fe в распознанных гемах.
# Таблица включает контакты Fe с белком и другими
# компонентами, а также с атомами самого гема.
# ============================================================

if not df_all_metal_coordination.empty:

    df_metal_coordination = (

        df_all_metal_coordination[

            (
                df_all_metal_coordination["metal_element"]
                == "FE"
            )

            &

            (
                df_all_metal_coordination["metal_component"]
                .isin(HEME_COMPONENT_CODES)
            )

        ]
        .copy()
        .reset_index(drop=True)

    )

else:

    df_metal_coordination = pd.DataFrame(
        columns=metal_columns
    )


# ============================================================
# 12. FE - OXY ANALYSIS
# ============================================================

if not df_all_metal_coordination.empty:

    df_fe_oxy = (

        df_all_metal_coordination[

            (
                df_all_metal_coordination["metal_element"]
                == "FE"
            )

            &

            (
                df_all_metal_coordination["neighbor_type"]
                == "OXY"
            )

            &

            (
                df_all_metal_coordination["neighbor_element"]
                == "O"
            )

        ]
        .copy()
        .reset_index(drop=True)

    )

else:

    df_fe_oxy = pd.DataFrame(
        columns=metal_columns
    )


# Совместимость с предыдущими именами переменных.

df_fe_oxy_coordination = df_fe_oxy.copy()


if not df_fe_oxy.empty:

    df_fe_oxy_molecules = (

        df_fe_oxy[

            [
                "model_id",
                "metal_component",
                "metal_chain",
                "metal_residue_number",
                "metal_atom",
                "neighbor_component",
                "neighbor_chain",
                "neighbor_residue_number",
                "distance"
            ]

        ]
        .drop_duplicates()
        .reset_index(drop=True)

    )

else:

    df_fe_oxy_molecules = pd.DataFrame(

        columns=[

            "model_id",
            "metal_component",
            "metal_chain",
            "metal_residue_number",
            "metal_atom",
            "neighbor_component",
            "neighbor_chain",
            "neighbor_residue_number",
            "distance"

        ]

    )


# ============================================================
# 13. SORTING
# ============================================================

if not df_interactions.empty:

    df_interactions = (

        df_interactions
        .sort_values(
            by=["distance", "protein_chain"]
        )
        .reset_index(drop=True)

    )


if not df_all_metal_coordination.empty:

    df_all_metal_coordination = (

        df_all_metal_coordination
        .sort_values(

            by=[
                "metal_element",
                "metal_chain",
                "metal_residue_number",
                "distance"
            ]

        )
        .reset_index(drop=True)

    )


if not df_metal_coordination.empty:

    df_metal_coordination = (

        df_metal_coordination
        .sort_values(
            by=[
                "metal_chain",
                "metal_residue_number",
                "distance"
            ]
        )
        .reset_index(drop=True)

    )


if not df_fe_oxy.empty:

    df_fe_oxy = (

        df_fe_oxy
        .sort_values(
            by=["distance"]
        )
        .reset_index(drop=True)

    )

    df_fe_oxy_coordination = df_fe_oxy.copy()


if not df_disulfide_bridges.empty:

    df_disulfide_bridges = (

        df_disulfide_bridges
        .sort_values(
            by=["distance"]
        )
        .reset_index(drop=True)

    )


# ============================================================
# 14. INTERACTION SUMMARY
# ============================================================

interaction_summary_columns = [

    "interaction_type",
    "contact_count",
    "mean_distance",
    "min_distance",
    "max_distance"

]


if not df_interactions.empty:

    df_interaction_summary = (

        df_interactions
        .groupby(
            "interaction_type",
            dropna=False
        )
        .agg(

            contact_count=("distance", "count"),

            mean_distance=("distance", "mean"),

            min_distance=("distance", "min"),

            max_distance=("distance", "max")

        )
        .reset_index()

    )

else:

    df_interaction_summary = pd.DataFrame(
        columns=interaction_summary_columns
    )


# ============================================================
# 15. METAL SUMMARIES
# ============================================================

metal_summary_group_columns = [

    "metal_element",
    "metal_component",
    "neighbor_type",
    "neighbor_element"

]


metal_summary_columns = [

    "metal_element",
    "metal_component",
    "neighbor_type",
    "neighbor_element",
    "coordination_contact_count",
    "mean_distance",
    "min_distance",
    "max_distance"

]


def build_metal_summary(dataframe):

    if dataframe.empty:
        return pd.DataFrame(
            columns=metal_summary_columns
        )

    return (

        dataframe
        .groupby(
            metal_summary_group_columns,
            dropna=False
        )
        .agg(

            coordination_contact_count=("distance", "count"),

            mean_distance=("distance", "mean"),

            min_distance=("distance", "min"),

            max_distance=("distance", "max")

        )
        .reset_index()

    )


df_metal_coordination_summary = build_metal_summary(
    df_metal_coordination
)


df_all_metal_coordination_summary = build_metal_summary(
    df_all_metal_coordination
)


# ============================================================
# 16. FE-OXY SUMMARY
# ============================================================

fe_oxy_summary_columns = [

    "metal_component",
    "neighbor_component",
    "neighbor_element",
    "contact_count",
    "mean_distance",
    "min_distance",
    "max_distance"

]


if not df_fe_oxy.empty:

    df_fe_oxy_summary = (

        df_fe_oxy
        .groupby(

            [
                "metal_component",
                "neighbor_component",
                "neighbor_element"
            ],

            dropna=False

        )
        .agg(

            contact_count=("distance", "count"),

            mean_distance=("distance", "mean"),

            min_distance=("distance", "min"),

            max_distance=("distance", "max")

        )
        .reset_index()

    )

else:

    df_fe_oxy_summary = pd.DataFrame(
        columns=fe_oxy_summary_columns
    )


# ============================================================
# 17. DISULFIDE SUMMARY
# ============================================================

disulfide_summary_columns = [

    "model_id",
    "chain_1",
    "chain_2",
    "bridge_count",
    "mean_distance",
    "min_distance",
    "max_distance"

]


if not df_disulfide_bridges.empty:

    df_disulfide_summary = (

        df_disulfide_bridges
        .groupby(
            ["model_id", "chain_1", "chain_2"],
            dropna=False
        )
        .agg(

            bridge_count=("distance", "count"),

            mean_distance=("distance", "mean"),

            min_distance=("distance", "min"),

            max_distance=("distance", "max")

        )
        .reset_index()

    )

else:

    df_disulfide_summary = pd.DataFrame(
        columns=disulfide_summary_columns
    )


# ============================================================
# 18. VALIDATION
# ============================================================

validation_results = {}


# 18.1. Проверка таблицы Fe в гемах.

if df_metal_coordination.empty:

    legacy_metal_valid = True

else:

    legacy_metal_valid = bool(

        (
            df_metal_coordination["metal_element"] == "FE"
        ).all()

        and

        (
            df_metal_coordination["metal_component"]
            .isin(HEME_COMPONENT_CODES)
        ).all()

        and

        (
            df_metal_coordination["neighbor_element"]
            .isin(POLAR_ELEMENTS)
        ).all()

        and

        (
            df_metal_coordination["distance"].notna()
        ).all()

        and

        (
            df_metal_coordination["distance"]
            <= METAL_COORDINATION_CUTOFF
        ).all()

    )


validation_results[
    "legacy_heme_iron_coordination_valid"
] = legacy_metal_valid


# 18.2. Проверка универсальной таблицы металлов.

if df_all_metal_coordination.empty:

    all_metal_valid = True

else:

    all_metal_valid = bool(

        (
            df_all_metal_coordination["metal_element"]
            .isin(METAL_ELEMENTS)
        ).all()

        and

        (
            df_all_metal_coordination["neighbor_element"]
            .isin(POLAR_ELEMENTS)
        ).all()

        and

        (
            df_all_metal_coordination["distance"].notna()
        ).all()

        and

        (
            df_all_metal_coordination["distance"]
            <= METAL_COORDINATION_CUTOFF
        ).all()

        and

        (
            df_all_metal_coordination["distance"] >= 0
        ).all()

    )


validation_results[
    "universal_metal_coordination_valid"
] = all_metal_valid


# 18.3. Строгая проверка старой таблицы
# как подмножества универсальной.

legacy_subset_columns = [

    "metal_component_id",
    "metal_atom",
    "metal_element",
    "neighbor_component_id",
    "neighbor_atom",
    "neighbor_element",
    "neighbor_type",
    "distance"

]


if df_metal_coordination.empty:

    legacy_subset_valid = True

elif df_all_metal_coordination.empty:

    legacy_subset_valid = False

else:

    legacy_check = df_metal_coordination[
        legacy_subset_columns
    ].merge(

        df_all_metal_coordination[
            legacy_subset_columns
        ].drop_duplicates(),

        on=legacy_subset_columns,

        how="left",

        indicator=True

    )

    legacy_subset_valid = bool(
        (legacy_check["_merge"] == "both").all()
    )


validation_results[
    "legacy_table_subset_of_universal_table"
] = legacy_subset_valid


# 18.4. Проверка Fe-OXY.

if df_fe_oxy.empty:

    fe_oxy_valid = True

else:

    fe_oxy_valid = bool(

        (
            df_fe_oxy["metal_element"] == "FE"
        ).all()

        and

        (
            df_fe_oxy["neighbor_type"] == "OXY"
        ).all()

        and

        (
            df_fe_oxy["neighbor_element"] == "O"
        ).all()

        and

        (
            df_fe_oxy["distance"]
            <= METAL_COORDINATION_CUTOFF
        ).all()

    )


validation_results[
    "fe_oxy_contacts_valid"
] = fe_oxy_valid


# 18.5. Проверка дисульфидных кандидатов.

if df_disulfide_bridges.empty:

    disulfide_valid = True

else:

    disulfide_valid = bool(

        (
            df_disulfide_bridges["distance"].notna()
        ).all()

        and

        (
            df_disulfide_bridges["distance"]
            <= DISULFIDE_CUTOFF
        ).all()

        and

        (
            df_disulfide_bridges["distance"] > 0
        ).all()

        and

        (
            df_disulfide_bridges["atom_1"] == "SG"
        ).all()

        and

        (
            df_disulfide_bridges["atom_2"] == "SG"
        ).all()

    )


validation_results[
    "disulfide_candidates_valid"
] = disulfide_valid


# 18.6. Проверка атомных контактов.

if df_interactions.empty:

    interaction_valid = True

else:

    interaction_valid = bool(

        (
            df_interactions["distance"].notna()
        ).all()

        and

        (
            df_interactions["distance"] >= 0
        ).all()

        and

        (
            df_interactions["distance"]
            <= CONTACT_CUTOFF
        ).all()

    )


validation_results[
    "protein_component_contacts_valid"
] = interaction_valid


# 18.7. Проверка выбранного компонента.

if selected_residue is None:

    selected_component_valid = True

else:

    selected_component_valid = bool(

        selected_component_id is not None

        and selected_component_model_id == analysis_model_id

        and len(selected_atoms) > 0

        and selected_coordinates.shape == (
            len(selected_atoms),
            3
        )

    )


validation_results[
    "selected_component_valid"
] = selected_component_valid


# ============================================================
# 19. SAVE CSV FILES
# ============================================================

legacy_csv_data = {

    f"{PDB_ID}_ligand_interactions.csv":
        df_interactions,

    f"{PDB_ID}_heme_iron_coordination.csv":
        df_metal_coordination,

    f"{PDB_ID}_heme_iron_coordination_summary.csv":
        df_metal_coordination_summary,

    f"{PDB_ID}_interaction_summary.csv":
        df_interaction_summary,

    f"{PDB_ID}_disulfide_bridges.csv":
        df_disulfide_bridges,

    f"{PDB_ID}_disulfide_summary.csv":
        df_disulfide_summary,

    f"{PDB_ID}_fe_oxy_distances.csv":
        df_fe_oxy,

    f"{PDB_ID}_fe_oxy_summary.csv":
        df_fe_oxy_summary

}


universal_csv_data = {

    f"{PDB_ID}_metal_coordination.csv":
        df_all_metal_coordination,

    f"{PDB_ID}_metal_coordination_summary.csv":
        df_all_metal_coordination_summary

}


all_csv_data = {
    **legacy_csv_data,
    **universal_csv_data
}


output_files = []

for filename, dataframe in all_csv_data.items():

    output_path = (
        PDB_RESULTS_DIR / filename
    )

    dataframe.to_csv(
        output_path,
        index=False,
        encoding="utf-8-sig"
    )

    output_files.append(
        output_path
    )


# Проверяем CSV-файлы.

csv_files_exist = all(
    path.is_file() and path.stat().st_size > 0
    for path in output_files
)

validation_results[
    "all_csv_files_exist"
] = csv_files_exist


# ============================================================
# 20. JSON REPORT
# ============================================================

interaction_analysis_report = {

    "pdb_id": PDB_ID,

    "analysis_status": (
        "COMPLETED"
        if selected_residue is not None
        else "COMPLETED_WITHOUT_SELECTED_COMPONENT"
    ),

    "selected_component": {

        "component": selected_component_code,

        "component_id": selected_component_id,

        "chain": selected_component_chain,

        "residue_number": selected_component_number,

        "model_id": selected_component_model_id

    },

    "model_scope": {

        "source_model_count": len(models),

        "analyzed_model_id": analysis_model_id,

        "analyzed_model_serial": analysis_model_serial,

        "scope": "First model only"

    },

    "parameters": {

        "contact_cutoff_A": CONTACT_CUTOFF,

        "polar_atom_proximity_cutoff_A": POLAR_CUTOFF,

        "hydrophobic_proximity_cutoff_A": HYDROPHOBIC_CUTOFF,

        "metal_coordination_cutoff_A":
            METAL_COORDINATION_CUTOFF,

        "disulfide_cutoff_A": DISULFIDE_CUTOFF

    },

    "structure": {

        "protein_residue_count": len(protein_residues),

        "nonprotein_component_count": len(nonprotein_residues),

        "water_residue_count": len(water_residues),

        "heme_component_count": len(heme_residues),

        "oxy_component_count": len(oxy_residues)

    },

    "interaction_analysis": {

        "total_contacts": int(
            len(df_interactions)
        ),

        "potential_polar_atom_proximities": int(

            (
                df_interactions["interaction_type"]
                == "Potential polar-atom proximity"
            ).sum()

        ) if not df_interactions.empty else 0,

        "potential_hydrophobic_proximities": int(

            (
                df_interactions["interaction_type"]
                == "Potential hydrophobic proximity"
            ).sum()

        ) if not df_interactions.empty else 0,

        "other_contacts": int(

            (
                df_interactions["interaction_type"]
                == "Other contact"
            ).sum()

        ) if not df_interactions.empty else 0

    },

    "metal_coordination": {

        "legacy_heme_iron_contact_count": int(
            len(df_metal_coordination)
        ),

        "all_metal_contact_count": int(
            len(df_all_metal_coordination)
        ),

        "recognized_metal_elements": sorted(
            list(METAL_ELEMENTS)
        ),

        "metal_element_distribution": (

            df_all_metal_coordination[
                "metal_element"
            ]
            .value_counts()
            .to_dict()

            if not df_all_metal_coordination.empty

            else {}

        )

    },

    "fe_oxy_analysis": {

        "fe_oxy_contact_count": int(
            len(df_fe_oxy)
        ),

        "oxy_component_count": int(

            df_fe_oxy["neighbor_component_id"].nunique()

        ) if not df_fe_oxy.empty else 0

    },

    "disulfide_analysis": {

        "potential_disulfide_bridge_count": int(
            len(df_disulfide_bridges)
        )

    },

    "validation": validation_results,

    "output_files": [
        path.name
        for path in output_files
    ],

    "interpretation_notes": [

        "Only the first model is analyzed.",

        "Atomic contacts are identified using geometric distance cutoffs.",

        "Polar-atom proximity is not a confirmed hydrogen bond.",

        "Hydrophobic proximity is a preliminary geometric candidate.",

        "Metal coordination candidates are not definitive chemical bonds.",

        "The legacy heme iron table contains contacts of Fe atoms "
        "in recognized heme components.",

        "The universal metal table includes recognized metals "
        "represented in nonprotein components.",

        "Potential disulfide bridges are identified by SG-SG distance "
        "and require additional chemical validation.",

        "Hydrogen atoms are excluded and the selected Bio.PDB "
        "alternate conformer is used.",

        "Water molecules are excluded from metal-neighbor analysis.",

        "The analysis does not establish donor-acceptor chemistry, "
        "protonation states or interaction angles."

    ]

}


# ============================================================
# 21. SAVE JSON REPORT
# ============================================================

json_path = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_interaction_analysis.json"
)


# Сначала сохраняем отчет с текущими проверками.
interaction_analysis_report["validation"] = (
    validation_results
)


with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(

        json_safe(interaction_analysis_report),

        file,

        ensure_ascii=False,

        indent=4,

        allow_nan=False

    )


json_file_valid = (
    json_path.is_file()
    and json_path.stat().st_size > 0
)

validation_results[
    "json_report_exists"
] = json_file_valid


# ============================================================
# 22. FINAL VALIDATION STATUS
# ============================================================

failed_validations = [

    name

    for name, result in validation_results.items()

    if not result

]


interaction_analysis_report["validation"] = (
    validation_results
)

interaction_analysis_report["validation_status"] = (

    "PASSED"
    if not failed_validations
    else "CHECK_REQUIRED"

)


# Обновляем JSON после завершения всех проверок.

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(

        json_safe(interaction_analysis_report),

        file,

        ensure_ascii=False,

        indent=4,

        allow_nan=False

    )


output_files.append(json_path)


# ============================================================
# 23. DISPLAY RESULTS
# ============================================================

print("\n" + "=" * 70)
print("PROTEIN-COMPONENT INTERACTION ANALYSIS")
print("=" * 70)

print("PDB ID:", PDB_ID)

print("Analyzed model:", analysis_model_id)

print("Selected component:", selected_component_code)

print("Protein residues:", len(protein_residues))

print("Nonprotein components:", len(nonprotein_residues))


print("-" * 70)
print("1. SELECTED COMPONENT - PROTEIN INTERACTIONS")

print(
    "Total contacts:",
    len(df_interactions)
)

if not df_interactions.empty:
    display(
        df_interactions.head(20)
    )


print("-" * 70)
print("2. LEGACY HEME IRON CONTACTS")

print(
    "Fe contacts in recognized hemes:",
    len(df_metal_coordination)
)

if not df_metal_coordination.empty:
    display(
        df_metal_coordination.head(20)
    )


print("-" * 70)
print("3. UNIVERSAL METAL CONTACTS")

print(
    "All metal contacts:",
    len(df_all_metal_coordination)
)

if not df_all_metal_coordination.empty:

    display(

        df_all_metal_coordination[
            "metal_element"
        ]
        .value_counts()
        .rename_axis("metal_element")
        .reset_index(name="contact_count")

    )

    display(
        df_all_metal_coordination.head(30)
    )


print("-" * 70)
print("4. FE - OXY CONTACTS")

print(
    "Fe-OXY contacts:",
    len(df_fe_oxy)
)

if not df_fe_oxy.empty:
    display(df_fe_oxy)


print("-" * 70)
print("5. POTENTIAL DISULFIDE BRIDGES")

print(
    "Potential bridges:",
    len(df_disulfide_bridges)
)

if not df_disulfide_bridges.empty:
    display(df_disulfide_bridges)


print("-" * 70)
print("6. VALIDATION")

for validation_name, validation_value in (
    validation_results.items()
):

    status = (
        "OK"
        if validation_value
        else "CHECK"
    )

    print(
        f"{validation_name}: {status}"
    )


print("-" * 70)
print("7. OUTPUT FILES")

for path in output_files:

    print(
        f"[{'OK' if path.is_file() else 'ERROR'}] "
        f"{path.name}"
    )


print("=" * 70)

print(
    "VALIDATION STATUS:",
    "PASSED"
    if not failed_validations
    else "CHECK_REQUIRED"
)

print("CELL 19 COMPLETED")
print("=" * 70)


if failed_validations:

    print("\nWARNING: Validation issues detected:")

    for name in failed_validations:
        print(" -", name)

else:

    print(
        "\nAll CELL 19 validation checks passed."
    )
```

```python

# ============================================================
# CELL 20. FINAL STRUCTURAL AND FUNCTIONAL METRICS
# ============================================================
# Назначение:
# 1. Собрать основные характеристики структуры.
# 2. Рассчитать метрики белковых цепей.
# 3. Сравнить аминокислотные последовательности цепей.
# 4. Систематизировать геометрические характеристики белка.
# 5. Систематизировать показатели качества структуры.
# 6. Оценить состав небелковых компонентов.
# 7. Систематизировать показатели binding site.
# 8. Систематизировать контакты выбранного компонента с белком.
# 9. Сравнить расстояния Fe-N во всех копиях HEM.
# 10. Систематизировать контакты Fe-OXY.
# 11. Включить результаты анализа дисульфидных мостиков.
# 12. Сформировать сводные таблицы и JSON-отчет.
#
# Согласованность:
# - PDB_RESULTS_DIR: индивидуальная папка структуры;
# - df_atoms: белковые атомы первой модели;
# - df_residues: белковые остатки первой модели;
# - component_structure: исходная структура Bio.PDB;
# - анализ Fe-N и Fe-OXY: первая модель;
# - CELL 19: используются готовые таблицы взаимодействий.
#
# Важно:
# Геометрическая близость не доказывает наличие химической связи.
# Отсутствующие данные не заменяются нулевыми значениями.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.Align import PairwiseAligner
from Bio.PDB import is_aa
from IPython.display import display


# ============================================================
# 1. INPUT VALIDATION
# ============================================================

required_variables = [
    "PDB_ID",
    "PDB_RESULTS_DIR",
    "df_atoms",
    "df_residues",
    "df_hetero",
    "df_water",
    "chain_sequences"
]

missing_variables = [
    name
    for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise NameError(
        "Не найдены обязательные переменные: "
        + ", ".join(missing_variables)
    )


PDB_ID = str(PDB_ID).strip().upper()

if len(PDB_ID) != 4 or not PDB_ID.isalnum():
    raise ValueError(f"Некорректный PDB_ID: {PDB_ID}")


PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


required_dataframes = {
    "df_atoms": df_atoms,
    "df_residues": df_residues,
    "df_hetero": df_hetero,
    "df_water": df_water
}

for name, dataframe in required_dataframes.items():

    if not isinstance(dataframe, pd.DataFrame):
        raise TypeError(f"{name} должен быть pandas DataFrame.")

    if "pdb_id" in dataframe.columns and not dataframe.empty:

        ids = (
            dataframe["pdb_id"]
            .dropna()
            .astype(str)
            .str.upper()
            .unique()
        )

        if len(ids) and any(value != PDB_ID for value in ids):
            raise ValueError(
                f"{name} содержит записи другого PDB: {ids}"
            )


if not isinstance(chain_sequences, dict):
    raise TypeError("chain_sequences должен быть словарем.")


if df_atoms.empty:
    raise ValueError(
        "df_atoms пуст. Невозможно рассчитать геометрию белка."
    )


if not {"x", "y", "z"}.issubset(df_atoms.columns):
    raise ValueError(
        "В df_atoms отсутствуют координаты x, y, z."
    )


# Проверка структуры Bio.PDB, если она доступна.

component_structure_available = (
    "component_structure" in globals()
    and component_structure is not None
)

if component_structure_available:

    structure_id = str(
        getattr(component_structure, "id", "")
    ).strip().upper()

    if structure_id and structure_id != PDB_ID:
        raise ValueError(
            f"component_structure относится к {structure_id}, "
            f"а текущий PDB_ID={PDB_ID}"
        )

    models = list(component_structure.get_models())

    if not models:
        raise ValueError("В component_structure нет моделей.")

    analysis_model = models[0]

    analysis_model_id = analysis_model.id

else:

    models = []

    analysis_model = None

    analysis_model_id = None


print("=" * 75)
print("CELL 20. FINAL STRUCTURAL AND FUNCTIONAL METRICS")
print("=" * 75)

print("PDB ID:", PDB_ID)
print("Output directory:", PDB_RESULTS_DIR)

if component_structure_available:

    print("Models in structure:", len(models))
    print("Analyzed model:", analysis_model_id)


# ============================================================
# 2. HELPER FUNCTIONS
# ============================================================

def numeric_values(series):

    return pd.to_numeric(
        series,
        errors="coerce"
    ).dropna()


def safe_mean(series):

    values = numeric_values(series)

    return (
        float(values.mean())
        if not values.empty
        else np.nan
    )


def safe_median(series):

    values = numeric_values(series)

    return (
        float(values.median())
        if not values.empty
        else np.nan
    )


def safe_std(series):

    values = numeric_values(series)

    return (
        float(values.std())
        if len(values) >= 2
        else np.nan
    )


def safe_percentage(numerator, denominator):

    if pd.isna(numerator) or pd.isna(denominator):
        return np.nan

    if denominator == 0:
        return np.nan

    return float(100.0 * numerator / denominator)


def get_first_existing_column(dataframe, candidates):

    for column in candidates:

        if column in dataframe.columns:
            return column

    return None


def dataframe_or_empty(variable_names):

    if isinstance(variable_names, str):
        variable_names = [variable_names]

    for name in variable_names:

        value = globals().get(name)

        if isinstance(value, pd.DataFrame):
            return value.copy(), name

    return pd.DataFrame(), None


def json_safe(value):

    if isinstance(value, dict):

        return {
            str(key): json_safe(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple)):

        return [
            json_safe(item)
            for item in value
        ]

    if isinstance(value, np.ndarray):
        return json_safe(value.tolist())

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):

        value = float(value)

        return value if np.isfinite(value) else None

    if isinstance(value, float):

        return value if np.isfinite(value) else None

    if isinstance(value, np.bool_):
        return bool(value)

    if isinstance(value, bool):
        return value

    if isinstance(value, Path):
        return str(value)

    if value is None:
        return None

    if isinstance(value, (str, int)):
        return value

    try:

        if pd.isna(value):
            return None

    except (TypeError, ValueError):
        pass

    return str(value)


def read_csv_if_exists(filename):

    path = PDB_RESULTS_DIR / filename

    if not path.exists():
        return pd.DataFrame()

    try:

        return pd.read_csv(path)

    except Exception as error:

        print(
            f"WARNING: Не удалось прочитать {filename}: {error}"
        )

        return pd.DataFrame()


def read_json_if_exists(filename):

    path = PDB_RESULTS_DIR / filename

    if not path.exists():
        return {}

    try:

        with open(path, "r", encoding="utf-8") as file:
            return json.load(file)

    except Exception as error:

        print(
            f"WARNING: Не удалось прочитать {filename}: {error}"
        )

        return {}


def add_metric(
    collection,
    category,
    metric,
    value,
    unit=None,
    status="calculated"
):

    collection.append({

        "category": category,

        "metric": metric,

        "value": json_safe(value),

        "unit": unit,

        "status": status

    })


def normalize_chain(value):

    if pd.isna(value):
        return ""

    return str(value).strip()


def get_atom_element(atom):

    element = getattr(atom, "element", None)

    if element is not None:

        element = str(element).strip().upper()

        if element not in {"", "?", ".", "X"}:
            return element

    atom_name = "".join(
        character
        for character in atom.get_fullname().strip().upper()
        if character.isalpha()
    )

    return atom_name[0] if atom_name else "UNKNOWN"


def get_valid_coordinate(atom):

    coordinate = np.asarray(
        atom.coord,
        dtype=float
    )

    if coordinate.shape != (3,):
        return None

    if not np.isfinite(coordinate).all():
        return None

    return coordinate


# ============================================================
# 3. GLOBAL STRUCTURE METRICS
# ============================================================

number_of_chains = len(chain_sequences)

number_of_protein_residues = len(df_residues)

number_of_protein_atoms = len(df_atoms)

number_of_water_residues = len(df_water)


# ------------------------------------------------------------
# 3.1. Atom element composition
# ------------------------------------------------------------

if "element" in df_atoms.columns:

    element_counts = (

        df_atoms["element"]

        .fillna("Unknown")

        .astype(str)

        .str.strip()

        .str.upper()

        .value_counts()

    )

else:

    element_counts = pd.Series(dtype=int)


number_of_elements = len(element_counts)


df_element_distribution = (

    element_counts

    .rename_axis("element")

    .reset_index(name="atom_count")

)

if not df_element_distribution.empty:

    df_element_distribution["percentage"] = (

        100.0
        * df_element_distribution["atom_count"]
        / number_of_protein_atoms

    )


# ------------------------------------------------------------
# 3.2. Non-protein component instances
# ------------------------------------------------------------

df_components_source, components_source_name = (
    dataframe_or_empty([
        "df_components",
        "df_component_analysis"
    ])
)


if not df_components_source.empty:

    component_code_column = get_first_existing_column(

        df_components_source,

        [
            "component",
            "residue_name",
            "resname",
            "component_code"
        ]

    )

    component_id_column = get_first_existing_column(

        df_components_source,

        [
            "component_id",
            "residue_key",
            "component_instance_id"
        ]

    )

    if component_code_column is not None:

        component_codes = (

            df_components_source[component_code_column]

            .dropna()

            .astype(str)

            .str.strip()

            .str.upper()

        )

        if component_id_column is not None:

            component_records = (

                df_components_source[
                    [component_id_column, component_code_column]
                ]

                .dropna(subset=[component_code_column])

                .drop_duplicates(subset=[component_id_column])

            )

        else:

            instance_columns = [

                column

                for column in [

                    "model_id",
                    "chain",
                    "chain_id",
                    "residue_number",
                    "residue_id",
                    "insertion_code",
                    "component"

                ]

                if column in df_components_source.columns

            ]

            if instance_columns:

                component_records = (
                    df_components_source[
                        instance_columns
                    ].drop_duplicates()
                )

            else:

                component_records = df_components_source.copy()

        number_of_component_instances = len(
            component_records
        )

        component_counts = component_codes.value_counts()

        number_of_unique_component_types = (
            component_codes.nunique()
        )

        component_count_status = (
            f"from {components_source_name}"
        )

    else:

        component_counts = pd.Series(dtype=int)

        number_of_component_instances = np.nan

        number_of_unique_component_types = np.nan

        component_count_status = (
            "component code column unavailable"
        )

elif not df_hetero.empty:

    component_code_column = get_first_existing_column(

        df_hetero,

        [
            "residue_name",
            "resname",
            "component"
        ]

    )

    component_id_column = get_first_existing_column(

        df_hetero,

        [
            "component_id",
            "residue_key"
        ]

    )

    if component_code_column is not None:

        if component_id_column is not None:

            component_records = (

                df_hetero[
                    [component_id_column, component_code_column]
                ]

                .drop_duplicates(subset=[component_id_column])

            )

        else:

            instance_columns = [

                column

                for column in [

                    "model_id",
                    "chain_id",
                    "chain",
                    "residue_number",
                    "residue_id",
                    "insertion_code",
                    component_code_column

                ]

                if column in df_hetero.columns

            ]

            component_records = (

                df_hetero[instance_columns]

                .drop_duplicates()

                if instance_columns

                else pd.DataFrame()

            )

        if not component_records.empty:

            component_counts = (

                component_records[component_code_column]

                .fillna("Unknown")

                .astype(str)

                .str.strip()

                .str.upper()

                .value_counts()

            )

            number_of_component_instances = len(
                component_records
            )

            number_of_unique_component_types = len(
                component_counts
            )

            component_count_status = (
                "calculated from df_hetero"
            )

        else:

            component_counts = pd.Series(dtype=int)

            number_of_component_instances = np.nan

            number_of_unique_component_types = np.nan

            component_count_status = (
                "component instance identifiers unavailable"
            )

    else:

        component_counts = pd.Series(dtype=int)

        number_of_component_instances = np.nan

        number_of_unique_component_types = np.nan

        component_count_status = (
            "component code column unavailable"
        )

else:

    component_counts = pd.Series(dtype=int)

    number_of_component_instances = 0

    number_of_unique_component_types = 0

    component_count_status = (
        "no non-water hetero components detected"
    )


global_metrics = {

    "PDB_ID": PDB_ID,

    "number_of_chains": number_of_chains,

    "number_of_protein_residues": number_of_protein_residues,

    "number_of_protein_atoms": number_of_protein_atoms,

    "number_of_component_instances": (
        number_of_component_instances
    ),

    "number_of_unique_component_types": (
        number_of_unique_component_types
    ),

    "number_of_water_residues": number_of_water_residues,

    "number_of_atom_elements": number_of_elements

}


df_global_metrics = pd.DataFrame(
    [global_metrics]
)


df_component_distribution = (

    component_counts

    .rename_axis("component")

    .reset_index(name="occurrence_count")

)


# ============================================================
# 4. PROTEIN CHAIN METRICS
# ============================================================

chain_metric_data = []


chain_column = get_first_existing_column(
    df_residues,
    ["chain_id", "chain"]
)

atom_chain_column = get_first_existing_column(
    df_atoms,
    ["chain_id", "chain"]
)


for chain_id, sequence in chain_sequences.items():

    chain_key = normalize_chain(chain_id)

    if chain_column is not None:

        chain_residues = df_residues[

            df_residues[chain_column]
            .apply(normalize_chain)
            == chain_key

        ]

    else:

        chain_residues = pd.DataFrame()


    if atom_chain_column is not None:

        chain_atoms = df_atoms[

            df_atoms[atom_chain_column]
            .apply(normalize_chain)
            == chain_key

        ]

    else:

        chain_atoms = pd.DataFrame()


    sequence_length = len(str(sequence))

    residue_count = len(chain_residues)

    atom_count = len(chain_atoms)


    if "missing_backbone_atom_count" in chain_residues.columns:

        missing_backbone_count = int(

            pd.to_numeric(

                chain_residues["missing_backbone_atom_count"],

                errors="coerce"

            ).fillna(0).sum()

        )

    elif "missing_backbone_atoms" in chain_residues.columns:

        missing_backbone_count = int(

            pd.to_numeric(

                chain_residues["missing_backbone_atoms"],

                errors="coerce"

            ).fillna(0).sum()

        )

    else:

        missing_backbone_count = np.nan


    mean_bfactor = (

        safe_mean(chain_atoms["bfactor"])

        if "bfactor" in chain_atoms.columns

        else np.nan

    )

    median_bfactor = (

        safe_median(chain_atoms["bfactor"])

        if "bfactor" in chain_atoms.columns

        else np.nan

    )

    mean_occupancy = (

        safe_mean(chain_atoms["occupancy"])

        if "occupancy" in chain_atoms.columns

        else np.nan

    )


    chain_metric_data.append({

        "chain_id": chain_key,

        "sequence_length": sequence_length,

        "residue_count": residue_count,

        "atom_count": atom_count,

        "atoms_per_residue": (

            atom_count / residue_count

            if residue_count > 0

            else np.nan

        ),

        "missing_backbone_atoms": missing_backbone_count,

        "mean_B_factor": mean_bfactor,

        "median_B_factor": median_bfactor,

        "mean_occupancy": mean_occupancy

    })


df_final_chain_metrics = pd.DataFrame(
    chain_metric_data
)


# ============================================================
# 5. SEQUENCE IDENTITY BETWEEN CHAINS
# ============================================================

aligner = PairwiseAligner()

aligner.mode = "global"

aligner.match_score = 2.0

aligner.mismatch_score = -1.0

aligner.open_gap_score = -3.0

aligner.extend_gap_score = -0.5


chain_identity_data = []

chain_ids = list(chain_sequences.keys())


for i in range(len(chain_ids)):

    for j in range(i + 1, len(chain_ids)):

        chain_a = chain_ids[i]

        chain_b = chain_ids[j]

        sequence_a = str(chain_sequences[chain_a])

        sequence_b = str(chain_sequences[chain_b])


        if not sequence_a or not sequence_b:

            chain_identity_data.append({

                "chain_1": str(chain_a),

                "chain_2": str(chain_b),

                "alignment_score": np.nan,

                "identical_positions": np.nan,

                "aligned_positions": 0,

                "sequence_identity_percent": np.nan,

                "comparison_status": "Empty sequence"

            })

            continue


        alignment = aligner.align(
            sequence_a,
            sequence_b
        )[0]


        alignment_indices = alignment.indices

        identical_positions = 0

        aligned_positions = 0


        for position_a, position_b in zip(

            alignment_indices[0],

            alignment_indices[1]

        ):

            if position_a < 0 or position_b < 0:
                continue

            aligned_positions += 1

            if sequence_a[position_a] == sequence_b[position_b]:
                identical_positions += 1


        identity = safe_percentage(
            identical_positions,
            aligned_positions
        )


        chain_identity_data.append({

            "chain_1": str(chain_a),

            "chain_2": str(chain_b),

            "alignment_score": float(alignment.score),

            "identical_positions": identical_positions,

            "aligned_positions": aligned_positions,

            "sequence_identity_percent": identity,

            "comparison_status": "Global sequence alignment"

        })


df_chain_identity = pd.DataFrame(
    chain_identity_data
)


# ============================================================
# 6. GEOMETRY METRICS
# ============================================================

geometry_metric_data = []


# CELL 9 output file.

df_geometry_file = read_csv_if_exists(
    f"{PDB_ID}_geometry.csv"
)


if not df_geometry_file.empty:

    for column in df_geometry_file.columns:

        if column.lower() in {
            "pdb_id",
            "status",
            "output_checks"
        }:
            continue

        if len(df_geometry_file) == 1:

            value = df_geometry_file.iloc[0][column]

            if pd.api.types.is_numeric_dtype(
                df_geometry_file[column]
            ):

                geometry_metric_data.append({

                    "metric": column,

                    "value": json_safe(value),

                    "unit": None,

                    "status": "from CELL 9"

                })


# Alternative: geometry variables from CELL 9.

if not geometry_metric_data:

    geometry_summary = globals().get(
        "geometry_summary",
        None
    )

    if isinstance(geometry_summary, dict):

        for metric_name, value in geometry_summary.items():

            if isinstance(value, (int, float, np.number)):

                geometry_metric_data.append({

                    "metric": str(metric_name),

                    "value": json_safe(value),

                    "unit": None,

                    "status": "from CELL 9"

                })


# Additional common geometry variables.

if "radius_of_gyration" in globals():

    if not any(
        "gyration" in str(item["metric"]).lower()
        for item in geometry_metric_data
    ):

        geometry_metric_data.append({

            "metric": "Geometric radius of gyration",

            "value": json_safe(radius_of_gyration),

            "unit": "A",

            "status": "from CELL 9"

        })


if "dimensions" in globals():

    try:

        dimensions_array = np.asarray(
            dimensions,
            dtype=float
        ).reshape(-1)

        if dimensions_array.size == 3:

            for axis, value in zip(
                ["X", "Y", "Z"],
                dimensions_array
            ):

                geometry_metric_data.append({

                    "metric": f"{axis} dimension",

                    "value": float(value),

                    "unit": "A",

                    "status": "from CELL 9"

                })

    except (TypeError, ValueError):
        pass


# No fabricated geometry values if CELL 9 data are unavailable.

if not geometry_metric_data:

    geometry_metric_data.append({

        "metric": "Geometry metrics",

        "value": np.nan,

        "unit": None,

        "status": "CELL 9 geometry results unavailable"

    })


df_final_geometry_metrics = pd.DataFrame(

    geometry_metric_data,

    columns=[
        "metric",
        "value",
        "unit",
        "status"
    ]

)


# ============================================================
# 7. STRUCTURE QUALITY METRICS
# ============================================================

quality_metric_data = []


# CELL 14 outputs may be available as variables or CSV files.

df_atom_quality_source, atom_quality_source_name = (
    dataframe_or_empty([
        "df_atom_quality",
        "df_structure_quality_atoms"
    ])
)


if df_atom_quality_source.empty:

    df_atom_quality_source = read_csv_if_exists(
        f"{PDB_ID}_atom_quality.csv"
    )

    if not df_atom_quality_source.empty:
        atom_quality_source_name = "CELL 14 CSV"


df_altloc_source, altloc_source_name = (
    dataframe_or_empty([
        "df_altloc_atoms",
        "df_altloc"
    ])
)


if df_altloc_source.empty:

    df_altloc_source = read_csv_if_exists(
        f"{PDB_ID}_altloc_atoms.csv"
    )


df_low_occupancy_source, _ = dataframe_or_empty([
    "df_low_occupancy_atoms",
    "df_low_occupancy"
])


if df_low_occupancy_source.empty:

    df_low_occupancy_source = read_csv_if_exists(
        f"{PDB_ID}_low_occupancy_atoms.csv"
    )


df_missing_occupancy_source, _ = dataframe_or_empty([
    "df_missing_occupancy_atoms"
])


if df_missing_occupancy_source.empty:

    df_missing_occupancy_source = read_csv_if_exists(
        f"{PDB_ID}_missing_occupancy_atoms.csv"
    )


df_invalid_occupancy_source, _ = dataframe_or_empty([
    "df_invalid_occupancy_atoms"
])


if df_invalid_occupancy_source.empty:

    df_invalid_occupancy_source = read_csv_if_exists(
        f"{PDB_ID}_invalid_occupancy_atoms.csv"
    )


if not df_atom_quality_source.empty:

    quality_atom_count = len(
        df_atom_quality_source
    )

    quality_status = (
        f"from {atom_quality_source_name}"
    )

    quality_mean_bfactor = (

        safe_mean(df_atom_quality_source["bfactor"])

        if "bfactor" in df_atom_quality_source.columns

        else np.nan

    )

    quality_median_bfactor = (

        safe_median(df_atom_quality_source["bfactor"])

        if "bfactor" in df_atom_quality_source.columns

        else np.nan

    )

    quality_mean_occupancy = (

        safe_mean(df_atom_quality_source["occupancy"])

        if "occupancy" in df_atom_quality_source.columns

        else np.nan

    )

else:

    quality_atom_count = np.nan

    quality_status = "CELL 14 atom-level table unavailable"

    quality_mean_bfactor = np.nan

    quality_median_bfactor = np.nan

    quality_mean_occupancy = np.nan


# Alternate locations are reported separately.
# Occupancy 0.5 in alternate conformers is not automatically
# interpreted as poor quality.

if not df_altloc_source.empty:

    altloc_record_count = len(
        df_altloc_source
    )

    altloc_status = "from CELL 14 altloc table"

elif (
    not df_atom_quality_source.empty
    and "altloc" in df_atom_quality_source.columns
):

    altloc_values = (

        df_atom_quality_source["altloc"]

        .fillna("")

        .astype(str)

        .str.strip()

    )

    altloc_record_count = int(
        (altloc_values != "").sum()
    )

    altloc_status = "from atom quality table"

else:

    altloc_record_count = np.nan

    altloc_status = "altloc data unavailable"


# Low occupancy count: use CELL 14's dedicated table if available.

if not df_low_occupancy_source.empty:

    low_occupancy_count = len(
        df_low_occupancy_source
    )

    low_occupancy_status = (
        "from CELL 14 low occupancy table"
    )

else:

    low_occupancy_count = np.nan

    low_occupancy_status = (
        "dedicated low occupancy table unavailable"
    )


missing_occupancy_count = (

    len(df_missing_occupancy_source)

    if not df_missing_occupancy_source.empty

    else np.nan

)


invalid_occupancy_count = (

    len(df_invalid_occupancy_source)

    if not df_invalid_occupancy_source.empty

    else np.nan

)


quality_metric_data.extend([

    {

        "metric": "Atom quality records",

        "value": quality_atom_count,

        "unit": "records",

        "status": quality_status

    },

    {

        "metric": "Alternate location atom records",

        "value": altloc_record_count,

        "unit": "records",

        "status": altloc_status

    },

    {

        "metric": "Low occupancy atom records",

        "value": low_occupancy_count,

        "unit": "records",

        "status": low_occupancy_status

    },

    {

        "metric": "Missing occupancy atom records",

        "value": missing_occupancy_count,

        "unit": "records",

        "status": (
            "from CELL 14"
            if pd.notna(missing_occupancy_count)
            else "data unavailable"
        )

    },

    {

        "metric": "Invalid occupancy atom records",

        "value": invalid_occupancy_count,

        "unit": "records",

        "status": (
            "from CELL 14"
            if pd.notna(invalid_occupancy_count)
            else "data unavailable"
        )

    },

    {

        "metric": "Mean B-factor",

        "value": quality_mean_bfactor,

        "unit": "A^2",

        "status": quality_status

    },

    {

        "metric": "Median B-factor",

        "value": quality_median_bfactor,

        "unit": "A^2",

        "status": quality_status

    },

    {

        "metric": "Mean occupancy",

        "value": quality_mean_occupancy,

        "unit": None,

        "status": quality_status

    }

])


# ------------------------------------------------------------
# 7.1. Missing backbone atoms
# ------------------------------------------------------------

missing_backbone_source, _ = dataframe_or_empty([
    "missing_backbone_df",
    "df_missing_backbone_atoms"
])


if missing_backbone_source.empty:

    missing_backbone_source = read_csv_if_exists(
        f"{PDB_ID}_missing_backbone_atoms.csv"
    )


if not missing_backbone_source.empty:

    missing_backbone_count = len(
        missing_backbone_source
    )

    missing_backbone_status = (
        "from CELL 6 missing backbone table"
    )

else:

    missing_backbone_count = np.nan

    missing_backbone_status = (
        "missing backbone table unavailable"
    )


quality_metric_data.append({

    "metric": "Residues with missing backbone atoms",

    "value": missing_backbone_count,

    "unit": "records",

    "status": missing_backbone_status

})


# ------------------------------------------------------------
# 7.2. Unobserved residues and atoms from mmCIF
# ------------------------------------------------------------

df_unobserved_residues_source, _ = dataframe_or_empty([
    "df_unobserved_residues",
    "df_missing_residues"
])


if df_unobserved_residues_source.empty:

    df_unobserved_residues_source = read_csv_if_exists(
        f"{PDB_ID}_unobserved_residues.csv"
    )


df_unobserved_atoms_source, _ = dataframe_or_empty([
    "df_unobserved_atoms",
    "df_missing_atoms"
])


if df_unobserved_atoms_source.empty:

    df_unobserved_atoms_source = read_csv_if_exists(
        f"{PDB_ID}_unobserved_atoms.csv"
    )


unobserved_residue_count = (

    len(df_unobserved_residues_source)

    if not df_unobserved_residues_source.empty

    else np.nan

)


unobserved_atom_count = (

    len(df_unobserved_atoms_source)

    if not df_unobserved_atoms_source.empty

    else np.nan

)


quality_metric_data.extend([

    {

        "metric": "Unobserved residue records in mmCIF",

        "value": unobserved_residue_count,

        "unit": "records",

        "status": (
            "from CELL 10"
            if pd.notna(unobserved_residue_count)
            else "data unavailable"
        )

    },

    {

        "metric": "Unobserved atom records in mmCIF",

        "value": unobserved_atom_count,

        "unit": "records",

        "status": (
            "from CELL 10"
            if pd.notna(unobserved_atom_count)
            else "data unavailable"
        )

    }

])


df_final_quality_metrics = pd.DataFrame(
    quality_metric_data
)


# ============================================================
# 8. COMPONENT METRICS
# ============================================================

component_metric_data = []


if not component_counts.empty:

    for component, count in component_counts.items():

        component_metric_data.append({

            "component": str(component),

            "occurrence_count": int(count),

            "status": component_count_status

        })


df_final_component_metrics = pd.DataFrame(

    component_metric_data,

    columns=[
        "component",
        "occurrence_count",
        "status"
    ]

)


# ============================================================
# 9. BINDING SITE METRICS
# ============================================================
# CELL 18 может создавать несколько таблиц:
# - df_binding_by_residue;
# - df_binding_by_chain;
# - df_binding_atom_contacts.
#
# Используем таблицу по остаткам для числа остатков
# и таблицу атомных контактов для числа пар атомов.
# ============================================================

df_binding_residue_source, binding_residue_source_name = (
    dataframe_or_empty([
        "df_binding_by_residue",
        "df_binding_residues",
        "df_binding"
    ])
)


df_binding_atom_source, binding_atom_source_name = (
    dataframe_or_empty([
        "df_binding_atom_contacts",
        "df_atom_contacts",
        "df_binding_contacts"
    ])
)


# Fallback to files if dataframes are unavailable.

if df_binding_residue_source.empty:

    df_binding_residue_source = read_csv_if_exists(
        f"{PDB_ID}_binding_site_by_residue.csv"
    )

    if not df_binding_residue_source.empty:
        binding_residue_source_name = "CELL 18 CSV"


if df_binding_atom_source.empty:

    df_binding_atom_source = read_csv_if_exists(
        f"{PDB_ID}_binding_site_atom_contacts.csv"
    )

    if not df_binding_atom_source.empty:
        binding_atom_source_name = "CELL 18 CSV"


selected_component_available = (
    globals().get("selected_component") is not None
)


if not selected_component_available:

    binding_residue_count = np.nan

    binding_atom_pair_count = np.nan

    binding_minimum_distance = np.nan

    binding_mean_residue_distance = np.nan

    binding_status = (
        "not applicable: no selected component"
    )

elif not df_binding_residue_source.empty:

    binding_residue_count = len(
        df_binding_residue_source
    )

    minimum_distance_column = get_first_existing_column(

        df_binding_residue_source,

        [
            "minimum_distance",
            "min_distance",
            "minimum_distance_A",
            "min_distance_A"
        ]

    )

    if minimum_distance_column is not None:

        minimum_distances = numeric_values(
            df_binding_residue_source[minimum_distance_column]
        )

        binding_minimum_distance = (

            float(minimum_distances.min())

            if not minimum_distances.empty

            else np.nan

        )

        binding_mean_residue_distance = (

            float(minimum_distances.mean())

            if not minimum_distances.empty

            else np.nan

        )

    else:

        binding_minimum_distance = np.nan

        binding_mean_residue_distance = np.nan


    if not df_binding_atom_source.empty:

        binding_atom_pair_count = len(
            df_binding_atom_source
        )

    else:

        atom_contacts_column = get_first_existing_column(

            df_binding_residue_source,

            [
                "atom_contacts",
                "number_of_atom_contacts",
                "contact_count"
            ]

        )

        if atom_contacts_column is not None:

            binding_atom_pair_count = int(

                pd.to_numeric(

                    df_binding_residue_source[atom_contacts_column],

                    errors="coerce"

                ).fillna(0).sum()

            )

        else:

            binding_atom_pair_count = np.nan


    binding_status = (
        f"from {binding_residue_source_name}"
    )

else:

    binding_residue_count = 0

    binding_atom_pair_count = (

        0

        if selected_component_available

        else np.nan

    )

    binding_minimum_distance = np.nan

    binding_mean_residue_distance = np.nan

    binding_status = (
        "no binding-site residue records available"
    )


df_final_binding_metrics = pd.DataFrame([

    {

        "metric": "Binding site residues",

        "value": binding_residue_count,

        "unit": "residues",

        "status": binding_status

    },

    {

        "metric": "Protein-component atom-pair contacts",

        "value": binding_atom_pair_count,

        "unit": "contacts",

        "status": (
            f"from {binding_atom_source_name}"
            if binding_atom_source_name
            else binding_status
        )

    },

    {

        "metric": "Minimum component-protein distance",

        "value": binding_minimum_distance,

        "unit": "A",

        "status": binding_status

    },

    {

        "metric": "Mean minimum residue distance",

        "value": binding_mean_residue_distance,

        "unit": "A",

        "status": binding_status

    }

])


# ============================================================
# 10. SELECTED COMPONENT INTERACTION METRICS
# ============================================================

df_interactions_source, interaction_source_name = (
    dataframe_or_empty([
        "df_interactions"
    ])
)


interaction_metric_data = []


if interaction_source_name is None:

    interaction_status = "CELL 19 interaction table unavailable"

    total_interactions = np.nan

else:

    total_interactions = len(
        df_interactions_source
    )

    interaction_status = (
        "from CELL 19"
        if not df_interactions_source.empty
        else "CELL 19 completed; no contacts found"
    )


interaction_metric_data.append({

    "metric": "Total component-protein atom-pair contacts",

    "value": total_interactions,

    "unit": "contacts",

    "status": interaction_status

})


if (
    not df_interactions_source.empty
    and "interaction_type" in df_interactions_source.columns
):

    interaction_counts = (
        df_interactions_source["interaction_type"]
        .value_counts()
    )

    for interaction_type, count in interaction_counts.items():

        interaction_metric_data.append({

            "metric": str(interaction_type),

            "value": int(count),

            "unit": "contacts",

            "status": "distance-based classification"

        })


df_final_interaction_metrics = pd.DataFrame(
    interaction_metric_data
)


# ============================================================
# 11. HEM IRON COORDINATION METRICS
# ============================================================
# CELL 19:
# df_metal_coordination = Fe contacts in recognized hemes.
# df_all_metal_coordination = universal metal table.
#
# Actual neighbor_type values:
# PROTEIN
# Same HEME component
# OXY
# OTHER_COMPONENT
#
# Distance column: distance
# ============================================================

df_heme_fe_source, heme_source_name = (
    dataframe_or_empty([
        "df_metal_coordination"
    ])
)


df_all_metal_source, all_metal_source_name = (
    dataframe_or_empty([
        "df_all_metal_coordination"
    ])
)


heme_metric_data = []


if heme_source_name is None:

    heme_status = "CELL 19 heme table unavailable"

    total_fe_contacts = np.nan

else:

    total_fe_contacts = len(
        df_heme_fe_source
    )

    heme_status = (
        "from CELL 19"
        if not df_heme_fe_source.empty
        else "no Fe contacts in recognized hemes"
    )


add_metric(
    heme_metric_data,
    "HEM iron coordination",
    "Potential Fe coordination contacts",
    total_fe_contacts,
    "contacts",
    heme_status
)


if not df_heme_fe_source.empty:

    if "neighbor_type" in df_heme_fe_source.columns:

        neighbor_types = (
            df_heme_fe_source["neighbor_type"]
            .astype(str)
            .str.upper()
        )

        protein_fe = df_heme_fe_source[
            neighbor_types == "PROTEIN"
        ]

        heme_fe = df_heme_fe_source[
            neighbor_types == "SAME HEME COMPONENT"
        ]

        oxy_fe = df_heme_fe_source[
            neighbor_types == "OXY"
        ]

        other_fe = df_heme_fe_source[
            neighbor_types == "OTHER_COMPONENT"
        ]

    else:

        protein_fe = pd.DataFrame()

        heme_fe = pd.DataFrame()

        oxy_fe = pd.DataFrame()

        other_fe = pd.DataFrame()


    add_metric(
        heme_metric_data,
        "HEM iron coordination",
        "Potential protein-Fe contacts",
        len(protein_fe),
        "contacts",
        heme_status
    )

    add_metric(
        heme_metric_data,
        "HEM iron coordination",
        "Intracomponent Fe contacts in HEM",
        len(heme_fe),
        "contacts",
        heme_status
    )

    add_metric(
        heme_metric_data,
        "HEM iron coordination",
        "Potential Fe-OXY contacts",
        len(oxy_fe),
        "contacts",
        heme_status
    )

    add_metric(
        heme_metric_data,
        "HEM iron coordination",
        "Potential Fe contacts with other components",
        len(other_fe),
        "contacts",
        heme_status
    )


    distance_column = get_first_existing_column(
        df_heme_fe_source,
        ["distance", "distance_A"]
    )

    if distance_column is not None:

        all_fe_distances = numeric_values(
            df_heme_fe_source[distance_column]
        )

        add_metric(
            heme_metric_data,
            "HEM iron coordination",
            "Minimum Fe-neighbor distance",
            (
                float(all_fe_distances.min())
                if not all_fe_distances.empty
                else np.nan
            ),
            "A",
            heme_status
        )


    if not protein_fe.empty and distance_column is not None:

        protein_distances = numeric_values(
            protein_fe[distance_column]
        )

        add_metric(
            heme_metric_data,
            "HEM iron coordination",
            "Minimum Fe-protein distance",
            (
                float(protein_distances.min())
                if not protein_distances.empty
                else np.nan
            ),
            "A",
            heme_status
        )


    if not oxy_fe.empty and distance_column is not None:

        oxy_distances = numeric_values(
            oxy_fe[distance_column]
        )

        add_metric(
            heme_metric_data,
            "HEM iron coordination",
            "Minimum Fe-OXY distance",
            (
                float(oxy_distances.min())
                if not oxy_distances.empty
                else np.nan
            ),
            "A",
            heme_status
        )


df_final_heme_metrics = pd.DataFrame(
    heme_metric_data
)


# ============================================================
# 12. Fe-N DISTANCES WITHIN ALL HEM COMPONENTS
# ============================================================
# Рассчитываем расстояния Fe-N отдельно для каждой копии HEM.
#
# Используется первая модель.
# Внутри HEM учитываются атомы N самого гемового компонента.
# ============================================================

fe_n_distance_data = []


if component_structure_available:

    for chain in analysis_model:

        for residue in chain:

            residue_code = residue.get_resname().strip().upper()

            if residue_code != "HEM":
                continue


            atoms = list(residue.get_atoms())


            fe_atoms = [

                atom

                for atom in atoms

                if get_atom_element(atom) == "FE"

            ]


            nitrogen_atoms = [

                atom

                for atom in atoms

                if get_atom_element(atom) == "N"

            ]


            for fe_atom in fe_atoms:

                fe_coordinate = get_valid_coordinate(
                    fe_atom
                )

                if fe_coordinate is None:
                    continue


                for nitrogen_atom in nitrogen_atoms:

                    nitrogen_coordinate = get_valid_coordinate(
                        nitrogen_atom
                    )

                    if nitrogen_coordinate is None:
                        continue


                    distance = float(

                        np.linalg.norm(
                            fe_coordinate - nitrogen_coordinate
                        )

                    )


                    fe_n_distance_data.append({

                        "pdb_id": PDB_ID,

                        "model_id": analysis_model_id,

                        "chain_id": str(chain.id),

                        "residue_number": int(residue.id[1]),

                        "insertion_code": str(residue.id[2]).strip(),

                        "heme_component": residue_code,

                        "iron_atom": fe_atom.get_name().strip(),

                        "nitrogen_atom": nitrogen_atom.get_name().strip(),

                        "distance_A": distance

                    })


df_fe_n_distances = pd.DataFrame(

    fe_n_distance_data,

    columns=[

        "pdb_id",
        "model_id",
        "chain_id",
        "residue_number",
        "insertion_code",
        "heme_component",
        "iron_atom",
        "nitrogen_atom",
        "distance_A"

    ]

)


if not df_fe_n_distances.empty:

    df_fe_n_summary = (

        df_fe_n_distances

        .groupby(

            [
                "model_id",
                "chain_id",
                "residue_number",
                "insertion_code",
                "heme_component"
            ],

            dropna=False

        )

        .agg(

            number_of_Fe_N_distances=(
                "distance_A",
                "count"
            ),

            minimum_distance_A=(
                "distance_A",
                "min"
            ),

            maximum_distance_A=(
                "distance_A",
                "max"
            ),

            mean_distance_A=(
                "distance_A",
                "mean"
            ),

            std_distance_A=(
                "distance_A",
                "std"
            )

        )

        .reset_index()

    )

else:

    df_fe_n_summary = pd.DataFrame(

        columns=[

            "model_id",
            "chain_id",
            "residue_number",
            "insertion_code",
            "heme_component",
            "number_of_Fe_N_distances",
            "minimum_distance_A",
            "maximum_distance_A",
            "mean_distance_A",
            "std_distance_A"

        ]

    )


# ============================================================
# 13. Fe-OXY DISTANCES
# ============================================================
# Считаем расстояния между Fe в HEM и кислородными атомами
# компонентов OXY в первой модели.
#
# Это отдельная геометрическая таблица.
# Она не заменяет химическую интерпретацию связывания O2.
# ============================================================

fe_oxy_distance_data = []


if component_structure_available:

    fe_records = []

    oxy_records = []


    for chain in analysis_model:

        for residue in chain:

            residue_code = residue.get_resname().strip().upper()


            if residue_code in {
                "HEM", "HEA", "HEB", "HEC",
                "HEO", "HEF", "HEG", "HEV"
            }:

                for atom in residue.get_atoms():

                    if get_atom_element(atom) != "FE":
                        continue

                    coordinate = get_valid_coordinate(atom)

                    if coordinate is None:
                        continue

                    fe_records.append({

                        "chain_id": str(chain.id),

                        "residue_number": int(residue.id[1]),

                        "heme_component": residue_code,

                        "atom": atom,

                        "coordinate": coordinate

                    })


            elif residue_code in {"OXY", "O2"}:

                for atom in residue.get_atoms():

                    if get_atom_element(atom) != "O":
                        continue

                    coordinate = get_valid_coordinate(atom)

                    if coordinate is None:
                        continue

                    oxy_records.append({

                        "chain_id": str(chain.id),

                        "residue_number": int(residue.id[1]),

                        "oxy_component": residue_code,

                        "atom": atom,

                        "coordinate": coordinate

                    })


    for fe_record in fe_records:

        for oxy_record in oxy_records:

            distance = float(

                np.linalg.norm(

                    fe_record["coordinate"]
                    -
                    oxy_record["coordinate"]

                )

            )


            fe_oxy_distance_data.append({

                "pdb_id": PDB_ID,

                "model_id": analysis_model_id,

                "heme_chain": fe_record["chain_id"],

                "heme_residue_number": (
                    fe_record["residue_number"]
                ),

                "heme_component": fe_record["heme_component"],

                "iron_atom": fe_record["atom"].get_name().strip(),

                "oxy_chain": oxy_record["chain_id"],

                "oxy_residue_number": (
                    oxy_record["residue_number"]
                ),

                "oxy_component": oxy_record["oxy_component"],

                "oxygen_atom": oxy_record["atom"].get_name().strip(),

                "distance_A": distance

            })


df_fe_oxy_distances = pd.DataFrame(

    fe_oxy_distance_data,

    columns=[

        "pdb_id",
        "model_id",
        "heme_chain",
        "heme_residue_number",
        "heme_component",
        "iron_atom",
        "oxy_chain",
        "oxy_residue_number",
        "oxy_component",
        "oxygen_atom",
        "distance_A"

    ]

)


if not df_fe_oxy_distances.empty:

    df_fe_oxy_summary = (

        df_fe_oxy_distances

        .groupby(

            [
                "heme_chain",
                "heme_residue_number",
                "heme_component",
                "oxy_chain",
                "oxy_residue_number",
                "oxy_component"
            ],

            dropna=False

        )

        .agg(

            number_of_Fe_OXY_distances=(
                "distance_A",
                "count"
            ),

            minimum_distance_A=(
                "distance_A",
                "min"
            ),

            mean_distance_A=(
                "distance_A",
                "mean"
            ),

            maximum_distance_A=(
                "distance_A",
                "max"
            )

        )

        .reset_index()

    )

else:

    df_fe_oxy_summary = pd.DataFrame(

        columns=[

            "heme_chain",
            "heme_residue_number",
            "heme_component",
            "oxy_chain",
            "oxy_residue_number",
            "oxy_component",
            "number_of_Fe_OXY_distances",
            "minimum_distance_A",
            "mean_distance_A",
            "maximum_distance_A"

        ]

    )


# ============================================================
# 14. DISULFIDE BRIDGE METRICS
# ============================================================

df_disulfide_source, disulfide_source_name = (
    dataframe_or_empty([
        "df_disulfide_bridges"
    ])
)


if disulfide_source_name is None:

    disulfide_count = np.nan

    disulfide_status = (
        "CELL 19 disulfide table unavailable"
    )

else:

    disulfide_count = len(
        df_disulfide_source
    )

    disulfide_status = (

        "potential SG-SG bridges from CELL 19"

        if disulfide_count > 0

        else "CELL 19 completed; no pairs within cutoff"

    )


disulfide_metric_data = []


add_metric(
    disulfide_metric_data,
    "Disulfide bridges",
    "Potential disulfide bridge count",
    disulfide_count,
    "bridges",
    disulfide_status
)


disulfide_distance_column = get_first_existing_column(

    df_disulfide_source,

    ["distance", "distance_A"]

)


if (
    not df_disulfide_source.empty
    and disulfide_distance_column is not None
):

    disulfide_distances = numeric_values(
        df_disulfide_source[disulfide_distance_column]
    )

    add_metric(
        disulfide_metric_data,
        "Disulfide bridges",
        "Minimum SG-SG distance",
        (
            float(disulfide_distances.min())
            if not disulfide_distances.empty
            else np.nan
        ),
        "A",
        disulfide_status
    )

    add_metric(
        disulfide_metric_data,
        "Disulfide bridges",
        "Mean SG-SG distance",
        (
            float(disulfide_distances.mean())
            if not disulfide_distances.empty
            else np.nan
        ),
        "A",
        disulfide_status
    )


df_final_disulfide_metrics = pd.DataFrame(
    disulfide_metric_data
)


# ============================================================
# 15. BUILD MASTER METRICS TABLE
# ============================================================

master_metric_data = []


# ------------------------------------------------------------
# 15.1. Global metrics
# ------------------------------------------------------------

for metric_name, metric_value in global_metrics.items():

    if metric_name == "PDB_ID":
        continue

    if "component" in metric_name:

        status = component_count_status

    else:

        status = "calculated"


    add_metric(
        master_metric_data,
        "Global structure",
        metric_name,
        metric_value,
        None,
        status
    )


# ------------------------------------------------------------
# 15.2. Append metric DataFrames
# ------------------------------------------------------------

def append_metric_dataframe(
    dataframe,
    category,
    metric_column="metric",
    value_column="value",
    unit_column="unit",
    status_column="status"
):

    if dataframe is None or dataframe.empty:
        return

    for _, row in dataframe.iterrows():

        unit = (
            row[unit_column]
            if unit_column in dataframe.columns
            else None
        )

        status = (
            row[status_column]
            if status_column in dataframe.columns
            else "calculated"
        )

        add_metric(
            master_metric_data,
            category,
            row[metric_column],
            row[value_column],
            unit,
            status
        )


append_metric_dataframe(
    df_final_geometry_metrics,
    "Geometry"
)

append_metric_dataframe(
    df_final_quality_metrics,
    "Structure quality"
)

append_metric_dataframe(
    df_final_binding_metrics,
    "Binding site"
)

append_metric_dataframe(
    df_final_interaction_metrics,
    "Interactions"
)

append_metric_dataframe(
    df_final_heme_metrics,
    "HEM iron coordination"
)

append_metric_dataframe(
    df_final_disulfide_metrics,
    "Disulfide bridges"
)


# ------------------------------------------------------------
# 15.3. Chain metrics
# ------------------------------------------------------------

for _, row in df_final_chain_metrics.iterrows():

    chain_id = row["chain_id"]

    for metric_name in [

        "sequence_length",
        "residue_count",
        "atom_count",
        "atoms_per_residue",
        "missing_backbone_atoms",
        "mean_B_factor",
        "median_B_factor",
        "mean_occupancy"

    ]:

        add_metric(
            master_metric_data,
            f"Chain {chain_id}",
            metric_name,
            row[metric_name],
            None,
            "calculated"
        )


# ------------------------------------------------------------
# 15.4. Sequence identity
# ------------------------------------------------------------

for _, row in df_chain_identity.iterrows():

    add_metric(
        master_metric_data,
        "Sequence comparison",
        f"Identity {row['chain_1']}-{row['chain_2']}",
        row["sequence_identity_percent"],
        "%",
        row["comparison_status"]
    )


# ------------------------------------------------------------
# 15.5. Component occurrence counts
# ------------------------------------------------------------

for _, row in df_final_component_metrics.iterrows():

    add_metric(
        master_metric_data,
        "Components",
        f"{row['component']} occurrences",
        row["occurrence_count"],
        "instances",
        row["status"]
    )


# ------------------------------------------------------------
# 15.6. Fe-N distances
# ------------------------------------------------------------

if not df_fe_n_summary.empty:

    for _, row in df_fe_n_summary.iterrows():

        category = (

            f"HEM Fe-N chain {row['chain_id']}, "
            f"residue {row['residue_number']}"

        )

        for metric_name in [

            "number_of_Fe_N_distances",
            "minimum_distance_A",
            "maximum_distance_A",
            "mean_distance_A",
            "std_distance_A"

        ]:

            unit = (
                "A"
                if metric_name.endswith("_A")
                else "distances"
            )

            add_metric(
                master_metric_data,
                category,
                metric_name,
                row[metric_name],
                unit,
                "calculated"
            )


# ------------------------------------------------------------
# 15.7. Fe-OXY distances
# ------------------------------------------------------------

if not df_fe_oxy_summary.empty:

    for _, row in df_fe_oxy_summary.iterrows():

        category = (

            f"HEM Fe-OXY chain {row['heme_chain']}, "
            f"residue {row['heme_residue_number']}"

        )

        for metric_name in [

            "number_of_Fe_OXY_distances",
            "minimum_distance_A",
            "mean_distance_A",
            "maximum_distance_A"

        ]:

            unit = (
                "A"
                if metric_name.endswith("_A")
                else "distances"
            )

            add_metric(
                master_metric_data,
                category,
                metric_name,
                row[metric_name],
                unit,
                "calculated"
            )


df_final_metrics = pd.DataFrame(

    master_metric_data,

    columns=[
        "category",
        "metric",
        "value",
        "unit",
        "status"
    ]

)


# ============================================================
# 16. DISPLAY RESULTS
# ============================================================

print("\n" + "=" * 75)
print("1. GLOBAL STRUCTURE METRICS")
print("=" * 75)

display(df_global_metrics)


print("\n" + "=" * 75)
print("2. ATOM ELEMENT DISTRIBUTION")
print("=" * 75)

display(df_element_distribution)


print("\n" + "=" * 75)
print("3. PROTEIN CHAIN METRICS")
print("=" * 75)

display(df_final_chain_metrics)


print("\n" + "=" * 75)
print("4. SEQUENCE IDENTITY BETWEEN CHAINS")
print("=" * 75)

display(df_chain_identity)


print("\n" + "=" * 75)
print("5. GEOMETRY METRICS")
print("=" * 75)

display(df_final_geometry_metrics)


print("\n" + "=" * 75)
print("6. STRUCTURE QUALITY METRICS")
print("=" * 75)

display(df_final_quality_metrics)


print("\n" + "=" * 75)
print("7. COMPONENT DISTRIBUTION")
print("=" * 75)

display(df_component_distribution)


print("\n" + "=" * 75)
print("8. BINDING SITE METRICS")
print("=" * 75)

display(df_final_binding_metrics)


print("\n" + "=" * 75)
print("9. SELECTED COMPONENT INTERACTION METRICS")
print("=" * 75)

display(df_final_interaction_metrics)


print("\n" + "=" * 75)
print("10. HEM IRON COORDINATION METRICS")
print("=" * 75)

display(df_final_heme_metrics)


print("\n" + "=" * 75)
print("11. Fe-N DISTANCES")
print("=" * 75)

display(df_fe_n_distances)


print("\nFe-N summary by HEM:")

display(df_fe_n_summary)


print("\n" + "=" * 75)
print("12. Fe-OXY DISTANCES")
print("=" * 75)

display(df_fe_oxy_distances)


print("\nFe-OXY summary:")

display(df_fe_oxy_summary)


print("\n" + "=" * 75)
print("13. DISULFIDE BRIDGE METRICS")
print("=" * 75)

display(df_final_disulfide_metrics)


if not df_disulfide_source.empty:

    print("\nPotential disulfide bridges:")

    display(df_disulfide_source)


print("\n" + "=" * 75)
print("14. MASTER METRICS TABLE")
print("=" * 75)

display(df_final_metrics)


# ============================================================
# 17. SAVE CSV FILES
# ============================================================

metrics_files = {

    "global": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_global_metrics.csv"
    ),

    "elements": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_element_distribution.csv"
    ),

    "chains": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_chain_metrics.csv"
    ),

    "chain_identity": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_chain_identity.csv"
    ),

    "geometry": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_geometry_metrics.csv"
    ),

    "quality": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_quality_metrics.csv"
    ),

    "components": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_component_metrics.csv"
    ),

    "component_distribution": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_component_distribution.csv"
    ),

    "binding": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_binding_metrics.csv"
    ),

    "interactions": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_interaction_metrics.csv"
    ),

    "heme": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_heme_metrics.csv"
    ),

    "fe_n_distances": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_fe_n_distances.csv"
    ),

    "fe_n_summary": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_fe_n_summary.csv"
    ),

    "fe_oxy_distances": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_fe_oxy_distances.csv"
    ),

    "fe_oxy_summary": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_fe_oxy_summary.csv"
    ),

    "disulfide_metrics": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_disulfide_metrics.csv"
    ),

    "all_metrics": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_all_final_metrics.csv"
    )

}


dataframes_to_save = {

    "global": df_global_metrics,

    "elements": df_element_distribution,

    "chains": df_final_chain_metrics,

    "chain_identity": df_chain_identity,

    "geometry": df_final_geometry_metrics,

    "quality": df_final_quality_metrics,

    "components": df_final_component_metrics,

    "component_distribution": df_component_distribution,

    "binding": df_final_binding_metrics,

    "interactions": df_final_interaction_metrics,

    "heme": df_final_heme_metrics,

    "fe_n_distances": df_fe_n_distances,

    "fe_n_summary": df_fe_n_summary,

    "fe_oxy_distances": df_fe_oxy_distances,

    "fe_oxy_summary": df_fe_oxy_summary,

    "disulfide_metrics": df_final_disulfide_metrics,

    "all_metrics": df_final_metrics

}


for name, dataframe in dataframes_to_save.items():

    dataframe.to_csv(

        metrics_files[name],

        index=False,

        encoding="utf-8-sig"

    )


# ============================================================
# 18. SAVE JSON SUMMARY
# ============================================================

final_metrics_json = {

    "pdb_id": PDB_ID,

    "model_scope": {

        "source_model_count": (
            len(models)
            if component_structure_available
            else None
        ),

        "analyzed_model_id": analysis_model_id,

        "scope": "First model only"

    },

    "global_metrics": global_metrics,

    "component_count_status": component_count_status,

    "data_sources": {

        "geometry": (
            "CELL 9 output"
        ),

        "structure_quality": (
            "CELL 14 output, where available"
        ),

        "binding_site": (
            "CELL 18 residue and atom contact tables, where available"
        ),

        "component_interactions": (
            "CELL 19 df_interactions"
        ),

        "heme_iron_coordination": (
            "CELL 19 df_metal_coordination"
        ),

        "universal_metal_coordination": (
            "CELL 19 df_all_metal_coordination"
        ),

        "disulfide_bridges": (
            "CELL 19 df_disulfide_bridges"
        )

    },

    "counts": {

        "chain_metric_records": len(
            df_final_chain_metrics
        ),

        "chain_identity_comparisons": len(
            df_chain_identity
        ),

        "geometry_metrics": len(
            df_final_geometry_metrics
        ),

        "quality_metrics": len(
            df_final_quality_metrics
        ),

        "component_types": len(
            df_final_component_metrics
        ),

        "binding_metrics": len(
            df_final_binding_metrics
        ),

        "interaction_metrics": len(
            df_final_interaction_metrics
        ),

        "heme_metrics": len(
            df_final_heme_metrics
        ),

        "Fe_N_distances": len(
            df_fe_n_distances
        ),

        "Fe_N_HEM_summaries": len(
            df_fe_n_summary
        ),

        "Fe_OXY_distances": len(
            df_fe_oxy_distances
        ),

        "potential_disulfide_bridges": (
            disulfide_count
        ),

        "master_metrics": len(
            df_final_metrics
        )

    },

    "interpretation_notes": [

        "Protein geometry is based on protein atoms from df_atoms.",

        "Water and non-protein components are excluded from protein geometry.",

        "Chain sequence identity is calculated using global pairwise alignment.",

        "Binding-site metrics refer to the selected component from CELL 17-18.",

        "Fe-N distances are calculated separately for each HEM component.",

        "Fe-OXY distances are calculated between HEM iron and OXY oxygen atoms.",

        "Metal contacts are geometric candidates, not definitive coordination bonds.",

        "Polar atom proximity is not proof of a hydrogen bond.",

        "Potential disulfide bridges require chemical connectivity confirmation.",

        "Alternate conformer occupancy values are not automatically treated as low quality.",

        "Missing source tables are reported as unavailable rather than zero."

    ],

    "metric_tables": {

        name: path.name

        for name, path in metrics_files.items()

    }

}


metrics_json_path = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_final_metrics.json"

)


with open(

    metrics_json_path,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        json_safe(final_metrics_json),

        file,

        ensure_ascii=False,

        indent=4,

        allow_nan=False

    )


# ============================================================
# 19. FINAL VALIDATION
# ============================================================

print("\n" + "=" * 75)
print("FINAL METRICS VALIDATION")
print("=" * 75)


validation_results = {

    "PDB_ID_valid": (
        len(PDB_ID) == 4
        and PDB_ID.isalnum()
    ),

    "PDB_RESULTS_DIR_exists": (
        PDB_RESULTS_DIR.exists()
    ),

    "df_atoms_not_empty": (
        not df_atoms.empty
    ),

    "df_residues_is_dataframe": (
        isinstance(df_residues, pd.DataFrame)
    ),

    "global_metrics_created": (
        not df_global_metrics.empty
    ),

    "chain_metrics_created": (
        isinstance(df_final_chain_metrics, pd.DataFrame)
    ),

    "master_metrics_created": (
        isinstance(df_final_metrics, pd.DataFrame)
    ),

    "Fe_N_distances_valid": (

        df_fe_n_distances.empty

        or (

            df_fe_n_distances["distance_A"].notna().all()

            and
            (df_fe_n_distances["distance_A"] >= 0).all()

        )

    ),

    "Fe_OXY_distances_valid": (

        df_fe_oxy_distances.empty

        or (

            df_fe_oxy_distances["distance_A"].notna().all()

            and
            (df_fe_oxy_distances["distance_A"] >= 0).all()

        )

    )

}


for name, path in metrics_files.items():

    validation_results[
        f"file_{name}_exists"
    ] = path.is_file() and path.stat().st_size > 0


validation_results["json_summary_exists"] = (

    metrics_json_path.is_file()

    and metrics_json_path.stat().st_size > 0

)


failed_validations = [

    name

    for name, valid in validation_results.items()

    if not valid

]


for name, valid in validation_results.items():

    print(
        f"{name}: {'OK' if valid else 'CHECK'}"
    )


print("\nSAVED FILES")

for name, path in metrics_files.items():

    print(

        f"[{'OK' if path.exists() else 'ERROR'}] "

        f"{path.name}",

        "|",

        path.stat().st_size if path.exists() else 0,

        "bytes"

    )


print(

    f"[{'OK' if metrics_json_path.exists() else 'ERROR'}] "

    f"{metrics_json_path.name}"

)


# ============================================================
# 20. UPDATE JSON WITH VALIDATION RESULTS
# ============================================================

final_metrics_json["validation"] = validation_results

final_metrics_json["validation_status"] = (

    "PASSED"

    if not failed_validations

    else "CHECK_REQUIRED"

)


with open(

    metrics_json_path,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        json_safe(final_metrics_json),

        file,

        ensure_ascii=False,

        indent=4,

        allow_nan=False

    )


# ============================================================
# 21. FINAL REPORT
# ============================================================

print("\n" + "=" * 75)
print("FINAL METRICS CALCULATION COMPLETED")
print("=" * 75)

print("PDB ID:", PDB_ID)

print("Protein chains:", number_of_chains)

print("Protein residues:", number_of_protein_residues)

print("Protein atoms:", number_of_protein_atoms)

print("Component instances:", number_of_component_instances)

print("Unique component types:", number_of_unique_component_types)

print("HEM Fe-N distances:", len(df_fe_n_distances))

print("Fe-OXY distances:", len(df_fe_oxy_distances))

print("Potential disulfide bridges:", disulfide_count)

print("Master metrics:", len(df_final_metrics))

print("Output directory:", PDB_RESULTS_DIR)


if failed_validations:

    print("\nWARNING: Some validation checks require attention:")

    for name in failed_validations:
        print(" -", name)

else:

    print("\nAll CELL 20 validation checks passed.")


print("=" * 75)
print("CELL 20 COMPLETED")
print("=" * 75)
```

# Создание единого файла с признаками белка

```python

# ============================================================
# CELL 22. PROTEIN FEATURE EXTRACTION
# ============================================================
# Назначение:
# 1. Собрать структурные данные из предыдущих ячеек.
# 2. Сформировать глобальные признаки белка.
# 3. Сформировать атомные и остаточные признаки.
# 4. Подключить геометрические признаки CELL 9.
# 5. Подключить метаданные и качество структуры CELL 10/14.
# 6. Подключить область связывания CELL 18.
# 7. Подключить взаимодействия и координацию металлов CELL 19.
# 8. Сформировать объединенную таблицу взаимодействий.
# 9. Сохранить CSV-таблицы и единый JSON.
#
# ВАЖНО:
# - Исходные DataFrame не изменяются.
# - Анализируется первая модель структуры.
# - Отсутствующие признаки не заменяются искусственными нулями.
# - Геометрические контакты не считаются доказанными
#   химическими связями.
# - Расстояния измеряются в Angstrom (Å).
# - Все результаты сохраняются в PDB_RESULTS_DIR/features.
# ============================================================


import json
import numpy as np
import pandas as pd

from pathlib import Path
from datetime import datetime


print("=" * 75)
print("CELL 22. PROTEIN FEATURE EXTRACTION")
print("=" * 75)


# ------------------------------------------------------------
# 1. CONFIGURATION AND INPUT VALIDATION
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "PDB_RESULTS_DIR"
]

for variable in required_variables:

    if variable not in globals():

        raise NameError(
            f"Required variable '{variable}' is not defined. "
            "Run the previous cells first."
        )


PDB_ID_VALUE = str(PDB_ID).strip().upper()

if len(PDB_ID_VALUE) != 4:

    raise ValueError(
        f"Invalid PDB_ID: {PDB_ID_VALUE}"
    )


PDB_RESULTS_DIR_VALUE = Path(
    PDB_RESULTS_DIR
)

PDB_RESULTS_DIR_VALUE.mkdir(
    parents=True,
    exist_ok=True
)


FEATURE_DIR = (
    PDB_RESULTS_DIR_VALUE
    / "features"
)

FEATURE_DIR.mkdir(
    parents=True,
    exist_ok=True
)


ANALYSIS_SCOPE = "first_model"

print("PDB ID:", PDB_ID_VALUE)
print("PDB results directory:", PDB_RESULTS_DIR_VALUE)
print("Feature directory:", FEATURE_DIR)
print("Analysis scope:", ANALYSIS_SCOPE)


# ------------------------------------------------------------
# 2. DATAFRAME DISCOVERY
# ------------------------------------------------------------
# Для каждой категории задаются возможные имена.
#
# Приоритет:
# 1. Первый найденный непустой DataFrame.
# 2. Если все найденные таблицы пустые,
#    используется первый найденный пустой DataFrame.
#
# Это позволяет не выбирать пустую таблицу,
# если альтернативный источник содержит данные.
# ------------------------------------------------------------


DATAFRAME_CANDIDATES = {

    "atoms": [
        "df_atoms"
    ],

    "residues": [
        "df_residues"
    ],

    "chains": [
        "df_chains",
        "df_chain_statistics"
    ],

    "interactions": [
        "df_ligand_interactions",
        "df_interactions"
    ],

    "binding_site": [
        "df_binding",
        "df_binding_site",
        "df_binding_site_residues",
        "df_binding_site_analysis"
    ],

    "binding_atom_contacts": [
        "df_binding_atom_contacts"
    ],

    "binding_by_chain": [
        "df_binding_by_chain"
    ],

    "metal_coordination": [
        "df_all_metal_coordination",
        "df_metal_coordination"
    ],

    "heme_iron_coordination": [
        "df_metal_coordination"
    ],

    "disulfide_bridges": [
        "df_disulfide_bridges"
    ],

    "components": [
        "df_components",
        "df_nonprotein_components",
        "df_hetero"
    ],

    "geometry": [
        "df_geometry",
        "df_geometric_metrics",
        "df_structural_metrics"
    ],

    "hydrogen_bonds": [
        "df_hydrogen_bonds"
    ],

    "hydrophobic_contacts": [
        "df_hydrophobic_contacts"
    ],

    "altloc_atoms": [
        "df_altloc_atoms"
    ],

    "altloc_occupancy": [
        "df_altloc_occupancy"
    ],

    "low_occupancy_atoms": [
        "df_low_occupancy_atoms"
    ],

    "missing_occupancy_atoms": [
        "df_missing_occupancy_atoms"
    ],

    "invalid_occupancy_atoms": [
        "df_invalid_occupancy_atoms"
    ],

    "atom_quality": [
        "df_atom_quality"
    ],

    "unobserved_residues": [
        "df_unobserved_residues"
    ],

    "unobserved_atoms": [
        "df_unobserved_atoms"
    ]

}


def get_dataframe(candidate_names):
    """
    Ищет DataFrame по списку имен.

    Возвращает:
        dataframe_copy
        source_name

    Исходные таблицы не изменяются.
    """

    found = []

    for name in candidate_names:

        value = globals().get(name, None)

        if isinstance(value, pd.DataFrame):

            found.append((name, value))

            if not value.empty:

                return value.copy(deep=True), name

    if found:

        name, dataframe = found[0]

        return dataframe.copy(deep=True), name

    return pd.DataFrame(), None


DATAFRAMES = {}

DATAFRAME_SOURCES = {}


for category, candidates in DATAFRAME_CANDIDATES.items():

    dataframe, source_name = get_dataframe(
        candidates
    )

    DATAFRAMES[category] = dataframe

    DATAFRAME_SOURCES[category] = source_name


print("\nDATAFRAME DISCOVERY")
print("-" * 75)


for category, dataframe in DATAFRAMES.items():

    source_name = DATAFRAME_SOURCES[category]

    if source_name is not None:

        print(
            f"{category:27s}: "
            f"{len(dataframe):8d} rows | "
            f"{source_name}"
        )

    else:

        print(
            f"{category:27s}: not available"
        )


# ------------------------------------------------------------
# 3. REQUIRED DATAFRAME VALIDATION
# ------------------------------------------------------------

for category in ["atoms", "residues"]:

    if DATAFRAME_SOURCES[category] is None:

        raise RuntimeError(
            f"Required source table '{category}' is missing."
        )


df_atoms_feature = DATAFRAMES["atoms"].copy(deep=True)

df_residues_feature = DATAFRAMES["residues"].copy(deep=True)


if df_atoms_feature.empty:

    raise ValueError(
        "df_atoms is empty. Protein atom features cannot be generated."
    )


if df_residues_feature.empty:

    raise ValueError(
        "df_residues is empty. Protein residue features cannot be generated."
    )


# Проверяем принадлежность таблиц текущему PDB.
def validate_pdb_dataframe(dataframe, table_name):

    if dataframe.empty:

        return

    pdb_column = None

    for candidate in ["pdb_id", "PDB_ID"]:

        if candidate in dataframe.columns:

            pdb_column = candidate

            break

    if pdb_column is None:

        return

    values = (

        dataframe[pdb_column]
        .dropna()
        .astype(str)
        .str.strip()
        .str.upper()
        .unique()

    )

    if len(values) > 0:

        if any(value != PDB_ID_VALUE for value in values):

            raise ValueError(
                f"{table_name} contains records from another PDB: "
                f"{values.tolist()}"
            )


for category, dataframe in DATAFRAMES.items():

    validate_pdb_dataframe(
        dataframe,
        category
    )


# ------------------------------------------------------------
# 4. HELPER FUNCTIONS
# ------------------------------------------------------------


def safe_value(value):
    """
    Преобразует NumPy/Pandas значения в JSON-compatible types.
    NaN, Inf, pd.NA и NaT преобразуются в None.
    """

    if value is None:

        return None

    if isinstance(value, dict):

        return {
            str(key): safe_value(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple, set)):

        return [
            safe_value(item)
            for item in value
        ]

    if isinstance(value, np.ndarray):

        return safe_value(
            value.tolist()
        )

    if isinstance(value, (np.integer,)):

        return int(value)

    if isinstance(value, (np.floating, float)):

        number = float(value)

        return (
            number
            if np.isfinite(number)
            else None
        )

    if isinstance(value, (np.bool_,)):

        return bool(value)

    if isinstance(value, (pd.Timestamp, datetime)):

        return value.isoformat()

    try:

        missing = pd.isna(value)

        if isinstance(missing, (bool, np.bool_)):

            if missing:

                return None

    except (TypeError, ValueError):

        pass

    if isinstance(value, (str, int, bool)):

        return value

    return str(value)


def find_column(dataframe, candidates):
    """
    Поиск столбца без учета регистра.
    """

    if dataframe.empty:

        return None

    lookup = {

        str(column).strip().lower(): column

        for column in dataframe.columns

    }

    for candidate in candidates:

        key = str(candidate).strip().lower()

        if key in lookup:

            return lookup[key]

    return None


def numeric_series(dataframe, candidates):
    """
    Получает числовой Series из первого найденного столбца.
    """

    column = find_column(
        dataframe,
        candidates
    )

    if column is None:

        return pd.Series(dtype=float)

    return pd.to_numeric(
        dataframe[column],
        errors="coerce"
    ).dropna()


def mean_value(dataframe, candidates):

    values = numeric_series(
        dataframe,
        candidates
    )

    if values.empty:

        return None

    return float(values.mean())


def min_value(dataframe, candidates):

    values = numeric_series(
        dataframe,
        candidates
    )

    if values.empty:

        return None

    return float(values.min())


def max_value(dataframe, candidates):

    values = numeric_series(
        dataframe,
        candidates
    )

    if values.empty:

        return None

    return float(values.max())


def unique_count(dataframe, candidates):
    """
    Число уникальных непустых значений.
    Возвращает None, если нужного столбца нет.
    """

    column = find_column(
        dataframe,
        candidates
    )

    if column is None:

        return None

    return int(
        dataframe[column].dropna().nunique()
    )


def count_by_column(dataframe, candidates, prefix):
    """
    Считает количество объектов по категориям.
    """

    column = find_column(
        dataframe,
        candidates
    )

    if column is None:

        return {}

    values = (

        dataframe[column]
        .dropna()
        .astype(str)
        .str.strip()
        .str.upper()

    )

    values = values[
        ~values.isin(["", "NAN", "NONE", "UNKNOWN"])
    ]

    if values.empty:

        return {}

    counts = values.value_counts()

    return {

        f"{prefix}_{key}": int(value)

        for key, value in counts.items()

    }


def first_nonempty_dataframe(*dataframes):
    """
    Возвращает первую непустую таблицу.
    Если все пустые, возвращает первую доступную.
    """

    available = [

        dataframe

        for dataframe in dataframes

        if isinstance(dataframe, pd.DataFrame)

    ]

    for dataframe in available:

        if not dataframe.empty:

            return dataframe.copy(deep=True)

    if available:

        return available[0].copy(deep=True)

    return pd.DataFrame()


def normalize_chain_value(value):
    """
    Нормализует идентификатор цепи.
    Пустая цепь представляется как <blank>.
    """

    if pd.isna(value):

        return "<blank>"

    value = str(value).strip()

    return value if value else "<blank>"


def dataframe_from_csv(filename):
    """
    Загружает вспомогательную таблицу из PDB_RESULTS_DIR,
    если она не найдена среди глобальных переменных.
    """

    path = PDB_RESULTS_DIR_VALUE / filename

    if not path.is_file():

        return pd.DataFrame()

    try:

        return pd.read_csv(path)

    except Exception as error:

        print(
            f"WARNING: Could not read {filename}: {error}"
        )

        return pd.DataFrame()


def dataframe_from_json(filename):
    """
    Загружает JSON из PDB_RESULTS_DIR.
    """

    path = PDB_RESULTS_DIR_VALUE / filename

    if not path.is_file():

        return None

    try:

        with open(
            path,
            "r",
            encoding="utf-8"
        ) as file:

            return json.load(file)

    except Exception as error:

        print(
            f"WARNING: Could not read {filename}: {error}"
        )

        return None


def flatten_numeric_dictionary(data, prefix=""):
    """
    Извлекает числовые значения из вложенного словаря.

    Списки и текстовые значения не преобразуются
    в искусственные числовые признаки.
    """

    result = {}

    if not isinstance(data, dict):

        return result

    for key, value in data.items():

        name = (
            f"{prefix}_{key}"
            if prefix
            else str(key)
        )

        if isinstance(value, dict):

            result.update(
                flatten_numeric_dictionary(
                    value,
                    name
                )
            )

        elif isinstance(value, (int, float, np.integer, np.floating)):

            number = safe_value(value)

            if number is not None:

                result[name] = number

    return result


def add_counts(target, counts):

    for key, value in counts.items():

        target[key] = value


# ------------------------------------------------------------
# 5. SOURCE DATAFRAMES
# ------------------------------------------------------------


df_binding = DATAFRAMES["binding_site"].copy(deep=True)

df_binding_atom_contacts = (
    DATAFRAMES["binding_atom_contacts"].copy(deep=True)
)

df_binding_by_chain = (
    DATAFRAMES["binding_by_chain"].copy(deep=True)
)


df_interactions = DATAFRAMES["interactions"].copy(deep=True)

df_metal_coordination = (
    DATAFRAMES["metal_coordination"].copy(deep=True)
)

df_heme_iron_coordination = (
    DATAFRAMES["heme_iron_coordination"].copy(deep=True)
)

df_disulfide_bridges = (
    DATAFRAMES["disulfide_bridges"].copy(deep=True)
)

df_components = (
    DATAFRAMES["components"].copy(deep=True)
)

df_geometry = (
    DATAFRAMES["geometry"].copy(deep=True)
)


df_hydrogen_bonds = (
    DATAFRAMES["hydrogen_bonds"].copy(deep=True)
)

df_hydrophobic_contacts = (
    DATAFRAMES["hydrophobic_contacts"].copy(deep=True)
)


# ------------------------------------------------------------
# 6. ATOM FEATURES
# ------------------------------------------------------------
# Сохраняем все исходные столбцы CELL 7.
# Добавляем нормализованные поля.
# ------------------------------------------------------------


df_atom_features = df_atoms_feature.copy(deep=True)


element_column = find_column(
    df_atom_features,
    ["element", "atom_element"]
)

if element_column is not None:

    df_atom_features["element_normalized"] = (

        df_atom_features[element_column]
        .fillna("")
        .astype(str)
        .str.strip()
        .str.upper()

    )


atom_name_column = find_column(
    df_atom_features,
    ["atom_name", "name", "atom"]
)

if atom_name_column is not None:

    df_atom_features["atom_name_normalized"] = (

        df_atom_features[atom_name_column]
        .fillna("")
        .astype(str)
        .str.strip()

    )


chain_column = find_column(
    df_atom_features,
    ["chain_id", "chain"]
)

if chain_column is not None:

    df_atom_features["chain_id_normalized"] = (

        df_atom_features[chain_column]
        .apply(normalize_chain_value)

    )


# ------------------------------------------------------------
# 7. RESIDUE FEATURES
# ------------------------------------------------------------
# Сохраняем все исходные столбцы CELL 6.
# ------------------------------------------------------------


df_residue_features = df_residues_feature.copy(deep=True)


residue_name_column = find_column(
    df_residue_features,
    [
        "residue_name",
        "resname",
        "residue"
    ]
)

if residue_name_column is not None:

    df_residue_features["residue_name_normalized"] = (

        df_residue_features[residue_name_column]
        .fillna("")
        .astype(str)
        .str.strip()
        .str.upper()

    )


chain_column = find_column(
    df_residue_features,
    ["chain_id", "chain"]
)

if chain_column is not None:

    df_residue_features["chain_id_normalized"] = (

        df_residue_features[chain_column]
        .apply(normalize_chain_value)

    )


# ------------------------------------------------------------
# 8. GEOMETRY FEATURES — CELL 9
# ------------------------------------------------------------
# CELL 9 может предоставлять:
# - geometry_summary;
# - dimensions;
# - отдельные числовые значения;
# - CSV/JSON с геометрическими характеристиками.
#
# Извлекаем доступные числовые признаки.
# ------------------------------------------------------------


geometry_features = {}


geometry_summary_source = globals().get(
    "geometry_summary",
    None
)

if isinstance(geometry_summary_source, dict):

    geometry_features.update(
        flatten_numeric_dictionary(
            geometry_summary_source,
            "geometry"
        )
    )


dimensions_source = globals().get(
    "dimensions",
    None
)

if isinstance(dimensions_source, dict):

    geometry_features.update(
        flatten_numeric_dictionary(
            dimensions_source,
            "dimensions"
        )
    )


# Дополнительные известные глобальные переменные.
GEOMETRY_VARIABLE_CANDIDATES = [

    "radius_of_gyration",
    "mean_distance_from_center",
    "median_distance_from_center",
    "max_distance_from_center",
    "bounding_box_volume",
    "aspect_ratio",
    "protein_atom_count",
    "total_protein_atoms"

]


for variable_name in GEOMETRY_VARIABLE_CANDIDATES:

    value = globals().get(
        variable_name,
        None
    )

    if isinstance(value, (int, float, np.integer, np.floating)):

        geometry_features[variable_name] = safe_value(value)


# Если геометрия доступна в DataFrame.
if not df_geometry.empty:

    for column in df_geometry.columns:

        if len(df_geometry) == 1:

            value = df_geometry.iloc[0][column]

            if isinstance(
                value,
                (int, float, np.integer, np.floating)
            ):

                geometry_features[
                    f"geometry_{column}"
                ] = safe_value(value)


# Если геометрия доступна только в JSON.
if not geometry_features:

    geometry_json = dataframe_from_json(
        f"{PDB_ID_VALUE}_geometry.json"
    )

    if isinstance(geometry_json, dict):

        geometry_features.update(
            flatten_numeric_dictionary(
                geometry_json,
                "geometry"
            )
        )


# ------------------------------------------------------------
# 9. METADATA AND STRUCTURE QUALITY — CELL 10 / CELL 14
# ------------------------------------------------------------


metadata_json = dataframe_from_json(
    f"{PDB_ID_VALUE}_metadata.json"
)


structure_quality_json = dataframe_from_json(
    f"{PDB_ID_VALUE}_structure_quality.json"
)


quality_features = {}


# ------------------------------------------------------------
# 9.1. Unobserved residues and atoms
# ------------------------------------------------------------


df_unobserved_residues = first_nonempty_dataframe(

    DATAFRAMES["unobserved_residues"],

    dataframe_from_csv(
        f"{PDB_ID_VALUE}_unobserved_residues.csv"
    )

)


df_unobserved_atoms = first_nonempty_dataframe(

    DATAFRAMES["unobserved_atoms"],

    dataframe_from_csv(
        f"{PDB_ID_VALUE}_unobserved_atoms.csv"
    )

)


unobserved_residue_available = (

    DATAFRAME_SOURCES["unobserved_residues"] is not None

    or not df_unobserved_residues.empty

)


unobserved_atom_available = (

    DATAFRAME_SOURCES["unobserved_atoms"] is not None

    or not df_unobserved_atoms.empty

)


quality_features["unobserved_residue_count"] = (

    int(len(df_unobserved_residues))

    if unobserved_residue_available

    else None

)


quality_features["unobserved_atom_record_count"] = (

    int(len(df_unobserved_atoms))

    if unobserved_atom_available

    else None

)


# ------------------------------------------------------------
# 9.2. Altloc and occupancy diagnostics
# ------------------------------------------------------------


QUALITY_TABLES = {

    "altloc_atoms": "altloc_atoms",

    "altloc_occupancy": "altloc_occupancy",

    "low_occupancy_atoms": "low_occupancy_atoms",

    "missing_occupancy_atoms": "missing_occupancy_atoms",

    "invalid_occupancy_atoms": "invalid_occupancy_atoms",

    "atom_quality": "atom_quality"

}


df_quality_tables = {}


for key, category in QUALITY_TABLES.items():

    dataframe = DATAFRAMES[category].copy(deep=True)

    if dataframe.empty:

        filename_map = {

            "altloc_atoms": f"{PDB_ID_VALUE}_altloc_atoms.csv",

            "altloc_occupancy": f"{PDB_ID_VALUE}_altloc_occupancy.csv",

            "low_occupancy_atoms": f"{PDB_ID_VALUE}_low_occupancy_atoms.csv",

            "missing_occupancy_atoms": f"{PDB_ID_VALUE}_missing_occupancy_atoms.csv",

            "invalid_occupancy_atoms": f"{PDB_ID_VALUE}_invalid_occupancy_atoms.csv",

            "atom_quality": f"{PDB_ID_VALUE}_atom_quality.csv"

        }

        dataframe = dataframe_from_csv(
            filename_map[key]
        )

    df_quality_tables[key] = dataframe


for key, dataframe in df_quality_tables.items():

    quality_features[f"{key}_record_count"] = (

        int(len(dataframe))

        if not dataframe.empty
        or DATAFRAME_SOURCES.get(key) is not None

        else None

    )


# Сводные показатели качества из JSON CELL 14.
if isinstance(structure_quality_json, dict):

    quality_payload = structure_quality_json.get(
        "summary",
        structure_quality_json
    )

    if isinstance(quality_payload, dict):

        for key, value in flatten_numeric_dictionary(
            quality_payload,
            "structure_quality"
        ).items():

            quality_features[key] = value


# ------------------------------------------------------------
# 10. BINDING SITE FEATURES — CELL 18
# ------------------------------------------------------------


df_binding_features = df_binding.copy(deep=True)

df_binding_atom_contact_features = (
    df_binding_atom_contacts.copy(deep=True)
)

df_binding_chain_features = (
    df_binding_by_chain.copy(deep=True)
)


# Расстояние до остатка.
binding_distance_column = find_column(
    df_binding_features,
    [
        "min_distance_A",
        "minimum_distance_A",
        "min_distance"
    ]
)

if binding_distance_column is not None:

    df_binding_features[
        "binding_distance_normalized_A"
    ] = pd.to_numeric(
        df_binding_features[binding_distance_column],
        errors="coerce"
    )


# Расстояние между атомами.
binding_atom_distance_column = find_column(
    df_binding_atom_contact_features,
    [
        "distance_A",
        "distance",
        "contact_distance_A"
    ]
)

if binding_atom_distance_column is not None:

    df_binding_atom_contact_features[
        "contact_distance_normalized_A"
    ] = pd.to_numeric(
        df_binding_atom_contact_features[
            binding_atom_distance_column
        ],
        errors="coerce"
    )


def get_binding_residue_count(dataframe):
    """
    Число уникальных остатков в области связывания.
    """

    if dataframe.empty:

        return 0

    residue_key_column = find_column(
        dataframe,
        ["residue_key"]
    )

    if residue_key_column is not None:

        return int(
            dataframe[residue_key_column]
            .dropna()
            .nunique()
        )

    required_columns = [

        find_column(dataframe, ["protein_chain", "chain_id"]),

        find_column(dataframe, ["residue_id", "residue_number"]),

        find_column(dataframe, ["insertion_code"])

    ]

    if all(column is not None for column in required_columns):

        return int(
            dataframe[required_columns]
            .drop_duplicates()
            .shape[0]
        )

    # Если идентификаторы остатков отсутствуют,
    # число строк не объявляется числом уникальных остатков.
    return None


def get_unique_contact_atom_count(dataframe, side):
    """
    Подсчитывает уникальные контактирующие атомы.

    Учитываются доступные идентификаторы:
    chain, residue, atom name, altloc.

    Если идентификаторов недостаточно, возвращается None.
    """

    if dataframe.empty:

        return 0

    if side == "protein":

        candidates = {

            "chain": ["protein_chain", "chain_id", "protein_chain_id"],

            "residue": ["residue_key", "protein_residue_key"],

            "atom": ["protein_atom", "protein_atom_name"],

            "altloc": ["protein_altloc"]

        }

    else:

        candidates = {

            "component": [
                "component_id",
                "component_residue_key"
            ],

            "atom": [
                "component_atom",
                "component_atom_name"
            ],

            "altloc": [
                "component_altloc"
            ]

        }

    columns = []

    for key in ["chain", "component", "residue", "atom", "altloc"]:

        if key not in candidates:

            continue

        column = find_column(
            dataframe,
            candidates[key]
        )

        if column is not None:

            columns.append(column)

    # Минимум необходим атомный идентификатор.
    atom_column = find_column(
        dataframe,
        candidates["atom"]
    )

    if atom_column is None:

        return None

    if side == "protein":

        residue_column = find_column(
            dataframe,
            candidates["residue"]
        )

        chain_column = find_column(
            dataframe,
            candidates["chain"]
        )

        if residue_column is None or chain_column is None:

            return None

    else:

        component_column = find_column(
            dataframe,
            candidates["component"]
        )

        if component_column is None:

            return None

    return int(
        dataframe[columns]
        .dropna(subset=[atom_column])
        .drop_duplicates()
        .shape[0]
    )


binding_available = (
    DATAFRAME_SOURCES["binding_site"] is not None
)

binding_atom_contacts_available = (
    DATAFRAME_SOURCES["binding_atom_contacts"] is not None
)

binding_chain_available = (
    DATAFRAME_SOURCES["binding_by_chain"] is not None
)


binding_residue_count = (

    get_binding_residue_count(df_binding_features)

    if binding_available

    else None

)


binding_protein_atom_count = (

    get_unique_contact_atom_count(
        df_binding_atom_contact_features,
        "protein"
    )

    if binding_atom_contacts_available

    else None

)


binding_component_atom_count = (

    get_unique_contact_atom_count(
        df_binding_atom_contact_features,
        "component"
    )

    if binding_atom_contacts_available

    else None

)


binding_features = {

    "binding_site_available": binding_available,

    "binding_atom_contacts_available": binding_atom_contacts_available,

    "binding_chain_statistics_available": binding_chain_available,

    "binding_site_residue_count": binding_residue_count,

    "binding_site_atom_pair_contact_count": (

        int(len(df_binding_atom_contact_features))

        if binding_atom_contacts_available

        else None

    ),

    "binding_site_min_distance_A": min_value(
        df_binding_features,
        [
            "min_distance_A",
            "minimum_distance_A",
            "min_distance"
        ]
    ),

    "binding_site_mean_residue_min_distance_A": mean_value(
        df_binding_features,
        [
            "min_distance_A",
            "minimum_distance_A",
            "min_distance"
        ]
    ),

    "binding_site_max_residue_min_distance_A": max_value(
        df_binding_features,
        [
            "min_distance_A",
            "maximum_distance_A",
            "min_distance"
        ]
    ),

    "binding_site_mean_atom_contact_distance_A": mean_value(
        df_binding_atom_contact_features,
        [
            "distance_A",
            "distance",
            "contact_distance_A"
        ]
    ),

    "binding_site_unique_protein_contact_atom_count": (
        binding_protein_atom_count
    ),

    "binding_site_unique_component_contact_atom_count": (
        binding_component_atom_count
    ),

    "binding_site_chain_count": unique_count(
        df_binding_features,
        [
            "protein_chain",
            "chain_id"
        ]
    )

}


# ------------------------------------------------------------
# 11. METAL FEATURES — CELL 19
# ------------------------------------------------------------


df_metal_features = df_metal_coordination.copy(deep=True)

df_heme_iron_features = (
    df_heme_iron_coordination.copy(deep=True)
)


metal_element_column = find_column(
    df_metal_features,
    [
        "metal_element",
        "metal",
        "metal_name",
        "metal_type"
    ]
)


metal_distance_column = find_column(
    df_metal_features,
    [
        "distance_A",
        "distance",
        "metal_distance_A"
    ]
)


if not df_metal_features.empty:

    if metal_element_column is not None:

        df_metal_features["metal_normalized"] = (

            df_metal_features[metal_element_column]
            .fillna("")
            .astype(str)
            .str.strip()
            .str.upper()

        )

    if metal_distance_column is not None:

        df_metal_features[
            "metal_distance_normalized_A"
        ] = pd.to_numeric(
            df_metal_features[metal_distance_column],
            errors="coerce"
        )


metal_features = {

    "metal_coordination_available": (
        DATAFRAME_SOURCES["metal_coordination"] is not None
    ),

    "metal_coordination_contact_count": (

        int(len(df_metal_features))

        if DATAFRAME_SOURCES["metal_coordination"] is not None

        else None

    ),

    "metal_coordination_mean_distance_A": mean_value(
        df_metal_features,
        [
            "distance_A",
            "distance",
            "metal_distance_A"
        ]
    ),

    "heme_iron_coordination_contact_count": (

        int(len(df_heme_iron_features))

        if DATAFRAME_SOURCES["heme_iron_coordination"] is not None

        else None

    )

}


if metal_element_column is not None:

    metal_groups = df_metal_features.groupby(
        metal_element_column,
        dropna=True
    )

    for metal_name, metal_group in metal_groups:

        metal_name = str(
            metal_name
        ).strip().upper()

        if not metal_name:

            continue

        metal_features[
            f"metal_contact_count_{metal_name}"
        ] = int(len(metal_group))

        distance_column = find_column(
            metal_group,
            [
                "distance_A",
                "distance",
                "metal_distance_A"
            ]
        )

        if distance_column is not None:

            distances = pd.to_numeric(
                metal_group[distance_column],
                errors="coerce"
            ).dropna()

            metal_features[
                f"metal_mean_distance_{metal_name}_A"
            ] = (

                float(distances.mean())

                if not distances.empty

                else None

            )


# ------------------------------------------------------------
# 12. COMPONENT FEATURES
# ------------------------------------------------------------


component_code_column = find_column(
    df_components,
    [
        "component_code",
        "component",
        "component_id",
        "comp_id",
        "resname",
        "residue_name"
    ]
)


component_features = {

    "component_table_available": (
        DATAFRAME_SOURCES["components"] is not None
    ),

    "component_instance_count": (

        int(len(df_components))

        if DATAFRAME_SOURCES["components"] is not None

        else None

    ),

    "component_type_count": (

        int(
            df_components[component_code_column]
            .dropna()
            .astype(str)
            .str.upper()
            .nunique()
        )

        if component_code_column is not None

        else None

    )

}


# Распределение типов компонентов.
component_type_counts = count_by_column(
    df_components,
    [
        "component_code",
        "component",
        "component_id",
        "comp_id",
        "resname",
        "residue_name"
    ],
    "component_instance_count"
)


add_counts(
    component_features,
    component_type_counts
)


# ------------------------------------------------------------
# 13. DISULFIDE BRIDGE FEATURES
# ------------------------------------------------------------


disulfide_features = {

    "disulfide_analysis_available": (
        DATAFRAME_SOURCES["disulfide_bridges"] is not None
    ),

    "potential_disulfide_bridge_count": (

        int(len(df_disulfide_bridges))

        if DATAFRAME_SOURCES["disulfide_bridges"] is not None

        else None

    )

}


# ------------------------------------------------------------
# 14. ATOM AND RESIDUE COMPOSITION
# ------------------------------------------------------------


atom_element_counts = count_by_column(
    df_atom_features,
    [
        "element",
        "element_normalized",
        "atom_element"
    ],
    "protein_atom_count_element"
)


residue_type_counts = count_by_column(
    df_residue_features,
    [
        "residue_name",
        "residue_name_normalized",
        "resname"
    ],
    "protein_residue_count"
)


# ------------------------------------------------------------
# 15. ADDITIONAL INTERACTION TABLES
# ------------------------------------------------------------


df_hydrogen_bond_features = (
    df_hydrogen_bonds.copy(deep=True)
)

df_hydrophobic_contact_features = (
    df_hydrophobic_contacts.copy(deep=True)
)


additional_interaction_features = {

    "hydrogen_bond_table_available": (
        DATAFRAME_SOURCES["hydrogen_bonds"] is not None
    ),

    "hydrogen_bond_record_count": (

        int(len(df_hydrogen_bond_features))

        if DATAFRAME_SOURCES["hydrogen_bonds"] is not None

        else None

    ),

    "hydrophobic_contact_table_available": (
        DATAFRAME_SOURCES["hydrophobic_contacts"] is not None
    ),

    "hydrophobic_contact_record_count": (

        int(len(df_hydrophobic_contact_features))

        if DATAFRAME_SOURCES["hydrophobic_contacts"] is not None

        else None

    )

}


# ------------------------------------------------------------
# 16. COMBINED INTERACTION FEATURES
# ------------------------------------------------------------
# Объединяются таблицы разной гранулярности.
#
# source_table и interaction_type сохраняют происхождение
# каждой записи.
#
# Не следует интерпретировать каждую строку как
# независимое химическое взаимодействие.
# ------------------------------------------------------------


interaction_frames = []


def append_interaction_table(
    dataframe,
    source_name,
    interaction_type
):

    if dataframe.empty:

        return

    frame = dataframe.copy(deep=True)

    frame["source_table"] = source_name

    frame["interaction_type"] = interaction_type

    interaction_frames.append(frame)


append_interaction_table(
    df_interactions,
    DATAFRAME_SOURCES["interactions"] or "not_available",
    "selected_component_contact"
)


append_interaction_table(
    df_binding_atom_contact_features,
    DATAFRAME_SOURCES["binding_atom_contacts"] or "not_available",
    "binding_site_atom_pair_contact"
)


append_interaction_table(
    df_metal_features,
    DATAFRAME_SOURCES["metal_coordination"] or "not_available",
    "potential_metal_coordination"
)


append_interaction_table(
    df_disulfide_bridges,
    DATAFRAME_SOURCES["disulfide_bridges"] or "not_available",
    "potential_disulfide_bridge"
)


append_interaction_table(
    df_hydrogen_bond_features,
    DATAFRAME_SOURCES["hydrogen_bonds"] or "not_available",
    "hydrogen_bond_table_record"
)


append_interaction_table(
    df_hydrophobic_contact_features,
    DATAFRAME_SOURCES["hydrophobic_contacts"] or "not_available",
    "hydrophobic_contact_table_record"
)


if interaction_frames:

    df_interaction_features = pd.concat(
        interaction_frames,
        ignore_index=True,
        sort=False
    )

else:

    df_interaction_features = pd.DataFrame(
        columns=[
            "source_table",
            "interaction_type"
        ]
    )


# ------------------------------------------------------------
# 17. GLOBAL FEATURES
# ------------------------------------------------------------


global_features = {

    "pdb_id": PDB_ID_VALUE,

    "analysis_scope": ANALYSIS_SCOPE,

    "protein_atom_count": int(len(df_atom_features)),

    "protein_residue_count": int(len(df_residue_features)),

    "protein_chain_count": None,

    "component_instance_count": component_features[
        "component_instance_count"
    ],

    "component_type_count": component_features[
        "component_type_count"
    ],

    "selected_component_contact_record_count": (

        int(len(df_interactions))

        if DATAFRAME_SOURCES["interactions"] is not None

        else None

    ),

    "selected_component_mean_contact_distance_A": mean_value(
        df_interactions,
        [
            "distance_A",
            "distance",
            "min_distance_A"
        ]
    ),

    "geometry_metric_record_count": (

        int(len(df_geometry))

        if DATAFRAME_SOURCES["geometry"] is not None

        else None

    )

}


# ------------------------------------------------------------
# 17.1. CHAIN COUNT
# ------------------------------------------------------------


chain_column = find_column(
    df_residue_features,
    [
        "chain_id",
        "chain"
    ]
)


if chain_column is not None:

    global_features["protein_chain_count"] = int(

        df_residue_features[chain_column]
        .apply(normalize_chain_value)
        .nunique()

    )


# Если доступен chain_sequences, он может содержать
# цепи, в которых отсутствуют распознанные остатки.
chain_sequences_source = globals().get(
    "chain_sequences",
    None
)


if isinstance(chain_sequences_source, dict):

    global_features["sequence_chain_count"] = int(
        len(chain_sequences_source)
    )


# ------------------------------------------------------------
# 17.2. BINDING SITE GLOBAL FEATURES
# ------------------------------------------------------------


global_features.update(
    binding_features
)


# ------------------------------------------------------------
# 17.3. METAL GLOBAL FEATURES
# ------------------------------------------------------------


global_features.update(
    metal_features
)


# ------------------------------------------------------------
# 17.4. COMPONENT GLOBAL FEATURES
# ------------------------------------------------------------


global_features.update(
    component_features
)


# ------------------------------------------------------------
# 17.5. DISULFIDE GLOBAL FEATURES
# ------------------------------------------------------------


global_features.update(
    disulfide_features
)


# ------------------------------------------------------------
# 17.6. QUALITY GLOBAL FEATURES
# ------------------------------------------------------------


global_features.update(
    quality_features
)


# ------------------------------------------------------------
# 17.7. ADDITIONAL INTERACTION FEATURES
# ------------------------------------------------------------


global_features.update(
    additional_interaction_features
)


# ------------------------------------------------------------
# 17.8. ATOM AND RESIDUE DISTRIBUTIONS
# ------------------------------------------------------------


add_counts(
    global_features,
    atom_element_counts
)


add_counts(
    global_features,
    residue_type_counts
)


# ------------------------------------------------------------
# 17.9. GEOMETRY FEATURES
# ------------------------------------------------------------


global_features.update(
    geometry_features
)


# ------------------------------------------------------------
# 18. FEATURE TABLES
# ------------------------------------------------------------


df_global_features = pd.DataFrame(
    [global_features]
)


feature_table_map = {

    "global_features": df_global_features,

    "atom_features": df_atom_features,

    "residue_features": df_residue_features,

    "interaction_features": df_interaction_features,

    "metal_features": df_metal_features,

    "binding_site_features": df_binding_features,

    "binding_atom_contact_features": (
        df_binding_atom_contact_features
    ),

    "binding_chain_features": (
        df_binding_chain_features
    ),

    "component_features": df_components,

    "disulfide_features": df_disulfide_bridges,

    "geometry_features": df_geometry,

    "altloc_features": df_quality_tables["altloc_atoms"],

    "altloc_occupancy_features": df_quality_tables["altloc_occupancy"],

    "low_occupancy_features": df_quality_tables["low_occupancy_atoms"],

    "missing_occupancy_features": df_quality_tables["missing_occupancy_atoms"],

    "invalid_occupancy_features": df_quality_tables["invalid_occupancy_atoms"],

    "atom_quality_features": df_quality_tables["atom_quality"],

    "unobserved_residue_features": df_unobserved_residues,

    "unobserved_atom_features": df_unobserved_atoms

}


# ------------------------------------------------------------
# 19. TABLE STATISTICS
# ------------------------------------------------------------


table_statistics = {}


for table_name, dataframe in feature_table_map.items():

    table_statistics[table_name] = {

        "rows": int(len(dataframe)),

        "columns": int(len(dataframe.columns)),

        "column_names": [
            str(column)
            for column in dataframe.columns
        ],

        "available": bool(
            table_name in [
                "global_features",
                "atom_features",
                "residue_features"
            ]
            or len(dataframe) > 0
        )

    }


# ------------------------------------------------------------
# 20. FEATURE DEFINITIONS
# ------------------------------------------------------------


feature_definitions = {

    "pdb_id": (
        "PDB identifier of the analyzed structure."
    ),

    "analysis_scope": (
        "Structural analysis scope. The current pipeline "
        "uses the first model of the PDB structure."
    ),

    "protein_atom_count": (
        "Number of protein atoms in df_atoms."
    ),

    "protein_residue_count": (
        "Number of amino-acid residue records in df_residues."
    ),

    "protein_chain_count": (
        "Number of unique protein chain identifiers."
    ),

    "component_instance_count": (
        "Number of non-protein component instances."
    ),

    "component_type_count": (
        "Number of unique non-protein component types."
    ),

    "selected_component_contact_record_count": (
        "Number of records in the selected-component interaction table."
    ),

    "selected_component_mean_contact_distance_A": (
        "Mean distance in the selected-component interaction table, Å."
    ),

    "binding_site_residue_count": (
        "Number of unique amino-acid residues in the binding site."
    ),

    "binding_site_atom_pair_contact_count": (
        "Number of protein-atom/component-atom contact pairs."
    ),

    "binding_site_min_distance_A": (
        "Minimum residue-level distance to the selected component, Å."
    ),

    "binding_site_mean_residue_min_distance_A": (
        "Mean of residue-level minimum distances, Å. "
        "This is not the mean of all atom-pair distances."
    ),

    "binding_site_max_residue_min_distance_A": (
        "Maximum of residue-level minimum distances, Å."
    ),

    "binding_site_mean_atom_contact_distance_A": (
        "Mean distance between atom pairs recorded as contacts, Å."
    ),

    "binding_site_unique_protein_contact_atom_count": (
        "Number of unique protein atoms participating in contacts."
    ),

    "binding_site_unique_component_contact_atom_count": (
        "Number of unique selected-component atoms participating in contacts."
    ),

    "metal_coordination_contact_count": (
        "Number of geometrically detected potential metal coordination contacts."
    ),

    "metal_coordination_mean_distance_A": (
        "Mean distance of potential metal coordination contacts, Å."
    ),

    "heme_iron_coordination_contact_count": (
        "Number of records in the specialized heme iron coordination table."
    ),

    "potential_disulfide_bridge_count": (
        "Number of potential disulfide bridge records."
    ),

    "unobserved_residue_count": (
        "Number of records in the unobserved-residue table."
    ),

    "unobserved_atom_record_count": (
        "Number of records in the unobserved-atom table."
    ),

    "geometry_features": (
        "Numerical structural geometry descriptors extracted from CELL 9."
    ),

    "quality_features": (
        "Structural quality indicators from CELL 10 and CELL 14."
    )

}


# ------------------------------------------------------------
# 21. JSON METADATA
# ------------------------------------------------------------


protein_features = {

    "metadata": {

        "pdb_id": PDB_ID_VALUE,

        "created_at": datetime.now().isoformat(
            timespec="seconds"
        ),

        "feature_extraction_cell": 22,

        "analysis_scope": ANALYSIS_SCOPE,

        "pdb_results_directory": str(
            PDB_RESULTS_DIR_VALUE
        ),

        "feature_directory": str(
            FEATURE_DIR
        ),

        "binding_site_cutoff_A": safe_value(
            globals().get("BINDING_CUTOFF", None)
        ),

        "interaction_cutoff_A": safe_value(
            globals().get("CONTACT_CUTOFF", None)
        ),

        "metal_coordination_cutoff_A": safe_value(
            globals().get("METAL_COORDINATION_CUTOFF", None)
        )

    },

    "global_features": safe_value(
        global_features
    ),

    "table_statistics": safe_value(
        table_statistics
    ),

    "dataframe_sources": safe_value(
        DATAFRAME_SOURCES
    ),

    "feature_definitions": safe_value(
        feature_definitions
    ),

    "geometry_features": safe_value(
        geometry_features
    ),

    "quality_features": safe_value(
        quality_features
    ),

    "analysis_notes": {

        "structural_scope": (
            "The feature tables describe the first structural model. "
            "Alternate conformations are handled separately by "
            "the structure quality analysis."
        ),

        "binding_site": (
            "Binding-site features originate from CELL 18. "
            "Residue-level and atom-pair-level statistics are kept separate."
        ),

        "geometry": (
            "Geometry descriptors originate from CELL 9. "
            "The bounding-box volume is not a physical molecular volume."
        ),

        "metal_coordination": (
            "Metal contacts are geometric candidates. "
            "They do not independently prove chemical coordination."
        ),

        "disulfide_bridges": (
            "Disulfide bridge records are potential geometrical matches "
            "based on the criteria used in CELL 19."
        ),

        "interaction_features": (
            "The combined interaction table contains records with "
            "different granularities. source_table and interaction_type "
            "preserve their provenance."
        ),

        "quality": (
            "Unobserved residues, alternate conformations and occupancy "
            "diagnostics are preserved as separate indicators."
        ),

        "missing_data": (
            "Unavailable source tables are not replaced with synthetic "
            "observations or artificial zero values."
        )

    }

}


# ------------------------------------------------------------
# 22. OUTPUT FILES
# ------------------------------------------------------------


OUTPUT_FILES = {}


for table_name in feature_table_map:

    OUTPUT_FILES[table_name] = (

        FEATURE_DIR
        / f"{PDB_ID_VALUE}_{table_name}.csv"

    )


JSON_FILE = (

    FEATURE_DIR
    / f"{PDB_ID_VALUE}_protein_features.json"

)


# ------------------------------------------------------------
# 23. SAVE CSV TABLES
# ------------------------------------------------------------


print("\n" + "=" * 75)
print("SAVING FEATURE TABLES")
print("=" * 75)


for table_name, dataframe in feature_table_map.items():

    output_path = OUTPUT_FILES[table_name]

    dataframe.to_csv(

        output_path,

        index=False,

        encoding="utf-8-sig"

    )

    if not output_path.is_file():

        raise RuntimeError(
            f"Output file was not saved: {output_path}"
        )

    print(

        f"{output_path.name:55s} "
        f"{len(dataframe):8d} rows | "
        f"{len(dataframe.columns):4d} columns"

    )


# ------------------------------------------------------------
# 24. SAVE JSON
# ------------------------------------------------------------


with open(

    JSON_FILE,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        safe_value(protein_features),

        file,

        ensure_ascii=False,

        indent=4,

        allow_nan=False

    )


if not JSON_FILE.is_file():

    raise RuntimeError(
        f"JSON file was not saved: {JSON_FILE}"
    )


# ------------------------------------------------------------
# 25. OUTPUT VALIDATION
# ------------------------------------------------------------


print("\n" + "=" * 75)
print("OUTPUT VALIDATION")
print("=" * 75)


for table_name, dataframe in feature_table_map.items():

    output_path = OUTPUT_FILES[table_name]

    if not output_path.is_file():

        raise RuntimeError(
            f"Missing output file: {output_path}"
        )

    # Пустые CSV допустимы для необязательных таблиц.
    # Поэтому проверяем существование, но не требуем
    # ненулевой размер файла.
    print(

        f"{table_name:35s}: "
        f"{'OK' if output_path.is_file() else 'ERROR'}"

    )


# Проверка JSON.
with open(
    JSON_FILE,
    "r",
    encoding="utf-8"
) as file:

    json_check = json.load(file)


if json_check["metadata"]["pdb_id"] != PDB_ID_VALUE:

    raise ValueError(
        "PDB ID mismatch in saved JSON."
    )


if "global_features" not in json_check:

    raise ValueError(
        "Global features are missing in JSON."
    )


# Проверка согласованности количества строк.
if len(df_atom_features) != len(df_atoms_feature):

    raise ValueError(
        "Atom feature row count mismatch."
    )


if len(df_residue_features) != len(df_residues_feature):

    raise ValueError(
        "Residue feature row count mismatch."
    )


expected_interaction_rows = sum(

    len(dataframe)

    for dataframe in [

        df_interactions,

        df_binding_atom_contact_features,

        df_metal_features,

        df_disulfide_bridges,

        df_hydrogen_bond_features,

        df_hydrophobic_contact_features

    ]

)


if len(df_interaction_features) != expected_interaction_rows:

    raise ValueError(
        "Combined interaction feature row count mismatch."
    )


# Проверяем, что исходные таблицы не были изменены.
if len(df_atoms_feature) != len(DATAFRAMES["atoms"]):

    raise ValueError(
        "Original atom table size changed unexpectedly."
    )


if len(df_residue_features) != len(DATAFRAMES["residues"]):

    raise ValueError(
        "Original residue table size changed unexpectedly."
    )


print("JSON structure: OK")
print("Atom feature rows: OK")
print("Residue feature rows: OK")
print("Interaction feature rows: OK")
print("Original DataFrame row counts: OK")


# ------------------------------------------------------------
# 26. FINAL SUMMARY
# ------------------------------------------------------------


print("\n" + "=" * 75)
print("PROTEIN FEATURE EXTRACTION SUMMARY")
print("=" * 75)


print("PDB ID:", PDB_ID_VALUE)

print("Analysis scope:", ANALYSIS_SCOPE)


print("\nSTRUCTURAL FEATURES")
print("-" * 75)

print(
    "Protein atoms:",
    global_features["protein_atom_count"]
)

print(
    "Protein residues:",
    global_features["protein_residue_count"]
)

print(
    "Protein chains:",
    global_features["protein_chain_count"]
)

print(
    "Component instances:",
    global_features["component_instance_count"]
)

print(
    "Component types:",
    global_features["component_type_count"]
)


print("\nGEOMETRY FEATURES")
print("-" * 75)

print(
    "Geometry feature count:",
    len(geometry_features)
)

for key, value in geometry_features.items():

    print(
        f"{key}: {value}"
    )


print("\nBINDING SITE FEATURES")
print("-" * 75)

print(
    "Binding-site residues:",
    binding_features["binding_site_residue_count"]
)

print(
    "Atom-pair contacts:",
    binding_features["binding_site_atom_pair_contact_count"]
)

print(
    "Minimum residue distance:",
    binding_features["binding_site_min_distance_A"]
)

print(
    "Mean residue minimum distance:",
    binding_features["binding_site_mean_residue_min_distance_A"]
)

print(
    "Mean atom-pair distance:",
    binding_features["binding_site_mean_atom_contact_distance_A"]
)


print("\nINTERACTION FEATURES")
print("-" * 75)

print(
    "Selected-component records:",
    len(df_interactions)
)

print(
    "Binding atom-pair contacts:",
    len(df_binding_atom_contact_features)
)

print(
    "Metal coordination candidates:",
    len(df_metal_features)
)

print(
    "Potential disulfide bridges:",
    len(df_disulfide_bridges)
)

print(
    "Combined interaction records:",
    len(df_interaction_features)
)


print("\nSTRUCTURE QUALITY")
print("-" * 75)

for key, value in quality_features.items():

    print(
        f"{key}: {value}"
    )


print("\nOUTPUT TABLES")
print("-" * 75)

for table_name, dataframe in feature_table_map.items():

    print(

        f"{table_name:35s}: "
        f"{len(dataframe):8d} rows × "
        f"{len(dataframe.columns):3d} columns"

    )


print("\nOUTPUT DIRECTORY")
print("-" * 75)

print(FEATURE_DIR)


print("\nJSON FILE")
print("-" * 75)

print(JSON_FILE)


print("\nSTATUS: COMPLETED")
print("CELL 22 COMPLETED")

print("=" * 75)
```

```python

# ============================================================
# CELL 23. PROTEIN SPATIAL REPRESENTATION
# ============================================================
#
# НАЗНАЧЕНИЕ:
#
# Создание двух пространственных представлений белка:
#
# 1. ATOM SPATIAL GRAPH
#    Узел = отдельный тяжёлый атом белка.
#    Ребро = пара атомов на расстоянии <= SPATIAL_CUTOFF_A.
#
# 2. RESIDUE SPATIAL GRAPH
#    Узел = аминокислотный остаток.
#    Ребро = два разных остатка, между которыми существует
#            хотя бы одна пара атомов на расстоянии <= cutoff.
#
# ДОПОЛНИТЕЛЬНО:
# - координаты атомов и центры остатков;
# - расстояния между контактирующими атомами;
# - количество атомных контактов между остатками;
# - статистика связности графов;
# - количество изолированных узлов;
# - средняя степень узла;
# - проверки идентификаторов и расстояний.
#
# ВАЖНО:
# - Используются атомы белка из CELL 7.
# - Рассматривается первая модель структуры.
# - Водород и дейтерий исключаются.
# - Пространственное ребро означает геометрическую близость,
#   а не подтверждённую химическую связь.
# - Расстояния измеряются в Angstrom (Å).
# ============================================================


import json
import numpy as np
import pandas as pd

from pathlib import Path
from datetime import datetime

from scipy.spatial import cKDTree
from scipy.sparse import coo_matrix
from scipy.sparse.csgraph import connected_components


print("=" * 75)
print("CELL 23. PROTEIN SPATIAL REPRESENTATION")
print("=" * 75)


# ============================================================
# 1. CONFIGURATION
# ============================================================

SPATIAL_CUTOFF_A = 5.0

EXCLUDE_HYDROGEN = True

ANALYSIS_SCOPE = "first_model"

# Допуск для проверки расстояний.
DISTANCE_TOLERANCE = 1e-6


# ============================================================
# 2. INPUT VALIDATION
# ============================================================

required_variables = [
    "PDB_ID",
    "PDB_RESULTS_DIR",
    "df_atoms"
]

for variable in required_variables:

    if variable not in globals():

        raise NameError(
            f"Required variable '{variable}' is not defined. "
            "Run the previous cells first."
        )


PDB_ID_VALUE = str(PDB_ID).strip().upper()

if len(PDB_ID_VALUE) != 4:

    raise ValueError(
        f"Invalid PDB_ID: {PDB_ID_VALUE}"
    )


if not isinstance(df_atoms, pd.DataFrame):

    raise TypeError(
        "df_atoms must be a pandas DataFrame."
    )


if df_atoms.empty:

    raise ValueError(
        "df_atoms is empty."
    )


PDB_RESULTS_DIR_VALUE = Path(
    PDB_RESULTS_DIR
)

PDB_RESULTS_DIR_VALUE.mkdir(
    parents=True,
    exist_ok=True
)


FEATURE_DIR = (
    PDB_RESULTS_DIR_VALUE
    / "features"
)

FEATURE_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("PDB ID:", PDB_ID_VALUE)
print("Analysis scope:", ANALYSIS_SCOPE)
print("Spatial cutoff:", SPATIAL_CUTOFF_A, "Å")
print("Input atom rows:", len(df_atoms))
print("Output directory:", FEATURE_DIR)


# ============================================================
# 3. HELPER FUNCTIONS
# ============================================================

def find_column(dataframe, candidates):
    """
    Поиск столбца без учета регистра,
    пробелов и символов подчёркивания.
    """

    normalized = {

        "".join(
            char.lower()
            for char in str(column)
            if char.isalnum()
        ): column

        for column in dataframe.columns

    }

    for candidate in candidates:

        key = "".join(
            char.lower()
            for char in str(candidate)
            if char.isalnum()
        )

        if key in normalized:

            return normalized[key]

    return None


def normalize_text(value, default=""):

    if pd.isna(value):

        return default

    value = str(value).strip()

    if value.lower() in ["nan", "none", "nat", "<na>"]:

        return default

    return value


def safe_json_value(value):
    """
    Преобразование значений в JSON-compatible формат.
    """

    if value is None:

        return None

    if isinstance(value, dict):

        return {
            str(key): safe_json_value(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple, set)):

        return [
            safe_json_value(item)
            for item in value
        ]

    if isinstance(value, np.ndarray):

        return safe_json_value(
            value.tolist()
        )

    if isinstance(value, (np.integer,)):

        return int(value)

    if isinstance(value, (np.floating, float)):

        number = float(value)

        return (
            number
            if np.isfinite(number)
            else None
        )

    if isinstance(value, (np.bool_,)):

        return bool(value)

    try:

        missing = pd.isna(value)

        if isinstance(missing, (bool, np.bool_)) and missing:

            return None

    except (TypeError, ValueError):

        pass

    if isinstance(value, (str, int, bool)):

        return value

    return str(value)


def graph_statistics(
    node_count,
    edge_array
):
    """
    Рассчитывает статистику неориентированного графа.

    Изолированные узлы также учитываются
    как отдельные компоненты связности.
    """

    edge_count = len(edge_array)

    if node_count == 0:

        return {

            "node_count": 0,

            "edge_count": 0,

            "component_count": 0,

            "largest_component_size": 0,

            "largest_component_fraction": None,

            "isolated_node_count": 0,

            "mean_degree": None,

            "max_degree": None,

            "graph_density": None

        }


    degrees = np.zeros(
        node_count,
        dtype=np.int64
    )

    if edge_count > 0:

        degrees += np.bincount(
            edge_array[:, 0],
            minlength=node_count
        )

        degrees += np.bincount(
            edge_array[:, 1],
            minlength=node_count
        )

        rows = np.concatenate([
            edge_array[:, 0],
            edge_array[:, 1]
        ])

        cols = np.concatenate([
            edge_array[:, 1],
            edge_array[:, 0]
        ])

        adjacency = coo_matrix(

            (
                np.ones(
                    len(rows),
                    dtype=np.int8
                ),

                (rows, cols)

            ),

            shape=(node_count, node_count)

        ).tocsr()

    else:

        adjacency = coo_matrix(
            (node_count, node_count)
        ).tocsr()


    component_count, labels = connected_components(

        adjacency,

        directed=False,

        return_labels=True

    )


    component_sizes = np.bincount(
        labels,
        minlength=component_count
    )


    largest_component_size = int(
        component_sizes.max()
    )


    if node_count > 1:

        graph_density = float(
            2 * edge_count
            / (node_count * (node_count - 1))
        )

    else:

        graph_density = 0.0


    return {

        "node_count": int(node_count),

        "edge_count": int(edge_count),

        "component_count": int(component_count),

        "largest_component_size": largest_component_size,

        "largest_component_fraction": float(
            largest_component_size / node_count
        ),

        "isolated_node_count": int(
            np.sum(degrees == 0)
        ),

        "mean_degree": float(
            degrees.mean()
        ),

        "max_degree": int(
            degrees.max()
        ),

        "graph_density": graph_density

    }


def get_sorted_pair_array(pairs):
    """
    Приводит пары к детерминированному порядку.
    """

    if len(pairs) == 0:

        return np.empty(
            (0, 2),
            dtype=np.int64
        )

    pairs = np.asarray(
        pairs,
        dtype=np.int64
    )

    pairs = np.sort(
        pairs,
        axis=1
    )

    order = np.lexsort(
        (
            pairs[:, 1],
            pairs[:, 0]
        )
    )

    return pairs[order]


# ============================================================
# 4. PREPARE ATOM DATA
# ============================================================
#
# Работаем с копией df_atoms.
# Исходная таблица не изменяется.
# ============================================================

df_spatial_source = df_atoms.copy(
    deep=True
).reset_index(
    drop=True
)


# ------------------------------------------------------------
# 4.1. Identify coordinates
# ------------------------------------------------------------

x_col = find_column(
    df_spatial_source,
    [
        "x",
        "coord_x",
        "x_coord",
        "x_coordinate",
        "Cartn_x"
    ]
)

y_col = find_column(
    df_spatial_source,
    [
        "y",
        "coord_y",
        "y_coord",
        "y_coordinate",
        "Cartn_y"
    ]
)

z_col = find_column(
    df_spatial_source,
    [
        "z",
        "coord_z",
        "z_coord",
        "z_coordinate",
        "Cartn_z"
    ]
)


if not all(
    column is not None
    for column in [x_col, y_col, z_col]
):

    raise ValueError(
        "Could not identify X, Y, Z coordinate columns. "
        f"Available columns: {df_spatial_source.columns.tolist()}"
    )


# ------------------------------------------------------------
# 4.2. Identify structural fields
# ------------------------------------------------------------

pdb_col = find_column(
    df_spatial_source,
    [
        "pdb_id",
        "PDB_ID"
    ]
)

element_col = find_column(
    df_spatial_source,
    [
        "element",
        "atom_element",
        "symbol"
    ]
)

chain_col = find_column(
    df_spatial_source,
    [
        "chain_id",
        "chain"
    ]
)

resname_col = find_column(
    df_spatial_source,
    [
        "residue_name",
        "resname",
        "residue"
    ]
)

resnum_col = find_column(
    df_spatial_source,
    [
        "residue_id",
        "residue_number",
        "resnum",
        "resseq"
    ]
)

icode_col = find_column(
    df_spatial_source,
    [
        "insertion_code",
        "icode",
        "insertion"
    ]
)

residue_key_col = find_column(
    df_spatial_source,
    [
        "residue_key"
    ]
)

atom_name_col = find_column(
    df_spatial_source,
    [
        "atom_name",
        "name",
        "atom"
    ]
)

atom_serial_col = find_column(
    df_spatial_source,
    [
        "atom_serial",
        "serial_number",
        "serial"
    ]
)

altloc_col = find_column(
    df_spatial_source,
    [
        "altloc",
        "alt_loc"
    ]
)


# ------------------------------------------------------------
# 4.3. Validate PDB ID
# ------------------------------------------------------------

if pdb_col is not None:

    pdb_values = (

        df_spatial_source[pdb_col]
        .dropna()
        .astype(str)
        .str.strip()
        .str.upper()
        .unique()

    )

    if len(pdb_values) > 0:

        if any(
            value != PDB_ID_VALUE
            for value in pdb_values
        ):

            raise ValueError(
                "df_atoms contains records from another PDB: "
                f"{pdb_values.tolist()}"
            )


# ------------------------------------------------------------
# 4.4. Validate required structural fields
# ------------------------------------------------------------

if element_col is None:

    raise ValueError(
        "Atom element column is missing."
    )


if chain_col is None:

    raise ValueError(
        "Chain identifier column is missing."
    )


if atom_name_col is None:

    raise ValueError(
        "Atom name column is missing."
    )


if residue_key_col is None:

    if resname_col is None or resnum_col is None:

        raise ValueError(
            "Cannot construct a unique residue identifier. "
            "Need residue_key or residue_name + residue_id."
        )


# ------------------------------------------------------------
# 4.5. Normalize coordinates
# ------------------------------------------------------------

for column in [x_col, y_col, z_col]:

    df_spatial_source[column] = pd.to_numeric(

        df_spatial_source[column],

        errors="coerce"

    )


coordinate_values = df_spatial_source[
    [x_col, y_col, z_col]
].to_numpy(
    dtype=np.float64
)


valid_coordinates = np.isfinite(
    coordinate_values
).all(
    axis=1
)


invalid_coordinate_count = int(
    (~valid_coordinates).sum()
)


df_spatial_source = df_spatial_source.loc[
    valid_coordinates
].copy()


# ------------------------------------------------------------
# 4.6. Normalize atom fields
# ------------------------------------------------------------

df_spatial_source["_element"] = (

    df_spatial_source[element_col]
    .apply(
        lambda value: normalize_text(
            value
        ).upper()
    )

)


df_spatial_source["_atom_name"] = (

    df_spatial_source[atom_name_col]
    .apply(
        lambda value: normalize_text(
            value
        )
    )

)


df_spatial_source["_chain_id"] = (

    df_spatial_source[chain_col]
    .apply(
        lambda value: normalize_text(
            value,
            default="<blank>"
        )
    )

)


# ------------------------------------------------------------
# 4.7. Exclude hydrogen and deuterium
# ------------------------------------------------------------

if EXCLUDE_HYDROGEN:

    before_filter = len(
        df_spatial_source
    )

    df_spatial_source = df_spatial_source[
        ~df_spatial_source["_element"].isin(
            ["H", "D"]
        )
    ].copy()

    excluded_hydrogen_count = (

        before_filter
        - len(df_spatial_source)

    )

else:

    excluded_hydrogen_count = 0


# ------------------------------------------------------------
# 4.8. Normalize residue information
# ------------------------------------------------------------

if resname_col is not None:

    df_spatial_source["_resname"] = (

        df_spatial_source[resname_col]
        .apply(
            lambda value: normalize_text(
                value,
                default="UNK"
            ).upper()
        )

    )

else:

    df_spatial_source["_resname"] = "UNK"


if resnum_col is not None:

    df_spatial_source["_resnum"] = (

        df_spatial_source[resnum_col]
        .apply(
            lambda value: normalize_text(
                value,
                default="UNKNOWN"
            )
        )

    )

else:

    df_spatial_source["_resnum"] = "UNKNOWN"


if icode_col is not None:

    df_spatial_source["_icode"] = (

        df_spatial_source[icode_col]
        .apply(
            lambda value: normalize_text(value)
        )

    )

else:

    df_spatial_source["_icode"] = ""


# ------------------------------------------------------------
# 4.9. Build stable residue identifiers
# ------------------------------------------------------------

if residue_key_col is not None:

    df_spatial_source["_residue_key"] = (

        df_spatial_source[residue_key_col]
        .apply(
            lambda value: normalize_text(value)
        )

    )

    missing_keys = (
        df_spatial_source["_residue_key"] == ""
    )

    if missing_keys.any():

        fallback_keys = (

            df_spatial_source["_chain_id"]
            + "|"
            + df_spatial_source["_resname"]
            + "|"
            + df_spatial_source["_resnum"]
            + "|"
            + df_spatial_source["_icode"]

        )

        df_spatial_source.loc[
            missing_keys,
            "_residue_key"
        ] = fallback_keys.loc[
            missing_keys
        ]

else:

    df_spatial_source["_residue_key"] = (

        df_spatial_source["_chain_id"]
        + "|"
        + df_spatial_source["_resname"]
        + "|"
        + df_spatial_source["_resnum"]
        + "|"
        + df_spatial_source["_icode"]

    )


if (
    df_spatial_source["_residue_key"]
    .eq("")
    .any()
):

    raise ValueError(
        "Some atoms have empty residue identifiers."
    )


# ------------------------------------------------------------
# 4.10. Check residue identifier consistency
# ------------------------------------------------------------
#
# Один residue_key должен соответствовать одному набору:
# chain_id, residue_name, residue_number, insertion_code.
#
# Если это условие нарушено, остатки нельзя безопасно
# объединять в один узел.
# ------------------------------------------------------------

residue_identity_columns = [

    "_chain_id",
    "_resname",
    "_resnum",
    "_icode"

]


residue_identity_check = (

    df_spatial_source[
        ["_residue_key"] + residue_identity_columns
    ]

    .drop_duplicates()

    .groupby(
        "_residue_key",
        sort=False
    )

    .size()

)


ambiguous_residue_keys = residue_identity_check[
    residue_identity_check > 1
]


if not ambiguous_residue_keys.empty:

    raise ValueError(

        "Some residue_key values refer to different residue identities. "
        f"Examples: {ambiguous_residue_keys.index[:10].tolist()}"

    )


# ------------------------------------------------------------
# 4.11. Check duplicate atom sites
# ------------------------------------------------------------
#
# CELL 7 использует выбранную конформацию Bio.PDB.
# Поэтому для одного остатка не должно быть повторных записей
# одного и того же atom_name + element.
#
# Если такие записи обнаружены, останавливаем построение графа,
# а не объединяем их автоматически.
# ------------------------------------------------------------

atom_site_columns = [

    "_residue_key",
    "_atom_name",
    "_element"

]


duplicate_atom_mask = df_spatial_source.duplicated(
    subset=atom_site_columns,
    keep=False
)


duplicate_atom_count = int(
    duplicate_atom_mask.sum()
)


if duplicate_atom_count > 0:

    duplicate_examples = (

        df_spatial_source.loc[
            duplicate_atom_mask,
            atom_site_columns
        ]

        .head(10)

        .to_dict("records")

    )

    raise ValueError(

        "Duplicate atom sites detected. "
        "The spatial graph cannot be built safely until "
        "the source atom table is checked. "
        f"Examples: {duplicate_examples}"

    )


# ------------------------------------------------------------
# 4.12. Unknown element diagnostics
# ------------------------------------------------------------

unknown_element_count = int(
    df_spatial_source["_element"].isin(
        ["", "UNKNOWN", "UNK", "NONE"]
    ).sum()
)


if unknown_element_count > 0:

    print(
        "WARNING:",
        unknown_element_count,
        "atoms have unknown elements."
    )


if len(df_spatial_source) < 2:

    raise ValueError(
        "Fewer than two valid heavy atoms remain."
    )


print("\nATOM DATA PREPARATION")
print("-" * 75)

print("Valid atom records:", len(df_spatial_source))

print("Invalid coordinate records:", invalid_coordinate_count)

print("Excluded H/D atoms:", excluded_hydrogen_count)

print("Unknown element records:", unknown_element_count)

print("Duplicate atom sites:", duplicate_atom_count)

print(
    "Unique residues:",
    df_spatial_source["_residue_key"].nunique()
)


# ============================================================
# 5. ATOM SPATIAL NODES
# ============================================================
#
# Каждый тяжёлый атом становится отдельным узлом.
#
# graph_atom_id:
#   локальный индекс атома в графе.
#
# source_atom_row:
#   индекс исходной строки df_atoms после подготовки.
#
# Все исходные столбцы сохраняются.
# ============================================================

df_atom_spatial_nodes = (
    df_spatial_source.copy(deep=True)
    .reset_index(drop=True)
)


df_atom_spatial_nodes["graph_atom_id"] = np.arange(
    len(df_atom_spatial_nodes),
    dtype=np.int64
)


df_atom_spatial_nodes["source_atom_row"] = np.arange(
    len(df_atom_spatial_nodes),
    dtype=np.int64
)


df_atom_spatial_nodes["pdb_id"] = PDB_ID_VALUE


df_atom_spatial_nodes["x_A"] = (
    df_atom_spatial_nodes[x_col].astype(float)
)

df_atom_spatial_nodes["y_A"] = (
    df_atom_spatial_nodes[y_col].astype(float)
)

df_atom_spatial_nodes["z_A"] = (
    df_atom_spatial_nodes[z_col].astype(float)
)


df_atom_spatial_nodes["element"] = (
    df_atom_spatial_nodes["_element"]
)

df_atom_spatial_nodes["atom_name"] = (
    df_atom_spatial_nodes["_atom_name"]
)

df_atom_spatial_nodes["chain_id"] = (
    df_atom_spatial_nodes["_chain_id"]
)

df_atom_spatial_nodes["residue_name"] = (
    df_atom_spatial_nodes["_resname"]
)

df_atom_spatial_nodes["residue_number"] = (
    df_atom_spatial_nodes["_resnum"]
)

df_atom_spatial_nodes["insertion_code"] = (
    df_atom_spatial_nodes["_icode"]
)

df_atom_spatial_nodes["residue_key"] = (
    df_atom_spatial_nodes["_residue_key"]
)


# ============================================================
# 6. ATOM SPATIAL NEIGHBOR SEARCH
# ============================================================
#
# Используется cKDTree.
#
# В отличие от полной матрицы N x N, дерево позволяет
# эффективно находить только пары соседних атомов.
#
# query_pairs возвращает каждую неориентированную пару
# только один раз.
# ============================================================

coordinates = df_atom_spatial_nodes[
    ["x_A", "y_A", "z_A"]
].to_numpy(
    dtype=np.float64
)


if not np.isfinite(coordinates).all():

    raise ValueError(
        "Non-finite coordinates remain in atom nodes."
    )


spatial_tree = cKDTree(
    coordinates
)


atom_pairs = spatial_tree.query_pairs(

    r=SPATIAL_CUTOFF_A,

    output_type="ndarray"

)


atom_pairs = get_sorted_pair_array(
    atom_pairs
)


print("\nATOM NEIGHBOR SEARCH")
print("-" * 75)

print("Atoms indexed:", len(coordinates))

print("Atom pairs found:", len(atom_pairs))


# ============================================================
# 7. ATOM SPATIAL EDGES
# ============================================================


if len(atom_pairs) > 0:

    atom_i = atom_pairs[:, 0]

    atom_j = atom_pairs[:, 1]


    distances = np.linalg.norm(

        coordinates[atom_i]
        - coordinates[atom_j],

        axis=1

    )


    node_elements = (
        df_atom_spatial_nodes["element"].to_numpy()
    )

    node_residues = (
        df_atom_spatial_nodes["residue_key"].to_numpy()
    )

    node_chains = (
        df_atom_spatial_nodes["chain_id"].to_numpy()
    )

    node_resnames = (
        df_atom_spatial_nodes["residue_name"].to_numpy()
    )

    node_resnums = (
        df_atom_spatial_nodes["residue_number"].to_numpy()
    )

    node_atom_names = (
        df_atom_spatial_nodes["atom_name"].to_numpy()
    )


    same_residue = (
        node_residues[atom_i]
        == node_residues[atom_j]
    )


    df_atom_spatial_edges = pd.DataFrame({

        "pdb_id": PDB_ID_VALUE,

        "atom_i": atom_i,

        "atom_j": atom_j,

        "atom_name_i": node_atom_names[atom_i],

        "atom_name_j": node_atom_names[atom_j],

        "element_i": node_elements[atom_i],

        "element_j": node_elements[atom_j],

        "residue_i": node_residues[atom_i],

        "residue_j": node_residues[atom_j],

        "chain_i": node_chains[atom_i],

        "chain_j": node_chains[atom_j],

        "resname_i": node_resnames[atom_i],

        "resname_j": node_resnames[atom_j],

        "resnum_i": node_resnums[atom_i],

        "resnum_j": node_resnums[atom_j],

        "distance_A": distances,

        "same_residue": same_residue,

        "same_chain": (
            node_chains[atom_i]
            == node_chains[atom_j]
        )

    })


else:

    df_atom_spatial_edges = pd.DataFrame(

        columns=[

            "pdb_id",

            "atom_i",
            "atom_j",

            "atom_name_i",
            "atom_name_j",

            "element_i",
            "element_j",

            "residue_i",
            "residue_j",

            "chain_i",
            "chain_j",

            "resname_i",
            "resname_j",

            "resnum_i",
            "resnum_j",

            "distance_A",

            "same_residue",

            "same_chain"

        ]

    )


# ------------------------------------------------------------
# 7.1. Separate intra-residue and inter-residue contacts
# ------------------------------------------------------------

if not df_atom_spatial_edges.empty:

    df_atom_intra_residue_edges = (

        df_atom_spatial_edges[
            df_atom_spatial_edges["same_residue"]
        ]

        .copy()

    )

    df_atom_inter_residue_edges = (

        df_atom_spatial_edges[
            ~df_atom_spatial_edges["same_residue"]
        ]

        .copy()

    )

else:

    df_atom_intra_residue_edges = (
        df_atom_spatial_edges.copy()
    )

    df_atom_inter_residue_edges = (
        df_atom_spatial_edges.copy()
    )


# ============================================================
# 8. RESIDUE SPATIAL NODES
# ============================================================
#
# Один узел = один аминокислотный остаток.
#
# Центр остатка:
# среднее арифметическое координат его тяжёлых атомов.
#
# Это геометрический центр, а не центр масс.
# ============================================================

df_residue_spatial_nodes = (

    df_atom_spatial_nodes

    .groupby(
        "residue_key",
        sort=False,
        as_index=False
    )

    .agg(

        chain_id=("chain_id", "first"),

        residue_name=("residue_name", "first"),

        residue_number=("residue_number", "first"),

        insertion_code=("insertion_code", "first"),

        atom_count=("graph_atom_id", "count"),

        center_x_A=("x_A", "mean"),

        center_y_A=("y_A", "mean"),

        center_z_A=("z_A", "mean")

    )

)


df_residue_spatial_nodes["pdb_id"] = PDB_ID_VALUE


df_residue_spatial_nodes["graph_residue_id"] = np.arange(
    len(df_residue_spatial_nodes),
    dtype=np.int64
)


# ------------------------------------------------------------
# 8.1. Residue node mapping
# ------------------------------------------------------------

residue_id_map = dict(

    zip(

        df_residue_spatial_nodes["residue_key"],

        df_residue_spatial_nodes["graph_residue_id"]

    )

)


if len(residue_id_map) != len(df_residue_spatial_nodes):

    raise ValueError(
        "Residue graph node identifiers are not unique."
    )


# ============================================================
# 9. RESIDUE SPATIAL EDGES
# ============================================================
#
# Остаточное ребро создаётся, если между двумя разными
# остатками обнаружена хотя бы одна пара атомов в пределах cutoff.
#
# Для каждой пары остатков:
#
# min_distance_A:
#   минимальное расстояние между атомами двух остатков.
#
# mean_contact_distance_A:
#   среднее расстояние среди всех атомных пар,
#   удовлетворяющих cutoff.
#
# atom_contact_count:
#   количество таких атомных пар.
#
# Это не расстояние между центрами остатков.
# ============================================================

if not df_atom_inter_residue_edges.empty:

    df_residue_spatial_edges = (

        df_atom_inter_residue_edges

        .groupby(
            ["residue_i", "residue_j"],
            sort=False,
            as_index=False
        )

        .agg(

            min_distance_A=("distance_A", "min"),

            mean_contact_distance_A=("distance_A", "mean"),

            atom_contact_count=("distance_A", "count")

        )

    )


    # Нормализуем порядок пар по graph_residue_id.
    residue_node_i = (

        df_residue_spatial_edges["residue_i"]
        .map(residue_id_map)
        .to_numpy(dtype=np.int64)

    )

    residue_node_j = (

        df_residue_spatial_edges["residue_j"]
        .map(residue_id_map)
        .to_numpy(dtype=np.int64)

    )


    swap_mask = (
        residue_node_i > residue_node_j
    )


    original_residue_i = (
        df_residue_spatial_edges["residue_i"].to_numpy()
    )

    original_residue_j = (
        df_residue_spatial_edges["residue_j"].to_numpy()
    )


    normalized_residue_i = original_residue_i.copy()

    normalized_residue_j = original_residue_j.copy()


    normalized_residue_i[swap_mask] = (
        original_residue_j[swap_mask]
    )

    normalized_residue_j[swap_mask] = (
        original_residue_i[swap_mask]
    )


    df_residue_spatial_edges["residue_i"] = (
        normalized_residue_i
    )

    df_residue_spatial_edges["residue_j"] = (
        normalized_residue_j
    )


    df_residue_spatial_edges["residue_node_i"] = np.minimum(
        residue_node_i,
        residue_node_j
    )

    df_residue_spatial_edges["residue_node_j"] = np.maximum(
        residue_node_i,
        residue_node_j
    )


    # Добавляем сведения о цепях и типах остатков.
    residue_metadata = (

        df_residue_spatial_nodes

        .set_index("graph_residue_id")

    )


    df_residue_spatial_edges["chain_i"] = (

        df_residue_spatial_edges["residue_node_i"]
        .map(residue_metadata["chain_id"])

    )

    df_residue_spatial_edges["chain_j"] = (

        df_residue_spatial_edges["residue_node_j"]
        .map(residue_metadata["chain_id"])

    )

    df_residue_spatial_edges["resname_i"] = (

        df_residue_spatial_edges["residue_node_i"]
        .map(residue_metadata["residue_name"])

    )

    df_residue_spatial_edges["resname_j"] = (

        df_residue_spatial_edges["residue_node_j"]
        .map(residue_metadata["residue_name"])

    )


    df_residue_spatial_edges["same_chain"] = (

        df_residue_spatial_edges["chain_i"]
        == df_residue_spatial_edges["chain_j"]

    )


    df_residue_spatial_edges["pdb_id"] = PDB_ID_VALUE


    df_residue_spatial_edges = (

        df_residue_spatial_edges

        .sort_values(
            [
                "residue_node_i",
                "residue_node_j"
            ]
        )

        .reset_index(drop=True)

    )


else:

    df_residue_spatial_edges = pd.DataFrame(

        columns=[

            "residue_i",
            "residue_j",

            "min_distance_A",
            "mean_contact_distance_A",

            "atom_contact_count",

            "residue_node_i",
            "residue_node_j",

            "chain_i",
            "chain_j",

            "resname_i",
            "resname_j",

            "same_chain",

            "pdb_id"

        ]

    )


# ============================================================
# 10. GRAPH CONNECTIVITY STATISTICS
# ============================================================


atom_edge_array = (

    df_atom_spatial_edges[
        ["atom_i", "atom_j"]
    ]

    .to_numpy(dtype=np.int64)

    if not df_atom_spatial_edges.empty

    else np.empty((0, 2), dtype=np.int64)

)


residue_edge_array = (

    df_residue_spatial_edges[
        ["residue_node_i", "residue_node_j"]
    ]

    .to_numpy(dtype=np.int64)

    if not df_residue_spatial_edges.empty

    else np.empty((0, 2), dtype=np.int64)

)


atom_graph_stats = graph_statistics(

    node_count=len(df_atom_spatial_nodes),

    edge_array=atom_edge_array

)


residue_graph_stats = graph_statistics(

    node_count=len(df_residue_spatial_nodes),

    edge_array=residue_edge_array

)


# ============================================================
# 11. DISTANCE STATISTICS
# ============================================================


if not df_atom_spatial_edges.empty:

    atom_distance_stats = {

        "min_distance_A": float(
            df_atom_spatial_edges["distance_A"].min()
        ),

        "mean_distance_A": float(
            df_atom_spatial_edges["distance_A"].mean()
        ),

        "median_distance_A": float(
            df_atom_spatial_edges["distance_A"].median()
        ),

        "max_distance_A": float(
            df_atom_spatial_edges["distance_A"].max()
        )

    }

else:

    atom_distance_stats = {

        "min_distance_A": None,

        "mean_distance_A": None,

        "median_distance_A": None,

        "max_distance_A": None

    }


if not df_residue_spatial_edges.empty:

    residue_distance_stats = {

        "min_distance_A": float(
            df_residue_spatial_edges["min_distance_A"].min()
        ),

        "mean_min_distance_A": float(
            df_residue_spatial_edges["min_distance_A"].mean()
        ),

        "mean_contact_distance_A": float(
            df_residue_spatial_edges[
                "mean_contact_distance_A"
            ].mean()
        ),

        "max_min_distance_A": float(
            df_residue_spatial_edges["min_distance_A"].max()
        )

    }

else:

    residue_distance_stats = {

        "min_distance_A": None,

        "mean_min_distance_A": None,

        "mean_contact_distance_A": None,

        "max_min_distance_A": None

    }


# ============================================================
# 12. CHAIN-LEVEL GRAPH STATISTICS
# ============================================================
#
# Дополнительно рассчитываем число узлов по цепям.
# Это помогает анализировать многосубъединичные комплексы.
# ============================================================

df_atom_nodes_by_chain = (

    df_atom_spatial_nodes

    .groupby(
        "chain_id",
        sort=False
    )

    .agg(

        atom_node_count=("graph_atom_id", "count"),

        residue_count=("residue_key", "nunique")

    )

    .reset_index()

)


df_atom_edges_by_chain = pd.DataFrame()


if not df_atom_spatial_edges.empty:

    df_atom_edges_by_chain = (

        df_atom_spatial_edges

        .groupby(
            "chain_i",
            sort=False
        )

        .agg(
            atom_edge_count=("distance_A", "count")
        )

        .reset_index()

        .rename(
            columns={
                "chain_i": "chain_id"
            }
        )

    )


if not df_atom_edges_by_chain.empty:

    df_chain_spatial_statistics = (

        df_atom_nodes_by_chain

        .merge(
            df_atom_edges_by_chain,
            on="chain_id",
            how="left"
        )

    )

else:

    df_chain_spatial_statistics = (
        df_atom_nodes_by_chain.copy()
    )

    df_chain_spatial_statistics["atom_edge_count"] = 0


df_chain_spatial_statistics["atom_edge_count"] = (

    df_chain_spatial_statistics["atom_edge_count"]
    .fillna(0)
    .astype(int)

)


# Межцепочечные атомные контакты.
if not df_atom_spatial_edges.empty:

    df_inter_chain_edges = (

        df_atom_spatial_edges[
            ~df_atom_spatial_edges["same_chain"]
        ]

        .copy()

    )

else:

    df_inter_chain_edges = (
        df_atom_spatial_edges.copy()
    )


# ============================================================
# 13. GRAPH SUMMARY
# ============================================================


spatial_graph_summary = {

    "metadata": {

        "pdb_id": PDB_ID_VALUE,

        "created_at": datetime.now().isoformat(
            timespec="seconds"
        ),

        "cell": 23,

        "analysis_scope": ANALYSIS_SCOPE,

        "spatial_cutoff_A": float(
            SPATIAL_CUTOFF_A
        ),

        "hydrogen_excluded": bool(
            EXCLUDE_HYDROGEN
        ),

        "distance_tolerance": float(
            DISTANCE_TOLERANCE
        ),

        "coordinate_columns": {

            "x": str(x_col),

            "y": str(y_col),

            "z": str(z_col)

        }

    },

    "input_quality": {

        "input_atom_rows": int(len(df_atoms)),

        "valid_coordinate_atom_rows": int(
            len(df_spatial_source)
        ),

        "invalid_coordinate_rows": invalid_coordinate_count,

        "excluded_hydrogen_deuterium_atoms": (
            excluded_hydrogen_count
        ),

        "unknown_element_count": unknown_element_count,

        "duplicate_atom_site_count": duplicate_atom_count,

        "unique_residue_count": int(
            df_spatial_source["_residue_key"].nunique()
        )

    },

    "atom_graph": {

        **atom_graph_stats,

        "distance_statistics": atom_distance_stats,

        "intra_residue_edge_count": int(
            len(df_atom_intra_residue_edges)
        ),

        "inter_residue_edge_count": int(
            len(df_atom_inter_residue_edges)
        ),

        "inter_chain_edge_count": int(
            len(df_inter_chain_edges)
        )

    },

    "residue_graph": {

        **residue_graph_stats,

        "distance_statistics": residue_distance_stats

    },

    "chain_statistics": {

        "chain_count": int(
            len(df_chain_spatial_statistics)
        ),

        "inter_chain_edge_count": int(
            len(df_inter_chain_edges)
        )

    },

    "interpretation": {

        "atom_edges": (
            "An atom edge represents spatial proximity "
            "within the specified cutoff. It is not automatically "
            "a chemical bond."
        ),

        "residue_edges": (
            "A residue edge exists when at least one atom pair "
            "between two different residues is within the cutoff."
        ),

        "residue_center": (
            "Residue coordinates represent the arithmetic mean "
            "of heavy-atom coordinates, not the center of mass."
        ),

        "distance": (
            "All distances are expressed in Angstrom."
        ),

        "model_scope": (
            "The input atom table is expected to represent "
            "the first structural model."
        )

    }

}


# ============================================================
# 14. OUTPUT FILES
# ============================================================


spatial_files = {

    "atom_nodes":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_atom_nodes.csv",

    "atom_edges":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_atom_edges.csv",

    "atom_intra_residue_edges":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_atom_intra_residue_edges.csv",

    "atom_inter_residue_edges":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_atom_inter_residue_edges.csv",

    "residue_nodes":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_residue_nodes.csv",

    "residue_edges":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_residue_edges.csv",

    "chain_statistics":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_chain_statistics.csv",

    "inter_chain_edges":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_inter_chain_edges.csv",

    "summary":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_graph_summary.json"

}


# ============================================================
# 15. SAVE CSV TABLES
# ============================================================


print("\n" + "=" * 75)
print("SAVING SPATIAL GRAPH TABLES")
print("=" * 75)


output_tables = {

    "atom_nodes": df_atom_spatial_nodes,

    "atom_edges": df_atom_spatial_edges,

    "atom_intra_residue_edges": (
        df_atom_intra_residue_edges
    ),

    "atom_inter_residue_edges": (
        df_atom_inter_residue_edges
    ),

    "residue_nodes": df_residue_spatial_nodes,

    "residue_edges": df_residue_spatial_edges,

    "chain_statistics": df_chain_spatial_statistics,

    "inter_chain_edges": df_inter_chain_edges

}


for table_name, dataframe in output_tables.items():

    output_path = spatial_files[table_name]

    dataframe.to_csv(

        output_path,

        index=False,

        encoding="utf-8-sig"

    )

    if not output_path.is_file():

        raise RuntimeError(
            f"Failed to save {output_path}"
        )

    print(

        f"{output_path.name:65s} "
        f"{len(dataframe):9d} rows"

    )


# ============================================================
# 16. SAVE JSON SUMMARY
# ============================================================


with open(

    spatial_files["summary"],

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        safe_json_value(spatial_graph_summary),

        file,

        ensure_ascii=False,

        indent=4,

        allow_nan=False

    )


# ============================================================
# 17. VALIDATION
# ============================================================


print("\n" + "=" * 75)
print("SPATIAL GRAPH VALIDATION")
print("=" * 75)


# ------------------------------------------------------------
# 17.1. Node count checks
# ------------------------------------------------------------

assert (
    len(df_atom_spatial_nodes)
    == atom_graph_stats["node_count"]
)

assert (
    len(df_residue_spatial_nodes)
    == residue_graph_stats["node_count"]
)


# ------------------------------------------------------------
# 17.2. Atom edge checks
# ------------------------------------------------------------

if not df_atom_spatial_edges.empty:

    assert (
        df_atom_spatial_edges["atom_i"]
        < df_atom_spatial_edges["atom_j"]
    ).all()

    assert (
        df_atom_spatial_edges["distance_A"]
        > 0
    ).all()

    assert (
        df_atom_spatial_edges["distance_A"]
        <= SPATIAL_CUTOFF_A + DISTANCE_TOLERANCE
    ).all()

    assert (
        df_atom_spatial_edges[
            ["atom_i", "atom_j"]
        ]

        .duplicated()

        .sum()
        == 0
    )


# ------------------------------------------------------------
# 17.3. Residue edge checks
# ------------------------------------------------------------

if not df_residue_spatial_edges.empty:

    assert (
        df_residue_spatial_edges["residue_node_i"]
        < df_residue_spatial_edges["residue_node_j"]
    ).all()

    assert (
        df_residue_spatial_edges["min_distance_A"]
        > 0
    ).all()

    assert (
        df_residue_spatial_edges["min_distance_A"]
        <= SPATIAL_CUTOFF_A + DISTANCE_TOLERANCE
    ).all()

    assert (
        df_residue_spatial_edges[
            ["residue_node_i", "residue_node_j"]
        ]

        .duplicated()

        .sum()
        == 0
    )


# ------------------------------------------------------------
# 17.4. Residue mapping checks
# ------------------------------------------------------------

if not df_residue_spatial_edges.empty:

    assert (
        df_residue_spatial_edges[
            ["residue_node_i", "residue_node_j"]
        ]

        .notna()

        .all()

        .all()
    )

    valid_node_ids = set(
        df_residue_spatial_nodes["graph_residue_id"]
    )

    edge_node_ids = set(

        df_residue_spatial_edges[
            ["residue_node_i", "residue_node_j"]
        ]

        .to_numpy()
        .ravel()

    )

    assert edge_node_ids.issubset(
        valid_node_ids
    )


# ------------------------------------------------------------
# 17.5. File checks
# ------------------------------------------------------------

for name, path in spatial_files.items():

    if not path.is_file():

        raise RuntimeError(
            f"Output file missing: {path}"
        )


# ------------------------------------------------------------
# 17.6. JSON check
# ------------------------------------------------------------

with open(
    spatial_files["summary"],
    "r",
    encoding="utf-8"
) as file:

    saved_summary = json.load(file)


if saved_summary["metadata"]["pdb_id"] != PDB_ID_VALUE:

    raise ValueError(
        "PDB ID mismatch in spatial graph JSON."
    )


# ============================================================
# 18. FINAL REPORT
# ============================================================


print("\n" + "=" * 75)
print("ATOM SPATIAL GRAPH")
print("=" * 75)

for key, value in atom_graph_stats.items():

    print(f"{key:35s}: {value}")


print("\nATOM DISTANCE STATISTICS")

for key, value in atom_distance_stats.items():

    print(f"{key:35s}: {value}")


print("\nINTRA-RESIDUE ATOM EDGES:",
      len(df_atom_intra_residue_edges))

print("INTER-RESIDUE ATOM EDGES:",
      len(df_atom_inter_residue_edges))

print("INTER-CHAIN ATOM EDGES:",
      len(df_inter_chain_edges))


print("\n" + "=" * 75)
print("RESIDUE SPATIAL GRAPH")
print("=" * 75)

for key, value in residue_graph_stats.items():

    print(f"{key:35s}: {value}")


print("\nRESIDUE DISTANCE STATISTICS")

for key, value in residue_distance_stats.items():

    print(f"{key:35s}: {value}")


print("\n" + "=" * 75)
print("CHAIN STATISTICS")
print("=" * 75)

print(
    df_chain_spatial_statistics.to_string(
        index=False
    )
)


print("\n" + "=" * 75)
print("SAVED FILES")
print("=" * 75)

for name, path in spatial_files.items():

    print(
        f"{name:30s}: "
        f"{'OK' if path.is_file() else 'MISSING'}"
    )


print("\nOutput directory:")
print(FEATURE_DIR)


print("\nSTATUS: COMPLETED")

print("CELL 23 COMPLETED")

print("=" * 75)
```

# подготовленное представление для следующего этапа машинного обучения.

```python
# ============================================================
# CELL 24. UNIFIED PROTEIN STRUCTURAL REPRESENTATION
# ============================================================
#
# НАЗНАЧЕНИЕ:
# Объединение результатов Cell 22 и Cell 23 в единое
# машинно-ориентированное представление структуры белка.
#
# СОЗДАВАЕМЫЕ ПРЕДСТАВЛЕНИЯ:
#
# ATOM GRAPH:
# X_atom                  - признаки атомов
# pos_atom                - координаты атомов
# edge_index_atom         - индексы атомных рёбер
# edge_attr_atom          - признаки атомных рёбер
# atom_node_ids           - исходные ID атомных узлов
#
# RESIDUE GRAPH:
# X_residue               - признаки аминокислотных остатков
# pos_residue             - координаты центров остатков
# edge_index_residue      - индексы остаточных рёбер
# edge_attr_residue       - признаки остаточных рёбер
# residue_node_ids        - исходные ID остатков
#
# GLOBAL:
# X_global                - глобальные признаки белка
#
# BINDING SITE:
# is_binding_site_atom    - атом принадлежит остатку binding site
# is_direct_contact_atom  - атом непосредственно контактирует
#                           с выбранным компонентом
# is_binding_site_residue - остаток входит в binding site
#
# ВАЖНО:
# Пространственные контакты не являются автоматически
# химическими связями.
#
# На этом этапе обучение модели не выполняется.
# ============================================================


import os
import json

import numpy as np
import pandas as pd

from pathlib import Path


print("=" * 70)
print("CELL 24. UNIFIED PROTEIN STRUCTURAL REPRESENTATION")
print("=" * 70)


# ============================================================
# 1. PATHS
# ============================================================

PDB_NAME = str(
    globals().get("PDB_ID", "UNKNOWN")
)


if "FEATURE_DIR" in globals():

    feature_dir_24 = Path(FEATURE_DIR)

elif "feature_dir" in globals():

    feature_dir_24 = Path(feature_dir)

else:

    feature_dir_24 = (
        Path("PDB_Protein_Analysis")
        / "results"
        / "features"
    )


output_dir_24 = (
    feature_dir_24.parent
    / "unified_representation"
)

output_dir_24.mkdir(
    parents=True,
    exist_ok=True
)


print(f"PDB ID: {PDB_NAME}")
print(f"Input directory: {feature_dir_24}")
print(f"Output directory: {output_dir_24}")


# ============================================================
# 2. LOADING CELL 22 AND CELL 23 DATA
# ============================================================


def load_csv_24(filename, required=True):

    path = feature_dir_24 / filename

    if not path.exists():

        if required:

            raise FileNotFoundError(
                f"Не найден файл: {path}\n"
                "Проверьте выполнение ячеек 22 и 23."
            )

        return pd.DataFrame()

    return pd.read_csv(path)


# ----------------------------
# CELL 23: SPATIAL GRAPHS
# ----------------------------

df_atom_nodes_24 = load_csv_24(
    f"{PDB_NAME}_spatial_atom_nodes.csv"
)

df_atom_edges_24 = load_csv_24(
    f"{PDB_NAME}_spatial_atom_edges.csv"
)

df_residue_nodes_24 = load_csv_24(
    f"{PDB_NAME}_spatial_residue_nodes.csv"
)

df_residue_edges_24 = load_csv_24(
    f"{PDB_NAME}_spatial_residue_edges.csv"
)


# ----------------------------
# CELL 22: PROTEIN FEATURES
# ----------------------------

df_global_24 = load_csv_24(
    f"{PDB_NAME}_global_features.csv",
    required=False
)

df_binding_24 = load_csv_24(
    f"{PDB_NAME}_binding_site_features.csv",
    required=False
)

df_interactions_24 = load_csv_24(
    f"{PDB_NAME}_interaction_features.csv",
    required=False
)

df_metal_24 = load_csv_24(
    f"{PDB_NAME}_metal_features.csv",
    required=False
)

df_atom_features_24 = load_csv_24(
    f"{PDB_NAME}_atom_features.csv",
    required=False
)

df_residue_features_24 = load_csv_24(
    f"{PDB_NAME}_residue_features.csv",
    required=False
)

df_binding_contacts_24 = load_csv_24(
    f"{PDB_NAME}_binding_atom_contact_features.csv",
    required=False
)


print("\nINPUT DATA")
print("-" * 70)


input_tables_24 = {

    "Atom nodes": df_atom_nodes_24,
    "Atom edges": df_atom_edges_24,

    "Residue nodes": df_residue_nodes_24,
    "Residue edges": df_residue_edges_24,

    "Global features": df_global_24,
    "Binding site": df_binding_24,

    "Binding atom contacts": df_binding_contacts_24,

    "Interactions": df_interactions_24,
    "Metal features": df_metal_24

}


for name, dataframe in input_tables_24.items():

    print(
        f"{name:27s}: {len(dataframe):8d} rows"
    )


if df_atom_nodes_24.empty:

    raise ValueError(
        "Таблица атомных узлов пуста."
    )


if df_residue_nodes_24.empty:

    raise ValueError(
        "Таблица остаточных узлов пуста."
    )


# ============================================================
# 3. HELPER FUNCTIONS
# ============================================================


def find_column_24(dataframe, candidates):
    """
    Поиск столбца с учётом регистра и разделителей.
    """

    normalized = {

        "".join(
            ch.lower()
            for ch in str(col)
            if ch.isalnum()
        ): col

        for col in dataframe.columns

    }

    for candidate in candidates:

        key = "".join(
            ch.lower()
            for ch in str(candidate)
            if ch.isalnum()
        )

        if key in normalized:

            return normalized[key]

    return None


def normalize_text_24(value):
    """
    Нормализация текстовых идентификаторов.
    """

    if pd.isna(value):

        return ""

    return str(value).strip()


def normalize_number_24(value):
    """
    Нормализация номера остатка.
    """

    if pd.isna(value):

        return None

    try:

        numeric_value = float(value)

        if not np.isfinite(numeric_value):

            raise ValueError(
                f"Некорректный номер: {value}"
            )

        if not numeric_value.is_integer():

            raise ValueError(
                f"Номер остатка не является целым: {value}"
            )

        return int(numeric_value)

    except (ValueError, TypeError):

        raise ValueError(
            f"Невозможно нормализовать номер остатка: {value}"
        )


def normalize_icode_24(value):
    """
    Пустой или отсутствующий insertion code -> ''.
    """

    if pd.isna(value):

        return ""

    return str(value).strip()


def make_residue_key_24(
    chain_id,
    residue_name,
    residue_number,
    insertion_code=""
):
    """
    Унифицированный идентификатор остатка.

    Формат:
    (
        chain_id,
        residue_name,
        residue_number,
        insertion_code
    )

    Исходный residue_key не используется,
    поскольку Cell 22 и Cell 23 могут формировать
    его в разных форматах.
    """

    return (

        normalize_text_24(chain_id),

        normalize_text_24(
            residue_name
        ).upper(),

        normalize_number_24(
            residue_number
        ),

        normalize_icode_24(
            insertion_code
        )

    )


def clean_numeric_matrix_24(dataframe):
    """
    Перевод признаков в float32.

    NaN и +/-inf заменяются нулями.

    Стандартизация не выполняется.
    """

    if dataframe.shape[1] == 0:

        return np.zeros(
            (len(dataframe), 0),
            dtype=np.float32
        )

    result = dataframe.copy()

    for col in result.columns:

        result[col] = pd.to_numeric(
            result[col],
            errors="coerce"
        )

        result[col] = result[col].replace(
            [np.inf, -np.inf],
            np.nan
        )

        result[col] = result[col].fillna(0.0)

    return result.to_numpy(
        dtype=np.float32
    )


def get_node_ids_24(
    dataframe,
    candidates,
    object_name
):
    """
    Получение исходных ID узлов.

    Если ID отсутствует, используются позиции строк.

    Проверяются:
    - отсутствие пропусков;
    - целочисленность;
    - уникальность.
    """

    col = find_column_24(
        dataframe,
        candidates
    )

    if col is None:

        print(
            f"WARNING: {object_name}: ID column not found. "
            "Используются позиции строк."
        )

        return np.arange(
            len(dataframe),
            dtype=np.int64
        )

    ids = pd.to_numeric(
        dataframe[col],
        errors="coerce"
    )

    if ids.isna().any():

        raise ValueError(
            f"{object_name}: обнаружены некорректные ID."
        )

    values = ids.to_numpy(
        dtype=np.float64
    )

    if not np.isfinite(values).all():

        raise ValueError(
            f"{object_name}: ID содержат NaN или inf."
        )

    if not np.all(
        values == np.floor(values)
    ):

        raise ValueError(
            f"{object_name}: обнаружены нецелочисленные ID."
        )

    ids = values.astype(
        np.int64
    )

    if len(np.unique(ids)) != len(ids):

        raise ValueError(
            f"{object_name}: ID узлов не уникальны."
        )

    return ids


def make_bidirectional_edges_24(
    edge_dataframe,
    source_col,
    target_col,
    feature_columns,
    node_ids,
    object_name
):
    """
    Создание двунаправленных рёбер.

    Исходные ID преобразуются в позиции строк
    соответствующей матрицы узлов.

    edge_index содержит позиции строк, а не исходные ID.
    """

    if edge_dataframe.empty:

        return (

            np.empty(
                (2, 0),
                dtype=np.int64
            ),

            np.empty(
                (0, len(feature_columns)),
                dtype=np.float32
            )

        )

    if source_col not in edge_dataframe.columns:

        raise ValueError(
            f"{object_name}: отсутствует {source_col}."
        )

    if target_col not in edge_dataframe.columns:

        raise ValueError(
            f"{object_name}: отсутствует {target_col}."
        )

    source_raw = pd.to_numeric(
        edge_dataframe[source_col],
        errors="coerce"
    )

    target_raw = pd.to_numeric(
        edge_dataframe[target_col],
        errors="coerce"
    )

    if source_raw.isna().any() or target_raw.isna().any():

        raise ValueError(
            f"{object_name}: некорректные индексы рёбер."
        )

    source_values = source_raw.to_numpy(
        dtype=np.float64
    )

    target_values = target_raw.to_numpy(
        dtype=np.float64
    )

    if not np.all(
        source_values == np.floor(source_values)
    ):

        raise ValueError(
            f"{object_name}: source ID нецелочисленные."
        )

    if not np.all(
        target_values == np.floor(target_values)
    ):

        raise ValueError(
            f"{object_name}: target ID нецелочисленные."
        )

    source_raw = source_values.astype(
        np.int64
    )

    target_raw = target_values.astype(
        np.int64
    )

    id_to_position = {

        int(node_id): position

        for position, node_id in enumerate(node_ids)

    }

    unknown_ids = (

        set(source_raw.tolist())
        | set(target_raw.tolist())

    ) - set(id_to_position.keys())

    if unknown_ids:

        raise ValueError(
            f"{object_name}: рёбра содержат неизвестные ID: "
            f"{list(unknown_ids)[:10]}"
        )

    source = np.array(

        [
            id_to_position[int(value)]
            for value in source_raw
        ],

        dtype=np.int64

    )

    target = np.array(

        [
            id_to_position[int(value)]
            for value in target_raw
        ],

        dtype=np.int64

    )

    if feature_columns:

        edge_features = clean_numeric_matrix_24(
            edge_dataframe[feature_columns]
        )

    else:

        edge_features = np.empty(
            (len(source), 0),
            dtype=np.float32
        )

    no_self_loop = source != target

    source = source[no_self_loop]
    target = target[no_self_loop]

    edge_features = edge_features[
        no_self_loop
    ]

    edge_index = np.vstack([

        np.concatenate([
            source,
            target
        ]),

        np.concatenate([
            target,
            source
        ])

    ])

    edge_attr = np.concatenate([

        edge_features,
        edge_features

    ], axis=0)

    return edge_index, edge_attr


# ============================================================
# 4. NODE IDENTIFIERS
# ============================================================

# Проверяем реальные поля атомных узлов.

required_atom_columns_24 = [

    "chain_id",
    "residue_name",
    "residue_number",
    "insertion_code"

]


for col in required_atom_columns_24:

    if col not in df_atom_nodes_24.columns:

        raise ValueError(
            f"В atom nodes отсутствует столбец: {col}"
        )


# Поля остаточных узлов.
required_residue_columns_24 = [

    "chain_id",
    "residue_name",
    "residue_number",
    "graph_residue_id"

]


for col in required_residue_columns_24:

    if col not in df_residue_nodes_24.columns:

        raise ValueError(
            f"В residue nodes отсутствует столбец: {col}"
        )


# Поля binding site.
required_binding_columns_24 = [

    "protein_chain",
    "residue_name",
    "residue_number",
    "insertion_code"

]


if not df_binding_24.empty:

    for col in required_binding_columns_24:

        if col not in df_binding_24.columns:

            raise ValueError(
                f"В binding site отсутствует столбец: {col}"
            )


# ============================================================
# 5. UNIFIED RESIDUE IDENTIFIERS
# ============================================================
#
# Cell 23:
# residue_key = A: :1:
#
# Cell 22:
# residue_key = A:HIS:240
#
# Исходные строки не сравниваем.
#
# Используем составной ключ:
# chain + residue name + residue number + insertion code.
# ============================================================


# ------------------------------------------------------------
# 5.1. Создаём карту insertion_code по атомам
# ------------------------------------------------------------

atom_icode_map_24 = {}


for row in df_atom_nodes_24[
    required_atom_columns_24
].itertuples(index=False, name=None):

    chain_id, residue_name, residue_number, insertion_code = row

    base_key = (

        normalize_text_24(chain_id),

        normalize_text_24(residue_name).upper(),

        normalize_number_24(residue_number)

    )

    icode = normalize_icode_24(
        insertion_code
    )

    atom_icode_map_24.setdefault(
        base_key,
        set()
    ).add(icode)


# ------------------------------------------------------------
# 5.2. Ключи атомных узлов
# ------------------------------------------------------------

atom_residue_keys_24 = [

    make_residue_key_24(*row)

    for row in df_atom_nodes_24[
        required_atom_columns_24
    ].itertuples(index=False, name=None)

]


# ------------------------------------------------------------
# 5.3. Ключи остаточных узлов
# ------------------------------------------------------------

residue_node_keys_24 = []


for row in df_residue_nodes_24[
    required_residue_columns_24[:3]
].itertuples(index=False, name=None):

    chain_id, residue_name, residue_number = row

    base_key = (

        normalize_text_24(chain_id),

        normalize_text_24(residue_name).upper(),

        normalize_number_24(residue_number)

    )

    possible_icodes = atom_icode_map_24.get(
        base_key,
        set()
    )

    if len(possible_icodes) == 0:

        raise ValueError(
            f"Для остатка {base_key} не найдены атомы."
        )

    if len(possible_icodes) > 1:

        raise ValueError(
            f"Для остатка {base_key} обнаружено несколько "
            f"insertion_code: {possible_icodes}."
        )

    insertion_code = next(
        iter(possible_icodes)
    )

    residue_node_keys_24.append(

        make_residue_key_24(

            chain_id,
            residue_name,
            residue_number,
            insertion_code

        )

    )


# ------------------------------------------------------------
# 5.4. Ключи binding site
# ------------------------------------------------------------

binding_residue_keys_24 = set()


if not df_binding_24.empty:

    for row in df_binding_24[
        required_binding_columns_24
    ].itertuples(index=False, name=None):

        protein_chain, residue_name, residue_number, insertion_code = row

        binding_residue_keys_24.add(

            make_residue_key_24(

                protein_chain,
                residue_name,
                residue_number,
                insertion_code

            )

        )


# ------------------------------------------------------------
# 5.5. Маски binding site
# ------------------------------------------------------------

is_binding_site_residue_24 = np.array(

    [
        key in binding_residue_keys_24
        for key in residue_node_keys_24
    ],

    dtype=np.float32

)


is_binding_site_atom_24 = np.array(

    [
        key in binding_residue_keys_24
        for key in atom_residue_keys_24
    ],

    dtype=np.float32

)


matched_residue_keys_24 = (

    set(residue_node_keys_24)
    & binding_residue_keys_24

)


unmatched_residue_keys_24 = (

    binding_residue_keys_24
    - set(residue_node_keys_24)

)


matched_residues_24 = int(
    is_binding_site_residue_24.sum()
)

matched_atoms_24 = int(
    is_binding_site_atom_24.sum()
)


print("\nRESIDUE IDENTIFIER VALIDATION")
print("-" * 70)

print(
    "Unique binding site residues:",
    len(binding_residue_keys_24)
)

print(
    "Matched residue nodes:",
    matched_residues_24
)

print(
    "Matched atom nodes:",
    matched_atoms_24
)

print(
    "Total residue nodes:",
    len(residue_node_keys_24)
)

print(
    "Total atom nodes:",
    len(atom_residue_keys_24)
)


if unmatched_residue_keys_24:

    raise ValueError(
        "Не все binding site residues найдены в residue graph: "
        f"{unmatched_residue_keys_24}"
    )


if matched_residues_24 != len(binding_residue_keys_24):

    raise ValueError(
        "Количество сопоставленных остатков не совпадает "
        "с количеством остатков binding site."
    )


print(
    "\nВсе binding site residues успешно сопоставлены."
)


# ------------------------------------------------------------
# 5.6. Совместимость с именами переменных Cell 24
# ------------------------------------------------------------

# Эти имена используются в последующих разделах.
# Повторное формирование ключей не требуется.

residue_keys_24 = residue_node_keys_24

atom_binding_mask_24 = is_binding_site_atom_24

residue_binding_mask_24 = is_binding_site_residue_24


# ============================================================
# ============================================================
# 6. DIRECT CONTACT ATOM IDENTIFICATION
# ============================================================
#
# Определяем атомы белка, непосредственно контактирующие
# с исследуемым компонентом.
#
# Используем реальные столбцы таблицы Cell 22:
#
# protein_chain
# residue_name
# residue_number
# insertion_code
# protein_atom
#
# Сопоставление выполняется по двум идентификаторам:
# 1. Составной ключ остатка.
# 2. Название атома.
#
# Результат:
# direct_contact_mask_24
#
# ============================================================


direct_contact_mask_24 = np.zeros(
    len(df_atom_nodes_24),
    dtype=np.float32
)

direct_contact_available_24 = False

direct_contact_mapping_info_24 = {

    "source": None,

    "contact_rows": int(
        len(df_binding_contacts_24)
    ),

    "unique_contact_atoms": 0,

    "matched_contact_atoms": 0,

    "unmatched_contact_atoms": 0

}


if not df_binding_contacts_24.empty:

    # --------------------------------------------------------
    # 6.1. Определение столбцов таблицы контактов
    # --------------------------------------------------------

    contact_chain_col_24 = find_column_24(
        df_binding_contacts_24,
        [
            "protein_chain",
            "chain_id"
        ]
    )

    contact_residue_name_col_24 = find_column_24(
        df_binding_contacts_24,
        [
            "protein_residue_name",
            "residue_name"
        ]
    )

    contact_residue_number_col_24 = find_column_24(
        df_binding_contacts_24,
        [
            "protein_residue_number",
            "residue_number"
        ]
    )

    contact_icode_col_24 = find_column_24(
        df_binding_contacts_24,
        [
            "protein_insertion_code",
            "insertion_code",
            "icode"
        ]
    )

    # ВАЖНО:
    # В твоей таблице используется именно protein_atom.
    # Добавили его в список допустимых названий.

    contact_atom_name_col_24 = find_column_24(
        df_binding_contacts_24,
        [
            "protein_atom_name",
            "protein_contact_atom",
            "closest_protein_atom",
            "protein_atom",
            "atom_name"
        ]
    )

    # Столбец имени атома в таблице атомных узлов.

    atom_atom_name_col_24 = find_column_24(
        df_atom_nodes_24,
        [
            "atom_name"
        ]
    )

    contact_columns_24 = {

        "chain": contact_chain_col_24,

        "residue_name": contact_residue_name_col_24,

        "residue_number": contact_residue_number_col_24,

        "insertion_code": contact_icode_col_24,

        "atom_name": contact_atom_name_col_24

    }

    missing_contact_columns_24 = [

        name

        for name, col in contact_columns_24.items()

        if col is None

    ]

    if atom_atom_name_col_24 is None:

        missing_contact_columns_24.append(
            "atom_name in atom nodes"
        )


    # --------------------------------------------------------
    # 6.2. Проверка доступности идентификаторов
    # --------------------------------------------------------

    if not missing_contact_columns_24:

        # ----------------------------------------------------
        # 6.3. Формирование множества контактных атомов
        # ----------------------------------------------------

        direct_contact_keys_24 = set()

        for _, row in df_binding_contacts_24.iterrows():

            residue_key = make_residue_key_24(

                row[contact_chain_col_24],

                row[contact_residue_name_col_24],

                row[contact_residue_number_col_24],

                row[contact_icode_col_24]

            )

            atom_name = normalize_text_24(
                row[contact_atom_name_col_24]
            ).upper()

            if atom_name:

                direct_contact_keys_24.add(

                    (
                        residue_key,
                        atom_name
                    )

                )


        # ----------------------------------------------------
        # 6.4. Сопоставление с атомными узлами
        # ----------------------------------------------------

        matched_contact_keys_24 = set()

        for atom_index, (_, row) in enumerate(
            df_atom_nodes_24.iterrows()
        ):

            residue_key = atom_residue_keys_24[
                atom_index
            ]

            atom_name = normalize_text_24(
                row[atom_atom_name_col_24]
            ).upper()

            atom_key = (
                residue_key,
                atom_name
            )

            if atom_key in direct_contact_keys_24:

                direct_contact_mask_24[
                    atom_index
                ] = 1.0

                matched_contact_keys_24.add(
                    atom_key
                )


        # ----------------------------------------------------
        # 6.5. Расчёт результатов сопоставления
        # ----------------------------------------------------

        unmatched_contact_keys_24 = (

            direct_contact_keys_24
            - matched_contact_keys_24

        )

        direct_contact_available_24 = True

        direct_contact_mapping_info_24 = {

            "source": "binding_atom_contact_features",

            "contact_rows": int(
                len(df_binding_contacts_24)
            ),

            "unique_contact_atoms": int(
                len(direct_contact_keys_24)
            ),

            "matched_contact_atoms": int(
                len(matched_contact_keys_24)
            ),

            "unmatched_contact_atoms": int(
                len(unmatched_contact_keys_24)
            )

        }

    else:

        print(
            "\nWARNING: Не удалось определить все "
            "идентификаторы прямых атомных контактов."
        )

        print(
            "Missing fields:",
            missing_contact_columns_24
        )

        print(
            "Available contact columns:",
            df_binding_contacts_24.columns.tolist()
        )


# ------------------------------------------------------------
# 6.6. Вывод результатов
# ------------------------------------------------------------

print("\nDIRECT CONTACT ATOMS")
print("-" * 70)

print(
    "Direct contact feature available:",
    direct_contact_available_24
)

print(
    "Direct contact atoms matched:",
    int(direct_contact_mask_24.sum())
)

print(
    "Direct contact mapping:",
    direct_contact_mapping_info_24
)
# ============================================================
# 7. ATOM NODE FEATURES
# ============================================================
#
# X_atom:
# - числовые характеристики атомов;
# - one-hot элемент;
# - is_binding_site_atom;
# - is_direct_contact_atom.
#
# Координаты хранятся отдельно в pos_atom.
# ============================================================


atom_element_col_24 = find_column_24(

    df_atom_nodes_24,

    ["element"]

)


if atom_element_col_24 is None:

    raise ValueError(
        "В атомных узлах отсутствует element."
    )


ELEMENTS_24 = [

    "C",
    "N",
    "O",
    "S",
    "P",
    "F",
    "CL",
    "BR",
    "I",
    "MG",
    "ZN",
    "FE",
    "CU",
    "CA",
    "NA",
    "K",
    "MN",
    "CO",
    "NI",
    "OTHER"

]


atom_elements_24 = (

    df_atom_nodes_24[
        atom_element_col_24
    ]

    .fillna("OTHER")

    .astype(str)

    .str.strip()

    .str.upper()

)


atom_elements_24 = atom_elements_24.where(

    atom_elements_24.isin(ELEMENTS_24),

    "OTHER"

)


atom_element_features_24 = np.column_stack([

    (
        atom_elements_24 == element
    ).astype(np.float32)

    for element in ELEMENTS_24

])


excluded_atom_columns_24 = {

    "graphatomid",
    "atomserial",
    "atomid",
    "serial",

    "resid",
    "residueid",
    "residuenumber",
    "resnum",
    "resseq",
    "modelid",

    "x",
    "y",
    "z",

    "xa",
    "ya",
    "za",

    "coordx",
    "coordy",
    "coordz",

    "xcoord",
    "ycoord",
    "zcoord",

    "chainid",
    "residuename",
    "residuekey",
    "insertioncode",
    "icode",

    "atomname",
    "atomfullname",
    "element",

    "heteroflag",
    "altloc",

    "_element",
    "_chainid",
    "_resname",
    "_resnum",
    "_icode",
    "_residuekey"

}


atom_numeric_columns_24 = []


for col in df_atom_nodes_24.columns:

    normalized = "".join(

        ch.lower()

        for ch in str(col)

        if ch.isalnum()

    )

    if normalized in excluded_atom_columns_24:

        continue

    converted = pd.to_numeric(

        df_atom_nodes_24[col],

        errors="coerce"

    )

    if converted.notna().any():

        if converted.nunique(dropna=True) > 1:

            atom_numeric_columns_24.append(col)


df_atom_numeric_24 = df_atom_nodes_24[
    atom_numeric_columns_24
].copy()


X_atom_numeric_24 = clean_numeric_matrix_24(
    df_atom_numeric_24
)


# Координаты атомов.

atom_coordinate_columns_24 = [

    find_column_24(
        df_atom_nodes_24,
        ["x_A"]
    ),

    find_column_24(
        df_atom_nodes_24,
        ["y_A"]
    ),

    find_column_24(
        df_atom_nodes_24,
        ["z_A"]
    )

]


if any(
    col is None
    for col in atom_coordinate_columns_24
):

    raise ValueError(
        "Не найдены координаты атомов x_A, y_A, z_A."
    )


pos_atom = df_atom_nodes_24[
    atom_coordinate_columns_24
].apply(
    pd.to_numeric,
    errors="coerce"
).to_numpy(
    dtype=np.float32
)


if not np.isfinite(pos_atom).all():

    raise ValueError(
        "В координатах атомов обнаружены NaN или inf."
    )


# Итоговая матрица атомных признаков.

X_atom = np.concatenate(

    [

        X_atom_numeric_24,

        atom_element_features_24,

        atom_binding_mask_24.reshape(-1, 1),

        direct_contact_mask_24.reshape(-1, 1)

    ],

    axis=1

).astype(np.float32)


atom_feature_names_24 = (

    [
        f"numeric_{col}"
        for col in atom_numeric_columns_24
    ]

    + [
        f"element_{element}"
        for element in ELEMENTS_24
    ]

    + [

        "is_binding_site_atom",

        "is_direct_contact_atom"

    ]

)


# ============================================================
# 8. RESIDUE NODE FEATURES
# ============================================================
#
# X_residue:
# - числовые признаки остатков;
# - one-hot аминокислоты;
# - is_binding_site_residue.
#
# Координаты центров хранятся отдельно.
# ============================================================


res_name_col_24 = "residue_name"


AMINO_ACIDS_24 = [

    "ALA",
    "ARG",
    "ASN",
    "ASP",
    "CYS",
    "GLN",
    "GLU",
    "GLY",
    "HIS",
    "ILE",
    "LEU",
    "LYS",
    "MET",
    "PHE",
    "PRO",
    "SER",
    "THR",
    "TRP",
    "TYR",
    "VAL",
    "OTHER"

]


residue_names_24 = (

    df_residue_nodes_24[
        res_name_col_24
    ]

    .fillna("OTHER")

    .astype(str)

    .str.strip()

    .str.upper()

)


residue_names_24 = residue_names_24.where(

    residue_names_24.isin(AMINO_ACIDS_24),

    "OTHER"

)


residue_type_features_24 = np.column_stack([

    (
        residue_names_24 == amino_acid
    ).astype(np.float32)

    for amino_acid in AMINO_ACIDS_24

])


excluded_residue_columns_24 = {

    "graphresidueid",
    "residueid",
    "resid",
    "residuenumber",
    "resnum",
    "resseq",
    "modelid",

    "chainid",
    "residuename",
    "residuekey",
    "insertioncode",
    "icode",

    "centerx",
    "centery",
    "centerz",

    "centerxa",
    "centerya",
    "centerza",

    "x",
    "y",
    "z",

    "xa",
    "ya",
    "za"

}


residue_numeric_columns_24 = []


for col in df_residue_nodes_24.columns:

    normalized = "".join(

        ch.lower()

        for ch in str(col)

        if ch.isalnum()

    )

    if normalized in excluded_residue_columns_24:

        continue

    converted = pd.to_numeric(

        df_residue_nodes_24[col],

        errors="coerce"

    )

    if converted.notna().any():

        if converted.nunique(dropna=True) > 1:

            residue_numeric_columns_24.append(col)


df_residue_numeric_24 = df_residue_nodes_24[
    residue_numeric_columns_24
].copy()


X_residue_numeric_24 = clean_numeric_matrix_24(
    df_residue_numeric_24
)


# Координаты центров остатков.

residue_coordinate_columns_24 = [

    find_column_24(
        df_residue_nodes_24,
        ["center_x_A"]
    ),

    find_column_24(
        df_residue_nodes_24,
        ["center_y_A"]
    ),

    find_column_24(
        df_residue_nodes_24,
        ["center_z_A"]
    )

]


if any(
    col is None
    for col in residue_coordinate_columns_24
):

    raise ValueError(
        "Не найдены координаты центров остатков."
    )


pos_residue = df_residue_nodes_24[
    residue_coordinate_columns_24
].apply(
    pd.to_numeric,
    errors="coerce"
).to_numpy(
    dtype=np.float32
)


if not np.isfinite(pos_residue).all():

    raise ValueError(
        "В координатах остатков обнаружены NaN или inf."
    )


X_residue = np.concatenate(

    [

        X_residue_numeric_24,

        residue_type_features_24,

        residue_binding_mask_24.reshape(-1, 1)

    ],

    axis=1

).astype(np.float32)


residue_feature_names_24 = (

    [
        f"numeric_{col}"
        for col in residue_numeric_columns_24
    ]

    + [
        f"amino_acid_{aa}"
        for aa in AMINO_ACIDS_24
    ]

    + [
        "is_binding_site_residue"
    ]

)


# ============================================================
# 9. NODE IDENTIFIERS
# ============================================================


atom_node_ids = get_node_ids_24(

    df_atom_nodes_24,

    ["graph_atom_id"],

    "Atom graph"

)


residue_node_ids = get_node_ids_24(

    df_residue_nodes_24,

    ["graph_residue_id"],

    "Residue graph"

)


# ============================================================
# 10. ATOM EDGES
# ============================================================


atom_edge_features_24 = [

    col

    for col in [

        "distance_A",

        "same_residue"

    ]

    if col in df_atom_edges_24.columns

]


edge_index_atom, edge_attr_atom = (
    make_bidirectional_edges_24(

        df_atom_edges_24,

        "atom_i",

        "atom_j",

        atom_edge_features_24,

        atom_node_ids,

        "Atom graph"

    )
)


atom_edge_feature_names_24 = (
    atom_edge_features_24.copy()
)


# ============================================================
# 11. RESIDUE EDGES
# ============================================================


residue_edge_features_24 = [

    col

    for col in [

        "min_distance_A",

        "mean_contact_distance_A",

        "atom_contact_count"

    ]

    if col in df_residue_edges_24.columns

]


edge_index_residue, edge_attr_residue = (
    make_bidirectional_edges_24(

        df_residue_edges_24,

        "residue_node_i",

        "residue_node_j",

        residue_edge_features_24,

        residue_node_ids,

        "Residue graph"

    )
)


residue_edge_feature_names_24 = (
    residue_edge_features_24.copy()
)


# ============================================================
# 12. GLOBAL PROTEIN FEATURES
# ============================================================
#
# Один вектор числовых характеристик структуры.
#
# Он хранится отдельно от атомного и остаточного графов.
# ============================================================


global_numeric_columns_24 = []


excluded_global_columns_24 = {

    "pdbid",
    "structureid",
    "proteinid",
    "modelid"

}


if len(df_global_24) > 1:

    raise ValueError(
        "Ожидалась одна строка глобальных признаков "
        "для одной структуры белка."
    )


if not df_global_24.empty:

    for col in df_global_24.columns:

        normalized = "".join(

            ch.lower()

            for ch in str(col)

            if ch.isalnum()

        )

        if normalized in excluded_global_columns_24:

            continue

        converted = pd.to_numeric(

            df_global_24[col],

            errors="coerce"

        )

        if converted.notna().any():

            global_numeric_columns_24.append(col)


if global_numeric_columns_24:

    X_global = clean_numeric_matrix_24(

        df_global_24[
            global_numeric_columns_24
        ]

    )[0]

else:

    X_global = np.empty(
        0,
        dtype=np.float32
    )


global_feature_names_24 = [

    str(col)

    for col in global_numeric_columns_24

]


# ============================================================
# 13. FINAL FEATURE TABLES
# ============================================================


df_ml_atom_features = pd.DataFrame(

    X_atom,

    columns=atom_feature_names_24

)


df_ml_atom_features.insert(

    0,

    "graph_atom_id",

    atom_node_ids

)


df_ml_residue_features = pd.DataFrame(

    X_residue,

    columns=residue_feature_names_24

)


df_ml_residue_features.insert(

    0,

    "graph_residue_id",

    residue_node_ids

)


# ============================================================
# 14. VALIDATION
# ============================================================


print("\nVALIDATION")
print("-" * 70)


# Размерности узлов и координат.

assert X_atom.shape[0] == len(atom_node_ids)

assert pos_atom.shape == (
    len(atom_node_ids),
    3
)

assert X_residue.shape[0] == len(residue_node_ids)

assert pos_residue.shape == (
    len(residue_node_ids),
    3
)


# Числовые значения.

assert np.isfinite(X_atom).all()

assert np.isfinite(X_residue).all()

assert np.isfinite(X_global).all()

assert np.isfinite(pos_atom).all()

assert np.isfinite(pos_residue).all()

assert np.isfinite(edge_attr_atom).all()

assert np.isfinite(edge_attr_residue).all()


# Проверка индексов рёбер.

if edge_index_atom.size:

    assert edge_index_atom.min() >= 0

    assert edge_index_atom.max() < len(X_atom)


if edge_index_residue.size:

    assert edge_index_residue.min() >= 0

    assert edge_index_residue.max() < len(X_residue)


# Проверка размерностей признаков.

assert X_atom.shape[1] == len(
    atom_feature_names_24
)

assert X_residue.shape[1] == len(
    residue_feature_names_24
)


# Проверка бинарных масок.

assert set(
    np.unique(atom_binding_mask_24)
).issubset({0.0, 1.0})


assert set(
    np.unique(direct_contact_mask_24)
).issubset({0.0, 1.0})


assert set(
    np.unique(residue_binding_mask_24)
).issubset({0.0, 1.0})


# Проверка соответствия числа узлов и масок.

assert len(atom_binding_mask_24) == len(X_atom)

assert len(direct_contact_mask_24) == len(X_atom)

assert len(residue_binding_mask_24) == len(X_residue)


print("Node dimensions: OK")
print("Coordinate dimensions: OK")
print("Edge index ranges: OK")
print("Finite numerical values: OK")
print("Feature names: OK")
print("Binding masks: OK")


# ============================================================
# 15. SAVE NPZ
# ============================================================


ml_data_path_24 = (

    output_dir_24

    / f"{PDB_NAME}_ml_graph_data.npz"

)


np.savez_compressed(

    ml_data_path_24,

    X_atom=X_atom,

    pos_atom=pos_atom,

    edge_index_atom=edge_index_atom,

    edge_attr_atom=edge_attr_atom,

    X_residue=X_residue,

    pos_residue=pos_residue,

    edge_index_residue=edge_index_residue,

    edge_attr_residue=edge_attr_residue,

    X_global=X_global,

    atom_node_ids=atom_node_ids,

    residue_node_ids=residue_node_ids,

    is_binding_site_atom=atom_binding_mask_24,

    is_direct_contact_atom=direct_contact_mask_24,

    is_binding_site_residue=residue_binding_mask_24

)


# ============================================================
# 16. SAVE FEATURE TABLES
# ============================================================


ml_atom_csv_24 = (

    output_dir_24

    / f"{PDB_NAME}_ml_atom_features.csv"

)


ml_residue_csv_24 = (

    output_dir_24

    / f"{PDB_NAME}_ml_residue_features.csv"

)


df_ml_atom_features.to_csv(

    ml_atom_csv_24,

    index=False

)


df_ml_residue_features.to_csv(

    ml_residue_csv_24,

    index=False

)


global_csv_24 = (

    output_dir_24

    / f"{PDB_NAME}_ml_global_features.csv"

)


if len(X_global) > 0:

    pd.DataFrame(

        [X_global],

        columns=global_feature_names_24

    ).to_csv(

        global_csv_24,

        index=False

    )


# ============================================================
# 17. FEATURE MANIFEST
# ============================================================


ml_manifest_24 = {

    "pdb_id": PDB_NAME,

    "description": (
        "Unified protein structural representation "
        "for machine learning and graph neural networks"
    ),

    "source_cells": [22, 23, 24],

    "atom_graph": {

        "node_count": int(
            len(X_atom)
        ),

        "feature_count": int(
            X_atom.shape[1]
        ),

        "coordinate_dimensions": 3,

        "edge_count_directed": int(
            edge_index_atom.shape[1]
        ),

        "feature_names": atom_feature_names_24,

        "edge_feature_names": atom_edge_feature_names_24,

        "node_id_field": "atom_node_ids"

    },

    "residue_graph": {

        "node_count": int(
            len(X_residue)
        ),

        "feature_count": int(
            X_residue.shape[1]
        ),

        "coordinate_dimensions": 3,

        "edge_count_directed": int(
            edge_index_residue.shape[1]
        ),

        "feature_names": residue_feature_names_24,

        "edge_feature_names": residue_edge_feature_names_24,

        "node_id_field": "residue_node_ids"

    },

    "global_features": {

        "feature_count": int(
            len(X_global)
        ),

        "feature_names": global_feature_names_24

    },

    "binding_site": {

        "binding_residue_count": int(
            len(binding_residue_keys_24)
        ),

        "matched_binding_residue_count": int(
            len(matched_residue_keys_24)
        ),

        "unmatched_binding_residue_count": int(
            len(unmatched_residue_keys_24)
        ),

        "binding_atom_count": int(
            atom_binding_mask_24.sum()
        ),

        "binding_residue_node_count": int(
            residue_binding_mask_24.sum()
        ),

        "direct_contact_atom_count": int(
            direct_contact_mask_24.sum()
        ),

        "direct_contact_feature_available": bool(
            direct_contact_available_24
        ),

        "direct_contact_mapping": (
            direct_contact_mapping_info_24
        )

    },

    "preprocessing": {

        "categorical_encoding": "one-hot",

        "missing_numeric_values": "zero imputation",

        "feature_standardization": False,

        "coordinates_separate_from_node_features": True,

        "edge_representation": "bidirectional",

        "edge_index_convention": (
            "zero-based row positions in node matrices"
        ),

        "node_identifiers_saved_separately": True,

        "training_performed": False

    },

    "files": {

        "ml_graph_data": ml_data_path_24.name,

        "atom_features": ml_atom_csv_24.name,

        "residue_features": ml_residue_csv_24.name,

        "global_features": (

            global_csv_24.name

            if len(X_global) > 0

            else None

        )

    }

}


manifest_path_24 = (

    output_dir_24

    / f"{PDB_NAME}_ml_manifest.json"

)


with open(

    manifest_path_24,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        ml_manifest_24,

        file,

        ensure_ascii=False,

        indent=4

    )


# ============================================================
# 18. FINAL REPORT
# ============================================================


print("\n" + "=" * 70)
print("UNIFIED PROTEIN REPRESENTATION SUMMARY")
print("=" * 70)


print("\nATOM GRAPH")
print("-" * 70)

print(
    f"Atom nodes:             {X_atom.shape[0]:,}"
)

print(
    f"Atom features:          {X_atom.shape[1]:,}"
)

print(
    f"Atom coordinates:       {pos_atom.shape}"
)

print(
    f"Atom edge index:        {edge_index_atom.shape}"
)

print(
    f"Atom edge features:     {edge_attr_atom.shape}"
)


print("\nRESIDUE GRAPH")
print("-" * 70)

print(
    f"Residue nodes:          {X_residue.shape[0]:,}"
)

print(
    f"Residue features:       {X_residue.shape[1]:,}"
)

print(
    f"Residue coordinates:    {pos_residue.shape}"
)

print(
    f"Residue edge index:     {edge_index_residue.shape}"
)

print(
    f"Residue edge features:  {edge_attr_residue.shape}"
)


print("\nGLOBAL FEATURES")
print("-" * 70)

print(
    f"Global feature count:   {len(X_global):,}"
)


print("\nBINDING SITE")
print("-" * 70)

print(
    "Binding-site residues:",
    len(binding_residue_keys_24)
)

print(
    "Matched binding residues:",
    len(matched_residue_keys_24)
)

print(
    "Binding-site atom count:",
    int(atom_binding_mask_24.sum())
)

print(
    "Direct contact atom count:",
    int(direct_contact_mask_24.sum())
)

print(
    "Binding-site residue nodes:",
    int(residue_binding_mask_24.sum())
)


print("\nDIRECT CONTACT MAPPING")
print("-" * 70)

for key, value in direct_contact_mapping_info_24.items():

    print(
        f"{key:30s}: {value}"
    )


# ============================================================
# 19. SAVE VALIDATION
# ============================================================


output_files_24 = [

    ml_data_path_24,

    ml_atom_csv_24,

    ml_residue_csv_24,

    manifest_path_24

]


if len(X_global) > 0:

    output_files_24.append(
        global_csv_24
    )


print("\nSAVED FILES")
print("-" * 70)


for path in output_files_24:

    exists = path.exists()

    print(

        f"{path.name:45s}"

        f"{'OK' if exists else 'NOT FOUND'}"

    )

    assert exists, (
        f"Файл не создан: {path}"
    )


# ============================================================
# 20. FINAL STATUS
# ============================================================


print("\n" + "=" * 70)
print("CELL 24 COMPLETED")
print("=" * 70)

print(
    "Unified protein representation successfully created."
)

print(
    f"\nOutput directory: {output_dir_24}"
)


print("\nMAIN ML OBJECTS")


main_arrays_24 = {

    "X_atom": X_atom,

    "pos_atom": pos_atom,

    "edge_index_atom": edge_index_atom,

    "edge_attr_atom": edge_attr_atom,

    "X_residue": X_residue,

    "pos_residue": pos_residue,

    "edge_index_residue": edge_index_residue,

    "edge_attr_residue": edge_attr_residue,

    "X_global": X_global,

    "atom_node_ids": atom_node_ids,

    "residue_node_ids": residue_node_ids,

    "is_binding_site_atom": atom_binding_mask_24,

    "is_direct_contact_atom": direct_contact_mask_24,

    "is_binding_site_residue": residue_binding_mask_24

}


for name, array in main_arrays_24.items():

    print(

        f"{name:30s}"

        f"shape={str(array.shape):18s}"

        f"dtype={array.dtype}"

    )


print("\nAll validation checks passed.")
```

Cell 25 — Geometric Binding Pocket Detection

Поиск и геометрическое описание потенциальных карманов.

Cell 26 — Pocket Feature Integration

Объединение признаков карманов с единым представлением белка.

Уровень 1. Глобальные признаки
Cell 22
Общие характеристики белка: количество атомов, остатков, цепей, компонентов и другие метрики.


Уровень 2. Локальные признаки
Cell 22
Характеристики атомов, аминокислот, участка связывания и взаимодействий.


Уровень 3. Пространственные признаки
Cell 23
Трёхмерные координаты, пространственные соседства и расстояния между атомами и остатками.
Что появится после выполнения
Для каждой структуры будут созданы пять файлов:
Файл	Содержание
1OCC_spatial_atom_nodes.csv	Атомы белка и их координаты
1OCC_spatial_atom_edges.csv	Пространственные контакты между атомами
1OCC_spatial_residue_nodes.csv	Аминокислотные остатки и их центры
1OCC_spatial_residue_edges.csv	Пространственные контакты между остатками
1OCC_spatial_graph_summary.json	Сводная статистика графов

Назначение основных признаков:
Признак	Что означает
x_A, y_A, z_A	Координаты атома в ангстремах
element	Химический элемент атома
residue_key	Идентификатор аминокислотного остатка
distance_A	Расстояние между двумя атомами
atom_contact_count	Число контактирующих пар атомов двух остатков
min_distance_A	Минимальное расстояние между остатками
center_x_A, center_y_A, center_z_A	Координаты геометрического центра остатка

Как это связано с конечной целью
Теперь у нас появляются три уровня представления белка:


results/features/

1OCC_protein_features.json

1OCC_global_features.csv

1OCC_atom_features.csv

1OCC_residue_features.csv

1OCC_interaction_features.csv

1OCC_metal_features.csv

Группа	Примеры признаков	Для чего нужны
Глобальные	protein_atom_count, residue_count, chain_count	Общая характеристика белка
Элементный состав	protein_atom_count_element_C, ..._N, ..._O	Химический состав
Аминокислотный состав	residue_count_HIS, residue_count_CYS	Состав белковой последовательности
Металлы	metal_contact_count_CU, metal_mean_distance_ZN	Характеристика металлоцентров
Взаимодействия	protein_component_contact_count	Характеристика контактов
Дисульфидные мостики	disulfide_bridge_count	Структурная стабилизация
Область связывания	binding_site_residue_count	Характеристика binding site

Эта ячейка создаёт единый экспорт структурных признаков, но пока не формирует окончательный входной тензор для VAE или Transformer.

Причина в том, что у нас пока три разных уровня данных:
1. Глобальные признаки белка — одна строка.
2. Признаки остатков — множество строк.
3. Пространственные взаимодействия — граф связей между атомами или остатками.
Их нельзя бездумно объединять в одну строку: при таком преобразовании потеряется информация о пространственной организации белка.

Для будущей модели SBDD потребуется отдельный этап преобразования этих таблиц в согласованное представление белкового кармана.

# Заключение по выполненной работе

Анализ трёхмерной структуры белка и подготовка структурных данных для машинного обучения в области Structure-Based Drug Design

## Что было сделано

В этом ноутбуке я выполнил загрузку, изучение и комплексный анализ трёхмерной структуры белка. В качестве исходных данных использовал структуры из базы данных Protein Data Bank (PDB) в формате mmCIF.

Работа включала исследование состава белка, его пространственной организации, полноты структурных данных, небелковых компонентов и возможных взаимодействий между ними и белком.

В дальнейшем я расширил анализатор, добавив извлечение структурных и функциональных признаков, построение пространственных графов на атомном и остаточном уровнях, а также формирование единого машинно-читаемого представления белковой структуры.

Таким образом, ноутбук выполняет не только анализ структуры белка, но и подготовку данных, которые в дальнейшем могут использоваться в моделях машинного обучения.

## цель этой работы

Главная цель ноутбука — разработать универсальный инструмент для анализа трёхмерных структур белков и преобразования полученной информации в формат, пригодный для дальнейшего использования в машинном обучении.

В первой части работы я реализовал анализ структуры белка, исследование его пространственной организации, обнаружение небелковых компонентов и изучение возможных взаимодействий между ними и белком.

Во второй части я перешёл от обычного структурного анализа к подготовке данных для нейросетевых моделей. Для этого извлёк числовые признаки, построил атомный и остаточный пространственные графы и объединил полученную информацию в единое машинно-читаемое представление.

Особое значение имеет возможность описывать белок одновременно на нескольких уровнях: на уровне отдельных атомов, аминокислотных остатков и общих характеристик всей структуры. Это позволяет сохранить не только химические и структурные свойства белка, но и информацию о пространственном расположении его элементов.

Полученное представление в дальнейшем может использоваться как входные данные для специальной нейросетевой архитектуры — Protein Encoder, задача которой будет заключаться в преобразовании структурной информации в компактное латентное представление белка (Protein Embedding).

В перспективе это представление планируется интегрировать с уже разработанной генеративной моделью молекул на основе VAE и Transformer. Это позволит перейти от генерации молекул только на основе их химической структуры к генерации с учётом структурных особенностей белковой мишени.

При этом важно отметить, что на текущем этапе реализованы структурный анализатор и подготовка данных для машинного обучения. Сам Protein Encoder, совместное обучение с генеративной моделью и генерация молекул с учётом белковой структуры являются следующими этапами разработки.

Также обнаруженный небелковый компонент не обязательно является лекарственным соединением. Это может быть природный кофактор, ион металла, субстрат или другая молекула, присутствующая в структуре. Поэтому выполненная работа представляет собой универсальную подготовительную основу для последующего исследования взаимодействий белок–лиганд и разработки методов Structure-Based Drug Design.

Краткий вариант:

В этом ноутбуке я скачал трёхмерную структуру белка из базы данных PDB и провёл её комплексный анализ. Сначала изучил состав белка: цепи, аминокислоты, атомы и дополнительные молекулы. Затем получил координаты атомов, проверил полноту и качество структурных данных и построил интерактивную 3D-модель с помощью NGLView.

На следующем этапе я реализовал поиск и идентификацию небелковых компонентов структуры. Выбрал один из компонентов, определил аминокислотные остатки в его ближайшем окружении и проанализировал возможные взаимодействия между ним и белком.

Все полученные результаты сохранил в таблицы и файлы JSON и CSV, а также сформировал итоговый отчёт в форматах DOCX и PDF.

Выполненная работа позволила мне разобраться в основных этапах анализа трёхмерных структур белков и подготовить основу для дальнейшего исследования взаимодействий белков с потенциальными лекарственными соединениями с использованием методов машинного обучения в области SBDD.
