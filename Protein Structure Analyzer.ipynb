# Задачи этапов

Этап 1 — загрузить структуру белка и подготовить данные для будущей модели.
В результате получаем не просто файл PDB, а подготовленное числовое представление белка.

Структура ноутбука и назначение ячеек
Ячейки	Раздел	Результат
1–3	Загрузка структуры	Загрузка структуры белка из базы данных PDB в формате PDB/mmCIF и подготовка исходных данных к анализу.
4	Чтение структуры	Получение информации о моделях, белковых цепях и общей организации структуры.
5	Аминокислотные последовательности	Извлечение аминокислотных последовательностей белковых цепей и сохранение их в формате FASTA.
6	Анализ аминокислотных остатков	Формирование таблицы аминокислотных остатков с указанием их типов, идентификаторов и характеристик.
7	Координаты атомов	Формирование таблицы атомов с их пространственными координатами X, Y, Z и основными характеристиками.
8	Анализ комплекса	Выявление белковых остатков, небелковых компонентов и молекул воды, входящих в состав исследуемой структуры.
9	Геометрия структуры	Расчёт пространственного центра, габаритных размеров и других геометрических характеристик белка.
10	Метаданные	Получение сведений о методе экспериментального определения структуры, разрешении и других структурных характеристиках.
11	Распределение аминокислот	Расчёт частоты встречаемости аминокислот и построение графиков их распределения в белке.
12	Сохранение результатов	Сохранение промежуточных результатов структурного анализа в файлы JSON и CSV.
13	Проверка последовательности	Сравнение экспериментальных аминокислотных последовательностей с референсными последовательностями с помощью выравнивания.
14	Полнота структуры	Проверка отсутствующих остатков и атомов, альтернативных положений атомов, значений occupancy и полноты структурных данных.
15	Разделение компонентов	Разделение структуры на белковые цепи, небелковые компоненты и молекулы воды с сохранением отдельных результатов.
16	3D-визуализация	Построение интерактивной трёхмерной модели белка и его компонентов с использованием NGLView.
17	Обнаружение и идентификация небелковых компонентов	Определение кодов небелковых компонентов, цепей, номеров остатков, атомного и элементного состава, а также пространственных координат. Выбор компонентов для дальнейшего анализа.
18	Анализ области связывания	Определение аминокислотных остатков белка, расположенных вблизи выбранных небелковых компонентов, и расчёт минимальных расстояний между ними.
19	Универсальный анализ взаимодействий белка с компонентами	Определение атомных контактов, потенциальных полярных взаимодействий, возможных водородных связей, гидрофобных контактов и координации различных металлов. Дополнительно анализируются дисульфидные мостики и другие структурные взаимодействия.
20–21	Итоговые структурные и функциональные характеристики	Сбор и систематизация основных характеристик белка, его цепей, пространственной организации и структурных взаимодействий. Формирование итоговых метрик и результатов анализа.
22	Извлечение структурных признаков (Feature Extraction)	Преобразование структурной информации в числовые признаки для машинного обучения. Формирование характеристик атомов, аминокислотных остатков, глобальных параметров белка и информации об области связывания.
23	Построение пространственных графов (Protein Graph Construction)	Представление белковой структуры в виде атомного и остаточного графов. Формирование узлов, пространственных рёбер, координат и характеристик связей между пространственно близкими элементами структуры.
24	Формирование единого машинного представления (Unified Protein Representation)	Объединение атомных и остаточных графов, числовых признаков, координат, глобальных характеристик и масок области связывания в единую структуру данных. Проверка корректности представления и сохранение результатов в форматах NPZ, CSV и JSON.

1. Какую задачу решает текущий ноутбук
Текущий ноутбук 01_PDB_Protein_Analysis.ipynb решает задачу структурного анализа белка и его окружения.

Он отвечает на вопросы:
- Как устроен белок?
- Из каких аминокислот он состоит?
- Где в пространстве находятся его атомы?
- Какие дополнительные молекулы присутствуют в структуре?
- Какие небелковые компоненты находятся рядом с белком?
- Какие аминокислотные остатки образуют окружение выбранного компонента?
- Какие возможные взаимодействия возникают между ними?
То есть ноутбук переводит исходный файл PDB/mmCIF в более удобный для дальнейшей обработки набор структурных данных.
2. Для чего нужны полученные CSV-файлы
Разные таблицы могут использоваться для разных задач.
Данные	Для чего пригодятся
Аминокислотные последовательности	Представление белка в виде последовательности аминокислот
Координаты атомов	Восстановление пространственной структуры белка
Таблицы аминокислотных остатков	Анализ состава и свойств белка
Небелковые компоненты	Получение информации о молекулах, находящихся в структуре
Координаты выбранного компонента	Определение его положения относительно белка
Аминокислоты в области связывания	Описание окружения потенциального участка связывания
Таблицы атомных контактов	Изучение возможных взаимодействий между белком и компонентом
Геометрические характеристики	Дополнительные признаки для анализа структуры

ДАЛЕЕ БУДУТ
Модуль	Назначение
02_PDB_Ligand_Analysis.ipynb	Изучение связанных с белком малых молекул и их аннотаций
03_Protein_Feature_Extraction.ipynb	Подготовка признаков белка для нейросети
04_Protein_Encoder.ipynb	Разработка и проверка Protein Encoder

Этап 2. Создание Protein Encoder
После изучения структур разрабатываем отдельный модуль:

ProteinEncoder

Его задача — преобразовать структуру белка в векторное представление, которое сможет использовать нейросеть.

3. Этап 3. Интеграция с текущей Conditional Transformer VAE
Сейчас твоя архитектура получает три источника информации:
- SELFIES Encoder;
- Property Encoder;
- Fingerprint Encoder.
В перспективе добавится четвёртый источник — Protein Encoder.

Это схема возможного расширения, а не готовое решение: для генерации под белок особенно важно передавать декодеру информацию о структуре белка. Поэтому в дальнейшем стоит рассмотреть передачу последовательности признаков белка через cross-attention, а не только одного общего вектора.

4. Этап 4. Подготовка обучающей выборки
Это важный момент: для обучения генерации под белок недостаточно взять один PDB-файл.

Нужны данные, связывающие структуры белков с соответствующими молекулами. Например, несколько белок-лигандных комплексов для выбранной мишени. Затем формируется набор пар:

(Protein Structure, SELFIES, Molecular Properties)

На нём уже можно будет обучать расширенную модель. Текущая VAE останется базовой моделью для сравнения.

# Этап 1

```python

# ============================================================
# CELL 2. PROJECT CONFIGURATION AND BATCH INPUT
# ============================================================
# Назначение:
# 1. Настройка путей проекта.
# 2. Загрузка списка отобранных PDB.
# 3. Выбор режима обработки: SINGLE / BATCH / TEST.
# 4. Формирование списка PDB для пакетной обработки.
# 5. Подготовка каталогов результатов.
#
# Режимы:
# SINGLE — анализ одной структуры.
# BATCH  — пакетная обработка выбранного списка или всего CSV.
# TEST   — тестовая обработка первых N структур.
#
# Для совместимости с CELL 4–24:
# PDB_ID указывает на первую структуру для обработки.
# PDB_IDS_TO_PROCESS содержит полный список структур.
#
# ВАЖНО:
# CELL 2 только формирует конфигурацию.
# Сам цикл пакетной обработки будет выполняться
# отдельной управляющей ячейкой.
# ============================================================

import os
import re
import warnings
import requests
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path

from Bio.PDB import MMCIFParser, is_aa
from Bio.PDB.MMCIF2Dict import MMCIF2Dict
from Bio.SeqUtils import seq1


# ============================================================
# 1. PROJECT PATHS
# ============================================================

PROJECT_ROOT = Path(
    r"C:\apex\APEX\МИФИ\СНО\medic"
)

DATASET_ROOT = (
    PROJECT_ROOT / "Protein_Pocket_Dataset"
)

SELECTION_CSV = (
    DATASET_ROOT
    / "05_final_dataset"
    / "protein_final_100.csv"
)

STRUCTURE_ANALYSIS_DIR = (
    DATASET_ROOT / "07_structure_analysis"
)

STRUCTURES_DIR = (
    STRUCTURE_ANALYSIS_DIR / "structures"
)

RESULTS_ROOT = (
    STRUCTURE_ANALYSIS_DIR / "results"
)


# ============================================================
# 2. PROCESSING MODE
# ============================================================

# SINGLE — одна структура.
# BATCH  — пакетная обработка.
# TEST   — тестовая обработка первых N структур.

PROCESSING_MODE = "SINGLE"

PROCESSING_MODE = str(
    PROCESSING_MODE
).strip().upper()

ALLOWED_MODES = {
    "SINGLE",
    "BATCH",
    "TEST"
}

if PROCESSING_MODE not in ALLOWED_MODES:
    raise ValueError(
        f"Неизвестный PROCESSING_MODE: {PROCESSING_MODE}. "
        f"Допустимые значения: {sorted(ALLOWED_MODES)}"
    )


# ============================================================
# 3. SINGLE MODE SETTINGS
# ============================================================

# Ручной выбор одной структуры.
# Имеет наивысший приоритет в режиме SINGLE.

PDB_ID_OVERRIDE = None

# Внешний ввод из управляющего ноутбука.
# Не перезаписываем значение, если оно уже существует.

PDB_ID_INPUT = globals().get(
    "PDB_ID_INPUT",
    None
)

# Переменная окружения.

ENV_PDB_ID = os.environ.get(
    "PDB_ID",
    ""
).strip().upper()


# ============================================================
# 4. BATCH MODE SETTINGS
# ============================================================

# Если None:
# BATCH — используются все структуры из CSV.
# TEST  — используются первые BATCH_TEST_SIZE структур.

# Можно задать собственный список:
# BATCH_PDB_IDS = ["1OCC", "1ABC", "2XYZ"]

BATCH_PDB_IDS = None

# Начальный индекс в списке структур.
# Индексация начинается с нуля.

BATCH_START_INDEX = 0

# Максимальное количество структур.
# None означает отсутствие ограничения.

BATCH_LIMIT = None

# Размер тестовой выборки в режиме TEST.

BATCH_TEST_SIZE = 5

# Разрешить обработку PDB, которых нет в selection CSV.
# По умолчанию запрещено для пакетного режима.

BATCH_ALLOW_EXTERNAL = False


# ============================================================
# 5. LOAD SELECTION REGISTRY
# ============================================================

if not SELECTION_CSV.exists():
    raise FileNotFoundError(
        f"Selection CSV not found:\n{SELECTION_CSV}\n"
        "Проверьте расположение protein_final_100.csv."
    )

selection_df = pd.read_csv(
    SELECTION_CSV,
    dtype={"pdb_id": "string"}
)

required_columns = {"pdb_id"}

missing_columns = (
    required_columns - set(selection_df.columns)
)

if missing_columns:
    raise ValueError(
        "В selection CSV отсутствуют обязательные "
        f"столбцы: {sorted(missing_columns)}"
    )

selection_df["pdb_id"] = (
    selection_df["pdb_id"]
    .astype("string")
    .str.strip()
    .str.upper()
)

selection_df = selection_df.dropna(
    subset=["pdb_id"]
).copy()

selection_df = selection_df[
    selection_df["pdb_id"] != ""
].copy()


# ------------------------------------------------------------
# 5.1. PDB ID FORMAT VALIDATION
# ------------------------------------------------------------

invalid_pdb_ids = selection_df.loc[
    ~selection_df["pdb_id"].str.fullmatch(
        r"[A-Z0-9]{4}",
        na=False
    ),
    "pdb_id"
].tolist()

if invalid_pdb_ids:
    raise ValueError(
        "Обнаружены некорректные PDB ID: "
        f"{invalid_pdb_ids[:10]}"
    )


# ------------------------------------------------------------
# 5.2. DUPLICATE VALIDATION
# ------------------------------------------------------------

duplicate_pdb_ids = (
    selection_df.loc[
        selection_df["pdb_id"].duplicated(keep=False),
        "pdb_id"
    ]
    .unique()
    .tolist()
)

if duplicate_pdb_ids:
    raise ValueError(
        "В selection CSV найдены повторяющиеся PDB ID: "
        f"{duplicate_pdb_ids[:10]}"
    )


# ------------------------------------------------------------
# 5.3. PRESERVE SELECTION ORDER
# ------------------------------------------------------------

if "selection_order" in selection_df.columns:

    selection_df["selection_order"] = pd.to_numeric(
        selection_df["selection_order"],
        errors="coerce"
    )

    selection_df = selection_df.sort_values(
        "selection_order",
        na_position="last"
    ).reset_index(drop=True)

else:
    selection_df = selection_df.reset_index(drop=True)


if selection_df.empty:
    raise ValueError(
        "В selection CSV отсутствуют записи PDB."
    )


if len(selection_df) != 100:
    warnings.warn(
        f"Ожидалось 100 структур, "
        f"но в CSV найдено {len(selection_df)}."
    )


# ============================================================
# 6. BUILD PDB PROCESSING LIST
# ============================================================

AVAILABLE_PDB_IDS = (
    selection_df["pdb_id"]
    .astype(str)
    .tolist()
)

# Проверка параметров пакетной обработки.

if not isinstance(BATCH_START_INDEX, int) or BATCH_START_INDEX < 0:
    raise ValueError(
        "BATCH_START_INDEX должен быть целым числом >= 0."
    )

if BATCH_LIMIT is not None:
    if not isinstance(BATCH_LIMIT, int) or BATCH_LIMIT <= 0:
        raise ValueError(
            "BATCH_LIMIT должен быть положительным целым "
            "числом или None."
        )

if not isinstance(BATCH_TEST_SIZE, int) or BATCH_TEST_SIZE <= 0:
    raise ValueError(
        "BATCH_TEST_SIZE должен быть положительным целым числом."
    )


# ------------------------------------------------------------
# 6.1. SOURCE LIST
# ------------------------------------------------------------

if BATCH_PDB_IDS is not None:

    if isinstance(BATCH_PDB_IDS, str):
        raise TypeError(
            "BATCH_PDB_IDS должен быть списком или другой "
            "последовательностью PDB ID, а не одной строкой."
        )

    source_pdb_ids = [
        str(pdb_id).strip().upper()
        for pdb_id in BATCH_PDB_IDS
    ]

    if not source_pdb_ids:
        raise ValueError(
            "BATCH_PDB_IDS задан, но список пуст."
        )

    invalid_batch_ids = [
        pdb_id
        for pdb_id in source_pdb_ids
        if not re.fullmatch(r"[A-Z0-9]{4}", pdb_id)
    ]

    if invalid_batch_ids:
        raise ValueError(
            "Некорректные PDB ID в BATCH_PDB_IDS: "
            f"{invalid_batch_ids[:10]}"
        )

    duplicate_batch_ids = sorted({
        pdb_id
        for pdb_id in source_pdb_ids
        if source_pdb_ids.count(pdb_id) > 1
    })

    if duplicate_batch_ids:
        raise ValueError(
            "Повторяющиеся PDB ID в BATCH_PDB_IDS: "
            f"{duplicate_batch_ids[:10]}"
        )

    external_batch_ids = [
        pdb_id
        for pdb_id in source_pdb_ids
        if pdb_id not in AVAILABLE_PDB_IDS
    ]

    if external_batch_ids and not BATCH_ALLOW_EXTERNAL:
        raise ValueError(
            "В BATCH_PDB_IDS найдены структуры, "
            "отсутствующие в selection CSV: "
            f"{external_batch_ids[:10]}\n"
            "Если это тестовые структуры, установите "
            "BATCH_ALLOW_EXTERNAL = True."
        )

    source_description = "custom PDB list"

else:

    source_pdb_ids = AVAILABLE_PDB_IDS.copy()

    source_description = "selection CSV"


# ------------------------------------------------------------
# 6.2. APPLY START INDEX
# ------------------------------------------------------------

if BATCH_START_INDEX >= len(source_pdb_ids):
    raise ValueError(
        f"BATCH_START_INDEX={BATCH_START_INDEX} выходит "
        f"за пределы списка из {len(source_pdb_ids)} структур."
    )

source_pdb_ids = source_pdb_ids[
    BATCH_START_INDEX:
]


# ------------------------------------------------------------
# 6.3. APPLY MODE LIMIT
# ------------------------------------------------------------

if PROCESSING_MODE == "TEST":

    effective_limit = BATCH_TEST_SIZE

elif BATCH_LIMIT is not None:

    effective_limit = BATCH_LIMIT

else:

    effective_limit = None


if effective_limit is not None:

    source_pdb_ids = source_pdb_ids[
        :effective_limit
    ]


if not source_pdb_ids:
    raise ValueError(
        "После применения настроек пакет обработки пуст."
    )


# ============================================================
# 7. SELECT CURRENT PDB AND PROCESSING LIST
# ============================================================

if PROCESSING_MODE == "SINGLE":

    # --------------------------------------------------------
    # SINGLE MODE
    # --------------------------------------------------------

    if PDB_ID_OVERRIDE is not None:

        PDB_ID = str(
            PDB_ID_OVERRIDE
        ).strip().upper()

        selection_source = "manual override"

    elif PDB_ID_INPUT is not None:

        PDB_ID = str(
            PDB_ID_INPUT
        ).strip().upper()

        selection_source = "external input"

    elif ENV_PDB_ID:

        PDB_ID = ENV_PDB_ID

        selection_source = "environment variable"

    else:

        PDB_ID = str(
            selection_df.iloc[0]["pdb_id"]
        )

        selection_source = "first CSV record"


    if not re.fullmatch(r"[A-Z0-9]{4}", PDB_ID):
        raise ValueError(
            f"Некорректный PDB ID: {PDB_ID!r}. "
            "Ожидается 4 буквенно-цифровых символа."
        )

    PDB_IDS_TO_PROCESS = [
        PDB_ID
    ]


else:

    # --------------------------------------------------------
    # BATCH / TEST MODE
    # --------------------------------------------------------

    PDB_IDS_TO_PROCESS = source_pdb_ids.copy()

    # PDB_ID нужен для совместимости с ячейками,
    # рассчитанными на обработку одной структуры.

    PDB_ID = PDB_IDS_TO_PROCESS[0]

    selection_source = source_description


# ------------------------------------------------------------
# 7.1. PROCESSING METADATA
# ------------------------------------------------------------

BATCH_SIZE = len(PDB_IDS_TO_PROCESS)

PDB_METADATA = {}

PDB_IS_SELECTED = False

matching_rows = selection_df.loc[
    selection_df["pdb_id"] == PDB_ID
]

if not matching_rows.empty:

    PDB_METADATA = (
        matching_rows.iloc[0].to_dict()
    )

    PDB_IS_SELECTED = True

elif PROCESSING_MODE == "SINGLE":

    warnings.warn(
        f"PDB {PDB_ID} отсутствует в "
        "protein_final_100.csv. "
        "Структура будет обработана как тестовая."
    )


# ============================================================
# 8. DIRECTORY CONFIGURATION
# ============================================================

PROJECT_DIR = STRUCTURE_ANALYSIS_DIR

DATA_DIR = STRUCTURES_DIR

# Каталог текущей структуры.
# В пакетном режиме это каталог первой структуры.

PDB_RESULTS_DIR = (
    RESULTS_ROOT / PDB_ID
)

# Совместимость с предыдущими ячейками.

RESULTS_DIR = PDB_RESULTS_DIR

CIF_PATH = (
    DATA_DIR / f"{PDB_ID}.cif"
)

# Создаём только общие каталоги и каталог первой структуры.
# Остальные каталоги будут создаваться управляющим циклом
# непосредственно перед обработкой каждой структуры.

for directory in [
    PROJECT_DIR,
    DATA_DIR,
    RESULTS_ROOT,
    PDB_RESULTS_DIR
]:
    directory.mkdir(
        parents=True,
        exist_ok=True
    )


# ============================================================
# 9. BATCH METADATA
# ============================================================

BATCH_CONFIG = {
    "processing_mode": PROCESSING_MODE,
    "selection_source": selection_source,
    "selection_csv": str(SELECTION_CSV),
    "total_selection_records": len(selection_df),
    "available_pdb_count": len(AVAILABLE_PDB_IDS),
    "batch_size": BATCH_SIZE,
    "batch_start_index": BATCH_START_INDEX,
    "batch_limit": effective_limit,
    "allow_external": BATCH_ALLOW_EXTERNAL,
    "pdb_ids": PDB_IDS_TO_PROCESS.copy()
}


# ============================================================
# 10. CONFIGURATION SUMMARY
# ============================================================

print("=" * 75)
print("PROTEIN STRUCTURE ANALYSIS")
print("=" * 75)

print(f"Processing mode:    {PROCESSING_MODE}")
print(f"Selection source:   {selection_source}")

print("-" * 75)

print(f"Project root:       {PROJECT_ROOT}")
print(f"Dataset root:       {DATASET_ROOT}")
print(f"Selection CSV:      {SELECTION_CSV}")

print("-" * 75)

print(f"Available PDBs:     {len(AVAILABLE_PDB_IDS)}")
print(f"Processing count:   {BATCH_SIZE}")
print(f"Start index:        {BATCH_START_INDEX}")
print(f"Processing limit:   {effective_limit}")

print("-" * 75)

print(f"Current PDB ID:     {PDB_ID}")
print(f"Is selected PDB:    {PDB_IS_SELECTED}")

if "selection_order" in PDB_METADATA:
    print(
        "Selection order:   ",
        PDB_METADATA["selection_order"]
    )

if "dominant_category" in PDB_METADATA:
    print(
        "Functional category:",
        PDB_METADATA["dominant_category"]
    )

print("-" * 75)

print(f"Project directory:  {PROJECT_DIR}")
print(f"Data directory:     {DATA_DIR}")
print(f"Results root:       {RESULTS_ROOT}")
print(f"Current PDB folder: {PDB_RESULTS_DIR}")
print(f"Current CIF file:   {CIF_PATH}")

print("-" * 75)

print("PDB IDS TO PROCESS:")

for i, pdb_id in enumerate(
    PDB_IDS_TO_PROCESS,
    start=1
):
    print(f"{i:3d}. {pdb_id}")

print("=" * 75)

print("CELL 2 CONFIGURATION COMPLETED")
```

```python
# ============================================================
# START TEXT REPORT
# ============================================================

import sys
from pathlib import Path

class TeeOutput:

    """Display output in Jupyter and simultaneously save it to a file."""

    def __init__(self, original_stream, log_file):
        self.original_stream = original_stream
        self.log_file = log_file

    def write(self, text):
        self.original_stream.write(text)
        self.log_file.write(text)
        self.log_file.flush()

    def flush(self):
        self.original_stream.flush()
        self.log_file.flush()

    def isatty(self):
        return self.original_stream.isatty()

    @property
    def encoding(self):
        return self.original_stream.encoding


# Create report file
REPORT_PATH = RESULTS_DIR / f"{PDB_ID}_analysis_report.txt"

report_file = open(
    REPORT_PATH,
    "w",
    encoding="utf-8"
)

# Save original output streams
original_stdout = sys.stdout
original_stderr = sys.stderr

# Redirect output to screen + file
sys.stdout = TeeOutput(original_stdout, report_file)
sys.stderr = TeeOutput(original_stderr, report_file)

print("=" * 70)
print("PROTEIN STRUCTURE ANALYSIS REPORT")
print("=" * 70)

print("PDB ID:", PDB_ID)
print("Report file:", REPORT_PATH)
print(" ")
```

```python
# ============================================================
# CELL 3. DOWNLOAD PDB STRUCTURE (mmCIF)
# ============================================================
# Назначение:
# Загрузка mmCIF-файла для текущей структуры PDB.
#
# Поддерживает:
# 1. Пакетную обработку разных PDB_ID.
# 2. Повторное использование ранее загруженных файлов.
# 3. Проверку HTTP-ответа и содержимого файла.
# 4. Повторные попытки при временных сетевых ошибках.
# 5. Безопасную замену повреждённого файла.
# ============================================================

import requests

from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry


# ============================================================
# 1. VALIDATE PDB ID
# ============================================================

PDB_ID = str(
    PDB_ID
).strip().upper()

if (
    len(PDB_ID) != 4
    or not PDB_ID.isalnum()
):
    raise ValueError(
        f"Некорректный PDB ID: {PDB_ID}. "
        "Ожидается 4 буквенно-цифровых символа."
    )


# ============================================================
# 2. PATH CONFIGURATION
# ============================================================

CIF_PATH = (
    DATA_DIR / f"{PDB_ID}.cif"
)

TEMP_CIF_PATH = (
    DATA_DIR / f"{PDB_ID}.cif.tmp"
)

URL = (
    f"https://files.rcsb.org/download/{PDB_ID}.cif"
)


# ============================================================
# 3. CIF FILE VALIDATION
# ============================================================

def validate_cif_file(path):
    """
    Минимальная структурная проверка mmCIF-файла.

    Проверяется:
    - существование файла;
    - минимальный размер;
    - наличие data-блока;
    - наличие секции atom_site.

    Полноценный разбор структуры выполняется
    в следующих ячейках через Bio.PDB.
    """

    path = Path(path)

    if not path.exists():
        return False

    try:

        if path.stat().st_size < 100:
            return False

        with open(
            path,
            "r",
            encoding="utf-8",
            errors="replace"
        ) as file:

            content = file.read(100_000)

        if not content.strip():
            return False

        if "data_" not in content:
            return False

        if "_atom_site." not in content:
            return False

        return True

    except (OSError, UnicodeError):
        return False


# ============================================================
# 4. PREPARE SESSION WITH RETRIES
# ============================================================

retry_strategy = Retry(
    total=3,
    connect=3,
    read=3,
    status=3,
    backoff_factor=2,
    status_forcelist=[
        429,
        500,
        502,
        503,
        504
    ],
    allowed_methods=[
        "GET"
    ],
    raise_on_status=False
)

adapter = HTTPAdapter(
    max_retries=retry_strategy
)

session = requests.Session()

session.mount(
    "https://",
    adapter
)

session.headers.update({
    "User-Agent": "ProteinStructureAnalyzer/1.0"
})


# ============================================================
# 5. CHECK EXISTING FILE
# ============================================================

if validate_cif_file(CIF_PATH):

    print(
        f"Используется существующий файл: {PDB_ID}"
    )

else:

    print(
        f"Требуется загрузка структуры {PDB_ID}."
    )

    if CIF_PATH.exists():

        print(
            "Существующий файл не прошёл проверку. "
            "Он будет заменён только после успешной загрузки."
        )


    # ========================================================
    # 6. DOWNLOAD STRUCTURE
    # ========================================================

    print(
        f"Загрузка структуры {PDB_ID} из RCSB PDB..."
    )

    try:

        response = session.get(
            URL,
            timeout=(15, 120)
        )

        response.raise_for_status()

        content = response.content

        if not content:
            raise RuntimeError(
                f"RCSB вернул пустой файл для {PDB_ID}."
            )

        # Проверка содержимого до записи.
        text_content = content[:100_000].decode(
            "utf-8",
            errors="replace"
        )

        if "data_" not in text_content:
            raise RuntimeError(
                f"Полученный файл {PDB_ID} "
                "не содержит корректного data-блока."
            )

        if "_atom_site." not in text_content:
            raise RuntimeError(
                f"В файле {PDB_ID} "
                "не найдена секция atom_site."
            )

        # ====================================================
        # 7. SAFE TEMPORARY WRITE
        # ====================================================

        try:

            TEMP_CIF_PATH.write_bytes(
                content
            )

            if not validate_cif_file(
                TEMP_CIF_PATH
            ):
                raise RuntimeError(
                    "Временный mmCIF-файл "
                    "не прошёл проверку."
                )

            # Основной файл заменяется только после проверки.
            TEMP_CIF_PATH.replace(
                CIF_PATH
            )

        finally:

            if TEMP_CIF_PATH.exists():
                TEMP_CIF_PATH.unlink()

    finally:

        response.close()


# ============================================================
# 8. FINAL VALIDATION
# ============================================================

if not validate_cif_file(CIF_PATH):

    raise RuntimeError(
        f"Файл структуры {PDB_ID} "
        f"не прошёл финальную проверку:\n{CIF_PATH}"
    )


# ============================================================
# 9. DOWNLOAD SUMMARY
# ============================================================

file_size_kb = (
    CIF_PATH.stat().st_size / 1024
)

print("\nSTRUCTURE DOWNLOAD")
print("-" * 60)

print("PDB ID:", PDB_ID)
print("URL:", URL)
print("File:", CIF_PATH.resolve())
print("Size:", f"{file_size_kb:.2f} KB")
print("Validation: PASSED")

print("-" * 60)
```

```python
# ============================================================
# CELL 4. ЧТЕНИЕ СТРУКТУРЫ БЕЛКА
# ============================================================
# Назначение:
# 1. Чтение mmCIF-файла.
# 2. Получение моделей и цепей.
# 3. Подсчёт белковых, небелковых остатков и воды.
# 4. Подсчёт атомов по категориям.
# 5. Диагностика альтернативных положений атомов.
# 6. Сохранение статистики структуры и отдельных цепей.
#
# Основной режим:
# - анализируется первая модель;
# - остальные модели учитываются только в model_count;
# - для атомов используются выбранные Bio.PDB варианты altloc.
# ============================================================

import json
from pathlib import Path

from Bio.PDB.Polypeptide import is_aa


print("=" * 70)
print("CELL 4. ЧТЕНИЕ СТРУКТУРЫ БЕЛКА")
print("=" * 70)


# ============================================================
# 1. INPUT VALIDATION
# ============================================================

required_variables = [
    "PDB_ID",
    "CIF_PATH",
    "RESULTS_DIR",
    "MMCIFParser",
    "pd"
]

for variable in required_variables:
    if variable not in globals():
        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение CELL 2 и CELL 3."
        )

PDB_ID = str(PDB_ID).strip().upper()

if len(PDB_ID) != 4 or not PDB_ID.isalnum():
    raise ValueError(
        f"Некорректный PDB ID: {PDB_ID}"
    )

cif_path = Path(CIF_PATH)

if not cif_path.is_file():
    raise FileNotFoundError(
        f"Файл структуры не найден: {cif_path}"
    )

if cif_path.stat().st_size == 0:
    raise ValueError(
        "Файл структуры пустой."
    )

# ВАЖНО:
# CELL 2 уже создала индивидуальный каталог:
# RESULTS_ROOT / PDB_ID.
#
# Повторно добавлять PDB_ID нельзя.

PDB_RESULTS_DIR = Path(RESULTS_DIR)
PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

# Проверяем соответствие имени файла текущему PDB.
if cif_path.stem.upper() != PDB_ID:
    raise ValueError(
        f"Имя CIF-файла ({cif_path.stem}) "
        f"не соответствует PDB_ID ({PDB_ID})."
    )

print(f"PDB ID: {PDB_ID}")
print(f"Файл структуры: {cif_path.resolve()}")
print(f"Каталог результатов: {PDB_RESULTS_DIR.resolve()}")
print(
    f"Размер файла: "
    f"{cif_path.stat().st_size / 1024:.2f} KB"
)


# ============================================================
# 2. READ mmCIF
# ============================================================

parser = MMCIFParser(QUIET=True)

try:
    structure = parser.get_structure(
        PDB_ID,
        str(cif_path)
    )

except Exception as e:
    raise RuntimeError(
        f"Ошибка чтения mmCIF для {PDB_ID}: {e}"
    ) from e

models = list(structure.get_models())

if not models:
    raise ValueError(
        f"В структуре {PDB_ID} не обнаружено моделей."
    )

# Основной анализ выполняется по первой модели.
model = models[0]

chains = list(model.get_chains())

if not chains:
    raise ValueError(
        f"В первой модели {PDB_ID} не обнаружено цепей."
    )


# ============================================================
# 3. RESIDUE CLASSIFICATION SETTINGS
# ============================================================

WATER_NAMES = {
    "HOH",
    "WAT",
    "DOD"
}

chain_statistics = []

total_altloc_atom_sites = 0
total_altloc_variants = 0
total_disordered_residues = 0


# ============================================================
# 4. CHAIN STATISTICS
# ============================================================

for chain in chains:

    residues = list(chain.get_residues())

    # Белковые аминокислотные остатки.
    protein_residues = [
        residue
        for residue in residues
        if is_aa(residue, standard=False)
    ]

    # Остатки воды.
    water_residues = [
        residue
        for residue in residues
        if residue.resname.strip().upper() in WATER_NAMES
    ]

    protein_residue_ids = {
        id(residue)
        for residue in protein_residues
    }

    water_residue_ids = {
        id(residue)
        for residue in water_residues
    }

    # Все остальные остатки считаются небелковыми.
    nonprotein_residues = [
        residue
        for residue in residues
        if (
            id(residue) not in protein_residue_ids
            and id(residue) not in water_residue_ids
        )
    ]

    protein_atoms = 0
    nonprotein_atoms = 0
    water_atoms = 0

    chain_altloc_atom_sites = 0
    chain_altloc_variants = 0

    for residue in residues:

        # get_atoms() возвращает выбранные варианты
        # атомных сайтов, а не все altloc-варианты.
        atoms = list(residue.get_atoms())

        if id(residue) in protein_residue_ids:
            protein_atoms += len(atoms)

        elif id(residue) in water_residue_ids:
            water_atoms += len(atoms)

        else:
            nonprotein_atoms += len(atoms)

        # Диагностика DisorderedAtom.
        for atom in atoms:

            if atom.is_disordered() == 2:

                variants = len(
                    atom.child_dict
                )

                if variants > 1:

                    chain_altloc_atom_sites += 1
                    chain_altloc_variants += variants

    total_atoms = (
        protein_atoms
        + nonprotein_atoms
        + water_atoms
    )

    # Отдельный показатель DisorderedResidue.
    chain_disordered_residues = sum(
        1
        for residue in residues
        if residue.is_disordered()
    )

    # Реальный ID используется для последующего анализа.
    chain_id = str(chain.id)

    # Отображаемый ID используется только в статистике.
    chain_id_display = (
        "<blank>"
        if chain_id == " "
        else chain_id
    )

    chain_statistics.append({

        "pdb_id": PDB_ID,

        "chain_id": chain_id,
        "chain_id_display": chain_id_display,

        "total_residues": len(residues),

        "protein_residues": len(protein_residues),
        "nonprotein_residues": len(nonprotein_residues),
        "water_residues": len(water_residues),

        "protein_atoms": protein_atoms,
        "nonprotein_atoms": nonprotein_atoms,
        "water_atoms": water_atoms,

        "total_atoms": total_atoms,

        "altloc_atom_sites": chain_altloc_atom_sites,
        "altloc_variants": chain_altloc_variants,

        "disordered_residues": chain_disordered_residues
    })

    total_altloc_atom_sites += chain_altloc_atom_sites
    total_altloc_variants += chain_altloc_variants
    total_disordered_residues += chain_disordered_residues


# ============================================================
# 5. FIRST-MODEL SUMMARY
# ============================================================

total_residues = sum(
    item["total_residues"]
    for item in chain_statistics
)

protein_residue_count = sum(
    item["protein_residues"]
    for item in chain_statistics
)

nonprotein_residue_count = sum(
    item["nonprotein_residues"]
    for item in chain_statistics
)

water_residue_count = sum(
    item["water_residues"]
    for item in chain_statistics
)

protein_atom_count = sum(
    item["protein_atoms"]
    for item in chain_statistics
)

nonprotein_atom_count = sum(
    item["nonprotein_atoms"]
    for item in chain_statistics
)

water_atom_count = sum(
    item["water_atoms"]
    for item in chain_statistics
)

total_atom_count = sum(
    item["total_atoms"]
    for item in chain_statistics
)


# ============================================================
# 6. VALIDATION
# ============================================================

calculated_atom_sum = (
    protein_atom_count
    + nonprotein_atom_count
    + water_atom_count
)

atom_count_check = (
    calculated_atom_sum == total_atom_count
)

residue_count_check = (
    protein_residue_count
    + nonprotein_residue_count
    + water_residue_count
    == total_residues
)

if total_atom_count == 0:
    raise ValueError(
        f"В первой модели {PDB_ID} не обнаружено атомов."
    )

if protein_residue_count == 0:
    raise ValueError(
        f"В первой модели {PDB_ID} не обнаружены "
        "белковые остатки."
    )

if not atom_count_check:
    raise ValueError(
        "Сумма категорий атомов не совпадает "
        "с общим количеством атомов."
    )

if not residue_count_check:
    raise ValueError(
        "Сумма категорий остатков не совпадает "
        "с общим количеством остатков."
    )


# ============================================================
# 7. STRUCTURE STATISTICS
# ============================================================

structure_statistics = {

    "pdb_id": PDB_ID,
    "source_file": str(cif_path.resolve()),

    "model_count": len(models),
    "analyzed_model": str(model.id),

    "analysis_uses_first_model_only": True,

    "chain_count": len(chains),

    "total_residues": total_residues,
    "protein_residues": protein_residue_count,
    "nonprotein_residues": nonprotein_residue_count,
    "water_residues": water_residue_count,

    "protein_atoms": protein_atom_count,
    "nonprotein_atoms": nonprotein_atom_count,
    "water_atoms": water_atom_count,
    "total_atoms": total_atom_count,

    "altloc_atom_sites": total_altloc_atom_sites,
    "altloc_variants": total_altloc_variants,

    "disordered_residues": total_disordered_residues,

    "atom_count_check": atom_count_check,
    "residue_count_check": residue_count_check,

    "status": "OK"
}


# ============================================================
# 8. SAVE RESULTS
# ============================================================

chain_statistics_df = pd.DataFrame(
    chain_statistics
)

chain_statistics_path = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_chain_statistics.csv"
)

structure_statistics_path = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_structure_statistics.json"
)

chain_statistics_df.to_csv(
    chain_statistics_path,
    index=False,
    encoding="utf-8-sig"
)

with open(
    structure_statistics_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        structure_statistics,
        file,
        ensure_ascii=False,
        indent=4
    )


# ============================================================
# 9. GENERAL OUTPUT
# ============================================================

print("\n" + "-" * 70)
print("ОБЩАЯ ИНФОРМАЦИЯ О СТРУКТУРЕ")
print("-" * 70)

print(f"Количество моделей:              {len(models)}")
print(f"Анализируемая модель:             {model.id}")
print(f"Количество цепей:                 {len(chains)}")

print(f"\nВсего остатков:                   {total_residues}")
print(f"Белковых остатков:                {protein_residue_count}")
print(f"Небелковых остатков:              {nonprotein_residue_count}")
print(f"Остатков воды:                    {water_residue_count}")

print(f"\nВсего атомов:                     {total_atom_count}")
print(f"Белковых атомов:                  {protein_atom_count}")
print(f"Небелковых атомов:                {nonprotein_atom_count}")
print(f"Атомов воды:                      {water_atom_count}")

print(f"\nАтомных сайтов с altloc:          {total_altloc_atom_sites}")
print(f"Альтернативных вариантов атомов:  {total_altloc_variants}")
print(f"DisorderedResidue:                {total_disordered_residues}")


# ============================================================
# 10. CHAIN OUTPUT
# ============================================================

print("\n" + "-" * 70)
print("СТАТИСТИКА ПО ЦЕПЯМ")
print("-" * 70)

for item in chain_statistics:

    print(f"\nChain {item['chain_id_display']}")

    print(
        f"  Остатки: всего={item['total_residues']}, "
        f"белковые={item['protein_residues']}, "
        f"небелковые={item['nonprotein_residues']}, "
        f"вода={item['water_residues']}"
    )

    print(
        f"  Атомы: всего={item['total_atoms']}, "
        f"белковые={item['protein_atoms']}, "
        f"небелковые={item['nonprotein_atoms']}, "
        f"вода={item['water_atoms']}"
    )

    print(
        f"  Altloc: сайтов={item['altloc_atom_sites']}, "
        f"вариантов={item['altloc_variants']}"
    )


# ============================================================
# 11. FINAL VALIDATION
# ============================================================

print("\n" + "-" * 70)
print("КОНТРОЛЬ РЕЗУЛЬТАТОВ")
print("-" * 70)

print(
    "Проверка суммы категорий атомов:",
    "OK" if atom_count_check else "ERROR"
)

print(
    "Проверка суммы категорий остатков:",
    "OK" if residue_count_check else "ERROR"
)

print(
    "Наличие белковых остатков:",
    "OK" if protein_residue_count > 0 else "ERROR"
)

print("\nСохранённые файлы:")
print(chain_statistics_path)
print(structure_statistics_path)

print("\nСтатус: OK")
print("Структура прочитана.")
print("Статистика подготовлена для последующих ячеек.")

print("=" * 70)
print("CELL 4 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# CELL 5. АНАЛИЗ АМИНОКИСЛОТНЫХ ПОСЛЕДОВАТЕЛЬНОСТЕЙ
# ============================================================
# Назначение:
# 1. Извлечение последовательностей белковых цепей.
# 2. Преобразование трехбуквенных обозначений в однобуквенный формат.
# 3. Обработка распространённых нестандартных аминокислот.
# 4. Контроль неизвестных и модифицированных остатков.
# 5. Сохранение FASTA и статистики для каждой структуры.
#
# ВАЖНО:
# - используются цепи первой модели из CELL 4;
# - последовательности разных цепей не объединяются;
# - исходные обозначения остатков сохраняются в структуре;
# - преобразование применяется только для последовательностей.
# ============================================================

import json
from pathlib import Path

from Bio.SeqUtils import seq1
from Bio.PDB.Polypeptide import is_aa


print("=" * 70)
print("CELL 5. АНАЛИЗ АМИНОКИСЛОТНЫХ ПОСЛЕДОВАТЕЛЬНОСТЕЙ")
print("=" * 70)


# ============================================================
# 1. INPUT VALIDATION
# ============================================================

required_variables = [
    "PDB_ID",
    "chains",
    "PDB_RESULTS_DIR",
    "model"
]

for variable in required_variables:

    if variable not in globals():
        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение CELL 4."
        )

PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

if not chains:
    raise ValueError(
        f"Для {PDB_ID} отсутствуют цепи."
    )

print(f"PDB ID: {PDB_ID}")
print(f"Анализируемая модель: {model.id}")
print(f"Количество цепей: {len(chains)}")
print(f"Каталог результатов: {PDB_RESULTS_DIR.resolve()}")


# ============================================================
# 2. CUSTOM AMINO ACID MAPPING
# ============================================================

CUSTOM_AMINO_ACIDS = {

    "MSE": "M",
    "SEC": "U",
    "PYL": "O",

    "HYP": "P",
    "CSO": "C",
    "CME": "C",

    "MLY": "K",
    "M3L": "K",
    "KCX": "K",

    "FME": "M",

    "SEP": "S",
    "TPO": "T",
    "PTR": "Y",

    "HID": "H",
    "HIE": "H",
    "HIP": "H"
}


# ============================================================
# 3. EXTRACT CHAIN SEQUENCES
# ============================================================

chain_sequences = {}
sequence_statistics = []

for chain in chains:

    sequence = []

    unknown_residues = []
    mapped_modified_residues = []

    protein_residues = [
        residue
        for residue in chain.get_residues()
        if is_aa(residue, standard=False)
    ]

    for residue in protein_residues:

        residue_name = (
            residue.resname.strip().upper()
        )

        amino_acid = seq1(
            residue_name,
            custom_map=CUSTOM_AMINO_ACIDS,
            undef_code="X"
        )

        sequence.append(amino_acid)

        if residue_name in CUSTOM_AMINO_ACIDS:

            mapped_modified_residues.append(
                residue_name
            )

        elif amino_acid == "X":

            unknown_residues.append(
                residue_name
            )

    sequence = "".join(sequence)

    # Сохраняем настоящий ID цепи.
    # Не заменяем пустой ID на <blank> в ключе словаря.
    chain_sequences[chain.id] = sequence

    chain_id_display = (
        "<blank>"
        if str(chain.id) == " "
        else str(chain.id)
    )

    mapped_modified_count = len(
        mapped_modified_residues
    )

    sequence_statistics.append({

        "pdb_id": PDB_ID,

        "chain_id": str(chain.id),
        "chain_id_display": chain_id_display,

        "residue_count": len(protein_residues),

        "sequence_length": len(sequence),

        "unknown_residue_count": sequence.count("X"),

        "unknown_residue_names": sorted(
            set(unknown_residues)
        ),

        # Точное название показателя.
        "mapped_modified_residue_count": mapped_modified_count,

        "mapped_modified_residue_names": sorted(
            set(mapped_modified_residues)
        ),

        # Сохранено для совместимости.
        # По смыслу равно количеству остатков,
        # обработанных через CUSTOM_AMINO_ACIDS.
        "nonstandard_residue_count": mapped_modified_count,

        "sequence_empty": len(sequence) == 0
    })


# ============================================================
# 4. SEQUENCE VALIDATION
# ============================================================

total_sequence_length = sum(
    item["sequence_length"]
    for item in sequence_statistics
)

total_unknown_count = sum(
    item["unknown_residue_count"]
    for item in sequence_statistics
)

total_mapped_modified_count = sum(
    item["mapped_modified_residue_count"]
    for item in sequence_statistics
)

empty_sequence_count = sum(
    item["sequence_empty"]
    for item in sequence_statistics
)

sequence_length_check = all(
    item["residue_count"] == item["sequence_length"]
    for item in sequence_statistics
)

chain_count_check = (
    len(chain_sequences) == len(chains)
)

if not chain_count_check:
    raise ValueError(
        "Количество извлечённых последовательностей "
        "не совпадает с количеством цепей."
    )

if not sequence_length_check:
    raise ValueError(
        "Обнаружено несоответствие между количеством "
        "белковых остатков и длиной последовательности."
    )

if total_sequence_length == 0:
    raise ValueError(
        f"Для {PDB_ID} не удалось извлечь "
        "ни одной аминокислотной последовательности."
    )


# ============================================================
# 5. SAVE FASTA
# ============================================================

FASTA_PATH = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_sequences.fasta"
)

with open(
    FASTA_PATH,
    "w",
    encoding="utf-8"
) as file:

    for chain_id, sequence in chain_sequences.items():

        if not sequence:
            continue

        fasta_chain_id = (
            "<blank>"
            if str(chain_id) == " "
            else str(chain_id)
        )

        file.write(
            f">{PDB_ID}|chain={fasta_chain_id}\n"
        )

        for i in range(0, len(sequence), 80):

            file.write(
                sequence[i:i + 80] + "\n"
            )


# Совместимость с предыдущими ячейками.
fasta_path = FASTA_PATH


# ============================================================
# 6. SEQUENCE SUMMARY
# ============================================================

sequence_summary = {

    "pdb_id": PDB_ID,

    "source_model": str(model.id),

    "uses_first_model_only": True,

    "chain_count": len(chain_sequences),

    "nonempty_chain_count": (
        len(chain_sequences) - empty_sequence_count
    ),

    "empty_chain_count": empty_sequence_count,

    "total_sequence_length": total_sequence_length,

    "unknown_residue_count": total_unknown_count,

    "mapped_modified_residue_count": (
        total_mapped_modified_count
    ),

    "sequence_length_check": sequence_length_check,

    "chain_count_check": chain_count_check,

    "fasta_path": str(
        FASTA_PATH.resolve()
    ),

    "chain_statistics": sequence_statistics
}


# ============================================================
# 7. SAVE STATISTICS
# ============================================================

sequence_statistics_df = pd.DataFrame(
    sequence_statistics
)

sequence_statistics_path = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_sequence_statistics.csv"
)

sequence_summary_path = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_sequence_summary.json"
)

sequence_statistics_df.to_csv(
    sequence_statistics_path,
    index=False,
    encoding="utf-8-sig"
)

with open(
    sequence_summary_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        sequence_summary,
        file,
        ensure_ascii=False,
        indent=4
    )


# ============================================================
# 8. DISPLAY SEQUENCES
# ============================================================

print("\n" + "-" * 70)
print("ПОСЛЕДОВАТЕЛЬНОСТИ БЕЛКОВЫХ ЦЕПЕЙ")
print("-" * 70)

for chain_id, sequence in chain_sequences.items():

    display_chain_id = (
        "<blank>"
        if str(chain_id) == " "
        else str(chain_id)
    )

    print(f"\nChain {display_chain_id}")

    print(
        f"Длина последовательности: {len(sequence)}"
    )

    print("Последовательность:")

    print(
        sequence
        if sequence
        else "Пустая последовательность"
    )

    print("-" * 70)


# ============================================================
# 9. TECHNICAL STATISTICS
# ============================================================

print("\n" + "-" * 70)
print("ТЕХНИЧЕСКАЯ СТАТИСТИКА")
print("-" * 70)

for item in sequence_statistics:

    print(
        f"\nChain {item['chain_id_display']}"
    )

    print(
        f"  Белковых остатков: "
        f"{item['residue_count']}"
    )

    print(
        f"  Длина последовательности: "
        f"{item['sequence_length']}"
    )

    print(
        f"  Неизвестных остатков (X): "
        f"{item['unknown_residue_count']}"
    )

    print(
        f"  Остатков из пользовательской карты: "
        f"{item['mapped_modified_residue_count']}"
    )

    if item["unknown_residue_names"]:

        print(
            "  Неизвестные обозначения:",
            item["unknown_residue_names"]
        )

    if item["mapped_modified_residue_names"]:

        print(
            "  Модифицированные обозначения:",
            item["mapped_modified_residue_names"]
        )


# ============================================================
# 10. FINAL VALIDATION
# ============================================================

if not FASTA_PATH.is_file():
    raise FileNotFoundError(
        f"FASTA-файл не был создан: {FASTA_PATH}"
    )

if FASTA_PATH.stat().st_size == 0:
    raise ValueError(
        f"FASTA-файл пустой: {FASTA_PATH}"
    )

print("\n" + "-" * 70)
print("ИТОГИ CELL 5")
print("-" * 70)

print(
    f"Количество цепей: {len(chain_sequences)}"
)

print(
    f"Общая длина последовательностей: "
    f"{total_sequence_length}"
)

print(
    f"Неизвестные аминокислоты (X): "
    f"{total_unknown_count}"
)

print(
    f"Остатки из пользовательской карты: "
    f"{total_mapped_modified_count}"
)

print(
    f"Цепей без последовательности: "
    f"{empty_sequence_count}"
)

print(
    "Проверка количества цепей:",
    "OK" if chain_count_check else "ERROR"
)

print(
    "Проверка длины последовательностей:",
    "OK" if sequence_length_check else "ERROR"
)

print(f"\nFASTA: {FASTA_PATH}")

print(
    f"Размер FASTA: "
    f"{FASTA_PATH.stat().st_size} bytes"
)

print(f"CSV статистики: {sequence_statistics_path}")
print(f"JSON статистики: {sequence_summary_path}")

print("\nСтатус: OK")
print("Последовательности подготовлены для дальнейшего анализа.")

print("=" * 70)
print("CELL 5 COMPLETED")
print("=" * 70)
```

```python

# ============================================================
# CELL 6. АНАЛИЗ АМИНОКИСЛОТНЫХ ОСТАТКОВ
# ============================================================
# Назначение:
# 1. Формирование таблицы белковых остатков.
# 2. Контроль основных атомов пептидного остова.
# 3. Проверка соответствия последовательностям CELL 5.
# 4. Сверка количества остатков с CELL 4.
# 5. Контроль повторяющихся идентификаторов.
# 6. Сохранение результатов отдельно для каждого PDB.
#
# ВАЖНО:
# 1. Сохраняется исходная идентификация цепей и остатков.
# 2. Проверка последовательностей выполняется по символам.
# 3. Для сверок используются исходные ID цепей.
# 4. Используется тот же первый model и выбранные altloc,
#    что и в CELL 4 и CELL 5.
# ============================================================

import json
import pandas as pd

from pathlib import Path
from Bio.SeqUtils import seq1
from Bio.PDB.Polypeptide import is_aa


print("=" * 70)
print("CELL 6. АНАЛИЗ АМИНОКИСЛОТНЫХ ОСТАТКОВ")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка входных данных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "chains",
    "chain_sequences",
    "PDB_RESULTS_DIR",
    "CUSTOM_AMINO_ACIDS",
    "chain_statistics"
]

for variable in required_variables:

    if variable not in globals():
        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )

PDB_ID = str(PDB_ID).strip().upper()

PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)
PDB_RESULTS_DIR.mkdir(parents=True, exist_ok=True)

if not isinstance(CUSTOM_AMINO_ACIDS, dict):
    raise TypeError(
        "CUSTOM_AMINO_ACIDS должна быть словарём."
    )

if not isinstance(chain_sequences, dict):
    raise TypeError(
        "chain_sequences должна быть словарём."
    )

print(f"PDB ID: {PDB_ID}")
print(f"Количество цепей: {len(chains)}")
print(f"Каталог результатов: {PDB_RESULTS_DIR.resolve()}")


# ------------------------------------------------------------
# 2. Настройки анализа
# ------------------------------------------------------------

BACKBONE_ATOMS = ["N", "CA", "C", "O"]


def normalize_chain_id(chain_id):
    """
    Представление ID цепи для отображения.

    Не используется для идентификации или сверок.
    """

    if chain_id is None or not str(chain_id).strip():
        return "<blank>"

    return str(chain_id)


def residue_identity(row):
    """
    Составная идентификация остатка.
    Исходные поля остаются отдельными в таблице.
    """

    return (
        row["chain_id"],
        row["hetero_flag"],
        row["residue_id"],
        row["insertion_code"]
    )


# ------------------------------------------------------------
# 3. Формирование записей по каждому остатку
# ------------------------------------------------------------

residue_records = []

for chain in chains:

    for residue in chain.get_residues():

        if not is_aa(residue, standard=False):
            continue

        residue_name = residue.resname.strip().upper()

        amino_acid = seq1(
            residue_name,
            custom_map=CUSTOM_AMINO_ACIDS,
            undef_code="X"
        )

        present_backbone_atoms = [
            atom_name
            for atom_name in BACKBONE_ATOMS
            if residue.has_id(atom_name)
        ]

        missing_backbone_atoms = [
            atom_name
            for atom_name in BACKBONE_ATOMS
            if not residue.has_id(atom_name)
        ]

        is_nonstandard = not is_aa(
            residue,
            standard=True
        )

        residue_records.append({

            "pdb_id": PDB_ID,

            "chain_id": chain.id,

            "chain_id_display": normalize_chain_id(chain.id),

            "hetero_flag": residue.id[0],

            "residue_id": int(residue.id[1]),

            "insertion_code": residue.id[2].strip(),

            "residue_name": residue_name,

            "amino_acid": amino_acid,

            "atom_count": len(list(residue.get_atoms())),

            "has_N": residue.has_id("N"),

            "has_CA": residue.has_id("CA"),

            "has_C": residue.has_id("C"),

            "has_O": residue.has_id("O"),

            "present_backbone_atoms": ",".join(
                present_backbone_atoms
            ),

            "missing_backbone_atoms": ",".join(
                missing_backbone_atoms
            ),

            "missing_backbone_atom_count": len(
                missing_backbone_atoms
            ),

            "is_nonstandard": is_nonstandard,

            "is_custom_mapped": (
                residue_name in CUSTOM_AMINO_ACIDS
            ),

            "is_unknown": amino_acid == "X"

        })


# ------------------------------------------------------------
# 4. Создание DataFrame
# ------------------------------------------------------------

RESIDUE_COLUMNS = [

    "pdb_id",
    "chain_id",
    "chain_id_display",
    "hetero_flag",
    "residue_id",
    "insertion_code",
    "residue_name",
    "amino_acid",
    "atom_count",
    "has_N",
    "has_CA",
    "has_C",
    "has_O",
    "present_backbone_atoms",
    "missing_backbone_atoms",
    "missing_backbone_atom_count",
    "is_nonstandard",
    "is_custom_mapped",
    "is_unknown"

]

df_residues = pd.DataFrame(
    residue_records,
    columns=RESIDUE_COLUMNS
)

if df_residues.empty:

    raise ValueError(
        "Белковые остатки не обнаружены. "
        "Проверьте структуру и результат CELL 4."
    )


# ------------------------------------------------------------
# 5. Проверка уникальности идентификаторов остатков
# ------------------------------------------------------------

residue_key_columns = [

    "chain_id",
    "hetero_flag",
    "residue_id",
    "insertion_code"

]

duplicate_residues_df = df_residues.loc[
    df_residues.duplicated(
        subset=residue_key_columns,
        keep=False
    )
].copy()

duplicate_residue_count = len(
    duplicate_residues_df
)

if duplicate_residue_count > 0:

    print(
        "ВНИМАНИЕ: обнаружены повторяющиеся "
        "идентификаторы остатков."
    )


# ------------------------------------------------------------
# 6. Контроль основных атомов
# ------------------------------------------------------------

missing_backbone_df = df_residues.loc[
    df_residues["missing_backbone_atom_count"] > 0
].copy()

complete_backbone_count = int(
    (df_residues["missing_backbone_atom_count"] == 0).sum()
)

incomplete_backbone_count = len(
    missing_backbone_df
)

missing_atom_counts = {

    atom_name: int(
        (~df_residues[f"has_{atom_name}"]).sum()
    )

    for atom_name in BACKBONE_ATOMS
}


# ------------------------------------------------------------
# 7. Статистика по цепям
# ------------------------------------------------------------

residue_chain_statistics = []

for chain_id, group in df_residues.groupby(
    "chain_id",
    sort=False,
    dropna=False
):

    residue_chain_statistics.append({

        "pdb_id": PDB_ID,

        "chain_id": chain_id,

        "chain_id_display": normalize_chain_id(chain_id),

        "residue_count": len(group),

        "atom_count": int(group["atom_count"].sum()),

        "complete_backbone_count": int(
            (group["missing_backbone_atom_count"] == 0).sum()
        ),

        "incomplete_backbone_count": int(
            (group["missing_backbone_atom_count"] > 0).sum()
        ),

        "unknown_residue_count": int(
            group["is_unknown"].sum()
        ),

        "nonstandard_residue_count": int(
            group["is_nonstandard"].sum()
        ),

        "custom_mapped_residue_count": int(
            group["is_custom_mapped"].sum()
        )

    })

df_residue_chain_statistics = pd.DataFrame(
    residue_chain_statistics
)


# ------------------------------------------------------------
# 8. Проверка соответствия CELL 5
# ------------------------------------------------------------

sequence_comparison = []

observed_chain_ids = set(
    df_residues["chain_id"].tolist()
)

expected_chain_ids = set(
    chain_sequences.keys()
)

missing_sequence_chains = (
    observed_chain_ids - expected_chain_ids
)

extra_sequence_chains = (
    expected_chain_ids - observed_chain_ids
)

for chain_id in chain_sequences:

    expected_sequence = chain_sequences[chain_id]

    chain_residues = df_residues.loc[
        df_residues["chain_id"] == chain_id
    ]

    observed_sequence = "".join(
        chain_residues["amino_acid"].tolist()
    )

    observed_count = len(chain_residues)

    sequence_length = len(expected_sequence)

    length_matches = (
        observed_count == sequence_length
    )

    sequence_matches = (
        observed_sequence == expected_sequence
    )

    sequence_comparison.append({

        "pdb_id": PDB_ID,

        "chain_id": chain_id,

        "chain_id_display": normalize_chain_id(chain_id),

        "residue_count": observed_count,

        "sequence_length": sequence_length,

        "observed_sequence": observed_sequence,

        "expected_sequence": expected_sequence,

        "length_matches": length_matches,

        "sequence_matches": sequence_matches,

        "matches": (
            length_matches and sequence_matches
        )

    })

df_sequence_comparison = pd.DataFrame(
    sequence_comparison
)

chain_set_match_check = (
    len(missing_sequence_chains) == 0
    and len(extra_sequence_chains) == 0
)

sequence_content_match_check = (
    not df_sequence_comparison.empty
    and bool(df_sequence_comparison["matches"].all())
)

sequence_match_check = (
    chain_set_match_check
    and sequence_content_match_check
)


# ------------------------------------------------------------
# 9. Проверка соответствия CELL 4
# ------------------------------------------------------------

cell4_comparison = []

cell4_chain_ids = {
    item["chain_id"]
    for item in chain_statistics
    if "chain_id" in item
}

cell6_chain_ids = set(
    df_residues["chain_id"].tolist()
)

cell4_chain_set_match = (
    cell4_chain_ids == cell6_chain_ids
)

for item in chain_statistics:

    chain_id = item["chain_id"]

    observed_count = int(
        (
            df_residues["chain_id"] == chain_id
        ).sum()
    )

    expected_count = int(
        item["protein_residues"]
    )

    cell4_comparison.append({

        "pdb_id": PDB_ID,

        "chain_id": chain_id,

        "chain_id_display": normalize_chain_id(chain_id),

        "cell4_protein_residues": expected_count,

        "cell6_protein_residues": observed_count,

        "matches": expected_count == observed_count

    })

df_cell4_comparison = pd.DataFrame(
    cell4_comparison
)

cell4_match_check = (
    not df_cell4_comparison.empty
    and bool(df_cell4_comparison["matches"].all())
    and cell4_chain_set_match
)


# ------------------------------------------------------------
# 10. Общая статистика
# ------------------------------------------------------------

total_residue_count = len(df_residues)

total_residue_atoms = int(
    df_residues["atom_count"].sum()
)

unknown_residue_count = int(
    df_residues["is_unknown"].sum()
)

nonstandard_residue_count = int(
    df_residues["is_nonstandard"].sum()
)

custom_mapped_residue_count = int(
    df_residues["is_custom_mapped"].sum()
)

residue_analysis_summary = {

    "pdb_id": PDB_ID,

    "residue_count": total_residue_count,

    "chain_count": int(
        df_residues["chain_id"].nunique()
    ),

    "atom_count": total_residue_atoms,

    "complete_backbone_count": complete_backbone_count,

    "incomplete_backbone_count": incomplete_backbone_count,

    "missing_backbone_atom_counts": missing_atom_counts,

    "unknown_residue_count": unknown_residue_count,

    "nonstandard_residue_count": nonstandard_residue_count,

    "custom_mapped_residue_count": custom_mapped_residue_count,

    "duplicate_residue_count": duplicate_residue_count,

    "sequence_chain_set_match_check": bool(
        chain_set_match_check
    ),

    "sequence_match_check": bool(
        sequence_match_check
    ),

    "missing_sequence_chains": [
        normalize_chain_id(x)
        for x in missing_sequence_chains
    ],

    "extra_sequence_chains": [
        normalize_chain_id(x)
        for x in extra_sequence_chains
    ],

    "cell4_chain_set_match_check": bool(
        cell4_chain_set_match
    ),

    "cell4_match_check": bool(
        cell4_match_check
    )

}


# ------------------------------------------------------------
# 11. Сохранение результатов
# ------------------------------------------------------------

RESIDUE_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_residue_analysis.csv"
)

RESIDUE_CHAIN_CSV_PATH = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_residue_chain_statistics.csv"
)

MISSING_BACKBONE_CSV_PATH = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_missing_backbone_atoms.csv"
)

SEQUENCE_CHECK_CSV_PATH = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_sequence_verification.csv"
)

CELL4_CHECK_CSV_PATH = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_cell4_verification.csv"
)

DUPLICATE_RESIDUES_CSV_PATH = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_duplicate_residues.csv"
)

RESIDUE_SUMMARY_JSON_PATH = (
    PDB_RESULTS_DIR /
    f"{PDB_ID}_residue_analysis_summary.json"
)

df_residues.to_csv(
    RESIDUE_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

df_residue_chain_statistics.to_csv(
    RESIDUE_CHAIN_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

missing_backbone_df.to_csv(
    MISSING_BACKBONE_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

df_sequence_comparison.to_csv(
    SEQUENCE_CHECK_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

df_cell4_comparison.to_csv(
    CELL4_CHECK_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

duplicate_residues_df.to_csv(
    DUPLICATE_RESIDUES_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

with open(
    RESIDUE_SUMMARY_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        residue_analysis_summary,
        file,
        ensure_ascii=False,
        indent=4
    )


# ------------------------------------------------------------
# 12. Вывод таблиц
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ТАБЛИЦА БЕЛКОВЫХ ОСТАТКОВ")
print("-" * 70)

display(df_residues.head(10))

print("\n" + "-" * 70)
print("СТАТИСТИКА ПО ЦЕПЯМ")
print("-" * 70)

display(df_residue_chain_statistics)

print("\n" + "-" * 70)
print("ПРОВЕРКА ПОСЛЕДОВАТЕЛЬНОСТЕЙ CELL 5")
print("-" * 70)

display(
    df_sequence_comparison.drop(
        columns=["observed_sequence", "expected_sequence"],
        errors="ignore"
    )
)

print("\n" + "-" * 70)
print("ПРОВЕРКА ДАННЫХ CELL 4")
print("-" * 70)

display(df_cell4_comparison)

print(
    "Наборы цепей CELL 4 и CELL 6:",
    "OK" if cell4_chain_set_match else "ERROR"
)


# ------------------------------------------------------------
# 13. Итоговая диагностика
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ИТОГОВАЯ СТАТИСТИКА")
print("-" * 70)

print(f"Всего белковых остатков: {total_residue_count}")
print(f"Всего атомов белковых остатков: {total_residue_atoms}")
print(f"Цепей: {df_residues['chain_id'].nunique()}")

print(f"Полностью представленных остатков: {complete_backbone_count}")
print(f"Остатков с отсутствующими атомами остова: {incomplete_backbone_count}")

print("\nОтсутствующие основные атомы:")

for atom_name, count in missing_atom_counts.items():
    print(f"  {atom_name}: {count}")

print(f"\nНеизвестные аминокислоты (X): {unknown_residue_count}")
print(f"Нестандартные аминокислоты: {nonstandard_residue_count}")
print(f"Остатки с пользовательским отображением: {custom_mapped_residue_count}")
print(f"Повторяющиеся идентификаторы: {duplicate_residue_count}")

print(
    "\nСоответствие последовательностям CELL 5:",
    "OK" if sequence_match_check else "ERROR"
)

print(
    "Соответствие подсчётам CELL 4:",
    "OK" if cell4_match_check else "ERROR"
)


# ------------------------------------------------------------
# 14. Проверка сохранённых файлов
# ------------------------------------------------------------

output_paths = [

    RESIDUE_CSV_PATH,
    RESIDUE_CHAIN_CSV_PATH,
    MISSING_BACKBONE_CSV_PATH,
    SEQUENCE_CHECK_CSV_PATH,
    CELL4_CHECK_CSV_PATH,
    DUPLICATE_RESIDUES_CSV_PATH,
    RESIDUE_SUMMARY_JSON_PATH

]

print("\n" + "-" * 70)
print("СОХРАНЁННЫЕ ФАЙЛЫ")
print("-" * 70)

for path in output_paths:

    file_ok = path.is_file() and path.stat().st_size > 0

    print(
        f"{path.name}: "
        f"{'OK' if file_ok else 'ERROR'}"
    )

    if not file_ok:
        raise RuntimeError(
            f"Файл не сохранён или пустой: {path}"
        )


# ------------------------------------------------------------
# 15. Итоговый статус
# ------------------------------------------------------------

if duplicate_residue_count > 0:
    raise ValueError(
        "Обнаружены повторяющиеся идентификаторы остатков."
    )

if not sequence_match_check:
    raise ValueError(
        "Последовательности CELL 5 и CELL 6 не совпадают. "
        "Проверьте набор цепей, порядок остатков "
        "и карту нестандартных аминокислот."
    )

if not cell4_match_check:
    raise ValueError(
        "Количество остатков или набор цепей "
        "не соответствует статистике CELL 4."
    )

print("\nСтатус: OK")
print("Таблица остатков подготовлена для дальнейшего анализа.")

print("=" * 70)
print("CELL 6 COMPLETED")
print("=" * 70)
```

```python

# ============================================================
# CELL 7. АНАЛИЗ АТОМОВ И ИХ КООРДИНАТ
# ============================================================
# Назначение:
# 1. Извлечение координат белковых атомов.
# 2. Сохранение химических и структурных характеристик атомов.
# 3. Контроль координат, occupancy, B-factor и altloc.
# 4. Формирование статистики по атомам, элементам и цепям.
# 5. Сверка с результатами CELL 4 и CELL 6.
# 6. Сохранение таблиц и итогового JSON-отчёта.
#
# ВАЖНО:
# 1. Анализируется первый model.
# 2. Используется тот же обход атомов, что в CELL 4 и CELL 6.
# 3. Учитывается выбранный Bio.PDB altloc-вариант.
# 4. В таблицу входят только белковые остатки.
# 5. Геометрическая и химическая достоверность координат
#    не устанавливается только по проверке конечности значений.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.PDB.Polypeptide import is_aa
from Bio.SeqUtils import seq1
from IPython.display import display


print("=" * 70)
print("CELL 7. АНАЛИЗ АТОМОВ И ИХ КООРДИНАТ")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка входных данных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "chains",
    "PDB_RESULTS_DIR",
    "df_residues",
    "CUSTOM_AMINO_ACIDS",
    "structure_statistics"
]

for variable in required_variables:

    if variable not in globals():
        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )

PDB_ID = str(PDB_ID).strip().upper()

PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)
PDB_RESULTS_DIR.mkdir(parents=True, exist_ok=True)

if not isinstance(CUSTOM_AMINO_ACIDS, dict):
    raise TypeError(
        "CUSTOM_AMINO_ACIDS должна быть словарём."
    )

if df_residues.empty:
    raise ValueError(
        "Таблица df_residues пуста. "
        "Проверьте результаты CELL 6."
    )

required_residue_columns = [
    "chain_id",
    "hetero_flag",
    "residue_id",
    "insertion_code",
    "residue_name",
    "atom_count"
]

missing_columns = [
    column
    for column in required_residue_columns
    if column not in df_residues.columns
]

if missing_columns:
    raise ValueError(
        f"В df_residues отсутствуют столбцы: {missing_columns}"
    )

if "chain_statistics" not in globals():
    raise NameError(
        "chain_statistics отсутствует. "
        "Невозможно выполнить полную сверку с CELL 4."
    )

if not isinstance(structure_statistics, dict):
    raise TypeError(
        "structure_statistics должен быть словарём."
    )

print(f"PDB ID: {PDB_ID}")
print(f"Количество цепей: {len(chains)}")
print(f"Каталог результатов: {PDB_RESULTS_DIR.resolve()}")


# ------------------------------------------------------------
# 2. Получение ожидаемых данных CELL 4
# ------------------------------------------------------------

cell4_expected_atoms = structure_statistics.get(
    "protein_atom_count"
)

if cell4_expected_atoms is None:
    raise KeyError(
        "В structure_statistics отсутствует protein_atom_count. "
        "Проверьте формат статистики CELL 4."
    )

cell4_expected_atoms = int(cell4_expected_atoms)

cell4_expected_residues = structure_statistics.get(
    "protein_residue_count"
)

if cell4_expected_residues is None:
    raise KeyError(
        "В structure_statistics отсутствует protein_residue_count. "
        "Проверьте формат статистики CELL 4."
    )

cell4_expected_residues = int(
    cell4_expected_residues
)


# ------------------------------------------------------------
# 3. Функция составного ключа остатка
# ------------------------------------------------------------

def make_residue_key(
    chain_id,
    hetero_flag,
    residue_id,
    insertion_code
):
    """
    Однозначный строковый ключ для таблицы атомов.

    Основные сверки выполняются по исходным полям,
    а не по этой строке.
    """

    return json.dumps(
        [
            chain_id,
            str(hetero_flag),
            int(residue_id),
            str(insertion_code)
        ],
        ensure_ascii=False,
        separators=(",", ":")
    )


# ------------------------------------------------------------
# 4. Извлечение координат белковых атомов
# ------------------------------------------------------------

atom_records = []

for chain in chains:

    for residue in chain.get_residues():

        if not is_aa(residue, standard=False):
            continue

        residue_name = residue.resname.strip().upper()

        amino_acid = seq1(
            residue_name,
            custom_map=CUSTOM_AMINO_ACIDS,
            undef_code="X"
        )

        residue_number = int(residue.id[1])
        insertion_code = residue.id[2].strip()
        hetero_flag = residue.id[0]

        residue_key = make_residue_key(
            chain.id,
            hetero_flag,
            residue_number,
            insertion_code
        )

        # Используется выбранный Bio.PDB altloc-вариант.

        for atom in residue.get_atoms():

            coord = np.asarray(
                atom.coord,
                dtype=float
            )

            element = (
                atom.element.strip().upper()
                if atom.element
                else "UNKNOWN"
            )

            atom_name = atom.get_name().strip()

            occupancy = atom.get_occupancy()
            bfactor = atom.get_bfactor()

            altloc = atom.get_altloc()

            atom_records.append({

                "pdb_id": PDB_ID,

                "chain_id": chain.id,

                "residue_key": residue_key,

                "hetero_flag": hetero_flag,

                "residue_id": residue_number,

                "insertion_code": insertion_code,

                "residue_name": residue_name,

                "amino_acid": amino_acid,

                "atom_serial": atom.get_serial_number(),

                "atom_name": atom_name,

                "atom_fullname": atom.get_fullname().strip(),

                "element": element,

                "x": float(coord[0]),

                "y": float(coord[1]),

                "z": float(coord[2]),

                "occupancy": occupancy,

                "bfactor": bfactor,

                "altloc": altloc.strip() if altloc else "",

                "is_backbone": atom_name in {
                    "N", "CA", "C", "O", "OXT"
                },

                "is_hydrogen": element in {"H", "D"}

            })


# ------------------------------------------------------------
# 5. Создание таблицы атомов
# ------------------------------------------------------------

ATOM_COLUMNS = [

    "pdb_id",
    "chain_id",
    "residue_key",
    "hetero_flag",
    "residue_id",
    "insertion_code",
    "residue_name",
    "amino_acid",
    "atom_serial",
    "atom_name",
    "atom_fullname",
    "element",
    "x",
    "y",
    "z",
    "occupancy",
    "bfactor",
    "altloc",
    "is_backbone",
    "is_hydrogen"

]

df_atoms = pd.DataFrame(
    atom_records,
    columns=ATOM_COLUMNS
)

if df_atoms.empty:
    raise ValueError(
        "Белковые атомы не обнаружены. "
        "Проверьте результаты CELL 4 и CELL 6."
    )

print(f"\nИзвлечено белковых атомов: {len(df_atoms)}")


# ------------------------------------------------------------
# 6. Проверка координат
# ------------------------------------------------------------

coordinate_columns = ["x", "y", "z"]

coordinate_array = df_atoms[
    coordinate_columns
].to_numpy(dtype=float)

finite_coordinate_mask = np.isfinite(
    coordinate_array
).all(axis=1)

invalid_coordinates_df = df_atoms.loc[
    ~finite_coordinate_mask
].copy()

invalid_coordinate_count = len(
    invalid_coordinates_df
)


# ------------------------------------------------------------
# 7. Проверка occupancy и B-factor
# ------------------------------------------------------------

# Преобразование к числовому типу.

occupancy_values = pd.to_numeric(
    df_atoms["occupancy"],
    errors="coerce"
)

bfactor_values = pd.to_numeric(
    df_atoms["bfactor"],
    errors="coerce"
)

# Разделяем пропущенные и некорректные значения.

missing_occupancy_count = int(
    df_atoms["occupancy"].isna().sum()
)

missing_bfactor_count = int(
    df_atoms["bfactor"].isna().sum()
)

non_numeric_occupancy_mask = (
    df_atoms["occupancy"].notna()
    & occupancy_values.isna()
)

non_numeric_bfactor_mask = (
    df_atoms["bfactor"].notna()
    & bfactor_values.isna()
)

invalid_occupancy_mask = (
    non_numeric_occupancy_mask
    | (
        occupancy_values.notna()
        & (
            ~np.isfinite(occupancy_values)
            | (occupancy_values < 0)
            | (occupancy_values > 1)
        )
    )
)

invalid_bfactor_mask = (
    non_numeric_bfactor_mask
    | (
        bfactor_values.notna()
        & (
            ~np.isfinite(bfactor_values)
            | (bfactor_values < 0)
        )
    )
)

invalid_occupancy_df = df_atoms.loc[
    invalid_occupancy_mask
].copy()

invalid_bfactor_df = df_atoms.loc[
    invalid_bfactor_mask
].copy()


# ------------------------------------------------------------
# 8. Статистика по химическим элементам
# ------------------------------------------------------------

element_statistics = (
    df_atoms
    .groupby("element", dropna=False)
    .agg(
        atom_count=("atom_name", "size")
    )
    .reset_index()
    .sort_values(
        "atom_count",
        ascending=False
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 9. Статистика по цепям
# ------------------------------------------------------------

chain_atom_statistics = (
    df_atoms
    .groupby("chain_id", sort=False, dropna=False)
    .agg(
        atom_count=("atom_name", "size"),
        residue_count=("residue_key", "nunique"),
        backbone_atom_count=("is_backbone", "sum"),
        hydrogen_atom_count=("is_hydrogen", "sum"),
        altloc_atom_count=(
            "altloc",
            lambda x: int((x != "").sum())
        ),
        mean_occupancy=("occupancy", "mean"),
        mean_bfactor=("bfactor", "mean")
    )
    .reset_index()
)


# ------------------------------------------------------------
# 10. Общая статистика
# ------------------------------------------------------------

total_protein_atoms = len(df_atoms)

residue_identity_columns = [
    "chain_id",
    "hetero_flag",
    "residue_id",
    "insertion_code"
]

total_protein_residues = int(
    df_atoms[residue_identity_columns]
    .drop_duplicates()
    .shape[0]
)

unique_elements = sorted(
    df_atoms["element"].dropna().unique().tolist()
)

backbone_atom_count = int(
    df_atoms["is_backbone"].sum()
)

hydrogen_atom_count = int(
    df_atoms["is_hydrogen"].sum()
)

altloc_atom_count = int(
    (df_atoms["altloc"] != "").sum()
)

mean_occupancy = occupancy_values.mean()

mean_bfactor = bfactor_values.mean()


# ------------------------------------------------------------
# 11. Сверка с CELL 4
# ------------------------------------------------------------

cell4_atom_check = (
    total_protein_atoms == cell4_expected_atoms
)

cell4_residue_check = (
    total_protein_residues == cell4_expected_residues
)


# ------------------------------------------------------------
# 12. Сверка с CELL 6
# ------------------------------------------------------------

cell6_expected_atoms = int(
    df_residues["atom_count"].sum()
)

cell6_expected_residues = len(df_residues)

cell6_atom_check = (
    total_protein_atoms == cell6_expected_atoms
)

cell6_residue_check = (
    total_protein_residues == cell6_expected_residues
)


# ------------------------------------------------------------
# 13. Сверка количества атомов для каждого остатка
# ------------------------------------------------------------

residue_key_columns = [
    "chain_id",
    "hetero_flag",
    "residue_id",
    "insertion_code",
    "residue_name"
]

# Фактическое количество атомов по CELL 7.

atoms_per_residue = (
    df_atoms
    .groupby(
        residue_key_columns,
        dropna=False
    )
    .agg(
        cell7_atom_count=("atom_name", "size")
    )
    .reset_index()
)

# Количество атомов по CELL 6.

residues_cell6 = df_residues[
    residue_key_columns + ["atom_count"]
].copy()

residues_cell6 = (
    residues_cell6
    .groupby(
        residue_key_columns,
        dropna=False
    )
    .agg(
        cell6_atom_count=("atom_count", "sum")
    )
    .reset_index()
)

# Сопоставление таблиц.

residue_atom_comparison = residues_cell6.merge(
    atoms_per_residue,
    on=residue_key_columns,
    how="outer",
    indicator=True,
    validate="one_to_one"
)

residue_atom_comparison["cell6_atom_count"] = (
    residue_atom_comparison["cell6_atom_count"]
    .fillna(0)
    .astype(int)
)

residue_atom_comparison["cell7_atom_count"] = (
    residue_atom_comparison["cell7_atom_count"]
    .fillna(0)
    .astype(int)
)

residue_atom_comparison["atom_count_match"] = (
    residue_atom_comparison["cell6_atom_count"]
    == residue_atom_comparison["cell7_atom_count"]
)

residue_atom_comparison["residue_present_in_both"] = (
    residue_atom_comparison["_merge"] == "both"
)

residue_atom_mismatch_df = residue_atom_comparison.loc[
    ~residue_atom_comparison["atom_count_match"]
    | ~residue_atom_comparison["residue_present_in_both"]
].copy()

residue_atom_check = residue_atom_mismatch_df.empty


# ------------------------------------------------------------
# 14. Проверка идентификаторов остатков
# ------------------------------------------------------------

# Проверяем, что составные идентификаторы CELL 6
# однозначно представлены в таблице CELL 7.

cell6_residue_ids = set(
    df_residues[residue_identity_columns]
    .itertuples(index=False, name=None)
)

cell7_residue_ids = set(
    df_atoms[residue_identity_columns]
    .drop_duplicates()
    .itertuples(index=False, name=None)
)

residue_identity_check = (
    cell6_residue_ids == cell7_residue_ids
)


# ------------------------------------------------------------
# 15. Формирование итогового отчёта
# ------------------------------------------------------------

atom_analysis_summary = {

    "pdb_id": PDB_ID,

    "protein_atom_count": int(total_protein_atoms),

    "protein_residue_count": int(total_protein_residues),

    "element_count": int(len(unique_elements)),

    "elements": unique_elements,

    "backbone_atom_count": int(backbone_atom_count),

    "hydrogen_atom_count": int(hydrogen_atom_count),

    "altloc_atom_count": int(altloc_atom_count),

    "invalid_coordinate_count": int(invalid_coordinate_count),

    "missing_occupancy_count": int(missing_occupancy_count),

    "missing_bfactor_count": int(missing_bfactor_count),

    "invalid_occupancy_count": int(len(invalid_occupancy_df)),

    "invalid_bfactor_count": int(len(invalid_bfactor_df)),

    "mean_occupancy": (
        float(mean_occupancy)
        if pd.notna(mean_occupancy)
        else None
    ),

    "mean_bfactor": (
        float(mean_bfactor)
        if pd.notna(mean_bfactor)
        else None
    ),

    "cell4_atom_check": bool(cell4_atom_check),

    "cell4_expected_atoms": int(cell4_expected_atoms),

    "cell4_residue_check": bool(cell4_residue_check),

    "cell4_expected_residues": int(cell4_expected_residues),

    "cell6_atom_check": bool(cell6_atom_check),

    "cell6_expected_atoms": int(cell6_expected_atoms),

    "cell6_residue_check": bool(cell6_residue_check),

    "cell6_expected_residues": int(cell6_expected_residues),

    "residue_identity_check": bool(residue_identity_check),

    "residue_atom_check": bool(residue_atom_check),

    "residue_atom_mismatch_count": int(
        len(residue_atom_mismatch_df)
    )

}


# ------------------------------------------------------------
# 16. Сохранение результатов
# ------------------------------------------------------------

ATOM_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_protein_atoms.csv"
)

ELEMENT_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_element_statistics.csv"
)

CHAIN_ATOM_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_chain_atom_statistics.csv"
)

INVALID_COORDINATES_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_invalid_coordinates.csv"
)

INVALID_OCCUPANCY_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_invalid_occupancy.csv"
)

INVALID_BFACTOR_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_invalid_bfactor.csv"
)

RESIDUE_COMPARISON_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_residue_atom_comparison.csv"
)

RESIDUE_MISMATCH_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_residue_atom_mismatches.csv"
)

ATOM_SUMMARY_JSON_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_atom_analysis_summary.json"
)

RESIDUE_EXPORT_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_residues.csv"
)


df_atoms.to_csv(
    ATOM_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

element_statistics.to_csv(
    ELEMENT_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

chain_atom_statistics.to_csv(
    CHAIN_ATOM_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

invalid_coordinates_df.to_csv(
    INVALID_COORDINATES_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

invalid_occupancy_df.to_csv(
    INVALID_OCCUPANCY_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

invalid_bfactor_df.to_csv(
    INVALID_BFACTOR_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

residue_atom_comparison.to_csv(
    RESIDUE_COMPARISON_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

residue_atom_mismatch_df.to_csv(
    RESIDUE_MISMATCH_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

df_residues.to_csv(
    RESIDUE_EXPORT_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

with open(
    ATOM_SUMMARY_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        atom_analysis_summary,
        file,
        ensure_ascii=False,
        indent=4
    )


# ------------------------------------------------------------
# 17. Вывод таблиц
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ПРИМЕР ТАБЛИЦЫ АТОМОВ")
print("-" * 70)

display(df_atoms.head(10))

print("\n" + "-" * 70)
print("РАСПРЕДЕЛЕНИЕ АТОМОВ ПО ЭЛЕМЕНТАМ")
print("-" * 70)

display(element_statistics)

print("\n" + "-" * 70)
print("СТАТИСТИКА ПО ЦЕПЯМ")
print("-" * 70)

display(chain_atom_statistics)

print("\n" + "-" * 70)
print("СВЕРКА КОЛИЧЕСТВА АТОМОВ ПО ОСТАТКАМ")
print("-" * 70)

display(residue_atom_comparison.head(10))


# ------------------------------------------------------------
# 18. Техническая статистика
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ТЕХНИЧЕСКАЯ СТАТИСТИКА")
print("-" * 70)

print(f"Всего белковых атомов: {total_protein_atoms}")
print(f"Белковых остатков: {total_protein_residues}")
print(f"Количество химических элементов: {len(unique_elements)}")
print(f"Элементы: {unique_elements}")
print(f"Атомов основного остова: {backbone_atom_count}")
print(f"Атомов водорода и дейтерия: {hydrogen_atom_count}")
print(f"Атомов с altloc: {altloc_atom_count}")

print(
    "Среднее occupancy:",
    f"{mean_occupancy:.4f}" if pd.notna(mean_occupancy) else "N/A"
)

print(
    "Средний B-factor:",
    f"{mean_bfactor:.4f}" if pd.notna(mean_bfactor) else "N/A"
)

print(f"Атомов с отсутствующим occupancy: {missing_occupancy_count}")
print(f"Атомов с отсутствующим B-factor: {missing_bfactor_count}")
print(f"Атомов с некорректными координатами: {invalid_coordinate_count}")
print(f"Атомов с некорректным occupancy: {len(invalid_occupancy_df)}")
print(f"Атомов с некорректным B-factor: {len(invalid_bfactor_df)}")

print(
    "Остатков с расхождением количества атомов:",
    len(residue_atom_mismatch_df)
)


# ------------------------------------------------------------
# 19. Контроль согласованности
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("СВЕРКА С ПРЕДЫДУЩИМИ ЯЧЕЙКАМИ")
print("-" * 70)

print("Количество атомов CELL 4:", cell4_expected_atoms)
print("Количество атомов CELL 6:", cell6_expected_atoms)
print("Количество атомов CELL 7:", total_protein_atoms)

print(
    "Сверка атомов с CELL 4:",
    "OK" if cell4_atom_check else "ERROR"
)

print(
    "Сверка остатков с CELL 4:",
    "OK" if cell4_residue_check else "ERROR"
)

print(
    "Сверка общего количества атомов с CELL 6:",
    "OK" if cell6_atom_check else "ERROR"
)

print(
    "Сверка количества остатков с CELL 6:",
    "OK" if cell6_residue_check else "ERROR"
)

print(
    "Сверка идентификаторов остатков:",
    "OK" if residue_identity_check else "ERROR"
)

print(
    "Сверка количества атомов по каждому остатку:",
    "OK" if residue_atom_check else "ERROR"
)


# ------------------------------------------------------------
# 20. Проверка сохранённых файлов
# ------------------------------------------------------------

output_paths = [

    ATOM_CSV_PATH,
    ELEMENT_CSV_PATH,
    CHAIN_ATOM_CSV_PATH,
    INVALID_COORDINATES_CSV_PATH,
    INVALID_OCCUPANCY_CSV_PATH,
    INVALID_BFACTOR_CSV_PATH,
    RESIDUE_COMPARISON_CSV_PATH,
    RESIDUE_MISMATCH_CSV_PATH,
    RESIDUE_EXPORT_CSV_PATH,
    ATOM_SUMMARY_JSON_PATH

]

print("\n" + "-" * 70)
print("СОХРАНЁННЫЕ ФАЙЛЫ")
print("-" * 70)

missing_output_files = []

for path in output_paths:

    file_ok = (
        path.is_file()
        and path.stat().st_size > 0
    )

    print(
        f"{path.name}: "
        f"{'OK' if file_ok else 'ERROR'}"
    )

    if not file_ok:
        missing_output_files.append(path)


# ------------------------------------------------------------
# 21. Финальная проверка
# ------------------------------------------------------------

# Совместимость со старым именем переменной.

atom_csv_path = ATOM_CSV_PATH
residue_csv_path = RESIDUE_EXPORT_CSV_PATH

if missing_output_files:

    raise IOError(
        "Не удалось сохранить некоторые результаты: "
        + ", ".join(
            path.name
            for path in missing_output_files
        )
    )

if invalid_coordinate_count > 0:

    raise ValueError(
        f"Обнаружено атомов с некорректными координатами: "
        f"{invalid_coordinate_count}"
    )

if not cell4_atom_check:
    raise ValueError(
        "Количество атомов CELL 7 не совпадает с CELL 4."
    )

if not cell4_residue_check:
    raise ValueError(
        "Количество остатков CELL 7 не совпадает с CELL 4."
    )

if not cell6_atom_check:
    raise ValueError(
        "Общее количество атомов CELL 7 не совпадает с CELL 6."
    )

if not cell6_residue_check:
    raise ValueError(
        "Количество остатков CELL 7 не совпадает с CELL 6."
    )

if not residue_identity_check:
    raise ValueError(
        "Наборы идентификаторов остатков CELL 6 и CELL 7 различаются."
    )

if not residue_atom_check:
    raise ValueError(
        "Обнаружены расхождения в количестве атомов "
        "для отдельных остатков между CELL 6 и CELL 7."
    )

print("\nСтатус: OK")
print("Координаты белковых атомов подготовлены.")

print("=" * 70)
print("CELL 7 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# CELL 8. ANALYSIS OF SMALL MOLECULES AND OTHER COMPONENTS
# ============================================================
# Назначение:
# Анализ воды, лигандов, ионов, кофакторов и других
# не-белковых компонентов структуры.
#
# Источник:
# - CELL 4: chains, structure_statistics;
# - CELL 2: PDB_ID, PDB_RESULTS_DIR.
#
# ВАЖНО:
# 1. Анализируется только первая модель структуры.
# 2. Белковые остатки исключаются через is_aa().
# 3. Вода выделяется отдельно.
# 4. df_hetero_atoms и df_water_atoms не содержат атомов белка.
# 5. Используются выбранные Bio.PDB альтернативные положения
#    атомов, а не все альтернативные конформеры.
# 6. Классификация предварительная и не заменяет
#    идентификацию по Chemical Component Dictionary.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.PDB import is_aa
from IPython.display import display


print("=" * 70)
print("CELL 8. ANALYSIS OF SMALL MOLECULES AND OTHER COMPONENTS")
print("=" * 70)


# ------------------------------------------------------------
# 1. CHECK INPUT VARIABLES
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "chains",
    "PDB_RESULTS_DIR",
    "structure_statistics",
]

missing_variables = [
    name for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise NameError(
        f"Отсутствуют обязательные переменные: {missing_variables}"
    )

PDB_ID = str(PDB_ID).strip().upper()
PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

if not PDB_ID:
    raise ValueError("PDB_ID не может быть пустым.")

if not isinstance(structure_statistics, dict):
    raise TypeError(
        "structure_statistics должен быть словарём."
    )

if not isinstance(chains, (list, tuple)):
    chains = list(chains)

if not chains:
    raise ValueError(
        "Список chains пуст. Проверьте CELL 4."
    )


# Проверяем, что анализируется каталог именно текущего PDB.
if PDB_RESULTS_DIR.name.upper() != PDB_ID:
    raise ValueError(
        f"Каталог результатов не соответствует PDB_ID: "
        f"{PDB_RESULTS_DIR}"
    )


# ------------------------------------------------------------
# 2. COMPONENT CLASSIFICATION SETTINGS
# ------------------------------------------------------------

WATER_CODES = {
    "HOH",
    "WAT",
    "DOD",
}

METAL_ELEMENTS = {
    "FE", "ZN", "MG", "MN", "CU", "CO",
    "NI", "CA", "NA", "K", "CD", "HG",
}


def classify_component(residue_name, elements, atom_count):
    """
    Предварительная классификация не-белкового компонента.

    Используются:
    - количество атомов;
    - набор химических элементов.

    Функция не определяет химическую структуру,
    тип связи или биологическую функцию компонента.
    """

    elements = {
        str(element).strip().upper()
        for element in elements
        if pd.notna(element) and str(element).strip()
    }

    if not elements:
        return "Unknown component"

    if atom_count == 1:
        if elements.issubset(METAL_ELEMENTS):
            return "Single metal atom / ion"
        return "Single atom"

    if (
        elements.intersection(METAL_ELEMENTS)
        and "C" in elements
    ):
        return "Metal-containing organic component"

    if elements.issubset(METAL_ELEMENTS):
        return "Metal-containing component"

    if "C" in elements:
        return "Organic non-protein component"

    return "Other non-protein component"


# ------------------------------------------------------------
# 3. DATA COLLECTION
# ------------------------------------------------------------

hetero_residue_records = []
hetero_atom_records = []

water_residue_records = []
water_atom_records = []

invalid_coordinate_records = []

protein_residue_count = 0

seen_residue_keys = set()

for chain in chains:

    chain_id = str(chain.id)

    for residue in chain:

        residue_name = str(
            residue.resname
        ).strip().upper()

        # Белковые остатки уже анализируются в CELL 6 и CELL 7.
        if is_aa(residue, standard=False):
            protein_residue_count += 1
            continue

        residue_id = residue.id

        hetero_flag = str(residue_id[0])
        residue_number = int(residue_id[1])
        insertion_code = str(residue_id[2]).strip()

        atoms = list(residue.get_atoms())

        atom_elements = [
            str(atom.element).strip().upper()
            for atom in atoms
            if getattr(atom, "element", None)
        ]

        atom_count = len(atoms)

        # Сохраняем настоящий chain_id, в том числе пустой.
        residue_key = (
            f"{chain_id}:{hetero_flag}:"
            f"{residue_number}:{insertion_code}"
        )

        if residue_key in seen_residue_keys:
            raise ValueError(
                f"Обнаружен повторяющийся residue_key: "
                f"{residue_key}"
            )

        seen_residue_keys.add(residue_key)

        component_id = f"{PDB_ID}|{residue_key}"

        is_water = residue_name in WATER_CODES

        if is_water:
            component_type = "Water"
        else:
            component_type = classify_component(
                residue_name=residue_name,
                elements=atom_elements,
                atom_count=atom_count,
            )

        residue_record = {
            "pdb_id": PDB_ID,
            "component_id": component_id,
            "residue_key": residue_key,
            "chain_id": chain_id,
            "hetero_flag": hetero_flag,
            "residue_number": residue_number,
            "insertion_code": insertion_code,
            "residue_name": residue_name,
            "component_type": component_type,
            "atom_count": atom_count,
            "elements": sorted(set(atom_elements)),
        }

        if is_water:
            water_residue_records.append(residue_record)
        else:
            hetero_residue_records.append(residue_record)

        # ----------------------------------------------------
        # Atom-level records
        # ----------------------------------------------------

        for atom in atoms:

            coord = np.asarray(
                atom.coord,
                dtype=float,
            )

            if coord.shape != (3,):
                raise ValueError(
                    f"Некорректная размерность координат: "
                    f"{component_id}, atom={atom.get_name()}"
                )

            atom_element = (
                str(atom.element).strip().upper()
                if getattr(atom, "element", None)
                else ""
            )

            occupancy = atom.get_occupancy()
            bfactor = atom.get_bfactor()

            atom_record = {
                "pdb_id": PDB_ID,
                "component_id": component_id,
                "residue_key": residue_key,
                "chain_id": chain_id,
                "hetero_flag": hetero_flag,
                "residue_number": residue_number,
                "insertion_code": insertion_code,
                "residue_name": residue_name,
                "component_type": component_type,
                "atom_serial": atom.get_serial_number(),
                "atom_name": atom.get_name().strip(),
                "atom_fullname": atom.get_fullname(),
                "element": atom_element,
                "x": float(coord[0]),
                "y": float(coord[1]),
                "z": float(coord[2]),
                "occupancy": (
                    float(occupancy)
                    if occupancy is not None
                    else np.nan
                ),
                "bfactor": (
                    float(bfactor)
                    if bfactor is not None
                    else np.nan
                ),
                "altloc": str(atom.get_altloc()).strip(),
            }

            if not np.isfinite(coord).all():
                invalid_coordinate_records.append(atom_record)

            if is_water:
                water_atom_records.append(atom_record)
            else:
                hetero_atom_records.append(atom_record)


# ------------------------------------------------------------
# 4. DATAFRAMES
# ------------------------------------------------------------

RESIDUE_COLUMNS = [
    "pdb_id",
    "component_id",
    "residue_key",
    "chain_id",
    "hetero_flag",
    "residue_number",
    "insertion_code",
    "residue_name",
    "component_type",
    "atom_count",
    "elements",
]

ATOM_COLUMNS = [
    "pdb_id",
    "component_id",
    "residue_key",
    "chain_id",
    "hetero_flag",
    "residue_number",
    "insertion_code",
    "residue_name",
    "component_type",
    "atom_serial",
    "atom_name",
    "atom_fullname",
    "element",
    "x",
    "y",
    "z",
    "occupancy",
    "bfactor",
    "altloc",
]

df_hetero = pd.DataFrame(
    hetero_residue_records,
    columns=RESIDUE_COLUMNS,
)

df_water = pd.DataFrame(
    water_residue_records,
    columns=RESIDUE_COLUMNS,
)

df_hetero_atoms = pd.DataFrame(
    hetero_atom_records,
    columns=ATOM_COLUMNS,
)

df_water_atoms = pd.DataFrame(
    water_atom_records,
    columns=ATOM_COLUMNS,
)

df_invalid_component_coordinates = pd.DataFrame(
    invalid_coordinate_records,
    columns=ATOM_COLUMNS,
)


# ------------------------------------------------------------
# 5. COMPONENT STATISTICS
# ------------------------------------------------------------

if not df_hetero.empty:

    df_component_statistics = (
        df_hetero
        .groupby(
            ["residue_name", "component_type"],
            dropna=False,
        )
        .agg(
            occurrence_count=("component_id", "count"),
            total_atom_count=("atom_count", "sum"),
            chains=("chain_id", "nunique"),
        )
        .reset_index()
        .sort_values(
            ["occurrence_count", "residue_name"],
            ascending=[False, True],
        )
        .reset_index(drop=True)
    )

else:

    df_component_statistics = pd.DataFrame(
        columns=[
            "residue_name",
            "component_type",
            "occurrence_count",
            "total_atom_count",
            "chains",
        ]
    )


unique_component_names = (
    sorted(
        df_hetero["residue_name"]
        .dropna()
        .unique()
        .tolist()
    )
    if not df_hetero.empty
    else []
)


# ------------------------------------------------------------
# 6. COUNT RECONCILIATION WITH CELL 4
# ------------------------------------------------------------

observed_counts = {
    "nonprotein_residues": len(df_hetero),
    "water_residues": len(df_water),
    "nonprotein_atoms": len(df_hetero_atoms),
    "water_atoms": len(df_water_atoms),
}

expected_keys = [
    "nonprotein_residues",
    "water_residues",
    "nonprotein_atoms",
    "water_atoms",
]

missing_statistics = [
    key for key in expected_keys
    if key not in structure_statistics
]

if missing_statistics:
    raise KeyError(
        "В structure_statistics отсутствуют поля: "
        f"{missing_statistics}"
    )

count_comparison = {}

for key in expected_keys:

    expected_value = int(structure_statistics[key])
    observed_value = int(observed_counts[key])

    count_comparison[key] = {
        "expected_cell_4": expected_value,
        "observed_cell_8": observed_value,
        "matches": expected_value == observed_value,
    }

count_mismatches = [
    key
    for key, values in count_comparison.items()
    if not values["matches"]
]


# ------------------------------------------------------------
# 7. INTERNAL CONSISTENCY AND COORDINATE VALIDATION
# ------------------------------------------------------------

invalid_coordinate_count = len(
    df_invalid_component_coordinates
)

coordinates_valid = (
    invalid_coordinate_count == 0
)

atom_count_consistency = (
    len(df_hetero_atoms)
    == int(df_hetero["atom_count"].sum())
    if not df_hetero.empty
    else len(df_hetero_atoms) == 0
)

water_atom_count_consistency = (
    len(df_water_atoms)
    == int(df_water["atom_count"].sum())
    if not df_water.empty
    else len(df_water_atoms) == 0
)

# Дополнительная проверка идентификаторов PDB.
for table_name, dataframe in {
    "df_hetero": df_hetero,
    "df_water": df_water,
    "df_hetero_atoms": df_hetero_atoms,
    "df_water_atoms": df_water_atoms,
}.items():

    if not dataframe.empty:

        if not dataframe["pdb_id"].eq(PDB_ID).all():
            raise ValueError(
                f"В {table_name} обнаружен чужой PDB_ID."
            )

if not atom_count_consistency:
    raise ValueError(
        "Количество атомов в df_hetero_atoms "
        "не совпадает с суммой atom_count в df_hetero."
    )

if not water_atom_count_consistency:
    raise ValueError(
        "Количество атомов в df_water_atoms "
        "не совпадает с суммой atom_count в df_water."
    )


# ------------------------------------------------------------
# 8. SAVE RESULTS
# ------------------------------------------------------------

output_files = {
    "hetero_residues": (
        PDB_RESULTS_DIR / "hetero_residues.csv"
    ),
    "hetero_atoms": (
        PDB_RESULTS_DIR / "hetero_atoms.csv"
    ),
    "water_residues": (
        PDB_RESULTS_DIR / "water_residues.csv"
    ),
    "water_atoms": (
        PDB_RESULTS_DIR / "water_atoms.csv"
    ),
    "component_statistics": (
        PDB_RESULTS_DIR / "component_statistics.csv"
    ),
    "invalid_coordinates": (
        PDB_RESULTS_DIR / "invalid_component_coordinates.csv"
    ),
    "summary": (
        PDB_RESULTS_DIR / "component_analysis_summary.json"
    ),
}

df_hetero.to_csv(
    output_files["hetero_residues"],
    index=False,
    encoding="utf-8-sig",
)

df_hetero_atoms.to_csv(
    output_files["hetero_atoms"],
    index=False,
    encoding="utf-8-sig",
)

df_water.to_csv(
    output_files["water_residues"],
    index=False,
    encoding="utf-8-sig",
)

df_water_atoms.to_csv(
    output_files["water_atoms"],
    index=False,
    encoding="utf-8-sig",
)

df_component_statistics.to_csv(
    output_files["component_statistics"],
    index=False,
    encoding="utf-8-sig",
)

df_invalid_component_coordinates.to_csv(
    output_files["invalid_coordinates"],
    index=False,
    encoding="utf-8-sig",
)


# ------------------------------------------------------------
# 9. SUMMARY
# ------------------------------------------------------------

component_analysis_summary = {
    "pdb_id": PDB_ID,

    "analysis_model": 0,
    "uses_first_model_only": True,

    "classification_method": (
        "Preliminary classification based on element composition "
        "and atom count"
    ),

    "chemical_identity_limitation": (
        "Classification does not establish exact chemical identity, "
        "bonding, biological function or interaction type. "
        "Chemical Component Dictionary metadata is required "
        "for more detailed identification."
    ),

    "altloc_policy": (
        "Uses Bio.PDB selected atom representation for disordered "
        "atoms; does not count every alternative conformer."
    ),

    "protein_residues_skipped": int(protein_residue_count),

    "nonprotein_residues": int(len(df_hetero)),
    "nonprotein_atoms": int(len(df_hetero_atoms)),

    "water_residues": int(len(df_water)),
    "water_atoms": int(len(df_water_atoms)),

    "unique_nonprotein_component_names": unique_component_names,
    "unique_nonprotein_component_name_count": len(unique_component_names),

    "component_type_counts": (
        {
            str(key): int(value)
            for key, value in
            df_hetero["component_type"].value_counts().items()
        }
        if not df_hetero.empty
        else {}
    ),

    "count_comparison_with_cell_4": count_comparison,

    "atom_count_consistency": bool(atom_count_consistency),
    "water_atom_count_consistency": bool(water_atom_count_consistency),

    "invalid_coordinate_count": int(invalid_coordinate_count),
    "coordinates_valid": bool(coordinates_valid),

    "status": (
        "OK"
        if (
            not count_mismatches
            and coordinates_valid
            and atom_count_consistency
            and water_atom_count_consistency
        )
        else "ERROR"
    ),

    "output_files": {
        key: str(path)
        for key, path in output_files.items()
    },
}


# ------------------------------------------------------------
# 10. OUTPUT CHECKS
# ------------------------------------------------------------

missing_output_files = [
    str(path)
    for path in output_files.values()
    if not path.is_file() or path.stat().st_size == 0
]

if missing_output_files:
    raise FileNotFoundError(
        f"Не созданы или пусты выходные файлы: "
        f"{missing_output_files}"
    )


# Записываем JSON после формирования всех проверок.
with open(
    output_files["summary"],
    "w",
    encoding="utf-8",
) as file:

    json.dump(
        component_analysis_summary,
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False,
    )


# Повторно проверяем JSON.
if (
    not output_files["summary"].is_file()
    or output_files["summary"].stat().st_size == 0
):
    raise RuntimeError(
        "Не удалось сохранить итоговый JSON CELL 8."
    )


# ------------------------------------------------------------
# 11. DISPLAY RESULTS
# ------------------------------------------------------------

print("\nRESIDUE COUNTS")
print("-" * 70)
print(f"Protein residues skipped: {protein_residue_count}")
print(f"Non-protein residues:     {len(df_hetero)}")
print(f"Water residues:           {len(df_water)}")

print("\nATOM COUNTS")
print("-" * 70)
print(f"Non-protein atoms:        {len(df_hetero_atoms)}")
print(f"Water atoms:              {len(df_water_atoms)}")

print("\nCOMPONENT TYPES")
print("-" * 70)

if not df_component_statistics.empty:
    display(df_component_statistics)
else:
    print("Не-белковые компоненты не обнаружены.")

print("\nCOUNT COMPARISON WITH CELL 4")
print("-" * 70)

display(
    pd.DataFrame.from_dict(
        count_comparison,
        orient="index",
    )
)

print("\nVALIDATION")
print("-" * 70)
print(
    f"Non-protein atom counts consistent: "
    f"{atom_count_consistency}"
)
print(
    f"Water atom counts consistent:       "
    f"{water_atom_count_consistency}"
)
print(f"Coordinates valid:                  {coordinates_valid}")
print(f"Invalid coordinates:                {invalid_coordinate_count}")

print("\nOUTPUT DIRECTORY")
print("-" * 70)
print(PDB_RESULTS_DIR)

print("\nSTATUS:", component_analysis_summary["status"])


# ------------------------------------------------------------
# 12. FINAL VALIDATION
# ------------------------------------------------------------

if count_mismatches:
    raise ValueError(
        "Расхождение количества компонентов с CELL 4: "
        f"{count_mismatches}"
    )

if not coordinates_valid:
    raise ValueError(
        "Обнаружены невалидные координаты не-белковых атомов. "
        "См. invalid_component_coordinates.csv"
    )

print("\nCELL 8 COMPLETED SUCCESSFULLY.")
print("=" * 70)
```

Что изменилось по сравнению со старой версией
Возможность	Старая версия	Новая версия
Поиск небелковых компонентов	Да	Да
Поиск воды	Да	Да
Определение цепи и номера остатка	Да	Да
Количество атомов	Да	Да
Элементный состав	Нет	Да
Классификация металлов и ионов	Нет	Да
Выделение органических компонентов	Нет	Да
Распознавание нестандартных аминокислот	Нет	Да
Группировка по типам компонентов	Нет	Да

Важное уточнение: классификация здесь основана на элементном составе и типе остатка. Она не определяет достоверно биологическую функцию молекулы. Например, органический небелковый компонент может быть природным кофактором, субстратом или лекарственным лигандом.

```python
# ============================================================
# CELL 9. GEOMETRIC ANALYSIS OF PROTEIN STRUCTURE
# ============================================================
# Назначение:
# Расчёт основных геометрических характеристик белковой структуры.
#
# Источник:
# - CELL 7: df_atoms — таблица атомов белковой части структуры;
# - CELL 2: PDB_ID, PDB_RESULTS_DIR.
#
# Рассчитываются:
# 1. Геометрический центр белковой структуры.
# 2. Минимальные и максимальные координаты.
# 3. Размеры структуры по трём осям.
# 4. Объём ограничивающего параллелепипеда.
# 5. Расстояния атомов от геометрического центра.
# 6. Среднее, медианное и максимальное расстояние.
# 7. Невзвешенный геометрический радиус инерции.
# 8. Соотношение размеров структуры по осям.
#
# ВАЖНО:
# 1. Используются только белковые атомы из CELL 7.
# 2. Вода и не-белковые компоненты из CELL 8 не включаются.
# 3. Радиус инерции рассчитывается без учёта атомных масс.
# 4. Объём ограничивающего параллелепипеда не является
#    физическим объёмом белковой молекулы.
# ============================================================

import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path


print("=" * 70)
print("CELL 9. GEOMETRIC ANALYSIS OF PROTEIN STRUCTURE")
print("=" * 70)


# ------------------------------------------------------------
# 1. CHECK INPUT VARIABLES
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "PDB_RESULTS_DIR",
    "df_atoms",
]

missing_variables = [
    name for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise NameError(
        f"Отсутствуют обязательные переменные: {missing_variables}"
    )

PDB_ID = str(PDB_ID).strip().upper()
PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

if not PDB_ID:
    raise ValueError("PDB_ID не может быть пустым.")

if PDB_RESULTS_DIR.name.upper() != PDB_ID:
    raise ValueError(
        f"Каталог результатов не соответствует PDB_ID: "
        f"{PDB_RESULTS_DIR}"
    )

if not isinstance(df_atoms, pd.DataFrame):
    raise TypeError(
        "df_atoms должен быть объектом pandas.DataFrame."
    )

required_columns = [
    "pdb_id",
    "chain_id",
    "residue_key",
    "x",
    "y",
    "z",
]

missing_columns = [
    column for column in required_columns
    if column not in df_atoms.columns
]

if missing_columns:
    raise KeyError(
        f"В df_atoms отсутствуют обязательные столбцы: "
        f"{missing_columns}"
    )

if df_atoms.empty:
    raise ValueError(
        "df_atoms пуст. Геометрический анализ невозможен."
    )


# ------------------------------------------------------------
# 2. STRUCTURE ID VALIDATION
# ------------------------------------------------------------

if df_atoms["pdb_id"].isna().any():
    raise ValueError(
        "В df_atoms обнаружены отсутствующие PDB ID."
    )

pdb_ids_in_table = (
    df_atoms["pdb_id"]
    .astype(str)
    .str.strip()
    .str.upper()
    .unique()
)

if len(pdb_ids_in_table) != 1:
    raise ValueError(
        "df_atoms содержит несколько PDB ID: "
        f"{pdb_ids_in_table.tolist()}"
    )

if pdb_ids_in_table[0] != PDB_ID:
    raise ValueError(
        f"Несовпадение PDB ID: CELL 2 = {PDB_ID}, "
        f"df_atoms = {pdb_ids_in_table[0]}"
    )


# ------------------------------------------------------------
# 3. IDENTIFIER VALIDATION
# ------------------------------------------------------------

if df_atoms["chain_id"].isna().any():
    raise ValueError(
        "В df_atoms обнаружены отсутствующие chain_id."
    )

if df_atoms["residue_key"].isna().any():
    raise ValueError(
        "В df_atoms обнаружены отсутствующие residue_key."
    )

if df_atoms["residue_key"].astype(str).str.strip().eq("").any():
    raise ValueError(
        "В df_atoms обнаружены пустые residue_key."
    )

# Пустой chain_id допустим для структур с пустым
# идентификатором цепи. Пропуски NaN при этом запрещены.


# ------------------------------------------------------------
# 4. COORDINATE VALIDATION
# ------------------------------------------------------------

coordinate_columns = ["x", "y", "z"]

# Преобразуем координаты в числовой формат.
# Некорректные значения преобразуются в NaN и будут выявлены ниже.

numeric_coordinates = df_atoms[
    coordinate_columns
].apply(
    pd.to_numeric,
    errors="coerce",
)

coordinates = numeric_coordinates.to_numpy(
    dtype=float
)

finite_coordinate_mask = np.isfinite(
    coordinates
).all(axis=1)

invalid_coordinate_count = int(
    (~finite_coordinate_mask).sum()
)

if invalid_coordinate_count > 0:

    invalid_coordinates = df_atoms.loc[
        ~finite_coordinate_mask
    ].copy()

    invalid_path = (
        PDB_RESULTS_DIR
        / "invalid_protein_coordinates.csv"
    )

    invalid_coordinates.to_csv(
        invalid_path,
        index=False,
        encoding="utf-8-sig",
    )

    raise ValueError(
        f"Обнаружено атомов с невалидными координатами: "
        f"{invalid_coordinate_count}. "
        f"См. {invalid_path}"
    )


# ------------------------------------------------------------
# 5. BASIC STRUCTURE PARAMETERS
# ------------------------------------------------------------

atom_count = int(len(df_atoms))

residue_count = int(
    df_atoms["residue_key"].nunique()
)

chain_count = int(
    df_atoms["chain_id"].nunique(dropna=False)
)

if atom_count == 0:
    raise ValueError(
        "Нет атомов для геометрического анализа."
    )

if residue_count == 0:
    raise ValueError(
        "Не обнаружены белковые остатки."
    )

if chain_count == 0:
    raise ValueError(
        "Не обнаружены белковые цепи."
    )


# ------------------------------------------------------------
# 6. GEOMETRIC CENTROID
# ------------------------------------------------------------
# Центроид — среднее арифметическое координат всех атомов.
# Каждый атом имеет одинаковый вес.

centroid = coordinates.mean(axis=0)

centroid_x = float(centroid[0])
centroid_y = float(centroid[1])
centroid_z = float(centroid[2])


# ------------------------------------------------------------
# 7. BOUNDING BOX
# ------------------------------------------------------------

min_coordinates = coordinates.min(axis=0)
max_coordinates = coordinates.max(axis=0)

dimensions = max_coordinates - min_coordinates

dimension_x = float(dimensions[0])
dimension_y = float(dimensions[1])
dimension_z = float(dimensions[2])

max_dimension = float(dimensions.max())
min_dimension = float(dimensions.min())

# Геометрический объём ограничивающего параллелепипеда.
# Это не физический объём белковой молекулы.

bounding_box_volume = float(
    np.prod(dimensions)
)


# ------------------------------------------------------------
# 8. DISTANCES FROM CENTROID
# ------------------------------------------------------------

distances_from_center = np.linalg.norm(
    coordinates - centroid,
    axis=1,
)

mean_distance = float(
    np.mean(distances_from_center)
)

median_distance = float(
    np.median(distances_from_center)
)

max_distance = float(
    np.max(distances_from_center)
)

min_distance = float(
    np.min(distances_from_center)
)


# ------------------------------------------------------------
# 9. UNWEIGHTED GEOMETRIC RADIUS OF GYRATION
# ------------------------------------------------------------
# Формула:
#
# Rg = sqrt(mean(|ri - r_center|^2))
#
# Здесь не используются атомные массы.
# Это геометрический показатель распределения атомов,
# а не mass-weighted биофизический радиус инерции.

unweighted_geometric_radius_of_gyration = float(
    np.sqrt(
        np.mean(distances_from_center ** 2)
    )
)


# ------------------------------------------------------------
# 10. ASPECT RATIO
# ------------------------------------------------------------

if min_dimension > 0:

    aspect_ratio = float(
        max_dimension / min_dimension
    )

else:

    aspect_ratio = None


# ------------------------------------------------------------
# 11. ATOM COUNT CHECK WITH CELL 7
# ------------------------------------------------------------
# Если CELL 7 создала total_protein_atoms,
# проверяем совпадение количества атомов.
#
# Если переменная отсутствует, проверка явно
# отмечается как недоступная.

if "total_protein_atoms" in globals():

    expected_protein_atom_count = int(
        total_protein_atoms
    )

    atom_count_check_available = True

    atom_count_matches_cell_7 = (
        atom_count == expected_protein_atom_count
    )

else:

    expected_protein_atom_count = None

    atom_count_check_available = False

    atom_count_matches_cell_7 = None


# ------------------------------------------------------------
# 12. GEOMETRY SUMMARY
# ------------------------------------------------------------

geometry_summary = {
    "pdb_id": PDB_ID,

    "analysis_model": 0,
    "uses_first_model_only": True,

    "source_table": "df_atoms",
    "source_table_scope": "protein_atoms_only",

    "atom_count": atom_count,
    "residue_count": residue_count,
    "chain_count": chain_count,

    "centroid": {
        "x": centroid_x,
        "y": centroid_y,
        "z": centroid_z,
        "method": "unweighted_arithmetic_mean_of_atom_coordinates",
    },

    "bounding_box": {
        "min_x": float(min_coordinates[0]),
        "min_y": float(min_coordinates[1]),
        "min_z": float(min_coordinates[2]),

        "max_x": float(max_coordinates[0]),
        "max_y": float(max_coordinates[1]),
        "max_z": float(max_coordinates[2]),

        "dimension_x": dimension_x,
        "dimension_y": dimension_y,
        "dimension_z": dimension_z,

        "volume": bounding_box_volume,

        "volume_interpretation": (
            "Geometric bounding-box volume; "
            "not the physical molecular volume."
        ),
    },

    "distance_from_centroid": {
        "minimum": min_distance,
        "mean": mean_distance,
        "median": median_distance,
        "maximum": max_distance,
    },

    "unweighted_geometric_radius_of_gyration": (
        unweighted_geometric_radius_of_gyration
    ),

    "radius_of_gyration_method": (
        "sqrt(mean(squared_distances_from_geometric_centroid)); "
        "unweighted_atom_coordinates"
    ),

    "aspect_ratio": aspect_ratio,

    "coordinate_validation": {
        "invalid_coordinate_count": invalid_coordinate_count,
        "all_coordinates_finite": True,
    },

    "atom_count_check_with_cell_7": {
        "available": atom_count_check_available,
        "expected_atom_count": expected_protein_atom_count,
        "observed_atom_count": atom_count,
        "matches": atom_count_matches_cell_7,
    },
}


# ------------------------------------------------------------
# 13. SAVE SUMMARY TABLE
# ------------------------------------------------------------

df_geometry_summary = pd.DataFrame([
    {
        "pdb_id": PDB_ID,

        "atom_count": atom_count,
        "residue_count": residue_count,
        "chain_count": chain_count,

        "centroid_x": centroid_x,
        "centroid_y": centroid_y,
        "centroid_z": centroid_z,

        "dimension_x": dimension_x,
        "dimension_y": dimension_y,
        "dimension_z": dimension_z,

        "bounding_box_volume": bounding_box_volume,

        "min_distance_from_centroid": min_distance,
        "mean_distance_from_centroid": mean_distance,
        "median_distance_from_centroid": median_distance,
        "max_distance_from_centroid": max_distance,

        "unweighted_geometric_radius_of_gyration": (
            unweighted_geometric_radius_of_gyration
        ),

        "aspect_ratio": aspect_ratio,
    }
])


# ------------------------------------------------------------
# 14. SAVE FILES
# ------------------------------------------------------------

geometry_csv_path = (
    PDB_RESULTS_DIR
    / "protein_geometry_summary.csv"
)

geometry_json_path = (
    PDB_RESULTS_DIR
    / "protein_geometry_summary.json"
)

geometry_plot_path = (
    PDB_RESULTS_DIR
    / "protein_structure_geometry.png"
)

df_geometry_summary.to_csv(
    geometry_csv_path,
    index=False,
    encoding="utf-8-sig",
)


# ------------------------------------------------------------
# 15. 3D STRUCTURE GEOMETRY PLOT
# ------------------------------------------------------------

fig = plt.figure(figsize=(10, 8))

ax = fig.add_subplot(
    111,
    projection="3d",
)

# Отображаем координаты атомов белковой структуры.
ax.scatter(
    coordinates[:, 0],
    coordinates[:, 1],
    coordinates[:, 2],
    s=2,
    alpha=0.5,
)

# Геометрический центр.
ax.scatter(
    centroid_x,
    centroid_y,
    centroid_z,
    s=80,
    marker="X",
    label="Geometric centroid",
)

ax.set_xlabel("X (Å)")
ax.set_ylabel("Y (Å)")
ax.set_zlabel("Z (Å)")

ax.set_title(
    f"Protein structure geometry: {PDB_ID}"
)

ax.legend()

fig.tight_layout()

fig.savefig(
    geometry_plot_path,
    dpi=200,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)


# ------------------------------------------------------------
# 16. OUTPUT VALIDATION AND FINAL STATUS
# ------------------------------------------------------------

output_files = {
    "geometry_csv": geometry_csv_path,
    "geometry_json": geometry_json_path,
    "geometry_plot": geometry_plot_path,
}

output_checks = {
    name: (
        path.is_file()
        and path.stat().st_size > 0
    )
    for name, path in output_files.items()
}

if not all(output_checks.values()):
    raise RuntimeError(
        "Не все выходные файлы CELL 9 успешно сохранены: "
        f"{output_checks}"
    )

if atom_count_matches_cell_7 is False:
    raise ValueError(
        "Количество атомов df_atoms не совпадает "
        "с total_protein_atoms из CELL 7."
    )


geometry_summary["output_checks"] = output_checks

geometry_summary["status"] = "OK"

with open(
    geometry_json_path,
    "w",
    encoding="utf-8",
) as file:

    json.dump(
        geometry_summary,
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False,
    )


# Проверяем итоговый JSON после повторной записи.
if (
    not geometry_json_path.is_file()
    or geometry_json_path.stat().st_size == 0
):
    raise RuntimeError(
        "Не удалось сохранить итоговый JSON CELL 9."
    )


# ------------------------------------------------------------
# 17. DISPLAY RESULTS
# ------------------------------------------------------------

print("\nSTRUCTURE PARAMETERS")
print("-" * 70)

print(f"Protein atoms:    {atom_count}")
print(f"Protein residues: {residue_count}")
print(f"Protein chains:   {chain_count}")

print("\nGEOMETRIC CENTROID")
print("-" * 70)

print(f"X: {centroid_x:.3f} Å")
print(f"Y: {centroid_y:.3f} Å")
print(f"Z: {centroid_z:.3f} Å")

print("\nBOUNDING BOX")
print("-" * 70)

print(f"X dimension: {dimension_x:.3f} Å")
print(f"Y dimension: {dimension_y:.3f} Å")
print(f"Z dimension: {dimension_z:.3f} Å")

print(
    f"Bounding box volume: {bounding_box_volume:.3f} Å³"
)

print("\nDISTANCES FROM GEOMETRIC CENTROID")
print("-" * 70)

print(f"Minimum: {min_distance:.3f} Å")
print(f"Mean:    {mean_distance:.3f} Å")
print(f"Median:  {median_distance:.3f} Å")
print(f"Maximum: {max_distance:.3f} Å")

print("\nRADIUS OF GYRATION")
print("-" * 70)

print(
    "Unweighted geometric radius of gyration: "
    f"{unweighted_geometric_radius_of_gyration:.3f} Å"
)

print("\nASPECT RATIO")
print("-" * 70)

print(
    f"{aspect_ratio:.3f}"
    if aspect_ratio is not None
    else "Not defined (at least one dimension is zero)"
)

print("\nCELL 7 ATOM COUNT CHECK")
print("-" * 70)

if atom_count_check_available:

    print(
        f"Expected: {expected_protein_atom_count}; "
        f"observed: {atom_count}; "
        f"matches: {atom_count_matches_cell_7}"
    )

else:

    print(
        "Skipped: total_protein_atoms is not defined."
    )

print("\nOUTPUT FILES")
print("-" * 70)

for name, path in output_files.items():
    print(f"{name}: {path}")

print("\nSTATUS: OK")
print("CELL 9 COMPLETED SUCCESSFULLY.")
print("=" * 70)
```

Это облако координат атомов, а не полноценная молекулярная визуализация. Для изучения пространственной структуры, вторичной структуры и взаимодействий удобнее использовать специализированные визуализаторы, например PyMOL.

Что добавили
Показатель	Что означает
Геометрический центр	Среднее положение всех атомов
Bounding box	Габариты структуры по X, Y и Z
Объём bounding box	Объём параллелепипеда, охватывающего все атомы
Максимальное расстояние от центра	Расстояние до наиболее удалённого атома
Радиус инерции	Характеристика пространственной компактности структуры

Особенно полезен радиус инерции (Radius of gyration). Он позволяет количественно описать, насколько компактно расположены атомы белка. Это может быть полезным структурным признаком при дальнейшем машинном обучении.

```python
# ============================================================
# CELL 10. МЕТАДАННЫЕ И КАЧЕСТВО СТРУКТУРЫ
# ============================================================
# Назначение:
# Извлечение метаданных из mmCIF и контроль качества
# структурных таблиц, сформированных в CELL 4–8.
#
# Анализируем первый модельный экземпляр структуры.
# Число моделей исходного mmCIF фиксируется отдельно.
#
# Важно:
# - записи unobserved/zero occupancy из mmCIF не равны
#   автоматически числу отсутствующих атомов;
# - occupancy = 0 отличается от отсутствующего occupancy;
# - Bio.PDB возвращает выбранное представление altloc,
#   а не все альтернативные конформации.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.PDB.MMCIF2Dict import MMCIF2Dict
from IPython.display import display


print("=" * 70)
print("CELL 10. МЕТАДАННЫЕ И КАЧЕСТВО СТРУКТУРЫ")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка входных данных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "CIF_PATH",
    "PDB_RESULTS_DIR",
    "models",
    "chains",
    "df_residues",
    "df_atoms",
    "df_hetero",
    "df_water",
    "df_hetero_atoms",
    "df_water_atoms"
]

for variable in required_variables:
    if variable not in globals():
        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()
CIF_PATH = Path(CIF_PATH)
PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

if PDB_RESULTS_DIR.name.upper() != PDB_ID:
    raise ValueError(
        f"Каталог результатов не соответствует PDB ID: "
        f"{PDB_RESULTS_DIR}"
    )

PDB_RESULTS_DIR.mkdir(parents=True, exist_ok=True)

if not CIF_PATH.is_file():
    raise FileNotFoundError(
        f"Файл mmCIF не найден: {CIF_PATH}"
    )


dataframe_variables = [
    "df_residues",
    "df_atoms",
    "df_hetero",
    "df_water",
    "df_hetero_atoms",
    "df_water_atoms"
]

for variable in dataframe_variables:
    if not isinstance(globals()[variable], pd.DataFrame):
        raise TypeError(
            f"{variable} должен быть pandas.DataFrame."
        )


if df_residues.empty or df_atoms.empty:
    raise ValueError(
        "Таблицы белковых остатков или атомов пусты."
    )


if not isinstance(models, (list, tuple)):
    models = list(models)

if not isinstance(chains, (list, tuple)):
    chains = list(chains)

if len(models) == 0:
    raise ValueError("В структуре Bio.PDB не найдено моделей.")

if len(chains) == 0:
    raise ValueError("В анализируемой модели не найдено цепей.")


print(f"PDB ID: {PDB_ID}")
print(f"mmCIF file: {CIF_PATH.name}")
print(f"Каталог результатов: {PDB_RESULTS_DIR}")


# ------------------------------------------------------------
# 2. Чтение исходного mmCIF
# ------------------------------------------------------------

cif_dict = MMCIF2Dict(str(CIF_PATH))

print(f"Количество полей mmCIF: {len(cif_dict)}")


# ------------------------------------------------------------
# 3. Функции обработки значений mmCIF
# ------------------------------------------------------------

def normalize_cif_value(value):
    """Преобразует маркеры отсутствующих значений mmCIF в None."""

    if value is None:
        return None

    value = str(value).strip()

    if value in {"", ".", "?"}:
        return None

    return value


def get_cif_values(data, key):
    """Возвращает все доступные значения указанного поля."""

    value = data.get(key)

    if value is None:
        return []

    if not isinstance(value, (list, tuple)):
        value = [value]

    return [
        normalized
        for item in value
        if (normalized := normalize_cif_value(item)) is not None
    ]


def get_cif_value(data, key):
    """Возвращает первое доступное значение поля."""

    values = get_cif_values(data, key)
    return values[0] if values else None


def get_cif_category(data, category_prefix):
    """
    Извлекает категорию mmCIF в DataFrame.

    Отдельно различает отсутствие категории и наличие
    категории без записей.
    """

    category_fields = {
        key: value
        for key, value in data.items()
        if key.startswith(category_prefix)
    }

    if not category_fields:
        return None, "Категория отсутствует в mmCIF"

    normalized_fields = {}

    for key, values in category_fields.items():
        if not isinstance(values, (list, tuple)):
            values = [values]

        normalized_fields[key] = [
            normalize_cif_value(value)
            for value in values
        ]

    record_lengths = {
        len(values)
        for values in normalized_fields.values()
    }

    if len(record_lengths) != 1:
        raise ValueError(
            f"Несогласованное число записей "
            f"в категории {category_prefix}: {record_lengths}"
        )

    record_count = next(iter(record_lengths))

    if record_count == 0:
        return (
            pd.DataFrame(columns=list(normalized_fields.keys())),
            "Категория присутствует, записи отсутствуют"
        )

    return (
        pd.DataFrame(normalized_fields),
        "Категория присутствует, имеются записи"
    )


def parse_numeric_value(value):
    """Преобразует значение в конечное число или None."""

    if value is None:
        return None

    try:
        result = float(value)
        return result if np.isfinite(result) else None
    except (TypeError, ValueError):
        return None


def get_unique_cif_values(data, key):
    """Возвращает уникальные значения поля без маркеров пропусков."""

    return sorted(set(get_cif_values(data, key)))


# ------------------------------------------------------------
# 4. Основные метаданные
# ------------------------------------------------------------

structure_title = get_cif_value(
    cif_dict,
    "_struct.title"
)

experimental_method_values = get_unique_cif_values(
    cif_dict,
    "_exptl.method"
)

experimental_method = (
    "; ".join(experimental_method_values)
    if experimental_method_values
    else None
)

method_lower = (
    experimental_method.lower()
    if experimental_method
    else ""
)


# ------------------------------------------------------------
# 5. Разрешение структуры
# ------------------------------------------------------------
# Рентгеновские поля и поле EM извлекаются отдельно.
# Неизвестный метод не позволяет уверенно выбрать
# одно из значений как основное разрешение.

resolution_xray_refine = get_cif_value(
    cif_dict,
    "_refine.ls_d_res_high"
)

resolution_xray_reflns = get_cif_value(
    cif_dict,
    "_reflns.d_resolution_high"
)

resolution_em = get_cif_value(
    cif_dict,
    "_em_3d_reconstruction.resolution"
)


resolution_xray = (
    resolution_xray_refine
    if resolution_xray_refine is not None
    else resolution_xray_reflns
)

resolution = None
resolution_source = "Not available in mmCIF"


is_em = (
    "electron microscopy" in method_lower
    or "electron cryo-microscopy" in method_lower
    or "electron cryo microscopy" in method_lower
)

is_xray = (
    "x-ray" in method_lower
    or "x ray" in method_lower
    or "x-ray diffraction" in method_lower
)


if is_em:
    if resolution_em is not None:
        resolution = resolution_em
        resolution_source = "Electron microscopy"
    else:
        resolution_source = "EM resolution not found"


elif is_xray:
    if resolution_xray_refine is not None:
        resolution = resolution_xray_refine
        resolution_source = "X-ray refinement"
    elif resolution_xray_reflns is not None:
        resolution = resolution_xray_reflns
        resolution_source = "X-ray diffraction data"
    else:
        resolution_source = "X-ray resolution not found"


else:
    available_resolutions = []

    if resolution_xray_refine is not None:
        available_resolutions.append(
            ("X-ray refinement", resolution_xray_refine)
        )

    elif resolution_xray_reflns is not None:
        available_resolutions.append(
            ("X-ray diffraction data", resolution_xray_reflns)
        )

    if resolution_em is not None:
        available_resolutions.append(
            ("Electron microscopy", resolution_em)
        )

    if len(available_resolutions) == 1:
        resolution_source, resolution = available_resolutions[0]

    elif len(available_resolutions) > 1:
        resolution_source = (
            "Multiple possible sources; method not identified"
        )


resolution_angstrom = parse_numeric_value(resolution)


deposition_date = get_cif_value(
    cif_dict,
    "_pdbx_database_status.recvd_initial_deposition_date"
)

release_date = get_cif_value(
    cif_dict,
    "_pdbx_database_status.release_date"
)

structure_method_details = get_cif_value(
    cif_dict,
    "_struct_keywords.pdbx_keywords"
)


# ------------------------------------------------------------
# 6. Число моделей в исходном mmCIF
# ------------------------------------------------------------

cif_model_numbers = get_unique_cif_values(
    cif_dict,
    "_atom_site.pdbx_PDB_model_num"
)

number_of_models_in_cif = (
    len(cif_model_numbers)
    if cif_model_numbers
    else None
)

number_of_models_parsed = len(models)

number_of_chains = len(chains)

number_of_protein_residues = len(df_residues)

number_of_protein_atoms = len(df_atoms)

number_of_nonwater_hetero_residues = len(df_hetero)

number_of_water_residues = len(df_water)


number_of_atom_chains = (
    int(df_atoms["chain_id"].nunique(dropna=False))
    if "chain_id" in df_atoms.columns
    else None
)


# ------------------------------------------------------------
# 7. Проверка PDB ID в координатных таблицах
# ------------------------------------------------------------

def check_pdb_id(dataframe, table_name):
    """Проверяет наличие и соответствие PDB ID в таблице."""

    if dataframe.empty:
        return {
            "table": table_name,
            "check": None,
            "message": "Таблица пуста"
        }

    if "pdb_id" not in dataframe.columns:
        return {
            "table": table_name,
            "check": False,
            "message": "Отсутствует столбец pdb_id"
        }

    values = (
        dataframe["pdb_id"]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    result = bool(
        values.notna().all()
        and values.eq(PDB_ID).all()
    )

    return {
        "table": table_name,
        "check": result,
        "message": (
            "Все записи относятся к текущей структуре"
            if result
            else "Обнаружены пропуски или другой PDB ID"
        )
    }


pdb_id_checks = [
    check_pdb_id(df_residues, "df_residues"),
    check_pdb_id(df_atoms, "df_atoms"),
    check_pdb_id(df_hetero, "df_hetero"),
    check_pdb_id(df_water, "df_water"),
    check_pdb_id(df_hetero_atoms, "df_hetero_atoms"),
    check_pdb_id(df_water_atoms, "df_water_atoms")
]


# ------------------------------------------------------------
# 8. Подсчёт occupancy
# ------------------------------------------------------------

def count_occupancy_quality(dataframe, table_name):
    """
    Подсчитывает:
    - occupancy = 0;
    - отсутствующие occupancy;
    - некорректные или бесконечные значения.
    """

    if dataframe is None:
        return {
            "table": table_name,
            "atom_count": 0,
            "zero_occupancy_count": None,
            "missing_occupancy_count": None,
            "invalid_occupancy_count": None,
            "occupancy_column_available": False
        }

    if dataframe.empty:
        return {
            "table": table_name,
            "atom_count": 0,
            "zero_occupancy_count": 0,
            "missing_occupancy_count": 0,
            "invalid_occupancy_count": 0,
            "occupancy_column_available": (
                "occupancy" in dataframe.columns
            )
        }

    if "occupancy" not in dataframe.columns:
        return {
            "table": table_name,
            "atom_count": len(dataframe),
            "zero_occupancy_count": None,
            "missing_occupancy_count": None,
            "invalid_occupancy_count": None,
            "occupancy_column_available": False
        }

    original = dataframe["occupancy"]

    numeric = pd.to_numeric(
        original,
        errors="coerce"
    )

    finite_mask = np.isfinite(
        numeric.fillna(np.nan).to_numpy(dtype=float)
    )

    missing_count = int(original.isna().sum())

    invalid_count = int(
        (original.notna() & numeric.isna()).sum()
        + (numeric.notna() & ~finite_mask).sum()
    )

    zero_count = int(
        (numeric == 0).sum()
    )

    return {
        "table": table_name,
        "atom_count": len(dataframe),
        "zero_occupancy_count": zero_count,
        "missing_occupancy_count": missing_count,
        "invalid_occupancy_count": invalid_count,
        "occupancy_column_available": True
    }


occupancy_records = [
    count_occupancy_quality(df_atoms, "Protein atoms"),
    count_occupancy_quality(df_hetero_atoms, "Hetero atoms"),
    count_occupancy_quality(df_water_atoms, "Water atoms")
]

df_occupancy_summary = pd.DataFrame(
    occupancy_records
)


# ------------------------------------------------------------
# 9. Основные метаданные
# ------------------------------------------------------------

metadata = {
    "PDB_ID": PDB_ID,
    "Title": structure_title,
    "Experimental_method": experimental_method,
    "Structure_keywords": structure_method_details,
    "Resolution": resolution,
    "Resolution_Angstrom": resolution_angstrom,
    "Resolution_source": resolution_source,
    "Resolution_X_ray_refinement": resolution_xray_refine,
    "Resolution_X_ray_diffraction": resolution_xray_reflns,
    "Resolution_EM": resolution_em,
    "Deposition_date": deposition_date,
    "Release_date": release_date,
    "Number_of_models_in_mmCIF": number_of_models_in_cif,
    "Number_of_models_parsed_by_Bio_PDB": number_of_models_parsed,
    "Number_of_chains_in_analyzed_model": number_of_chains,
    "Number_of_chains_in_protein_atom_table": number_of_atom_chains,
    "Number_of_protein_residues": number_of_protein_residues,
    "Number_of_protein_atoms": number_of_protein_atoms,
    "Number_of_nonwater_hetero_residues": number_of_nonwater_hetero_residues,
    "Number_of_water_residues": number_of_water_residues,
    "Analyzed_model_scope": "First model",
    "Coordinate_unit": "Angstrom",
    "Altloc_policy": "Bio.PDB selected atom representation"
}


# ------------------------------------------------------------
# 10. Неразрешённые остатки
# ------------------------------------------------------------

(
    df_unobserved_residues,
    residues_category_status
) = get_cif_category(
    cif_dict,
    "_pdbx_unobs_or_zero_occ_residues."
)

unobserved_residue_count = (
    None
    if df_unobserved_residues is None
    else len(df_unobserved_residues)
)


# ------------------------------------------------------------
# 11. Неразрешённые атомы
# ------------------------------------------------------------

(
    df_unobserved_atoms,
    atoms_category_status
) = get_cif_category(
    cif_dict,
    "_pdbx_unobs_or_zero_occ_atoms."
)

unobserved_atom_count = (
    None
    if df_unobserved_atoms is None
    else len(df_unobserved_atoms)
)


df_unobserved_summary = pd.DataFrame([
    {
        "Parameter": "Unobserved / zero-occupancy residue records",
        "Count": unobserved_residue_count,
        "Status": residues_category_status
    },
    {
        "Parameter": "Unobserved / zero-occupancy atom records",
        "Count": unobserved_atom_count,
        "Status": atoms_category_status
    }
])


# ------------------------------------------------------------
# 12. Сводные таблицы
# ------------------------------------------------------------

df_metadata = pd.DataFrame([
    {"Parameter": key, "Value": value}
    for key, value in metadata.items()
])


print("\n" + "-" * 70)
print("ОСНОВНЫЕ МЕТАДАННЫЕ")
print("-" * 70)

for key, value in metadata.items():
    print(f"{key}: {value}")


print("\nНЕРАЗРЕШЁННЫЕ ОСТАТКИ И АТОМЫ")
display(df_unobserved_summary)

print("\nOCCUPANCY")
display(df_occupancy_summary)


if df_unobserved_residues is not None:
    print("\nЗаписи о неразрешённых остатках:")
    display(df_unobserved_residues.head(20))

if df_unobserved_atoms is not None:
    print("\nЗаписи о неразрешённых атомах:")
    display(df_unobserved_atoms.head(20))


# ------------------------------------------------------------
# 13. Сохранение таблиц
# ------------------------------------------------------------

METADATA_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_metadata.csv"
)

UNOBSERVED_RESIDUES_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_unobserved_residues.csv"
)

UNOBSERVED_ATOMS_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_unobserved_atoms.csv"
)

UNOBSERVED_SUMMARY_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_unobserved_summary.csv"
)

OCCUPANCY_SUMMARY_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_occupancy_summary.csv"
)


df_metadata.to_csv(
    METADATA_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

if df_unobserved_residues is None:
    pd.DataFrame().to_csv(
        UNOBSERVED_RESIDUES_CSV_PATH,
        index=False,
        encoding="utf-8-sig"
    )
else:
    df_unobserved_residues.to_csv(
        UNOBSERVED_RESIDUES_CSV_PATH,
        index=False,
        encoding="utf-8-sig"
    )

if df_unobserved_atoms is None:
    pd.DataFrame().to_csv(
        UNOBSERVED_ATOMS_CSV_PATH,
        index=False,
        encoding="utf-8-sig"
    )
else:
    df_unobserved_atoms.to_csv(
        UNOBSERVED_ATOMS_CSV_PATH,
        index=False,
        encoding="utf-8-sig"
    )

df_unobserved_summary.to_csv(
    UNOBSERVED_SUMMARY_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

df_occupancy_summary.to_csv(
    OCCUPANCY_SUMMARY_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 14. Контроль согласованности
# ------------------------------------------------------------

checks = {
    "At least one parsed model": number_of_models_parsed > 0,
    "At least one chain": number_of_chains > 0,
    "Protein residues available": number_of_protein_residues > 0,
    "Protein atoms available": number_of_protein_atoms > 0,
    "PDB ID checks": all(
        item["check"] is not False
        for item in pdb_id_checks
    )
}

# Дополнительная проверка числа моделей.
if number_of_models_in_cif is not None:
    checks["mmCIF and Bio.PDB model counts agree"] = (
        number_of_models_in_cif == number_of_models_parsed
    )


# Согласование количества атомов с CELL 4, если статистика доступна.
if "structure_statistics" in globals():
    expected_atoms = structure_statistics.get(
        "protein_atoms"
    )

    if expected_atoms is not None:
        checks["Protein atom count agrees with CELL 4"] = (
            int(expected_atoms) == number_of_protein_atoms
        )


df_pdb_id_checks = pd.DataFrame(pdb_id_checks)

print("\n" + "-" * 70)
print("КОНТРОЛЬ СОГЛАСОВАННОСТИ")
print("-" * 70)

display(df_pdb_id_checks)

for check_name, check_result in checks.items():
    print(
        f"{check_name}: "
        f"{'OK' if check_result else 'ERROR'}"
    )


# ------------------------------------------------------------
# 15. JSON со сведениями о качестве
# ------------------------------------------------------------

quality_summary = {
    "pdb_id": PDB_ID,

    "unobserved_residues": {
        "record_count": unobserved_residue_count,
        "category_status": residues_category_status
    },

    "unobserved_atoms": {
        "record_count": unobserved_atom_count,
        "category_status": atoms_category_status
    },

    "zero_occupancy": occupancy_records,

    "pdb_id_checks": pdb_id_checks,

    "checks": checks,

    "analysis_scope": "First model",

    "interpretation": {
        "unobserved_categories":
            "Counts are mmCIF category records, not necessarily unique missing atoms or residues.",
        "zero_occupancy_tables":
            "Counts refer to atoms in extracted coordinate tables with occupancy equal to zero.",
        "altloc":
            "Bio.PDB selected atom representation is used; alternative conformations are not counted as separate atoms."
    }
}


METADATA_JSON_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_metadata.json"
)

metadata_complete = {
    "metadata": metadata,
    "quality": quality_summary
}

with open(
    METADATA_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        metadata_complete,
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


# ------------------------------------------------------------
# 16. Проверка сохранённых файлов
# ------------------------------------------------------------

output_paths = [
    METADATA_CSV_PATH,
    METADATA_JSON_PATH,
    UNOBSERVED_RESIDUES_CSV_PATH,
    UNOBSERVED_ATOMS_CSV_PATH,
    UNOBSERVED_SUMMARY_CSV_PATH,
    OCCUPANCY_SUMMARY_CSV_PATH
]

output_checks = {}

print("\n" + "-" * 70)
print("ПРОВЕРКА СОХРАНЁННЫХ ФАЙЛОВ")
print("-" * 70)

for path in output_paths:
    file_exists = path.is_file() and path.stat().st_size > 0
    output_checks[path.name] = bool(file_exists)

    print(
        f"{path.name}: "
        f"{'OK' if file_exists else 'ERROR'}"
    )


# Сохраняем диагностику даже при ошибках проверок.
quality_summary["output_checks"] = output_checks

with open(
    METADATA_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        {
            "metadata": metadata,
            "quality": quality_summary
        },
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


if not all(output_checks.values()):
    raise RuntimeError(
        "Не все файлы метаданных успешно сохранены."
    )

if not all(checks.values()):
    raise ValueError(
        "Обнаружены несоответствия при контроле качества. "
        "Смотрите сохранённый metadata.json."
    )


print("\nИнтерпретация:")
print(
    "- Число моделей mmCIF и число моделей Bio.PDB "
    "проверяются отдельно."
)
print(
    "- Число цепей, остатков и атомов относится "
    "к первому модельному экземпляру."
)
print(
    "- Записи unobserved/zero occupancy не равны "
    "автоматически количеству отсутствующих атомов."
)
print(
    "- Отсутствие категории mmCIF не доказывает "
    "полное отсутствие структурных проблем."
)

print("\nСтатус: OK")
print("Метаданные и сведения о качестве сохранены.")
print("=" * 70)
print("CELL 10 COMPLETED")
print("=" * 70)
```

Здесь важно учитывать ограничение: отсутствие записей в этих полях не гарантирует, что структура полностью разрешена. Для полноценного анализа необходимо также учитывать пропуски в последовательности, occupancy, альтернативные положения атомов и особенности конкретной структуры.

```python
# ============================================================
# CELL 11. AMINO ACID DISTRIBUTION
# ============================================================
# Назначение:
# Анализ аминокислотного состава белковой структуры.
#
# Рассчитывается:
# 1. Количество остатков каждого типа.
# 2. Процентное содержание аминокислот.
# 3. Накопительный процент.
# 4. Распределение аминокислот по цепям.
# 5. Доля неизвестных, неоднозначных и некорректных записей.
#
# Источник: df_residues из CELL 6.
#
# Вода и небелковые компоненты не учитываются.
# U (селеноцистеин) и O (пирролизин) включаются.
# ============================================================

import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from IPython.display import display


print("=" * 70)
print("CELL 11. AMINO ACID DISTRIBUTION")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка входных данных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "df_residues",
    "PDB_RESULTS_DIR"
]

for variable in required_variables:
    if variable not in globals():
        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()
PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

if PDB_RESULTS_DIR.name.upper() != PDB_ID:
    raise ValueError(
        f"Каталог результатов не соответствует PDB ID: "
        f"{PDB_RESULTS_DIR}"
    )

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


if not isinstance(df_residues, pd.DataFrame):
    raise TypeError(
        "df_residues должен быть pandas.DataFrame."
    )

if df_residues.empty:
    raise ValueError(
        "Таблица df_residues пуста."
    )


required_columns = [
    "pdb_id",
    "chain_id",
    "amino_acid",
    "residue_key"
]

missing_columns = [
    column
    for column in required_columns
    if column not in df_residues.columns
]

if missing_columns:
    raise ValueError(
        "В df_residues отсутствуют обязательные столбцы: "
        f"{missing_columns}"
    )


# ------------------------------------------------------------
# 2. Проверка принадлежности текущей структуре
# ------------------------------------------------------------

source_pdb_ids = (
    df_residues["pdb_id"]
    .astype("string")
    .str.strip()
    .str.upper()
)

pdb_id_check = bool(
    source_pdb_ids.notna().all()
    and source_pdb_ids.eq(PDB_ID).all()
)

if not pdb_id_check:
    raise ValueError(
        f"Несоответствие PDB ID. "
        f"Текущая структура: {PDB_ID}. "
        "В df_residues присутствуют пропуски "
        "или записи другой структуры."
    )


# ------------------------------------------------------------
# 3. Проверка уникальности остатков
# ------------------------------------------------------------
# residue_key должен однозначно идентифицировать остаток
# внутри текущей структуры.

residue_keys = df_residues["residue_key"]

if residue_keys.isna().any():
    raise ValueError(
        "В df_residues обнаружены остатки без residue_key."
    )

if residue_keys.astype(str).str.strip().eq("").any():
    raise ValueError(
        "В df_residues обнаружены пустые residue_key."
    )

duplicate_residue_mask = residue_keys.duplicated(
    keep=False
)

duplicate_residue_count = int(
    duplicate_residue_mask.sum()
)

if duplicate_residue_count > 0:
    duplicate_path = (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_duplicate_residue_keys.csv"
    )

    df_residues.loc[
        duplicate_residue_mask
    ].to_csv(
        duplicate_path,
        index=False,
        encoding="utf-8-sig"
    )

    raise ValueError(
        f"Обнаружены повторяющиеся residue_key: "
        f"{duplicate_residue_count} строк. "
        f"Диагностика сохранена: {duplicate_path}"
    )


# Пустая строка chain_id допустима: это blank chain.
# Пропущенный chain_id не допускается.

if df_residues["chain_id"].isna().any():
    raise ValueError(
        "В df_residues обнаружены остатки без chain_id."
    )


print(f"PDB ID: {PDB_ID}")
print(f"Количество остатков: {len(df_residues)}")
print(f"Каталог результатов: {PDB_RESULTS_DIR}")
print("Уникальность residue_key: OK")


# ------------------------------------------------------------
# 4. Подготовка аминокислотных обозначений
# ------------------------------------------------------------

amino_acid_series = (
    df_residues["amino_acid"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
)


# Стандартные аминокислоты.
STANDARD_AMINO_ACIDS = [
    "A", "R", "N", "D", "C",
    "Q", "E", "G", "H", "I",
    "L", "K", "M", "F", "P",
    "S", "T", "W", "Y", "V"
]


# Дополнительные распознаваемые аминокислоты.
# U — селеноцистеин.
# O — пирролизин.

ADDITIONAL_AMINO_ACIDS = [
    "U",
    "O"
]


VALID_AMINO_ACIDS = set(
    STANDARD_AMINO_ACIDS
    + ADDITIONAL_AMINO_ACIDS
)


# Неоднозначные обозначения:
# B — аспарагин или аспартат;
# Z — глутамин или глутамат;
# J — лейцин или изолейцин.

AMBIGUOUS_AMINO_ACIDS = {
    "B",
    "Z",
    "J"
}


# Неизвестная аминокислота.
UNKNOWN_AMINO_ACIDS = {
    "X"
}


# ------------------------------------------------------------
# 5. Классификация записей
# ------------------------------------------------------------

is_single_letter = amino_acid_series.str.len().eq(1)

is_valid_amino_acid = (
    amino_acid_series.isin(VALID_AMINO_ACIDS)
)

is_ambiguous_amino_acid = (
    amino_acid_series.isin(AMBIGUOUS_AMINO_ACIDS)
)

is_unknown_amino_acid = (
    amino_acid_series.isin(UNKNOWN_AMINO_ACIDS)
)

is_missing_amino_acid = (
    amino_acid_series.eq("")
)

is_invalid_amino_acid = ~(
    is_valid_amino_acid
    | is_ambiguous_amino_acid
    | is_unknown_amino_acid
    | is_missing_amino_acid
)


# ------------------------------------------------------------
# 6. Сводка классификации обозначений
# ------------------------------------------------------------

total_residue_records = len(df_residues)

counted_amino_acid_mask = is_valid_amino_acid

ambiguous_count = int(
    is_ambiguous_amino_acid.sum()
)

unknown_count = int(
    is_unknown_amino_acid.sum()
)

missing_count = int(
    is_missing_amino_acid.sum()
)

invalid_count = int(
    is_invalid_amino_acid.sum()
)

counted_amino_acid_count = int(
    counted_amino_acid_mask.sum()
)


classification_total = (
    counted_amino_acid_count
    + ambiguous_count
    + unknown_count
    + missing_count
    + invalid_count
)

classification_check = (
    classification_total == total_residue_records
)


if not classification_check:
    raise ValueError(
        "Ошибка классификации аминокислотных обозначений."
    )


df_amino_acid_classification = pd.DataFrame([
    {
        "category": "Counted amino acids",
        "count": counted_amino_acid_count
    },
    {
        "category": "Ambiguous amino acids",
        "count": ambiguous_count
    },
    {
        "category": "Unknown amino acids",
        "count": unknown_count
    },
    {
        "category": "Missing amino acid values",
        "count": missing_count
    },
    {
        "category": "Invalid amino acid values",
        "count": invalid_count
    }
])


df_amino_acid_classification["percentage"] = (
    df_amino_acid_classification["count"]
    / total_residue_records
    * 100
).round(2)


print("\nКЛАССИФИКАЦИЯ АМИНОКИСЛОТНЫХ ОБОЗНАЧЕНИЙ")
display(df_amino_acid_classification)


# ------------------------------------------------------------
# 7. Основное распределение аминокислот
# ------------------------------------------------------------

valid_amino_acid_series = (
    amino_acid_series[counted_amino_acid_mask]
)


if valid_amino_acid_series.empty:
    raise ValueError(
        "Не найдено пригодных для основного анализа "
        "аминокислотных остатков."
    )


amino_acid_counts = (
    valid_amino_acid_series
    .value_counts()
    .sort_values(
        ascending=False,
        kind="stable"
    )
)


amino_acid_distribution = pd.DataFrame({
    "amino_acid": amino_acid_counts.index,
    "count": amino_acid_counts.values
})


# ------------------------------------------------------------
# 8. Процентное содержание и накопительный процент
# ------------------------------------------------------------

total_residues = int(
    amino_acid_distribution["count"].sum()
)


amino_acid_distribution["percentage_raw"] = (
    amino_acid_distribution["count"]
    / total_residues
    * 100
)


amino_acid_distribution["percentage"] = (
    amino_acid_distribution["percentage_raw"]
    .round(2)
)


amino_acid_distribution["cumulative_percentage"] = (
    amino_acid_distribution["percentage_raw"]
    .cumsum()
    .round(2)
)


percentage_sum_raw = float(
    amino_acid_distribution["percentage_raw"].sum()
)

percentage_sum = float(
    amino_acid_distribution["percentage"].sum()
)


amino_acid_distribution = (
    amino_acid_distribution
    .drop(columns=["percentage_raw"])
)


# ------------------------------------------------------------
# 9. Дополнительная статистика
# ------------------------------------------------------------

number_of_amino_acid_types = len(
    amino_acid_distribution
)


most_common_amino_acid = str(
    amino_acid_distribution.iloc[0]["amino_acid"]
)

most_common_amino_acid_count = int(
    amino_acid_distribution.iloc[0]["count"]
)


least_common_amino_acid = str(
    amino_acid_distribution.iloc[-1]["amino_acid"]
)

least_common_amino_acid_count = int(
    amino_acid_distribution.iloc[-1]["count"]
)


excluded_residue_count = (
    ambiguous_count
    + unknown_count
    + missing_count
    + invalid_count
)


excluded_percentage = (
    excluded_residue_count
    / total_residue_records
    * 100
)


# ------------------------------------------------------------
# 10. Распределение аминокислот по цепям
# ------------------------------------------------------------

chain_data = df_residues[
    ["chain_id", "amino_acid"]
].copy()


chain_data["amino_acid"] = (
    chain_data["amino_acid"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
)


chain_data = chain_data.loc[
    chain_data["amino_acid"].isin(VALID_AMINO_ACIDS)
].copy()


# Исходный chain_id сохраняется.
# Отображаемое обозначение blank chain формируется отдельно.

chain_data["chain_id_display"] = (
    chain_data["chain_id"]
    .astype(str)
    .map(
        lambda value: (
            "<blank>"
            if value == ""
            else value
        )
    )
)


chain_amino_acid_distribution = (
    chain_data
    .groupby(
        ["chain_id_display", "amino_acid"],
        dropna=False
    )
    .size()
    .reset_index(name="count")
)


if not chain_amino_acid_distribution.empty:

    chain_amino_acid_distribution["percentage"] = (
        chain_amino_acid_distribution
        .groupby("chain_id_display")["count"]
        .transform(
            lambda values:
                values / values.sum() * 100
        )
        .round(2)
    )

    chain_amino_acid_distribution = (
        chain_amino_acid_distribution
        .sort_values(
            ["chain_id_display", "count", "amino_acid"],
            ascending=[True, False, True]
        )
        .reset_index(drop=True)
    )

else:
    chain_amino_acid_distribution = pd.DataFrame(
        columns=[
            "chain_id_display",
            "amino_acid",
            "count",
            "percentage"
        ]
    )


# ------------------------------------------------------------
# 11. Статистика по цепям
# ------------------------------------------------------------

chain_residue_counts = (
    chain_data
    .groupby("chain_id_display")
    .size()
    .reset_index(name="counted_residue_count")
)


number_of_chains = int(
    df_residues["chain_id"].nunique(dropna=False)
)


# ------------------------------------------------------------
# 12. Вывод результатов
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ОБЩЕЕ РАСПРЕДЕЛЕНИЕ АМИНОКИСЛОТ")
print("-" * 70)

display(amino_acid_distribution)


print(f"\nTotal residue records: {total_residue_records}")
print(f"Counted amino acid residues: {total_residues}")
print(f"Number of amino acid types: {number_of_amino_acid_types}")

print(f"Ambiguous residues: {ambiguous_count}")
print(f"Unknown residues: {unknown_count}")
print(f"Missing amino acid values: {missing_count}")
print(f"Invalid amino acid values: {invalid_count}")

print(f"Excluded residue records: {excluded_residue_count}")
print(f"Excluded percentage: {excluded_percentage:.2f}%")

print(
    f"Most common amino acid: "
    f"{most_common_amino_acid} "
    f"({most_common_amino_acid_count})"
)

print(
    f"Least common amino acid: "
    f"{least_common_amino_acid} "
    f"({least_common_amino_acid_count})"
)

print(
    f"Sum of exact percentages: "
    f"{percentage_sum_raw:.6f}%"
)

print(
    f"Sum of displayed percentages: "
    f"{percentage_sum:.2f}%"
)


print("\n" + "-" * 70)
print("РАСПРЕДЕЛЕНИЕ АМИНОКИСЛОТ ПО ЦЕПЯМ")
print("-" * 70)

display(chain_amino_acid_distribution)


print("\nКоличество учитываемых остатков по цепям:")
display(chain_residue_counts)


# ------------------------------------------------------------
# 13. Сохранение общей таблицы
# ------------------------------------------------------------

AMINO_ACID_CSV_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_amino_acid_distribution.csv"
)

amino_acid_distribution.to_csv(
    AMINO_ACID_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 14. Сохранение распределения по цепям
# ------------------------------------------------------------

AMINO_ACID_CHAIN_CSV_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_amino_acid_distribution_by_chain.csv"
)

chain_amino_acid_distribution.to_csv(
    AMINO_ACID_CHAIN_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 15. Сохранение классификации обозначений
# ------------------------------------------------------------

AMINO_ACID_CLASSIFICATION_CSV_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_amino_acid_classification.csv"
)

df_amino_acid_classification.to_csv(
    AMINO_ACID_CLASSIFICATION_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 16. Диаграмма количества аминокислот
# ------------------------------------------------------------

AMINO_ACID_PLOT_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_amino_acid_distribution.png"
)

fig, ax = plt.subplots(figsize=(12, 5))

ax.bar(
    amino_acid_distribution["amino_acid"],
    amino_acid_distribution["count"]
)

ax.set_title(
    f"Amino acid distribution: {PDB_ID}"
)

ax.set_xlabel("Amino acid")
ax.set_ylabel("Number of residues")

ax.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    AMINO_ACID_PLOT_PATH,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)


# ------------------------------------------------------------
# 17. Диаграмма процентного содержания
# ------------------------------------------------------------

AMINO_ACID_PERCENTAGE_PLOT_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_amino_acid_percentage.png"
)

fig, ax = plt.subplots(figsize=(12, 5))

ax.bar(
    amino_acid_distribution["amino_acid"],
    amino_acid_distribution["percentage"]
)

ax.set_title(
    f"Amino acid percentage distribution: {PDB_ID}"
)

ax.set_xlabel("Amino acid")
ax.set_ylabel("Percentage (%)")

ax.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    AMINO_ACID_PERCENTAGE_PLOT_PATH,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)


# ------------------------------------------------------------
# 18. Проверка результатов
# ------------------------------------------------------------

count_check = (
    total_residues + excluded_residue_count
    == total_residue_records
)


percentage_check = np.isclose(
    percentage_sum_raw,
    100.0,
    atol=1e-8
)


chain_count_check = (
    int(chain_amino_acid_distribution["count"].sum())
    == total_residues
)


classification_check = (
    classification_total
    == total_residue_records
)


unique_residue_check = (
    not df_residues["residue_key"].duplicated().any()
)


print("\n" + "-" * 70)
print("VERIFICATION")
print("-" * 70)

checks = {
    "PDB ID check": pdb_id_check,
    "Unique residue_key check": unique_residue_check,
    "Residue count check": count_check,
    "Percentage check": percentage_check,
    "Chain distribution check": chain_count_check,
    "Classification check": classification_check
}


for check_name, check_result in checks.items():
    print(
        f"{check_name}: "
        f"{'OK' if check_result else 'ERROR'}"
    )


if not all(checks.values()):
    raise ValueError(
        "Ошибка проверки распределения аминокислот."
    )


# ------------------------------------------------------------
# 19. Итоговая сводка
# ------------------------------------------------------------

amino_acid_summary = {
    "pdb_id": PDB_ID,

    "total_residue_records": total_residue_records,

    "counted_amino_acid_residues": total_residues,

    "excluded_residue_records": excluded_residue_count,

    "excluded_percentage": float(excluded_percentage),

    "ambiguous_residue_count": ambiguous_count,

    "unknown_residue_count": unknown_count,

    "missing_amino_acid_count": missing_count,

    "invalid_amino_acid_count": invalid_count,

    "number_of_amino_acid_types": number_of_amino_acid_types,

    "most_common_amino_acid": most_common_amino_acid,

    "most_common_amino_acid_count": most_common_amino_acid_count,

    "least_common_amino_acid": least_common_amino_acid,

    "least_common_amino_acid_count": least_common_amino_acid_count,

    "percentage_sum": percentage_sum_raw,

    "chain_count": number_of_chains,

    "included_amino_acid_codes": sorted(VALID_AMINO_ACIDS),

    "ambiguous_codes": sorted(AMBIGUOUS_AMINO_ACIDS),

    "unknown_codes": sorted(UNKNOWN_AMINO_ACIDS),

    "checks": {
        key: bool(value)
        for key, value in checks.items()
    }
}


# ------------------------------------------------------------
# 20. Проверка сохранённых файлов
# ------------------------------------------------------------

AMINO_ACID_JSON_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_amino_acid_summary.json"
)


output_paths = [
    AMINO_ACID_CSV_PATH,
    AMINO_ACID_CHAIN_CSV_PATH,
    AMINO_ACID_CLASSIFICATION_CSV_PATH,
    AMINO_ACID_PLOT_PATH,
    AMINO_ACID_PERCENTAGE_PLOT_PATH
]


output_checks = {}

print("\n" + "-" * 70)
print("SAVED FILES")
print("-" * 70)

for path in output_paths:

    file_exists = (
        path.is_file()
        and path.stat().st_size > 0
    )

    output_checks[path.name] = bool(file_exists)

    print(
        f"{path.name}: "
        f"{'OK' if file_exists else 'ERROR'}"
    )


if not all(output_checks.values()):
    raise RuntimeError(
        "Не все файлы анализа аминокислот "
        "успешно сохранены."
    )


# ------------------------------------------------------------
# 21. Сохранение итогового JSON
# ------------------------------------------------------------

amino_acid_summary["output_checks"] = output_checks
amino_acid_summary["status"] = "OK"


with open(
    AMINO_ACID_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        amino_acid_summary,
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


if (
    not AMINO_ACID_JSON_PATH.is_file()
    or AMINO_ACID_JSON_PATH.stat().st_size == 0
):
    raise RuntimeError(
        "Итоговый JSON не был сохранён."
    )


# ------------------------------------------------------------
# 22. Итоговый статус
# ------------------------------------------------------------

print("\nИтог:")

print(
    "Распределение рассчитано по белковым остаткам "
    "из df_residues."
)

print(
    "В основной расчёт включены стандартные аминокислоты, "
    "селеноцистеин (U) и пирролизин (O)."
)

print(
    "Неоднозначные обозначения B, Z, J и неизвестное "
    "обозначение X учтены отдельно."
)

print(
    "Распределение по цепям рассчитано независимо "
    "для каждой цепи."
)

print("\nСтатус: OK")
print("=" * 70)
print("CELL 11 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# CELL 12. SAVE ANALYSIS RESULTS
# ============================================================
# Назначение:
# Централизованное сохранение результатов анализа структуры.
#
# Используются результаты CELL 4–11.
#
# Формируются:
# 1. Итоговая сводка JSON.
# 2. Краткая сводная таблица CSV.
# 3. Реестр файлов текущего PDB.
#
# Все результаты сохраняются в отдельном каталоге PDB_ID.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from IPython.display import display


print("=" * 70)
print("CELL 12. SAVE ANALYSIS RESULTS")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка основных переменных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "PDB_RESULTS_DIR",
    "chain_sequences",
    "df_residues",
    "df_atoms",
    "df_hetero",
    "df_water"
]

for variable in required_variables:
    if variable not in globals():
        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()

if not PDB_ID:
    raise ValueError("PDB_ID не может быть пустым.")


PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

if PDB_RESULTS_DIR.name.upper() != PDB_ID:
    raise ValueError(
        "Каталог результатов не соответствует текущему PDB_ID: "
        f"{PDB_RESULTS_DIR}"
    )


PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


if not isinstance(chain_sequences, dict):
    raise TypeError(
        "chain_sequences должен быть словарём."
    )


dataframe_variables = [
    "df_residues",
    "df_atoms",
    "df_hetero",
    "df_water"
]


for variable in dataframe_variables:
    if not isinstance(globals()[variable], pd.DataFrame):
        raise TypeError(
            f"{variable} должен быть pandas.DataFrame."
        )


if df_residues.empty:
    raise ValueError(
        "df_residues пуста. "
        "Невозможно сформировать итоговую сводку."
    )


if df_atoms.empty:
    raise ValueError(
        "df_atoms пуста. "
        "Невозможно сформировать итоговую сводку."
    )


print(f"PDB ID: {PDB_ID}")
print(f"Results directory: {PDB_RESULTS_DIR}")


# ------------------------------------------------------------
# 2. Проверка принадлежности таблиц текущей структуре
# ------------------------------------------------------------

def validate_dataframe_pdb_id(dataframe, table_name):
    """
    Проверяет:
    - наличие столбца pdb_id;
    - отсутствие пропусков в непустой таблице;
    - соответствие всех записей текущему PDB_ID.

    Пустые таблицы допускаются, например если структура
    не содержит небелковых компонентов или воды.
    """

    if "pdb_id" not in dataframe.columns:
        raise ValueError(
            f"В {table_name} отсутствует столбец pdb_id."
        )

    if dataframe.empty:
        return {
            "table": table_name,
            "row_count": 0,
            "status": True,
            "message": "Пустая таблица; проверка PDB ID не требуется"
        }

    source_ids = (
        dataframe["pdb_id"]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    if source_ids.isna().any():
        return {
            "table": table_name,
            "row_count": len(dataframe),
            "status": False,
            "message": "Обнаружены пропущенные PDB ID"
        }

    matches = source_ids.eq(PDB_ID)

    if not matches.all():
        found_ids = (
            source_ids.dropna()
            .unique()
            .tolist()
        )

        return {
            "table": table_name,
            "row_count": len(dataframe),
            "status": False,
            "message": (
                f"Ожидался {PDB_ID}; обнаружены значения {found_ids}"
            )
        }

    return {
        "table": table_name,
        "row_count": len(dataframe),
        "status": True,
        "message": "Все записи относятся к текущему PDB"
    }


pdb_id_checks = [
    validate_dataframe_pdb_id(
        globals()[variable],
        variable
    )
    for variable in dataframe_variables
]


# Дополнительные таблицы CELL 8, если они доступны.

optional_dataframe_variables = [
    "df_hetero_atoms",
    "df_water_atoms"
]

for variable in optional_dataframe_variables:
    if variable in globals():
        dataframe = globals()[variable]

        if not isinstance(dataframe, pd.DataFrame):
            raise TypeError(
                f"{variable} должен быть pandas.DataFrame."
            )

        pdb_id_checks.append(
            validate_dataframe_pdb_id(
                dataframe,
                variable
            )
        )


pdb_id_check = all(
    item["status"]
    for item in pdb_id_checks
)


print(
    "DataFrame PDB_ID consistency:",
    "OK" if pdb_id_check else "ERROR"
)


if not pdb_id_check:
    display(pd.DataFrame(pdb_id_checks))

    raise ValueError(
        "Обнаружено несоответствие PDB ID "
        "в структурных таблицах."
    )


# ------------------------------------------------------------
# 3. Функция преобразования значений для JSON
# ------------------------------------------------------------

def make_json_serializable(value):
    """
    Рекурсивно преобразует NumPy и pandas в JSON-совместимые
    типы данных.

    NaN, Infinity, NaT и pd.NA преобразуются в None.
    """

    if value is None:
        return None

    if isinstance(value, Path):
        return str(value)

    if isinstance(value, pd.DataFrame):
        return [
            make_json_serializable(record)
            for record in value.to_dict(orient="records")
        ]

    if isinstance(value, pd.Series):
        return make_json_serializable(
            value.to_dict()
        )

    if isinstance(value, dict):
        return {
            str(key): make_json_serializable(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple, set)):
        return [
            make_json_serializable(item)
            for item in value
        ]

    if isinstance(value, np.ndarray):
        return make_json_serializable(
            value.tolist()
        )

    if isinstance(value, np.generic):
        return make_json_serializable(
            value.item()
        )

    if isinstance(value, (pd.Timestamp, pd.Timedelta)):
        return value.isoformat()

    if isinstance(value, (float, np.floating)):
        if not np.isfinite(value):
            return None
        return float(value)

    if isinstance(value, (int, np.integer)):
        return int(value)

    if isinstance(value, (bool, np.bool_)):
        return bool(value)

    if value is pd.NA or value is pd.NaT:
        return None

    # Проверка скалярных пропущенных значений.
    try:
        missing_value = pd.isna(value)

        if (
            isinstance(missing_value, (bool, np.bool_))
            and missing_value
        ):
            return None

    except (TypeError, ValueError):
        pass

    # Строки, None и стандартные JSON-типы остаются без изменений.
    if isinstance(value, (str, bool)):
        return value

    return value


# ------------------------------------------------------------
# 4. Информация о последовательностях белковых цепей
# ------------------------------------------------------------

chain_lengths = {
    str(chain_id): len(str(sequence))
    for chain_id, sequence in chain_sequences.items()
}


chain_information = {
    "sequence_chains": list(chain_lengths.keys()),

    "chain_lengths": chain_lengths,

    "number_of_sequence_chains": len(chain_sequences),

    "total_sequence_length": int(
        sum(chain_lengths.values())
    )
}


# ------------------------------------------------------------
# 5. Общая статистика структуры
# ------------------------------------------------------------

number_of_models_parsed = (
    len(models)
    if "models" in globals()
    else None
)


number_of_chains = (
    len(chains)
    if "chains" in globals()
    else None
)


number_of_sequence_chains = len(chain_sequences)


number_of_protein_residues = len(df_residues)

number_of_protein_atoms = len(df_atoms)

number_of_hetero_residues = len(df_hetero)

number_of_water_residues = len(df_water)


# Отдельно сохраняем число атомов небелковых компонентов
# и воды, если соответствующие таблицы существуют.

number_of_hetero_atoms = (
    len(df_hetero_atoms)
    if "df_hetero_atoms" in globals()
    else None
)

number_of_water_atoms = (
    len(df_water_atoms)
    if "df_water_atoms" in globals()
    else None
)


structure_summary = {
    "PDB_ID": PDB_ID,

    "number_of_models_parsed": number_of_models_parsed,

    "number_of_chains_in_analyzed_model": number_of_chains,

    "number_of_sequence_chains": number_of_sequence_chains,

    "protein_residues": number_of_protein_residues,

    "protein_atoms": number_of_protein_atoms,

    "hetero_residues": number_of_hetero_residues,

    "hetero_atoms": number_of_hetero_atoms,

    "water_residues": number_of_water_residues,

    "water_atoms": number_of_water_atoms,

    "chain_information": chain_information,

    "analysis_scope": "First model"
}


# ------------------------------------------------------------
# 6. Добавление геометрических характеристик
# ------------------------------------------------------------

if "geometry_summary" in globals():

    structure_summary["geometry"] = geometry_summary

elif "dimensions" in globals():

    structure_summary["geometry"] = {
        "coordinate_dimensions": np.asarray(
            dimensions,
            dtype=float
        ).tolist()
    }

else:

    structure_summary["geometry"] = None


# ------------------------------------------------------------
# 7. Добавление метаданных CELL 10
# ------------------------------------------------------------

if "metadata" in globals():

    structure_summary["metadata"] = metadata

else:

    structure_summary["metadata"] = None


# ------------------------------------------------------------
# 8. Добавление результатов контроля качества CELL 10
# ------------------------------------------------------------

quality_summary = {
    "unobserved_residue_records": (
        unobserved_residue_count
        if "unobserved_residue_count" in globals()
        else None
    ),

    "unobserved_atom_records": (
        unobserved_atom_count
        if "unobserved_atom_count" in globals()
        else None
    ),

    "residues_category_status": (
        residues_category_status
        if "residues_category_status" in globals()
        else "Not checked"
    ),

    "atoms_category_status": (
        atoms_category_status
        if "atoms_category_status" in globals()
        else "Not checked"
    ),

    "occupancy_summary": (
        df_occupancy_summary
        if "df_occupancy_summary" in globals()
        else None
    )
}


structure_summary["quality_information"] = quality_summary


# ------------------------------------------------------------
# 9. Добавление результатов CELL 11
# ------------------------------------------------------------

if "amino_acid_summary" in globals():

    structure_summary["amino_acid_analysis"] = (
        amino_acid_summary
    )

else:

    structure_summary["amino_acid_analysis"] = None


if "amino_acid_distribution" in globals():

    structure_summary["amino_acid_distribution"] = (
        amino_acid_distribution
    )

else:

    structure_summary["amino_acid_distribution"] = None


if "chain_amino_acid_distribution" in globals():

    structure_summary["amino_acid_distribution_by_chain"] = (
        chain_amino_acid_distribution
    )

else:

    structure_summary["amino_acid_distribution_by_chain"] = None


if "df_amino_acid_classification" in globals():

    structure_summary["amino_acid_classification"] = (
        df_amino_acid_classification
    )

else:

    structure_summary["amino_acid_classification"] = None


# ------------------------------------------------------------
# 10. Дополнительные проверки согласованности
# ------------------------------------------------------------

consistency_checks = {}


# Согласование длины последовательностей с таблицей остатков.
# Здесь сравниваются только общие количества, а не порядок
# или биологическая идентичность цепей.

sequence_length_total = int(
    sum(chain_lengths.values())
)

consistency_checks["Sequence length equals residue table size"] = (
    sequence_length_total == number_of_protein_residues
)


# Согласование числа цепей, если доступен список цепей.
if number_of_chains is not None:

    consistency_checks["Sequence chain count does not exceed model chain count"] = (
        number_of_sequence_chains <= number_of_chains
    )


# Согласование таблиц CELL 8 с числом строк.
if "df_hetero_atoms" in globals():

    consistency_checks["Hetero atom table available"] = (
        isinstance(df_hetero_atoms, pd.DataFrame)
    )

if "df_water_atoms" in globals():

    consistency_checks["Water atom table available"] = (
        isinstance(df_water_atoms, pd.DataFrame)
    )


# Результаты проверок CELL 10 и CELL 11, если они доступны.
if "checks" in globals() and isinstance(checks, dict):

    # Переменная checks могла быть переопределена CELL 11.
    # Поэтому сохраняем её отдельно, не меняя исходный словарь.
    previous_checks = dict(checks)

else:

    previous_checks = {}


if "amino_acid_summary" in globals():

    aa_checks = amino_acid_summary.get("checks", {})

    if isinstance(aa_checks, dict):

        for key, value in aa_checks.items():

            consistency_checks[
                f"CELL 11: {key}"
            ] = bool(value)


if "quality_summary" in globals():
    # Не используем quality_summary из одноимённой переменной
    # CELL 10 как источник статусов: она могла быть заменена
    # локальной сводкой в этой ячейке.
    pass


structure_summary["consistency_checks"] = consistency_checks

structure_summary["pdb_id_checks"] = pdb_id_checks

structure_summary["previous_cell_checks"] = previous_checks


# ------------------------------------------------------------
# 11. Формирование краткой сводной таблицы CSV
# ------------------------------------------------------------

SUMMARY_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_summary.csv"
)


summary_table = pd.DataFrame([{
    "PDB_ID": PDB_ID,

    "Number of models parsed": number_of_models_parsed,

    "Number of chains in analyzed model": number_of_chains,

    "Number of sequence chains": number_of_sequence_chains,

    "Protein residues": number_of_protein_residues,

    "Protein atoms": number_of_protein_atoms,

    "Hetero residues": number_of_hetero_residues,

    "Hetero atoms": number_of_hetero_atoms,

    "Water residues": number_of_water_residues,

    "Water atoms": number_of_water_atoms,

    "Total sequence length": sequence_length_total
}])


summary_table.to_csv(
    SUMMARY_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 12. Первичное сохранение JSON
# ------------------------------------------------------------

SUMMARY_JSON_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_summary.json"
)


summary = make_json_serializable(
    structure_summary
)


with open(
    SUMMARY_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        summary,
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


# ------------------------------------------------------------
# 13. Инвентаризация файлов текущего PDB
# ------------------------------------------------------------
# Учитываются файлы непосредственно в каталоге PDB_ID.
# Вложенные каталоги не обходятся.
#
# Сам реестр исключается из собственного содержимого.

OUTPUT_FILES_CSV_PATH = (
    PDB_RESULTS_DIR / f"{PDB_ID}_output_files.csv"
)


file_records = []


for path in sorted(PDB_RESULTS_DIR.iterdir()):

    if not path.is_file():
        continue

    if path == OUTPUT_FILES_CSV_PATH:
        continue

    file_size = path.stat().st_size

    file_records.append({
        "file_name": path.name,

        "extension": path.suffix.lower(),

        "size_bytes": int(file_size),

        "size_kb": round(file_size / 1024, 2),

        "exists": True
    })


df_output_files = pd.DataFrame(
    file_records,
    columns=[
        "file_name",
        "extension",
        "size_bytes",
        "size_kb",
        "exists"
    ]
)


# ------------------------------------------------------------
# 14. Сохранение реестра файлов
# ------------------------------------------------------------

df_output_files.to_csv(
    OUTPUT_FILES_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


print("\nРеестр файлов сохранён:")
print(OUTPUT_FILES_CSV_PATH)

print(
    "Количество найденных файлов "
    "(без самого реестра):",
    len(df_output_files)
)


# ------------------------------------------------------------
# 15. Проверка сохранённых файлов
# ------------------------------------------------------------

important_files = [
    SUMMARY_JSON_PATH,
    SUMMARY_CSV_PATH,
    OUTPUT_FILES_CSV_PATH
]


output_checks = {}

print("\n" + "-" * 70)
print("FILE VERIFICATION")
print("-" * 70)


for path in important_files:

    file_exists = (
        path.is_file()
        and path.stat().st_size > 0
    )

    output_checks[path.name] = bool(file_exists)

    print(
        f"{path.name}: "
        f"{'OK' if file_exists else 'ERROR'}"
    )


if not all(output_checks.values()):

    raise RuntimeError(
        "Не все итоговые файлы были успешно сохранены."
    )


# ------------------------------------------------------------
# 16. Формирование итогового статуса
# ------------------------------------------------------------

all_consistency_checks_passed = all(
    bool(value)
    for value in consistency_checks.values()
)

structure_summary["output_checks"] = output_checks

structure_summary["status"] = (
    "OK"
    if all_consistency_checks_passed
    else "WARNING"
)

structure_summary["file_inventory"] = {
    "direct_files_in_pdb_directory": len(df_output_files) + 1,
    "recursive_scan": False,
    "inventory_file": OUTPUT_FILES_CSV_PATH.name
}


# Повторное сохранение JSON уже с итоговым статусом.
summary = make_json_serializable(
    structure_summary
)

with open(
    SUMMARY_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        summary,
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


# ------------------------------------------------------------
# 17. Финальная проверка JSON
# ------------------------------------------------------------

if (
    not SUMMARY_JSON_PATH.is_file()
    or SUMMARY_JSON_PATH.stat().st_size == 0
):
    raise RuntimeError(
        "Итоговый JSON не был сохранён."
    )


# Проверяем, что JSON действительно читается.
with open(
    SUMMARY_JSON_PATH,
    "r",
    encoding="utf-8"
) as file:

    saved_summary = json.load(file)


if saved_summary.get("PDB_ID") != PDB_ID:

    raise ValueError(
        "PDB ID в сохранённом JSON не соответствует "
        "текущей структуре."
    )


# ------------------------------------------------------------
# 18. Вывод общей сводки
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("STRUCTURE SUMMARY")
print("-" * 70)

print(f"PDB ID: {PDB_ID}")

print(
    f"Number of parsed models: "
    f"{number_of_models_parsed}"
)

print(
    f"Number of chains in analyzed model: "
    f"{number_of_chains}"
)

print(
    f"Number of chains with sequences: "
    f"{number_of_sequence_chains}"
)

print(
    f"Protein residues: "
    f"{number_of_protein_residues}"
)

print(
    f"Protein atoms: "
    f"{number_of_protein_atoms}"
)

print(
    f"Hetero residues: "
    f"{number_of_hetero_residues}"
)

print(
    f"Water residues: "
    f"{number_of_water_residues}"
)

print(
    f"Total sequence length: "
    f"{sequence_length_total}"
)


print("\nChain sequence lengths:")

for chain_id, length in chain_lengths.items():

    display_chain_id = (
        "<blank>"
        if chain_id == ""
        else chain_id
    )

    print(
        f"Chain {display_chain_id}: {length} residues"
    )


# ------------------------------------------------------------
# 19. Вывод реестра файлов
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("OUTPUT FILES")
print("-" * 70)

display(df_output_files)


# ------------------------------------------------------------
# 20. Вывод проверок
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("CONSISTENCY CHECKS")
print("-" * 70)

for check_name, check_result in consistency_checks.items():

    print(
        f"{check_name}: "
        f"{'OK' if check_result else 'WARNING'}"
    )


# ------------------------------------------------------------
# 21. Итоговый статус
# ------------------------------------------------------------

total_pdb_files = len(df_output_files) + 1

print("\n" + "-" * 70)
print("ANALYSIS RESULTS SAVED")
print("-" * 70)

print(f"Results directory: {PDB_RESULTS_DIR}")

print(
    f"Number of files in PDB directory: "
    f"{total_pdb_files}"
)

print("Files from other PDB structures: excluded")

print(
    "Consistency status:",
    "OK" if all_consistency_checks_passed else "WARNING"
)

print(
    "Output status:",
    "OK" if all(output_checks.values()) else "ERROR"
)

if not all_consistency_checks_passed:

    print(
        "\nВНИМАНИЕ: часть дополнительных проверок "
        "не пройдена. Смотрите summary.json."
    )

print("\nСтатус сохранения: OK")

print("=" * 70)
print("CELL 12 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# CELL 13. PROTEIN SEQUENCE VALIDATION
# ============================================================
# Назначение:
# Сравнение последовательностей белковых цепей, извлечённых
# из координатной модели, с референсными последовательностями
# полимерных сущностей из mmCIF.
#
# Анализ:
# 1. Загрузка референсных последовательностей из mmCIF.
# 2. Определение соответствия chain_id -> entity_id.
# 3. Учёт модифицированных аминокислотных остатков.
# 4. Глобальное выравнивание последовательностей.
# 5. Расчёт identity и coverage.
# 6. Подсчёт несовпадений и gap-позиций.
# 7. Контроль полноты сравнения цепей.
#
# ВАЖНО:
# - Сравниваются последовательности, а не координаты.
# - Gap не доказывает отсутствие остатка в координатной модели.
# - Основной способ сопоставления — через mmCIF entity_id.
# - Резервный поиск по последовательностям помечается отдельно.
# - Используется первый модельный контекст структуры.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.Align import PairwiseAligner
from Bio.PDB.MMCIF2Dict import MMCIF2Dict


print("=" * 70)
print("CELL 13. PROTEIN SEQUENCE VALIDATION")
print("=" * 70)


# ------------------------------------------------------------
# 1. ПРОВЕРКА ВХОДНЫХ ДАННЫХ
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "CIF_PATH",
    "chain_sequences"
]

for variable in required_variables:

    if variable not in globals():

        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()

CIF_PATH = Path(CIF_PATH)

if not PDB_ID:

    raise ValueError("PDB_ID не может быть пустым.")

if not CIF_PATH.is_file():

    raise FileNotFoundError(
        f"Файл mmCIF не найден: {CIF_PATH}"
    )

if not isinstance(chain_sequences, dict) or not chain_sequences:

    raise ValueError(
        "chain_sequences должен быть непустым словарём."
    )


# CELL2 уже определяет RESULTS_DIR как папку текущего PDB.
if "PDB_RESULTS_DIR" in globals():

    PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

elif "RESULTS_DIR" in globals():

    PDB_RESULTS_DIR = Path(RESULTS_DIR)

else:

    raise NameError(
        "Не определена PDB_RESULTS_DIR или RESULTS_DIR."
    )


PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


if PDB_RESULTS_DIR.name != PDB_ID:

    raise ValueError(
        f"Папка результатов не соответствует PDB_ID: "
        f"{PDB_RESULTS_DIR}"
    )


print(f"PDB ID: {PDB_ID}")
print(f"CIF path: {CIF_PATH}")
print(f"Results directory: {PDB_RESULTS_DIR}")
print(f"Experimental chains: {len(chain_sequences)}")


# ------------------------------------------------------------
# 2. ЗАГРУЗКА MMCIF
# ------------------------------------------------------------

cif_dict = MMCIF2Dict(str(CIF_PATH))


# ------------------------------------------------------------
# 3. ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ
# ------------------------------------------------------------

def cif_values(value):

    """
    Преобразование значения MMCIF2Dict в список строк.
    """

    if value is None:

        return []

    if isinstance(value, str):

        return [value]

    return list(value)


def clean_cif_value(value):

    """
    Нормализация значений mmCIF.
    """

    if value is None:

        return None

    value = str(value).strip()

    if value in {"", ".", "?"}:

        return None

    return value


def normalize_sequence(sequence):

    """
    Нормализация аминокислотной последовательности.
    """

    if sequence is None:

        return ""

    return (
        str(sequence)
        .replace("\n", "")
        .replace("\r", "")
        .replace(" ", "")
        .replace("\t", "")
        .upper()
    )


def get_first_model_number(cif_data):

    """
    Определение номера первой модели из atom_site.
    """

    model_values = cif_values(
        cif_data.get(
            "_atom_site.pdbx_PDB_model_num"
        )
    )

    valid_models = []

    for value in model_values:

        value = clean_cif_value(value)

        if value is not None and value not in valid_models:

            valid_models.append(value)

    if valid_models:

        return valid_models[0]

    return "1"


# ------------------------------------------------------------
# 4. ЗАГРУЗКА РЕФЕРЕНСНЫХ ПОСЛЕДОВАТЕЛЬНОСТЕЙ
# ------------------------------------------------------------

reference_sequences = cif_values(
    cif_dict.get(
        "_entity_poly.pdbx_seq_one_letter_code_can"
    )
)

reference_sequence_source = (
    "_entity_poly.pdbx_seq_one_letter_code_can"
)


if not reference_sequences:

    reference_sequences = cif_values(
        cif_dict.get(
            "_entity_poly.pdbx_seq_one_letter_code"
        )
    )

    reference_sequence_source = (
        "_entity_poly.pdbx_seq_one_letter_code"
    )


entity_ids = cif_values(
    cif_dict.get(
        "_entity_poly.entity_id"
    )
)


if not reference_sequences:

    raise ValueError(
        "В mmCIF отсутствуют референсные последовательности."
    )


if len(entity_ids) != len(reference_sequences):

    raise ValueError(
        "Количество entity_id не совпадает "
        "с количеством референсных последовательностей."
    )


reference_records = []

for entity_id, sequence in zip(
    entity_ids,
    reference_sequences
):

    entity_id = clean_cif_value(entity_id)

    sequence = normalize_sequence(sequence)

    if entity_id is None or not sequence:

        continue

    reference_records.append({

        "entity_id": entity_id,

        "sequence": sequence,

        "sequence_length": len(sequence),

        "sequence_source": reference_sequence_source

    })


df_reference = pd.DataFrame(
    reference_records
)


if df_reference.empty:

    raise ValueError(
        "Не удалось сформировать таблицу референсных последовательностей."
    )


if df_reference["entity_id"].duplicated().any():

    raise ValueError(
        "Обнаружены повторяющиеся entity_id."
    )


known_entities = set(
    df_reference["entity_id"]
)


print("\n" + "-" * 70)
print("REFERENCE SEQUENCES")
print("-" * 70)

print(f"Reference source: {reference_sequence_source}")
print(f"Reference entities: {len(df_reference)}")

display(
    df_reference[
        [
            "entity_id",
            "sequence_length",
            "sequence_source"
        ]
    ]
)


# ------------------------------------------------------------
# 5. ОПРЕДЕЛЕНИЕ ПЕРВОЙ МОДЕЛИ
# ------------------------------------------------------------

FIRST_MODEL_NUMBER = get_first_model_number(
    cif_dict
)


print(
    f"\nFirst model number: {FIRST_MODEL_NUMBER}"
)


# ------------------------------------------------------------
# 6. СОПОСТАВЛЕНИЕ AUTH_CHAIN -> ENTITY_ID
# ------------------------------------------------------------

# Для связи используем atom_site:
# auth_asym_id — авторский идентификатор цепи;
# label_entity_id — идентификатор полимерной сущности.
#
# Не фильтруем только по group_PDB == ATOM:
# модифицированные аминокислоты могут находиться в HETATM.
#
# Для надёжного определения принадлежности используем
# label_entity_id и наличие соответствующей entity_poly.

atom_site_auth = cif_values(
    cif_dict.get(
        "_atom_site.auth_asym_id"
    )
)

atom_site_entity = cif_values(
    cif_dict.get(
        "_atom_site.label_entity_id"
    )
)

atom_site_model = cif_values(
    cif_dict.get(
        "_atom_site.pdbx_PDB_model_num"
    )
)


if not atom_site_auth or not atom_site_entity:

    raise ValueError(
        "В mmCIF отсутствуют обязательные поля "
        "_atom_site.auth_asym_id или _atom_site.label_entity_id."
    )


if len(atom_site_auth) != len(atom_site_entity):

    raise ValueError(
        "Количество записей auth_asym_id и label_entity_id "
        "не совпадает."
    )


if atom_site_model and len(atom_site_model) != len(atom_site_auth):

    raise ValueError(
        "Количество записей pdbx_PDB_model_num "
        "не совпадает с atom_site."
    )


auth_chain_to_entities = {}

for i, (auth_id, entity_id) in enumerate(
    zip(
        atom_site_auth,
        atom_site_entity
    )
):

    auth_id = clean_cif_value(auth_id)

    entity_id = clean_cif_value(entity_id)

    if auth_id is None or entity_id is None:

        continue

    if entity_id not in known_entities:

        continue

    # Если поле номера модели отсутствует,
    # используем доступные записи как единый контекст.
    if atom_site_model:

        model_number = clean_cif_value(
            atom_site_model[i]
        )

        if model_number != FIRST_MODEL_NUMBER:

            continue

    auth_chain_to_entities.setdefault(
        auth_id,
        set()
    ).add(entity_id)


# ------------------------------------------------------------
# 7. РЕЗЕРВНАЯ СВЯЗЬ ЧЕРЕЗ POLY_SEQ_SCHEME
# ------------------------------------------------------------

# Используем pdb_strand_id как дополнительный источник.
# В этой категории могут быть перечислены несколько
# авторских идентификаторов цепей.

scheme_auth = cif_values(
    cif_dict.get(
        "_pdbx_poly_seq_scheme.pdb_strand_id"
    )
)

scheme_entity = cif_values(
    cif_dict.get(
        "_pdbx_poly_seq_scheme.entity_id"
    )
)


if scheme_auth and scheme_entity:

    if len(scheme_auth) != len(scheme_entity):

        raise ValueError(
            "Некорректная длина колонок "
            "_pdbx_poly_seq_scheme."
        )

    for auth_ids, entity_id in zip(
        scheme_auth,
        scheme_entity
    ):

        entity_id = clean_cif_value(
            entity_id
        )

        if entity_id is None or entity_id not in known_entities:

            continue

        if clean_cif_value(auth_ids) is None:

            continue

        chain_ids = [
            item.strip()
            for item in str(auth_ids).split(",")
            if item.strip()
        ]

        for chain_id in chain_ids:

            auth_chain_to_entities.setdefault(
                chain_id,
                set()
            ).add(entity_id)


# ------------------------------------------------------------
# 8. ФОРМИРОВАНИЕ КАРТЫ CHAIN -> ENTITY
# ------------------------------------------------------------

chain_entity_mapping = {}

mapping_records = []


for chain_id in chain_sequences:

    chain_id = str(chain_id)

    candidate_entities = sorted(
        auth_chain_to_entities.get(
            chain_id,
            set()
        )
    )


    if len(candidate_entities) == 1:

        mapping_status = "Direct"

        entity_id = candidate_entities[0]

    elif len(candidate_entities) > 1:

        mapping_status = "Ambiguous"

        entity_id = None

    else:

        mapping_status = "Not found"

        entity_id = None


    chain_entity_mapping[chain_id] = {

        "entity_id": entity_id,

        "mapping_status": mapping_status,

        "candidate_entities": candidate_entities

    }


    mapping_records.append({

        "chain_id": chain_id,

        "entity_id": entity_id,

        "mapping_status": mapping_status,

        "candidate_entities": ";".join(
            candidate_entities
        ),

        "candidate_count": len(candidate_entities)

    })


df_chain_mapping = pd.DataFrame(
    mapping_records
)


print("\n" + "-" * 70)
print("CHAIN TO ENTITY MAPPING")
print("-" * 70)

display(df_chain_mapping)


# ------------------------------------------------------------
# 9. НАСТРОЙКА ГЛОБАЛЬНОГО ВЫРАВНИВАНИЯ
# ------------------------------------------------------------

aligner = PairwiseAligner()

aligner.mode = "global"

aligner.match_score = 2.0

aligner.mismatch_score = -1.0

aligner.open_gap_score = -3.0

aligner.extend_gap_score = -0.5


# ------------------------------------------------------------
# 10. РАСЧЁТ МЕТРИК ВЫРАВНИВАНИЯ
# ------------------------------------------------------------

def calculate_alignment_metrics(
    reference_sequence,
    experimental_sequence,
    alignment
):

    """
    Метрики глобального выравнивания.

    Identity:
        Совпадения / число парных позиций.

    Reference coverage:
        Доля референсных остатков, имеющих пару
        в экспериментальной последовательности.

    Experimental coverage:
        Доля экспериментальных остатков, имеющих пару
        в референсной последовательности.

    Gaps:
        Разрывы в соответствующих последовательностях.
    """

    indices = alignment.indices

    reference_indices = indices[0]

    experimental_indices = indices[1]


    paired_positions = (

        (reference_indices >= 0)
        &
        (experimental_indices >= 0)

    )


    aligned_count = int(
        paired_positions.sum()
    )


    if aligned_count == 0:

        raise ValueError(
            "Выравнивание не содержит сопоставленных позиций."
        )


    paired_reference_indices = (
        reference_indices[paired_positions]
    )

    paired_experimental_indices = (
        experimental_indices[paired_positions]
    )


    matches = sum(

        reference_sequence[int(ref_idx)]
        == experimental_sequence[int(exp_idx)]

        for ref_idx, exp_idx in zip(
            paired_reference_indices,
            paired_experimental_indices
        )

    )


    mismatches = aligned_count - matches


    identity = (
        matches / aligned_count * 100
    )


    reference_coverage = (

        len(np.unique(paired_reference_indices))
        / len(reference_sequence)
        * 100

    )


    experimental_coverage = (

        len(np.unique(paired_experimental_indices))
        / len(experimental_sequence)
        * 100

    )


    missing_reference = int(

        np.sum(
            (reference_indices >= 0)
            &
            (experimental_indices < 0)
        )

    )


    extra_experimental = int(

        np.sum(
            (reference_indices < 0)
            &
            (experimental_indices >= 0)
        )

    )


    reference_self_score = aligner.score(
        reference_sequence,
        reference_sequence
    )

    experimental_self_score = aligner.score(
        experimental_sequence,
        experimental_sequence
    )


    normalization_score = max(
        reference_self_score,
        experimental_self_score
    )


    normalized_score = (

        alignment.score / normalization_score

        if normalization_score > 0

        else 0

    )


    return {

        "alignment_score": float(alignment.score),

        "normalized_score": float(normalized_score),

        "matches": int(matches),

        "mismatches": int(mismatches),

        "aligned_positions": aligned_count,

        "identity_percent": float(identity),

        "reference_coverage_percent": float(
            reference_coverage
        ),

        "experimental_coverage_percent": float(
            experimental_coverage
        ),

        "missing_reference_residues": missing_reference,

        "extra_experimental_residues": extra_experimental

    }


# ------------------------------------------------------------
# 11. СРАВНЕНИЕ ПОСЛЕДОВАТЕЛЬНОСТЕЙ
# ------------------------------------------------------------

comparison_records = []

best_alignments = {}

best_alignment_details = {}


for chain_id, experimental_sequence in chain_sequences.items():

    chain_id = str(chain_id)

    experimental_sequence = normalize_sequence(
        experimental_sequence
    )

    if not experimental_sequence:

        print(
            f"WARNING: Chain {chain_id} has an empty sequence."
        )

        continue


    mapping = chain_entity_mapping[chain_id]

    mapped_entity = mapping["entity_id"]

    mapping_status = mapping["mapping_status"]


    # Прямое сопоставление:
    # только с entity_id, определённым через mmCIF.
    #
    # Резервный поиск:
    # используется при отсутствии или неоднозначности карты.

    if mapped_entity is not None:

        candidate_references = df_reference[
            df_reference["entity_id"] == mapped_entity
        ]

        comparison_method = "Mapped entity"

    else:

        candidate_references = df_reference

        comparison_method = "Fallback search"


    best_result = None

    best_alignment = None

    best_reference_sequence = None

    best_rank = None


    print(
        f"\nProcessing chain {chain_id}: "
        f"{len(experimental_sequence)} residues"
    )

    print(
        f"Mapping status: {mapping_status}"
    )

    print(
        f"Comparison method: {comparison_method}"
    )


    for _, reference in candidate_references.iterrows():

        reference_sequence = reference["sequence"]

        if not reference_sequence:

            continue


        alignments = aligner.align(
            reference_sequence,
            experimental_sequence
        )

        alignment = alignments[0]


        metrics = calculate_alignment_metrics(
            reference_sequence,
            experimental_sequence,
            alignment
        )


        # Ранжирование:
        # 1. Нормализованный score.
        # 2. Identity.
        # 3. Минимальное покрытие двух последовательностей.

        result_rank = (

            metrics["normalized_score"],

            metrics["identity_percent"],

            min(
                metrics["reference_coverage_percent"],
                metrics["experimental_coverage_percent"]
            )

        )


        if best_result is None or result_rank > best_rank:

            best_rank = result_rank

            best_alignment = alignment

            best_reference_sequence = reference_sequence

            best_result = {

                "chain_id": chain_id,

                "entity_id": reference["entity_id"],

                "reference_length": len(reference_sequence),

                "experimental_length": len(experimental_sequence),

                **metrics

            }


    if best_result is None:

        print(
            f"WARNING: No reference found for chain {chain_id}."
        )

        continue


    best_result["mapping_status"] = mapping_status

    best_result["comparison_method"] = comparison_method

    best_result["candidate_entities"] = (
        mapping["candidate_entities"]
    )


    # Резервное сопоставление не считается подтверждённым
    # соответствием биологической сущности.
    if comparison_method == "Fallback search":

        best_result["mapping_confidence"] = "Unverified"

    else:

        best_result["mapping_confidence"] = "Direct"


    comparison_records.append({

        "chain_id": chain_id,

        "entity_id": best_result["entity_id"],

        "mapping_status": mapping_status,

        "comparison_method": comparison_method,

        "mapping_confidence": best_result["mapping_confidence"],

        "reference_length": best_result["reference_length"],

        "experimental_length": best_result["experimental_length"],

        "alignment_score": best_result["alignment_score"],

        "normalized_score": round(
            best_result["normalized_score"],
            4
        ),

        "matches": best_result["matches"],

        "mismatches": best_result["mismatches"],

        "identity_percent": round(
            best_result["identity_percent"],
            2
        ),

        "reference_coverage_percent": round(
            best_result["reference_coverage_percent"],
            2
        ),

        "experimental_coverage_percent": round(
            best_result["experimental_coverage_percent"],
            2
        ),

        "aligned_positions": best_result["aligned_positions"],

        "missing_reference_residues": (
            best_result["missing_reference_residues"]
        ),

        "extra_experimental_residues": (
            best_result["extra_experimental_residues"]
        )

    })


    best_alignments[chain_id] = best_alignment

    best_alignment_details[chain_id] = {

        "entity_id": best_result["entity_id"],

        "reference_sequence": best_reference_sequence,

        "experimental_sequence": experimental_sequence,

        "alignment": best_alignment,

        "metrics": best_result

    }


# ------------------------------------------------------------
# 12. ИТОГОВАЯ ТАБЛИЦА СРАВНЕНИЯ
# ------------------------------------------------------------

df_reference_sequence_comparison = pd.DataFrame(
    comparison_records
)


if df_reference_sequence_comparison.empty:

    raise ValueError(
        "Не удалось выполнить сравнение ни для одной цепи."
    )


print("\n" + "-" * 70)
print("REFERENCE SEQUENCE COMPARISON")
print("-" * 70)

display(
    df_reference_sequence_comparison
)


# ------------------------------------------------------------
# 13. ПРОВЕРКА ПОЛНОТЫ СРАВНЕНИЯ
# ------------------------------------------------------------

experimental_chain_ids = set(
    map(str, chain_sequences.keys())
)

compared_chain_ids = set(
    df_reference_sequence_comparison["chain_id"]
)

not_compared_chains = sorted(
    experimental_chain_ids - compared_chain_ids
)


direct_mapping_count = int(
    (
        df_reference_sequence_comparison["comparison_method"]
        == "Mapped entity"
    ).sum()
)

fallback_mapping_count = int(
    (
        df_reference_sequence_comparison["comparison_method"]
        == "Fallback search"
    ).sum()
)

ambiguous_mapping_count = int(
    (
        df_reference_sequence_comparison["mapping_status"]
        == "Ambiguous"
    ).sum()
)

not_found_mapping_count = int(
    (
        df_reference_sequence_comparison["mapping_status"]
        == "Not found"
    ).sum()
)


print("\n" + "-" * 70)
print("CHAIN VALIDATION")
print("-" * 70)

print(
    f"Experimental chains: {len(experimental_chain_ids)}"
)

print(
    f"Compared chains: {len(compared_chain_ids)}"
)

print(
    f"Not compared chains: {len(not_compared_chains)}"
)

print(
    f"Direct mappings: {direct_mapping_count}"
)

print(
    f"Fallback mappings: {fallback_mapping_count}"
)

print(
    f"Ambiguous mappings: {ambiguous_mapping_count}"
)

print(
    f"Not found mappings: {not_found_mapping_count}"
)

if not_compared_chains:

    print(
        "WARNING: Some experimental chains were not compared:",
        not_compared_chains
    )

else:

    print(
        "All experimental chains were compared."
    )


# ------------------------------------------------------------
# 14. СОХРАНЕНИЕ РЕФЕРЕНСНЫХ ПОСЛЕДОВАТЕЛЬНОСТЕЙ
# ------------------------------------------------------------

REFERENCE_SEQUENCE_CSV_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_reference_sequences.csv"

)

df_reference.to_csv(
    REFERENCE_SEQUENCE_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 15. СОХРАНЕНИЕ КАРТЫ CHAIN -> ENTITY
# ------------------------------------------------------------

CHAIN_MAPPING_CSV_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_chain_entity_mapping.csv"

)

df_chain_mapping.to_csv(
    CHAIN_MAPPING_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 16. СОХРАНЕНИЕ РЕЗУЛЬТАТОВ СРАВНЕНИЯ
# ------------------------------------------------------------

SEQUENCE_COMPARISON_CSV_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_reference_sequence_comparison.csv"

)

df_reference_sequence_comparison.to_csv(
    SEQUENCE_COMPARISON_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 17. СОХРАНЕНИЕ ТЕКСТОВЫХ ВЫРАВНИВАНИЙ
# ------------------------------------------------------------

ALIGNMENTS_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_sequence_alignments.txt"

)

with open(
    ALIGNMENTS_PATH,
    "w",
    encoding="utf-8"
) as file:

    file.write(
        f"REFERENCE SEQUENCE ALIGNMENTS: {PDB_ID}\n"
    )

    file.write("=" * 80 + "\n\n")


    for chain_id, details in best_alignment_details.items():

        metrics = details["metrics"]

        file.write(
            f"CHAIN: {chain_id}\n"
        )

        file.write(
            f"ENTITY ID: {details['entity_id']}\n"
        )

        file.write(
            f"REFERENCE LENGTH: "
            f"{len(details['reference_sequence'])}\n"
        )

        file.write(
            f"EXPERIMENTAL LENGTH: "
            f"{len(details['experimental_sequence'])}\n"
        )

        file.write(
            f"IDENTITY: "
            f"{metrics['identity_percent']:.2f}%\n"
        )

        file.write(
            f"REFERENCE COVERAGE: "
            f"{metrics['reference_coverage_percent']:.2f}%\n"
        )

        file.write(
            f"EXPERIMENTAL COVERAGE: "
            f"{metrics['experimental_coverage_percent']:.2f}%\n\n"
        )

        file.write(
            str(details["alignment"])
        )

        file.write("\n\n")

        file.write("-" * 80 + "\n\n")


# ------------------------------------------------------------
# 18. ИНТЕРПРЕТАЦИЯ РЕЗУЛЬТАТОВ
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("INTERPRETATION")
print("-" * 70)


for _, row in df_reference_sequence_comparison.iterrows():

    print(
        f"\nChain: {row['chain_id']}"
    )

    print(
        f"Reference entity: {row['entity_id']}"
    )

    print(
        f"Mapping method: {row['comparison_method']}"
    )

    print(
        f"Identity: {row['identity_percent']:.2f}%"
    )

    print(
        f"Reference coverage: "
        f"{row['reference_coverage_percent']:.2f}%"
    )

    print(
        f"Experimental coverage: "
        f"{row['experimental_coverage_percent']:.2f}%"
    )

    print(
        f"Aligned positions: {row['aligned_positions']}"
    )

    print(
        f"Missing reference positions in alignment: "
        f"{row['missing_reference_residues']}"
    )

    print(
        f"Extra experimental positions in alignment: "
        f"{row['extra_experimental_residues']}"
    )


    if row["comparison_method"] == "Fallback search":

        print(
            "WARNING: Reference selected by sequence search. "
            "Entity correspondence is not independently confirmed."
        )


    if (

        row["identity_percent"] >= 99

        and row["reference_coverage_percent"] >= 99

        and row["experimental_coverage_percent"] >= 99

    ):

        print(
            "Almost complete sequence correspondence."
        )

    elif row["identity_percent"] >= 95:

        print(
            "High sequence identity, but differences "
            "or coverage gaps are present."
        )

    else:

        print(
            "Noticeable sequence differences. "
            "The alignment should be examined."
        )


# ------------------------------------------------------------
# 19. ФОРМИРОВАНИЕ JSON-СВОДКИ
# ------------------------------------------------------------

SEQUENCE_VALIDATION_JSON_PATH = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_sequence_validation.json"

)


sequence_validation_summary = {

    "PDB_ID": PDB_ID,

    "first_model_number": FIRST_MODEL_NUMBER,

    "reference_sequence_source": reference_sequence_source,

    "number_of_reference_sequences": int(
        len(df_reference)
    ),

    "number_of_experimental_chains": int(
        len(experimental_chain_ids)
    ),

    "number_of_compared_chains": int(
        len(compared_chain_ids)
    ),

    "number_of_uncompared_chains": int(
        len(not_compared_chains)
    ),

    "number_of_direct_mappings": direct_mapping_count,

    "number_of_fallback_mappings": fallback_mapping_count,

    "number_of_ambiguous_mappings": ambiguous_mapping_count,

    "number_of_not_found_mappings": not_found_mapping_count,

    "not_compared_chains": not_compared_chains,

    "chain_entity_mapping": df_chain_mapping.to_dict(
        orient="records"
    ),

    "results": df_reference_sequence_comparison.to_dict(
        orient="records"
    ),

    "interpretation": {

        "identity": (
            "Percentage of identical residues among "
            "aligned residue pairs."
        ),

        "reference_coverage": (
            "Percentage of reference residues participating "
            "in paired alignment positions."
        ),

        "experimental_coverage": (
            "Percentage of experimental residues participating "
            "in paired alignment positions."
        ),

        "gap_interpretation": (
            "Alignment gaps do not independently prove "
            "absence of residues from experimental coordinates."
        ),

        "fallback_interpretation": (
            "Fallback sequence search identifies the best "
            "sequence candidate but does not independently "
            "confirm biological entity correspondence."
        )

    }

}


# ------------------------------------------------------------
# 20. JSON-СОВМЕСТИМОЕ ПРЕОБРАЗОВАНИЕ
# ------------------------------------------------------------

def make_json_serializable(value):

    if value is None:

        return None

    if value is pd.NA or value is pd.NaT:

        return None

    if isinstance(value, dict):

        return {

            str(key): make_json_serializable(item)

            for key, item in value.items()

        }

    if isinstance(value, (list, tuple, set)):

        return [

            make_json_serializable(item)

            for item in value

        ]

    if isinstance(value, np.ndarray):

        return make_json_serializable(
            value.tolist()
        )

    if isinstance(value, np.generic):

        return make_json_serializable(
            value.item()
        )

    if isinstance(value, (pd.Timestamp,)):

        return value.isoformat()

    if isinstance(value, Path):

        return str(value)

    if isinstance(value, float):

        return float(value) if np.isfinite(value) else None

    if isinstance(value, (str, int, bool)):

        return value

    if pd.isna(value):

        return None

    return str(value)


with open(
    SEQUENCE_VALIDATION_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        make_json_serializable(
            sequence_validation_summary
        ),
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


# ------------------------------------------------------------
# 21. ПРОВЕРКА СОХРАНЁННЫХ ФАЙЛОВ
# ------------------------------------------------------------

output_paths = [

    REFERENCE_SEQUENCE_CSV_PATH,

    CHAIN_MAPPING_CSV_PATH,

    SEQUENCE_COMPARISON_CSV_PATH,

    ALIGNMENTS_PATH,

    SEQUENCE_VALIDATION_JSON_PATH

]


print("\n" + "-" * 70)
print("SAVED FILES")
print("-" * 70)


output_checks = {}

for path in output_paths:

    exists = path.is_file()

    output_checks[path.name] = exists

    print(
        f"{path.name}: "
        f"{'OK' if exists else 'ERROR'}"
    )


if not all(output_checks.values()):

    raise RuntimeError(
        "Не все результаты проверки последовательностей сохранены."
    )


# ------------------------------------------------------------
# 22. ИТОГОВЫЙ СТАТУС
# ------------------------------------------------------------

sequence_validation_status = (

    "OK"

    if (

        not not_compared_chains

        and fallback_mapping_count == 0

        and ambiguous_mapping_count == 0

    )

    else "WARNING"

)


# Дополняем JSON результатами проверок сохранения.
sequence_validation_summary["output_checks"] = output_checks

sequence_validation_summary["status"] = (
    sequence_validation_status
)


with open(
    SEQUENCE_VALIDATION_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        make_json_serializable(
            sequence_validation_summary
        ),
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


print("\n" + "-" * 70)
print("SEQUENCE VALIDATION SUMMARY")
print("-" * 70)

print(f"PDB ID: {PDB_ID}")

print(
    f"Reference entities: {len(df_reference)}"
)

print(
    f"Experimental chains: {len(experimental_chain_ids)}"
)

print(
    f"Compared chains: {len(compared_chain_ids)}"
)

print(
    f"Uncompared chains: {len(not_compared_chains)}"
)

print(
    f"Direct mappings: {direct_mapping_count}"
)

print(
    f"Fallback mappings: {fallback_mapping_count}"
)

print(
    f"Ambiguous mappings: {ambiguous_mapping_count}"
)

print(
    f"Not found mappings: {not_found_mapping_count}"
)

print(
    f"Results directory: {PDB_RESULTS_DIR}"
)

print(
    f"Status: {sequence_validation_status}"
)

print("=" * 70)
print("CELL 13 COMPLETED")
print("=" * 70)
```

Что анализируем: экспериментальная последовательность может быть короче референсной, потому что некоторые участки белка не удалось определить в экспериментальной структуре.

Важно: приведённое выравнивание показывает сходство и длины последовательностей, но для точного подсчёта пропущенных остатков нужно учитывать соответствие цепей и нумерацию остатков. Это сделаем в следующей ячейке.
Что теперь будет в результатах
Показатель	Значение
reference_length	Длина референсной последовательности
experimental_length	Длина экспериментальной последовательности
alignment_score	Оценка качества выравнивания
identity_percent	Процент совпадения аминокислот
reference_coverage_percent	Процент покрытия референсной последовательности
experimental_coverage_percent	Процент покрытия экспериментальной последовательности
aligned_positions	Количество сопоставленных позиций
missing_reference_residues	Количество позиций референса без соответствующего остатка в экспериментальной последовательности
extra_experimental_residues	Количество экспериментальных позиций без соответствия в референсе

Важное замечание
Здесь сравниваются две последовательности:
- референсная — полная последовательность белка из аннотации PDB;
- экспериментальная — последовательность, извлечённая из атомной структуры.
Если часть белка не видна в экспериментальной структуре, она может отсутствовать в экспериментальной последовательности. Это не обязательно означает, что белок был повреждён: такие участки могут быть просто не разрешены экспериментальным методом.

Что добавилось
Показатель	Назначение
Пропущенные остатки	Выявление неразрешённых участков белка
Пропущенные атомы	Выявление неполного атомного состава остатков
Альтернативные положения	Определение атомов с несколькими координатными вариантами
Occupancy	Оценка заполненности атомных позиций
Распределение occupancy	Визуальная оценка заполненности
Статистика по цепям	Сравнение качества представления разных цепей
Общая сводка	Подготовка результатов для итогового отчёта

Важные научные уточнения
1. Пропущенные остатки и пропущенные атомы — не одно и то же.

Пропущенный остаток означает, что целый аминокислотный остаток не представлен в координатной модели. Пропущенные атомы означают, что остаток присутствует, но часть его атомов не разрешена.

2. Occupancy меньше 1 не обязательно означает плохое качество структуры.

Это может отражать частичную заселённость позиции или несколько конформаций молекулы.

3. Альтернативные положения нельзя считать независимыми атомами.

Они представляют несколько возможных координатных вариантов одного атома или группы атомов.

4. Количество пропущенных остатков не следует автоматически делить на количество наблюдаемых остатков.

Для корректного процента полноты нужен полный референсный состав каждой цепи и корректное сопоставление идентификаторов остатков. Результаты ячейки 13 могут стать основой для такого расчёта, но простого вычитания количеств недостаточно.

5. В этой ячейке df_atoms должен содержать атомы белка.

Если таблица содержит воду и другие небелковые компоненты, статистика occupancy и altloc будет относиться ко всему содержимому таблицы, а не только к белку.

```python
# ============================================================
# CELL 14. АНАЛИЗ ПОЛНОТЫ СТРУКТУРЫ И АЛЬТЕРНАТИВНЫХ ПОЛОЖЕНИЙ
# ============================================================
# Назначение:
# 1. Извлечение записей о неразрешённых остатках и атомах из mmCIF.
# 2. Анализ всех белковых атомных конформеров во всех моделях.
# 3. Разделение уникальных атомных позиций и altloc-конформеров.
# 4. Анализ occupancy: нормальные, низкие, отсутствующие,
#    некорректные значения.
# 5. Анализ качества структуры по цепям.
# 6. Сверка первой модели с результатами CELL 6 и CELL 7.
#
# ВАЖНО:
# - CELL 6 и CELL 7 содержат данные первой модели.
# - CELL 14 анализирует все модели, если их несколько.
# - Сравнение с CELL 7 проводится только для первой модели.
# - altloc-конформеры не считаются отдельными уникальными
#   атомными позициями.
# - Отсутствие записей в mmCIF-категориях не доказывает,
#   что все атомы и остатки структуры разрешены.
# - Переменные CELL 10 и CELL 12 не перезаписываются.
# ============================================================

import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from Bio.PDB.Polypeptide import is_aa
from Bio.PDB.MMCIF2Dict import MMCIF2Dict


print("=" * 70)
print("CELL 14. STRUCTURE COMPLETENESS AND ALTLOC ANALYSIS")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка входных данных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "CIF_PATH",
    "structure",
    "df_atoms",
    "RESULTS_DIR"
]

for variable in required_variables:

    if variable not in globals():

        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()

if not PDB_ID:
    raise ValueError("PDB_ID не может быть пустым.")


# CELL 2 уже задаёт RESULTS_DIR как папку конкретного PDB.
# Если PDB_RESULTS_DIR существует, используем его.

if "PDB_RESULTS_DIR" in globals():

    CELL14_RESULTS_DIR = Path(PDB_RESULTS_DIR)

else:

    CELL14_RESULTS_DIR = Path(RESULTS_DIR)


CELL14_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


if not isinstance(df_atoms, pd.DataFrame):

    raise TypeError(
        "df_atoms должен быть pandas DataFrame."
    )


if df_atoms.empty:

    raise ValueError(
        "df_atoms не содержит белковых атомов."
    )


required_atom_columns = {
    "chain_id",
    "atom_name",
    "occupancy",
    "altloc"
}

missing_columns = (
    required_atom_columns - set(df_atoms.columns)
)

if missing_columns:

    raise KeyError(
        "В df_atoms отсутствуют столбцы: "
        f"{sorted(missing_columns)}"
    )


if not Path(CIF_PATH).is_file():

    raise FileNotFoundError(
        f"Файл mmCIF не найден: {CIF_PATH}"
    )


# ------------------------------------------------------------
# 2. Получение словаря mmCIF
# ------------------------------------------------------------

# Используем уже созданный cif_dict, если он доступен.
# В противном случае читаем CIF самостоятельно.

if (
    "cif_dict" in globals()
    and isinstance(cif_dict, dict)
):

    CELL14_CIF_DICT = cif_dict

else:

    CELL14_CIF_DICT = MMCIF2Dict(
        str(CIF_PATH)
    )


def cell14_get_cif_values(key):
    """Возвращает значение поля mmCIF в виде списка."""

    value = CELL14_CIF_DICT.get(key, [])

    if value is None:
        return []

    if isinstance(value, (list, tuple, np.ndarray)):
        return list(value)

    return [value]


def cell14_clean_cif_value(value):
    """Преобразует служебные обозначения mmCIF в None."""

    if value is None:
        return None

    value = str(value).strip()

    if value in {"", ".", "?"}:
        return None

    return value


def cell14_category_exists(prefix):
    """Проверяет наличие хотя бы одного поля категории."""

    return any(
        key.startswith(prefix)
        for key in CELL14_CIF_DICT.keys()
    )


def cell14_build_category_dataframe(prefix, field_mapping):
    """
    Формирует DataFrame из категории mmCIF.

    Для каждого столбца проверяются альтернативные имена
    полей auth/label. Выбор значения выполняется отдельно
    для каждой записи, а не сразу для всего столбца.
    """

    available_fields = {}

    for column_name, possible_keys in field_mapping.items():

        selected_key = next(
            (
                key
                for key in possible_keys
                if key in CELL14_CIF_DICT
            ),
            None
        )

        if selected_key is not None:

            available_fields[column_name] = (
                selected_key,
                cell14_get_cif_values(selected_key)
            )

    if not available_fields:

        return (
            pd.DataFrame(columns=list(field_mapping.keys())),
            0
        )

    lengths = [
        len(values)
        for _, values in available_fields.values()
    ]

    max_length = max(lengths, default=0)

    # Поля одной категории обычно имеют одинаковую длину.
    # Несовпадение фиксируем, но не теряем доступные записи.

    if len(set(lengths)) > 1:

        print(
            f"WARNING: разные длины полей категории {prefix}: "
            f"{sorted(set(lengths))}"
        )

    records = []

    for i in range(max_length):

        record = {}

        for column_name in field_mapping:

            if column_name not in available_fields:

                record[column_name] = None
                continue

            _, values = available_fields[column_name]

            record[column_name] = (
                cell14_clean_cif_value(values[i])
                if i < len(values)
                else None
            )

        if any(
            value is not None
            for value in record.values()
        ):

            records.append(record)

    result = pd.DataFrame(
        records,
        columns=list(field_mapping.keys())
    )

    return result, len(result)


# ------------------------------------------------------------
# 3. Извлечение неразрешённых остатков
# ------------------------------------------------------------

RESIDUE_CATEGORY = (
    "_pdbx_unobs_or_zero_occ_residues."
)

cell14_residue_category_present = (
    cell14_category_exists(RESIDUE_CATEGORY)
)


residue_field_mapping = {

    "record_id": [
        "_pdbx_unobs_or_zero_occ_residues.id"
    ],

    "chain_id": [
        "_pdbx_unobs_or_zero_occ_residues.auth_asym_id",
        "_pdbx_unobs_or_zero_occ_residues.label_asym_id"
    ],

    "residue_name": [
        "_pdbx_unobs_or_zero_occ_residues.auth_comp_id",
        "_pdbx_unobs_or_zero_occ_residues.label_comp_id"
    ],

    "sequence_id": [
        "_pdbx_unobs_or_zero_occ_residues.auth_seq_id",
        "_pdbx_unobs_or_zero_occ_residues.label_seq_id"
    ],

    "insertion_code": [
        "_pdbx_unobs_or_zero_occ_residues.PDB_ins_code"
    ],

    "model_number": [
        "_pdbx_unobs_or_zero_occ_residues.PDB_model_num"
    ]

}


df_cell14_missing_residues, cell14_residue_record_count = (
    cell14_build_category_dataframe(
        RESIDUE_CATEGORY,
        residue_field_mapping
    )
)


print("\n" + "-" * 70)
print("UNOBSERVED OR ZERO-OCCUPANCY RESIDUES")
print("-" * 70)

print(
    "Category present:",
    cell14_residue_category_present
)

print(
    "Extracted records:",
    cell14_residue_record_count
)

if not df_cell14_missing_residues.empty:

    display(
        df_cell14_missing_residues.head(20)
    )

else:

    print("No records extracted.")

if not cell14_residue_category_present:

    print(
        "The category is absent. "
        "This does not prove that all residues are observed."
    )


# ------------------------------------------------------------
# 4. Извлечение неразрешённых атомов
# ------------------------------------------------------------

ATOM_CATEGORY = (
    "_pdbx_unobs_or_zero_occ_atoms."
)

cell14_atom_category_present = (
    cell14_category_exists(ATOM_CATEGORY)
)


atom_field_mapping = {

    "record_id": [
        "_pdbx_unobs_or_zero_occ_atoms.id"
    ],

    "chain_id": [
        "_pdbx_unobs_or_zero_occ_atoms.auth_asym_id",
        "_pdbx_unobs_or_zero_occ_atoms.label_asym_id"
    ],

    "residue_name": [
        "_pdbx_unobs_or_zero_occ_atoms.auth_comp_id",
        "_pdbx_unobs_or_zero_occ_atoms.label_comp_id"
    ],

    "sequence_id": [
        "_pdbx_unobs_or_zero_occ_atoms.auth_seq_id",
        "_pdbx_unobs_or_zero_occ_atoms.label_seq_id"
    ],

    "atom_name": [
        "_pdbx_unobs_or_zero_occ_atoms.auth_atom_id",
        "_pdbx_unobs_or_zero_occ_atoms.label_atom_id"
    ],

    "insertion_code": [
        "_pdbx_unobs_or_zero_occ_atoms.PDB_ins_code"
    ],

    "model_number": [
        "_pdbx_unobs_or_zero_occ_atoms.PDB_model_num"
    ]

}


df_cell14_missing_atoms, cell14_atom_record_count = (
    cell14_build_category_dataframe(
        ATOM_CATEGORY,
        atom_field_mapping
    )
)


print("\n" + "-" * 70)
print("UNOBSERVED OR ZERO-OCCUPANCY ATOMS")
print("-" * 70)

print(
    "Category present:",
    cell14_atom_category_present
)

print(
    "Extracted records:",
    cell14_atom_record_count
)

if not df_cell14_missing_atoms.empty:

    display(
        df_cell14_missing_atoms.head(20)
    )

else:

    print("No records extracted.")

if not cell14_atom_category_present:

    print(
        "The category is absent. "
        "No conclusion about missing atoms can be made."
    )


# ------------------------------------------------------------
# 5. Извлечение всех белковых атомных конформеров
# ------------------------------------------------------------

models_cell14 = list(
    structure.get_models()
)

if not models_cell14:

    raise ValueError(
        "В структуре отсутствуют модели."
    )


all_atom_records = []

model_id_mapping = {}


for model_index, model in enumerate(
    models_cell14,
    start=1
):

    model_id_mapping[model_index] = model.id

    for chain in model:

        chain_id = (
            ""
            if chain.id is None
            else str(chain.id).strip()
        )

        for residue in chain:

            if not is_aa(
                residue,
                standard=False
            ):
                continue

            hetero_flag = str(
                residue.id[0]
            ).strip()

            residue_id = int(
                residue.id[1]
            )

            insertion_code = str(
                residue.id[2]
            ).strip()

            residue_name = str(
                residue.resname
            ).strip()

            # Ключ включает hetero_flag, чтобы исключить
            # возможное совпадение разных типов остатков.

            residue_key = (
                f"{chain_id}:"
                f"{hetero_flag}:"
                f"{residue_id}:"
                f"{insertion_code}"
            )

            # get_unpacked_list() раскрывает альтернативные
            # положения DisorderedAtom.

            for atom in residue.get_unpacked_list():

                altloc = str(
                    atom.get_altloc()
                ).strip()

                if altloc in {"", ".", "?"}:
                    altloc = None

                raw_occupancy = atom.get_occupancy()

                try:

                    occupancy = (
                        float(raw_occupancy)
                        if raw_occupancy is not None
                        else np.nan
                    )

                except (TypeError, ValueError):

                    occupancy = np.nan

                element = str(
                    atom.element
                ).strip().upper()

                coord = np.asarray(
                    atom.coord,
                    dtype=float
                )

                all_atom_records.append({

                    "model_index": model_index,

                    "model_id": str(model.id),

                    "chain_id": chain_id,

                    "residue_key": residue_key,

                    "hetero_flag": hetero_flag,

                    "residue_id": residue_id,

                    "insertion_code": insertion_code,

                    "residue_name": residue_name,

                    "atom_name": atom.get_name().strip(),

                    "element": element,

                    "x": float(coord[0]),

                    "y": float(coord[1]),

                    "z": float(coord[2]),

                    "occupancy": occupancy,

                    "bfactor": atom.get_bfactor(),

                    "altloc": altloc

                })


df_cell14_atom_quality = pd.DataFrame(
    all_atom_records
)


if df_cell14_atom_quality.empty:

    raise ValueError(
        "Не удалось извлечь белковые атомы из структуры."
    )


# Проверка координат.

coordinate_columns = ["x", "y", "z"]

df_cell14_atom_quality["coordinates_valid"] = (
    np.isfinite(
        df_cell14_atom_quality[coordinate_columns]
    ).all(axis=1)
)


invalid_coordinate_count = int(
    (~df_cell14_atom_quality["coordinates_valid"]).sum()
)


print("\n" + "-" * 70)
print("EXTRACTED PROTEIN ATOMS")
print("-" * 70)

print(
    "Number of models:",
    len(models_cell14)
)

print(
    "Atom conformer records:",
    len(df_cell14_atom_quality)
)

print(
    "Records with invalid coordinates:",
    invalid_coordinate_count
)


# ------------------------------------------------------------
# 6. Уникальные атомные позиции и альтернативные положения
# ------------------------------------------------------------

# Атомная позиция определяется моделью, цепью, остатком
# и именем атома. Разные altloc одной позиции не создают
# новые уникальные атомные позиции.

atom_site_columns = [

    "model_index",

    "chain_id",

    "residue_key",

    "atom_name"

]


df_cell14_unique_atom_sites = (
    df_cell14_atom_quality[
        atom_site_columns
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)


cell14_unique_atom_site_count = len(
    df_cell14_unique_atom_sites
)


# Все записи, у которых задан altloc.

df_cell14_altloc_atoms = (
    df_cell14_atom_quality[
        df_cell14_atom_quality["altloc"].notna()
    ].copy()
)


cell14_altloc_atom_record_count = len(
    df_cell14_altloc_atoms
)


# Число уникальных позиций, для которых есть altloc.

df_cell14_altloc_sites = (
    df_cell14_altloc_atoms[
        atom_site_columns
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)


cell14_altloc_unique_site_count = len(
    df_cell14_altloc_sites
)


# Дополнительные записи конформеров сверх одной записи
# на каждую уникальную атомную позицию.

cell14_extra_conformer_records = (
    len(df_cell14_atom_quality)
    - cell14_unique_atom_site_count
)


print("\n" + "-" * 70)
print("ALTERNATIVE ATOM POSITIONS")
print("-" * 70)

print(
    "All atom conformer records:",
    len(df_cell14_atom_quality)
)

print(
    "Unique atom sites:",
    cell14_unique_atom_site_count
)

print(
    "Additional conformer records:",
    cell14_extra_conformer_records
)

print(
    "Atom records with altloc:",
    cell14_altloc_atom_record_count
)

print(
    "Unique atom sites with altloc:",
    cell14_altloc_unique_site_count
)


if not df_cell14_altloc_atoms.empty:

    print("\nAltloc label distribution:")

    display(
        df_cell14_altloc_atoms["altloc"]
        .value_counts()
        .rename_axis("altloc")
        .reset_index(name="atom_record_count")
    )

    print("\nExamples of alternative atom records:")

    display(
        df_cell14_altloc_atoms.head(30)
    )

else:

    print(
        "Alternative atom positions were not found."
    )


# ------------------------------------------------------------
# 7. Анализ occupancy
# ------------------------------------------------------------

# Порог можно менять в зависимости от требований анализа.
# Значение 0.99 позволяет отдельно учитывать значения
# ниже полной occupancy, не называя их автоматически
# доказательством плохого качества структуры.

LOW_OCCUPANCY_THRESHOLD = 0.99


df_cell14_occupancy_known = (
    df_cell14_atom_quality[
        df_cell14_atom_quality["occupancy"].notna()
    ].copy()
)


df_cell14_invalid_occupancy = (
    df_cell14_occupancy_known[
        (
            df_cell14_occupancy_known["occupancy"] < 0
        )
        |
        (
            df_cell14_occupancy_known["occupancy"] > 1
        )
    ].copy()
)


df_cell14_occupancy_valid = (
    df_cell14_occupancy_known[
        (
            df_cell14_occupancy_known["occupancy"] >= 0
        )
        &
        (
            df_cell14_occupancy_known["occupancy"] <= 1
        )
    ].copy()
)


df_cell14_low_occupancy = (
    df_cell14_occupancy_valid[
        df_cell14_occupancy_valid["occupancy"]
        < LOW_OCCUPANCY_THRESHOLD
    ].copy()
)


df_cell14_missing_occupancy = (
    df_cell14_atom_quality[
        df_cell14_atom_quality["occupancy"].isna()
    ].copy()
)


print("\n" + "-" * 70)
print("OCCUPANCY ANALYSIS")
print("-" * 70)

print(
    "Atoms with known occupancy:",
    len(df_cell14_occupancy_known)
)

print(
    "Atoms with valid occupancy [0, 1]:",
    len(df_cell14_occupancy_valid)
)

print(
    f"Atoms with occupancy < {LOW_OCCUPANCY_THRESHOLD}:",
    len(df_cell14_low_occupancy)
)

print(
    "Atoms with missing occupancy:",
    len(df_cell14_missing_occupancy)
)

print(
    "Atoms with invalid occupancy:",
    len(df_cell14_invalid_occupancy)
)


if not df_cell14_occupancy_valid.empty:

    print(
        "\nMean occupancy:",
        round(
            df_cell14_occupancy_valid["occupancy"].mean(),
            4
        )
    )

    print(
        "Median occupancy:",
        round(
            df_cell14_occupancy_valid["occupancy"].median(),
            4
        )
    )

    print(
        "Minimum occupancy:",
        round(
            df_cell14_occupancy_valid["occupancy"].min(),
            4
        )
    )

    print(
        "Maximum occupancy:",
        round(
            df_cell14_occupancy_valid["occupancy"].max(),
            4
        )
    )

    print("\nExamples of low occupancy atoms:")

    display(
        df_cell14_low_occupancy.head(30)
    )


# ------------------------------------------------------------
# 8. Распределение occupancy
# ------------------------------------------------------------

CELL14_OCCUPANCY_PLOT_PATH = (
    CELL14_RESULTS_DIR
    / f"{PDB_ID}_occupancy_distribution.png"
)


if not df_cell14_occupancy_valid.empty:

    fig, ax = plt.subplots(
        figsize=(9, 5)
    )

    ax.hist(
        df_cell14_occupancy_valid["occupancy"],
        bins=20,
        edgecolor="black"
    )

    ax.set_xlabel("Occupancy")
    ax.set_ylabel("Number of atom conformers")

    ax.set_title(
        f"Occupancy distribution: {PDB_ID}"
    )

    ax.grid(
        axis="y",
        alpha=0.3
    )

    plt.tight_layout()

    plt.savefig(
        CELL14_OCCUPANCY_PLOT_PATH,
        dpi=150,
        bbox_inches="tight"
    )

    plt.show()

    plt.close(fig)

else:

    print(
        "Occupancy plot was not created: "
        "no valid occupancy values."
    )


# ------------------------------------------------------------
# 9. Сводка качества по цепям
# ------------------------------------------------------------

chain_summary_records = []


for (model_index, chain_id), group in (
    df_cell14_atom_quality.groupby(
        ["model_index", "chain_id"],
        dropna=False
    )
):

    chain_altloc = group[
        group["altloc"].notna()
    ]

    chain_valid_occupancy = group[
        group["occupancy"].notna()
        &
        (group["occupancy"] >= 0)
        &
        (group["occupancy"] <= 1)
    ]

    chain_low_occupancy = chain_valid_occupancy[
        chain_valid_occupancy["occupancy"]
        < LOW_OCCUPANCY_THRESHOLD
    ]

    chain_missing_occupancy = group[
        group["occupancy"].isna()
    ]

    chain_invalid_occupancy = group[
        group["occupancy"].notna()
        &
        (
            (group["occupancy"] < 0)
            |
            (group["occupancy"] > 1)
        )
    ]

    chain_summary_records.append({

        "model_index": int(model_index),

        "model_id": str(
            model_id_mapping[int(model_index)]
        ),

        "chain_id": chain_id,

        "chain_id_display": (
            chain_id if chain_id else "<blank>"
        ),

        "atom_conformer_count": len(group),

        "unique_residue_count": (
            group["residue_key"].nunique()
        ),

        "unique_atom_site_count": (
            group[atom_site_columns]
            .drop_duplicates()
            .shape[0]
        ),

        "altloc_atom_record_count": len(
            chain_altloc
        ),

        "altloc_unique_site_count": (
            chain_altloc[atom_site_columns]
            .drop_duplicates()
            .shape[0]
        ),

        "low_occupancy_count": len(
            chain_low_occupancy
        ),

        "missing_occupancy_count": len(
            chain_missing_occupancy
        ),

        "invalid_occupancy_count": len(
            chain_invalid_occupancy
        ),

        "invalid_coordinate_count": int(
            (~group["coordinates_valid"]).sum()
        )

    })


df_cell14_quality_by_chain = pd.DataFrame(
    chain_summary_records
)


print("\n" + "-" * 70)
print("STRUCTURE QUALITY BY CHAIN")
print("-" * 70)

display(
    df_cell14_quality_by_chain
)


# ------------------------------------------------------------
# 10. Сверка с CELL 6 и CELL 7
# ------------------------------------------------------------

# CELL 6 и CELL 7 используют первую модель.
# Поэтому сравнение выполняется только с моделью № 1.

first_model_index = 1

df_cell14_first_model = (
    df_cell14_atom_quality[
        df_cell14_atom_quality["model_index"]
        == first_model_index
    ].copy()
)


first_model_atom_count = len(
    df_cell14_first_model
)


first_model_unique_site_count = (
    df_cell14_first_model[
        atom_site_columns
    ]
    .drop_duplicates()
    .shape[0]
)


cell7_atom_count = len(df_atoms)


cell7_atom_count_matches = (
    cell7_atom_count == first_model_atom_count
)


cell7_unique_site_count_matches = (
    cell7_atom_count == first_model_unique_site_count
)


# Сравнение числа остатков с CELL 6, если таблица доступна.

cell6_residue_count = None
cell6_residue_count_matches = None

if "df_residues" in globals():

    if isinstance(df_residues, pd.DataFrame):

        cell6_residue_count = len(df_residues)

        first_model_residue_count = (
            df_cell14_first_model[
                [
                    "chain_id",
                    "residue_key"
                ]
            ]
            .drop_duplicates()
            .shape[0]
        )

        cell6_residue_count_matches = (
            cell6_residue_count
            == first_model_residue_count
        )

else:

    first_model_residue_count = (
        df_cell14_first_model[
            [
                "chain_id",
                "residue_key"
            ]
        ]
        .drop_duplicates()
        .shape[0]
    )


print("\n" + "-" * 70)
print("CELL 6 / CELL 7 CONSISTENCY")
print("-" * 70)

print(
    "First model atom conformer records:",
    first_model_atom_count
)

print(
    "First model unique atom sites:",
    first_model_unique_site_count
)

print(
    "CELL 7 atom records:",
    cell7_atom_count
)

print(
    "CELL 7 matches first-model atom records:",
    cell7_atom_count_matches
)

print(
    "CELL 7 matches first-model unique atom sites:",
    cell7_unique_site_count_matches
)

if cell6_residue_count is not None:

    print(
        "CELL 6 residue count:",
        cell6_residue_count
    )

    print(
        "First-model residue count:",
        first_model_residue_count
    )

    print(
        "CELL 6 matches first-model residue count:",
        cell6_residue_count_matches
    )


# Сверка является диагностической.
# Если число атомов не совпало, это не всегда означает
# ошибку структуры: нужно проверить политику altloc
# в CELL 7 и обработку DisorderedResidue.


# ------------------------------------------------------------
# 11. Общая сводка полноты структуры
# ------------------------------------------------------------

total_atom_conformer_records = len(
    df_cell14_atom_quality
)


total_unique_atom_sites = (
    cell14_unique_atom_site_count
)


total_residues_all_models = (
    df_cell14_atom_quality[
        [
            "model_index",
            "chain_id",
            "residue_key"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)


first_model_residue_count = (
    df_cell14_first_model[
        [
            "chain_id",
            "residue_key"
        ]
    ]
    .drop_duplicates()
    .shape[0]
)


cell14_quality_parameters = [

    "Number of models",

    "Observed protein atom conformer records (all models)",

    "Unique protein atom sites (all models)",

    "Additional conformer records",

    "Observed protein residues (all models)",

    "Observed protein residues (first model)",

    "Unobserved / zero-occupancy residue records",

    "Unobserved / zero-occupancy atom records",

    "Atom records with alternative positions",

    "Unique atom sites with alternative positions",

    "Atom records with low occupancy",

    "Atom records with missing occupancy",

    "Atom records with invalid occupancy",

    "Atom records with invalid coordinates",

    "Reference residue category present",

    "Reference atom category present",

    "CELL 7 atom count",

    "CELL 7 matches first-model atom records",

    "CELL 6 residue count",

    "CELL 6 matches first-model residue count"

]


cell14_quality_values = [

    len(models_cell14),

    total_atom_conformer_records,

    total_unique_atom_sites,

    cell14_extra_conformer_records,

    total_residues_all_models,

    first_model_residue_count,

    cell14_residue_record_count,

    cell14_atom_record_count,

    cell14_altloc_atom_record_count,

    cell14_altloc_unique_site_count,

    len(df_cell14_low_occupancy),

    len(df_cell14_missing_occupancy),

    len(df_cell14_invalid_occupancy),

    invalid_coordinate_count,

    cell14_residue_category_present,

    cell14_atom_category_present,

    cell7_atom_count,

    cell7_atom_count_matches,

    cell6_residue_count,

    cell6_residue_count_matches

]


df_cell14_quality_summary = pd.DataFrame({

    "Parameter": cell14_quality_parameters,

    "Value": cell14_quality_values

})


print("\n" + "-" * 70)
print("OVERALL STRUCTURE QUALITY SUMMARY")
print("-" * 70)

display(
    df_cell14_quality_summary
)


# ------------------------------------------------------------
# 12. Пересечение altloc и низкой occupancy
# ------------------------------------------------------------

# Один конформер может одновременно иметь altloc
# и occupancy ниже установленного порога.
# Поэтому эти показатели нельзя суммировать.

cell14_altloc_low_occupancy_count = int(

    (
        df_cell14_atom_quality["altloc"].notna()
        &
        df_cell14_atom_quality["occupancy"].notna()
        &
        (df_cell14_atom_quality["occupancy"] >= 0)
        &
        (df_cell14_atom_quality["occupancy"] < 1)
        &
        (
            df_cell14_atom_quality["occupancy"]
            < LOW_OCCUPANCY_THRESHOLD
        )
    ).sum()

)


print("\n" + "-" * 70)
print("OVERLAP CHECK")
print("-" * 70)

print(
    "Altloc atom records:",
    cell14_altloc_atom_record_count
)

print(
    "Low occupancy atom records:",
    len(df_cell14_low_occupancy)
)

print(
    "Both altloc and low occupancy:",
    cell14_altloc_low_occupancy_count
)

print(
    "These categories overlap and must not be added "
    "as independent problem counts."
)


# ------------------------------------------------------------
# 13. Подготовка JSON-сводки
# ------------------------------------------------------------

def cell14_json_safe(value):
    """Преобразует значения NumPy/pandas в JSON-совместимый вид."""

    if value is None:
        return None

    if isinstance(value, dict):

        return {
            str(key): cell14_json_safe(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple)):

        return [
            cell14_json_safe(item)
            for item in value
        ]

    if isinstance(value, np.ndarray):

        return [
            cell14_json_safe(item)
            for item in value.tolist()
        ]

    if isinstance(value, (np.integer,)):

        return int(value)

    if isinstance(value, (np.floating, float)):

        value = float(value)

        return (
            value
            if np.isfinite(value)
            else None
        )

    if isinstance(value, (np.bool_, bool)):

        return bool(value)

    if pd.isna(value):

        return None

    if isinstance(value, (str, int)):

        return value

    return str(value)


cell14_quality_information = {

    "PDB_ID": PDB_ID,

    "analysis_scope": (
        "All protein models from the original mmCIF structure. "
        "CELL 6 and CELL 7 consistency checks use the first model."
    ),

    "number_of_models": len(models_cell14),

    "first_model_id": str(
        models_cell14[0].id
    ),

    "observed_protein_atom_conformer_records": (
        total_atom_conformer_records
    ),

    "unique_protein_atom_sites": (
        total_unique_atom_sites
    ),

    "additional_conformer_records": (
        cell14_extra_conformer_records
    ),

    "observed_protein_residues_all_models": (
        total_residues_all_models
    ),

    "observed_protein_residues_first_model": (
        first_model_residue_count
    ),

    "unobserved_residue_category_present": (
        cell14_residue_category_present
    ),

    "unobserved_residue_records": (
        cell14_residue_record_count
    ),

    "unobserved_atom_category_present": (
        cell14_atom_category_present
    ),

    "unobserved_atom_records": (
        cell14_atom_record_count
    ),

    "altloc_atom_record_count": (
        cell14_altloc_atom_record_count
    ),

    "altloc_unique_site_count": (
        cell14_altloc_unique_site_count
    ),

    "low_occupancy_threshold": (
        LOW_OCCUPANCY_THRESHOLD
    ),

    "low_occupancy_atom_record_count": (
        len(df_cell14_low_occupancy)
    ),

    "missing_occupancy_atom_record_count": (
        len(df_cell14_missing_occupancy)
    ),

    "invalid_occupancy_atom_record_count": (
        len(df_cell14_invalid_occupancy)
    ),

    "invalid_coordinate_record_count": (
        invalid_coordinate_count
    ),

    "altloc_and_low_occupancy_overlap": (
        cell14_altloc_low_occupancy_count
    ),

    "mean_occupancy": (
        float(
            df_cell14_occupancy_valid["occupancy"].mean()
        )
        if not df_cell14_occupancy_valid.empty
        else None
    ),

    "median_occupancy": (
        float(
            df_cell14_occupancy_valid["occupancy"].median()
        )
        if not df_cell14_occupancy_valid.empty
        else None
    ),

    "minimum_occupancy": (
        float(
            df_cell14_occupancy_valid["occupancy"].min()
        )
        if not df_cell14_occupancy_valid.empty
        else None
    ),

    "maximum_occupancy": (
        float(
            df_cell14_occupancy_valid["occupancy"].max()
        )
        if not df_cell14_occupancy_valid.empty
        else None
    ),

    "cell7_comparison": {

        "cell7_atom_count": cell7_atom_count,

        "first_model_atom_conformer_count": (
            first_model_atom_count
        ),

        "first_model_unique_atom_site_count": (
            first_model_unique_site_count
        ),

        "matches_first_model_atom_records": (
            cell7_atom_count_matches
        ),

        "matches_first_model_unique_atom_sites": (
            cell7_unique_site_count_matches
        )

    },

    "cell6_comparison": {

        "cell6_residue_count": cell6_residue_count,

        "first_model_residue_count": (
            first_model_residue_count
        ),

        "matches_first_model_residue_count": (
            cell6_residue_count_matches
        )

    },

    "interpretation_note": (
        "Altloc conformers and unique atom sites are reported "
        "separately. Low occupancy and altloc categories overlap. "
        "Missing mmCIF categories do not prove that the structure "
        "is complete. Unobserved category record counts are not "
        "necessarily equal to the number of all missing atoms "
        "or residues."
    )

}


# ------------------------------------------------------------
# 14. Сохранение результатов
# ------------------------------------------------------------

CELL14_OUTPUTS = {}


def cell14_save_dataframe(name, dataframe):

    path = (
        CELL14_RESULTS_DIR
        / f"{PDB_ID}_{name}.csv"
    )

    dataframe.to_csv(
        path,
        index=False,
        encoding="utf-8-sig"
    )

    CELL14_OUTPUTS[name] = path

    return path


CELL14_QUALITY_SUMMARY_CSV = (
    cell14_save_dataframe(
        "structure_quality_summary",
        df_cell14_quality_summary
    )
)


CELL14_QUALITY_BY_CHAIN_CSV = (
    cell14_save_dataframe(
        "structure_quality_by_chain",
        df_cell14_quality_by_chain
    )
)


CELL14_MISSING_RESIDUES_CSV = (
    cell14_save_dataframe(
        "unobserved_residues",
        df_cell14_missing_residues
    )
)


CELL14_MISSING_ATOMS_CSV = (
    cell14_save_dataframe(
        "unobserved_atoms",
        df_cell14_missing_atoms
    )
)


CELL14_ALTLOC_ATOMS_CSV = (
    cell14_save_dataframe(
        "altloc_atoms",
        df_cell14_altloc_atoms
    )
)


CELL14_LOW_OCCUPANCY_CSV = (
    cell14_save_dataframe(
        "low_occupancy_atoms",
        df_cell14_low_occupancy
    )
)


CELL14_ATOM_QUALITY_CSV = (
    cell14_save_dataframe(
        "atom_quality",
        df_cell14_atom_quality
    )
)


CELL14_INVALID_OCCUPANCY_CSV = (
    cell14_save_dataframe(
        "invalid_occupancy_atoms",
        df_cell14_invalid_occupancy
    )
)


CELL14_MISSING_OCCUPANCY_CSV = (
    cell14_save_dataframe(
        "missing_occupancy_atoms",
        df_cell14_missing_occupancy
    )
)


CELL14_JSON_PATH = (
    CELL14_RESULTS_DIR
    / f"{PDB_ID}_structure_quality.json"
)


with open(
    CELL14_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        cell14_json_safe(cell14_quality_information),
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


CELL14_OUTPUTS["structure_quality_json"] = (
    CELL14_JSON_PATH
)


if not df_cell14_occupancy_valid.empty:

    CELL14_OUTPUTS["occupancy_distribution"] = (
        CELL14_OCCUPANCY_PLOT_PATH
    )


# ------------------------------------------------------------
# 15. Переменные для совместимости с CELL 12
# ------------------------------------------------------------

# Не перезаписываем:
# - quality_summary из CELL 10;
# - unobserved_residue_count из CELL 10;
# - unobserved_atom_count из CELL 10;
# - residues_category_status из CELL 10;
# - atoms_category_status из CELL 10.
#
# Для CELL 14 используются собственные имена.

cell14_unobserved_residue_count = (
    cell14_residue_record_count
)

cell14_unobserved_atom_count = (
    cell14_atom_record_count
)

cell14_residues_category_status = (
    "Present"
    if cell14_residue_category_present
    else "Not present in mmCIF"
)

cell14_atoms_category_status = (
    "Present"
    if cell14_atom_category_present
    else "Not present in mmCIF"
)


# ------------------------------------------------------------
# 16. Проверка сохранённых файлов
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("SAVED FILES")
print("-" * 70)


cell14_output_check_records = []

for name, path in CELL14_OUTPUTS.items():

    exists = (
        path.is_file()
        and path.stat().st_size > 0
    )

    cell14_output_check_records.append({

        "file_type": name,

        "file_name": path.name,

        "exists": bool(exists),

        "file_size_bytes": (
            path.stat().st_size
            if path.is_file()
            else 0
        )

    })

    print(
        f"{path.name}: "
        f"{'OK' if exists else 'ERROR'}"
    )


df_cell14_output_checks = pd.DataFrame(
    cell14_output_check_records
)


cell14_all_outputs_saved = bool(
    df_cell14_output_checks["exists"].all()
)


if not cell14_all_outputs_saved:

    raise RuntimeError(
        "Не все результаты CELL 14 были сохранены."
    )


# ------------------------------------------------------------
# 17. Итог
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CELL 14 SUMMARY")
print("=" * 70)

print(
    f"PDB ID: {PDB_ID}"
)

print(
    f"Models analyzed: {len(models_cell14)}"
)

print(
    f"Protein atom conformer records: "
    f"{total_atom_conformer_records}"
)

print(
    f"Unique protein atom sites: "
    f"{total_unique_atom_sites}"
)

print(
    f"Additional conformer records: "
    f"{cell14_extra_conformer_records}"
)

print(
    f"Observed protein residues: "
    f"{total_residues_all_models}"
)

print(
    f"Altloc atom records: "
    f"{cell14_altloc_atom_record_count}"
)

print(
    f"Low occupancy atom records: "
    f"{len(df_cell14_low_occupancy)}"
)

print(
    f"Missing occupancy atom records: "
    f"{len(df_cell14_missing_occupancy)}"
)

print(
    f"Invalid occupancy atom records: "
    f"{len(df_cell14_invalid_occupancy)}"
)

print(
    f"Unobserved residue records: "
    f"{cell14_residue_record_count}"
)

print(
    f"Unobserved atom records: "
    f"{cell14_atom_record_count}"
)

print(
    f"Invalid coordinate records: "
    f"{invalid_coordinate_count}"
)

print(
    f"Saved output files: "
    f"{len(CELL14_OUTPUTS)}"
)

print(
    f"Results directory: "
    f"{CELL14_RESULTS_DIR}"
)

print("\nСтатус: OK")

print("=" * 70)
print("CELL 14 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# CELL 15. SEPARATE PROTEIN CHAINS AND HETERO COMPONENTS
# ============================================================
# Назначение:
# 1. Разделение белковых атомов и остатков.
# 2. Отдельное сохранение небелковых компонентов и воды.
# 3. Формирование сводки небелковых компонентов.
# 4. Сохранение отдельных файлов белковых цепей.
# 5. Проверка согласованности с CELL 6–8.
#
# ВАЖНО:
# - Используется первая модель структуры.
# - CELL 6 и CELL 7 уже содержат белковые данные.
# - Альтернативные положения атомов анализируются в CELL 14.
# - Небелковые компоненты не считаются автоматически
#   лекарственными лигандами.
# - Для цепей с идентификаторами длиннее одного символа
#   используется mmCIF, поскольку PDB ограничивает
#   длину идентификатора цепи одним символом.
# ============================================================

import json
import re
import numpy as np
import pandas as pd

from pathlib import Path

from Bio.PDB import PDBIO, Select
from Bio.PDB.MMCIFIO import MMCIFIO
from Bio.PDB.Polypeptide import is_aa


print("=" * 70)
print("CELL 15. SEPARATE PROTEIN CHAINS AND HETERO COMPONENTS")
print("=" * 70)


# ------------------------------------------------------------
# 1. Проверка входных данных
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "structure",
    "PDB_RESULTS_DIR",
    "df_atoms",
    "df_residues",
    "df_hetero",
    "df_water"
]

for variable in required_variables:

    if variable not in globals():

        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()

if not re.fullmatch(r"[A-Z0-9]{4}", PDB_ID):

    raise ValueError(
        f"Некорректный PDB ID: {PDB_ID}"
    )


PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

if PDB_RESULTS_DIR.name.upper() != PDB_ID:

    raise ValueError(
        "Папка результатов не соответствует текущему PDB ID.\n"
        f"PDB_ID: {PDB_ID}\n"
        f"Папка: {PDB_RESULTS_DIR}"
    )


PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


input_tables = {

    "df_atoms": df_atoms,

    "df_residues": df_residues,

    "df_hetero": df_hetero,

    "df_water": df_water

}


for table_name, table in input_tables.items():

    if not isinstance(table, pd.DataFrame):

        raise TypeError(
            f"{table_name} должен быть pandas DataFrame."
        )


if df_atoms.empty:

    raise ValueError(
        "Таблица df_atoms не содержит белковых атомов."
    )


if df_residues.empty:

    raise ValueError(
        "Таблица df_residues не содержит белковых остатков."
    )


required_atom_columns = {

    "pdb_id",

    "chain_id",

    "residue_key",

    "atom_name"

}


missing_atom_columns = (
    required_atom_columns - set(df_atoms.columns)
)


if missing_atom_columns:

    raise KeyError(
        "В df_atoms отсутствуют столбцы: "
        f"{sorted(missing_atom_columns)}"
    )


required_residue_columns = {

    "pdb_id",

    "chain_id",

    "residue_key"

}


missing_residue_columns = (
    required_residue_columns - set(df_residues.columns)
)


if missing_residue_columns:

    raise KeyError(
        "В df_residues отсутствуют столбцы: "
        f"{sorted(missing_residue_columns)}"
    )


# ------------------------------------------------------------
# 2. Проверка принадлежности таблиц текущей структуре
# ------------------------------------------------------------

def validate_table_pdb_id(table_name, table, required=True):

    if "pdb_id" not in table.columns:

        if required:

            raise KeyError(
                f"В {table_name} отсутствует столбец pdb_id."
            )

        print(
            f"WARNING: в {table_name} нет столбца pdb_id."
        )

        return

    if table["pdb_id"].isna().any():

        raise ValueError(
            f"{table_name} содержит пустые значения pdb_id."
        )

    pdb_ids = (
        table["pdb_id"]
        .astype(str)
        .str.strip()
        .str.upper()
        .unique()
    )

    if set(pdb_ids) != {PDB_ID}:

        raise ValueError(
            f"{table_name} содержит PDB ID, "
            f"не соответствующие текущей структуре: "
            f"{sorted(pdb_ids)}"
        )


validate_table_pdb_id(
    "df_atoms",
    df_atoms
)

validate_table_pdb_id(
    "df_residues",
    df_residues
)

validate_table_pdb_id(
    "df_hetero",
    df_hetero,
    required=False
)

validate_table_pdb_id(
    "df_water",
    df_water,
    required=False
)


print(f"PDB ID: {PDB_ID}")
print(f"Results directory: {PDB_RESULTS_DIR}")


# ------------------------------------------------------------
# 3. Выбор первой модели
# ------------------------------------------------------------

models = list(
    structure.get_models()
)


if not models:

    raise ValueError(
        "В структуре отсутствуют модели."
    )


first_model = models[0]

MODEL_INDEX = 1


print("\nMODEL INFORMATION")

print(
    f"Total models in structure: {len(models)}"
)

print(
    f"Selected model ID: {first_model.id}"
)

print(
    f"Selected model index: {MODEL_INDEX}"
)

print(
    "Analysis scope: first model only"
)


# ------------------------------------------------------------
# 4. Вспомогательные функции
# ------------------------------------------------------------

def normalize_chain_id(value):
    """
    Сохраняет пустой идентификатор цепи как пустую строку.
    """

    if value is None or pd.isna(value):

        return ""

    return str(value).strip()


def display_chain_id(value):

    normalized = normalize_chain_id(value)

    return normalized if normalized else "<blank>"


def safe_filename(value):

    value = str(value).strip()

    if not value:

        return "blank"

    return re.sub(
        r"[^A-Za-z0-9_-]",
        "_",
        value
    )


def make_residue_key(chain_id, hetero_flag, residue_id, insertion_code):

    """
    Формат ключа согласован с CELL 6–8:
    chain_id:hetero_flag:residue_id:insertion_code
    """

    return (
        f"{chain_id}:"
        f"{hetero_flag}:"
        f"{residue_id}:"
        f"{insertion_code}"
    )


def json_safe(value):
    """
    Рекурсивное преобразование значений pandas/NumPy
    в JSON-совместимые типы.
    """

    if value is None:

        return None

    if isinstance(value, dict):

        return {
            str(key): json_safe(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple, set)):

        return [
            json_safe(item)
            for item in value
        ]

    if isinstance(value, np.ndarray):

        return [
            json_safe(item)
            for item in value.tolist()
        ]

    if isinstance(value, (np.integer,)):

        return int(value)

    if isinstance(value, (np.floating, float)):

        value = float(value)

        return (
            value
            if np.isfinite(value)
            else None
        )

    if isinstance(value, (np.bool_, bool)):

        return bool(value)

    if isinstance(value, Path):

        return str(value)

    if isinstance(value, pd.Timestamp):

        return value.isoformat()

    if value is pd.NA or value is pd.NaT:

        return None

    if isinstance(value, (str, int)):

        return value

    try:

        missing = pd.isna(value)

        if isinstance(missing, (bool, np.bool_)) and missing:

            return None

    except (TypeError, ValueError):

        pass

    return str(value)


# ------------------------------------------------------------
# 5. Определение белковых цепей первой модели
# ------------------------------------------------------------

protein_chain_ids = set()

protein_residue_records = []

structure_atom_count = 0


for chain in first_model:

    chain_id = normalize_chain_id(chain.id)

    for residue in chain:

        if not is_aa(
            residue,
            standard=False
        ):

            continue

        protein_chain_ids.add(chain_id)

        residue_atoms = list(
            residue.get_atoms()
        )

        structure_atom_count += len(
            residue_atoms
        )

        hetero_flag = str(
            residue.id[0]
        ).strip()

        residue_id = int(
            residue.id[1]
        )

        insertion_code = str(
            residue.id[2]
        ).strip()


        residue_key = make_residue_key(

            chain_id,

            hetero_flag,

            residue_id,

            insertion_code

        )


        protein_residue_records.append({

            "pdb_id": PDB_ID,

            "model_index": MODEL_INDEX,

            "model_id": first_model.id,

            "chain_id": chain_id,

            "chain_id_display": display_chain_id(
                chain_id
            ),

            "residue_key": residue_key,

            "hetero_flag": hetero_flag,

            "residue_id": residue_id,

            "insertion_code": insertion_code,

            "residue_name": residue.resname.strip(),

            "atom_count": len(residue_atoms)

        })


df_structure_protein_residues = pd.DataFrame(
    protein_residue_records
)


if not protein_chain_ids:

    raise ValueError(
        "В первой модели не обнаружены белковые цепи."
    )


if df_structure_protein_residues["residue_key"].duplicated().any():

    raise RuntimeError(
        "Обнаружены повторяющиеся residue_key "
        "в таблице остатков первой модели."
    )


print("\n" + "-" * 70)
print("PROTEIN STRUCTURE")
print("-" * 70)

print(
    f"Protein chains: {len(protein_chain_ids)}"
)

print(
    f"Protein residues: {len(df_structure_protein_residues)}"
)

print(
    f"Protein atoms in first model: {structure_atom_count}"
)

print(
    "Chain IDs:",
    [
        display_chain_id(x)
        for x in sorted(protein_chain_ids)
    ]
)


# ------------------------------------------------------------
# 6. Формирование отдельных таблиц
# ------------------------------------------------------------

# CELL 6 и CELL 7 уже содержат белковые данные.
# Повторная фильтрация не выполняется.

df_protein_atoms = df_atoms.copy()

df_protein_residues = df_residues.copy()


# Небелковые компоненты.

df_hetero_components = df_hetero.copy()


# Сохраняем исходное имя для совместимости с последующими
# ячейками, если оно где-либо используется.
#
# ВАЖНО: это не подтверждённый набор лекарственных лигандов.
# Это все небелковые компоненты из CELL 8.

df_ligand_candidates = df_hetero_components.copy()


# Вода.

df_water_components = df_water.copy()


# ------------------------------------------------------------
# 7. Проверка согласованности CELL 6 и CELL 7
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("DATAFRAME CONSISTENCY CHECK")
print("-" * 70)


# 7.1. Проверка количества атомов.

cell7_atom_count = len(
    df_protein_atoms
)

atom_count_match = (
    cell7_atom_count == structure_atom_count
)


print(
    f"Atoms in structure: {structure_atom_count}"
)

print(
    f"Atoms in CELL 7: {cell7_atom_count}"
)

print(
    f"Atom count matches: {atom_count_match}"
)


# 7.2. Проверка количества остатков.

structure_residue_count = len(
    df_structure_protein_residues
)

cell6_residue_count = len(
    df_protein_residues
)

residue_count_match = (
    structure_residue_count == cell6_residue_count
)


print(
    f"Residues in structure: {structure_residue_count}"
)

print(
    f"Residues in CELL 6: {cell6_residue_count}"
)

print(
    f"Residue count matches: {residue_count_match}"
)


# 7.3. Сверка residue_key.

structure_residue_keys = set(
    df_structure_protein_residues["residue_key"]
    .astype(str)
)

cell6_residue_keys = set(
    df_protein_residues["residue_key"]
    .astype(str)
)


residue_keys_match = (
    structure_residue_keys == cell6_residue_keys
)


missing_in_cell6 = sorted(
    structure_residue_keys - cell6_residue_keys
)

extra_in_cell6 = sorted(
    cell6_residue_keys - structure_residue_keys
)


print(
    f"Residue keys match: {residue_keys_match}"
)

print(
    f"Missing residue keys in CELL 6: {len(missing_in_cell6)}"
)

print(
    f"Extra residue keys in CELL 6: {len(extra_in_cell6)}"
)


# 7.4. Проверка цепей.

structure_chain_ids = set(
    protein_chain_ids
)


cell7_chain_ids = set(
    df_protein_atoms["chain_id"]
    .map(normalize_chain_id)
)


cell6_chain_ids = set(
    df_protein_residues["chain_id"]
    .map(normalize_chain_id)
)


chain_ids_match = (
    structure_chain_ids == cell7_chain_ids
)


residue_chain_ids_match = (
    structure_chain_ids == cell6_chain_ids
)


print(
    "Chain IDs in structure:",
    sorted(structure_chain_ids)
)

print(
    "Chain IDs in CELL 7:",
    sorted(cell7_chain_ids)
)

print(
    "Chain IDs in CELL 6:",
    sorted(cell6_chain_ids)
)

print(
    f"Structure / CELL 7 chain match: {chain_ids_match}"
)

print(
    f"Structure / CELL 6 chain match: "
    f"{residue_chain_ids_match}"
)


# 7.5. Проверка уникальности residue_key в исходных таблицах.

cell6_duplicate_keys = int(
    df_protein_residues["residue_key"].duplicated().sum()
)


cell7_residue_key_count = (
    df_protein_atoms["residue_key"].nunique()
)


cell7_residue_key_match = (
    cell7_residue_key_count == structure_residue_count
)


print(
    f"Duplicate residue keys in CELL 6: "
    f"{cell6_duplicate_keys}"
)

print(
    f"Unique residue keys in CELL 7: "
    f"{cell7_residue_key_count}"
)

print(
    f"CELL 7 residue key count matches structure: "
    f"{cell7_residue_key_match}"
)


# Критические проверки.

if not atom_count_match:

    raise RuntimeError(
        "Количество белковых атомов в структуре "
        "не совпадает с CELL 7. "
        "Проверьте обработку altloc."
    )


if not residue_count_match:

    raise RuntimeError(
        "Количество белковых остатков в структуре "
        "не совпадает с CELL 6."
    )


if not residue_keys_match:

    raise RuntimeError(
        "Наборы residue_key не совпадают между структурой "
        "и CELL 6."
    )


if not chain_ids_match:

    raise RuntimeError(
        "Наборы цепей не совпадают между структурой "
        "и CELL 7."
    )


if not residue_chain_ids_match:

    raise RuntimeError(
        "Наборы цепей не совпадают между структурой "
        "и CELL 6."
    )


if cell6_duplicate_keys > 0:

    raise RuntimeError(
        "В CELL 6 обнаружены повторяющиеся residue_key."
    )


if not cell7_residue_key_match:

    raise RuntimeError(
        "Число уникальных residue_key в CELL 7 "
        "не совпадает с количеством остатков структуры."
    )


# ------------------------------------------------------------
# 8. Сводка небелковых компонентов
# ------------------------------------------------------------

df_component_summary = pd.DataFrame(

    columns=[
        "residue_name",
        "component_type",
        "occurrence_count"
    ]

)


if not df_hetero_components.empty:

    residue_name_col = next(

        (
            column
            for column in [
                "residue_name",
                "resname",
                "component_name"
            ]
            if column in df_hetero_components.columns
        ),

        None

    )


    component_type_col = next(

        (
            column
            for column in [
                "component_type",
                "classification",
                "component_class"
            ]
            if column in df_hetero_components.columns
        ),

        None

    )


    if residue_name_col is not None:

        component_data = df_hetero_components.copy()


        component_data["residue_name"] = (
            component_data[residue_name_col]
            .fillna("UNKNOWN")
            .astype(str)
            .str.strip()
        )


        if component_type_col is not None:

            component_data["component_type"] = (
                component_data[component_type_col]
                .fillna("Not classified")
                .astype(str)
                .str.strip()
            )

        else:

            component_data["component_type"] = (
                "Not classified"
            )


        df_component_summary = (

            component_data
            .groupby(
                [
                    "residue_name",
                    "component_type"
                ],
                dropna=False
            )
            .size()
            .reset_index(
                name="occurrence_count"
            )

        )


print("\n" + "-" * 70)
print("HETERO COMPONENTS")
print("-" * 70)

print(
    f"Total hetero component records: "
    f"{len(df_hetero_components)}"
)


if not df_component_summary.empty:

    display(
        df_component_summary
    )

else:

    print(
        "No hetero components found or residue names unavailable."
    )


# ------------------------------------------------------------
# 9. Сохранение отдельных файлов белковых цепей
# ------------------------------------------------------------

class ProteinChainSelect(Select):

    """
    Сохраняет одну белковую цепь выбранной модели.
    """

    def __init__(self, chain_id, model_id):

        self.chain_id = chain_id

        self.model_id = model_id


    def accept_model(self, model):

        return model.id == self.model_id


    def accept_chain(self, chain):

        return (
            normalize_chain_id(chain.id)
            == self.chain_id
        )


    def accept_residue(self, residue):

        return is_aa(
            residue,
            standard=False
        )


CHAIN_DIR = (
    PDB_RESULTS_DIR / "protein_chains"
)


CHAIN_DIR.mkdir(
    parents=True,
    exist_ok=True
)


saved_chain_files = []

chain_file_records = []


for chain_index, chain_id in enumerate(
    sorted(protein_chain_ids),
    start=1
):

    safe_chain = safe_filename(
        chain_id
    )


    output_stem = (
        f"{PDB_ID}_chain_{chain_index:02d}_{safe_chain}"
    )


    selector = ProteinChainSelect(
        chain_id,
        first_model.id
    )


    # PDB поддерживает только однобуквенный chain ID.
    # Для многосимвольных идентификаторов сохраняем mmCIF,
    # чтобы не потерять исходный идентификатор цепи.

    if len(chain_id) <= 1:

        output_path = (
            CHAIN_DIR / f"{output_stem}.pdb"
        )

        io = PDBIO()

        io.set_structure(
            structure
        )

        io.save(
            str(output_path),
            selector
        )

        output_format = "PDB"

    else:

        output_path = (
            CHAIN_DIR / f"{output_stem}.cif"
        )

        io = MMCIFIO()

        io.set_structure(
            structure
        )

        io.save(
            str(output_path),
            selector
        )

        output_format = "mmCIF"


    if not output_path.is_file():

        raise FileNotFoundError(
            f"Файл цепи не сохранён: {output_path}"
        )


    if output_path.stat().st_size == 0:

        raise RuntimeError(
            f"Файл цепи пустой: {output_path}"
        )


    saved_chain_files.append(
        output_path
    )


    chain_file_records.append({

        "pdb_id": PDB_ID,

        "model_index": MODEL_INDEX,

        "model_id": str(first_model.id),

        "chain_id": chain_id,

        "chain_id_display": display_chain_id(
            chain_id
        ),

        "file_name": output_path.name,

        "file_format": output_format,

        "file_size_bytes": output_path.stat().st_size

    })


    print(
        f"Saved chain {display_chain_id(chain_id)}: "
        f"{output_path.name} [{output_format}]"
    )


df_chain_files = pd.DataFrame(
    chain_file_records
)


# ------------------------------------------------------------
# 10. Проверка сохранённых цепей
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("CHAIN FILE VALIDATION")
print("-" * 70)


if len(saved_chain_files) != len(protein_chain_ids):

    raise RuntimeError(
        "Количество сохранённых файлов "
        "не совпадает с количеством белковых цепей."
    )


for path in saved_chain_files:

    if not path.is_file() or path.stat().st_size == 0:

        raise RuntimeError(
            f"Некорректный файл цепи: {path}"
        )


    print(
        f"{path.name}: "
        f"{path.stat().st_size:,} bytes"
    )


# ------------------------------------------------------------
# 11. Сохранение отдельных таблиц
# ------------------------------------------------------------

output_tables = {

    "separated_protein_atoms": (
        df_protein_atoms
    ),

    "separated_protein_residues": (
        df_protein_residues
    ),

    "separated_hetero_components": (
        df_hetero_components
    ),

    "separated_water_components": (
        df_water_components
    ),

    "separated_structure_protein_residues": (
        df_structure_protein_residues
    ),

    "separated_component_summary": (
        df_component_summary
    ),

    "separated_chain_files": (
        df_chain_files
    )

}


saved_table_files = {}


for name, table in output_tables.items():

    output_path = (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_{name}.csv"
    )


    table.to_csv(

        output_path,

        index=False,

        encoding="utf-8-sig"

    )


    saved_table_files[name] = output_path


    print(
        f"Saved: {output_path.name}"
    )


# ------------------------------------------------------------
# 12. Сохранение JSON-сводки
# ------------------------------------------------------------

SEPARATION_JSON_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_component_separation.json"
)


separation_summary = {

    "PDB_ID": PDB_ID,

    "analysis_scope": {

        "selected_model_index": MODEL_INDEX,

        "selected_model_id": str(first_model.id),

        "total_models_in_structure": len(models),

        "note": (
            "Protein chain separation uses the first model. "
            "This is consistent with CELL 4-11."
        )

    },

    "protein": {

        "number_of_protein_chains": len(
            protein_chain_ids
        ),

        "protein_chain_ids": sorted(
            protein_chain_ids
        ),

        "number_of_protein_residues": len(
            df_protein_residues
        ),

        "number_of_protein_atoms": len(
            df_protein_atoms
        )

    },

    "hetero_components": {

        "number_of_component_records": len(
            df_hetero_components
        ),

        "classification_available": (
            "component_type" in df_component_summary.columns
            and not df_component_summary.empty
        ),

        "component_summary_records": (
            df_component_summary.to_dict(
                orient="records"
            )
        ),

        "note": (
            "Hetero components are not automatically "
            "classified as drug-like ligands."
        )

    },

    "water": {

        "number_of_water_component_records": len(
            df_water_components
        )

    },

    "validation": {

        "atom_count_matches_structure": atom_count_match,

        "residue_count_matches_structure": residue_count_match,

        "residue_keys_match": residue_keys_match,

        "chain_ids_match_structure": chain_ids_match,

        "residue_chain_ids_match_structure": (
            residue_chain_ids_match
        ),

        "cell7_residue_key_count_matches": (
            cell7_residue_key_match
        ),

        "number_of_saved_chain_files": len(
            saved_chain_files
        )

    },

    "saved_chain_files": (

        df_chain_files.to_dict(
            orient="records"
        )

    )

}


with open(
    SEPARATION_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(

        json_safe(separation_summary),

        file,

        ensure_ascii=False,

        indent=4,

        allow_nan=False

    )


# ------------------------------------------------------------
# 13. Итоговая статистика
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("COMPONENT SEPARATION SUMMARY")
print("=" * 70)

print(
    f"Protein chains: {len(protein_chain_ids)}"
)

print(
    f"Protein residues: {len(df_protein_residues)}"
)

print(
    f"Protein atoms: {len(df_protein_atoms)}"
)

print(
    f"Hetero components: {len(df_hetero_components)}"
)

print(
    f"Water components: {len(df_water_components)}"
)

print(
    f"Saved chain files: {len(saved_chain_files)}"
)

print(
    f"Output directory: {PDB_RESULTS_DIR}"
)


# ------------------------------------------------------------
# 14. Проверка сохранённых результатов
# ------------------------------------------------------------

all_output_files = (

    list(saved_table_files.values())

    + saved_chain_files

    + [SEPARATION_JSON_PATH]

)


print("\n" + "-" * 70)
print("OUTPUT FILE VERIFICATION")
print("-" * 70)


output_check_records = []


for path in all_output_files:

    exists = (

        path.is_file()

        and path.stat().st_size > 0

    )


    output_check_records.append({

        "file_name": path.name,

        "exists": bool(exists),

        "file_size_bytes": (
            path.stat().st_size
            if path.is_file()
            else 0
        )

    })


    print(
        f"{path.name}: "
        f"{'OK' if exists else 'ERROR'}"
    )


df_cell15_output_checks = pd.DataFrame(
    output_check_records
)


all_outputs_saved = bool(
    df_cell15_output_checks["exists"].all()
)


if not all_outputs_saved:

    raise RuntimeError(
        "Не все результаты разделения структуры сохранены."
    )


# ------------------------------------------------------------
# 15. Итог
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("CELL 15 FINAL STATUS")
print("-" * 70)

print(
    f"Atom count validation: "
    f"{'OK' if atom_count_match else 'ERROR'}"
)

print(
    f"Residue count validation: "
    f"{'OK' if residue_count_match else 'ERROR'}"
)

print(
    f"Residue key validation: "
    f"{'OK' if residue_keys_match else 'ERROR'}"
)

print(
    f"Chain validation: "
    f"{'OK' if chain_ids_match else 'ERROR'}"
)

print(
    f"Saved output files: {len(all_output_files)}"
)

print(
    f"Results directory: {PDB_RESULTS_DIR}"
)

print("\nСтатус: OK")

print("=" * 70)
print("CELL 15 COMPLETED")
print("=" * 70)
```

Что изменилось по сравнению с исходным кодом
Элемент	Исходная версия	Новая версия
Белковые цепи	Используются ключи chain_sequences	Определяются непосредственно по structure
Белковые остатки	Копируется df_residues	Фильтруются по белковым остаткам структуры
Белковые атомы	Копируется df_atoms	Фильтруются по белковым остаткам структуры
Гетерокомпоненты	Копируются из df_hetero	Без изменений
Вода	Копируется из df_water	Без изменений
PDB-файлы	Сохраняются цепи из chain_sequences	Сохраняются все обнаруженные белковые цепи

Важное замечание: код предполагает, что в df_atoms и df_residues есть колонки с идентификатором цепи и номером остатка. Если они называются иначе или residue_id имеет нестандартный формат, фильтрацию нужно будет адаптировать под реальные названия колонок.

Ещё один нюанс: формат PDB имеет ограничения на идентификаторы цепей. Если исходный файл mmCIF использует многосимвольные идентификаторы цепей, сохранение в PDB может привести к потере информации. Для таких структур предпочтительнее сохранять отдельные цепи в mmCIF.

```python
# ============================================================
# CELL 17. DETECTION AND IDENTIFICATION OF NON-PROTEIN COMPONENTS
# ============================================================
# Назначение:
# 1. Обнаружение небелковых компонентов первой модели.
# 2. Исключение воды и распознанных аминокислотных остатков.
# 3. Определение элементного состава.
# 4. Предварительная классификация компонентов.
# 5. Получение химической информации из mmCIF _chem_comp.
# 6. Контролируемый выбор компонента без случайного fallback.
# 7. Сохранение таблиц компонентов и их атомов.
#
# ВАЖНО:
# - Используется объект structure из CELL 4.
# - Повторная загрузка структуры через MMCIFParser не выполняется.
# - MMCIF2Dict используется только для чтения _chem_comp.
# - Анализируется первая модель.
# - Альтернативные положения атомов сохраняются.
# - Альтернативные варианты остатков учитываются отдельно.
# - number_of_atoms = число уникальных атомных позиций.
# - number_of_atom_conformers = число записей атомов
#   с учетом альтернативных положений.
# - Классификация компонентов предварительная.
# - Небелковый компонент не считается автоматически лигандом.
# - При отсутствии однозначного выбора selected_component=None.
# ============================================================

import json
import warnings
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.PDB import is_aa
from Bio.PDB.MMCIF2Dict import MMCIF2Dict


print("=" * 70)
print("CELL 17. NON-PROTEIN COMPONENT DETECTION")
print("=" * 70)


# ------------------------------------------------------------
# 1. ПРОВЕРКА ВХОДНЫХ ДАННЫХ
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "CIF_PATH",
    "PDB_RESULTS_DIR",
    "structure"
]

for variable in required_variables:
    if variable not in globals():
        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()
CIF_PATH = Path(CIF_PATH)
PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

if not PDB_ID or PDB_ID == "UNKNOWN":
    raise ValueError("Некорректный PDB_ID.")

if not CIF_PATH.is_file():
    raise FileNotFoundError(
        f"Structure file not found: {CIF_PATH}"
    )

if CIF_PATH.stat().st_size == 0:
    raise ValueError("Structure file is empty.")

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ------------------------------------------------------------
# 2. ИСПОЛЬЗОВАНИЕ СТРУКТУРЫ ИЗ CELL 4
# ------------------------------------------------------------

# Повторно не парсим структуру.
# Это сохраняет идентичность объектов Bio.PDB
# для последующего использования в CELL18.

cell17_models = list(structure.get_models())

if not cell17_models:
    raise ValueError(
        f"No models found in structure {PDB_ID}."
    )

model = cell17_models[0]

MODEL_INDEX = 0

MODEL_SERIAL_NUMBER = getattr(
    model,
    "serial_num",
    None
)

if MODEL_SERIAL_NUMBER is None:
    MODEL_SERIAL_NUMBER = model.id


structure_id = str(
    getattr(structure, "id", "")
).strip().upper()

if structure_id and structure_id != PDB_ID:
    raise ValueError(
        f"Structure ID mismatch: "
        f"{structure_id} != {PDB_ID}"
    )


# mmCIF читается отдельно только для химического справочника.
cell17_cif_dict = MMCIF2Dict(str(CIF_PATH))


print("PDB ID:", PDB_ID)
print("Structure:", CIF_PATH)
print("Total models in file:", len(cell17_models))
print("Analyzed model index:", MODEL_INDEX)
print("Model serial number:", MODEL_SERIAL_NUMBER)


# ------------------------------------------------------------
# 3. НАСТРОЙКИ
# ------------------------------------------------------------

WATER_CODES = {
    "HOH",
    "WAT",
    "DOD"
}

METAL_ELEMENTS = {
    "FE", "ZN", "MG", "MN", "CU",
    "CO", "NI", "CA", "NA", "K",
    "CD", "HG"
}

# Условия выбора целевого компонента.
#
# Для анализа гема A в структуре 1OCC:
# PREFERRED_COMPONENT_NAME = "HEA"
# PREFERRED_COMPONENT_CHAIN = "A"
#
# Для пакетной обработки эти параметры следует
# задавать индивидуально для каждой структуры.
#
# Если компонент не задан, автоматический выбор
# не производится.

PREFERRED_COMPONENT_NAME = "HEM"
PREFERRED_COMPONENT_CHAIN = "A"

# Дополнительное уточнение, если в одной цепи
# обнаружено несколько экземпляров одного компонента.
#
# None означает, что номер остатка не ограничивается.

PREFERRED_COMPONENT_RESIDUE_NUMBER = None

# Необязательный явный выбор по component_index.
# Используется только при ручном подтверждении.
#
# None — выбирать по имени, цепи и номеру остатка.

SELECT_COMPONENT_INDEX = None

IMPORTANT_COMPONENT_CODES = [
    "HEM", "HEA", "HEB",
    "NDP", "NAD", "NAP",
    "FAD", "FMN",
    "ATP", "ADP", "GTP", "GDP",
    "SAM", "PLP", "COA", "TPP",
    "MG", "ZN", "FE", "CU",
    "PO4", "SO4", "ACT", "OXY"
]


# ------------------------------------------------------------
# 4. ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ
# ------------------------------------------------------------

def cell17_is_null(value):
    """Безопасная проверка скалярного значения на пропуск."""

    if value is None:
        return True

    if value is pd.NA or value is pd.NaT:
        return True

    try:
        result = pd.isna(value)

        if isinstance(result, (bool, np.bool_)):
            return bool(result)

    except (TypeError, ValueError):
        pass

    return False


def clean_cif_value(value):
    """Нормализация значения mmCIF."""

    if cell17_is_null(value):
        return None

    value = str(value).strip()

    if value in {"", ".", "?"}:
        return None

    return value


def get_cif_column(cif_data, key):
    """Безопасное получение столбца mmCIF."""

    values = cif_data.get(key, [])

    if values is None:
        return []

    if isinstance(values, str):
        values = [values]

    return [
        clean_cif_value(value)
        for value in values
    ]


def normalize_chain_id(chain_id):
    """Отображение пустого идентификатора цепи."""

    value = str(chain_id)

    return value if value.strip() else "<blank>"


def normalize_component_code(value):
    """Нормализация кода компонента."""

    if cell17_is_null(value):
        return None

    value = str(value).strip().upper()

    return value if value else None


def get_element(atom):
    """Получение нормализованного химического элемента."""

    element = str(
        getattr(atom, "element", "")
    ).strip().upper()

    return element if element else "UNKNOWN"


def get_unique_atom_site_key(atom):
    """
    Идентификатор атомной позиции.

    Альтернативные конформеры одной позиции
    имеют одинаковые имя атома и элемент.
    """

    return (
        str(atom.get_name()).strip(),
        get_element(atom)
    )


def get_concrete_residues(residue):
    """
    Возвращает конкретные варианты остатка.

    Для DisorderedResidue учитываются отдельные
    варианты аминокислот/компонентов.
    """

    if getattr(residue, "is_disordered", lambda: 0)() == 2:

        variants = residue.disordered_get_list()

        return list(variants)

    return [residue]


def classify_component(elements, atom_count):
    """
    Предварительная классификация по элементному составу.

    Не определяет биологическую функцию компонента.
    """

    element_set = set(elements.keys())

    has_carbon = "C" in element_set

    metal_elements = sorted(
        element_set.intersection(METAL_ELEMENTS)
    )

    if metal_elements:

        if has_carbon:
            return "Metal-containing organic component"

        if atom_count == 1:
            return "Metal ion"

        return "Metal-containing inorganic component"

    if atom_count == 1:
        return "Single-atom component / ion"

    if has_carbon:
        return "Organic non-protein component"

    return "Inorganic non-protein component"


def json_safe(value):
    """Рекурсивное преобразование значений в JSON-формат."""

    if isinstance(value, dict):
        return {
            str(key): json_safe(val)
            for key, val in value.items()
        }

    if isinstance(value, (list, tuple)):
        return [
            json_safe(val)
            for val in value
        ]

    if isinstance(value, np.ndarray):
        return json_safe(value.tolist())

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):
        value = float(value)
        return value if np.isfinite(value) else None

    if isinstance(value, float):
        return value if np.isfinite(value) else None

    if isinstance(value, np.bool_):
        return bool(value)

    if cell17_is_null(value):
        return None

    if isinstance(value, (str, int, bool)):
        return value

    return str(value)


# ------------------------------------------------------------
# 5. CHEMICAL COMPONENT DICTIONARY
# ------------------------------------------------------------

chem_comp_fields = {
    "component": "_chem_comp.id",
    "name": "_chem_comp.name",
    "type": "_chem_comp.type",
    "formula": "_chem_comp.formula",
    "formal_charge": "_chem_comp.pdbx_formal_charge",
    "synonyms": "_chem_comp.pdbx_synonyms",
    "mon_nstd_flag": "_chem_comp.mon_nstd_flag"
}

chem_comp_columns = {
    name: get_cif_column(cell17_cif_dict, cif_key)
    for name, cif_key in chem_comp_fields.items()
}

chem_comp_max_length = max(
    (len(values) for values in chem_comp_columns.values()),
    default=0
)

chem_comp_records = []

for i in range(chem_comp_max_length):

    record = {}

    for column, values in chem_comp_columns.items():

        record[column] = (
            values[i]
            if i < len(values)
            else None
        )

    chem_comp_records.append(record)


df_chem_comp_dictionary = pd.DataFrame(
    chem_comp_records,
    columns=list(chem_comp_fields.keys())
)


if not df_chem_comp_dictionary.empty:

    df_chem_comp_dictionary["component"] = (
        df_chem_comp_dictionary["component"]
        .fillna("")
        .astype(str)
        .str.strip()
        .str.upper()
    )

    df_chem_comp_dictionary = (
        df_chem_comp_dictionary[
            df_chem_comp_dictionary["component"] != ""
        ]
        .drop_duplicates(
            subset=["component"],
            keep="first"
        )
        .reset_index(drop=True)
    )


chem_comp_lookup = {
    str(row["component"]): row.to_dict()
    for _, row in df_chem_comp_dictionary.iterrows()
    if not cell17_is_null(row["component"])
}


print("\nChemical component dictionary:")
print("Records:", len(df_chem_comp_dictionary))


# ------------------------------------------------------------
# 6. ОБНАРУЖЕНИЕ НЕБЕЛКОВЫХ КОМПОНЕНТОВ
# ------------------------------------------------------------

# Основной список кандидатов.
#
# Каждая запись содержит конкретный объект остатка,
# который принадлежит исходному объекту structure.
#
# ligand_residues оставляется как совместимый alias,
# но фактически содержит кандидатов небелковых компонентов.

nonprotein_residue_candidates = []

ligand_residues = nonprotein_residue_candidates

water_count = 0

standard_amino_acid_count = 0
modified_amino_acid_count = 0

total_residue_count = 0
total_residue_variant_count = 0

for chain in model:

    for residue_container in chain:

        total_residue_count += 1

        concrete_residues = get_concrete_residues(
            residue_container
        )

        for residue in concrete_residues:

            total_residue_variant_count += 1

            residue_name = (
                str(residue.resname).strip().upper()
            )

            all_atoms = residue.get_unpacked_list()

            if not all_atoms:
                continue

            if residue_name in WATER_CODES:
                water_count += 1
                continue

            if is_aa(residue, standard=True):
                standard_amino_acid_count += 1
                continue

            if is_aa(residue, standard=False):
                modified_amino_acid_count += 1
                continue

            nonprotein_residue_candidates.append({
                "chain": chain,
                "residue": residue
            })


recognized_amino_acid_count = (
    standard_amino_acid_count
    + modified_amino_acid_count
)


print("\n" + "-" * 70)
print("DETECTION RESULTS")
print("-" * 70)

print("Residue containers:", total_residue_count)
print("Concrete residue variants:", total_residue_variant_count)
print("Water residues:", water_count)

print(
    "Standard amino acid residues excluded:",
    standard_amino_acid_count
)

print(
    "Modified / non-standard amino acid residues excluded:",
    modified_amino_acid_count
)

print(
    "Recognized amino acid residues excluded:",
    recognized_amino_acid_count
)

print(
    "Potential non-protein component instances:",
    len(nonprotein_residue_candidates)
)


# ------------------------------------------------------------
# 7. СБОР ИНФОРМАЦИИ О КАЖДОМ КОМПОНЕНТЕ
# ------------------------------------------------------------

components_info = []
component_atoms_data = []

# component_index -> реальный объект Bio.PDB Residue.
component_objects = {}

for item in nonprotein_residue_candidates:

    chain = item["chain"]
    residue = item["residue"]

    chain_id = str(chain.id)
    chain_id_display = normalize_chain_id(chain_id)

    residue_id = residue.id

    residue_name = (
        str(residue.resname).strip().upper()
    )

    residue_number = residue_id[1]

    insertion_code = str(
        residue_id[2]
    ).strip()

    hetero_flag = str(
        residue_id[0]
    ).strip()

    all_atoms = residue.get_unpacked_list()
    unique_atoms = list(residue.get_atoms())

    if not all_atoms or not unique_atoms:
        continue

    unique_site_keys = {
        get_unique_atom_site_key(atom)
        for atom in all_atoms
    }

    number_of_atom_sites = len(unique_site_keys)
    number_of_atom_conformers = len(all_atoms)

    if number_of_atom_sites == 0:
        continue

    coordinates = np.asarray(
        [atom.coord for atom in unique_atoms],
        dtype=float
    )

    if coordinates.ndim != 2 or coordinates.shape[1] != 3:
        raise ValueError(
            f"Invalid coordinate shape in component {residue_name}."
        )

    if not np.isfinite(coordinates).all():
        raise ValueError(
            f"Invalid coordinates in component {residue_name}, "
            f"chain {chain_id}, residue {residue_number}."
        )

    centroid = coordinates.mean(axis=0)

    elements = {}

    for atom in unique_atoms:

        element = get_element(atom)

        elements[element] = (
            elements.get(element, 0) + 1
        )

    metal_elements = sorted(
        set(elements.keys()).intersection(METAL_ELEMENTS)
    )

    component_type = classify_component(
        elements,
        number_of_atom_sites
    )

    dictionary_info = chem_comp_lookup.get(
        residue_name,
        {}
    )

    component_name = dictionary_info.get("name")
    component_formula = dictionary_info.get("formula")
    component_synonyms = dictionary_info.get("synonyms")
    component_chem_type = dictionary_info.get("type")
    component_formal_charge = dictionary_info.get("formal_charge")
    component_mon_nstd_flag = dictionary_info.get("mon_nstd_flag")

    component_id = (
        f"{PDB_ID}|model_{MODEL_INDEX}|"
        f"chain_{chain_id}|{hetero_flag}|"
        f"{residue_number}|{insertion_code}|{residue_name}"
    )

    component_index = len(components_info) + 1

    info = {
        "component_index": component_index,
        "component_id": component_id,

        "component": residue_name,
        "component_name": component_name,
        "component_formula": component_formula,
        "component_synonyms": component_synonyms,
        "chemical_component_type": component_chem_type,
        "formal_charge": component_formal_charge,
        "mon_nstd_flag": component_mon_nstd_flag,

        "component_type": component_type,

        "model_index": MODEL_INDEX,
        "model_id": model.id,
        "model_serial_number": MODEL_SERIAL_NUMBER,

        "chain": chain_id,
        "chain_display": chain_id_display,

        "residue_number": residue_number,
        "insertion_code": insertion_code,
        "hetero_flag": hetero_flag,

        "number_of_atoms": number_of_atom_sites,
        "number_of_atom_conformers": number_of_atom_conformers,

        "number_of_alternative_atom_records": (
            number_of_atom_conformers - number_of_atom_sites
        ),

        "elements": elements,
        "metal_elements": metal_elements,

        "center_x": float(centroid[0]),
        "center_y": float(centroid[1]),
        "center_z": float(centroid[2])
    }

    components_info.append(info)

    component_objects[component_index] = residue

    # Сохраняем все атомные конформеры.
    for atom in all_atoms:

        element = get_element(atom)

        component_atoms_data.append({

            "component_index": component_index,
            "component_id": component_id,

            "component": residue_name,
            "component_name": component_name,
            "component_type": component_type,

            "model_index": MODEL_INDEX,
            "model_id": model.id,
            "model_serial_number": MODEL_SERIAL_NUMBER,

            "chain": chain_id,
            "chain_display": chain_id_display,

            "residue_number": residue_number,
            "insertion_code": insertion_code,
            "hetero_flag": hetero_flag,

            "atom_name": atom.get_name(),
            "element": element,

            "x": float(atom.coord[0]),
            "y": float(atom.coord[1]),
            "z": float(atom.coord[2]),

            "occupancy": atom.occupancy,
            "B_factor": atom.bfactor,
            "altloc": atom.get_altloc()

        })


# ------------------------------------------------------------
# 8. СОЗДАНИЕ DATAFRAME
# ------------------------------------------------------------

COMPONENT_COLUMNS = [
    "component_index",
    "component_id",
    "component",
    "component_name",
    "component_formula",
    "component_synonyms",
    "chemical_component_type",
    "formal_charge",
    "mon_nstd_flag",
    "component_type",
    "model_index",
    "model_id",
    "model_serial_number",
    "chain",
    "chain_display",
    "residue_number",
    "insertion_code",
    "hetero_flag",
    "number_of_atoms",
    "number_of_atom_conformers",
    "number_of_alternative_atom_records",
    "elements",
    "metal_elements",
    "center_x",
    "center_y",
    "center_z"
]

ATOM_COLUMNS = [
    "component_index",
    "component_id",
    "component",
    "component_name",
    "component_type",
    "model_index",
    "model_id",
    "model_serial_number",
    "chain",
    "chain_display",
    "residue_number",
    "insertion_code",
    "hetero_flag",
    "atom_name",
    "element",
    "x",
    "y",
    "z",
    "occupancy",
    "B_factor",
    "altloc"
]

df_components = pd.DataFrame(
    components_info,
    columns=COMPONENT_COLUMNS
)

df_component_atoms = pd.DataFrame(
    component_atoms_data,
    columns=ATOM_COLUMNS
)


print("\n" + "=" * 70)
print("DETECTED NON-PROTEIN COMPONENTS")
print("=" * 70)

if df_components.empty:

    print("No non-protein components detected.")

else:

    display(df_components)

    print("\nComponent classification:")

    display(
        df_components["component_type"]
        .value_counts()
        .rename_axis("component_type")
        .reset_index(name="count")
    )


# ------------------------------------------------------------
# 9. СТАТИСТИКА ПО КОДАМ КОМПОНЕНТОВ
# ------------------------------------------------------------

if not df_components.empty:

    df_component_occurrences = (
        df_components
        .groupby("component", dropna=False)
        .agg(
            occurrence_count=("component_index", "count"),
            total_atom_count=("number_of_atoms", "sum"),
            total_atom_conformer_count=(
                "number_of_atom_conformers",
                "sum"
            ),
            chains=(
                "chain_display",
                lambda x: ", ".join(
                    sorted(set(x.astype(str)))
                )
            )
        )
        .reset_index()
    )

else:

    df_component_occurrences = pd.DataFrame(
        columns=[
            "component",
            "occurrence_count",
            "total_atom_count",
            "total_atom_conformer_count",
            "chains"
        ]
    )


number_of_component_instances = len(df_components)

number_of_unique_component_types = (
    df_components["component"].nunique()
    if not df_components.empty
    else 0
)


print("\nComponent occurrences:")
display(df_component_occurrences)

print(
    "\nNumber of component instances:",
    number_of_component_instances
)

print(
    "Number of unique component types:",
    number_of_unique_component_types
)


# ------------------------------------------------------------
# 10. ПРОВЕРКА ВАЖНЫХ КОМПОНЕНТОВ
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("IMPORTANT COMPONENTS")
print("=" * 70)

for component_code in IMPORTANT_COMPONENT_CODES:

    print(f"\n{component_code}:")

    if df_components.empty:
        print("Not detected.")
        continue

    selected_rows = df_components[
        df_components["component"] == component_code
    ]

    if selected_rows.empty:
        print("Not detected.")
        continue

    display(
        selected_rows[
            [
                "component",
                "component_name",
                "component_formula",
                "chain_display",
                "residue_number",
                "number_of_atoms",
                "elements",
                "metal_elements"
            ]
        ]
    )


# ------------------------------------------------------------
# 11. КОНТРОЛИРУЕМЫЙ ВЫБОР КОМПОНЕНТА
# ------------------------------------------------------------

selected_component = None
selected_component_index = None
selected_info = None

SELECTED_COMPONENT_NAME = None
SELECTED_COMPONENT_CHAIN = None

SELECTION_METHOD = "No component selected"
SELECTION_STATUS = "NOT_SELECTED"
SELECTION_CANDIDATE_COUNT = 0

selected_rows = pd.DataFrame(
    columns=COMPONENT_COLUMNS
)


if df_components.empty:

    SELECTION_STATUS = "NO_COMPONENTS"

    SELECTION_METHOD = (
        "No non-protein component instances detected"
    )

    print("\nNo component available for selection.")

else:

    print("\nAVAILABLE COMPONENTS")
    print("-" * 70)

    for _, row in df_components.iterrows():

        print(
            f"{int(row['component_index'])}. "
            f"{row['component']} | "
            f"Chain: {row['chain_display']} | "
            f"Residue: {row['residue_number']} | "
            f"Atoms: {row['number_of_atoms']}"
        )

    # --------------------------------------------------------
    # 11.1. Явный выбор по component_index
    # --------------------------------------------------------

    if SELECT_COMPONENT_INDEX is not None:

        selected_rows = df_components[
            df_components["component_index"]
            == int(SELECT_COMPONENT_INDEX)
        ]

        if len(selected_rows) == 1:

            SELECTION_STATUS = "SELECTED"
            SELECTION_METHOD = "Explicit component_index"

        else:

            SELECTION_STATUS = "INVALID_COMPONENT_INDEX"

            SELECTION_METHOD = (
                "Explicit component_index not found"
            )

    # --------------------------------------------------------
    # 11.2. Выбор по имени и дополнительным условиям
    # --------------------------------------------------------

    elif PREFERRED_COMPONENT_NAME is not None:

        preferred_name = normalize_component_code(
            PREFERRED_COMPONENT_NAME
        )

        selected_rows = df_components[
            df_components["component"] == preferred_name
        ].copy()

        if PREFERRED_COMPONENT_CHAIN is not None:

            selected_rows = selected_rows[
                selected_rows["chain"]
                == str(PREFERRED_COMPONENT_CHAIN)
            ]

        if PREFERRED_COMPONENT_RESIDUE_NUMBER is not None:

            selected_rows = selected_rows[
                selected_rows["residue_number"]
                == PREFERRED_COMPONENT_RESIDUE_NUMBER
            ]

        SELECTION_CANDIDATE_COUNT = len(selected_rows)

        if len(selected_rows) == 1:

            SELECTION_STATUS = "SELECTED"

            SELECTION_METHOD = (
                "Unique match by configured component criteria"
            )

        elif len(selected_rows) == 0:

            SELECTION_STATUS = "PREFERRED_COMPONENT_NOT_FOUND"

            SELECTION_METHOD = (
                "Configured component criteria did not match"
            )

        else:

            SELECTION_STATUS = "AMBIGUOUS_MATCH"

            SELECTION_METHOD = (
                "Multiple components match configured criteria"
            )

    else:

        SELECTION_STATUS = "NO_SELECTION_CRITERIA"

        SELECTION_METHOD = (
            "No explicit component selection criteria provided"
        )


    # --------------------------------------------------------
    # 11.3. Получение объекта только при однозначном выборе
    # --------------------------------------------------------

    if SELECTION_STATUS == "SELECTED":

        selected_info = selected_rows.iloc[0].to_dict()

        selected_component_index = int(
            selected_info["component_index"]
        )

        selected_component = component_objects.get(
            selected_component_index
        )

        if selected_component is None:

            raise RuntimeError(
                "Selected component object was not found."
            )

        selected_chain_object = (
            selected_component.get_parent()
        )

        selected_model_object = (
            selected_chain_object.get_parent()
        )

        if selected_model_object is not model:

            raise RuntimeError(
                "Selected component does not belong to "
                "the analyzed first model."
            )

        SELECTED_COMPONENT_NAME = (
            selected_info["component"]
        )

        SELECTED_COMPONENT_CHAIN = (
            selected_info["chain"]
        )

        print("\n" + "=" * 70)
        print("SELECTED COMPONENT")
        print("=" * 70)

        print("Selection status:", SELECTION_STATUS)
        print("Selection method:", SELECTION_METHOD)

        print("Component:", selected_info["component"])
        print("Chemical name:", selected_info["component_name"])
        print("Formula:", selected_info["component_formula"])

        print("Chain:", selected_info["chain_display"])
        print("Residue number:", selected_info["residue_number"])
        print("Model:", selected_info["model_serial_number"])

        print(
            "Unique atom sites:",
            selected_info["number_of_atoms"]
        )

        print(
            "Atom conformers:",
            selected_info["number_of_atom_conformers"]
        )

        print(
            "Elemental composition:",
            selected_info["elements"]
        )

        print(
            "Metal elements:",
            selected_info["metal_elements"]
        )

        print(
            "Geometric center:",
            np.round(
                [
                    selected_info["center_x"],
                    selected_info["center_y"],
                    selected_info["center_z"]
                ],
                3
            )
        )

    else:

        # Никакого выбора первого компонента.
        selected_component = None
        selected_component_index = None
        selected_info = None

        SELECTED_COMPONENT_NAME = None
        SELECTED_COMPONENT_CHAIN = None

        print("\n" + "=" * 70)
        print("COMPONENT NOT SELECTED")
        print("=" * 70)

        print("Selection status:", SELECTION_STATUS)
        print("Selection method:", SELECTION_METHOD)
        print("Matching candidates:", SELECTION_CANDIDATE_COUNT)

        if SELECTION_STATUS == "PREFERRED_COMPONENT_NOT_FOUND":

            print(
                "Configured component was not found. "
                "No fallback selection was performed."
            )

        elif SELECTION_STATUS == "AMBIGUOUS_MATCH":

            print(
                "Several components match the criteria. "
                "Specify residue number or component_index."
            )


# ------------------------------------------------------------
# 12. АТОМЫ ВЫБРАННОГО КОМПОНЕНТА
# ------------------------------------------------------------

if selected_component is not None:

    df_selected_atoms = df_component_atoms[
        df_component_atoms["component_index"]
        == selected_component_index
    ].copy()

else:

    df_selected_atoms = pd.DataFrame(
        columns=ATOM_COLUMNS
    )


print("\nSelected component atom records:")
print(len(df_selected_atoms))

print(
    "Selected component unique atom sites:",
    (
        len({
            (row["atom_name"], row["element"])
            for _, row in df_selected_atoms.iterrows()
        })
        if not df_selected_atoms.empty
        else 0
    )
)

print(
    "Selected component atom conformers:",
    len(df_selected_atoms)
)


# ------------------------------------------------------------
# 13. СОПОСТАВЛЕНИЕ С CELL 8
# ------------------------------------------------------------

cell8_comparison = {
    "available": False,
    "cell8_component_count": None,
    "cell17_component_count": len(df_components),
    "difference": None,
    "status": "Not checked"
}

if (
    "df_hetero" in globals()
    and isinstance(df_hetero, pd.DataFrame)
):

    cell8_comparison["available"] = True
    cell8_comparison["cell8_component_count"] = len(df_hetero)

    # df_hetero может содержать строки атомов или остатков.
    # Поэтому простое сравнение количества строк
    # не считается полноценной проверкой идентичности.

    if "residue_key" in df_hetero.columns:

        cell8_unique_residues = (
            df_hetero["residue_key"]
            .dropna()
            .nunique()
        )

        cell8_comparison["cell8_unique_residue_count"] = (
            int(cell8_unique_residues)
        )

        cell8_comparison["difference"] = (
            len(df_components) - cell8_unique_residues
        )

        cell8_comparison["status"] = (
            "Unique residue counts compared; identity not fully verified"
        )

    else:

        cell8_comparison["status"] = (
            "Row count only; granularity is not confirmed"
        )

    print("\nCELL 8 COMPARISON")
    print("-" * 70)

    print(
        "CELL 8 rows:",
        len(df_hetero)
    )

    print(
        "CELL 17 component instances:",
        len(df_components)
    )

    print(
        "Status:",
        cell8_comparison["status"]
    )

else:

    print(
        "\nCELL 8 comparison skipped: df_hetero unavailable."
    )


# ------------------------------------------------------------
# 14. СОХРАНЕНИЕ CSV
# ------------------------------------------------------------

COMPONENTS_CSV_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_components.csv"
)

COMPONENT_ATOMS_CSV_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_component_atoms.csv"
)

COMPONENT_OCCURRENCES_CSV_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_component_occurrences.csv"
)

CHEM_COMP_CSV_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_chemical_component_dictionary.csv"
)

SELECTED_COMPONENT_ATOMS_CSV_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_selected_component_atoms.csv"
)


df_components.to_csv(
    COMPONENTS_CSV_PATH,
    index=False
)

df_component_atoms.to_csv(
    COMPONENT_ATOMS_CSV_PATH,
    index=False
)

df_component_occurrences.to_csv(
    COMPONENT_OCCURRENCES_CSV_PATH,
    index=False
)

df_chem_comp_dictionary.to_csv(
    CHEM_COMP_CSV_PATH,
    index=False
)

df_selected_atoms.to_csv(
    SELECTED_COMPONENT_ATOMS_CSV_PATH,
    index=False
)


# ------------------------------------------------------------
# 15. СОХРАНЕНИЕ JSON
# ------------------------------------------------------------

selection_info = {
    "PDB_ID": PDB_ID,

    "selection_status": SELECTION_STATUS,
    "selection_method": SELECTION_METHOD,
    "matching_candidate_count": SELECTION_CANDIDATE_COUNT,

    "selected_component_index": selected_component_index,
    "selected_component": selected_info,

    "preferred_component_name": PREFERRED_COMPONENT_NAME,
    "preferred_component_chain": PREFERRED_COMPONENT_CHAIN,
    "preferred_component_residue_number": (
        PREFERRED_COMPONENT_RESIDUE_NUMBER
    ),

    "selected_component_name": SELECTED_COMPONENT_NAME,
    "selected_component_chain": SELECTED_COMPONENT_CHAIN,

    "number_of_component_instances": number_of_component_instances,
    "number_of_unique_component_types": number_of_unique_component_types
}


component_info = {
    "PDB_ID": PDB_ID,

    "model_scope": {
        "total_models_in_file": len(cell17_models),
        "analyzed_model_index": MODEL_INDEX,
        "analyzed_model_id": model.id,
        "model_serial_number": MODEL_SERIAL_NUMBER
    },

    "components": components_info,

    "component_occurrences": (
        df_component_occurrences.to_dict(orient="records")
    ),

    "selection": selection_info,

    "cell8_comparison": cell8_comparison,

    "classification_method": (
        "Preliminary classification based on elemental "
        "composition and unique atom sites"
    ),

    "chemical_dictionary_source": (
        "mmCIF _chem_comp category"
    ),

    "atom_count_definition": {
        "number_of_atoms": (
            "Number of unique atom sites, identified by atom name and element"
        ),

        "number_of_atom_conformers": (
            "Number of atom records including alternate locations"
        ),

        "number_of_alternative_atom_records": (
            "Difference between conformer records and unique atom sites"
        )
    },

    "limitations": [
        "Only the first structural model is analyzed.",
        "Non-protein classification is based on Bio.PDB residue recognition.",
        "Unrecognized modified amino acids may be classified as non-protein candidates.",
        "Element-based classification does not establish biological function.",
        "No automatic fallback component selection is performed.",
        "A selected component is not automatically a drug-like ligand.",
        "The geometric center uses selected atom coordinates.",
        "Alternative residue variants are represented as separate component candidates."
    ]
}


component_info = json_safe(component_info)

COMPONENT_INFO_PATH = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_component_info.json"
)

with open(
    COMPONENT_INFO_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        component_info,
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


# ------------------------------------------------------------
# 16. ПРОВЕРКИ
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("VALIDATION")
print("=" * 70)


# Проверка количества атомных конформеров.

if not df_component_atoms.empty:

    atom_conformer_counts = (
        df_component_atoms
        .groupby("component_index")
        .size()
        .to_dict()
    )

else:

    atom_conformer_counts = {}


for _, row in df_components.iterrows():

    index = int(row["component_index"])

    actual_count = atom_conformer_counts.get(index, 0)

    expected_count = int(
        row["number_of_atom_conformers"]
    )

    if actual_count != expected_count:

        raise ValueError(
            f"Atom conformer count mismatch for component {index}: "
            f"{actual_count} != {expected_count}"
        )


print("Atom conformer counts: OK")


# Проверка уникальных атомных позиций.

for _, row in df_components.iterrows():

    index = int(row["component_index"])

    component_atom_rows = df_component_atoms[
        df_component_atoms["component_index"] == index
    ]

    actual_unique_sites = len({
        (
            str(atom_name).strip(),
            str(element).strip().upper()
        )
        for atom_name, element in zip(
            component_atom_rows["atom_name"],
            component_atom_rows["element"]
        )
    })

    expected_unique_sites = int(
        row["number_of_atoms"]
    )

    if actual_unique_sites != expected_unique_sites:

        raise ValueError(
            f"Unique atom site count mismatch for component {index}: "
            f"{actual_unique_sites} != {expected_unique_sites}"
        )


print("Unique atom site counts: OK")


# Проверка соответствия component_index и объекта Bio.PDB.

if selected_component is not None:

    if (
        component_objects.get(selected_component_index)
        is not selected_component
    ):

        raise ValueError(
            "Selected component object mapping mismatch."
        )

    if selected_info["component"] != SELECTED_COMPONENT_NAME:
        raise ValueError("Selected component name mismatch.")

    if selected_info["chain"] != SELECTED_COMPONENT_CHAIN:
        raise ValueError("Selected component chain mismatch.")

    if (
        len(df_selected_atoms)
        != selected_info["number_of_atom_conformers"]
    ):

        raise ValueError(
            "Selected component atom count mismatch."
        )

    print("Selected component object mapping: OK")

else:

    print(
        "Selected component: Not selected; "
        "downstream binding analysis must be skipped."
    )


# Проверка координат.

if not df_component_atoms.empty:

    coordinates = df_component_atoms[
        ["x", "y", "z"]
    ].to_numpy(dtype=float)

    if not np.isfinite(coordinates).all():

        raise ValueError(
            "Invalid component atom coordinates."
        )


print("Coordinates: OK")


# Проверка количества экземпляров.

if len(df_components) != len(component_objects):

    raise ValueError(
        "Component object mapping count mismatch."
    )


print("Component instance count: OK")


# Проверка уникальности component_index.

if (
    not df_components.empty
    and df_components["component_index"].duplicated().any()
):

    raise ValueError(
        "Duplicate component_index values detected."
    )


print("Component indices: OK")


# ------------------------------------------------------------
# 17. ПРОВЕРКА СОХРАНЕННЫХ ФАЙЛОВ
# ------------------------------------------------------------

files_to_check = [
    COMPONENTS_CSV_PATH,
    COMPONENT_ATOMS_CSV_PATH,
    COMPONENT_OCCURRENCES_CSV_PATH,
    CHEM_COMP_CSV_PATH,
    SELECTED_COMPONENT_ATOMS_CSV_PATH,
    COMPONENT_INFO_PATH
]

print("\nSaved files:")

for path in files_to_check:

    print(
        f"- {path.name}: "
        f"{'OK' if path.is_file() else 'ERROR'}"
    )


if not all(path.is_file() for path in files_to_check):

    raise RuntimeError(
        "Не все файлы анализа компонентов были сохранены."
    )


# ------------------------------------------------------------
# 18. ИТОГ
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("COMPONENT ANALYSIS COMPLETED")
print("=" * 70)

print("PDB ID:", PDB_ID)

print(
    "Analyzed model:",
    MODEL_INDEX,
    "| serial:",
    MODEL_SERIAL_NUMBER
)

print(
    "Non-protein component instances:",
    number_of_component_instances
)

print(
    "Unique component types:",
    number_of_unique_component_types
)

print("Water residues:", water_count)

print(
    "Standard amino acid residues excluded:",
    standard_amino_acid_count
)

print(
    "Modified amino acid residues excluded:",
    modified_amino_acid_count
)

print(
    "Component atom records:",
    len(df_component_atoms)
)

print("Selection status:", SELECTION_STATUS)

print(
    "Selected component:",
    (
        f"{SELECTED_COMPONENT_NAME}, "
        f"chain {normalize_chain_id(SELECTED_COMPONENT_CHAIN)}"
        if selected_component is not None
        else "None"
    )
)

print("Selection method:", SELECTION_METHOD)

print("\nOutput directory:", PDB_RESULTS_DIR)

print("\nSTATUS: OK")
print("CELL 17 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# CELL 18. BINDING SITE ANALYSIS
# ============================================================
# Назначение:
# 1. Найти аминокислотные остатки рядом с выбранным компонентом.
# 2. Рассчитать минимальные расстояния между атомами.
# 3. Определить количество атомных пар в пределах cutoff.
# 4. Сформировать таблицы области пространственных контактов.
# 5. Выполнить статистику по белковым цепям.
# 6. Сохранить результаты в CSV и JSON.
#
# Пакетная обработка:
# - используется первая модель;
# - selected_component должен происходить из CELL 17;
# - при отсутствии выбранного компонента анализ пропускается;
# - пустые таблицы также сохраняются.
#
# Методика:
# - используются тяжелые атомы;
# - водород и дейтерий исключаются;
# - для DisorderedAtom используется выбранный Bio.PDB конформер;
# - контакт определяется как расстояние <= BINDING_CUTOFF;
# - контакт не является доказательством химической связи.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.PDB import is_aa


print("=" * 70)
print("CELL 18. BINDING SITE ANALYSIS")
print("=" * 70)


# ------------------------------------------------------------
# 1. ПРОВЕРКА ВХОДНЫХ ДАННЫХ
# ------------------------------------------------------------

required_variables = [
    "PDB_ID",
    "PDB_RESULTS_DIR",
    "selected_component",
    "structure"
]

for variable in required_variables:

    if variable not in globals():
        raise NameError(
            f"Переменная {variable} не определена. "
            "Проверьте выполнение предыдущих ячеек."
        )


PDB_ID = str(PDB_ID).strip().upper()

if len(PDB_ID) != 4 or not PDB_ID.isalnum():
    raise ValueError(f"Некорректный PDB_ID: {PDB_ID}")


PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ------------------------------------------------------------
# 2. ПРОВЕРКА ЕДИНСТВА ОБЪЕКТА СТРУКТУРЫ
# ------------------------------------------------------------

cell18_models = list(structure.get_models())

if not cell18_models:
    raise ValueError(
        f"В структуре {PDB_ID} отсутствуют модели."
    )

first_model = cell18_models[0]

selected_component_available = (
    selected_component is not None
)

if selected_component_available:

    selected_component_chain = (
        selected_component.get_parent()
    )

    selected_model = (
        selected_component_chain.get_parent()
    )

    # CELL 17 и CELL 18 должны использовать один объект.
    if selected_model is not first_model:

        raise ValueError(
            "Выбранный компонент принадлежит другому объекту "
            "структуры. Повторно выполните CELL 17 после CELL 4."
        )

    structure_id = str(
        getattr(structure, "id", "")
    ).strip().upper()

    if structure_id and structure_id != PDB_ID:
        raise ValueError(
            f"Structure ID mismatch: {structure_id} != {PDB_ID}"
        )


# ------------------------------------------------------------
# 3. ПАРАМЕТРЫ АНАЛИЗА
# ------------------------------------------------------------

BINDING_CUTOFF = 5.0  # Å

HYDROGEN_ELEMENTS = {
    "H",
    "D"
}

ALTLOC_POLICY = (
    "Bio.PDB selected conformer; "
    "alternative conformers are not counted simultaneously"
)


# ------------------------------------------------------------
# 4. ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ
# ------------------------------------------------------------

def get_heavy_atoms(residue):
    """
    Возвращает тяжелые атомы остатка.

    Используется get_atoms(), а не get_unpacked_list(),
    чтобы не учитывать альтернативные конформации
    одновременно.
    """

    result = []

    for atom in residue.get_atoms():

        element = str(
            atom.element
        ).strip().upper()

        if element in HYDROGEN_ELEMENTS:
            continue

        result.append(atom)

    return result


def get_atom_element(atom):

    element = str(
        atom.element
    ).strip().upper()

    return element if element else "UNKNOWN"


def get_atom_altloc(atom):

    altloc = atom.get_altloc()

    if altloc is None:
        return ""

    return str(altloc).strip()


def normalize_chain_id(chain_id):

    value = str(chain_id)

    if not value.strip():
        return "<blank>"

    return value


def get_residue_key(chain, residue):

    """
    Формирует идентификатор остатка:
    цепь + hetero flag + номер + insertion code.
    """

    hetero_flag, residue_number, insertion_code = residue.id

    return (
        f"{chain.id}:"
        f"{hetero_flag}:"
        f"{residue_number}:"
        f"{str(insertion_code).strip()}"
    )


def json_safe(value):

    """Преобразует значения NumPy/Pandas в JSON-совместимый формат."""

    if isinstance(value, dict):

        return {
            str(key): json_safe(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple)):

        return [
            json_safe(item)
            for item in value
        ]

    if isinstance(value, np.ndarray):

        return json_safe(value.tolist())

    if isinstance(value, np.integer):

        return int(value)

    if isinstance(value, np.floating):

        value = float(value)

        return value if np.isfinite(value) else None

    if isinstance(value, float):

        return value if np.isfinite(value) else None

    if isinstance(value, np.bool_):

        return bool(value)

    if value is None:
        return None

    if isinstance(value, (str, int, bool)):
        return value

    try:

        if pd.isna(value):
            return None

    except (TypeError, ValueError):

        pass

    return str(value)


# ------------------------------------------------------------
# 5. ИНИЦИАЛИЗАЦИЯ
# ------------------------------------------------------------

binding_data = []
atom_contact_data = []

protein_residue_count = 0
checked_residue_count = 0
invalid_residue_count = 0
residues_without_heavy_atoms = 0

selected_component_name = None
selected_chain = None
selected_model_id = None
selected_model_serial = None
selected_component_number = None
selected_component_index_value = None
selected_component_id = None

component_atoms = []
component_coordinates = np.empty((0, 3), dtype=float)


# ------------------------------------------------------------
# 6. ПОДГОТОВКА ВЫБРАННОГО КОМПОНЕНТА
# ------------------------------------------------------------

if selected_component_available:

    selected_component_chain = (
        selected_component.get_parent()
    )

    selected_model = (
        selected_component_chain.get_parent()
    )

    # Проверка принадлежности первой модели.
    if selected_model is not first_model:
        raise ValueError(
            "Selected component does not belong to first model."
        )

    # Проверка, что компонент присутствует в первой модели.
    component_found = any(
        residue is selected_component
        for chain in first_model
        for residue in chain
    )

    if not component_found:
        raise ValueError(
            "Выбранный компонент не найден "
            "в первой модели текущей структуры."
        )

    selected_component_name = (
        selected_component.resname.strip().upper()
    )

    selected_chain = str(
        selected_component_chain.id
    )

    selected_model_id = selected_model.id

    selected_model_serial = getattr(
        selected_model,
        "serial_num",
        None
    )

    selected_residue_id = selected_component.id

    selected_component_number = (
        selected_residue_id[1]
    )

    selected_component_index_value = globals().get(
        "selected_component_index",
        None
    )

    if selected_component_index_value is not None:
        selected_component_index_value = int(
            selected_component_index_value
        )

    # Берем исходный идентификатор из CELL 17.
    # Не формируем его повторно по другой схеме.
    if (
        "selected_info" in globals()
        and isinstance(selected_info, dict)
    ):

        selected_component_id = selected_info.get(
            "component_id"
        )

    if not selected_component_id:

        raise ValueError(
            "Не удалось получить component_id из CELL 17. "
            "Повторно выполните CELL 17."
        )

    # Получаем тяжелые атомы выбранного компонента.
    component_atoms = get_heavy_atoms(
        selected_component
    )

    if not component_atoms:
        raise ValueError(
            "Выбранный компонент не содержит тяжелых атомов."
        )

    component_coordinates = np.asarray(
        [
            atom.coord
            for atom in component_atoms
        ],
        dtype=float
    )

    if (
        component_coordinates.ndim != 2
        or component_coordinates.shape[1] != 3
    ):
        raise ValueError(
            "Некорректная размерность координат компонента."
        )

    if not np.isfinite(component_coordinates).all():
        raise ValueError(
            "В координатах выбранного компонента "
            "обнаружены NaN или Inf."
        )

    print("PDB ID:", PDB_ID)

    print(
        "Selected component:",
        selected_component_name
    )

    print(
        "Component chain:",
        normalize_chain_id(selected_chain)
    )

    print(
        "Component residue number:",
        selected_component_number
    )

    print(
        "Model ID:",
        selected_model_id
    )

    print(
        "Component heavy atoms:",
        len(component_atoms)
    )

    print(
        "Search radius:",
        BINDING_CUTOFF,
        "Å"
    )

else:

    selected_model = None

    print("PDB ID:", PDB_ID)
    print("No component selected in CELL 17.")
    print("Binding site analysis will be skipped.")


# ------------------------------------------------------------
# 7. ПОИСК БЕЛКОВЫХ ОСТАТКОВ ВОКРУГ КОМПОНЕНТА
# ------------------------------------------------------------

if selected_component_available:

    for chain in selected_model:

        for residue in chain:

            if not is_aa(residue, standard=False):
                continue

            # Исключаем сам выбранный компонент,
            # если он по какой-либо причине распознан как аминокислота.
            if residue is selected_component:
                continue

            protein_residue_count += 1

            protein_atoms = get_heavy_atoms(residue)

            if not protein_atoms:

                residues_without_heavy_atoms += 1

                continue

            protein_coordinates = np.asarray(
                [
                    atom.coord
                    for atom in protein_atoms
                ],
                dtype=float
            )

            checked_residue_count += 1

            if (
                protein_coordinates.ndim != 2
                or protein_coordinates.shape[1] != 3
                or not np.isfinite(protein_coordinates).all()
            ):

                invalid_residue_count += 1

                continue

            # Матрица расстояний:
            # строки — атомы белкового остатка;
            # столбцы — атомы компонента.
            distances = np.linalg.norm(
                protein_coordinates[:, None, :]
                -
                component_coordinates[None, :, :],
                axis=2
            )

            if not np.isfinite(distances).all():

                invalid_residue_count += 1

                continue

            min_flat_index = int(
                np.argmin(distances)
            )

            protein_atom_index, component_atom_index = (
                np.unravel_index(
                    min_flat_index,
                    distances.shape
                )
            )

            min_distance = float(
                distances[
                    protein_atom_index,
                    component_atom_index
                ]
            )

            contact_mask = (
                distances <= BINDING_CUTOFF
            )

            # Число атомных пар, а не уникальных взаимодействий.
            atom_contacts = int(
                contact_mask.sum()
            )

            protein_contact_atoms = int(
                np.any(
                    contact_mask,
                    axis=1
                ).sum()
            )

            component_contact_atoms = int(
                np.any(
                    contact_mask,
                    axis=0
                ).sum()
            )

            if min_distance > BINDING_CUTOFF:
                continue

            closest_protein_atom = protein_atoms[
                protein_atom_index
            ]

            closest_component_atom = component_atoms[
                component_atom_index
            ]

            hetero_flag, residue_number, insertion_code = (
                residue.id
            )

            insertion_code = str(
                insertion_code
            ).strip()

            residue_name = (
                residue.resname.strip().upper()
            )

            residue_key = get_residue_key(
                chain,
                residue
            )

            chain_id_display = normalize_chain_id(
                chain.id
            )

            # ------------------------------------------------
            # 7.1. Информация об остатке
            # ------------------------------------------------

            binding_data.append({

                "pdb_id": PDB_ID,

                "component_id": selected_component_id,

                "component_index": (
                    selected_component_index_value
                ),

                "component": selected_component_name,

                "component_chain": selected_chain,

                "component_residue_number": (
                    selected_component_number
                ),

                "model_id": selected_model_id,

                "protein_chain": str(chain.id),

                "protein_chain_display": chain_id_display,

                "residue_key": residue_key,

                "hetero_flag": str(hetero_flag),

                "residue_name": residue_name,

                "residue_number": residue_number,

                "insertion_code": insertion_code,

                "min_distance_A": min_distance,

                "atom_contacts": atom_contacts,

                "protein_contact_atoms": (
                    protein_contact_atoms
                ),

                "component_contact_atoms": (
                    component_contact_atoms
                ),

                "closest_protein_atom": (
                    closest_protein_atom.get_name()
                ),

                "closest_protein_element": (
                    get_atom_element(closest_protein_atom)
                ),

                "closest_protein_altloc": (
                    get_atom_altloc(closest_protein_atom)
                ),

                "closest_component_atom": (
                    closest_component_atom.get_name()
                ),

                "closest_component_element": (
                    get_atom_element(closest_component_atom)
                ),

                "closest_component_altloc": (
                    get_atom_altloc(closest_component_atom)
                )

            })

            # ------------------------------------------------
            # 7.2. Информация об атомных парах
            # ------------------------------------------------

            contact_indices = np.argwhere(
                contact_mask
            )

            for protein_index, component_index in contact_indices:

                protein_atom = protein_atoms[
                    protein_index
                ]

                component_atom = component_atoms[
                    component_index
                ]

                component_atom_id = (
                    f"{selected_component_id}|"
                    f"{component_atom.get_name().strip()}|"
                    f"{get_atom_element(component_atom)}|"
                    f"{get_atom_altloc(component_atom)}"
                )

                atom_contact_data.append({

                    "pdb_id": PDB_ID,

                    "component_id": selected_component_id,

                    "component_index": (
                        selected_component_index_value
                    ),

                    "component": selected_component_name,

                    "component_chain": selected_chain,

                    "component_residue_number": (
                        selected_component_number
                    ),

                    "model_id": selected_model_id,

                    "protein_chain": str(chain.id),

                    "protein_chain_display": chain_id_display,

                    "residue_name": residue_name,

                    "residue_number": residue_number,

                    "insertion_code": insertion_code,

                    "residue_key": residue_key,

                    "protein_atom": (
                        protein_atom.get_name().strip()
                    ),

                    "protein_element": (
                        get_atom_element(protein_atom)
                    ),

                    "protein_altloc": (
                        get_atom_altloc(protein_atom)
                    ),

                    "component_atom": (
                        component_atom.get_name().strip()
                    ),

                    "component_element": (
                        get_atom_element(component_atom)
                    ),

                    "component_altloc": (
                        get_atom_altloc(component_atom)
                    ),

                    "component_atom_id": component_atom_id,

                    "distance_A": float(
                        distances[
                            protein_index,
                            component_index
                        ]
                    )

                })


# ------------------------------------------------------------
# 8. СОЗДАНИЕ ТАБЛИЦ РЕЗУЛЬТАТОВ
# ------------------------------------------------------------

binding_columns = [

    "pdb_id",
    "component_id",
    "component_index",
    "component",
    "component_chain",
    "component_residue_number",
    "model_id",
    "protein_chain",
    "protein_chain_display",
    "residue_key",
    "hetero_flag",
    "residue_name",
    "residue_number",
    "insertion_code",
    "min_distance_A",
    "atom_contacts",
    "protein_contact_atoms",
    "component_contact_atoms",
    "closest_protein_atom",
    "closest_protein_element",
    "closest_protein_altloc",
    "closest_component_atom",
    "closest_component_element",
    "closest_component_altloc"

]

atom_contact_columns = [

    "pdb_id",
    "component_id",
    "component_index",
    "component",
    "component_chain",
    "component_residue_number",
    "model_id",
    "protein_chain",
    "protein_chain_display",
    "residue_name",
    "residue_number",
    "insertion_code",
    "residue_key",
    "protein_atom",
    "protein_element",
    "protein_altloc",
    "component_atom",
    "component_element",
    "component_altloc",
    "component_atom_id",
    "distance_A"

]


df_binding = pd.DataFrame(
    binding_data,
    columns=binding_columns
)

df_binding_atom_contacts = pd.DataFrame(
    atom_contact_data,
    columns=atom_contact_columns
)


if not df_binding.empty:

    df_binding = (
        df_binding
        .sort_values(
            [
                "min_distance_A",
                "protein_chain",
                "residue_number"
            ]
        )
        .reset_index(drop=True)
    )


if not df_binding_atom_contacts.empty:

    df_binding_atom_contacts = (
        df_binding_atom_contacts
        .sort_values(
            [
                "distance_A",
                "protein_chain",
                "residue_number"
            ]
        )
        .reset_index(drop=True)
    )


print("\n" + "=" * 70)
print("BINDING SITE RESULTS")
print("=" * 70)


if not df_binding.empty:

    print(
        "Nearby amino acid residues:",
        len(df_binding)
    )

    print(
        "Atom-pair contacts:",
        len(df_binding_atom_contacts)
    )

    print("\nNearby protein residues:")
    display(df_binding)

    print("\nIndividual atom contacts:")
    display(
        df_binding_atom_contacts.head(30)
    )

else:

    print(
        "No amino acid residues found "
        "within the specified radius."
    )


# ------------------------------------------------------------
# 9. СВОДКА ПО БЕЛКОВЫМ ЦЕПЯМ
# ------------------------------------------------------------

chain_columns = [

    "protein_chain",
    "protein_chain_display",
    "number_of_residues",
    "minimum_distance_A",
    "maximum_residue_min_distance_A",
    "total_atom_contacts",
    "protein_contact_atoms",
    "unique_component_contact_atoms"

]


chain_records = []

if not df_binding.empty:

    for chain_id, chain_group in df_binding.groupby(
        "protein_chain",
        dropna=False
    ):

        chain_contacts = df_binding_atom_contacts[
            df_binding_atom_contacts["protein_chain"]
            == chain_id
        ]

        unique_component_atoms = (
            chain_contacts["component_atom_id"].nunique()
            if not chain_contacts.empty
            else 0
        )

        # Уникальные атомы белка в пределах цепи.
        unique_protein_atoms = (
            chain_contacts[
                [
                    "residue_key",
                    "protein_atom",
                    "protein_element",
                    "protein_altloc"
                ]
            ]
            .drop_duplicates()
            .shape[0]
        )

        chain_records.append({

            "protein_chain": chain_id,

            "protein_chain_display": (
                chain_group["protein_chain_display"].iloc[0]
            ),

            "number_of_residues": len(chain_group),

            "minimum_distance_A": float(
                chain_group["min_distance_A"].min()
            ),

            "maximum_residue_min_distance_A": float(
                chain_group["min_distance_A"].max()
            ),

            "total_atom_contacts": int(
                chain_group["atom_contacts"].sum()
            ),

            "protein_contact_atoms": int(
                unique_protein_atoms
            ),

            "unique_component_contact_atoms": int(
                unique_component_atoms
            )

        })


df_binding_by_chain = pd.DataFrame(
    chain_records,
    columns=chain_columns
)


print("\nBinding site summary by chain:")

if not df_binding_by_chain.empty:
    display(df_binding_by_chain)
else:
    print("No contacts to summarize.")


# ------------------------------------------------------------
# 10. ОБЩАЯ СТАТИСТИКА
# ------------------------------------------------------------

if not df_binding.empty:

    minimum_distance = float(
        df_binding["min_distance_A"].min()
    )

    maximum_residue_min_distance = float(
        df_binding["min_distance_A"].max()
    )

    total_atom_contacts = int(
        df_binding["atom_contacts"].sum()
    )

    average_contacts = float(
        df_binding["atom_contacts"].mean()
    )

    unique_protein_contact_atoms = int(

        df_binding_atom_contacts[

            [
                "protein_chain",
                "residue_key",
                "protein_atom",
                "protein_element",
                "protein_altloc"
            ]

        ]
        .drop_duplicates()
        .shape[0]

    )

    unique_component_contact_atoms = int(
        df_binding_atom_contacts[
            "component_atom_id"
        ].nunique()
    )

else:

    minimum_distance = None

    maximum_residue_min_distance = None

    total_atom_contacts = 0

    average_contacts = None

    unique_protein_contact_atoms = 0

    unique_component_contact_atoms = 0


analysis_status = (
    "COMPLETED"
    if selected_component_available
    else "SKIPPED_NO_COMPONENT"
)


binding_summary = {

    "PDB_ID": PDB_ID,

    "analysis_status": analysis_status,

    "component": selected_component_name,

    "component_id": selected_component_id,

    "component_index": selected_component_index_value,

    "component_chain": selected_chain,

    "component_residue_number": selected_component_number,

    "model_id": selected_model_id,

    "model_serial_number": selected_model_serial,

    "binding_cutoff_A": BINDING_CUTOFF,

    "hydrogen_atoms_excluded": True,

    "altloc_policy": ALTLOC_POLICY,

    "protein_residue_count": protein_residue_count,

    "checked_residue_count": checked_residue_count,

    "residues_without_heavy_atoms": (
        residues_without_heavy_atoms
    ),

    "invalid_residue_count": invalid_residue_count,

    "nearby_residue_count": len(df_binding),

    "atom_pair_contact_count": total_atom_contacts,

    "unique_protein_contact_atom_count": (
        unique_protein_contact_atoms
    ),

    "unique_component_contact_atom_count": (
        unique_component_contact_atoms
    ),

    "minimum_distance_A": minimum_distance,

    "maximum_residue_min_distance_A": (
        maximum_residue_min_distance
    ),

    "average_contacts_per_residue": average_contacts,

    "analysis_scope": (
        "Protein amino acid residues in the first model "
        "around the selected component"
    ),

    "distance_interpretation": (
        "maximum_residue_min_distance_A is the largest "
        "of the per-residue minimum distances; it is not "
        "the maximum distance among all atom pairs"
    ),

    "contact_count_interpretation": (
        "atom_pair_contact_count is the number of atom pairs "
        "within the cutoff, not the number of chemical bonds"
    ),

    "interpretation": (
        "Spatial contacts within the specified cutoff; "
        "not proof of chemical bonding"
    )

}


print("\nOVERALL STATISTICS")
print("-" * 70)

print("Analysis status:", analysis_status)

print(
    "Protein residues in model:",
    protein_residue_count
)

print(
    "Checked protein residues:",
    checked_residue_count
)

print(
    "Residues without heavy atoms:",
    residues_without_heavy_atoms
)

print(
    "Residues with invalid coordinates:",
    invalid_residue_count
)

print(
    "Nearby amino acid residues:",
    len(df_binding)
)

print(
    "Total atom-pair contacts:",
    total_atom_contacts
)

print(
    "Unique protein contact atoms:",
    unique_protein_contact_atoms
)

print(
    "Unique component contact atoms:",
    unique_component_contact_atoms
)

if minimum_distance is not None:

    print(
        "Minimum distance:",
        round(minimum_distance, 3),
        "Å"
    )

    print(
        "Maximum residue minimum distance:",
        round(maximum_residue_min_distance, 3),
        "Å"
    )

    print(
        "Average contacts per residue:",
        round(average_contacts, 2)
    )


# ------------------------------------------------------------
# 11. СОХРАНЕНИЕ CSV
# ------------------------------------------------------------

BINDING_FILE = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_binding_site.csv"
)

BINDING_ATOM_CONTACTS_FILE = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_binding_atom_contacts.csv"
)

BINDING_SUMMARY_FILE = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_binding_site_summary.csv"
)


df_binding.to_csv(
    BINDING_FILE,
    index=False
)

df_binding_atom_contacts.to_csv(
    BINDING_ATOM_CONTACTS_FILE,
    index=False
)

df_binding_by_chain.to_csv(
    BINDING_SUMMARY_FILE,
    index=False
)


# ------------------------------------------------------------
# 12. СОХРАНЕНИЕ JSON
# ------------------------------------------------------------

BINDING_JSON_FILE = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_binding_site.json"
)


with open(
    BINDING_JSON_FILE,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_safe(binding_summary),
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


# ------------------------------------------------------------
# 13. ПРОВЕРКИ
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("VALIDATION")
print("=" * 70)


# 13.1. Проверка расстояний.
if not df_binding_atom_contacts.empty:

    if not (
        df_binding_atom_contacts["distance_A"]
        <= BINDING_CUTOFF
    ).all():

        raise ValueError(
            "Some atom contacts exceed the specified cutoff."
        )

    if (
        df_binding_atom_contacts["distance_A"] < 0
    ).any():

        raise ValueError(
            "Negative interatomic distance detected."
        )


print("Contact distance threshold: OK")


# 13.2. Проверка количества атомных пар.
if (
    len(df_binding_atom_contacts)
    != total_atom_contacts
):

    raise ValueError(
        "Mismatch between atom contact records "
        "and calculated contact count."
    )


print("Atom-pair contact count: OK")


# 13.3. Проверка количества остатков.
if len(df_binding) != len(binding_data):

    raise ValueError(
        "Mismatch in binding residue count."
    )


print("Binding residue count: OK")


# 13.4. Проверка уникальных атомов компонента.
if not df_binding_atom_contacts.empty:

    calculated_unique_component_atoms = int(
        df_binding_atom_contacts[
            "component_atom_id"
        ].nunique()
    )

    if (
        calculated_unique_component_atoms
        != unique_component_contact_atoms
    ):

        raise ValueError(
            "Mismatch in unique component contact atom count."
        )


print("Unique component contact atoms: OK")


# 13.5. Проверка согласованности таблиц.
if not df_binding_atom_contacts.empty:

    contact_residues = set(
        df_binding_atom_contacts["residue_key"]
    )

    binding_residues = set(
        df_binding["residue_key"]
    )

    if not contact_residues.issubset(binding_residues):

        raise ValueError(
            "Some atom contacts refer to residues "
            "missing from the binding site table."
        )


print("Contact-to-residue consistency: OK")


# 13.6. Проверка идентификатора компонента.
if selected_component_available:

    if (
        not df_binding.empty
        and not (
            df_binding["component_id"] == selected_component_id
        ).all()
    ):

        raise ValueError(
            "Component ID mismatch in binding table."
        )


print("Component ID consistency: OK")


# 13.7. Проверка выходных файлов.
saved_files = [

    BINDING_FILE,

    BINDING_ATOM_CONTACTS_FILE,

    BINDING_SUMMARY_FILE,

    BINDING_JSON_FILE

]


for path in saved_files:

    if not path.is_file():

        raise RuntimeError(
            f"Output file was not saved: {path}"
        )

    if path.stat().st_size == 0:

        raise RuntimeError(
            f"Output file is empty: {path}"
        )


print("Output files: OK")


# ------------------------------------------------------------
# 14. ИТОГ
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("BINDING SITE ANALYSIS COMPLETED")
print("=" * 70)

print("PDB ID:", PDB_ID)

print("Analysis status:", analysis_status)

print(
    "Selected component:",
    (
        selected_component_name
        if selected_component_available
        else "None"
    )
)

print(
    "Component chain:",
    (
        normalize_chain_id(selected_chain)
        if selected_component_available
        else "None"
    )
)

print(
    "Binding cutoff:",
    BINDING_CUTOFF,
    "Å"
)

print("Nearby residues:", len(df_binding))

print("Atom-pair contacts:", total_atom_contacts)

print(
    "Unique protein contact atoms:",
    unique_protein_contact_atoms
)

print(
    "Unique component contact atoms:",
    unique_component_contact_atoms
)

print("\nSaved files:")

for path in saved_files:

    print(
        f"- {path.name}: "
        f"{'OK' if path.is_file() else 'ERROR'}"
    )


print("\nOutput directory:", PDB_RESULTS_DIR)

print("\nSTATUS:", analysis_status)

print("CELL 18 COMPLETED")
print("=" * 70)
```

Результат ячейки: таблица ближайших аминокислотных остатков с указанием цепи, номера остатка, минимального расстояния до гема и числа атомных контактов в пределах 5 Å.
3. Что означают новые показатели
Показатель	Значение
min_distance_A	Минимальное расстояние между атомами остатка и компонента
atom_contacts	Количество пар атомов на расстоянии не более 5 Å
protein_contact_atoms	Количество уникальных атомов белка, участвующих в контактах
component_contact_atoms	Количество уникальных атомов компонента, участвующих в контактах
closest_protein_atom	Название ближайшего атома белка
closest_component_atom	Название ближайшего атома компонента

4. Важное методическое уточнение
Радиус 5 Å — это критерий поиска соседних остатков, а не доказательство химического взаимодействия.

Например:
- расстояние 3,0 Å может соответствовать потенциальной водородной связи, но нужно проверить типы атомов и геометрию;
- расстояние 4,5 Å указывает на пространственную близость, но само по себе не доказывает наличие конкретного взаимодействия;
- контакты с металлами требуют отдельного анализа координационного окружения.
Также новая версия анализирует только модель, в которой находится выбранный компонент. Это важно для структур с несколькими моделями.

```python
# ============================================================
# CELL 19. PROTEIN-COMPONENT INTERACTION ANALYSIS
# ============================================================
# Назначение:
# 1. Найти атомные контакты выбранного компонента с белком.
# 2. Выделить потенциальные контакты полярных атомов.
# 3. Выделить потенциальные гидрофобные контакты.
# 4. Найти геометрические контакты Fe в гемах.
# 5. Выполнить универсальный поиск контактов распознаваемых
#    металлов в небелковых компонентах.
# 6. Найти контакты Fe с компонентами OXY.
# 7. Найти потенциальные дисульфидные мостики CYS-SG-SG.
# 8. Сохранить результаты в CSV и JSON.
#
# Согласованность:
# - используется исходный объект structure из CELL 4;
# - анализируется первая модель;
# - selected_component должен принадлежать этой модели;
# - используется выбранный Bio.PDB конформер;
# - водородные атомы исключаются;
# - результаты сохраняются в PDB_RESULTS_DIR.
#
# ВАЖНО:
# Геометрические контакты не являются доказательством
# химической связи, водородной связи или координации металла.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.PDB import is_aa


print("=" * 70)
print("CELL 19. PROTEIN-COMPONENT INTERACTION ANALYSIS")
print("=" * 70)


# ============================================================
# 1. PARAMETERS
# ============================================================

CONTACT_CUTOFF = 4.0
POLAR_CUTOFF = 3.5
HYDROPHOBIC_CUTOFF = 4.0

METAL_COORDINATION_CUTOFF = 2.8
DISULFIDE_CUTOFF = 2.3

METAL_ELEMENTS = {
    "FE", "ZN", "MG", "MN", "CU",
    "CO", "NI", "CA", "CD", "HG"
}

POLAR_ELEMENTS = {"N", "O", "S"}

# Для более консервативного выделения потенциальных
# гидрофобных контактов используется пара C-C.
HYDROPHOBIC_ELEMENTS = {"C"}

HYDROGEN_ELEMENTS = {"H", "D"}

WATER_CODES = {"HOH", "WAT", "DOD"}

HEME_COMPONENT_CODES = {
    "HEM", "HEA", "HEB", "HEC",
    "HEO", "HEF", "HEG", "HEV"
}

OXYGEN_COMPONENT_CODES = {
    "OXY", "O2"
}


# ============================================================
# 2. INPUT VALIDATION
# ============================================================

required_variables = [
    "PDB_ID",
    "PDB_RESULTS_DIR",
    "structure"
]

missing_variables = [
    name
    for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise NameError(
        "Не найдены обязательные переменные: "
        + ", ".join(missing_variables)
    )


PDB_ID = str(PDB_ID).strip().upper()

if len(PDB_ID) != 4 or not PDB_ID.isalnum():
    raise ValueError(
        f"Некорректный PDB_ID: {PDB_ID}"
    )


PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


if structure is None:
    raise ValueError(
        "Переменная structure не содержит структуру."
    )


structure_id = str(
    getattr(structure, "id", "")
).strip().upper()

if structure_id and structure_id != PDB_ID:
    raise ValueError(
        f"Несоответствие PDB ID: "
        f"structure.id={structure_id}, "
        f"PDB_ID={PDB_ID}"
    )


# Не перезаписываем глобальную переменную models
# из предыдущих ячеек.

cell19_models = list(
    structure.get_models()
)

if not cell19_models:
    raise ValueError(
        "В структуре не обнаружено ни одной модели."
    )


# Анализируется первая модель.

cell19_model = cell19_models[0]

cell19_model_id = cell19_model.id

cell19_model_serial = getattr(
    cell19_model,
    "serial_num",
    None
)


# Выбранный компонент формируется в CELL 17.
# Его отсутствие допустимо: остальные виды анализа
# всё равно будут выполнены.

selected_component = globals().get(
    "selected_component",
    None
)


# ============================================================
# 3. HELPER FUNCTIONS
# ============================================================

def get_atom_element(atom):
    """
    Возвращает химический элемент атома.

    Приоритет:
    1. atom.element из Bio.PDB.
    2. Осторожное определение по первой букве имени.

    Для mmCIF предпочтительно использовать atom.element.
    """

    element = getattr(
        atom,
        "element",
        None
    )

    if element is not None:

        element = str(element).strip().upper()

        if element not in {"", "?", ".", "X"}:
            return element

    atom_name = str(
        atom.get_fullname()
    ).strip().upper()

    letters = "".join(
        character
        for character in atom_name
        if character.isalpha()
    )

    if not letters:
        return "UNKNOWN"

    # Не определяем двухбуквенные элементы только по имени:
    # например, CA может быть атомом C-alpha, а не кальцием.
    return letters[0]


def get_residue_code(residue):

    return str(
        residue.get_resname()
    ).strip().upper()


def get_residue_number(residue):

    return int(residue.id[1])


def get_insertion_code(residue):

    code = residue.id[2]

    if code is None:
        return ""

    return str(code).strip()


def get_hetero_flag(residue):

    return str(
        residue.id[0]
    ).strip()


def get_chain_id(residue):

    # Пустой идентификатор цепи сохраняется как "".
    return str(
        residue.get_parent().id
    )


def get_model_id(residue):

    return residue.get_parent().get_parent().id


def get_residue_component_id(residue):
    """
    Устойчивый идентификатор остатка/компонента.

    Включает:
    - PDB ID;
    - модель;
    - цепь;
    - hetero-флаг;
    - номер остатка;
    - код вставки;
    - код компонента.
    """

    return (
        f"{PDB_ID}|"
        f"model_{get_model_id(residue)}|"
        f"chain_{get_chain_id(residue)}|"
        f"{get_hetero_flag(residue)}|"
        f"{get_residue_number(residue)}|"
        f"{get_insertion_code(residue)}|"
        f"{get_residue_code(residue)}"
    )


def get_atom_altloc(atom):

    altloc = atom.get_altloc()

    if altloc is None:
        return ""

    return str(altloc).strip()


def get_valid_atoms_and_coordinates(residue):
    """
    Возвращает тяжелые атомы с корректными координатами.

    get_atoms() использует выбранный Bio.PDB конформер
    для DisorderedAtom.
    """

    atoms = []
    coordinates = []

    for atom in residue.get_atoms():

        element = get_atom_element(atom)

        if element in HYDROGEN_ELEMENTS:
            continue

        coordinate = np.asarray(
            atom.coord,
            dtype=float
        )

        if coordinate.shape != (3,):
            continue

        if not np.isfinite(coordinate).all():
            continue

        atoms.append(atom)
        coordinates.append(coordinate)

    if not coordinates:
        return (
            [],
            np.empty((0, 3), dtype=float)
        )

    return (
        atoms,
        np.asarray(coordinates, dtype=float)
    )


def get_component_type(residue):
    """
    Предварительная классификация небелкового компонента.
    Не определяет биологическую функцию компонента.
    """

    code = get_residue_code(residue)

    if code in HEME_COMPONENT_CODES:
        return "HEME"

    if code in OXYGEN_COMPONENT_CODES:
        return "OXY"

    if code in WATER_CODES:
        return "WATER"

    atoms = list(residue.get_atoms())

    if any(
        get_atom_element(atom) in METAL_ELEMENTS
        for atom in atoms
    ):
        return "METAL_CONTAINING_COMPONENT"

    return "OTHER"


def is_protein_residue(residue):

    try:
        return bool(
            is_aa(
                residue,
                standard=False
            )
        )

    except Exception:
        return False


def is_water_residue(residue):

    return (
        get_residue_code(residue)
        in WATER_CODES
    )


def is_heme_residue(residue):
    """
    Распознает гемовый компонент, содержащий Fe.
    """

    if get_residue_code(residue) not in HEME_COMPONENT_CODES:
        return False

    return any(
        get_atom_element(atom) == "FE"
        for atom in residue.get_atoms()
    )


def get_metal_atoms(residue):

    return [
        atom
        for atom in residue.get_atoms()
        if get_atom_element(atom) in METAL_ELEMENTS
    ]


def get_atom_distance(atom_1, atom_2):

    coord_1 = np.asarray(
        atom_1.coord,
        dtype=float
    )

    coord_2 = np.asarray(
        atom_2.coord,
        dtype=float
    )

    if (
        coord_1.shape != (3,)
        or coord_2.shape != (3,)
        or not np.isfinite(coord_1).all()
        or not np.isfinite(coord_2).all()
    ):
        return np.nan

    return float(
        np.linalg.norm(coord_1 - coord_2)
    )


def get_atom_record(atom, residue):

    return {
        "model_id": get_model_id(residue),
        "chain": get_chain_id(residue),
        "residue": get_residue_code(residue),
        "residue_number": get_residue_number(residue),
        "insertion_code": get_insertion_code(residue),
        "hetero_flag": get_hetero_flag(residue),
        "atom": atom.get_name().strip(),
        "element": get_atom_element(atom),
        "altloc": get_atom_altloc(atom),
        "component_id": get_residue_component_id(residue)
    }


def json_safe(value):
    """
    Преобразует значения NumPy и pandas в типы,
    поддерживаемые стандартным JSON-сериализатором.
    """

    if isinstance(value, dict):
        return {
            str(key): json_safe(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple)):
        return [
            json_safe(item)
            for item in value
        ]

    if isinstance(value, np.ndarray):
        return json_safe(value.tolist())

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):

        value = float(value)

        return (
            value
            if np.isfinite(value)
            else None
        )

    if isinstance(value, float):
        return (
            value
            if np.isfinite(value)
            else None
        )

    if isinstance(value, np.bool_):
        return bool(value)

    if value is None:
        return None

    if isinstance(value, (str, int, bool)):
        return value

    try:
        missing = pd.isna(value)

        if isinstance(missing, (bool, np.bool_)) and missing:
            return None

    except (TypeError, ValueError):
        pass

    return str(value)


def build_distance_summary(
    dataframe,
    group_columns,
    count_column,
    output_columns
):
    """
    Универсальное построение сводной таблицы расстояний.
    """

    if dataframe.empty:
        return pd.DataFrame(
            columns=output_columns
        )

    result = (
        dataframe
        .groupby(
            group_columns,
            dropna=False
        )
        .agg(
            contact_count=("distance", "count"),
            mean_distance=("distance", "mean"),
            min_distance=("distance", "min"),
            max_distance=("distance", "max")
        )
        .reset_index()
    )

    if count_column != "contact_count":
        result = result.rename(
            columns={
                "contact_count": count_column
            }
        )

    return result


# ============================================================
# 4. STRUCTURE COLLECTION
# ============================================================

all_residues = [
    residue
    for chain in cell19_model
    for residue in chain
]


protein_residues = [
    residue
    for residue in all_residues
    if is_protein_residue(residue)
]


water_residues = [
    residue
    for residue in all_residues
    if is_water_residue(residue)
]


nonprotein_residues = [
    residue
    for residue in all_residues
    if (
        not is_protein_residue(residue)
        and not is_water_residue(residue)
    )
]


heme_residues = [
    residue
    for residue in nonprotein_residues
    if is_heme_residue(residue)
]


oxy_residues = [
    residue
    for residue in nonprotein_residues
    if get_residue_code(residue) in OXYGEN_COMPONENT_CODES
]


# ============================================================
# 5. SELECTED COMPONENT VALIDATION
# ============================================================

selected_residue = selected_component

selected_atoms = []

selected_coordinates = np.empty(
    (0, 3),
    dtype=float
)

selected_component_code = None
selected_component_id = None
selected_component_chain = None
selected_component_number = None
selected_component_model_id = None


if selected_residue is not None:

    if not hasattr(selected_residue, "get_atoms"):
        raise TypeError(
            "selected_component должен быть объектом "
            "Residue из Bio.PDB."
        )

    # Проверяем, что компонент принадлежит именно первой
    # модели исходного объекта structure из CELL 4.
    component_found = any(
        residue is selected_residue
        for residue in all_residues
    )

    if not component_found:
        raise ValueError(
            "selected_component не принадлежит первой модели "
            "объекта structure из CELL 4. Проверьте CELL 17: "
            "выбранный компонент и анализируемая структура "
            "должны происходить из одного объекта Bio.PDB."
        )

    if is_protein_residue(selected_residue):
        raise ValueError(
            "selected_component определен как белковый остаток. "
            "Для этой ячейки требуется небелковый компонент."
        )

    if is_water_residue(selected_residue):
        raise ValueError(
            "Вода не может использоваться как выбранный "
            "компонент в данном анализе."
        )

    selected_component_code = get_residue_code(
        selected_residue
    )

    selected_component_id = get_residue_component_id(
        selected_residue
    )

    selected_component_chain = get_chain_id(
        selected_residue
    )

    selected_component_number = get_residue_number(
        selected_residue
    )

    selected_component_model_id = get_model_id(
        selected_residue
    )

    selected_atoms, selected_coordinates = (
        get_valid_atoms_and_coordinates(
            selected_residue
        )
    )

    if not selected_atoms:
        raise ValueError(
            "Выбранный компонент не содержит тяжелых атомов "
            "с корректными координатами."
        )


# ============================================================
# 6. STRUCTURE INFORMATION
# ============================================================

print("\nSTRUCTURE INFORMATION")

print("PDB ID:", PDB_ID)
print("Model count in source:", len(cell19_models))
print("Analyzed model ID:", cell19_model_id)
print("Protein residues:", len(protein_residues))
print("Nonprotein components:", len(nonprotein_residues))
print("Water residues:", len(water_residues))
print("HEME components:", len(heme_residues))
print("OXY components:", len(oxy_residues))
print("Selected component:", selected_component_code)


# ============================================================
# 7. SELECTED COMPONENT - PROTEIN INTERACTIONS
# ============================================================

interaction_data = []


if selected_residue is not None:

    for protein_residue in protein_residues:

        protein_atoms, protein_coordinates = (
            get_valid_atoms_and_coordinates(
                protein_residue
            )
        )

        if not protein_atoms:
            continue

        # Матрица расстояний:
        # строки — атомы выбранного компонента;
        # столбцы — атомы белкового остатка.

        distances = np.linalg.norm(
            selected_coordinates[:, None, :]
            -
            protein_coordinates[None, :, :],
            axis=2
        )

        component_indices, protein_indices = np.where(
            distances <= CONTACT_CUTOFF
        )

        for component_index, protein_index in zip(
            component_indices,
            protein_indices
        ):

            component_atom = selected_atoms[
                component_index
            ]

            protein_atom = protein_atoms[
                protein_index
            ]

            component_element = get_atom_element(
                component_atom
            )

            protein_element = get_atom_element(
                protein_atom
            )

            distance = float(
                distances[
                    component_index,
                    protein_index
                ]
            )

            # Это предварительная геометрическая классификация.
            # Она не устанавливает химическую природу взаимодействия.

            if (
                component_element in POLAR_ELEMENTS
                and protein_element in POLAR_ELEMENTS
                and distance <= POLAR_CUTOFF
            ):

                interaction_type = (
                    "Potential polar-atom proximity"
                )

            elif (
                component_element in HYDROPHOBIC_ELEMENTS
                and protein_element in HYDROPHOBIC_ELEMENTS
                and distance <= HYDROPHOBIC_CUTOFF
            ):

                interaction_type = (
                    "Potential hydrophobic proximity"
                )

            else:

                interaction_type = "Other contact"

            component_record = get_atom_record(
                component_atom,
                selected_residue
            )

            protein_record = get_atom_record(
                protein_atom,
                protein_residue
            )

            interaction_data.append({

                "pdb_id": PDB_ID,
                "model_id": cell19_model_id,

                "component": selected_component_code,
                "component_id": selected_component_id,

                "component_chain": component_record["chain"],
                "component_residue_number": (
                    component_record["residue_number"]
                ),

                "component_atom": component_record["atom"],
                "component_element": component_element,
                "component_altloc": component_record["altloc"],

                "protein_chain": protein_record["chain"],
                "protein_residue": protein_record["residue"],

                "protein_residue_number": (
                    protein_record["residue_number"]
                ),

                "protein_insertion_code": (
                    protein_record["insertion_code"]
                ),

                "protein_atom": protein_record["atom"],
                "protein_element": protein_element,
                "protein_altloc": protein_record["altloc"],

                "protein_component_id": (
                    protein_record["component_id"]
                ),

                "distance": distance,
                "interaction_type": interaction_type

            })


# ============================================================
# 8. UNIVERSAL METAL COORDINATION ANALYSIS
# ============================================================
# Рассматриваются металлы в небелковых компонентах.
#
# Соседние атомы:
# - N/O/S в том же компоненте;
# - N/O/S в белковых остатках;
# - N/O/S в других небелковых компонентах.
#
# Вода исключена.
#
# Один порог 2.8 Å используется как предварительный
# геометрический фильтр, а не как универсальный критерий
# координации всех металлов.
# ============================================================

metal_coordination_data = []


def append_metal_contact(
    metal_atom,
    metal_residue,
    neighbor_atom,
    neighbor_residue,
    neighbor_type
):
    """
    Добавляет одну пару металл-соседний атом,
    если расстояние удовлетворяет геометрическому порогу.
    """

    metal_element = get_atom_element(
        metal_atom
    )

    neighbor_element = get_atom_element(
        neighbor_atom
    )

    if neighbor_element not in POLAR_ELEMENTS:
        return

    distance = get_atom_distance(
        metal_atom,
        neighbor_atom
    )

    if (
        not np.isfinite(distance)
        or distance > METAL_COORDINATION_CUTOFF
    ):
        return

    metal_record = get_atom_record(
        metal_atom,
        metal_residue
    )

    neighbor_record = get_atom_record(
        neighbor_atom,
        neighbor_residue
    )

    metal_component = get_residue_code(
        metal_residue
    )

    neighbor_component = get_residue_code(
        neighbor_residue
    )

    metal_coordination_data.append({

        "pdb_id": PDB_ID,
        "model_id": cell19_model_id,

        "metal_element": metal_element,
        "metal_component": metal_component,
        "component": metal_component,

        "metal_chain": metal_record["chain"],
        "metal_residue_number": (
            metal_record["residue_number"]
        ),

        "metal_atom": metal_record["atom"],
        "metal_component_id": metal_record["component_id"],

        "neighbor_type": neighbor_type,
        "neighbor_component": neighbor_component,

        "neighbor_component_id": (
            neighbor_record["component_id"]
        ),

        "neighbor_chain": neighbor_record["chain"],
        "neighbor_residue": neighbor_record["residue"],

        "neighbor_residue_number": (
            neighbor_record["residue_number"]
        ),

        "neighbor_atom": neighbor_record["atom"],
        "neighbor_element": neighbor_element,

        "distance": distance

    })


for metal_residue in nonprotein_residues:

    metal_atoms = get_metal_atoms(
        metal_residue
    )

    if not metal_atoms:
        continue

    metal_is_heme = is_heme_residue(
        metal_residue
    )

    # --------------------------------------------------------
    # 8.1. Атомы того же компонента
    # --------------------------------------------------------

    for metal_atom in metal_atoms:

        for neighbor_atom in metal_residue.get_atoms():

            if neighbor_atom is metal_atom:
                continue

            neighbor_type = (
                "Same HEME component"
                if metal_is_heme
                else "Same component"
            )

            append_metal_contact(
                metal_atom,
                metal_residue,
                neighbor_atom,
                metal_residue,
                neighbor_type
            )

    # --------------------------------------------------------
    # 8.2. Белковые атомы
    # --------------------------------------------------------

    for metal_atom in metal_atoms:

        for protein_residue in protein_residues:

            protein_atoms, _ = (
                get_valid_atoms_and_coordinates(
                    protein_residue
                )
            )

            for neighbor_atom in protein_atoms:

                append_metal_contact(
                    metal_atom,
                    metal_residue,
                    neighbor_atom,
                    protein_residue,
                    "PROTEIN"
                )

    # --------------------------------------------------------
    # 8.3. Другие небелковые компоненты
    # --------------------------------------------------------

    for metal_atom in metal_atoms:

        for neighbor_residue in nonprotein_residues:

            if neighbor_residue is metal_residue:
                continue

            neighbor_component_type = get_component_type(
                neighbor_residue
            )

            if neighbor_component_type == "OXY":
                neighbor_type = "OXY"
            else:
                neighbor_type = "OTHER_COMPONENT"

            neighbor_atoms, _ = (
                get_valid_atoms_and_coordinates(
                    neighbor_residue
                )
            )

            for neighbor_atom in neighbor_atoms:

                append_metal_contact(
                    metal_atom,
                    metal_residue,
                    neighbor_atom,
                    neighbor_residue,
                    neighbor_type
                )


# ============================================================
# 9. DISULFIDE BRIDGE ANALYSIS
# ============================================================
# Ищутся пары CYS-SG-SG с расстоянием не более 2.3 Å.
#
# Результат является геометрическим кандидатом.
# Дополнительная проверка химического состояния серы,
# альтернативных конформеров и связности не выполняется.
# ============================================================

disulfide_data = []

cysteine_residues = []


for residue in protein_residues:

    if get_residue_code(residue) != "CYS":
        continue

    if "SG" not in residue:
        continue

    sg_atom = residue["SG"]

    if get_atom_element(sg_atom) != "S":
        continue

    coordinate = np.asarray(
        sg_atom.coord,
        dtype=float
    )

    if not np.isfinite(coordinate).all():
        continue

    cysteine_residues.append(
        (residue, sg_atom)
    )


for i in range(len(cysteine_residues)):

    residue_1, atom_1 = cysteine_residues[i]

    for j in range(i + 1, len(cysteine_residues)):

        residue_2, atom_2 = cysteine_residues[j]

        distance = get_atom_distance(
            atom_1,
            atom_2
        )

        if (
            not np.isfinite(distance)
            or distance > DISULFIDE_CUTOFF
            or distance <= 0
        ):
            continue

        disulfide_data.append({

            "pdb_id": PDB_ID,
            "model_id": cell19_model_id,

            "chain_1": get_chain_id(residue_1),
            "residue_1": "CYS",

            "residue_number_1": (
                get_residue_number(residue_1)
            ),

            "insertion_code_1": (
                get_insertion_code(residue_1)
            ),

            "residue_component_id_1": (
                get_residue_component_id(residue_1)
            ),

            "atom_1": "SG",

            "chain_2": get_chain_id(residue_2),
            "residue_2": "CYS",

            "residue_number_2": (
                get_residue_number(residue_2)
            ),

            "insertion_code_2": (
                get_insertion_code(residue_2)
            ),

            "residue_component_id_2": (
                get_residue_component_id(residue_2)
            ),

            "atom_2": "SG",

            "distance": distance,

            "interaction_type": (
                "Potential disulfide bridge"
            )

        })


# ============================================================
# 10. DATAFRAME CREATION
# ============================================================

interaction_columns = [

    "pdb_id",
    "model_id",
    "component",
    "component_id",
    "component_chain",
    "component_residue_number",
    "component_atom",
    "component_element",
    "component_altloc",
    "protein_chain",
    "protein_residue",
    "protein_residue_number",
    "protein_insertion_code",
    "protein_atom",
    "protein_element",
    "protein_altloc",
    "protein_component_id",
    "distance",
    "interaction_type"

]


metal_columns = [

    "pdb_id",
    "model_id",
    "metal_element",
    "metal_component",
    "component",
    "metal_chain",
    "metal_residue_number",
    "metal_atom",
    "metal_component_id",
    "neighbor_type",
    "neighbor_component",
    "neighbor_component_id",
    "neighbor_chain",
    "neighbor_residue",
    "neighbor_residue_number",
    "neighbor_atom",
    "neighbor_element",
    "distance"

]


disulfide_columns = [

    "pdb_id",
    "model_id",
    "chain_1",
    "residue_1",
    "residue_number_1",
    "insertion_code_1",
    "residue_component_id_1",
    "atom_1",
    "chain_2",
    "residue_2",
    "residue_number_2",
    "insertion_code_2",
    "residue_component_id_2",
    "atom_2",
    "distance",
    "interaction_type"

]


df_interactions = pd.DataFrame(
    interaction_data,
    columns=interaction_columns
)


df_all_metal_coordination = pd.DataFrame(
    metal_coordination_data,
    columns=metal_columns
)


df_disulfide_bridges = pd.DataFrame(
    disulfide_data,
    columns=disulfide_columns
)


# ============================================================
# 11. LEGACY HEME IRON TABLE
# ============================================================
# Контакты Fe в распознанных гемовых компонентах.
# Включает контакты Fe с:
# - атомами самого гема;
# - белковыми остатками;
# - другими небелковыми компонентами.
# ============================================================

if not df_all_metal_coordination.empty:

    df_metal_coordination = (

        df_all_metal_coordination.loc[

            (
                df_all_metal_coordination["metal_element"]
                == "FE"
            )
            &
            (
                df_all_metal_coordination["metal_component"]
                .isin(HEME_COMPONENT_CODES)
            )

        ]
        .copy()
        .reset_index(drop=True)

    )

else:

    df_metal_coordination = pd.DataFrame(
        columns=metal_columns
    )


# ============================================================
# 12. FE - OXY ANALYSIS
# ============================================================

if not df_all_metal_coordination.empty:

    df_fe_oxy = (

        df_all_metal_coordination.loc[

            (
                df_all_metal_coordination["metal_element"]
                == "FE"
            )
            &
            (
                df_all_metal_coordination["neighbor_type"]
                == "OXY"
            )
            &
            (
                df_all_metal_coordination["neighbor_element"]
                == "O"
            )

        ]
        .copy()
        .reset_index(drop=True)

    )

else:

    df_fe_oxy = pd.DataFrame(
        columns=metal_columns
    )


# Совместимость с предыдущими именами переменных.

df_fe_oxy_coordination = df_fe_oxy.copy()


if not df_fe_oxy.empty:

    df_fe_oxy_molecules = (

        df_fe_oxy[

            [
                "model_id",
                "metal_component",
                "metal_chain",
                "metal_residue_number",
                "metal_atom",
                "neighbor_component",
                "neighbor_chain",
                "neighbor_residue_number",
                "distance"
            ]

        ]
        .drop_duplicates()
        .reset_index(drop=True)

    )

else:

    df_fe_oxy_molecules = pd.DataFrame(

        columns=[

            "model_id",
            "metal_component",
            "metal_chain",
            "metal_residue_number",
            "metal_atom",
            "neighbor_component",
            "neighbor_chain",
            "neighbor_residue_number",
            "distance"

        ]

    )


# ============================================================
# 13. SORTING
# ============================================================

if not df_interactions.empty:

    df_interactions = (

        df_interactions
        .sort_values(
            by=["distance", "protein_chain"]
        )
        .reset_index(drop=True)

    )


if not df_all_metal_coordination.empty:

    df_all_metal_coordination = (

        df_all_metal_coordination
        .sort_values(

            by=[
                "metal_element",
                "metal_chain",
                "metal_residue_number",
                "distance"
            ]

        )
        .reset_index(drop=True)

    )


if not df_metal_coordination.empty:

    df_metal_coordination = (

        df_metal_coordination
        .sort_values(
            by=[
                "metal_chain",
                "metal_residue_number",
                "distance"
            ]
        )
        .reset_index(drop=True)

    )


if not df_fe_oxy.empty:

    df_fe_oxy = (

        df_fe_oxy
        .sort_values(
            by=["distance"]
        )
        .reset_index(drop=True)

    )

    df_fe_oxy_coordination = df_fe_oxy.copy()


if not df_disulfide_bridges.empty:

    df_disulfide_bridges = (

        df_disulfide_bridges
        .sort_values(
            by=["distance"]
        )
        .reset_index(drop=True)

    )


# ============================================================
# 14. INTERACTION SUMMARY
# ============================================================

interaction_summary_columns = [

    "interaction_type",
    "contact_count",
    "mean_distance",
    "min_distance",
    "max_distance"

]


if not df_interactions.empty:

    df_interaction_summary = (

        df_interactions
        .groupby(
            "interaction_type",
            dropna=False
        )
        .agg(

            contact_count=("distance", "count"),
            mean_distance=("distance", "mean"),
            min_distance=("distance", "min"),
            max_distance=("distance", "max")

        )
        .reset_index()

    )

else:

    df_interaction_summary = pd.DataFrame(
        columns=interaction_summary_columns
    )


# ============================================================
# 15. METAL SUMMARIES
# ============================================================

metal_summary_group_columns = [

    "metal_element",
    "metal_component",
    "neighbor_type",
    "neighbor_element"

]


metal_summary_columns = [

    "metal_element",
    "metal_component",
    "neighbor_type",
    "neighbor_element",
    "coordination_contact_count",
    "mean_distance",
    "min_distance",
    "max_distance"

]


def build_metal_summary(dataframe):

    if dataframe.empty:
        return pd.DataFrame(
            columns=metal_summary_columns
        )

    return (

        dataframe
        .groupby(
            metal_summary_group_columns,
            dropna=False
        )
        .agg(

            coordination_contact_count=("distance", "count"),
            mean_distance=("distance", "mean"),
            min_distance=("distance", "min"),
            max_distance=("distance", "max")

        )
        .reset_index()

    )


df_metal_coordination_summary = build_metal_summary(
    df_metal_coordination
)


df_all_metal_coordination_summary = build_metal_summary(
    df_all_metal_coordination
)


# ============================================================
# 16. FE-OXY SUMMARY
# ============================================================

fe_oxy_summary_columns = [

    "metal_component",
    "neighbor_component",
    "neighbor_element",
    "contact_count",
    "mean_distance",
    "min_distance",
    "max_distance"

]


if not df_fe_oxy.empty:

    df_fe_oxy_summary = (

        df_fe_oxy
        .groupby(

            [
                "metal_component",
                "neighbor_component",
                "neighbor_element"
            ],

            dropna=False

        )
        .agg(

            contact_count=("distance", "count"),
            mean_distance=("distance", "mean"),
            min_distance=("distance", "min"),
            max_distance=("distance", "max")

        )
        .reset_index()

    )

else:

    df_fe_oxy_summary = pd.DataFrame(
        columns=fe_oxy_summary_columns
    )


# ============================================================
# 17. DISULFIDE SUMMARY
# ============================================================

disulfide_summary_columns = [

    "model_id",
    "chain_1",
    "chain_2",
    "bridge_count",
    "mean_distance",
    "min_distance",
    "max_distance"

]


if not df_disulfide_bridges.empty:

    df_disulfide_summary = (

        df_disulfide_bridges
        .groupby(
            ["model_id", "chain_1", "chain_2"],
            dropna=False
        )
        .agg(

            bridge_count=("distance", "count"),
            mean_distance=("distance", "mean"),
            min_distance=("distance", "min"),
            max_distance=("distance", "max")

        )
        .reset_index()

    )

else:

    df_disulfide_summary = pd.DataFrame(
        columns=disulfide_summary_columns
    )


# ============================================================
# 18. VALIDATION
# ============================================================

validation_results = {}


# 18.1. Проверка таблицы Fe в гемах.

if df_metal_coordination.empty:

    legacy_metal_valid = True

else:

    legacy_metal_valid = bool(

        (
            df_metal_coordination["metal_element"] == "FE"
        ).all()

        and

        (
            df_metal_coordination["metal_component"]
            .isin(HEME_COMPONENT_CODES)
        ).all()

        and

        (
            df_metal_coordination["neighbor_element"]
            .isin(POLAR_ELEMENTS)
        ).all()

        and

        (
            df_metal_coordination["distance"].notna()
        ).all()

        and

        (
            df_metal_coordination["distance"]
            <= METAL_COORDINATION_CUTOFF
        ).all()

    )


validation_results[
    "legacy_heme_iron_coordination_valid"
] = legacy_metal_valid


# 18.2. Проверка универсальной таблицы металлов.

if df_all_metal_coordination.empty:

    all_metal_valid = True

else:

    all_metal_valid = bool(

        (
            df_all_metal_coordination["metal_element"]
            .isin(METAL_ELEMENTS)
        ).all()

        and

        (
            df_all_metal_coordination["neighbor_element"]
            .isin(POLAR_ELEMENTS)
        ).all()

        and

        (
            df_all_metal_coordination["distance"].notna()
        ).all()

        and

        (
            df_all_metal_coordination["distance"]
            <= METAL_COORDINATION_CUTOFF
        ).all()

        and

        (
            df_all_metal_coordination["distance"] >= 0
        ).all()

    )


validation_results[
    "universal_metal_coordination_valid"
] = all_metal_valid


# 18.3. Проверка старой таблицы как подмножества
# универсальной таблицы.

legacy_subset_columns = [

    "metal_component_id",
    "metal_atom",
    "metal_element",
    "neighbor_component_id",
    "neighbor_atom",
    "neighbor_element",
    "neighbor_type",
    "distance"

]


if df_metal_coordination.empty:

    legacy_subset_valid = True

elif df_all_metal_coordination.empty:

    legacy_subset_valid = False

else:

    legacy_check = df_metal_coordination[
        legacy_subset_columns
    ].merge(

        df_all_metal_coordination[
            legacy_subset_columns
        ].drop_duplicates(),

        on=legacy_subset_columns,

        how="left",

        indicator=True

    )

    legacy_subset_valid = bool(
        (legacy_check["_merge"] == "both").all()
    )


validation_results[
    "legacy_table_subset_of_universal_table"
] = legacy_subset_valid


# 18.4. Проверка Fe-OXY.

if df_fe_oxy.empty:

    fe_oxy_valid = True

else:

    fe_oxy_valid = bool(

        (
            df_fe_oxy["metal_element"] == "FE"
        ).all()

        and

        (
            df_fe_oxy["neighbor_type"] == "OXY"
        ).all()

        and

        (
            df_fe_oxy["neighbor_element"] == "O"
        ).all()

        and

        (
            df_fe_oxy["distance"]
            <= METAL_COORDINATION_CUTOFF
        ).all()

    )


validation_results[
    "fe_oxy_contacts_valid"
] = fe_oxy_valid


# 18.5. Проверка дисульфидных кандидатов.

if df_disulfide_bridges.empty:

    disulfide_valid = True

else:

    disulfide_valid = bool(

        (
            df_disulfide_bridges["distance"].notna()
        ).all()

        and

        (
            df_disulfide_bridges["distance"]
            <= DISULFIDE_CUTOFF
        ).all()

        and

        (
            df_disulfide_bridges["distance"] > 0
        ).all()

        and

        (
            df_disulfide_bridges["atom_1"] == "SG"
        ).all()

        and

        (
            df_disulfide_bridges["atom_2"] == "SG"
        ).all()

    )


validation_results[
    "disulfide_candidates_valid"
] = disulfide_valid


# 18.6. Проверка атомных контактов.

if df_interactions.empty:

    interaction_valid = True

else:

    interaction_valid = bool(

        (
            df_interactions["distance"].notna()
        ).all()

        and

        (
            df_interactions["distance"] >= 0
        ).all()

        and

        (
            df_interactions["distance"]
            <= CONTACT_CUTOFF
        ).all()

    )


validation_results[
    "protein_component_contacts_valid"
] = interaction_valid


# 18.7. Проверка выбранного компонента.

if selected_residue is None:

    selected_component_valid = True

else:

    selected_component_valid = bool(

        selected_component_id is not None

        and selected_component_model_id == cell19_model_id

        and len(selected_atoms) > 0

        and selected_coordinates.shape == (
            len(selected_atoms),
            3
        )

    )


validation_results[
    "selected_component_valid"
] = selected_component_valid


# ============================================================
# 19. SAVE CSV FILES
# ============================================================

legacy_csv_data = {

    f"{PDB_ID}_ligand_interactions.csv":
        df_interactions,

    f"{PDB_ID}_heme_iron_coordination.csv":
        df_metal_coordination,

    f"{PDB_ID}_heme_iron_coordination_summary.csv":
        df_metal_coordination_summary,

    f"{PDB_ID}_interaction_summary.csv":
        df_interaction_summary,

    f"{PDB_ID}_disulfide_bridges.csv":
        df_disulfide_bridges,

    f"{PDB_ID}_disulfide_summary.csv":
        df_disulfide_summary,

    f"{PDB_ID}_fe_oxy_distances.csv":
        df_fe_oxy,

    f"{PDB_ID}_fe_oxy_summary.csv":
        df_fe_oxy_summary

}


universal_csv_data = {

    f"{PDB_ID}_metal_coordination.csv":
        df_all_metal_coordination,

    f"{PDB_ID}_metal_coordination_summary.csv":
        df_all_metal_coordination_summary

}


all_csv_data = {
    **legacy_csv_data,
    **universal_csv_data
}


output_files = []

for filename, dataframe in all_csv_data.items():

    output_path = (
        PDB_RESULTS_DIR / filename
    )

    dataframe.to_csv(
        output_path,
        index=False,
        encoding="utf-8-sig"
    )

    output_files.append(
        output_path
    )


csv_files_exist = all(
    path.is_file() and path.stat().st_size > 0
    for path in output_files
)

validation_results[
    "all_csv_files_exist"
] = csv_files_exist


# ============================================================
# 20. JSON REPORT
# ============================================================

interaction_analysis_report = {

    "pdb_id": PDB_ID,

    "analysis_status": (
        "COMPLETED"
        if selected_residue is not None
        else "COMPLETED_WITHOUT_SELECTED_COMPONENT"
    ),

    "selected_component": {

        "component": selected_component_code,
        "component_id": selected_component_id,
        "chain": selected_component_chain,
        "residue_number": selected_component_number,
        "model_id": selected_component_model_id

    },

    "model_scope": {

        "source_model_count": len(cell19_models),

        "analyzed_model_id": cell19_model_id,

        "analyzed_model_serial": cell19_model_serial,

        "scope": "First model only"

    },

    "parameters": {

        "contact_cutoff_A": CONTACT_CUTOFF,

        "polar_atom_proximity_cutoff_A": POLAR_CUTOFF,

        "hydrophobic_proximity_cutoff_A": HYDROPHOBIC_CUTOFF,

        "metal_coordination_cutoff_A":
            METAL_COORDINATION_CUTOFF,

        "disulfide_cutoff_A": DISULFIDE_CUTOFF

    },

    "structure": {

        "protein_residue_count": len(protein_residues),

        "nonprotein_component_count": len(nonprotein_residues),

        "water_residue_count": len(water_residues),

        "heme_component_count": len(heme_residues),

        "oxy_component_count": len(oxy_residues)

    },

    "interaction_analysis": {

        "total_contacts": int(
            len(df_interactions)
        ),

        "potential_polar_atom_proximities": int(

            (
                df_interactions["interaction_type"]
                == "Potential polar-atom proximity"
            ).sum()

        ) if not df_interactions.empty else 0,

        "potential_hydrophobic_proximities": int(

            (
                df_interactions["interaction_type"]
                == "Potential hydrophobic proximity"
            ).sum()

        ) if not df_interactions.empty else 0,

        "other_contacts": int(

            (
                df_interactions["interaction_type"]
                == "Other contact"
            ).sum()

        ) if not df_interactions.empty else 0

    },

    "metal_coordination": {

        "legacy_heme_iron_contact_count": int(
            len(df_metal_coordination)
        ),

        "all_metal_contact_count": int(
            len(df_all_metal_coordination)
        ),

        "recognized_metal_elements": sorted(
            list(METAL_ELEMENTS)
        ),

        "metal_element_distribution": (

            df_all_metal_coordination[
                "metal_element"
            ]
            .value_counts()
            .to_dict()

            if not df_all_metal_coordination.empty

            else {}

        )

    },

    "fe_oxy_analysis": {

        "fe_oxy_contact_count": int(
            len(df_fe_oxy)
        ),

        "oxy_component_count": int(

            df_fe_oxy["neighbor_component_id"].nunique()

        ) if not df_fe_oxy.empty else 0

    },

    "disulfide_analysis": {

        "potential_disulfide_bridge_count": int(
            len(df_disulfide_bridges)
        )

    },

    "validation": validation_results,

    "output_files": [
        path.name
        for path in output_files
    ],

    "interpretation_notes": [

        "Only the first model is analyzed.",

        "The original structure object from CELL 4 is used.",

        "Atomic contacts are identified using geometric distance cutoffs.",

        "Polar-atom proximity is not a confirmed hydrogen bond.",

        "Hydrophobic proximity is a preliminary geometric candidate.",

        "Metal contacts are preliminary geometric candidates, "
        "not definitive coordination bonds.",

        "The same metal distance threshold is used as a screening "
        "criterion for all recognized elements.",

        "The heme iron table contains contacts of Fe atoms "
        "in recognized heme components.",

        "The universal metal table includes recognized metals "
        "represented in nonprotein components.",

        "Potential disulfide bridges are identified by SG-SG distance "
        "and require additional chemical validation.",

        "Hydrogen atoms are excluded and the selected Bio.PDB "
        "alternate conformer is used.",

        "Water molecules are excluded from metal-neighbor analysis.",

        "The analysis does not establish donor-acceptor chemistry, "
        "protonation states, coordination geometry or interaction angles."

    ]

}


# ============================================================
# 21. SAVE JSON REPORT AND FINAL VALIDATION
# ============================================================

json_path = (
    PDB_RESULTS_DIR
    / f"{PDB_ID}_interaction_analysis.json"
)


# Сначала сохраняем JSON, чтобы проверить реальное создание файла.

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_safe(interaction_analysis_report),
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


json_file_valid = (
    json_path.is_file()
    and json_path.stat().st_size > 0
)

validation_results[
    "json_report_exists"
] = json_file_valid


failed_validations = [

    name

    for name, result in validation_results.items()

    if not result

]


interaction_analysis_report["validation"] = (
    validation_results
)

interaction_analysis_report["validation_status"] = (

    "PASSED"
    if not failed_validations
    else "CHECK_REQUIRED"

)


# Обновляем JSON с окончательными результатами проверок.

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_safe(interaction_analysis_report),
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


output_files.append(json_path)


# ============================================================
# 22. DISPLAY RESULTS
# ============================================================

print("\n" + "=" * 70)
print("PROTEIN-COMPONENT INTERACTION ANALYSIS")
print("=" * 70)

print("PDB ID:", PDB_ID)
print("Analyzed model:", cell19_model_id)
print("Selected component:", selected_component_code)

print("Protein residues:", len(protein_residues))
print("Nonprotein components:", len(nonprotein_residues))


print("-" * 70)
print("1. SELECTED COMPONENT - PROTEIN INTERACTIONS")

print(
    "Total contacts:",
    len(df_interactions)
)

if not df_interactions.empty:
    display(
        df_interactions.head(20)
    )


print("-" * 70)
print("2. LEGACY HEME IRON CONTACTS")

print(
    "Fe contacts in recognized hemes:",
    len(df_metal_coordination)
)

if not df_metal_coordination.empty:
    display(
        df_metal_coordination.head(20)
    )


print("-" * 70)
print("3. UNIVERSAL METAL CONTACTS")

print(
    "All metal contacts:",
    len(df_all_metal_coordination)
)

if not df_all_metal_coordination.empty:

    display(

        df_all_metal_coordination[
            "metal_element"
        ]
        .value_counts()
        .rename_axis("metal_element")
        .reset_index(name="contact_count")

    )

    display(
        df_all_metal_coordination.head(30)
    )


print("-" * 70)
print("4. FE - OXY CONTACTS")

print(
    "Fe-OXY contacts:",
    len(df_fe_oxy)
)

if not df_fe_oxy.empty:
    display(df_fe_oxy)


print("-" * 70)
print("5. POTENTIAL DISULFIDE BRIDGES")

print(
    "Potential bridges:",
    len(df_disulfide_bridges)
)

if not df_disulfide_bridges.empty:
    display(df_disulfide_bridges)


print("-" * 70)
print("6. VALIDATION")

for validation_name, validation_value in (
    validation_results.items()
):

    status = (
        "OK"
        if validation_value
        else "CHECK"
    )

    print(
        f"{validation_name}: {status}"
    )


print("-" * 70)
print("7. OUTPUT FILES")

for path in output_files:

    print(
        f"[{'OK' if path.is_file() else 'ERROR'}] "
        f"{path.name}"
    )


print("=" * 70)

print(
    "VALIDATION STATUS:",
    "PASSED"
    if not failed_validations
    else "CHECK_REQUIRED"
)

print("CELL 19 COMPLETED")
print("=" * 70)


if failed_validations:

    print("\nWARNING: Validation issues detected:")

    for name in failed_validations:
        print(" -", name)

else:

    print(
        "\nAll CELL 19 validation checks passed."
    )
```

```python
# ============================================================
# CELL 20. FINAL STRUCTURAL AND FUNCTIONAL METRICS
# ============================================================
# Назначение:
# 1. Собрать основные характеристики структуры.
# 2. Рассчитать метрики белковых цепей.
# 3. Сравнить аминокислотные последовательности цепей.
# 4. Систематизировать геометрические характеристики белка.
# 5. Систематизировать показатели качества структуры.
# 6. Оценить состав небелковых компонентов.
# 7. Систематизировать показатели binding site.
# 8. Систематизировать контакты выбранного компонента с белком.
# 9. Сравнить расстояния Fe-N во всех копиях HEM.
# 10. Систематизировать контакты Fe-OXY.
# 11. Включить результаты анализа дисульфидных мостиков.
# 12. Сформировать сводные таблицы и JSON-отчет.
#
# Согласованность:
# - PDB_RESULTS_DIR — индивидуальная папка структуры;
# - df_atoms — белковые атомы первой модели;
# - df_residues — белковые остатки первой модели;
# - structure — исходная структура Bio.PDB из CELL 4;
# - component_structure — совместимая структура из CELL 17,
#   если она ссылается на исходный объект;
# - Fe-N и Fe-OXY — первая модель;
# - CELL 18 — таблицы binding site и атомных контактов;
# - CELL 19 — таблицы взаимодействий, Fe-контактов и S-S.
#
# Важно:
# Геометрическая близость не доказывает наличие химической связи.
# Отсутствующие данные не заменяются нулевыми значениями.
# ============================================================

import json
import re
import numpy as np
import pandas as pd

from pathlib import Path
from Bio.Align import PairwiseAligner
from IPython.display import display


# ============================================================
# 1. INPUT VALIDATION
# ============================================================

required_variables = [
    "PDB_ID",
    "PDB_RESULTS_DIR",
    "df_atoms",
    "df_residues",
    "df_hetero",
    "df_water",
    "chain_sequences"
]

missing_variables = [
    name
    for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise NameError(
        "Не найдены обязательные переменные: "
        + ", ".join(missing_variables)
    )


PDB_ID = str(PDB_ID).strip().upper()

if len(PDB_ID) != 4 or not PDB_ID.isalnum():
    raise ValueError(f"Некорректный PDB_ID: {PDB_ID}")


PDB_RESULTS_DIR = Path(PDB_RESULTS_DIR)

PDB_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


required_dataframes = {
    "df_atoms": df_atoms,
    "df_residues": df_residues,
    "df_hetero": df_hetero,
    "df_water": df_water
}

for name, dataframe in required_dataframes.items():

    if not isinstance(dataframe, pd.DataFrame):
        raise TypeError(
            f"{name} должен быть pandas DataFrame."
        )

    if "pdb_id" in dataframe.columns and not dataframe.empty:

        ids = (
            dataframe["pdb_id"]
            .dropna()
            .astype(str)
            .str.upper()
            .unique()
        )

        if len(ids) and any(value != PDB_ID for value in ids):
            raise ValueError(
                f"{name} содержит записи другого PDB: {ids}"
            )


if not isinstance(chain_sequences, dict):
    raise TypeError(
        "chain_sequences должен быть словарем."
    )


if df_atoms.empty:
    raise ValueError(
        "df_atoms пуст. Невозможно рассчитать геометрию белка."
    )


if not {"x", "y", "z"}.issubset(df_atoms.columns):
    raise ValueError(
        "В df_atoms отсутствуют координаты x, y, z."
    )


# ------------------------------------------------------------
# 1.1. Структура Bio.PDB
# ------------------------------------------------------------
# Предпочтительно использовать исходный объект structure из CELL 4.
# component_structure из CELL 17 допускается как fallback,
# но не должен быть независимой повторно загруженной копией,
# если далее требуется сопоставление объектов CELL 17 и CELL 18.
# ------------------------------------------------------------

cell20_structure = globals().get("structure")

if cell20_structure is None:
    cell20_structure = globals().get("component_structure")

component_structure_available = (
    cell20_structure is not None
    and hasattr(cell20_structure, "get_models")
)

if component_structure_available:

    structure_id = str(
        getattr(cell20_structure, "id", "")
    ).strip().upper()

    if structure_id and structure_id != PDB_ID:
        raise ValueError(
            f"Структура относится к {structure_id}, "
            f"а текущий PDB_ID={PDB_ID}"
        )

    cell20_models = list(
        cell20_structure.get_models()
    )

    if not cell20_models:
        raise ValueError(
            "В структуре Bio.PDB отсутствуют модели."
        )

    analysis_model = cell20_models[0]

    analysis_model_id = analysis_model.id

else:

    cell20_models = []

    analysis_model = None

    analysis_model_id = None


print("=" * 75)
print("CELL 20. FINAL STRUCTURAL AND FUNCTIONAL METRICS")
print("=" * 75)

print("PDB ID:", PDB_ID)
print("Output directory:", PDB_RESULTS_DIR)

if component_structure_available:

    print("Models in structure:", len(cell20_models))
    print("Analyzed model:", analysis_model_id)

else:

    print(
        "WARNING: Bio.PDB structure unavailable. "
        "Fe-N and Fe-OXY calculations will be skipped."
    )


# ============================================================
# 2. HELPER FUNCTIONS
# ============================================================

def numeric_values(series):

    return pd.to_numeric(
        series,
        errors="coerce"
    ).dropna()


def safe_mean(series):

    values = numeric_values(series)

    return (
        float(values.mean())
        if not values.empty
        else np.nan
    )


def safe_median(series):

    values = numeric_values(series)

    return (
        float(values.median())
        if not values.empty
        else np.nan
    )


def safe_std(series):

    values = numeric_values(series)

    return (
        float(values.std())
        if len(values) >= 2
        else np.nan
    )


def safe_percentage(numerator, denominator):

    if pd.isna(numerator) or pd.isna(denominator):
        return np.nan

    if denominator == 0:
        return np.nan

    return float(
        100.0 * numerator / denominator
    )


def get_first_existing_column(dataframe, candidates):

    if dataframe is None:
        return None

    for column in candidates:

        if column in dataframe.columns:
            return column

    return None


def dataframe_or_empty(variable_names):

    if isinstance(variable_names, str):
        variable_names = [variable_names]

    for name in variable_names:

        value = globals().get(name)

        if isinstance(value, pd.DataFrame):
            return value.copy(), name

    return pd.DataFrame(), None


def json_safe(value):

    """Рекурсивное преобразование значений в JSON-совместимый формат."""

    if isinstance(value, dict):

        return {
            str(key): json_safe(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple)):

        return [
            json_safe(item)
            for item in value
        ]

    if isinstance(value, np.ndarray):
        return json_safe(value.tolist())

    if isinstance(value, (np.integer,)):
        return int(value)

    if isinstance(value, (np.floating, float)):

        value = float(value)

        return (
            value
            if np.isfinite(value)
            else None
        )

    if isinstance(value, (np.bool_, bool)):
        return bool(value)

    if isinstance(value, Path):
        return str(value)

    if value is None:
        return None

    if isinstance(value, (str, int)):
        return value

    # pd.NA, pd.NaT и другие скалярные пропуски.
    try:

        missing = pd.isna(value)

        if isinstance(missing, (bool, np.bool_)) and missing:
            return None

    except (TypeError, ValueError):
        pass

    return str(value)


def read_csv_if_exists(filenames):

    if isinstance(filenames, str):
        filenames = [filenames]

    for filename in filenames:

        path = PDB_RESULTS_DIR / filename

        if not path.is_file():
            continue

        try:

            dataframe = pd.read_csv(path)

            print(
                f"Loaded: {filename} "
                f"({len(dataframe)} records)"
            )

            return dataframe

        except Exception as error:

            print(
                f"WARNING: Не удалось прочитать "
                f"{filename}: {error}"
            )

    return pd.DataFrame()


def read_json_if_exists(filename):

    path = PDB_RESULTS_DIR / filename

    if not path.is_file():
        return {}

    try:

        with open(path, "r", encoding="utf-8") as file:
            return json.load(file)

    except Exception as error:

        print(
            f"WARNING: Не удалось прочитать "
            f"{filename}: {error}"
        )

        return {}


def add_metric(
    collection,
    category,
    metric,
    value,
    unit=None,
    status="calculated"
):

    collection.append({

        "category": category,

        "metric": metric,

        "value": json_safe(value),

        "unit": unit,

        "status": status

    })


def normalize_chain(value):

    if value is None:
        return ""

    try:

        if pd.isna(value):
            return ""

    except (TypeError, ValueError):
        pass

    return str(value).strip()


def normalize_category(value):

    if value is None:
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(value).strip().upper().replace("_", " ")
    )


def get_atom_element(atom):

    element = getattr(atom, "element", None)

    if element is not None:

        element = str(element).strip().upper()

        if element not in {"", "?", ".", "X"}:
            return element

    atom_name = "".join(
        character
        for character in atom.get_fullname().strip().upper()
        if character.isalpha()
    )

    return atom_name[0] if atom_name else "UNKNOWN"


def get_valid_coordinate(atom):

    coordinate = np.asarray(
        atom.coord,
        dtype=float
    )

    if coordinate.shape != (3,):
        return None

    if not np.isfinite(coordinate).all():
        return None

    return coordinate


def get_residue_number(residue):

    try:
        return int(residue.id[1])

    except (TypeError, ValueError, IndexError):
        return None


def get_insertion_code(residue):

    try:
        return str(residue.id[2]).strip()

    except (TypeError, IndexError):
        return ""


def get_residue_hetero_flag(residue):

    try:
        return str(residue.id[0]).strip()

    except (TypeError, IndexError):
        return ""


def get_selected_atoms(residue):

    """
    Возвращает выбранное Bio.PDB представление атомов.
    Для DisorderedAtom используется выбранный конформер,
    что соответствует обычному get_atoms() в CELL 17–19.
    """

    return list(residue.get_atoms())


# ============================================================
# 3. GLOBAL STRUCTURE METRICS
# ============================================================

number_of_chains = len(chain_sequences)

number_of_protein_residues = len(df_residues)

number_of_protein_atoms = len(df_atoms)

number_of_water_residues = len(df_water)


# ------------------------------------------------------------
# 3.1. Проверка фактического количества белковых цепей
# ------------------------------------------------------------

residue_chain_column = get_first_existing_column(
    df_residues,
    ["chain_id", "chain"]
)

if residue_chain_column is not None:

    actual_residue_chains = (
        df_residues[residue_chain_column]
        .apply(normalize_chain)
        .nunique()
    )

else:

    actual_residue_chains = np.nan


# ------------------------------------------------------------
# 3.2. Element composition
# ------------------------------------------------------------

if "element" in df_atoms.columns:

    element_counts = (

        df_atoms["element"]

        .fillna("Unknown")

        .astype(str)

        .str.strip()

        .str.upper()

        .replace("", "UNKNOWN")

        .value_counts()

    )

else:

    element_counts = pd.Series(dtype=int)


number_of_elements = len(element_counts)


df_element_distribution = (

    element_counts

    .rename_axis("element")

    .reset_index(name="atom_count")

)

if not df_element_distribution.empty:

    df_element_distribution["percentage"] = (

        100.0
        * df_element_distribution["atom_count"]
        / number_of_protein_atoms

    )


# ============================================================
# 4. NON-PROTEIN COMPONENT METRICS
# ============================================================

# CELL 17: df_components — предпочтительный источник.
# CELL 8: df_hetero — резервный источник.
#
# Количество экземпляров и количество типов компонентов
# рассчитываются отдельно.
# ============================================================

df_components_source, components_source_name = (
    dataframe_or_empty([
        "df_components",
        "df_component_analysis"
    ])
)


if df_components_source.empty:

    df_components_source = read_csv_if_exists(
        [
            f"{PDB_ID}_components.csv",
            f"{PDB_ID}_component_occurrences.csv"
        ]
    )

    if not df_components_source.empty:
        components_source_name = "CELL 17 CSV"


if not df_components_source.empty:

    component_code_column = get_first_existing_column(

        df_components_source,

        [
            "component",
            "residue_name",
            "resname",
            "component_code"
        ]

    )

    if component_code_column is not None:

        component_codes = (

            df_components_source[component_code_column]

            .dropna()

            .astype(str)

            .str.strip()

            .str.upper()

        )

        component_codes = component_codes[
            component_codes != ""
        ]

        # CELL 17 components.csv — таблица экземпляров компонентов.
        # Не удаляем строки по одному component_id:
        # один и тот же код может встречаться в нескольких местах.

        number_of_component_instances = len(
            df_components_source
        )

        number_of_unique_component_types = (
            component_codes.nunique()
        )

        component_counts = component_codes.value_counts()

        component_count_status = (
            f"from {components_source_name}"
        )

    else:

        component_counts = pd.Series(dtype=int)

        number_of_component_instances = np.nan

        number_of_unique_component_types = np.nan

        component_count_status = (
            "component code column unavailable"
        )


elif not df_hetero.empty:

    component_code_column = get_first_existing_column(

        df_hetero,

        [
            "residue_name",
            "resname",
            "component"
        ]

    )

    if component_code_column is not None:

        instance_columns = [

            column

            for column in [

                "model_id",
                "chain_id",
                "chain",
                "residue_key",
                "residue_number",
                "residue_id",
                "insertion_code",
                component_code_column

            ]

            if column in df_hetero.columns

        ]

        if instance_columns:

            component_records = (
                df_hetero[instance_columns].drop_duplicates()
            )

        else:

            component_records = df_hetero.copy()


        component_codes = (

            component_records[component_code_column]

            .dropna()

            .astype(str)

            .str.strip()

            .str.upper()

        )

        component_codes = component_codes[
            component_codes != ""
        ]

        number_of_component_instances = len(
            component_records
        )

        number_of_unique_component_types = (
            component_codes.nunique()
        )

        component_counts = component_codes.value_counts()

        component_count_status = (
            "calculated from df_hetero"
        )

    else:

        component_counts = pd.Series(dtype=int)

        number_of_component_instances = np.nan

        number_of_unique_component_types = np.nan

        component_count_status = (
            "component code column unavailable"
        )

else:

    component_counts = pd.Series(dtype=int)

    number_of_component_instances = 0

    number_of_unique_component_types = 0

    component_count_status = (
        "no non-water hetero components detected"
    )


global_metrics = {

    "PDB_ID": PDB_ID,

    "number_of_chains_from_sequence_dictionary": number_of_chains,

    "number_of_chains_in_residue_table": actual_residue_chains,

    "number_of_protein_residues": number_of_protein_residues,

    "number_of_protein_atoms": number_of_protein_atoms,

    "number_of_component_instances": (
        number_of_component_instances
    ),

    "number_of_unique_component_types": (
        number_of_unique_component_types
    ),

    "number_of_water_residues": number_of_water_residues,

    "number_of_atom_elements": number_of_elements

}


df_global_metrics = pd.DataFrame(
    [global_metrics]
)


df_component_distribution = (

    component_counts

    .rename_axis("component")

    .reset_index(name="occurrence_count")

)


# ============================================================
# 5. PROTEIN CHAIN METRICS
# ============================================================

chain_metric_data = []


chain_column = get_first_existing_column(
    df_residues,
    ["chain_id", "chain"]
)

atom_chain_column = get_first_existing_column(
    df_atoms,
    ["chain_id", "chain"]
)


for chain_id, sequence in chain_sequences.items():

    chain_key = normalize_chain(chain_id)

    if chain_column is not None:

        chain_residues = df_residues[

            df_residues[chain_column]
            .apply(normalize_chain)
            == chain_key

        ]

    else:

        chain_residues = pd.DataFrame()


    if atom_chain_column is not None:

        chain_atoms = df_atoms[

            df_atoms[atom_chain_column]
            .apply(normalize_chain)
            == chain_key

        ]

    else:

        chain_atoms = pd.DataFrame()


    sequence_length = len(str(sequence))

    residue_count = len(chain_residues)

    atom_count = len(chain_atoms)


    missing_backbone_count = np.nan

    missing_backbone_column = get_first_existing_column(

        chain_residues,

        [
            "missing_backbone_atom_count",
            "missing_backbone_atoms"
        ]

    )

    if missing_backbone_column is not None:

        values = chain_residues[missing_backbone_column]

        if pd.api.types.is_numeric_dtype(values):

            missing_backbone_count = int(
                pd.to_numeric(
                    values,
                    errors="coerce"
                ).fillna(0).sum()
            )


    mean_bfactor = (

        safe_mean(chain_atoms["bfactor"])

        if "bfactor" in chain_atoms.columns

        else np.nan

    )

    median_bfactor = (

        safe_median(chain_atoms["bfactor"])

        if "bfactor" in chain_atoms.columns

        else np.nan

    )

    mean_occupancy = (

        safe_mean(chain_atoms["occupancy"])

        if "occupancy" in chain_atoms.columns

        else np.nan

    )


    chain_metric_data.append({

        "chain_id": chain_key,

        "sequence_length": sequence_length,

        "residue_count": residue_count,

        "atom_count": atom_count,

        "atoms_per_residue": (

            atom_count / residue_count

            if residue_count > 0

            else np.nan

        ),

        "missing_backbone_atoms": missing_backbone_count,

        "mean_B_factor": mean_bfactor,

        "median_B_factor": median_bfactor,

        "mean_occupancy": mean_occupancy

    })


df_final_chain_metrics = pd.DataFrame(
    chain_metric_data
)


# ============================================================
# 6. SEQUENCE IDENTITY BETWEEN CHAINS
# ============================================================

aligner = PairwiseAligner()

aligner.mode = "global"

aligner.match_score = 2.0

aligner.mismatch_score = -1.0

aligner.open_gap_score = -3.0

aligner.extend_gap_score = -0.5


chain_identity_data = []

chain_ids = list(chain_sequences.keys())


for i in range(len(chain_ids)):

    for j in range(i + 1, len(chain_ids)):

        chain_a = chain_ids[i]

        chain_b = chain_ids[j]

        sequence_a = str(chain_sequences[chain_a])

        sequence_b = str(chain_sequences[chain_b])


        if not sequence_a or not sequence_b:

            chain_identity_data.append({

                "chain_1": str(chain_a),

                "chain_2": str(chain_b),

                "alignment_score": np.nan,

                "identical_positions": np.nan,

                "aligned_positions": 0,

                "sequence_identity_percent": np.nan,

                "comparison_status": "Empty sequence"

            })

            continue


        alignment = aligner.align(
            sequence_a,
            sequence_b
        )[0]


        alignment_indices = alignment.indices

        identical_positions = 0

        aligned_positions = 0


        for position_a, position_b in zip(

            alignment_indices[0],

            alignment_indices[1]

        ):

            # В знаменатель входят только позиции,
            # в которых имеются остатки обеих цепей.
            if position_a < 0 or position_b < 0:
                continue

            aligned_positions += 1

            if sequence_a[position_a] == sequence_b[position_b]:
                identical_positions += 1


        identity = safe_percentage(
            identical_positions,
            aligned_positions
        )


        chain_identity_data.append({

            "chain_1": str(chain_a),

            "chain_2": str(chain_b),

            "alignment_score": float(alignment.score),

            "identical_positions": identical_positions,

            "aligned_positions": aligned_positions,

            "sequence_identity_percent": identity,

            "comparison_status": (
                "Global alignment; gaps excluded from identity denominator"
            )

        })


df_chain_identity = pd.DataFrame(
    chain_identity_data
)


# ============================================================
# 7. GEOMETRY METRICS
# ============================================================

geometry_metric_data = []


df_geometry_file = read_csv_if_exists(
    f"{PDB_ID}_geometry.csv"
)


if not df_geometry_file.empty:

    for column in df_geometry_file.columns:

        if column.lower() in {
            "pdb_id",
            "status",
            "output_checks"
        }:
            continue

        if len(df_geometry_file) == 1:

            value = df_geometry_file.iloc[0][column]

            if pd.api.types.is_numeric_dtype(
                df_geometry_file[column]
            ):

                geometry_metric_data.append({

                    "metric": column,

                    "value": json_safe(value),

                    "unit": None,

                    "status": "from CELL 9"

                })


if not geometry_metric_data:

    geometry_summary = globals().get(
        "geometry_summary",
        None
    )

    if isinstance(geometry_summary, dict):

        for metric_name, value in geometry_summary.items():

            if isinstance(value, (int, float, np.number)):

                geometry_metric_data.append({

                    "metric": str(metric_name),

                    "value": json_safe(value),

                    "unit": None,

                    "status": "from CELL 9"

                })


if "radius_of_gyration" in globals():

    if not any(
        "gyration" in str(item["metric"]).lower()
        for item in geometry_metric_data
    ):

        geometry_metric_data.append({

            "metric": "Geometric radius of gyration",

            "value": json_safe(radius_of_gyration),

            "unit": "A",

            "status": "from CELL 9"

        })


if "dimensions" in globals():

    try:

        dimensions_array = np.asarray(
            dimensions,
            dtype=float
        ).reshape(-1)

        if dimensions_array.size == 3:

            for axis, value in zip(
                ["X", "Y", "Z"],
                dimensions_array
            ):

                geometry_metric_data.append({

                    "metric": f"{axis} dimension",

                    "value": float(value),

                    "unit": "A",

                    "status": "from CELL 9"

                })

    except (TypeError, ValueError):
        pass


if not geometry_metric_data:

    geometry_metric_data.append({

        "metric": "Geometry metrics",

        "value": np.nan,

        "unit": None,

        "status": "CELL 9 geometry results unavailable"

    })


df_final_geometry_metrics = pd.DataFrame(

    geometry_metric_data,

    columns=[
        "metric",
        "value",
        "unit",
        "status"
    ]

)


# ============================================================
# 8. STRUCTURE QUALITY METRICS
# ============================================================

quality_metric_data = []


df_atom_quality_source, atom_quality_source_name = (
    dataframe_or_empty([
        "df_atom_quality",
        "df_structure_quality_atoms"
    ])
)


if df_atom_quality_source.empty:

    df_atom_quality_source = read_csv_if_exists(
        f"{PDB_ID}_atom_quality.csv"
    )

    if not df_atom_quality_source.empty:
        atom_quality_source_name = "CELL 14 CSV"


df_altloc_source, altloc_source_name = (
    dataframe_or_empty([
        "df_altloc_atoms",
        "df_altloc"
    ])
)


if df_altloc_source.empty:

    df_altloc_source = read_csv_if_exists(
        f"{PDB_ID}_altloc_atoms.csv"
    )

    if not df_altloc_source.empty:
        altloc_source_name = "CELL 14 CSV"


df_low_occupancy_source, low_occupancy_source_name = (
    dataframe_or_empty([
        "df_low_occupancy_atoms",
        "df_low_occupancy"
    ])
)


if df_low_occupancy_source.empty:

    df_low_occupancy_source = read_csv_if_exists(
        f"{PDB_ID}_low_occupancy_atoms.csv"
    )

    if not df_low_occupancy_source.empty:
        low_occupancy_source_name = "CELL 14 CSV"


df_missing_occupancy_source, _ = dataframe_or_empty([
    "df_missing_occupancy_atoms"
])


if df_missing_occupancy_source.empty:

    df_missing_occupancy_source = read_csv_if_exists(
        f"{PDB_ID}_missing_occupancy_atoms.csv"
    )


df_invalid_occupancy_source, _ = dataframe_or_empty([
    "df_invalid_occupancy_atoms"
])


if df_invalid_occupancy_source.empty:

    df_invalid_occupancy_source = read_csv_if_exists(
        f"{PDB_ID}_invalid_occupancy_atoms.csv"
    )


if not df_atom_quality_source.empty:

    quality_atom_count = len(
        df_atom_quality_source
    )

    quality_status = (
        f"from {atom_quality_source_name}"
    )

    quality_mean_bfactor = (

        safe_mean(df_atom_quality_source["bfactor"])

        if "bfactor" in df_atom_quality_source.columns

        else np.nan

    )

    quality_median_bfactor = (

        safe_median(df_atom_quality_source["bfactor"])

        if "bfactor" in df_atom_quality_source.columns

        else np.nan

    )

    quality_mean_occupancy = (

        safe_mean(df_atom_quality_source["occupancy"])

        if "occupancy" in df_atom_quality_source.columns

        else np.nan

    )

else:

    quality_atom_count = np.nan

    quality_status = (
        "CELL 14 atom-level table unavailable"
    )

    quality_mean_bfactor = np.nan

    quality_median_bfactor = np.nan

    quality_mean_occupancy = np.nan


# ------------------------------------------------------------
# 8.1. Alternate conformations
# ------------------------------------------------------------

if not df_altloc_source.empty:

    altloc_record_count = len(
        df_altloc_source
    )

    altloc_status = (
        f"from {altloc_source_name or 'CELL 14'}"
    )

elif (
    not df_atom_quality_source.empty
    and "altloc" in df_atom_quality_source.columns
):

    altloc_values = (

        df_atom_quality_source["altloc"]

        .fillna("")

        .astype(str)

        .str.strip()

    )

    altloc_record_count = int(
        (altloc_values != "").sum()
    )

    altloc_status = (
        "from atom quality table"
    )

else:

    altloc_record_count = np.nan

    altloc_status = (
        "altloc data unavailable"
    )


# ------------------------------------------------------------
# 8.2. Occupancy records
# ------------------------------------------------------------

if not df_low_occupancy_source.empty:

    low_occupancy_count = len(
        df_low_occupancy_source
    )

    low_occupancy_status = (
        f"from {low_occupancy_source_name or 'CELL 14'}"
    )

else:

    low_occupancy_count = np.nan

    low_occupancy_status = (
        "dedicated low occupancy table unavailable"
    )


missing_occupancy_count = (

    len(df_missing_occupancy_source)

    if not df_missing_occupancy_source.empty

    else np.nan

)


invalid_occupancy_count = (

    len(df_invalid_occupancy_source)

    if not df_invalid_occupancy_source.empty

    else np.nan

)


quality_metric_data.extend([

    {

        "metric": "Atom quality records",

        "value": quality_atom_count,

        "unit": "records",

        "status": quality_status

    },

    {

        "metric": "Alternate location atom records",

        "value": altloc_record_count,

        "unit": "records",

        "status": altloc_status

    },

    {

        "metric": "Low occupancy atom records",

        "value": low_occupancy_count,

        "unit": "records",

        "status": low_occupancy_status

    },

    {

        "metric": "Missing occupancy atom records",

        "value": missing_occupancy_count,

        "unit": "records",

        "status": (
            "from CELL 14"
            if pd.notna(missing_occupancy_count)
            else "data unavailable"
        )

    },

    {

        "metric": "Invalid occupancy atom records",

        "value": invalid_occupancy_count,

        "unit": "records",

        "status": (
            "from CELL 14"
            if pd.notna(invalid_occupancy_count)
            else "data unavailable"
        )

    },

    {

        "metric": "Mean B-factor",

        "value": quality_mean_bfactor,

        "unit": "A^2",

        "status": quality_status

    },

    {

        "metric": "Median B-factor",

        "value": quality_median_bfactor,

        "unit": "A^2",

        "status": quality_status

    },

    {

        "metric": "Mean occupancy",

        "value": quality_mean_occupancy,

        "unit": None,

        "status": quality_status

    }

])


# ------------------------------------------------------------
# 8.3. Missing backbone atoms
# ------------------------------------------------------------

missing_backbone_source, missing_backbone_source_name = (
    dataframe_or_empty([
        "missing_backbone_df",
        "df_missing_backbone_atoms"
    ])
)


if missing_backbone_source.empty:

    missing_backbone_source = read_csv_if_exists(
        f"{PDB_ID}_missing_backbone_atoms.csv"
    )

    if not missing_backbone_source.empty:
        missing_backbone_source_name = "CELL 6 CSV"


if not missing_backbone_source.empty:

    missing_backbone_count = len(
        missing_backbone_source
    )

    missing_backbone_status = (
        f"from {missing_backbone_source_name or 'CELL 6'}"
    )

else:

    missing_backbone_count = np.nan

    missing_backbone_status = (
        "missing backbone table unavailable"
    )


quality_metric_data.append({

    "metric": "Missing backbone atom records",

    "value": missing_backbone_count,

    "unit": "records",

    "status": missing_backbone_status

})


# ------------------------------------------------------------
# 8.4. Unobserved residues and atoms from mmCIF
# ------------------------------------------------------------

df_unobserved_residues_source, _ = dataframe_or_empty([
    "df_unobserved_residues",
    "df_missing_residues"
])


if df_unobserved_residues_source.empty:

    df_unobserved_residues_source = read_csv_if_exists(
        f"{PDB_ID}_unobserved_residues.csv"
    )


df_unobserved_atoms_source, _ = dataframe_or_empty([
    "df_unobserved_atoms",
    "df_missing_atoms"
])


if df_unobserved_atoms_source.empty:

    df_unobserved_atoms_source = read_csv_if_exists(
        f"{PDB_ID}_unobserved_atoms.csv"
    )


unobserved_residue_count = (

    len(df_unobserved_residues_source)

    if not df_unobserved_residues_source.empty

    else np.nan

)


unobserved_atom_count = (

    len(df_unobserved_atoms_source)

    if not df_unobserved_atoms_source.empty

    else np.nan

)


quality_metric_data.extend([

    {

        "metric": "Unobserved residue records in mmCIF",

        "value": unobserved_residue_count,

        "unit": "records",

        "status": (
            "from CELL 10"
            if pd.notna(unobserved_residue_count)
            else "data unavailable"
        )

    },

    {

        "metric": "Unobserved atom records in mmCIF",

        "value": unobserved_atom_count,

        "unit": "records",

        "status": (
            "from CELL 10"
            if pd.notna(unobserved_atom_count)
            else "data unavailable"
        )

    }

])


df_final_quality_metrics = pd.DataFrame(
    quality_metric_data
)


# ============================================================
# 9. COMPONENT DISTRIBUTION TABLE
# ============================================================

component_metric_data = []

if not component_counts.empty:

    for component, count in component_counts.items():

        component_metric_data.append({

            "component": str(component),

            "occurrence_count": int(count),

            "status": component_count_status

        })


df_final_component_metrics = pd.DataFrame(

    component_metric_data,

    columns=[
        "component",
        "occurrence_count",
        "status"
    ]

)


# ============================================================
# 10. BINDING SITE METRICS
# ============================================================
# CELL 18:
# - df_binding — таблица по остаткам binding site;
# - df_binding_atom_contacts — атомные пары контактов.
#
# Если переменные недоступны, используются имена файлов
# CELL 18: *_binding_site.csv и *_binding_atom_contacts.csv.
# ============================================================

df_binding_residue_source, binding_residue_source_name = (
    dataframe_or_empty([
        "df_binding",
        "df_binding_by_residue",
        "df_binding_residues"
    ])
)


df_binding_atom_source, binding_atom_source_name = (
    dataframe_or_empty([
        "df_binding_atom_contacts",
        "df_atom_contacts",
        "df_binding_contacts"
    ])
)


if df_binding_residue_source.empty:

    df_binding_residue_source = read_csv_if_exists(
        [
            f"{PDB_ID}_binding_site.csv",
            f"{PDB_ID}_binding_site_by_residue.csv"
        ]
    )

    if not df_binding_residue_source.empty:
        binding_residue_source_name = "CELL 18 CSV"


if df_binding_atom_source.empty:

    df_binding_atom_source = read_csv_if_exists(
        [
            f"{PDB_ID}_binding_atom_contacts.csv",
            f"{PDB_ID}_binding_site_atom_contacts.csv"
        ]
    )

    if not df_binding_atom_source.empty:
        binding_atom_source_name = "CELL 18 CSV"


selected_component_available = (
    globals().get("selected_component") is not None
)


if not selected_component_available:

    binding_residue_count = np.nan

    binding_atom_pair_count = np.nan

    binding_minimum_distance = np.nan

    binding_mean_residue_distance = np.nan

    binding_status = (
        "not applicable: no selected component"
    )

else:

    if not df_binding_residue_source.empty:

        binding_residue_count = len(
            df_binding_residue_source
        )

        minimum_distance_column = get_first_existing_column(

            df_binding_residue_source,

            [
                "minimum_distance",
                "min_distance",
                "minimum_distance_A",
                "min_distance_A"
            ]

        )

        if minimum_distance_column is not None:

            minimum_distances = numeric_values(
                df_binding_residue_source[minimum_distance_column]
            )

            binding_minimum_distance = (

                float(minimum_distances.min())

                if not minimum_distances.empty

                else np.nan

            )

            binding_mean_residue_distance = (

                float(minimum_distances.mean())

                if not minimum_distances.empty

                else np.nan

            )

        else:

            binding_minimum_distance = np.nan

            binding_mean_residue_distance = np.nan


        binding_status = (
            f"from {binding_residue_source_name}"
        )

    else:

        binding_residue_count = 0

        binding_minimum_distance = np.nan

        binding_mean_residue_distance = np.nan

        binding_status = (
            "no binding-site residue records available"
        )


    if not df_binding_atom_source.empty:

        binding_atom_pair_count = len(
            df_binding_atom_source
        )

        binding_atom_status = (
            f"from {binding_atom_source_name}"
        )

    elif not df_binding_residue_source.empty:

        atom_contacts_column = get_first_existing_column(

            df_binding_residue_source,

            [
                "atom_contacts",
                "number_of_atom_contacts",
                "contact_count"
            ]

        )

        if atom_contacts_column is not None:

            binding_atom_pair_count = int(

                pd.to_numeric(

                    df_binding_residue_source[atom_contacts_column],

                    errors="coerce"

                ).fillna(0).sum()

            )

            binding_atom_status = (
                "summed from residue-level contact counts"
            )

        else:

            binding_atom_pair_count = np.nan

            binding_atom_status = (
                "atom contact table unavailable"
            )

    else:

        binding_atom_pair_count = np.nan

        binding_atom_status = (
            "atom contact table unavailable"
        )


df_final_binding_metrics = pd.DataFrame([

    {

        "metric": "Binding site residues",

        "value": binding_residue_count,

        "unit": "residues",

        "status": binding_status

    },

    {

        "metric": "Protein-component atom-pair contacts",

        "value": binding_atom_pair_count,

        "unit": "contacts",

        "status": (
            binding_atom_status
            if selected_component_available
            else binding_status
        )

    },

    {

        "metric": "Minimum component-protein distance",

        "value": binding_minimum_distance,

        "unit": "A",

        "status": binding_status

    },

    {

        "metric": "Mean minimum residue distance",

        "value": binding_mean_residue_distance,

        "unit": "A",

        "status": binding_status

    }

])


# ============================================================
# 11. SELECTED COMPONENT INTERACTION METRICS
# ============================================================

df_interactions_source, interaction_source_name = (
    dataframe_or_empty([
        "df_interactions"
    ])
)


interaction_metric_data = []


if interaction_source_name is None:

    interaction_status = (
        "CELL 19 interaction table unavailable"
    )

    total_interactions = np.nan

else:

    total_interactions = len(
        df_interactions_source
    )

    interaction_status = (

        "from CELL 19"

        if not df_interactions_source.empty

        else "CELL 19 completed; no contacts found"

    )


add_metric(
    interaction_metric_data,
    "Interactions",
    "Total component-protein atom-pair contacts",
    total_interactions,
    "contacts",
    interaction_status
)


if (
    not df_interactions_source.empty
    and "interaction_type" in df_interactions_source.columns
):

    interaction_counts = (
        df_interactions_source["interaction_type"]
        .fillna("Unknown")
        .astype(str)
        .value_counts()
    )

    for interaction_type, count in interaction_counts.items():

        interaction_metric_data.append({

            "category": "Interactions",

            "metric": str(interaction_type),

            "value": int(count),

            "unit": "records",

            "status": "distance-based classification"

        })


df_final_interaction_metrics = pd.DataFrame(

    interaction_metric_data,

    columns=[
        "category",
        "metric",
        "value",
        "unit",
        "status"
    ]

)


# ============================================================
# 12. HEM IRON COORDINATION METRICS
# ============================================================
# CELL 19:
# df_metal_coordination — контакты Fe в распознанных гемах.
# df_all_metal_coordination — общая таблица металлических контактов.
#
# Возможные neighbor_type:
# PROTEIN
# Same HEME component
# OXY
# OTHER_COMPONENT
#
# Названия категорий нормализуются перед сравнением.
# ============================================================

df_heme_fe_source, heme_source_name = (
    dataframe_or_empty([
        "df_metal_coordination"
    ])
)


df_all_metal_source, all_metal_source_name = (
    dataframe_or_empty([
        "df_all_metal_coordination"
    ])
)


heme_metric_data = []


if heme_source_name is None:

    heme_status = (
        "CELL 19 heme table unavailable"
    )

    total_fe_contacts = np.nan

else:

    total_fe_contacts = len(
        df_heme_fe_source
    )

    heme_status = (

        "from CELL 19"

        if not df_heme_fe_source.empty

        else "no Fe contacts in recognized hemes"

    )


add_metric(
    heme_metric_data,
    "HEM iron coordination",
    "Potential Fe coordination contacts",
    total_fe_contacts,
    "contacts",
    heme_status
)


if not df_heme_fe_source.empty:

    if "neighbor_type" in df_heme_fe_source.columns:

        normalized_neighbor_types = (

            df_heme_fe_source["neighbor_type"]

            .fillna("UNKNOWN")

            .apply(normalize_category)

        )

        protein_fe = df_heme_fe_source[
            normalized_neighbor_types == "PROTEIN"
        ]

        heme_fe = df_heme_fe_source[
            normalized_neighbor_types == "SAME HEME COMPONENT"
        ]

        oxy_fe = df_heme_fe_source[
            normalized_neighbor_types == "OXY"
        ]

        other_fe = df_heme_fe_source[
            normalized_neighbor_types == "OTHER COMPONENT"
        ]

    else:

        protein_fe = pd.DataFrame()

        heme_fe = pd.DataFrame()

        oxy_fe = pd.DataFrame()

        other_fe = pd.DataFrame()


    add_metric(
        heme_metric_data,
        "HEM iron coordination",
        "Potential protein-Fe contacts",
        len(protein_fe),
        "contacts",
        heme_status
    )

    add_metric(
        heme_metric_data,
        "HEM iron coordination",
        "Intracomponent Fe contacts in HEM",
        len(heme_fe),
        "contacts",
        heme_status
    )

    add_metric(
        heme_metric_data,
        "HEM iron coordination",
        "Potential Fe-OXY contacts",
        len(oxy_fe),
        "contacts",
        heme_status
    )

    add_metric(
        heme_metric_data,
        "HEM iron coordination",
        "Potential Fe contacts with other components",
        len(other_fe),
        "contacts",
        heme_status
    )


    distance_column = get_first_existing_column(
        df_heme_fe_source,
        ["distance", "distance_A"]
    )

    if distance_column is not None:

        all_fe_distances = numeric_values(
            df_heme_fe_source[distance_column]
        )

        add_metric(
            heme_metric_data,
            "HEM iron coordination",
            "Minimum Fe-neighbor distance",
            (
                float(all_fe_distances.min())
                if not all_fe_distances.empty
                else np.nan
            ),
            "A",
            heme_status
        )


    if not protein_fe.empty and distance_column is not None:

        protein_distances = numeric_values(
            protein_fe[distance_column]
        )

        add_metric(
            heme_metric_data,
            "HEM iron coordination",
            "Minimum Fe-protein distance",
            (
                float(protein_distances.min())
                if not protein_distances.empty
                else np.nan
            ),
            "A",
            heme_status
        )


    if not oxy_fe.empty and distance_column is not None:

        oxy_distances = numeric_values(
            oxy_fe[distance_column]
        )

        add_metric(
            heme_metric_data,
            "HEM iron coordination",
            "Minimum Fe-OXY distance",
            (
                float(oxy_distances.min())
                if not oxy_distances.empty
                else np.nan
            ),
            "A",
            heme_status
        )


df_final_heme_metrics = pd.DataFrame(

    heme_metric_data,

    columns=[
        "category",
        "metric",
        "value",
        "unit",
        "status"
    ]

)


# ============================================================
# 13. Fe-N DISTANCES WITHIN ALL HEM COMPONENTS
# ============================================================
# Расстояния рассчитываются отдельно для каждой копии HEM.
# Учитываются атомы Fe и N внутри одного остатка HEM.
# Используется первая модель и выбранное Bio.PDB представление
# альтернативных конформеров.
# ============================================================

fe_n_distance_data = []


if component_structure_available:

    for chain in analysis_model:

        for residue in chain:

            residue_code = (
                residue.get_resname().strip().upper()
            )

            if residue_code != "HEM":
                continue


            atoms = get_selected_atoms(residue)


            fe_atoms = [

                atom

                for atom in atoms

                if get_atom_element(atom) == "FE"

            ]


            nitrogen_atoms = [

                atom

                for atom in atoms

                if get_atom_element(atom) == "N"

            ]


            for fe_atom in fe_atoms:

                fe_coordinate = get_valid_coordinate(
                    fe_atom
                )

                if fe_coordinate is None:
                    continue


                for nitrogen_atom in nitrogen_atoms:

                    nitrogen_coordinate = get_valid_coordinate(
                        nitrogen_atom
                    )

                    if nitrogen_coordinate is None:
                        continue


                    distance = float(

                        np.linalg.norm(
                            fe_coordinate - nitrogen_coordinate
                        )

                    )


                    fe_n_distance_data.append({

                        "pdb_id": PDB_ID,

                        "model_id": analysis_model_id,

                        "chain_id": str(chain.id),

                        "residue_number": get_residue_number(residue),

                        "insertion_code": get_insertion_code(residue),

                        "heme_component": residue_code,

                        "iron_atom": fe_atom.get_name().strip(),

                        "nitrogen_atom": nitrogen_atom.get_name().strip(),

                        "distance_A": distance

                    })


df_fe_n_distances = pd.DataFrame(

    fe_n_distance_data,

    columns=[

        "pdb_id",
        "model_id",
        "chain_id",
        "residue_number",
        "insertion_code",
        "heme_component",
        "iron_atom",
        "nitrogen_atom",
        "distance_A"

    ]

)


if not df_fe_n_distances.empty:

    df_fe_n_summary = (

        df_fe_n_distances

        .groupby(

            [
                "model_id",
                "chain_id",
                "residue_number",
                "insertion_code",
                "heme_component"
            ],

            dropna=False

        )

        .agg(

            number_of_Fe_N_distances=(
                "distance_A",
                "count"
            ),

            minimum_distance_A=(
                "distance_A",
                "min"
            ),

            maximum_distance_A=(
                "distance_A",
                "max"
            ),

            mean_distance_A=(
                "distance_A",
                "mean"
            ),

            std_distance_A=(
                "distance_A",
                "std"
            )

        )

        .reset_index()

    )

else:

    df_fe_n_summary = pd.DataFrame(

        columns=[

            "model_id",
            "chain_id",
            "residue_number",
            "insertion_code",
            "heme_component",
            "number_of_Fe_N_distances",
            "minimum_distance_A",
            "maximum_distance_A",
            "mean_distance_A",
            "std_distance_A"

        ]

    )


# ============================================================
# 14. Fe-OXY DISTANCES
# ============================================================
# Рассчитываются геометрические расстояния между Fe в HEM
# и кислородными атомами остатков OXY/O2 первой модели.
#
# Это геометрическая оценка, а не доказательство связывания O2.
# ============================================================

fe_oxy_distance_data = []


if component_structure_available:

    fe_records = []

    oxy_records = []


    for chain in analysis_model:

        for residue in chain:

            residue_code = (
                residue.get_resname().strip().upper()
            )

            atoms = get_selected_atoms(residue)


            # В этой таблице учитываются только HEM,
            # чтобы область анализа совпадала с Fe-N.
            if residue_code == "HEM":

                for atom in atoms:

                    if get_atom_element(atom) != "FE":
                        continue

                    coordinate = get_valid_coordinate(atom)

                    if coordinate is None:
                        continue

                    fe_records.append({

                        "chain_id": str(chain.id),

                        "residue_number": get_residue_number(residue),

                        "insertion_code": get_insertion_code(residue),

                        "heme_component": residue_code,

                        "atom": atom,

                        "coordinate": coordinate

                    })


            elif residue_code in {"OXY", "O2"}:

                for atom in atoms:

                    if get_atom_element(atom) != "O":
                        continue

                    coordinate = get_valid_coordinate(atom)

                    if coordinate is None:
                        continue

                    oxy_records.append({

                        "chain_id": str(chain.id),

                        "residue_number": get_residue_number(residue),

                        "insertion_code": get_insertion_code(residue),

                        "oxy_component": residue_code,

                        "atom": atom,

                        "coordinate": coordinate

                    })


    for fe_record in fe_records:

        for oxy_record in oxy_records:

            distance = float(

                np.linalg.norm(

                    fe_record["coordinate"]
                    -
                    oxy_record["coordinate"]

                )

            )


            fe_oxy_distance_data.append({

                "pdb_id": PDB_ID,

                "model_id": analysis_model_id,

                "heme_chain": fe_record["chain_id"],

                "heme_residue_number": (
                    fe_record["residue_number"]
                ),

                "heme_insertion_code": (
                    fe_record["insertion_code"]
                ),

                "heme_component": fe_record["heme_component"],

                "iron_atom": fe_record["atom"].get_name().strip(),

                "oxy_chain": oxy_record["chain_id"],

                "oxy_residue_number": (
                    oxy_record["residue_number"]
                ),

                "oxy_insertion_code": (
                    oxy_record["insertion_code"]
                ),

                "oxy_component": oxy_record["oxy_component"],

                "oxygen_atom": oxy_record["atom"].get_name().strip(),

                "distance_A": distance

            })


df_fe_oxy_distances = pd.DataFrame(

    fe_oxy_distance_data,

    columns=[

        "pdb_id",
        "model_id",
        "heme_chain",
        "heme_residue_number",
        "heme_insertion_code",
        "heme_component",
        "iron_atom",
        "oxy_chain",
        "oxy_residue_number",
        "oxy_insertion_code",
        "oxy_component",
        "oxygen_atom",
        "distance_A"

    ]

)


if not df_fe_oxy_distances.empty:

    df_fe_oxy_summary = (

        df_fe_oxy_distances

        .groupby(

            [
                "heme_chain",
                "heme_residue_number",
                "heme_insertion_code",
                "heme_component",
                "oxy_chain",
                "oxy_residue_number",
                "oxy_insertion_code",
                "oxy_component"
            ],

            dropna=False

        )

        .agg(

            number_of_Fe_OXY_distances=(
                "distance_A",
                "count"
            ),

            minimum_distance_A=(
                "distance_A",
                "min"
            ),

            mean_distance_A=(
                "distance_A",
                "mean"
            ),

            maximum_distance_A=(
                "distance_A",
                "max"
            )

        )

        .reset_index()

    )

else:

    df_fe_oxy_summary = pd.DataFrame(

        columns=[

            "heme_chain",
            "heme_residue_number",
            "heme_insertion_code",
            "heme_component",
            "oxy_chain",
            "oxy_residue_number",
            "oxy_insertion_code",
            "oxy_component",
            "number_of_Fe_OXY_distances",
            "minimum_distance_A",
            "mean_distance_A",
            "maximum_distance_A"

        ]

    )


# ============================================================
# 15. DISULFIDE BRIDGE METRICS
# ============================================================

df_disulfide_source, disulfide_source_name = (
    dataframe_or_empty([
        "df_disulfide_bridges"
    ])
)


if disulfide_source_name is None:

    df_disulfide_source = read_csv_if_exists(
        f"{PDB_ID}_disulfide_bridges.csv"
    )

    if not df_disulfide_source.empty:
        disulfide_source_name = "CELL 19 CSV"


if disulfide_source_name is None:

    disulfide_count = np.nan

    disulfide_status = (
        "CELL 19 disulfide table unavailable"
    )

else:

    disulfide_count = len(
        df_disulfide_source
    )

    disulfide_status = (

        "potential SG-SG contacts from CELL 19"

        if disulfide_count > 0

        else "CELL 19 completed; no pairs within cutoff"

    )


disulfide_metric_data = []


add_metric(
    disulfide_metric_data,
    "Disulfide bridges",
    "Potential disulfide bridge count",
    disulfide_count,
    "pairs",
    disulfide_status
)


disulfide_distance_column = get_first_existing_column(

    df_disulfide_source,

    ["distance", "distance_A"]

)


if (
    not df_disulfide_source.empty
    and disulfide_distance_column is not None
):

    disulfide_distances = numeric_values(
        df_disulfide_source[disulfide_distance_column]
    )

    add_metric(
        disulfide_metric_data,
        "Disulfide bridges",
        "Minimum SG-SG distance",
        (
            float(disulfide_distances.min())
            if not disulfide_distances.empty
            else np.nan
        ),
        "A",
        disulfide_status
    )

    add_metric(
        disulfide_metric_data,
        "Disulfide bridges",
        "Mean SG-SG distance",
        (
            float(disulfide_distances.mean())
            if not disulfide_distances.empty
            else np.nan
        ),
        "A",
        disulfide_status
    )


df_final_disulfide_metrics = pd.DataFrame(

    disulfide_metric_data,

    columns=[
        "category",
        "metric",
        "value",
        "unit",
        "status"
    ]

)


# ============================================================
# 16. BUILD MASTER METRICS TABLE
# ============================================================

master_metric_data = []


# ------------------------------------------------------------
# 16.1. Global metrics
# ------------------------------------------------------------

for metric_name, metric_value in global_metrics.items():

    if metric_name == "PDB_ID":
        continue

    if "component" in metric_name:

        status = component_count_status

    elif "chains_from_sequence" in metric_name:

        status = "from chain_sequences"

    elif "chains_in_residue" in metric_name:

        status = "from df_residues"

    else:

        status = "calculated"


    add_metric(
        master_metric_data,
        "Global structure",
        metric_name,
        metric_value,
        None,
        status
    )


# ------------------------------------------------------------
# 16.2. Append metric DataFrames
# ------------------------------------------------------------

def append_metric_dataframe(
    dataframe,
    category,
    metric_column="metric",
    value_column="value",
    unit_column="unit",
    status_column="status"
):

    if dataframe is None or dataframe.empty:
        return

    for _, row in dataframe.iterrows():

        unit = (
            row[unit_column]
            if unit_column in dataframe.columns
            else None
        )

        status = (
            row[status_column]
            if status_column in dataframe.columns
            else "calculated"
        )

        add_metric(
            master_metric_data,
            category,
            row[metric_column],
            row[value_column],
            unit,
            status
        )


append_metric_dataframe(
    df_final_geometry_metrics,
    "Geometry"
)

append_metric_dataframe(
    df_final_quality_metrics,
    "Structure quality"
)

append_metric_dataframe(
    df_final_binding_metrics,
    "Binding site"
)

append_metric_dataframe(
    df_final_interaction_metrics,
    "Interactions",
    metric_column="metric",
    value_column="value"
)

append_metric_dataframe(
    df_final_heme_metrics,
    "HEM iron coordination"
)

append_metric_dataframe(
    df_final_disulfide_metrics,
    "Disulfide bridges"
)


# ------------------------------------------------------------
# 16.3. Chain metrics
# ------------------------------------------------------------

for _, row in df_final_chain_metrics.iterrows():

    chain_id = row["chain_id"]

    for metric_name in [

        "sequence_length",
        "residue_count",
        "atom_count",
        "atoms_per_residue",
        "missing_backbone_atoms",
        "mean_B_factor",
        "median_B_factor",
        "mean_occupancy"

    ]:

        add_metric(
            master_metric_data,
            f"Chain {chain_id}",
            metric_name,
            row[metric_name],
            None,
            "calculated from first-model tables"
        )


# ------------------------------------------------------------
# 16.4. Sequence identity
# ------------------------------------------------------------

for _, row in df_chain_identity.iterrows():

    add_metric(
        master_metric_data,
        "Sequence comparison",
        f"Identity {row['chain_1']}-{row['chain_2']}",
        row["sequence_identity_percent"],
        "%",
        row["comparison_status"]
    )


# ------------------------------------------------------------
# 16.5. Component occurrence counts
# ------------------------------------------------------------

for _, row in df_final_component_metrics.iterrows():

    add_metric(
        master_metric_data,
        "Components",
        f"{row['component']} occurrences",
        row["occurrence_count"],
        "instances",
        row["status"]
    )


# ------------------------------------------------------------
# 16.6. Fe-N distances
# ------------------------------------------------------------

if not df_fe_n_summary.empty:

    for _, row in df_fe_n_summary.iterrows():

        category = (

            f"HEM Fe-N chain {row['chain_id']}, "
            f"residue {row['residue_number']}"

        )

        for metric_name in [

            "number_of_Fe_N_distances",
            "minimum_distance_A",
            "maximum_distance_A",
            "mean_distance_A",
            "std_distance_A"

        ]:

            unit = (
                "A"
                if metric_name.endswith("_A")
                else "distances"
            )

            add_metric(
                master_metric_data,
                category,
                metric_name,
                row[metric_name],
                unit,
                "calculated from first model"
            )


# ------------------------------------------------------------
# 16.7. Fe-OXY distances
# ------------------------------------------------------------

if not df_fe_oxy_summary.empty:

    for _, row in df_fe_oxy_summary.iterrows():

        category = (

            f"HEM Fe-OXY chain {row['heme_chain']}, "
            f"residue {row['heme_residue_number']}"

        )

        for metric_name in [

            "number_of_Fe_OXY_distances",
            "minimum_distance_A",
            "mean_distance_A",
            "maximum_distance_A"

        ]:

            unit = (
                "A"
                if metric_name.endswith("_A")
                else "distances"
            )

            add_metric(
                master_metric_data,
                category,
                metric_name,
                row[metric_name],
                unit,
                "calculated from first model"
            )


df_final_metrics = pd.DataFrame(

    master_metric_data,

    columns=[
        "category",
        "metric",
        "value",
        "unit",
        "status"
    ]

)


# ============================================================
# 17. DISPLAY RESULTS
# ============================================================

print("\n" + "=" * 75)
print("1. GLOBAL STRUCTURE METRICS")
print("=" * 75)

display(df_global_metrics)


print("\n" + "=" * 75)
print("2. ATOM ELEMENT DISTRIBUTION")
print("=" * 75)

display(df_element_distribution)


print("\n" + "=" * 75)
print("3. PROTEIN CHAIN METRICS")
print("=" * 75)

display(df_final_chain_metrics)


print("\n" + "=" * 75)
print("4. SEQUENCE IDENTITY BETWEEN CHAINS")
print("=" * 75)

display(df_chain_identity)


print("\n" + "=" * 75)
print("5. GEOMETRY METRICS")
print("=" * 75)

display(df_final_geometry_metrics)


print("\n" + "=" * 75)
print("6. STRUCTURE QUALITY METRICS")
print("=" * 75)

display(df_final_quality_metrics)


print("\n" + "=" * 75)
print("7. COMPONENT DISTRIBUTION")
print("=" * 75)

display(df_component_distribution)


print("\n" + "=" * 75)
print("8. BINDING SITE METRICS")
print("=" * 75)

display(df_final_binding_metrics)


print("\n" + "=" * 75)
print("9. SELECTED COMPONENT INTERACTION METRICS")
print("=" * 75)

display(df_final_interaction_metrics)


print("\n" + "=" * 75)
print("10. HEM IRON COORDINATION METRICS")
print("=" * 75)

display(df_final_heme_metrics)


print("\n" + "=" * 75)
print("11. Fe-N DISTANCES")
print("=" * 75)

display(df_fe_n_distances)

print("\nFe-N summary by HEM:")

display(df_fe_n_summary)


print("\n" + "=" * 75)
print("12. Fe-OXY DISTANCES")
print("=" * 75)

display(df_fe_oxy_distances)

print("\nFe-OXY summary:")

display(df_fe_oxy_summary)


print("\n" + "=" * 75)
print("13. DISULFIDE BRIDGE METRICS")
print("=" * 75)

display(df_final_disulfide_metrics)


if not df_disulfide_source.empty:

    print("\nPotential disulfide bridges:")

    display(df_disulfide_source)


print("\n" + "=" * 75)
print("14. MASTER METRICS TABLE")
print("=" * 75)

display(df_final_metrics)


# ============================================================
# 18. SAVE CSV FILES
# ============================================================

metrics_files = {

    "global": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_global_metrics.csv"
    ),

    "elements": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_element_distribution.csv"
    ),

    "chains": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_chain_metrics.csv"
    ),

    "chain_identity": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_chain_identity.csv"
    ),

    "geometry": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_geometry_metrics.csv"
    ),

    "quality": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_quality_metrics.csv"
    ),

    "components": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_component_metrics.csv"
    ),

    "component_distribution": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_component_distribution.csv"
    ),

    "binding": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_binding_metrics.csv"
    ),

    "interactions": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_interaction_metrics.csv"
    ),

    "heme": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_heme_metrics.csv"
    ),

    "fe_n_distances": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_fe_n_distances.csv"
    ),

    "fe_n_summary": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_fe_n_summary.csv"
    ),

    "fe_oxy_distances": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_fe_oxy_distances.csv"
    ),

    "fe_oxy_summary": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_fe_oxy_summary.csv"
    ),

    "disulfide_metrics": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_final_disulfide_metrics.csv"
    ),

    "all_metrics": (
        PDB_RESULTS_DIR
        / f"{PDB_ID}_all_final_metrics.csv"
    )

}


dataframes_to_save = {

    "global": df_global_metrics,

    "elements": df_element_distribution,

    "chains": df_final_chain_metrics,

    "chain_identity": df_chain_identity,

    "geometry": df_final_geometry_metrics,

    "quality": df_final_quality_metrics,

    "components": df_final_component_metrics,

    "component_distribution": df_component_distribution,

    "binding": df_final_binding_metrics,

    "interactions": df_final_interaction_metrics,

    "heme": df_final_heme_metrics,

    "fe_n_distances": df_fe_n_distances,

    "fe_n_summary": df_fe_n_summary,

    "fe_oxy_distances": df_fe_oxy_distances,

    "fe_oxy_summary": df_fe_oxy_summary,

    "disulfide_metrics": df_final_disulfide_metrics,

    "all_metrics": df_final_metrics

}


for name, dataframe in dataframes_to_save.items():

    dataframe.to_csv(

        metrics_files[name],

        index=False,

        encoding="utf-8-sig"

    )


# ============================================================
# 19. FINAL VALIDATION
# ============================================================

print("\n" + "=" * 75)
print("FINAL METRICS VALIDATION")
print("=" * 75)


validation_results = {

    "PDB_ID_valid": (
        len(PDB_ID) == 4
        and PDB_ID.isalnum()
    ),

    "PDB_RESULTS_DIR_exists": (
        PDB_RESULTS_DIR.exists()
    ),

    "df_atoms_not_empty": (
        not df_atoms.empty
    ),

    "df_residues_is_dataframe": (
        isinstance(df_residues, pd.DataFrame)
    ),

    "global_metrics_created": (
        not df_global_metrics.empty
    ),

    "chain_metrics_created": (
        isinstance(df_final_chain_metrics, pd.DataFrame)
    ),

    "master_metrics_created": (
        isinstance(df_final_metrics, pd.DataFrame)
    ),

    "Fe_N_distances_valid": (

        df_fe_n_distances.empty

        or (

            df_fe_n_distances["distance_A"].notna().all()

            and
            np.isfinite(
                df_fe_n_distances["distance_A"].to_numpy(
                    dtype=float
                )
            ).all()

            and
            (df_fe_n_distances["distance_A"] >= 0).all()

        )

    ),

    "Fe_OXY_distances_valid": (

        df_fe_oxy_distances.empty

        or (

            df_fe_oxy_distances["distance_A"].notna().all()

            and
            np.isfinite(
                df_fe_oxy_distances["distance_A"].to_numpy(
                    dtype=float
                )
            ).all()

            and
            (df_fe_oxy_distances["distance_A"] >= 0).all()

        )

    ),

    "component_count_not_negative": (

        pd.isna(number_of_component_instances)

        or number_of_component_instances >= 0

    ),

    "binding_contact_count_not_negative": (

        pd.isna(binding_atom_pair_count)

        or binding_atom_pair_count >= 0

    )

}


# ------------------------------------------------------------
# 19.1. Проверка файлов CSV
# ------------------------------------------------------------

for name, path in metrics_files.items():

    validation_results[
        f"file_{name}_exists"
    ] = (
        path.is_file()
        and path.stat().st_size > 0
    )


failed_validations = [

    name

    for name, valid in validation_results.items()

    if not valid

]


for name, valid in validation_results.items():

    print(
        f"{name}: {'OK' if valid else 'CHECK'}"
    )


print("\nSAVED FILES")

for name, path in metrics_files.items():

    print(

        f"[{'OK' if path.exists() else 'ERROR'}] "

        f"{path.name}",

        "|",

        path.stat().st_size if path.exists() else 0,

        "bytes"

    )


# ============================================================
# 20. SAVE JSON SUMMARY
# ============================================================

final_metrics_json = {

    "pdb_id": PDB_ID,

    "model_scope": {

        "source_model_count": (
            len(cell20_models)
            if component_structure_available
            else None
        ),

        "analyzed_model_id": analysis_model_id,

        "scope": "First model only"

    },

    "global_metrics": global_metrics,

    "component_count_status": component_count_status,

    "data_sources": {

        "protein_atoms": (
            "CELL 7 df_atoms; first model"
        ),

        "protein_residues": (
            "CELL 6 df_residues; first model"
        ),

        "hetero_components": (
            components_source_name
            or "CELL 8 df_hetero"
        ),

        "geometry": (
            "CELL 9 output"
        ),

        "structure_quality": (
            "CELL 14 output, where available"
        ),

        "binding_site": (
            "CELL 18 residue and atom contact tables, where available"
        ),

        "component_interactions": (
            "CELL 19 df_interactions"
        ),

        "heme_iron_coordination": (
            "CELL 19 df_metal_coordination"
        ),

        "universal_metal_coordination": (
            "CELL 19 df_all_metal_coordination"
        ),

        "disulfide_bridges": (
            "CELL 19 df_disulfide_bridges"
        )

    },

    "counts": {

        "chain_metric_records": len(
            df_final_chain_metrics
        ),

        "chain_identity_comparisons": len(
            df_chain_identity
        ),

        "geometry_metrics": len(
            df_final_geometry_metrics
        ),

        "quality_metrics": len(
            df_final_quality_metrics
        ),

        "component_types": len(
            df_final_component_metrics
        ),

        "binding_metrics": len(
            df_final_binding_metrics
        ),

        "interaction_metrics": len(
            df_final_interaction_metrics
        ),

        "heme_metrics": len(
            df_final_heme_metrics
        ),

        "Fe_N_distances": len(
            df_fe_n_distances
        ),

        "Fe_N_HEM_summaries": len(
            df_fe_n_summary
        ),

        "Fe_OXY_distances": len(
            df_fe_oxy_distances
        ),

        "Fe_OXY_summaries": len(
            df_fe_oxy_summary
        ),

        "potential_disulfide_bridges": (
            disulfide_count
        ),

        "master_metrics": len(
            df_final_metrics
        )

    },

    "interpretation_notes": [

        "Protein geometry is based on protein atoms from df_atoms.",

        "Water and non-protein components are excluded from protein geometry.",

        "Chain sequence identity is calculated using global pairwise alignment.",

        "Gaps are excluded from the sequence identity denominator.",

        "Binding-site metrics refer to the selected component from CELL 17-18.",

        "Fe-N distances are calculated separately for each HEM component.",

        "Fe-OXY distances are calculated between HEM iron and OXY/O2 oxygen atoms.",

        "Fe-N and Fe-OXY distances use the selected Bio.PDB atom conformer.",

        "Metal contacts are geometric candidates, not definitive coordination bonds.",

        "Polar atom proximity is not proof of a hydrogen bond.",

        "Potential disulfide bridges require chemical connectivity confirmation.",

        "Alternate conformer occupancy values are not automatically treated as low quality.",

        "Missing source tables are reported as unavailable rather than zero.",

        "Component occurrence counts refer to structural instances, not unique chemical identities."

    ],

    "metric_tables": {

        name: path.name

        for name, path in metrics_files.items()

    },

    "validation": validation_results,

    "validation_status": (

        "PASSED"

        if not failed_validations

        else "CHECK_REQUIRED"

    )

}


metrics_json_path = (

    PDB_RESULTS_DIR
    / f"{PDB_ID}_final_metrics.json"

)


with open(

    metrics_json_path,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        json_safe(final_metrics_json),

        file,

        ensure_ascii=False,

        indent=4,

        allow_nan=False

    )


validation_results["json_summary_exists"] = (

    metrics_json_path.is_file()

    and metrics_json_path.stat().st_size > 0

)


# Если JSON не удалось создать, это отдельная ошибка.
if not validation_results["json_summary_exists"]:

    failed_validations.append(
        "json_summary_exists"
    )


# ============================================================
# 21. FINAL REPORT
# ============================================================

print("\n" + "=" * 75)
print("FINAL METRICS CALCULATION COMPLETED")
print("=" * 75)

print("PDB ID:", PDB_ID)

print("Protein chains from sequence dictionary:", number_of_chains)

print("Protein chains in residue table:", actual_residue_chains)

print("Protein residues:", number_of_protein_residues)

print("Protein atoms:", number_of_protein_atoms)

print("Component instances:", number_of_component_instances)

print("Unique component types:", number_of_unique_component_types)

print("HEM Fe-N distances:", len(df_fe_n_distances))

print("Fe-OXY distances:", len(df_fe_oxy_distances))

print("Potential disulfide bridges:", disulfide_count)

print("Master metrics:", len(df_final_metrics))

print("Output directory:", PDB_RESULTS_DIR)


if failed_validations:

    print("\nWARNING: Some validation checks require attention:")

    for name in sorted(set(failed_validations)):
        print(" -", name)

else:

    print("\nAll CELL 20 validation checks passed.")


print("\nJSON SUMMARY:")

print(
    f"[{'OK' if metrics_json_path.exists() else 'ERROR'}] "
    f"{metrics_json_path.name}"
)


print("=" * 75)
print("CELL 20 COMPLETED")
print("=" * 75)
```

# Создание единого файла с признаками белка

```python
# ============================================================
# CELL 22. PROTEIN FEATURE EXTRACTION
# ============================================================
# Назначение:
# 1. Собрать структурные данные из предыдущих ячеек.
# 2. Сформировать глобальные, атомные и остаточные признаки.
# 3. Подключить геометрические признаки CELL 9.
# 4. Подключить метаданные и показатели качества CELL 10/14.
# 5. Подключить область связывания CELL 18.
# 6. Подключить взаимодействия и координацию металлов CELL 19.
# 7. Сформировать объединённую таблицу записей о взаимодействиях.
# 8. Сохранить CSV-таблицы и единый JSON.
#
# ВАЖНО:
# - Исходные DataFrame не изменяются.
# - Атомные и остаточные признаки относятся к первой модели.
# - Таблицы качества CELL 14 могут включать все модели и altloc.
# - Отсутствующие признаки не заменяются искусственными нулями.
# - Геометрические контакты не доказывают химическую связь.
# - Расстояния измеряются в Angstrom (Å).
# - Результаты сохраняются в PDB_RESULTS_DIR/features.
# ============================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from datetime import datetime


print("=" * 75)
print("CELL 22. PROTEIN FEATURE EXTRACTION")
print("=" * 75)


# ------------------------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------------------------

for variable in ["PDB_ID", "PDB_RESULTS_DIR"]:
    if variable not in globals():
        raise NameError(
            f"Required variable '{variable}' is not defined. "
            "Run the previous cells first."
        )

PDB_ID_VALUE = str(PDB_ID).strip().upper()

if len(PDB_ID_VALUE) != 4:
    raise ValueError(f"Invalid PDB_ID: {PDB_ID_VALUE}")

PDB_RESULTS_DIR_VALUE = Path(PDB_RESULTS_DIR)
PDB_RESULTS_DIR_VALUE.mkdir(parents=True, exist_ok=True)

FEATURE_DIR = PDB_RESULTS_DIR_VALUE / "features"
FEATURE_DIR.mkdir(parents=True, exist_ok=True)

ANALYSIS_SCOPE = "first_model"

print("PDB ID:", PDB_ID_VALUE)
print("PDB results directory:", PDB_RESULTS_DIR_VALUE)
print("Feature directory:", FEATURE_DIR)
print("Analysis scope:", ANALYSIS_SCOPE)


# ------------------------------------------------------------
# 2. HELPER FUNCTIONS
# ------------------------------------------------------------

def is_missing_scalar(value):
    """Проверка пропуска без ошибки на списках и массивах."""
    if value is None:
        return True

    try:
        result = pd.isna(value)
        return bool(result) if np.isscalar(result) else False
    except (TypeError, ValueError):
        return False


def safe_value(value):
    """Рекурсивное преобразование значений в JSON-compatible types."""

    if value is None or value is pd.NA or value is pd.NaT:
        return None

    if isinstance(value, dict):
        return {
            str(key): safe_value(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple, set)):
        return [safe_value(item) for item in value]

    if isinstance(value, np.ndarray):
        return safe_value(value.tolist())

    if isinstance(value, (np.integer,)):
        return int(value)

    if isinstance(value, (np.floating, float)):
        number = float(value)
        return number if np.isfinite(number) else None

    if isinstance(value, (np.bool_, bool)):
        return bool(value)

    if isinstance(value, (pd.Timestamp, datetime)):
        return value.isoformat()

    if isinstance(value, Path):
        return str(value)

    if is_missing_scalar(value):
        return None

    if isinstance(value, (str, int)):
        return value

    return str(value)


def find_column(dataframe, candidates):
    """Поиск столбца без учёта регистра."""

    if not isinstance(dataframe, pd.DataFrame) or dataframe.empty:
        return None

    lookup = {
        str(column).strip().lower(): column
        for column in dataframe.columns
    }

    for candidate in candidates:
        key = str(candidate).strip().lower()
        if key in lookup:
            return lookup[key]

    return None


def get_dataframe(candidate_names):
    """
    Возвращает копию первого непустого DataFrame.
    Если все найденные таблицы пустые, возвращает первую пустую.
    """

    found = []

    for name in candidate_names:
        value = globals().get(name)

        if isinstance(value, pd.DataFrame):
            found.append((name, value))

            if not value.empty:
                return value.copy(deep=True), name

    if found:
        name, dataframe = found[0]
        return dataframe.copy(deep=True), name

    return pd.DataFrame(), None


def dataframe_from_csv(filename):
    path = PDB_RESULTS_DIR_VALUE / filename

    if not path.is_file():
        return pd.DataFrame()

    try:
        return pd.read_csv(path)
    except Exception as error:
        print(f"WARNING: Could not read {filename}: {error}")
        return pd.DataFrame()


def dataframe_from_json(filename):
    path = PDB_RESULTS_DIR_VALUE / filename

    if not path.is_file():
        return None

    try:
        with open(path, "r", encoding="utf-8") as file:
            return json.load(file)
    except Exception as error:
        print(f"WARNING: Could not read {filename}: {error}")
        return None


def numeric_series(dataframe, candidates):
    column = find_column(dataframe, candidates)

    if column is None:
        return pd.Series(dtype=float)

    values = pd.to_numeric(dataframe[column], errors="coerce")
    return values[np.isfinite(values)]


def mean_value(dataframe, candidates):
    values = numeric_series(dataframe, candidates)
    return float(values.mean()) if not values.empty else None


def min_value(dataframe, candidates):
    values = numeric_series(dataframe, candidates)
    return float(values.min()) if not values.empty else None


def max_value(dataframe, candidates):
    values = numeric_series(dataframe, candidates)
    return float(values.max()) if not values.empty else None


def unique_count(dataframe, candidates):
    column = find_column(dataframe, candidates)

    if column is None:
        return None

    values = dataframe[column].dropna()
    return int(values.nunique())


def count_by_column(dataframe, candidates, prefix):
    """Подсчёт записей по значениям категориального столбца."""

    column = find_column(dataframe, candidates)

    if column is None:
        return {}

    values = (
        dataframe[column]
        .dropna()
        .astype(str)
        .str.strip()
        .str.upper()
    )

    values = values[
        ~values.isin(["", "NAN", "NONE", "UNKNOWN", "N/A"])
    ]

    if values.empty:
        return {}

    return {
        f"{prefix}_{key}": int(value)
        for key, value in values.value_counts().items()
    }


def normalize_chain_value(value):
    """
    Сохраняет настоящий пустой chain ID как <blank>.
    Не превращает NaN в строку 'nan'.
    """

    if is_missing_scalar(value):
        return "<blank>"

    value = str(value).strip()

    if value.lower() in {"nan", "none", "<na>"}:
        return "<blank>"

    return value if value else "<blank>"


def flatten_numeric_dictionary(data, prefix=""):
    """Извлекает конечные числовые значения из вложенного словаря."""

    result = {}

    if not isinstance(data, dict):
        return result

    for key, value in data.items():
        name = f"{prefix}_{key}" if prefix else str(key)

        if isinstance(value, dict):
            result.update(
                flatten_numeric_dictionary(value, name)
            )

        elif isinstance(value, (int, float, np.integer, np.floating)):
            number = safe_value(value)

            if number is not None:
                result[name] = number

    return result


def validate_pdb_dataframe(dataframe, table_name):
    """Проверяет PDB ID, если он присутствует в таблице."""

    if dataframe.empty:
        return

    pdb_column = find_column(dataframe, ["pdb_id", "PDB_ID"])

    if pdb_column is None:
        return

    values = (
        dataframe[pdb_column]
        .dropna()
        .astype(str)
        .str.strip()
        .str.upper()
        .unique()
    )

    mismatches = [
        value for value in values
        if value != PDB_ID_VALUE
    ]

    if mismatches:
        raise ValueError(
            f"{table_name} contains records from another PDB: "
            f"{mismatches}"
        )


def first_available_dataframe(*dataframes):
    """Первая непустая таблица; иначе первая доступная пустая."""

    available = [
        dataframe.copy(deep=True)
        for dataframe in dataframes
        if isinstance(dataframe, pd.DataFrame)
    ]

    for dataframe in available:
        if not dataframe.empty:
            return dataframe

    return available[0] if available else pd.DataFrame()


def add_counts(target, counts):
    target.update(counts)


# ------------------------------------------------------------
# 3. DATAFRAME DISCOVERY
# ------------------------------------------------------------

DATAFRAME_CANDIDATES = {
    "atoms": ["df_atoms"],
    "residues": ["df_residues"],
    "chains": ["df_chains", "df_chain_statistics"],

    "interactions": [
        "df_ligand_interactions",
        "df_interactions"
    ],

    "binding_site": [
        "df_binding",
        "df_binding_site",
        "df_binding_site_residues",
        "df_binding_site_analysis"
    ],

    "binding_atom_contacts": ["df_binding_atom_contacts"],
    "binding_by_chain": ["df_binding_by_chain"],

    # Общая таблица металлов: предпочтительно все металлы.
    "all_metal_coordination": [
        "df_all_metal_coordination",
        "df_metal_coordination"
    ],

    # Специализированная таблица CELL 19 для гемового железа.
    "heme_iron_coordination": ["df_metal_coordination"],

    "disulfide_bridges": ["df_disulfide_bridges"],

    "components": [
        "df_components",
        "df_nonprotein_components",
        "df_hetero"
    ],

    "geometry": [
        "df_geometry",
        "df_geometric_metrics",
        "df_structural_metrics"
    ],

    "hydrogen_bonds": ["df_hydrogen_bonds"],
    "hydrophobic_contacts": ["df_hydrophobic_contacts"],

    "altloc_atoms": ["df_altloc_atoms"],
    "altloc_occupancy": ["df_altloc_occupancy"],
    "low_occupancy_atoms": ["df_low_occupancy_atoms"],
    "missing_occupancy_atoms": ["df_missing_occupancy_atoms"],
    "invalid_occupancy_atoms": ["df_invalid_occupancy_atoms"],
    "atom_quality": ["df_atom_quality"],

    "unobserved_residues": ["df_unobserved_residues"],
    "unobserved_atoms": ["df_unobserved_atoms"]
}


DATAFRAMES = {}
DATAFRAME_SOURCES = {}

for category, candidates in DATAFRAME_CANDIDATES.items():
    dataframe, source_name = get_dataframe(candidates)

    DATAFRAMES[category] = dataframe
    DATAFRAME_SOURCES[category] = source_name


print("\nDATAFRAME DISCOVERY")
print("-" * 75)

for category, dataframe in DATAFRAMES.items():
    source_name = DATAFRAME_SOURCES[category]

    if source_name is None:
        print(f"{category:30s}: not available")
    else:
        print(
            f"{category:30s}: {len(dataframe):8d} rows | "
            f"{source_name}"
        )


# ------------------------------------------------------------
# 4. REQUIRED INPUT VALIDATION
# ------------------------------------------------------------

for category in ["atoms", "residues"]:
    if DATAFRAME_SOURCES[category] is None:
        raise RuntimeError(
            f"Required source table '{category}' is missing."
        )

df_atoms_feature = DATAFRAMES["atoms"].copy(deep=True)
df_residues_feature = DATAFRAMES["residues"].copy(deep=True)

if df_atoms_feature.empty:
    raise ValueError("df_atoms is empty.")

if df_residues_feature.empty:
    raise ValueError("df_residues is empty.")

for category, dataframe in DATAFRAMES.items():
    validate_pdb_dataframe(dataframe, category)


# ------------------------------------------------------------
# 5. SOURCE DATAFRAMES
# ------------------------------------------------------------

df_interactions = DATAFRAMES["interactions"].copy(deep=True)

df_binding = DATAFRAMES["binding_site"].copy(deep=True)
df_binding_atom_contacts = (
    DATAFRAMES["binding_atom_contacts"].copy(deep=True)
)
df_binding_by_chain = (
    DATAFRAMES["binding_by_chain"].copy(deep=True)
)

df_all_metal_coordination = (
    DATAFRAMES["all_metal_coordination"].copy(deep=True)
)
df_heme_iron_coordination = (
    DATAFRAMES["heme_iron_coordination"].copy(deep=True)
)

df_disulfide_bridges = (
    DATAFRAMES["disulfide_bridges"].copy(deep=True)
)
df_components = DATAFRAMES["components"].copy(deep=True)
df_geometry = DATAFRAMES["geometry"].copy(deep=True)

df_hydrogen_bonds = DATAFRAMES["hydrogen_bonds"].copy(deep=True)
df_hydrophobic_contacts = (
    DATAFRAMES["hydrophobic_contacts"].copy(deep=True)
)


# ------------------------------------------------------------
# 6. ATOM FEATURES
# ------------------------------------------------------------

df_atom_features = df_atoms_feature.copy(deep=True)

element_column = find_column(
    df_atom_features,
    ["element", "atom_element"]
)

if element_column is not None:
    df_atom_features["element_normalized"] = (
        df_atom_features[element_column]
        .fillna("")
        .astype(str)
        .str.strip()
        .str.upper()
    )

atom_name_column = find_column(
    df_atom_features,
    ["atom_name", "name", "atom"]
)

if atom_name_column is not None:
    df_atom_features["atom_name_normalized"] = (
        df_atom_features[atom_name_column]
        .fillna("")
        .astype(str)
        .str.strip()
    )

atom_chain_column = find_column(
    df_atom_features,
    ["chain_id", "chain"]
)

if atom_chain_column is not None:
    df_atom_features["chain_id_normalized"] = (
        df_atom_features[atom_chain_column]
        .apply(normalize_chain_value)
    )


# ------------------------------------------------------------
# 7. RESIDUE FEATURES
# ------------------------------------------------------------

df_residue_features = df_residues_feature.copy(deep=True)

residue_name_column = find_column(
    df_residue_features,
    ["residue_name", "resname", "residue"]
)

if residue_name_column is not None:
    df_residue_features["residue_name_normalized"] = (
        df_residue_features[residue_name_column]
        .fillna("")
        .astype(str)
        .str.strip()
        .str.upper()
    )

residue_chain_column = find_column(
    df_residue_features,
    ["chain_id", "chain"]
)

if residue_chain_column is not None:
    df_residue_features["chain_id_normalized"] = (
        df_residue_features[residue_chain_column]
        .apply(normalize_chain_value)
    )


# ------------------------------------------------------------
# 8. GEOMETRY FEATURES — CELL 9
# ------------------------------------------------------------

geometry_features = {}

geometry_summary_source = globals().get("geometry_summary")

if isinstance(geometry_summary_source, dict):
    geometry_features.update(
        flatten_numeric_dictionary(
            geometry_summary_source,
            "geometry"
        )
    )

dimensions_source = globals().get("dimensions")

if isinstance(dimensions_source, dict):
    geometry_features.update(
        flatten_numeric_dictionary(
            dimensions_source,
            "dimensions"
        )
    )

GEOMETRY_VARIABLE_CANDIDATES = [
    "radius_of_gyration",
    "mean_distance_from_center",
    "median_distance_from_center",
    "max_distance_from_center",
    "bounding_box_volume",
    "aspect_ratio",
    "protein_atom_count",
    "total_protein_atoms"
]

for variable_name in GEOMETRY_VARIABLE_CANDIDATES:
    value = globals().get(variable_name)

    if isinstance(value, (int, float, np.integer, np.floating)):
        normalized_value = safe_value(value)

        if normalized_value is not None:
            geometry_features[variable_name] = normalized_value


# Значения из CSV могут быть записаны как строки.
if not df_geometry.empty and len(df_geometry) == 1:
    for column in df_geometry.columns:
        value = pd.to_numeric(
            pd.Series([df_geometry.iloc[0][column]]),
            errors="coerce"
        ).iloc[0]

        if pd.notna(value) and np.isfinite(float(value)):
            geometry_features.setdefault(
                f"geometry_{column}",
                float(value)
            )


# JSON используется только для дополнения отсутствующих признаков.
geometry_json = dataframe_from_json(
    f"{PDB_ID_VALUE}_geometry.json"
)

if isinstance(geometry_json, dict):
    json_geometry_features = flatten_numeric_dictionary(
        geometry_json,
        "geometry_json"
    )

    for key, value in json_geometry_features.items():
        geometry_features.setdefault(key, value)


# ------------------------------------------------------------
# 9. QUALITY METRICS — CELL 10 / CELL 14
# ------------------------------------------------------------

metadata_json = dataframe_from_json(
    f"{PDB_ID_VALUE}_metadata.json"
)

structure_quality_json = dataframe_from_json(
    f"{PDB_ID_VALUE}_structure_quality.json"
)


# 9.1. Unobserved residues / atoms

df_unobserved_residues = first_available_dataframe(
    DATAFRAMES["unobserved_residues"],
    dataframe_from_csv(
        f"{PDB_ID_VALUE}_unobserved_residues.csv"
    )
)

df_unobserved_atoms = first_available_dataframe(
    DATAFRAMES["unobserved_atoms"],
    dataframe_from_csv(
        f"{PDB_ID_VALUE}_unobserved_atoms.csv"
    )
)

unobserved_residue_available = (
    DATAFRAME_SOURCES["unobserved_residues"] is not None
    or not df_unobserved_residues.empty
)

unobserved_atom_available = (
    DATAFRAME_SOURCES["unobserved_atoms"] is not None
    or not df_unobserved_atoms.empty
)

quality_features = {
    "unobserved_residue_record_count": (
        int(len(df_unobserved_residues))
        if unobserved_residue_available else None
    ),

    "unobserved_atom_record_count": (
        int(len(df_unobserved_atoms))
        if unobserved_atom_available else None
    ),

    "quality_tables_scope": (
        "CELL 14 tables may include all models and alternate conformers; "
        "they are not restricted to the first-model atom table."
    )
}


# 9.2. Altloc and occupancy diagnostics

QUALITY_FILENAME_MAP = {
    "altloc_atoms": f"{PDB_ID_VALUE}_altloc_atoms.csv",
    "altloc_occupancy": f"{PDB_ID_VALUE}_altloc_occupancy.csv",
    "low_occupancy_atoms": f"{PDB_ID_VALUE}_low_occupancy_atoms.csv",
    "missing_occupancy_atoms": f"{PDB_ID_VALUE}_missing_occupancy_atoms.csv",
    "invalid_occupancy_atoms": f"{PDB_ID_VALUE}_invalid_occupancy_atoms.csv",
    "atom_quality": f"{PDB_ID_VALUE}_atom_quality.csv"
}

df_quality_tables = {}

for key in QUALITY_FILENAME_MAP:
    dataframe = DATAFRAMES[key].copy(deep=True)

    if dataframe.empty:
        dataframe_from_file = dataframe_from_csv(
            QUALITY_FILENAME_MAP[key]
        )

        if not dataframe_from_file.empty:
            dataframe = dataframe_from_file

    df_quality_tables[key] = dataframe

    source_available = (
        DATAFRAME_SOURCES.get(key) is not None
        or not dataframe.empty
    )

    quality_features[f"{key}_record_count"] = (
        int(len(dataframe))
        if source_available else None
    )


# Сводные показатели CELL 14, если JSON содержит summary.
if isinstance(structure_quality_json, dict):
    quality_payload = structure_quality_json.get(
        "summary",
        structure_quality_json
    )

    quality_features.update(
        flatten_numeric_dictionary(
            quality_payload,
            "structure_quality"
        )
    )


# ------------------------------------------------------------
# 10. BINDING SITE FEATURES — CELL 18
# ------------------------------------------------------------

df_binding_features = df_binding.copy(deep=True)

df_binding_atom_contact_features = (
    df_binding_atom_contacts.copy(deep=True)
)

df_binding_chain_features = (
    df_binding_by_chain.copy(deep=True)
)


def get_binding_residue_count(dataframe):
    """Подсчитывает уникальные остатки, а не строки таблицы."""

    if dataframe.empty:
        return 0

    residue_key = find_column(dataframe, ["residue_key"])

    if residue_key is not None:
        return int(dataframe[residue_key].dropna().nunique())

    chain_col = find_column(
        dataframe,
        ["protein_chain", "chain_id", "protein_chain_id"]
    )

    residue_col = find_column(
        dataframe,
        ["residue_id", "residue_number", "residue_seq"]
    )

    insertion_col = find_column(
        dataframe,
        ["insertion_code", "icode"]
    )

    if chain_col is None or residue_col is None:
        return None

    key_columns = [chain_col, residue_col]

    if insertion_col is not None:
        key_columns.append(insertion_col)

    return int(
        dataframe[key_columns]
        .drop_duplicates()
        .shape[0]
    )


def get_unique_contact_atom_count(dataframe, side):
    """
    Число уникальных атомов, участвующих в контактных парах.
    Нужны идентификаторы цепи/остатка/атома для белка
    или компонента/атома для небелковой стороны.
    """

    if dataframe.empty:
        return 0

    if side == "protein":
        chain_col = find_column(
            dataframe,
            ["protein_chain", "chain_id", "protein_chain_id"]
        )

        residue_col = find_column(
            dataframe,
            ["protein_residue_key", "residue_key"]
        )

        atom_col = find_column(
            dataframe,
            ["protein_atom", "protein_atom_name"]
        )

        if chain_col is None or residue_col is None or atom_col is None:
            return None

        key_columns = [chain_col, residue_col, atom_col]

    else:
        component_col = find_column(
            dataframe,
            ["component_residue_key", "component_id"]
        )

        atom_col = find_column(
            dataframe,
            ["component_atom", "component_atom_name"]
        )

        if component_col is None or atom_col is None:
            return None

        key_columns = [component_col, atom_col]

    altloc_col = find_column(
        dataframe,
        [
            "protein_altloc" if side == "protein"
            else "component_altloc"
        ]
    )

    if altloc_col is not None:
        key_columns.append(altloc_col)

    return int(
        dataframe[key_columns]
        .dropna(subset=[atom_col])
        .drop_duplicates()
        .shape[0]
    )


binding_available = DATAFRAME_SOURCES["binding_site"] is not None

binding_atom_contacts_available = (
    DATAFRAME_SOURCES["binding_atom_contacts"] is not None
)

binding_chain_available = (
    DATAFRAME_SOURCES["binding_by_chain"] is not None
)

binding_distance_candidates = [
    "min_distance_A",
    "minimum_distance_A",
    "min_distance"
]

binding_atom_distance_candidates = [
    "distance_A",
    "distance",
    "contact_distance_A"
]

binding_features = {
    "binding_site_available": binding_available,

    "binding_atom_contacts_available": (
        binding_atom_contacts_available
    ),

    "binding_chain_statistics_available": (
        binding_chain_available
    ),

    "binding_site_residue_count": (
        get_binding_residue_count(df_binding_features)
        if binding_available else None
    ),

    "binding_site_atom_pair_record_count": (
        int(len(df_binding_atom_contact_features))
        if binding_atom_contacts_available else None
    ),

    "binding_site_min_residue_distance_A": (
        min_value(df_binding_features, binding_distance_candidates)
    ),

    "binding_site_mean_residue_min_distance_A": (
        mean_value(df_binding_features, binding_distance_candidates)
    ),

    "binding_site_max_residue_min_distance_A": (
        max_value(df_binding_features, binding_distance_candidates)
    ),

    "binding_site_mean_atom_pair_distance_A": (
        mean_value(
            df_binding_atom_contact_features,
            binding_atom_distance_candidates
        )
    ),

    "binding_site_unique_protein_contact_atom_count": (
        get_unique_contact_atom_count(
            df_binding_atom_contact_features,
            "protein"
        )
        if binding_atom_contacts_available else None
    ),

    "binding_site_unique_component_contact_atom_count": (
        get_unique_contact_atom_count(
            df_binding_atom_contact_features,
            "component"
        )
        if binding_atom_contacts_available else None
    ),

    "binding_site_chain_count": (
        unique_count(
            df_binding_features,
            ["protein_chain", "chain_id"]
        )
        if binding_available else None
    )
}


# ------------------------------------------------------------
# 11. METAL FEATURES — CELL 19
# ------------------------------------------------------------

df_metal_features = df_all_metal_coordination.copy(deep=True)
df_heme_iron_features = df_heme_iron_coordination.copy(deep=True)

metal_element_column = find_column(
    df_metal_features,
    ["metal_element", "metal", "metal_name", "metal_type"]
)

metal_distance_candidates = [
    "distance_A",
    "distance",
    "metal_distance_A"
]

metal_distance_column = find_column(
    df_metal_features,
    metal_distance_candidates
)

if not df_metal_features.empty:
    if metal_element_column is not None:
        df_metal_features["metal_normalized"] = (
            df_metal_features[metal_element_column]
            .fillna("")
            .astype(str)
            .str.strip()
            .str.upper()
        )

    if metal_distance_column is not None:
        df_metal_features["metal_distance_normalized_A"] = (
            pd.to_numeric(
                df_metal_features[metal_distance_column],
                errors="coerce"
            )
        )


metal_features = {
    "metal_coordination_table_available": (
        DATAFRAME_SOURCES["all_metal_coordination"] is not None
    ),

    "metal_coordination_record_count": (
        int(len(df_metal_features))
        if DATAFRAME_SOURCES["all_metal_coordination"] is not None
        else None
    ),

    "metal_coordination_mean_distance_A": (
        mean_value(df_metal_features, metal_distance_candidates)
    ),

    "heme_iron_coordination_table_available": (
        DATAFRAME_SOURCES["heme_iron_coordination"] is not None
    ),

    "heme_iron_coordination_record_count": (
        int(len(df_heme_iron_features))
        if DATAFRAME_SOURCES["heme_iron_coordination"] is not None
        else None
    )
}


if metal_element_column is not None and not df_metal_features.empty:
    for metal_name, metal_group in df_metal_features.groupby(
        metal_element_column,
        dropna=True
    ):
        metal_name = str(metal_name).strip().upper()

        if not metal_name:
            continue

        metal_features[
            f"metal_contact_record_count_{metal_name}"
        ] = int(len(metal_group))

        metal_features[
            f"metal_mean_distance_{metal_name}_A"
        ] = mean_value(
            metal_group,
            metal_distance_candidates
        )


# ------------------------------------------------------------
# 12. COMPONENT FEATURES
# ------------------------------------------------------------

component_code_column = find_column(
    df_components,
    [
        "component_code",
        "comp_id",
        "resname",
        "residue_name",
        "component"
    ]
)

# Для экземпляров используем идентификатор конкретного остатка,
# если он доступен. Иначе используем строки таблицы как записи,
# не называя их гарантированно уникальными экземплярами.
component_instance_key_columns = []

for candidates in [
    ["component_instance_id", "instance_id", "component_residue_key"],
    ["residue_key"],
]:
    column = find_column(df_components, candidates)

    if column is not None:
        component_instance_key_columns.append(column)
        break

if not component_instance_key_columns:
    chain_col = find_column(df_components, ["chain_id", "chain"])
    residue_id_col = find_column(
        df_components,
        ["residue_id", "residue_number", "resseq"]
    )
    insertion_col = find_column(
        df_components,
        ["insertion_code", "icode"]
    )

    if chain_col is not None and residue_id_col is not None:
        component_instance_key_columns = [
            chain_col,
            residue_id_col
        ]

        if insertion_col is not None:
            component_instance_key_columns.append(insertion_col)


if DATAFRAME_SOURCES["components"] is None:
    component_instance_count = None
    component_instance_count_method = "unavailable"

elif component_instance_key_columns:
    component_instance_count = int(
        df_components[component_instance_key_columns]
        .drop_duplicates()
        .shape[0]
    )
    component_instance_count_method = (
        "unique component instance identifiers"
    )

else:
    component_instance_count = int(len(df_components))
    component_instance_count_method = (
        "number of records; unique instances could not be verified"
    )


if component_code_column is not None:
    component_type_count = int(
        df_components[component_code_column]
        .dropna()
        .astype(str)
        .str.strip()
        .str.upper()
        .replace("", np.nan)
        .nunique()
    )
else:
    component_type_count = None


component_features = {
    "component_table_available": (
        DATAFRAME_SOURCES["components"] is not None
    ),

    "component_instance_count": component_instance_count,

    "component_type_count": component_type_count,

    "component_instance_count_method": (
        component_instance_count_method
    )
}


component_type_counts = count_by_column(
    df_components,
    [
        "component_code",
        "comp_id",
        "resname",
        "residue_name",
        "component"
    ],
    "component_record_count"
)

add_counts(component_features, component_type_counts)


# ------------------------------------------------------------
# 13. DISULFIDE BRIDGE FEATURES
# ------------------------------------------------------------

disulfide_features = {
    "disulfide_analysis_available": (
        DATAFRAME_SOURCES["disulfide_bridges"] is not None
    ),

    "potential_disulfide_pair_record_count": (
        int(len(df_disulfide_bridges))
        if DATAFRAME_SOURCES["disulfide_bridges"] is not None
        else None
    )
}


# ------------------------------------------------------------
# 14. ATOM AND RESIDUE COMPOSITION
# ------------------------------------------------------------

atom_element_counts = count_by_column(
    df_atom_features,
    ["element", "element_normalized", "atom_element"],
    "protein_atom_count_element"
)

residue_type_counts = count_by_column(
    df_residue_features,
    [
        "residue_name",
        "residue_name_normalized",
        "resname"
    ],
    "protein_residue_count"
)


# ------------------------------------------------------------
# 15. ADDITIONAL INTERACTION TABLES
# ------------------------------------------------------------

df_hydrogen_bond_features = df_hydrogen_bonds.copy(deep=True)
df_hydrophobic_contact_features = (
    df_hydrophobic_contacts.copy(deep=True)
)

additional_interaction_features = {
    "hydrogen_bond_table_available": (
        DATAFRAME_SOURCES["hydrogen_bonds"] is not None
    ),

    "hydrogen_bond_record_count": (
        int(len(df_hydrogen_bond_features))
        if DATAFRAME_SOURCES["hydrogen_bonds"] is not None
        else None
    ),

    "hydrophobic_contact_table_available": (
        DATAFRAME_SOURCES["hydrophobic_contacts"] is not None
    ),

    "hydrophobic_contact_record_count": (
        int(len(df_hydrophobic_contact_features))
        if DATAFRAME_SOURCES["hydrophobic_contacts"] is not None
        else None
    )
}


# ------------------------------------------------------------
# 16. COMBINED INTERACTION TABLE
# ------------------------------------------------------------

interaction_frames = []


def append_interaction_table(
    dataframe,
    source_name,
    interaction_type
):
    if dataframe.empty:
        return

    frame = dataframe.copy(deep=True)

    # Сохраняем исходные значения, если такие столбцы уже были.
    if "source_table" in frame.columns:
        frame["original_source_table"] = frame["source_table"]

    if "interaction_type" in frame.columns:
        frame["original_interaction_type"] = (
            frame["interaction_type"]
        )

    frame["source_table"] = source_name
    frame["interaction_type"] = interaction_type

    interaction_frames.append(frame)


append_interaction_table(
    df_interactions,
    DATAFRAME_SOURCES["interactions"] or "unknown_source",
    "selected_component_interaction_record"
)

append_interaction_table(
    df_binding_atom_contact_features,
    DATAFRAME_SOURCES["binding_atom_contacts"] or "unknown_source",
    "binding_site_atom_pair_contact_record"
)

append_interaction_table(
    df_metal_features,
    DATAFRAME_SOURCES["all_metal_coordination"] or "unknown_source",
    "potential_metal_coordination_record"
)

append_interaction_table(
    df_disulfide_bridges,
    DATAFRAME_SOURCES["disulfide_bridges"] or "unknown_source",
    "potential_disulfide_pair_record"
)

append_interaction_table(
    df_hydrogen_bond_features,
    DATAFRAME_SOURCES["hydrogen_bonds"] or "unknown_source",
    "hydrogen_bond_table_record"
)

append_interaction_table(
    df_hydrophobic_contact_features,
    DATAFRAME_SOURCES["hydrophobic_contacts"] or "unknown_source",
    "hydrophobic_contact_table_record"
)

if interaction_frames:
    df_interaction_features = pd.concat(
        interaction_frames,
        ignore_index=True,
        sort=False
    )
else:
    df_interaction_features = pd.DataFrame(
        columns=["source_table", "interaction_type"]
    )


# ------------------------------------------------------------
# 17. GLOBAL FEATURES
# ------------------------------------------------------------

global_features = {
    "pdb_id": PDB_ID_VALUE,
    "analysis_scope": ANALYSIS_SCOPE,

    "protein_atom_count": int(len(df_atom_features)),
    "protein_residue_count": int(len(df_residue_features)),

    "protein_chain_count": None,

    "component_instance_count": (
        component_features["component_instance_count"]
    ),

    "component_type_count": (
        component_features["component_type_count"]
    ),

    "selected_component_interaction_record_count": (
        int(len(df_interactions))
        if DATAFRAME_SOURCES["interactions"] is not None
        else None
    ),

    "selected_component_mean_interaction_distance_A": (
        mean_value(
            df_interactions,
            [
                "distance_A",
                "distance",
                "min_distance_A"
            ]
        )
    ),

    "geometry_source_table_row_count": (
        int(len(df_geometry))
        if DATAFRAME_SOURCES["geometry"] is not None
        else None
    )
}


# Число цепей по остаточной таблице.
chain_column = find_column(
    df_residue_features,
    ["chain_id", "chain"]
)

if chain_column is not None:
    global_features["protein_chain_count"] = int(
        df_residue_features[chain_column]
        .apply(normalize_chain_value)
        .nunique()
    )


# Число цепей, представленных в chain_sequences, — отдельный показатель.
chain_sequences_source = globals().get("chain_sequences")

if isinstance(chain_sequences_source, dict):
    global_features["sequence_chain_count"] = int(
        len(chain_sequences_source)
    )


global_features.update(binding_features)
global_features.update(metal_features)
global_features.update(component_features)
global_features.update(disulfide_features)
global_features.update(quality_features)
global_features.update(additional_interaction_features)

add_counts(global_features, atom_element_counts)
add_counts(global_features, residue_type_counts)

global_features.update(geometry_features)


# ------------------------------------------------------------
# 18. FEATURE TABLES
# ------------------------------------------------------------

df_global_features = pd.DataFrame([global_features])

feature_table_map = {
    "global_features": df_global_features,

    "atom_features": df_atom_features,
    "residue_features": df_residue_features,

    "interaction_features": df_interaction_features,

    "metal_features": df_metal_features,
    "heme_iron_coordination_features": df_heme_iron_features,

    "binding_site_features": df_binding_features,
    "binding_atom_contact_features": df_binding_atom_contact_features,
    "binding_chain_features": df_binding_chain_features,

    "component_features": df_components,
    "disulfide_features": df_disulfide_bridges,

    "geometry_features": df_geometry,

    "altloc_features": df_quality_tables["altloc_atoms"],
    "altloc_occupancy_features": df_quality_tables["altloc_occupancy"],
    "low_occupancy_features": df_quality_tables["low_occupancy_atoms"],
    "missing_occupancy_features": df_quality_tables["missing_occupancy_atoms"],
    "invalid_occupancy_features": df_quality_tables["invalid_occupancy_atoms"],
    "atom_quality_features": df_quality_tables["atom_quality"],

    "unobserved_residue_features": df_unobserved_residues,
    "unobserved_atom_features": df_unobserved_atoms
}


# ------------------------------------------------------------
# 19. TABLE STATISTICS
# ------------------------------------------------------------

table_statistics = {}

for table_name, dataframe in feature_table_map.items():
    source_available = (
        table_name in [
            "global_features",
            "atom_features",
            "residue_features"
        ]
        or len(dataframe) > 0
    )

    table_statistics[table_name] = {
        "rows": int(len(dataframe)),
        "columns": int(len(dataframe.columns)),
        "column_names": [
            str(column)
            for column in dataframe.columns
        ],
        "available": bool(source_available)
    }


# ------------------------------------------------------------
# 20. FEATURE DEFINITIONS
# ------------------------------------------------------------

feature_definitions = {
    "protein_atom_count": (
        "Number of protein atom records in the first-model atom table."
    ),

    "protein_residue_count": (
        "Number of protein residue records in the first-model residue table."
    ),

    "protein_chain_count": (
        "Number of distinct chain identifiers represented in the residue table."
    ),

    "sequence_chain_count": (
        "Number of entries in chain_sequences; may differ from the "
        "number of chains represented in the residue table."
    ),

    "component_instance_count": (
        "Number of component instances when instance identifiers are available; "
        "otherwise the number of records is reported with a limitation note."
    ),

    "component_type_count": (
        "Number of distinct component codes."
    ),

    "selected_component_interaction_record_count": (
        "Number of records in the selected-component interaction table."
    ),

    "binding_site_residue_count": (
        "Number of unique protein residues represented in the binding-site table."
    ),

    "binding_site_atom_pair_record_count": (
        "Number of atom-pair contact records; not a count of unique chemical bonds."
    ),

    "metal_coordination_record_count": (
        "Number of records in the general metal-coordination table."
    ),

    "heme_iron_coordination_record_count": (
        "Number of records in the specialized heme-iron coordination table."
    ),

    "potential_disulfide_pair_record_count": (
        "Number of potential SG-SG pair records; not independently confirmed bonds."
    ),

    "unobserved_residue_record_count": (
        "Number of records in the unobserved-residue source table."
    ),

    "unobserved_atom_record_count": (
        "Number of records in the unobserved-atom source table."
    ),

    "geometry_features": (
        "Numerical geometry descriptors obtained from CELL 9."
    ),

    "quality_features": (
        "Quality indicators from CELL 10 and CELL 14. "
        "Their scope may include all models and alternate conformers."
    )
}


# ------------------------------------------------------------
# 21. JSON METADATA
# ------------------------------------------------------------

protein_features = {
    "metadata": {
        "pdb_id": PDB_ID_VALUE,

        "created_at": datetime.now().isoformat(
            timespec="seconds"
        ),

        "feature_extraction_cell": 22,
        "analysis_scope": ANALYSIS_SCOPE,

        "pdb_results_directory": str(PDB_RESULTS_DIR_VALUE),
        "feature_directory": str(FEATURE_DIR),

        "binding_site_cutoff_A": safe_value(
            globals().get("BINDING_CUTOFF")
        ),

        "interaction_cutoff_A": safe_value(
            globals().get("CONTACT_CUTOFF")
        ),

        "metal_coordination_cutoff_A": safe_value(
            globals().get("METAL_COORDINATION_CUTOFF")
        )
    },

    "global_features": safe_value(global_features),

    "table_statistics": safe_value(table_statistics),

    "dataframe_sources": safe_value(DATAFRAME_SOURCES),

    "feature_definitions": safe_value(feature_definitions),

    "geometry_features": safe_value(geometry_features),

    "quality_features": safe_value(quality_features),

    "analysis_notes": {
        "structural_scope": (
            "Protein atom and residue feature tables describe the first model. "
            "Quality-control tables may describe all models and alternate conformers."
        ),

        "components": (
            "Non-protein components are not automatically drug-like ligands. "
            "They may include cofactors, ions, buffers, additives or other molecules."
        ),

        "binding_site": (
            "Residue-level and atom-pair-level binding-site statistics are "
            "kept separate. Contact records are geometric observations."
        ),

        "geometry": (
            "Geometry descriptors originate from CELL 9. "
            "Bounding-box volume, if present, is not a physical molecular volume."
        ),

        "metal_coordination": (
            "Metal-contact records are geometric candidates and do not "
            "independently prove chemical coordination."
        ),

        "disulfide_bridges": (
            "Disulfide records are potential geometric pairs and require "
            "chemical/contextual confirmation."
        ),

        "interaction_features": (
            "The combined interaction table contains records with different "
            "granularities. source_table and interaction_type preserve provenance."
        ),

        "quality": (
            "Unobserved records, alternate conformers and occupancy diagnostics "
            "are retained as separate indicators. These categories may overlap."
        ),

        "missing_data": (
            "Unavailable source tables are represented as unavailable/null, "
            "not as artificial zero-valued observations."
        )
    }
}


# ------------------------------------------------------------
# 22. OUTPUT FILES
# ------------------------------------------------------------

OUTPUT_FILES = {
    table_name: (
        FEATURE_DIR / f"{PDB_ID_VALUE}_{table_name}.csv"
    )
    for table_name in feature_table_map
}

JSON_FILE = (
    FEATURE_DIR / f"{PDB_ID_VALUE}_protein_features.json"
)


# ------------------------------------------------------------
# 23. SAVE CSV TABLES
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("SAVING FEATURE TABLES")
print("=" * 75)

for table_name, dataframe in feature_table_map.items():
    output_path = OUTPUT_FILES[table_name]

    dataframe.to_csv(
        output_path,
        index=False,
        encoding="utf-8-sig"
    )

    if not output_path.is_file():
        raise RuntimeError(
            f"Output file was not saved: {output_path}"
        )

    print(
        f"{output_path.name:58s} "
        f"{len(dataframe):8d} rows | "
        f"{len(dataframe.columns):4d} columns"
    )


# ------------------------------------------------------------
# 24. SAVE JSON
# ------------------------------------------------------------

with open(JSON_FILE, "w", encoding="utf-8") as file:
    json.dump(
        safe_value(protein_features),
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )

if not JSON_FILE.is_file():
    raise RuntimeError(
        f"JSON file was not saved: {JSON_FILE}"
    )


# ------------------------------------------------------------
# 25. OUTPUT VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("OUTPUT VALIDATION")
print("=" * 75)

validation_results = {}

for table_name, dataframe in feature_table_map.items():
    output_path = OUTPUT_FILES[table_name]

    valid = (
        output_path.is_file()
        and output_path.stat().st_size >= 0
    )

    validation_results[f"{table_name}_file_exists"] = bool(valid)

    print(
        f"{table_name:40s}: "
        f"{'OK' if valid else 'ERROR'}"
    )

    if not valid:
        raise RuntimeError(
            f"Output validation failed: {output_path}"
        )


# Проверка JSON: файл должен читаться после сохранения.
with open(JSON_FILE, "r", encoding="utf-8") as file:
    json_check = json.load(file)

if json_check.get("metadata", {}).get("pdb_id") != PDB_ID_VALUE:
    raise ValueError("PDB ID mismatch in saved JSON.")

if "global_features" not in json_check:
    raise ValueError("Global features are missing in JSON.")

validation_results["json_readable"] = True
validation_results["json_pdb_id_matches"] = True


# Проверка сохранения числа атомных и остаточных записей.
if len(df_atom_features) != len(df_atoms_feature):
    raise ValueError("Atom feature row count mismatch.")

if len(df_residue_features) != len(df_residues_feature):
    raise ValueError("Residue feature row count mismatch.")

validation_results["atom_row_count_matches"] = True
validation_results["residue_row_count_matches"] = True


# Проверка числа строк объединённой таблицы.
expected_interaction_rows = sum(
    len(dataframe)
    for dataframe in [
        df_interactions,
        df_binding_atom_contact_features,
        df_metal_features,
        df_disulfide_bridges,
        df_hydrogen_bond_features,
        df_hydrophobic_contact_features
    ]
)

if len(df_interaction_features) != expected_interaction_rows:
    raise ValueError(
        "Combined interaction feature row count mismatch."
    )

validation_results["interaction_row_count_matches"] = True


# Проверка, что копии исходных таблиц сохранили количество строк.
if len(df_atoms_feature) != len(DATAFRAMES["atoms"]):
    raise ValueError("Original atom table row count changed.")

if len(df_residue_features) != len(DATAFRAMES["residues"]):
    raise ValueError("Original residue table row count changed.")

validation_results["source_atom_row_count_preserved"] = True
validation_results["source_residue_row_count_preserved"] = True


# Дополнительная проверка конечности числовых расстояний.
for table_name, dataframe, candidates in [
    (
        "binding_atom_contacts",
        df_binding_atom_contact_features,
        binding_atom_distance_candidates
    ),
    (
        "metal_coordination",
        df_metal_features,
        metal_distance_candidates
    )
]:
    distance_values = numeric_series(dataframe, candidates)

    validation_results[
        f"{table_name}_finite_distance_values"
    ] = bool(
        distance_values.empty
        or np.isfinite(distance_values).all()
    )


# Записываем результаты проверки в JSON.
protein_features["validation"] = safe_value(validation_results)

with open(JSON_FILE, "w", encoding="utf-8") as file:
    json.dump(
        safe_value(protein_features),
        file,
        ensure_ascii=False,
        indent=4,
        allow_nan=False
    )


print("\nJSON structure: OK")
print("Atom feature rows: OK")
print("Residue feature rows: OK")
print("Combined interaction rows: OK")
print("Source DataFrame row counts: OK")
print("Validation metadata added to JSON: OK")


# ------------------------------------------------------------
# 26. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PROTEIN FEATURE EXTRACTION SUMMARY")
print("=" * 75)

print("PDB ID:", PDB_ID_VALUE)
print("Analysis scope:", ANALYSIS_SCOPE)

print("\nSTRUCTURAL FEATURES")
print("-" * 75)

for key in [
    "protein_atom_count",
    "protein_residue_count",
    "protein_chain_count",
    "sequence_chain_count",
    "component_instance_count",
    "component_type_count"
]:
    print(f"{key}: {global_features.get(key)}")


print("\nGEOMETRY FEATURES")
print("-" * 75)

print("Geometry feature count:", len(geometry_features))

for key, value in geometry_features.items():
    print(f"{key}: {value}")


print("\nBINDING SITE FEATURES")
print("-" * 75)

for key, value in binding_features.items():
    print(f"{key}: {value}")


print("\nMETAL FEATURES")
print("-" * 75)

for key, value in metal_features.items():
    print(f"{key}: {value}")


print("\nINTERACTION RECORDS")
print("-" * 75)

print("Selected-component records:", len(df_interactions))
print("Binding atom-pair records:", len(df_binding_atom_contact_features))
print("Metal coordination records:", len(df_metal_features))
print("Heme iron coordination records:", len(df_heme_iron_features))
print("Potential disulfide pair records:", len(df_disulfide_bridges))
print("Combined interaction records:", len(df_interaction_features))


print("\nSTRUCTURE QUALITY")
print("-" * 75)

for key, value in quality_features.items():
    print(f"{key}: {value}")


print("\nOUTPUT TABLES")
print("-" * 75)

for table_name, dataframe in feature_table_map.items():
    print(
        f"{table_name:40s}: "
        f"{len(dataframe):8d} rows × "
        f"{len(dataframe.columns):3d} columns"
    )


print("\nOUTPUT DIRECTORY")
print("-" * 75)
print(FEATURE_DIR)

print("\nJSON FILE")
print("-" * 75)
print(JSON_FILE)

print("\nVALIDATION")
print("-" * 75)

for key, value in validation_results.items():
    print(f"{key}: {value}")


print("\nSTATUS: COMPLETED")
print("CELL 22 COMPLETED")
print("=" * 75)
```

```python
# ============================================================
# CELL 23. PROTEIN SPATIAL REPRESENTATION
# ============================================================
#
# НАЗНАЧЕНИЕ:
# 1. Построение атомного пространственного графа.
# 2. Построение остаточного пространственного графа.
# 3. Расчёт геометрических контактов между атомами.
# 4. Агрегация атомных контактов до уровня остатков.
# 5. Расчёт характеристик связности графов.
# 6. Анализ пространственных контактов внутри и между цепями.
#
# ИСТОЧНИК:
# - CELL 7: df_atoms.
#
# ОБЛАСТЬ АНАЛИЗА:
# - первая модель структуры;
# - атомы белковых остатков из CELL 7;
# - выбранные CELL 7 конформации атомов;
# - водород и дейтерий исключаются;
# - расстояния измеряются в Å.
#
# ВАЖНО:
# Пространственное ребро означает геометрическую близость,
# а не подтверждённую химическую связь.
#
# Центр остатка рассчитывается как среднее координат его
# тяжёлых атомов, а не как центр масс.
# ============================================================


import json
import re
import numpy as np
import pandas as pd

from pathlib import Path
from datetime import datetime

from scipy.spatial import cKDTree
from scipy.sparse import coo_matrix
from scipy.sparse.csgraph import connected_components


print("=" * 75)
print("CELL 23. PROTEIN SPATIAL REPRESENTATION")
print("=" * 75)


# ============================================================
# 1. CONFIGURATION
# ============================================================

SPATIAL_CUTOFF_A = 5.0

EXCLUDE_HYDROGEN = True

ANALYSIS_SCOPE = "first_model"

DISTANCE_TOLERANCE = 1e-6


if (
    not np.isfinite(SPATIAL_CUTOFF_A)
    or SPATIAL_CUTOFF_A <= 0
):
    raise ValueError(
        "SPATIAL_CUTOFF_A must be a positive finite number."
    )


# ============================================================
# 2. INPUT VALIDATION
# ============================================================

required_variables = [
    "PDB_ID",
    "PDB_RESULTS_DIR",
    "df_atoms"
]

for variable in required_variables:

    if variable not in globals():

        raise NameError(
            f"Required variable '{variable}' is not defined. "
            "Run the previous cells first."
        )


PDB_ID_VALUE = str(PDB_ID).strip().upper()


if not re.fullmatch(r"[A-Z0-9]{4}", PDB_ID_VALUE):

    raise ValueError(
        f"Invalid PDB_ID: {PDB_ID_VALUE}"
    )


if not isinstance(df_atoms, pd.DataFrame):

    raise TypeError(
        "df_atoms must be a pandas DataFrame."
    )


if df_atoms.empty:

    raise ValueError(
        "df_atoms is empty."
    )


PDB_RESULTS_DIR_VALUE = Path(PDB_RESULTS_DIR)

FEATURE_DIR = (
    PDB_RESULTS_DIR_VALUE
    / "features"
)

FEATURE_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("PDB ID:", PDB_ID_VALUE)
print("Analysis scope:", ANALYSIS_SCOPE)
print("Spatial cutoff:", SPATIAL_CUTOFF_A, "Å")
print("Input atom rows:", len(df_atoms))
print("Output directory:", FEATURE_DIR)


# ============================================================
# 3. HELPER FUNCTIONS
# ============================================================

def find_column(dataframe, candidates):
    """
    Поиск столбца без учёта регистра,
    пробелов и символов подчёркивания.
    """

    normalized = {

        "".join(
            char.lower()
            for char in str(column)
            if char.isalnum()
        ): column

        for column in dataframe.columns

    }

    for candidate in candidates:

        key = "".join(
            char.lower()
            for char in str(candidate)
            if char.isalnum()
        )

        if key in normalized:
            return normalized[key]

    return None


def is_missing_scalar(value):
    """
    Проверка пропуска для скалярных значений.
    """

    try:

        result = pd.isna(value)

        return (
            isinstance(result, (bool, np.bool_))
            and bool(result)
        )

    except (TypeError, ValueError):

        return False


def normalize_text(value, default=""):

    if value is None or is_missing_scalar(value):
        return default

    value = str(value).strip()

    if value.lower() in {
        "nan",
        "none",
        "nat",
        "<na>"
    }:
        return default

    return value


def safe_json_value(value):
    """
    Рекурсивное преобразование объектов
    в JSON-compatible формат.
    """

    if value is None:
        return None

    if isinstance(value, dict):

        return {
            str(key): safe_json_value(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple, set)):

        return [
            safe_json_value(item)
            for item in value
        ]

    if isinstance(value, np.ndarray):

        return safe_json_value(
            value.tolist()
        )

    if isinstance(value, (np.integer,)):

        return int(value)

    if isinstance(value, (np.floating, float)):

        number = float(value)

        return (
            number
            if np.isfinite(number)
            else None
        )

    if isinstance(value, (np.bool_,)):

        return bool(value)

    if isinstance(value, Path):

        return str(value)

    if isinstance(value, (pd.Timestamp, datetime)):

        return value.isoformat()

    if is_missing_scalar(value):

        return None

    if isinstance(value, (str, int, bool)):

        return value

    return str(value)


def get_sorted_pair_array(pairs):
    """
    Приведение неориентированных пар
    к детерминированному порядку.
    """

    pairs = np.asarray(
        pairs,
        dtype=np.int64
    )

    if pairs.size == 0:

        return np.empty(
            (0, 2),
            dtype=np.int64
        )

    pairs = pairs.reshape(-1, 2)

    pairs = np.sort(
        pairs,
        axis=1
    )

    pairs = np.unique(
        pairs,
        axis=0
    )

    order = np.lexsort(
        (
            pairs[:, 1],
            pairs[:, 0]
        )
    )

    return pairs[order]


def graph_statistics(node_count, edge_array):
    """
    Статистика неориентированного графа.

    Изолированные узлы учитываются как отдельные
    компоненты связности.
    """

    edge_count = len(edge_array)

    if node_count == 0:

        return {

            "node_count": 0,
            "edge_count": 0,
            "component_count": 0,
            "largest_component_size": 0,
            "largest_component_fraction": None,
            "isolated_node_count": 0,
            "mean_degree": None,
            "max_degree": None,
            "graph_density": None

        }

    degrees = np.zeros(
        node_count,
        dtype=np.int64
    )

    if edge_count > 0:

        degrees += np.bincount(
            edge_array[:, 0],
            minlength=node_count
        )

        degrees += np.bincount(
            edge_array[:, 1],
            minlength=node_count
        )

        rows = np.concatenate([
            edge_array[:, 0],
            edge_array[:, 1]
        ])

        cols = np.concatenate([
            edge_array[:, 1],
            edge_array[:, 0]
        ])

        adjacency = coo_matrix(

            (
                np.ones(
                    len(rows),
                    dtype=np.int8
                ),

                (rows, cols)

            ),

            shape=(node_count, node_count)

        ).tocsr()

    else:

        adjacency = coo_matrix(
            (node_count, node_count)
        ).tocsr()

    component_count, labels = connected_components(

        adjacency,

        directed=False,

        return_labels=True

    )

    component_sizes = np.bincount(
        labels,
        minlength=component_count
    )

    largest_component_size = int(
        component_sizes.max()
    )

    if node_count > 1:

        graph_density = float(
            2 * edge_count
            / (node_count * (node_count - 1))
        )

    else:

        graph_density = 0.0

    return {

        "node_count": int(node_count),

        "edge_count": int(edge_count),

        "component_count": int(component_count),

        "largest_component_size": largest_component_size,

        "largest_component_fraction": float(
            largest_component_size / node_count
        ),

        "isolated_node_count": int(
            np.sum(degrees == 0)
        ),

        "mean_degree": float(
            degrees.mean()
        ),

        "max_degree": int(
            degrees.max()
        ),

        "graph_density": graph_density

    }


def empty_dataframe(columns):

    return pd.DataFrame(
        columns=columns
    )


# ============================================================
# 4. PREPARE ATOM DATA
# ============================================================
#
# Исходный df_atoms не изменяется.
# ============================================================

df_spatial_source = df_atoms.copy(
    deep=True
)

df_spatial_source["_source_index"] = (
    df_spatial_source.index.astype(str)
)

df_spatial_source = df_spatial_source.reset_index(
    drop=True
)


# ------------------------------------------------------------
# 4.1. Coordinate columns
# ------------------------------------------------------------

x_col = find_column(
    df_spatial_source,
    [
        "x",
        "coord_x",
        "x_coord",
        "x_coordinate",
        "Cartn_x"
    ]
)

y_col = find_column(
    df_spatial_source,
    [
        "y",
        "coord_y",
        "y_coord",
        "y_coordinate",
        "Cartn_y"
    ]
)

z_col = find_column(
    df_spatial_source,
    [
        "z",
        "coord_z",
        "z_coord",
        "z_coordinate",
        "Cartn_z"
    ]
)


if not all(
    column is not None
    for column in [x_col, y_col, z_col]
):

    raise ValueError(
        "Could not identify X, Y, Z coordinate columns. "
        f"Available columns: {df_spatial_source.columns.tolist()}"
    )


# ------------------------------------------------------------
# 4.2. Structural columns
# ------------------------------------------------------------

pdb_col = find_column(
    df_spatial_source,
    ["pdb_id"]
)

element_col = find_column(
    df_spatial_source,
    ["element", "atom_element", "symbol"]
)

chain_col = find_column(
    df_spatial_source,
    ["chain_id", "chain"]
)

resname_col = find_column(
    df_spatial_source,
    ["residue_name", "resname"]
)

resnum_col = find_column(
    df_spatial_source,
    [
        "residue_id",
        "residue_number",
        "resnum",
        "resseq"
    ]
)

icode_col = find_column(
    df_spatial_source,
    [
        "insertion_code",
        "icode",
        "insertion"
    ]
)

hetero_col = find_column(
    df_spatial_source,
    [
        "hetero_flag",
        "hetero"
    ]
)

residue_key_col = find_column(
    df_spatial_source,
    ["residue_key"]
)

atom_name_col = find_column(
    df_spatial_source,
    ["atom_name", "name", "atom"]
)

atom_serial_col = find_column(
    df_spatial_source,
    ["atom_serial", "serial_number", "serial"]
)

altloc_col = find_column(
    df_spatial_source,
    ["altloc", "alt_loc"]
)


# ------------------------------------------------------------
# 4.3. Required columns
# ------------------------------------------------------------

required_columns = {

    "element": element_col,
    "chain_id": chain_col,
    "atom_name": atom_name_col

}

missing_columns = [
    name
    for name, column in required_columns.items()
    if column is None
]

if missing_columns:

    raise ValueError(
        "Required structural columns are missing: "
        f"{missing_columns}"
    )


if residue_key_col is None:

    if (
        resname_col is None
        or resnum_col is None
    ):

        raise ValueError(
            "Cannot construct residue identifiers. "
            "Need residue_key or residue_name + residue_id."
        )


# ------------------------------------------------------------
# 4.4. PDB ID consistency
# ------------------------------------------------------------

if pdb_col is not None:

    pdb_values = (

        df_spatial_source[pdb_col]

        .apply(
            lambda value: normalize_text(value).upper()
        )

    )

    pdb_values = sorted({
        value
        for value in pdb_values
        if value
    })

    if any(
        value != PDB_ID_VALUE
        for value in pdb_values
    ):

        raise ValueError(
            "df_atoms contains records from another PDB: "
            f"{pdb_values}"
        )


# ------------------------------------------------------------
# 4.5. Coordinate conversion
# ------------------------------------------------------------

for column in [x_col, y_col, z_col]:

    df_spatial_source[column] = pd.to_numeric(
        df_spatial_source[column],
        errors="coerce"
    )


coordinate_values = df_spatial_source[
    [x_col, y_col, z_col]
].to_numpy(
    dtype=np.float64
)


valid_coordinates = np.isfinite(
    coordinate_values
).all(axis=1)


invalid_coordinate_count = int(
    (~valid_coordinates).sum()
)


df_spatial_source = df_spatial_source.loc[
    valid_coordinates
].copy()


# ------------------------------------------------------------
# 4.6. Normalize atom fields
# ------------------------------------------------------------

df_spatial_source["_element"] = (

    df_spatial_source[element_col]

    .apply(
        lambda value: normalize_text(value).upper()
    )

)


df_spatial_source["_atom_name"] = (

    df_spatial_source[atom_name_col]

    .apply(
        lambda value: normalize_text(value)
    )

)


# Пустая цепь сохраняется как "".
# Отдельное значение "<blank>" используется только для отображения.

df_spatial_source["_chain_id"] = (

    df_spatial_source[chain_col]

    .apply(
        lambda value: normalize_text(value)
    )

)


if resname_col is not None:

    df_spatial_source["_resname"] = (

        df_spatial_source[resname_col]

        .apply(
            lambda value: normalize_text(
                value,
                default="UNK"
            ).upper()
        )

    )

else:

    df_spatial_source["_resname"] = "UNK"


if resnum_col is not None:

    df_spatial_source["_resnum"] = (

        df_spatial_source[resnum_col]

        .apply(
            lambda value: normalize_text(
                value,
                default="UNKNOWN"
            )
        )

    )

else:

    df_spatial_source["_resnum"] = "UNKNOWN"


if icode_col is not None:

    df_spatial_source["_icode"] = (

        df_spatial_source[icode_col]

        .apply(
            lambda value: normalize_text(value)
        )

    )

else:

    df_spatial_source["_icode"] = ""


if hetero_col is not None:

    df_spatial_source["_hetero_flag"] = (

        df_spatial_source[hetero_col]

        .apply(
            lambda value: normalize_text(
                value,
                default=" "
            )
        )

    )

else:

    df_spatial_source["_hetero_flag"] = " "


# ------------------------------------------------------------
# 4.7. Exclude hydrogen and deuterium
# ------------------------------------------------------------

before_hydrogen_filter = len(
    df_spatial_source
)

if EXCLUDE_HYDROGEN:

    df_spatial_source = df_spatial_source[
        ~df_spatial_source["_element"].isin(
            ["H", "D"]
        )
    ].copy()


excluded_hydrogen_count = (

    before_hydrogen_filter
    - len(df_spatial_source)

)


# ------------------------------------------------------------
# 4.8. Exclude records with unknown elements
# ------------------------------------------------------------

unknown_element_mask = (

    df_spatial_source["_element"]

    .isin([
        "",
        "UNKNOWN",
        "UNK",
        "NONE"
    ])

)

unknown_element_count = int(
    unknown_element_mask.sum()
)


df_spatial_source = df_spatial_source[
    ~unknown_element_mask
].copy()


# ------------------------------------------------------------
# 4.9. Build residue identifiers
# ------------------------------------------------------------
#
# Если residue_key из CELL 7 доступен, используем его.
# Это предпочтительнее самостоятельной реконструкции ключа.
#
# При отсутствии residue_key создаём ключ с учётом:
# chain_id + hetero_flag + residue_number + insertion_code.
# ------------------------------------------------------------

if residue_key_col is not None:

    df_spatial_source["_residue_key"] = (

        df_spatial_source[residue_key_col]

        .apply(
            lambda value: normalize_text(value)
        )

    )

    missing_keys = (
        df_spatial_source["_residue_key"] == ""
    )

    if missing_keys.any():

        fallback_keys = (

            df_spatial_source["_chain_id"]
            + "|"
            + df_spatial_source["_hetero_flag"]
            + "|"
            + df_spatial_source["_resnum"]
            + "|"
            + df_spatial_source["_icode"]

        )

        df_spatial_source.loc[
            missing_keys,
            "_residue_key"
        ] = fallback_keys.loc[
            missing_keys
        ]

else:

    df_spatial_source["_residue_key"] = (

        df_spatial_source["_chain_id"]
        + "|"
        + df_spatial_source["_hetero_flag"]
        + "|"
        + df_spatial_source["_resnum"]
        + "|"
        + df_spatial_source["_icode"]

    )


if (
    df_spatial_source["_residue_key"]
    .eq("")
    .any()
):

    raise ValueError(
        "Some atoms have empty residue identifiers."
    )


# ------------------------------------------------------------
# 4.10. Residue identity consistency
# ------------------------------------------------------------
#
# Один residue_key должен соответствовать одному набору
# структурных идентификаторов.
# ------------------------------------------------------------

residue_identity_columns = [

    "_chain_id",
    "_hetero_flag",
    "_resname",
    "_resnum",
    "_icode"

]


residue_identity_check = (

    df_spatial_source[
        ["_residue_key"] + residue_identity_columns
    ]

    .drop_duplicates()

    .groupby(
        "_residue_key",
        sort=False
    )

    .size()

)


ambiguous_residue_keys = residue_identity_check[
    residue_identity_check > 1
]


if not ambiguous_residue_keys.empty:

    raise ValueError(

        "Some residue_key values refer to different residue identities. "
        "Check residue identifiers in CELL 7. "
        f"Examples: {ambiguous_residue_keys.index[:10].tolist()}"

    )


# ------------------------------------------------------------
# 4.11. Atom site consistency
# ------------------------------------------------------------
#
# CELL 7 формирует таблицу с выбранной конформацией.
# Поэтому в пределах одного остатка не должно быть
# повторов одного atom_name + element.
#
# Повторы не удаляем автоматически: они могут указывать
# на проблему исходной таблицы или идентификаторов.
# ------------------------------------------------------------

atom_site_columns = [

    "_residue_key",
    "_atom_name",
    "_element"

]


duplicate_atom_mask = df_spatial_source.duplicated(
    subset=atom_site_columns,
    keep=False
)


duplicate_atom_count = int(
    duplicate_atom_mask.sum()
)


if duplicate_atom_count > 0:

    duplicate_examples = (

        df_spatial_source.loc[
            duplicate_atom_mask,
            atom_site_columns
        ]

        .head(10)

        .to_dict("records")

    )

    raise ValueError(

        "Duplicate atom sites detected. "
        "Check CELL 7 atom selection and residue identifiers. "
        f"Examples: {duplicate_examples}"

    )


if len(df_spatial_source) < 2:

    raise ValueError(
        "Fewer than two valid heavy atoms remain."
    )


print("\nATOM DATA PREPARATION")
print("-" * 75)

print("Valid heavy atom records:", len(df_spatial_source))

print("Invalid coordinate records:", invalid_coordinate_count)

print("Excluded H/D records:", excluded_hydrogen_count)

print("Unknown element records:", unknown_element_count)

print("Duplicate atom sites:", duplicate_atom_count)

print(
    "Unique residues:",
    df_spatial_source["_residue_key"].nunique()
)


# ============================================================
# 5. ATOM SPATIAL NODES
# ============================================================

df_atom_spatial_nodes = (

    df_spatial_source

    .copy(deep=True)

    .reset_index(drop=True)

)


df_atom_spatial_nodes["graph_atom_id"] = np.arange(
    len(df_atom_spatial_nodes),
    dtype=np.int64
)


df_atom_spatial_nodes["source_atom_row"] = (
    df_atom_spatial_nodes["_source_index"]
)


df_atom_spatial_nodes["pdb_id"] = PDB_ID_VALUE


df_atom_spatial_nodes["x_A"] = (
    df_atom_spatial_nodes[x_col].astype(float)
)

df_atom_spatial_nodes["y_A"] = (
    df_atom_spatial_nodes[y_col].astype(float)
)

df_atom_spatial_nodes["z_A"] = (
    df_atom_spatial_nodes[z_col].astype(float)
)


df_atom_spatial_nodes["element"] = (
    df_atom_spatial_nodes["_element"]
)

df_atom_spatial_nodes["atom_name"] = (
    df_atom_spatial_nodes["_atom_name"]
)

df_atom_spatial_nodes["chain_id"] = (
    df_atom_spatial_nodes["_chain_id"]
)

df_atom_spatial_nodes["chain_id_display"] = (

    df_atom_spatial_nodes["chain_id"]

    .replace("", "<blank>")

)

df_atom_spatial_nodes["residue_name"] = (
    df_atom_spatial_nodes["_resname"]
)

df_atom_spatial_nodes["residue_number"] = (
    df_atom_spatial_nodes["_resnum"]
)

df_atom_spatial_nodes["insertion_code"] = (
    df_atom_spatial_nodes["_icode"]
)

df_atom_spatial_nodes["hetero_flag"] = (
    df_atom_spatial_nodes["_hetero_flag"]
)

df_atom_spatial_nodes["residue_key"] = (
    df_atom_spatial_nodes["_residue_key"]
)


# Проверка уникальности идентификаторов атомных узлов.

if df_atom_spatial_nodes["graph_atom_id"].duplicated().any():

    raise ValueError(
        "Atom graph node IDs are not unique."
    )


# ============================================================
# 6. ATOM SPATIAL NEIGHBOR SEARCH
# ============================================================

coordinates = df_atom_spatial_nodes[
    ["x_A", "y_A", "z_A"]
].to_numpy(
    dtype=np.float64
)


if not np.isfinite(coordinates).all():

    raise ValueError(
        "Non-finite coordinates remain in atom nodes."
    )


spatial_tree = cKDTree(
    coordinates
)


atom_pairs = spatial_tree.query_pairs(

    r=SPATIAL_CUTOFF_A,

    output_type="ndarray"

)


atom_pairs = get_sorted_pair_array(
    atom_pairs
)


print("\nATOM NEIGHBOR SEARCH")
print("-" * 75)

print("Atoms indexed:", len(coordinates))

print("Atom pairs found:", len(atom_pairs))


# ============================================================
# 7. ATOM SPATIAL EDGES
# ============================================================

atom_edge_columns = [

    "pdb_id",

    "atom_i",
    "atom_j",

    "atom_name_i",
    "atom_name_j",

    "element_i",
    "element_j",

    "residue_i",
    "residue_j",

    "chain_i",
    "chain_j",

    "resname_i",
    "resname_j",

    "resnum_i",
    "resnum_j",

    "distance_A",

    "same_residue",
    "same_chain"

]


if len(atom_pairs) > 0:

    atom_i = atom_pairs[:, 0]
    atom_j = atom_pairs[:, 1]

    distances = np.linalg.norm(
        coordinates[atom_i] - coordinates[atom_j],
        axis=1
    )

    node_elements = (
        df_atom_spatial_nodes["element"].to_numpy()
    )

    node_residues = (
        df_atom_spatial_nodes["residue_key"].to_numpy()
    )

    node_chains = (
        df_atom_spatial_nodes["chain_id"].to_numpy()
    )

    node_resnames = (
        df_atom_spatial_nodes["residue_name"].to_numpy()
    )

    node_resnums = (
        df_atom_spatial_nodes["residue_number"].to_numpy()
    )

    node_atom_names = (
        df_atom_spatial_nodes["atom_name"].to_numpy()
    )

    same_residue = (
        node_residues[atom_i]
        == node_residues[atom_j]
    )

    same_chain = (
        node_chains[atom_i]
        == node_chains[atom_j]
    )

    df_atom_spatial_edges = pd.DataFrame({

        "pdb_id": PDB_ID_VALUE,

        "atom_i": atom_i,
        "atom_j": atom_j,

        "atom_name_i": node_atom_names[atom_i],
        "atom_name_j": node_atom_names[atom_j],

        "element_i": node_elements[atom_i],
        "element_j": node_elements[atom_j],

        "residue_i": node_residues[atom_i],
        "residue_j": node_residues[atom_j],

        "chain_i": node_chains[atom_i],
        "chain_j": node_chains[atom_j],

        "resname_i": node_resnames[atom_i],
        "resname_j": node_resnames[atom_j],

        "resnum_i": node_resnums[atom_i],
        "resnum_j": node_resnums[atom_j],

        "distance_A": distances,

        "same_residue": same_residue,

        "same_chain": same_chain

    })

else:

    df_atom_spatial_edges = empty_dataframe(
        atom_edge_columns
    )


# ------------------------------------------------------------
# 7.1. Intra-residue and inter-residue atom edges
# ------------------------------------------------------------

df_atom_intra_residue_edges = (

    df_atom_spatial_edges[
        df_atom_spatial_edges["same_residue"]
    ]

    .copy()

)


df_atom_inter_residue_edges = (

    df_atom_spatial_edges[
        ~df_atom_spatial_edges["same_residue"]
    ]

    .copy()

)


# ============================================================
# 8. RESIDUE SPATIAL NODES
# ============================================================
#
# Центр остатка = среднее координат тяжёлых атомов.
# ============================================================

df_residue_spatial_nodes = (

    df_atom_spatial_nodes

    .groupby(
        "residue_key",
        sort=False,
        as_index=False
    )

    .agg(

        chain_id=("chain_id", "first"),

        chain_id_display=("chain_id_display", "first"),

        hetero_flag=("hetero_flag", "first"),

        residue_name=("residue_name", "first"),

        residue_number=("residue_number", "first"),

        insertion_code=("insertion_code", "first"),

        atom_count=("graph_atom_id", "count"),

        center_x_A=("x_A", "mean"),

        center_y_A=("y_A", "mean"),

        center_z_A=("z_A", "mean")

    )

)


df_residue_spatial_nodes["pdb_id"] = PDB_ID_VALUE


df_residue_spatial_nodes["graph_residue_id"] = np.arange(
    len(df_residue_spatial_nodes),
    dtype=np.int64
)


residue_id_map = dict(

    zip(

        df_residue_spatial_nodes["residue_key"],

        df_residue_spatial_nodes["graph_residue_id"]

    )

)


if len(residue_id_map) != len(df_residue_spatial_nodes):

    raise ValueError(
        "Residue graph node identifiers are not unique."
    )


# ============================================================
# 9. RESIDUE SPATIAL EDGES
# ============================================================
#
# Сначала нормализуем порядок пар остатков по graph_residue_id,
# затем группируем атомные контакты.
#
# Благодаря этому одна пара остатков всегда представлена
# одним ребром независимо от порядка атомных индексов.
# ============================================================

residue_edge_columns = [

    "residue_i",
    "residue_j",

    "min_distance_A",
    "mean_contact_distance_A",

    "atom_contact_count",

    "residue_node_i",
    "residue_node_j",

    "chain_i",
    "chain_j",

    "resname_i",
    "resname_j",

    "same_chain",

    "pdb_id"

]


if not df_atom_inter_residue_edges.empty:

    df_residue_atom_pairs = (
        df_atom_inter_residue_edges.copy()
    )

    df_residue_atom_pairs["_node_i"] = (

        df_residue_atom_pairs["residue_i"]
        .map(residue_id_map)

    )

    df_residue_atom_pairs["_node_j"] = (

        df_residue_atom_pairs["residue_j"]
        .map(residue_id_map)

    )


    if (
        df_residue_atom_pairs[
            ["_node_i", "_node_j"]
        ].isna().any().any()
    ):

        raise ValueError(
            "Some residue contacts cannot be mapped to graph nodes."
        )


    df_residue_atom_pairs["_node_i"] = (
        df_residue_atom_pairs["_node_i"].astype(np.int64)
    )

    df_residue_atom_pairs["_node_j"] = (
        df_residue_atom_pairs["_node_j"].astype(np.int64)
    )


    # Нормализация порядка узлов.

    df_residue_atom_pairs["_residue_node_i"] = np.minimum(
        df_residue_atom_pairs["_node_i"],
        df_residue_atom_pairs["_node_j"]
    )

    df_residue_atom_pairs["_residue_node_j"] = np.maximum(
        df_residue_atom_pairs["_node_i"],
        df_residue_atom_pairs["_node_j"]
    )


    df_residue_spatial_edges = (

        df_residue_atom_pairs

        .groupby(
            [
                "_residue_node_i",
                "_residue_node_j"
            ],
            sort=False,
            as_index=False
        )

        .agg(

            min_distance_A=("distance_A", "min"),

            mean_contact_distance_A=("distance_A", "mean"),

            atom_contact_count=("distance_A", "count")

        )

    )


    # Идентификаторы остатков из таблицы узлов.

    residue_metadata = (

        df_residue_spatial_nodes

        .set_index("graph_residue_id")

    )


    df_residue_spatial_edges["residue_node_i"] = (
        df_residue_spatial_edges["_residue_node_i"]
    )

    df_residue_spatial_edges["residue_node_j"] = (
        df_residue_spatial_edges["_residue_node_j"]
    )


    df_residue_spatial_edges["residue_i"] = (

        df_residue_spatial_edges["residue_node_i"]

        .map(residue_metadata["residue_key"])

    )

    df_residue_spatial_edges["residue_j"] = (

        df_residue_spatial_edges["residue_node_j"]

        .map(residue_metadata["residue_key"])

    )


    df_residue_spatial_edges["chain_i"] = (

        df_residue_spatial_edges["residue_node_i"]

        .map(residue_metadata["chain_id"])

    )

    df_residue_spatial_edges["chain_j"] = (

        df_residue_spatial_edges["residue_node_j"]

        .map(residue_metadata["chain_id"])

    )


    df_residue_spatial_edges["resname_i"] = (

        df_residue_spatial_edges["residue_node_i"]

        .map(residue_metadata["residue_name"])

    )

    df_residue_spatial_edges["resname_j"] = (

        df_residue_spatial_edges["residue_node_j"]

        .map(residue_metadata["residue_name"])

    )


    df_residue_spatial_edges["same_chain"] = (

        df_residue_spatial_edges["chain_i"]
        == df_residue_spatial_edges["chain_j"]

    )


    df_residue_spatial_edges["pdb_id"] = PDB_ID_VALUE


    df_residue_spatial_edges = (

        df_residue_spatial_edges

        .drop(
            columns=[
                "_residue_node_i",
                "_residue_node_j"
            ]
        )

        .sort_values(
            [
                "residue_node_i",
                "residue_node_j"
            ]
        )

        .reset_index(drop=True)

    )

else:

    df_residue_spatial_edges = empty_dataframe(
        residue_edge_columns
    )


# ============================================================
# 10. GRAPH CONNECTIVITY STATISTICS
# ============================================================

atom_edge_array = (

    df_atom_spatial_edges[
        ["atom_i", "atom_j"]
    ]

    .to_numpy(dtype=np.int64)

    if not df_atom_spatial_edges.empty

    else np.empty((0, 2), dtype=np.int64)

)


residue_edge_array = (

    df_residue_spatial_edges[
        ["residue_node_i", "residue_node_j"]
    ]

    .to_numpy(dtype=np.int64)

    if not df_residue_spatial_edges.empty

    else np.empty((0, 2), dtype=np.int64)

)


atom_graph_stats = graph_statistics(

    node_count=len(df_atom_spatial_nodes),

    edge_array=atom_edge_array

)


residue_graph_stats = graph_statistics(

    node_count=len(df_residue_spatial_nodes),

    edge_array=residue_edge_array

)


# ============================================================
# 11. DISTANCE STATISTICS
# ============================================================

if not df_atom_spatial_edges.empty:

    atom_distance_stats = {

        "min_distance_A": float(
            df_atom_spatial_edges["distance_A"].min()
        ),

        "mean_distance_A": float(
            df_atom_spatial_edges["distance_A"].mean()
        ),

        "median_distance_A": float(
            df_atom_spatial_edges["distance_A"].median()
        ),

        "max_distance_A": float(
            df_atom_spatial_edges["distance_A"].max()
        )

    }

else:

    atom_distance_stats = {

        "min_distance_A": None,
        "mean_distance_A": None,
        "median_distance_A": None,
        "max_distance_A": None

    }


if not df_residue_spatial_edges.empty:

    residue_distance_stats = {

        "min_distance_A": float(
            df_residue_spatial_edges["min_distance_A"].min()
        ),

        "mean_min_distance_A": float(
            df_residue_spatial_edges["min_distance_A"].mean()
        ),

        "mean_contact_distance_A": float(
            df_residue_spatial_edges[
                "mean_contact_distance_A"
            ].mean()
        ),

        "max_min_distance_A": float(
            df_residue_spatial_edges["min_distance_A"].max()
        )

    }

else:

    residue_distance_stats = {

        "min_distance_A": None,
        "mean_min_distance_A": None,
        "mean_contact_distance_A": None,
        "max_min_distance_A": None

    }


# ============================================================
# 12. CHAIN-LEVEL GRAPH STATISTICS
# ============================================================
#
# Для каждой цепи отдельно считаются:
# - количество атомных узлов;
# - количество остатков;
# - внут­рицепочечные атомные рёбра;
# - межцепочечные рёбра, инцидентные этой цепи;
# - количество цепей-партнёров по межцепочечным контактам.
#
# Межцепочечное ребро учитывается для обеих участвующих цепей.
# ============================================================

df_atom_nodes_by_chain = (

    df_atom_spatial_nodes

    .groupby(
        "chain_id",
        sort=False
    )

    .agg(

        atom_node_count=("graph_atom_id", "count"),

        residue_count=("residue_key", "nunique")

    )

    .reset_index()

)


df_chain_spatial_statistics = (
    df_atom_nodes_by_chain.copy()
)


df_chain_spatial_statistics[
    "intra_chain_atom_edge_count"
] = 0


df_chain_spatial_statistics[
    "inter_chain_atom_edge_incident_count"
] = 0


df_chain_spatial_statistics[
    "inter_chain_partner_count"
] = 0


if not df_atom_spatial_edges.empty:

    # Внутрицепочечные рёбра.

    intra_chain_counts = (

        df_atom_spatial_edges[
            df_atom_spatial_edges["same_chain"]
        ]

        .groupby("chain_i")
        .size()

    )

    df_chain_spatial_statistics[
        "intra_chain_atom_edge_count"
    ] = (

        df_chain_spatial_statistics["chain_id"]

        .map(intra_chain_counts)

        .fillna(0)

        .astype(int)

    )


    # Межцепочечные рёбра.
    # Каждое такое ребро учитывается для обеих цепей.

    df_inter_chain_edges = (

        df_atom_spatial_edges[
            ~df_atom_spatial_edges["same_chain"]
        ]

        .copy()

    )


    incident_counts = pd.concat([

        df_inter_chain_edges["chain_i"],

        df_inter_chain_edges["chain_j"]

    ]).value_counts()


    df_chain_spatial_statistics[
        "inter_chain_atom_edge_incident_count"
    ] = (

        df_chain_spatial_statistics["chain_id"]

        .map(incident_counts)

        .fillna(0)

        .astype(int)

    )


    # Число уникальных цепей-партнёров.

    partner_pairs = pd.concat([

        df_inter_chain_edges[
            ["chain_i", "chain_j"]
        ].rename(
            columns={
                "chain_i": "chain_id",
                "chain_j": "partner_chain"
            }
        ),

        df_inter_chain_edges[
            ["chain_j", "chain_i"]
        ].rename(
            columns={
                "chain_j": "chain_id",
                "chain_i": "partner_chain"
            }
        )

    ], ignore_index=True)


    if not partner_pairs.empty:

        partner_counts = (

            partner_pairs

            .drop_duplicates()

            .groupby("chain_id")["partner_chain"]

            .nunique()

        )

        df_chain_spatial_statistics[
            "inter_chain_partner_count"
        ] = (

            df_chain_spatial_statistics["chain_id"]

            .map(partner_counts)

            .fillna(0)

            .astype(int)

        )

else:

    df_inter_chain_edges = (
        df_atom_spatial_edges.copy()
    )


df_chain_spatial_statistics["chain_id_display"] = (

    df_chain_spatial_statistics["chain_id"]

    .replace("", "<blank>")

)


df_chain_spatial_statistics = (

    df_chain_spatial_statistics

    [[

        "chain_id",
        "chain_id_display",

        "atom_node_count",
        "residue_count",

        "intra_chain_atom_edge_count",

        "inter_chain_atom_edge_incident_count",

        "inter_chain_partner_count"

    ]]

    .sort_values(
        "chain_id",
        kind="stable"
    )

    .reset_index(drop=True)

)


# ============================================================
# 13. GRAPH SUMMARY
# ============================================================

spatial_graph_summary = {

    "metadata": {

        "pdb_id": PDB_ID_VALUE,

        "created_at": datetime.now().isoformat(
            timespec="seconds"
        ),

        "cell": 23,

        "analysis_scope": ANALYSIS_SCOPE,

        "spatial_cutoff_A": float(
            SPATIAL_CUTOFF_A
        ),

        "hydrogen_excluded": bool(
            EXCLUDE_HYDROGEN
        ),

        "distance_tolerance_A": float(
            DISTANCE_TOLERANCE
        ),

        "coordinate_columns": {

            "x": str(x_col),
            "y": str(y_col),
            "z": str(z_col)

        }

    },

    "input_quality": {

        "input_atom_rows": int(len(df_atoms)),

        "valid_heavy_atom_rows": int(
            len(df_spatial_source)
        ),

        "invalid_coordinate_rows": invalid_coordinate_count,

        "excluded_hydrogen_deuterium_records": (
            excluded_hydrogen_count
        ),

        "unknown_element_records": unknown_element_count,

        "duplicate_atom_site_count": duplicate_atom_count,

        "unique_residue_count": int(
            df_spatial_source["_residue_key"].nunique()
        )

    },

    "atom_graph": {

        **atom_graph_stats,

        "distance_statistics": atom_distance_stats,

        "intra_residue_edge_count": int(
            len(df_atom_intra_residue_edges)
        ),

        "inter_residue_edge_count": int(
            len(df_atom_inter_residue_edges)
        ),

        "inter_chain_edge_count": int(
            len(df_inter_chain_edges)
        )

    },

    "residue_graph": {

        **residue_graph_stats,

        "distance_statistics": residue_distance_stats

    },

    "chain_statistics": {

        "chain_count": int(
            len(df_chain_spatial_statistics)
        ),

        "inter_chain_edge_count": int(
            len(df_inter_chain_edges)
        )

    },

    "interpretation": {

        "atom_edges": (
            "An atom edge represents spatial proximity "
            "within the specified cutoff. It is not automatically "
            "a chemical bond."
        ),

        "residue_edges": (
            "A residue edge exists when at least one heavy-atom pair "
            "between two different residues is within the cutoff."
        ),

        "residue_center": (
            "Residue coordinates represent the arithmetic mean "
            "of heavy-atom coordinates, not the center of mass."
        ),

        "distance": (
            "All distances are expressed in Angstrom."
        ),

        "model_scope": (
            "The analysis uses the first-model atom table from CELL 7 "
            "and its selected atom conformations."
        ),

        "chain_statistics": (
            "Inter-chain atom edges are counted for both participating "
            "chains in the incident-edge statistic."
        ),

        "chemical_interpretation": (
            "Spatial proximity alone does not establish a chemical bond, "
            "hydrogen bond, metal coordination, or functional interaction."
        )

    }

}


# ============================================================
# 14. OUTPUT FILES
# ============================================================

spatial_files = {

    "atom_nodes":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_atom_nodes.csv",

    "atom_edges":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_atom_edges.csv",

    "atom_intra_residue_edges":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_atom_intra_residue_edges.csv",

    "atom_inter_residue_edges":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_atom_inter_residue_edges.csv",

    "residue_nodes":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_residue_nodes.csv",

    "residue_edges":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_residue_edges.csv",

    "chain_statistics":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_chain_statistics.csv",

    "inter_chain_edges":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_inter_chain_edges.csv",

    "summary":
        FEATURE_DIR / f"{PDB_ID_VALUE}_spatial_graph_summary.json"

}


# ============================================================
# 15. SAVE CSV TABLES
# ============================================================

print("\n" + "=" * 75)
print("SAVING SPATIAL GRAPH TABLES")
print("=" * 75)


output_tables = {

    "atom_nodes": df_atom_spatial_nodes,

    "atom_edges": df_atom_spatial_edges,

    "atom_intra_residue_edges": (
        df_atom_intra_residue_edges
    ),

    "atom_inter_residue_edges": (
        df_atom_inter_residue_edges
    ),

    "residue_nodes": df_residue_spatial_nodes,

    "residue_edges": df_residue_spatial_edges,

    "chain_statistics": df_chain_spatial_statistics,

    "inter_chain_edges": df_inter_chain_edges

}


for table_name, dataframe in output_tables.items():

    output_path = spatial_files[table_name]

    dataframe.to_csv(

        output_path,

        index=False,

        encoding="utf-8-sig"

    )

    if not output_path.is_file():

        raise RuntimeError(
            f"Failed to save {output_path}"
        )

    print(
        f"{output_path.name:65s} "
        f"{len(dataframe):9d} rows"
    )


# ============================================================
# 16. SAVE JSON SUMMARY
# ============================================================

with open(

    spatial_files["summary"],

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        safe_json_value(spatial_graph_summary),

        file,

        ensure_ascii=False,

        indent=4,

        allow_nan=False

    )


# ============================================================
# 17. VALIDATION
# ============================================================

print("\n" + "=" * 75)
print("SPATIAL GRAPH VALIDATION")
print("=" * 75)


# ------------------------------------------------------------
# 17.1. Node count checks
# ------------------------------------------------------------

assert (
    len(df_atom_spatial_nodes)
    == atom_graph_stats["node_count"]
)

assert (
    len(df_residue_spatial_nodes)
    == residue_graph_stats["node_count"]
)


# ------------------------------------------------------------
# 17.2. Atom node checks
# ------------------------------------------------------------

assert (
    df_atom_spatial_nodes["graph_atom_id"]
    .is_unique
)

assert (
    df_atom_spatial_nodes["residue_key"]
    .notna()
    .all()
)

assert (
    df_atom_spatial_nodes[
        ["x_A", "y_A", "z_A"]
    ].to_numpy(dtype=float)
    .__class__
    is np.ndarray
)


# ------------------------------------------------------------
# 17.3. Atom edge checks
# ------------------------------------------------------------

if not df_atom_spatial_edges.empty:

    assert (
        df_atom_spatial_edges["atom_i"]
        < df_atom_spatial_edges["atom_j"]
    ).all()

    assert np.isfinite(
        df_atom_spatial_edges["distance_A"].to_numpy()
    ).all()

    # Нулевые расстояния допускаются:
    # они могут указывать на совпадающие координаты,
    # но не должны автоматически удаляться.

    assert (
        df_atom_spatial_edges["distance_A"]
        >= 0
    ).all()

    assert (
        df_atom_spatial_edges["distance_A"]
        <= SPATIAL_CUTOFF_A + DISTANCE_TOLERANCE
    ).all()

    assert (
        df_atom_spatial_edges[
            ["atom_i", "atom_j"]
        ]

        .duplicated()

        .sum()
        == 0
    )


# ------------------------------------------------------------
# 17.4. Residue edge checks
# ------------------------------------------------------------

if not df_residue_spatial_edges.empty:

    assert (
        df_residue_spatial_edges["residue_node_i"]
        < df_residue_spatial_edges["residue_node_j"]
    ).all()

    assert np.isfinite(
        df_residue_spatial_edges[
            "min_distance_A"
        ].to_numpy()
    ).all()

    assert (
        df_residue_spatial_edges["min_distance_A"]
        >= 0
    ).all()

    assert (
        df_residue_spatial_edges["min_distance_A"]
        <= SPATIAL_CUTOFF_A + DISTANCE_TOLERANCE
    ).all()

    assert (
        df_residue_spatial_edges[
            ["residue_node_i", "residue_node_j"]
        ]

        .duplicated()

        .sum()
        == 0
    )

    assert (
        df_residue_spatial_edges["atom_contact_count"]
        > 0
    ).all()


# ------------------------------------------------------------
# 17.5. Residue mapping checks
# ------------------------------------------------------------

if not df_residue_spatial_edges.empty:

    assert (
        df_residue_spatial_edges[
            ["residue_node_i", "residue_node_j"]
        ]

        .notna()

        .all()

        .all()
    )

    valid_node_ids = set(
        df_residue_spatial_nodes["graph_residue_id"]
    )

    edge_node_ids = set(

        df_residue_spatial_edges[
            ["residue_node_i", "residue_node_j"]
        ]

        .to_numpy()
        .ravel()

    )

    assert edge_node_ids.issubset(
        valid_node_ids
    )


# ------------------------------------------------------------
# 17.6. Cross-level consistency checks
# ------------------------------------------------------------
#
# Каждое остаточное ребро должно соответствовать хотя бы
# одному межостаточному атомному контакту.
# Сумма atom_contact_count должна совпадать с числом
# межостаточных атомных рёбер.
# ------------------------------------------------------------

if not df_residue_spatial_edges.empty:

    assert (

        int(
            df_residue_spatial_edges[
                "atom_contact_count"
            ].sum()
        )

        == len(df_atom_inter_residue_edges)

    )

else:

    assert len(df_atom_inter_residue_edges) == 0


# ------------------------------------------------------------
# 17.7. Chain statistics checks
# ------------------------------------------------------------

assert (
    df_chain_spatial_statistics["chain_id"]
    .is_unique
)

assert (
    df_chain_spatial_statistics[
        "atom_node_count"
    ].sum()
    == len(df_atom_spatial_nodes)
)

assert (
    df_chain_spatial_statistics[
        "intra_chain_atom_edge_count"
    ].sum()
    == int(
        df_atom_spatial_edges["same_chain"].sum()
    )
)

assert (
    df_chain_spatial_statistics[
        "inter_chain_atom_edge_incident_count"
    ].sum()
    == 2 * len(df_inter_chain_edges)
)


# ------------------------------------------------------------
# 17.8. File checks
# ------------------------------------------------------------

for name, path in spatial_files.items():

    if not path.is_file():

        raise RuntimeError(
            f"Output file missing: {path}"
        )

    if path.stat().st_size == 0:

        raise RuntimeError(
            f"Output file is empty: {path}"
        )


# ------------------------------------------------------------
# 17.9. JSON check
# ------------------------------------------------------------

with open(
    spatial_files["summary"],
    "r",
    encoding="utf-8"
) as file:

    saved_summary = json.load(file)


if saved_summary["metadata"]["pdb_id"] != PDB_ID_VALUE:

    raise ValueError(
        "PDB ID mismatch in spatial graph JSON."
    )


if (
    saved_summary["metadata"]["spatial_cutoff_A"]
    != SPATIAL_CUTOFF_A
):

    raise ValueError(
        "Spatial cutoff mismatch in JSON summary."
    )


# ============================================================
# 18. FINAL REPORT
# ============================================================

print("\n" + "=" * 75)
print("ATOM SPATIAL GRAPH")
print("=" * 75)

for key, value in atom_graph_stats.items():

    print(f"{key:35s}: {value}")


print("\nATOM DISTANCE STATISTICS")

for key, value in atom_distance_stats.items():

    print(f"{key:35s}: {value}")


print("\nINTRA-RESIDUE ATOM EDGES:",
      len(df_atom_intra_residue_edges))

print("INTER-RESIDUE ATOM EDGES:",
      len(df_atom_inter_residue_edges))

print("INTER-CHAIN ATOM EDGES:",
      len(df_inter_chain_edges))


print("\n" + "=" * 75)
print("RESIDUE SPATIAL GRAPH")
print("=" * 75)

for key, value in residue_graph_stats.items():

    print(f"{key:35s}: {value}")


print("\nRESIDUE DISTANCE STATISTICS")

for key, value in residue_distance_stats.items():

    print(f"{key:35s}: {value}")


print("\n" + "=" * 75)
print("CHAIN STATISTICS")
print("=" * 75)

print(
    df_chain_spatial_statistics.to_string(
        index=False
    )
)


print("\n" + "=" * 75)
print("SAVED FILES")
print("=" * 75)

for name, path in spatial_files.items():

    print(
        f"{name:30s}: "
        f"{'OK' if path.is_file() else 'MISSING'}"
    )


print("\nOutput directory:")
print(FEATURE_DIR)


print("\nVALIDATION: PASSED")

print("CELL 23 COMPLETED")

print("=" * 75)
```

# подготовленное представление для следующего этапа машинного обучения.

```python
# ============================================================
# CELL 24. UNIFIED PROTEIN STRUCTURAL REPRESENTATION
# ============================================================
#
# НАЗНАЧЕНИЕ:
# Объединение результатов CELL 22 и CELL 23 в единое
# машинно-ориентированное представление структуры белка.
#
# СОЗДАВАЕМЫЕ ПРЕДСТАВЛЕНИЯ:
#
# ATOM GRAPH:
# X_atom                  - признаки атомов
# pos_atom                - координаты атомов
# edge_index_atom         - индексы атомных рёбер
# edge_attr_atom          - признаки атомных рёбер
# atom_node_ids           - исходные ID атомных узлов
#
# RESIDUE GRAPH:
# X_residue               - признаки аминокислотных остатков
# pos_residue             - координаты центров остатков
# edge_index_residue      - индексы остаточных рёбер
# edge_attr_residue       - признаки остаточных рёбер
# residue_node_ids        - исходные ID остаточных узлов
#
# GLOBAL:
# X_global                - глобальные признаки белка
#
# BINDING SITE:
# is_binding_site_atom    - атом принадлежит остатку binding site
# is_direct_contact_atom  - атом непосредственно присутствует
#                           в таблице атомных контактов
# is_binding_site_residue - остаток входит в binding site
#
# ВАЖНО:
# 1. Геометрическая близость не доказывает химическую связь.
# 2. Binding-site mask и direct-contact mask различаются.
# 3. Исходные ID узлов не используются как числовые признаки.
# 4. edge_index содержит позиции строк матриц узлов.
# 5. Обучение модели в этой ячейке не выполняется.
# ============================================================


import os
import json
import warnings

import numpy as np
import pandas as pd

from pathlib import Path


print("=" * 70)
print("CELL 24. UNIFIED PROTEIN STRUCTURAL REPRESENTATION")
print("=" * 70)


# ============================================================
# 1. PATHS AND SETTINGS
# ============================================================

PDB_NAME = str(
    globals().get("PDB_ID", "UNKNOWN")
).strip().upper()


if PDB_NAME in ("", "UNKNOWN", "NONE", "NAN"):

    raise ValueError(
        "Не удалось определить PDB_ID."
    )


if "FEATURE_DIR" in globals():

    feature_dir_24 = Path(
        FEATURE_DIR
    )

elif "feature_dir" in globals():

    feature_dir_24 = Path(
        feature_dir
    )

else:

    raise NameError(
        "Не найдена FEATURE_DIR. "
        "Сначала выполните CELL 22."
    )


output_dir_24 = (
    feature_dir_24.parent
    / "unified_representation"
)

output_dir_24.mkdir(
    parents=True,
    exist_ok=True
)


# Порог для контроля полноты сопоставления.
# Не является научным критерием качества структуры.

STRICT_BINDING_MAPPING_24 = True


# Сохранять признаки наличия пропусков.
# Это позволяет отличать реальный ноль от импутации.

ADD_MISSING_INDICATORS_24 = True


print(f"PDB ID: {PDB_NAME}")
print(f"Input directory: {feature_dir_24}")
print(f"Output directory: {output_dir_24}")


# ============================================================
# 2. LOADING CELL 22 AND CELL 23 DATA
# ============================================================


def load_csv_24(filename, required=True):

    path = feature_dir_24 / filename

    if not path.exists():

        if required:

            raise FileNotFoundError(
                f"Не найден файл: {path}\n"
                "Проверьте выполнение CELL 22 и CELL 23."
            )

        return pd.DataFrame()

    try:

        return pd.read_csv(
            path,
            low_memory=False
        )

    except Exception as exc:

        raise RuntimeError(
            f"Ошибка чтения файла {path}: {exc}"
        ) from exc


# ------------------------------------------------------------
# CELL 23: SPATIAL GRAPHS
# ------------------------------------------------------------

df_atom_nodes_24 = load_csv_24(
    f"{PDB_NAME}_spatial_atom_nodes.csv"
)

df_atom_edges_24 = load_csv_24(
    f"{PDB_NAME}_spatial_atom_edges.csv"
)

df_residue_nodes_24 = load_csv_24(
    f"{PDB_NAME}_spatial_residue_nodes.csv"
)

df_residue_edges_24 = load_csv_24(
    f"{PDB_NAME}_spatial_residue_edges.csv"
)


# ------------------------------------------------------------
# CELL 22: PROTEIN FEATURES
# ------------------------------------------------------------

df_global_24 = load_csv_24(
    f"{PDB_NAME}_global_features.csv",
    required=False
)

df_binding_24 = load_csv_24(
    f"{PDB_NAME}_binding_site_features.csv",
    required=False
)

df_interactions_24 = load_csv_24(
    f"{PDB_NAME}_interaction_features.csv",
    required=False
)

df_metal_24 = load_csv_24(
    f"{PDB_NAME}_metal_features.csv",
    required=False
)

df_atom_features_24 = load_csv_24(
    f"{PDB_NAME}_atom_features.csv",
    required=False
)

df_residue_features_24 = load_csv_24(
    f"{PDB_NAME}_residue_features.csv",
    required=False
)

df_binding_contacts_24 = load_csv_24(
    f"{PDB_NAME}_binding_atom_contact_features.csv",
    required=False
)


print("\nINPUT DATA")
print("-" * 70)


input_tables_24 = {

    "Atom nodes": df_atom_nodes_24,
    "Atom edges": df_atom_edges_24,

    "Residue nodes": df_residue_nodes_24,
    "Residue edges": df_residue_edges_24,

    "Global features": df_global_24,
    "Binding site": df_binding_24,

    "Binding atom contacts": df_binding_contacts_24,

    "Interactions": df_interactions_24,
    "Metal features": df_metal_24,

    "Atom features": df_atom_features_24,
    "Residue features": df_residue_features_24

}


for name, dataframe in input_tables_24.items():

    print(
        f"{name:27s}: {len(dataframe):8d} rows"
    )


if df_atom_nodes_24.empty:

    raise ValueError(
        "Таблица атомных узлов пуста."
    )


if df_residue_nodes_24.empty:

    raise ValueError(
        "Таблица остаточных узлов пуста."
    )


# ============================================================
# 3. HELPER FUNCTIONS
# ============================================================


def is_missing_24(value):
    """
    Безопасная проверка пропусков.
    """

    if value is None:

        return True

    try:

        result = pd.isna(value)

        if isinstance(result, (bool, np.bool_)):

            return bool(result)

    except Exception:

        pass

    return False


def normalize_text_24(value):
    """
    Нормализация текстовых идентификаторов.

    Пустая цепь остаётся пустой строкой.
    """

    if is_missing_24(value):

        return ""

    return str(value).strip()


def normalize_chain_24(value):
    """
    Нормализация идентификатора цепи.

    Не заменяем пустую цепь на строку 'nan'.
    """

    return normalize_text_24(value)


def normalize_number_24(value):
    """
    Нормализация номера остатка.
    """

    if is_missing_24(value):

        raise ValueError(
            "Номер остатка отсутствует."
        )

    try:

        numeric_value = float(value)

    except (ValueError, TypeError):

        raise ValueError(
            f"Некорректный номер остатка: {value}"
        )

    if not np.isfinite(numeric_value):

        raise ValueError(
            f"Номер остатка содержит NaN/inf: {value}"
        )

    if not numeric_value.is_integer():

        raise ValueError(
            f"Номер остатка не является целым: {value}"
        )

    return int(numeric_value)


def normalize_icode_24(value):
    """
    Пустой или отсутствующий insertion code -> ''.
    """

    return normalize_text_24(value)


def normalize_residue_name_24(value):

    return normalize_text_24(
        value
    ).upper()


def find_column_24(dataframe, candidates):
    """
    Поиск столбца с учётом регистра и разделителей.
    """

    normalized = {

        "".join(
            ch.lower()
            for ch in str(col)
            if ch.isalnum()
        ): col

        for col in dataframe.columns

    }

    for candidate in candidates:

        key = "".join(
            ch.lower()
            for ch in str(candidate)
            if ch.isalnum()
        )

        if key in normalized:

            return normalized[key]

    return None


def make_residue_key_24(
    chain_id,
    residue_name,
    residue_number,
    insertion_code=""
):
    """
    Составной идентификатор белкового остатка.

    Используется для сопоставления атомов, остатков
    и binding-site записей.

    Формат:
    (chain_id, residue_name, residue_number, insertion_code)

    Предполагается, что CELL 23 содержит белковые узлы.
    Для небелковых компонентов этот ключ не применяется.
    """

    return (

        normalize_chain_24(chain_id),

        normalize_residue_name_24(
            residue_name
        ),

        normalize_number_24(
            residue_number
        ),

        normalize_icode_24(
            insertion_code
        )

    )


def clean_numeric_matrix_24(
    dataframe,
    add_missing_indicators=False
):
    """
    Перевод признаков в float32.

    NaN и +/-inf заменяются нулями.

    При необходимости добавляются бинарные признаки
    наличия пропусков.

    Стандартизация не выполняется.
    """

    if dataframe.shape[1] == 0:

        return np.zeros(
            (len(dataframe), 0),
            dtype=np.float32
        ), []

    numeric = pd.DataFrame(
        index=dataframe.index
    )

    feature_names = []

    missing_blocks = []

    for col in dataframe.columns:

        converted = pd.to_numeric(
            dataframe[col],
            errors="coerce"
        )

        converted = converted.replace(
            [np.inf, -np.inf],
            np.nan
        )

        missing_mask = converted.isna().astype(
            np.float32
        )

        numeric[col] = converted.fillna(
            0.0
        ).astype(np.float32)

        feature_names.append(
            str(col)
        )

        if (
            add_missing_indicators
            and missing_mask.any()
        ):

            missing_blocks.append(
                missing_mask.to_numpy(
                    dtype=np.float32
                ).reshape(-1, 1)
            )

            feature_names.append(
                f"{col}__missing"
            )

    result = numeric.to_numpy(
        dtype=np.float32
    )

    if missing_blocks:

        result = np.concatenate(
            [result] + missing_blocks,
            axis=1
        )

        # Порядок имён должен совпадать с порядком матрицы.
        feature_names = []

        for col in dataframe.columns:

            feature_names.append(
                str(col)
            )

            converted = pd.to_numeric(
                dataframe[col],
                errors="coerce"
            )

            converted = converted.replace(
                [np.inf, -np.inf],
                np.nan
            )

            if converted.isna().any():

                feature_names.append(
                    f"{col}__missing"
                )

        # Формируем матрицу в том же порядке:
        # значение признака, затем его missing indicator.

        blocks = []

        for col in dataframe.columns:

            converted = pd.to_numeric(
                dataframe[col],
                errors="coerce"
            )

            converted = converted.replace(
                [np.inf, -np.inf],
                np.nan
            )

            blocks.append(
                converted.fillna(0.0).to_numpy(
                    dtype=np.float32
                ).reshape(-1, 1)
            )

            if converted.isna().any():

                blocks.append(
                    converted.isna().to_numpy(
                        dtype=np.float32
                    ).reshape(-1, 1)
                )

        result = np.concatenate(
            blocks,
            axis=1
        )

    return result.astype(np.float32), feature_names


def get_node_ids_24(
    dataframe,
    candidates,
    object_name
):
    """
    Получение исходных ID узлов.

    Проверяются:
    - отсутствие пропусков;
    - целочисленность;
    - уникальность.

    При отсутствии ID используются позиции строк.
    """

    col = find_column_24(
        dataframe,
        candidates
    )

    if col is None:

        warnings.warn(
            f"{object_name}: ID column not found. "
            "Используются позиции строк."
        )

        return np.arange(
            len(dataframe),
            dtype=np.int64
        )

    ids = pd.to_numeric(
        dataframe[col],
        errors="coerce"
    )

    if ids.isna().any():

        raise ValueError(
            f"{object_name}: обнаружены некорректные ID."
        )

    values = ids.to_numpy(
        dtype=np.float64
    )

    if not np.isfinite(values).all():

        raise ValueError(
            f"{object_name}: ID содержат NaN или inf."
        )

    if not np.all(
        values == np.floor(values)
    ):

        raise ValueError(
            f"{object_name}: обнаружены нецелочисленные ID."
        )

    if np.any(
        np.abs(values) > np.iinfo(np.int64).max
    ):

        raise ValueError(
            f"{object_name}: ID выходят за диапазон int64."
        )

    ids = values.astype(
        np.int64
    )

    if len(np.unique(ids)) != len(ids):

        raise ValueError(
            f"{object_name}: ID узлов не уникальны."
        )

    return ids


def make_bidirectional_edges_24(
    edge_dataframe,
    source_col,
    target_col,
    feature_columns,
    node_ids,
    object_name
):
    """
    Создание двунаправленных рёбер.

    Исходные ID преобразуются в позиции строк матрицы узлов.

    Повторные записи одной и той же неориентированной
    связи удаляются.

    edge_index содержит позиции строк, а не исходные ID.
    """

    if edge_dataframe.empty:

        return (

            np.empty(
                (2, 0),
                dtype=np.int64
            ),

            np.empty(
                (0, len(feature_columns)),
                dtype=np.float32
            )

        )

    if source_col not in edge_dataframe.columns:

        raise ValueError(
            f"{object_name}: отсутствует {source_col}. "
            f"Доступные колонки: {edge_dataframe.columns.tolist()}"
        )

    if target_col not in edge_dataframe.columns:

        raise ValueError(
            f"{object_name}: отсутствует {target_col}. "
            f"Доступные колонки: {edge_dataframe.columns.tolist()}"
        )

    source_raw = pd.to_numeric(
        edge_dataframe[source_col],
        errors="coerce"
    )

    target_raw = pd.to_numeric(
        edge_dataframe[target_col],
        errors="coerce"
    )

    if source_raw.isna().any() or target_raw.isna().any():

        raise ValueError(
            f"{object_name}: некорректные индексы рёбер."
        )

    source_values = source_raw.to_numpy(
        dtype=np.float64
    )

    target_values = target_raw.to_numpy(
        dtype=np.float64
    )

    if not np.isfinite(source_values).all():

        raise ValueError(
            f"{object_name}: source содержит NaN/inf."
        )

    if not np.isfinite(target_values).all():

        raise ValueError(
            f"{object_name}: target содержит NaN/inf."
        )

    if not np.all(
        source_values == np.floor(source_values)
    ):

        raise ValueError(
            f"{object_name}: source ID нецелочисленные."
        )

    if not np.all(
        target_values == np.floor(target_values)
    ):

        raise ValueError(
            f"{object_name}: target ID нецелочисленные."
        )

    source_raw = source_values.astype(
        np.int64
    )

    target_raw = target_values.astype(
        np.int64
    )

    id_to_position = {

        int(node_id): position

        for position, node_id in enumerate(node_ids)

    }

    known_ids = set(
        id_to_position.keys()
    )

    unknown_ids = (

        set(source_raw.tolist())
        | set(target_raw.tolist())

    ) - known_ids

    if unknown_ids:

        raise ValueError(
            f"{object_name}: рёбра содержат неизвестные ID: "
            f"{list(unknown_ids)[:10]}"
        )

    source = np.array(

        [
            id_to_position[int(value)]
            for value in source_raw
        ],

        dtype=np.int64

    )

    target = np.array(

        [
            id_to_position[int(value)]
            for value in target_raw
        ],

        dtype=np.int64

    )

    if feature_columns:

        edge_features, _ = clean_numeric_matrix_24(
            edge_dataframe[feature_columns],
            add_missing_indicators=False
        )

    else:

        edge_features = np.empty(
            (len(source), 0),
            dtype=np.float32
        )

    # Удаляем self-loops.

    no_self_loop = source != target

    source = source[no_self_loop]
    target = target[no_self_loop]

    edge_features = edge_features[
        no_self_loop
    ]

    # Удаляем дубликаты неориентированных рёбер.
    # Сохраняем первую встретившуюся запись.

    unique_edges = {}

    for i, (src, dst) in enumerate(
        zip(source, target)
    ):

        key = (
            min(int(src), int(dst)),
            max(int(src), int(dst))
        )

        if key not in unique_edges:

            unique_edges[key] = edge_features[i]

    unique_pairs = list(
        unique_edges.keys()
    )

    if not unique_pairs:

        return (

            np.empty(
                (2, 0),
                dtype=np.int64
            ),

            np.empty(
                (0, len(feature_columns)),
                dtype=np.float32
            )

        )

    src_unique = np.array(
        [pair[0] for pair in unique_pairs],
        dtype=np.int64
    )

    dst_unique = np.array(
        [pair[1] for pair in unique_pairs],
        dtype=np.int64
    )

    features_unique = np.stack(
        [
            unique_edges[pair]
            for pair in unique_pairs
        ],
        axis=0
    ).astype(np.float32)

    edge_index = np.vstack([

        np.concatenate([
            src_unique,
            dst_unique
        ]),

        np.concatenate([
            dst_unique,
            src_unique
        ])

    ])

    edge_attr = np.concatenate([

        features_unique,
        features_unique

    ], axis=0)

    return edge_index, edge_attr


def get_numeric_feature_columns_24(
    dataframe,
    excluded_columns
):
    """
    Выбор числовых признаков.

    Исключаются:
    - идентификаторы;
    - номера остатков;
    - координаты;
    - технические поля;
    - категориальные идентификаторы.

    Столбцы с единственным значением исключаются,
    поскольку не несут вариативной информации
    внутри одной структуры.
    """

    result = []

    for col in dataframe.columns:

        normalized = "".join(

            ch.lower()

            for ch in str(col)

            if ch.isalnum()

        )

        if normalized in excluded_columns:

            continue

        converted = pd.to_numeric(
            dataframe[col],
            errors="coerce"
        )

        converted = converted.replace(
            [np.inf, -np.inf],
            np.nan
        )

        if converted.notna().sum() == 0:

            continue

        if converted.nunique(dropna=True) <= 1:

            continue

        result.append(col)

    return result


# ============================================================
# 4. VALIDATE NODE TABLE SCHEMAS
# ============================================================


atom_chain_col_24 = find_column_24(
    df_atom_nodes_24,
    ["chain_id"]
)

atom_resname_col_24 = find_column_24(
    df_atom_nodes_24,
    ["residue_name"]
)

atom_resnum_col_24 = find_column_24(
    df_atom_nodes_24,
    ["residue_number", "resnum", "resseq"]
)

atom_icode_col_24 = find_column_24(
    df_atom_nodes_24,
    ["insertion_code", "icode"]
)

atom_name_col_24 = find_column_24(
    df_atom_nodes_24,
    ["atom_name"]
)


required_atom_mapping_24 = {

    "chain_id": atom_chain_col_24,

    "residue_name": atom_resname_col_24,

    "residue_number": atom_resnum_col_24,

    "atom_name": atom_name_col_24

}


missing_atom_mapping_24 = [

    name

    for name, col in required_atom_mapping_24.items()

    if col is None

]


if missing_atom_mapping_24:

    raise ValueError(
        "В atom nodes отсутствуют обязательные поля: "
        f"{missing_atom_mapping_24}. "
        f"Доступные колонки: {df_atom_nodes_24.columns.tolist()}"
    )


res_chain_col_24 = find_column_24(
    df_residue_nodes_24,
    ["chain_id"]
)

res_resname_col_24 = find_column_24(
    df_residue_nodes_24,
    ["residue_name"]
)

res_resnum_col_24 = find_column_24(
    df_residue_nodes_24,
    ["residue_number", "resnum", "resseq"]
)

res_icode_col_24 = find_column_24(
    df_residue_nodes_24,
    ["insertion_code", "icode"]
)

res_graph_id_col_24 = find_column_24(
    df_residue_nodes_24,
    ["graph_residue_id"]
)


required_res_mapping_24 = {

    "chain_id": res_chain_col_24,

    "residue_name": res_resname_col_24,

    "residue_number": res_resnum_col_24,

    "graph_residue_id": res_graph_id_col_24

}


missing_res_mapping_24 = [

    name

    for name, col in required_res_mapping_24.items()

    if col is None

]


if missing_res_mapping_24:

    raise ValueError(
        "В residue nodes отсутствуют обязательные поля: "
        f"{missing_res_mapping_24}. "
        f"Доступные колонки: {df_residue_nodes_24.columns.tolist()}"
    )


# ============================================================
# 5. NODE IDENTIFIERS
# ============================================================


atom_node_ids = get_node_ids_24(

    df_atom_nodes_24,

    ["graph_atom_id", "atom_node_id"],

    "Atom graph"

)


residue_node_ids = get_node_ids_24(

    df_residue_nodes_24,

    ["graph_residue_id"],

    "Residue graph"

)


# ============================================================
# 6. UNIFIED RESIDUE IDENTIFIERS
# ============================================================


# ------------------------------------------------------------
# 6.1. Atom residue keys
# ------------------------------------------------------------

atom_residue_keys_24 = []


for _, row in df_atom_nodes_24.iterrows():

    insertion_code = (

        row[atom_icode_col_24]

        if atom_icode_col_24 is not None

        else ""

    )

    key = make_residue_key_24(

        row[atom_chain_col_24],

        row[atom_resname_col_24],

        row[atom_resnum_col_24],

        insertion_code

    )

    atom_residue_keys_24.append(
        key
    )


# ------------------------------------------------------------
# 6.2. Проверяем уникальность соответствия атомов остаткам
# ------------------------------------------------------------

atom_residue_key_set_24 = set(
    atom_residue_keys_24
)


# ------------------------------------------------------------
# 6.3. Residue node keys
# ------------------------------------------------------------

residue_node_keys_24 = []


for _, row in df_residue_nodes_24.iterrows():

    insertion_code = (

        row[res_icode_col_24]

        if res_icode_col_24 is not None

        else ""

    )

    key = make_residue_key_24(

        row[res_chain_col_24],

        row[res_resname_col_24],

        row[res_resnum_col_24],

        insertion_code

    )

    residue_node_keys_24.append(
        key
    )


if len(set(residue_node_keys_24)) != len(
    residue_node_keys_24
):

    raise ValueError(
        "В residue nodes обнаружены повторяющиеся "
        "составные идентификаторы остатков."
    )


# Каждый остаточный узел должен иметь атомы.

residues_without_atoms_24 = (

    set(residue_node_keys_24)
    - atom_residue_key_set_24

)


if residues_without_atoms_24:

    raise ValueError(
        "Для некоторых остаточных узлов не найдены "
        "соответствующие атомы: "
        f"{list(residues_without_atoms_24)[:10]}"
    )


# Каждый атом должен относиться к известному остатку.

unknown_atom_residues_24 = (

    atom_residue_key_set_24
    - set(residue_node_keys_24)

)


if unknown_atom_residues_24:

    raise ValueError(
        "В atom nodes обнаружены остатки, отсутствующие "
        "в residue nodes: "
        f"{list(unknown_atom_residues_24)[:10]}"
    )


# ============================================================
# 7. BINDING SITE RESIDUE IDENTIFIERS
# ============================================================


binding_residue_keys_24 = set()

binding_available_24 = not df_binding_24.empty


if binding_available_24:

    binding_chain_col_24 = find_column_24(
        df_binding_24,
        [
            "protein_chain",
            "chain_id"
        ]
    )

    binding_resname_col_24 = find_column_24(
        df_binding_24,
        [
            "protein_residue_name",
            "residue_name"
        ]
    )

    binding_resnum_col_24 = find_column_24(
        df_binding_24,
        [
            "protein_residue_number",
            "residue_number",
            "resnum",
            "resseq"
        ]
    )

    binding_icode_col_24 = find_column_24(
        df_binding_24,
        [
            "protein_insertion_code",
            "insertion_code",
            "icode"
        ]
    )

    binding_columns_24 = {

        "chain": binding_chain_col_24,

        "residue_name": binding_resname_col_24,

        "residue_number": binding_resnum_col_24

    }

    missing_binding_columns_24 = [

        name

        for name, col in binding_columns_24.items()

        if col is None

    ]

    if missing_binding_columns_24:

        raise ValueError(
            "В binding-site table отсутствуют необходимые "
            "поля: "
            f"{missing_binding_columns_24}. "
            f"Доступные колонки: {df_binding_24.columns.tolist()}"
        )

    for _, row in df_binding_24.iterrows():

        insertion_code = (

            row[binding_icode_col_24]

            if binding_icode_col_24 is not None

            else ""

        )

        binding_residue_keys_24.add(

            make_residue_key_24(

                row[binding_chain_col_24],

                row[binding_resname_col_24],

                row[binding_resnum_col_24],

                insertion_code

            )

        )


# ============================================================
# 8. BINDING SITE MASKS
# ============================================================


is_binding_site_residue_24 = np.array(

    [

        key in binding_residue_keys_24

        for key in residue_node_keys_24

    ],

    dtype=np.float32

)


is_binding_site_atom_24 = np.array(

    [

        key in binding_residue_keys_24

        for key in atom_residue_keys_24

    ],

    dtype=np.float32

)


matched_residue_keys_24 = (

    set(residue_node_keys_24)
    & binding_residue_keys_24

)


unmatched_residue_keys_24 = (

    binding_residue_keys_24
    - set(residue_node_keys_24)

)


matched_residues_24 = int(
    is_binding_site_residue_24.sum()
)

matched_atoms_24 = int(
    is_binding_site_atom_24.sum()
)


print("\nRESIDUE IDENTIFIER VALIDATION")
print("-" * 70)

print(
    "Binding-site source available:",
    binding_available_24
)

print(
    "Unique binding-site residues:",
    len(binding_residue_keys_24)
)

print(
    "Matched residue nodes:",
    matched_residues_24
)

print(
    "Matched atom nodes:",
    matched_atoms_24
)

print(
    "Total residue nodes:",
    len(residue_node_keys_24)
)

print(
    "Total atom nodes:",
    len(atom_residue_keys_24)
)


if unmatched_residue_keys_24:

    message = (
        "Некоторые binding-site residues не найдены "
        "в residue graph: "
        f"{list(unmatched_residue_keys_24)[:10]}"
    )

    if STRICT_BINDING_MAPPING_24:

        raise ValueError(message)

    warnings.warn(message)


if binding_available_24 and not unmatched_residue_keys_24:

    print(
        "Все binding-site residues успешно сопоставлены."
    )


# Совместимость с именами переменных CELL 24.

residue_keys_24 = residue_node_keys_24

atom_binding_mask_24 = is_binding_site_atom_24

residue_binding_mask_24 = is_binding_site_residue_24


# ============================================================
# 9. DIRECT CONTACT ATOM IDENTIFICATION
# ============================================================


direct_contact_mask_24 = np.zeros(
    len(df_atom_nodes_24),
    dtype=np.float32
)


direct_contact_available_24 = False


direct_contact_mapping_info_24 = {

    "source": None,

    "contact_rows": int(
        len(df_binding_contacts_24)
    ),

    "unique_contact_atoms": 0,

    "matched_contact_atoms": 0,

    "unmatched_contact_atoms": 0,

    "status": "unavailable"

}


if not df_binding_contacts_24.empty:

    contact_chain_col_24 = find_column_24(
        df_binding_contacts_24,
        [
            "protein_chain",
            "chain_id"
        ]
    )

    contact_resname_col_24 = find_column_24(
        df_binding_contacts_24,
        [
            "protein_residue_name",
            "residue_name"
        ]
    )

    contact_resnum_col_24 = find_column_24(
        df_binding_contacts_24,
        [
            "protein_residue_number",
            "residue_number",
            "resnum",
            "resseq"
        ]
    )

    contact_icode_col_24 = find_column_24(
        df_binding_contacts_24,
        [
            "protein_insertion_code",
            "insertion_code",
            "icode"
        ]
    )

    contact_atom_name_col_24 = find_column_24(
        df_binding_contacts_24,
        [
            "protein_atom_name",
            "protein_contact_atom",
            "closest_protein_atom",
            "protein_atom",
            "atom_name"
        ]
    )

    contact_columns_24 = {

        "chain": contact_chain_col_24,

        "residue_name": contact_resname_col_24,

        "residue_number": contact_resnum_col_24,

        "atom_name": contact_atom_name_col_24

    }

    missing_contact_columns_24 = [

        name

        for name, col in contact_columns_24.items()

        if col is None

    ]

    if missing_contact_columns_24:

        warnings.warn(
            "Невозможно построить direct-contact mask. "
            f"Отсутствуют поля: {missing_contact_columns_24}. "
            f"Доступные поля: {df_binding_contacts_24.columns.tolist()}"
        )

        direct_contact_mapping_info_24["status"] = (
            "missing_required_columns"
        )

    else:

        direct_contact_keys_24 = set()

        for _, row in df_binding_contacts_24.iterrows():

            insertion_code = (

                row[contact_icode_col_24]

                if contact_icode_col_24 is not None

                else ""

            )

            residue_key = make_residue_key_24(

                row[contact_chain_col_24],

                row[contact_resname_col_24],

                row[contact_resnum_col_24],

                insertion_code

            )

            atom_name = normalize_text_24(
                row[contact_atom_name_col_24]
            ).upper()

            if atom_name:

                direct_contact_keys_24.add(

                    (
                        residue_key,
                        atom_name
                    )

                )

        matched_contact_keys_24 = set()

        for atom_index, (_, row) in enumerate(
            df_atom_nodes_24.iterrows()
        ):

            residue_key = atom_residue_keys_24[
                atom_index
            ]

            atom_name = normalize_text_24(
                row[atom_name_col_24]
            ).upper()

            atom_key = (
                residue_key,
                atom_name
            )

            if atom_key in direct_contact_keys_24:

                direct_contact_mask_24[
                    atom_index
                ] = 1.0

                matched_contact_keys_24.add(
                    atom_key
                )

        unmatched_contact_keys_24 = (

            direct_contact_keys_24
            - matched_contact_keys_24

        )

        direct_contact_available_24 = True

        direct_contact_mapping_info_24 = {

            "source": "binding_atom_contact_features",

            "contact_rows": int(
                len(df_binding_contacts_24)
            ),

            "unique_contact_atoms": int(
                len(direct_contact_keys_24)
            ),

            "matched_contact_atoms": int(
                len(matched_contact_keys_24)
            ),

            "unmatched_contact_atoms": int(
                len(unmatched_contact_keys_24)
            ),

            "status": "available"

        }


print("\nDIRECT CONTACT ATOMS")
print("-" * 70)

print(
    "Direct-contact feature available:",
    direct_contact_available_24
)

print(
    "Direct-contact atoms matched:",
    int(direct_contact_mask_24.sum())
)

print(
    "Direct-contact mapping:",
    direct_contact_mapping_info_24
)


# ============================================================
# 10. ATOM NODE FEATURES
# ============================================================


ELEMENTS_24 = [

    "C",
    "N",
    "O",
    "S",
    "P",
    "F",
    "CL",
    "BR",
    "I",
    "MG",
    "ZN",
    "FE",
    "CU",
    "CA",
    "NA",
    "K",
    "MN",
    "CO",
    "NI",
    "OTHER"

]


atom_element_col_24 = find_column_24(
    df_atom_nodes_24,
    ["element"]
)


if atom_element_col_24 is None:

    raise ValueError(
        "В атомных узлах отсутствует element."
    )


atom_elements_24 = (

    df_atom_nodes_24[
        atom_element_col_24
    ]

    .fillna("OTHER")

    .astype(str)

    .str.strip()

    .str.upper()

)


atom_elements_24 = atom_elements_24.where(

    atom_elements_24.isin(ELEMENTS_24),

    "OTHER"

)


atom_element_features_24 = np.column_stack([

    (
        atom_elements_24 == element
    ).astype(np.float32)

    for element in ELEMENTS_24

])


excluded_atom_columns_24 = {

    "graphatomid",
    "atomnodeid",
    "atomserial",
    "atomid",
    "serial",

    "resid",
    "residueid",
    "residuenumber",
    "resnum",
    "resseq",
    "modelid",

    "x",
    "y",
    "z",

    "xa",
    "ya",
    "za",

    "coordx",
    "coordy",
    "coordz",

    "xcoord",
    "ycoord",
    "zcoord",

    "chainid",
    "residuename",
    "residuekey",
    "insertioncode",
    "icode",

    "atomname",
    "atomfullname",
    "element",

    "heteroflag",
    "altloc",

    "proteinid",
    "pdbid",
    "structureid"

}


atom_numeric_columns_24 = get_numeric_feature_columns_24(

    df_atom_nodes_24,

    excluded_atom_columns_24

)


df_atom_numeric_24 = df_atom_nodes_24[
    atom_numeric_columns_24
].copy()


X_atom_numeric_24, atom_numeric_feature_names_24 = (
    clean_numeric_matrix_24(

        df_atom_numeric_24,

        add_missing_indicators=ADD_MISSING_INDICATORS_24

    )
)


# ------------------------------------------------------------
# 10.1. Atom coordinates
# ------------------------------------------------------------

atom_coordinate_columns_24 = [

    find_column_24(
        df_atom_nodes_24,
        ["x_A", "x"]
    ),

    find_column_24(
        df_atom_nodes_24,
        ["y_A", "y"]
    ),

    find_column_24(
        df_atom_nodes_24,
        ["z_A", "z"]
    )

]


if any(
    col is None
    for col in atom_coordinate_columns_24
):

    raise ValueError(
        "Не найдены координаты атомов. "
        f"Доступные колонки: {df_atom_nodes_24.columns.tolist()}"
    )


pos_atom = df_atom_nodes_24[
    atom_coordinate_columns_24
].apply(
    pd.to_numeric,
    errors="coerce"
).to_numpy(
    dtype=np.float32
)


if not np.isfinite(pos_atom).all():

    raise ValueError(
        "В координатах атомов обнаружены NaN или inf."
    )


# ------------------------------------------------------------
# 10.2. Final atom features
# ------------------------------------------------------------

X_atom = np.concatenate(

    [

        X_atom_numeric_24,

        atom_element_features_24,

        atom_binding_mask_24.reshape(-1, 1),

        direct_contact_mask_24.reshape(-1, 1)

    ],

    axis=1

).astype(np.float32)


atom_feature_names_24 = (

    [
        f"numeric_{col}"
        for col in atom_numeric_feature_names_24
    ]

    + [
        f"element_{element}"
        for element in ELEMENTS_24
    ]

    + [

        "is_binding_site_atom",

        "is_direct_contact_atom"

    ]

)


# ============================================================
# 11. RESIDUE NODE FEATURES
# ============================================================


AMINO_ACIDS_24 = [

    "ALA",
    "ARG",
    "ASN",
    "ASP",
    "CYS",
    "GLN",
    "GLU",
    "GLY",
    "HIS",
    "ILE",
    "LEU",
    "LYS",
    "MET",
    "PHE",
    "PRO",
    "SER",
    "THR",
    "TRP",
    "TYR",
    "VAL",
    "OTHER"

]


residue_names_24 = (

    df_residue_nodes_24[
        res_resname_col_24
    ]

    .fillna("OTHER")

    .astype(str)

    .str.strip()

    .str.upper()

)


residue_names_24 = residue_names_24.where(

    residue_names_24.isin(AMINO_ACIDS_24),

    "OTHER"

)


residue_type_features_24 = np.column_stack([

    (
        residue_names_24 == amino_acid
    ).astype(np.float32)

    for amino_acid in AMINO_ACIDS_24

])


excluded_residue_columns_24 = {

    "graphresidueid",
    "residueid",
    "resid",
    "residuenumber",
    "resnum",
    "resseq",
    "modelid",

    "chainid",
    "residuename",
    "residuekey",
    "insertioncode",
    "icode",

    "centerx",
    "centery",
    "centerz",

    "centerxa",
    "centerya",
    "centerza",

    "x",
    "y",
    "z",

    "xa",
    "ya",
    "za",

    "pdbid",
    "proteinid",
    "structureid"

}


residue_numeric_columns_24 = get_numeric_feature_columns_24(

    df_residue_nodes_24,

    excluded_residue_columns_24

)


df_residue_numeric_24 = df_residue_nodes_24[
    residue_numeric_columns_24
].copy()


X_residue_numeric_24, residue_numeric_feature_names_24 = (
    clean_numeric_matrix_24(

        df_residue_numeric_24,

        add_missing_indicators=ADD_MISSING_INDICATORS_24

    )
)


# ------------------------------------------------------------
# 11.1. Residue center coordinates
# ------------------------------------------------------------

residue_coordinate_columns_24 = [

    find_column_24(
        df_residue_nodes_24,
        ["center_x_A", "center_x"]
    ),

    find_column_24(
        df_residue_nodes_24,
        ["center_y_A", "center_y"]
    ),

    find_column_24(
        df_residue_nodes_24,
        ["center_z_A", "center_z"]
    )

]


if any(
    col is None
    for col in residue_coordinate_columns_24
):

    raise ValueError(
        "Не найдены координаты центров остатков. "
        f"Доступные колонки: {df_residue_nodes_24.columns.tolist()}"
    )


pos_residue = df_residue_nodes_24[
    residue_coordinate_columns_24
].apply(
    pd.to_numeric,
    errors="coerce"
).to_numpy(
    dtype=np.float32
)


if not np.isfinite(pos_residue).all():

    raise ValueError(
        "В координатах центров остатков обнаружены NaN или inf."
    )


# ------------------------------------------------------------
# 11.2. Final residue features
# ------------------------------------------------------------

X_residue = np.concatenate(

    [

        X_residue_numeric_24,

        residue_type_features_24,

        residue_binding_mask_24.reshape(-1, 1)

    ],

    axis=1

).astype(np.float32)


residue_feature_names_24 = (

    [
        f"numeric_{col}"
        for col in residue_numeric_feature_names_24
    ]

    + [
        f"amino_acid_{aa}"
        for aa in AMINO_ACIDS_24
    ]

    + [
        "is_binding_site_residue"
    ]

)


# ============================================================
# 12. ATOM EDGES
# ============================================================


atom_edge_features_24 = [

    col

    for col in [

        "distance_A",

        "same_residue"

    ]

    if col in df_atom_edges_24.columns

]


edge_index_atom, edge_attr_atom = (
    make_bidirectional_edges_24(

        df_atom_edges_24,

        "atom_i",

        "atom_j",

        atom_edge_features_24,

        atom_node_ids,

        "Atom graph"

    )
)


atom_edge_feature_names_24 = (
    atom_edge_features_24.copy()
)


# ============================================================
# 13. RESIDUE EDGES
# ============================================================


residue_edge_features_24 = [

    col

    for col in [

        "min_distance_A",

        "mean_contact_distance_A",

        "atom_contact_count"

    ]

    if col in df_residue_edges_24.columns

]


edge_index_residue, edge_attr_residue = (
    make_bidirectional_edges_24(

        df_residue_edges_24,

        "residue_node_i",

        "residue_node_j",

        residue_edge_features_24,

        residue_node_ids,

        "Residue graph"

    )
)


residue_edge_feature_names_24 = (
    residue_edge_features_24.copy()
)


# ============================================================
# 14. GLOBAL PROTEIN FEATURES
# ============================================================


excluded_global_columns_24 = {

    "pdbid",
    "structureid",
    "proteinid",
    "modelid"

}


if len(df_global_24) > 1:

    raise ValueError(
        "Ожидалась одна строка глобальных признаков "
        "для одной структуры белка."
    )


global_numeric_columns_24 = []


if not df_global_24.empty:

    for col in df_global_24.columns:

        normalized = "".join(

            ch.lower()

            for ch in str(col)

            if ch.isalnum()

        )

        if normalized in excluded_global_columns_24:

            continue

        converted = pd.to_numeric(

            df_global_24[col],

            errors="coerce"

        )

        converted = converted.replace(
            [np.inf, -np.inf],
            np.nan
        )

        if converted.notna().any():

            global_numeric_columns_24.append(col)


if global_numeric_columns_24:

    X_global_matrix_24, global_feature_names_24 = (
        clean_numeric_matrix_24(

            df_global_24[
                global_numeric_columns_24
            ],

            add_missing_indicators=ADD_MISSING_INDICATORS_24

        )
    )

    X_global = X_global_matrix_24[0]

else:

    X_global = np.empty(
        0,
        dtype=np.float32
    )

    global_feature_names_24 = []


# ============================================================
# 15. FINAL FEATURE TABLES
# ============================================================


df_ml_atom_features = pd.DataFrame(

    X_atom,

    columns=atom_feature_names_24

)


df_ml_atom_features.insert(

    0,

    "graph_atom_id",

    atom_node_ids

)


df_ml_residue_features = pd.DataFrame(

    X_residue,

    columns=residue_feature_names_24

)


df_ml_residue_features.insert(

    0,

    "graph_residue_id",

    residue_node_ids

)


# Дополнительная таблица соответствия узлов остаткам.
# Она нужна для интерпретации предсказаний модели.

df_atom_node_mapping_24 = pd.DataFrame({

    "graph_atom_id": atom_node_ids,

    "chain_id": [

        key[0]
        for key in atom_residue_keys_24

    ],

    "residue_name": [

        key[1]
        for key in atom_residue_keys_24

    ],

    "residue_number": [

        key[2]
        for key in atom_residue_keys_24

    ],

    "insertion_code": [

        key[3]
        for key in atom_residue_keys_24

    ],

    "atom_name": (

        df_atom_nodes_24[
            atom_name_col_24
        ].astype(str).to_numpy()

    )

})


df_residue_node_mapping_24 = pd.DataFrame({

    "graph_residue_id": residue_node_ids,

    "chain_id": [

        key[0]
        for key in residue_node_keys_24

    ],

    "residue_name": [

        key[1]
        for key in residue_node_keys_24

    ],

    "residue_number": [

        key[2]
        for key in residue_node_keys_24

    ],

    "insertion_code": [

        key[3]
        for key in residue_node_keys_24

    ]

})


# ============================================================
# 16. VALIDATION
# ============================================================


print("\nVALIDATION")
print("-" * 70)


# ------------------------------------------------------------
# 16.1. Node dimensions
# ------------------------------------------------------------

assert X_atom.shape[0] == len(atom_node_ids)

assert pos_atom.shape == (
    len(atom_node_ids),
    3
)

assert X_residue.shape[0] == len(residue_node_ids)

assert pos_residue.shape == (
    len(residue_node_ids),
    3
)


# ------------------------------------------------------------
# 16.2. Finite numerical values
# ------------------------------------------------------------

assert np.isfinite(X_atom).all()

assert np.isfinite(X_residue).all()

assert np.isfinite(X_global).all()

assert np.isfinite(pos_atom).all()

assert np.isfinite(pos_residue).all()

assert np.isfinite(edge_attr_atom).all()

assert np.isfinite(edge_attr_residue).all()


# ------------------------------------------------------------
# 16.3. Edge index ranges
# ------------------------------------------------------------

if edge_index_atom.size:

    assert edge_index_atom.min() >= 0

    assert edge_index_atom.max() < len(X_atom)


if edge_index_residue.size:

    assert edge_index_residue.min() >= 0

    assert edge_index_residue.max() < len(X_residue)


# ------------------------------------------------------------
# 16.4. Feature dimensions
# ------------------------------------------------------------

assert X_atom.shape[1] == len(
    atom_feature_names_24
)

assert X_residue.shape[1] == len(
    residue_feature_names_24
)

assert X_global.shape[0] == len(
    global_feature_names_24
)


# ------------------------------------------------------------
# 16.5. Binary masks
# ------------------------------------------------------------

assert set(
    np.unique(atom_binding_mask_24)
).issubset({0.0, 1.0})


assert set(
    np.unique(direct_contact_mask_24)
).issubset({0.0, 1.0})


assert set(
    np.unique(residue_binding_mask_24)
).issubset({0.0, 1.0})


# ------------------------------------------------------------
# 16.6. Mask dimensions
# ------------------------------------------------------------

assert len(atom_binding_mask_24) == len(X_atom)

assert len(direct_contact_mask_24) == len(X_atom)

assert len(residue_binding_mask_24) == len(X_residue)


# ------------------------------------------------------------
# 16.7. Bidirectional edge consistency
# ------------------------------------------------------------

def validate_bidirectional_edges_24(
    edge_index,
    object_name
):

    if edge_index.size == 0:

        return True

    directed = set(

        zip(

            edge_index[0].tolist(),

            edge_index[1].tolist()

        )

    )

    for source, target in directed:

        if (target, source) not in directed:

            raise AssertionError(
                f"{object_name}: отсутствует обратное ребро "
                f"{target} -> {source}."
            )

    return True


validate_bidirectional_edges_24(
    edge_index_atom,
    "Atom graph"
)

validate_bidirectional_edges_24(
    edge_index_residue,
    "Residue graph"
)


# ------------------------------------------------------------
# 16.8. No self-loops
# ------------------------------------------------------------

if edge_index_atom.size:

    assert not np.any(
        edge_index_atom[0] == edge_index_atom[1]
    )


if edge_index_residue.size:

    assert not np.any(
        edge_index_residue[0] == edge_index_residue[1]
    )


# ------------------------------------------------------------
# 16.9. Node mapping consistency
# ------------------------------------------------------------

assert len(df_atom_node_mapping_24) == len(X_atom)

assert len(df_residue_node_mapping_24) == len(X_residue)


print("Node dimensions: OK")
print("Coordinate dimensions: OK")
print("Edge index ranges: OK")
print("Finite numerical values: OK")
print("Feature names: OK")
print("Binding masks: OK")
print("Bidirectional edges: OK")
print("No self-loops: OK")
print("Node mappings: OK")


# ============================================================
# 17. SAVE NPZ
# ============================================================


ml_data_path_24 = (

    output_dir_24

    / f"{PDB_NAME}_ml_graph_data.npz"

)


np.savez_compressed(

    ml_data_path_24,

    X_atom=X_atom,

    pos_atom=pos_atom,

    edge_index_atom=edge_index_atom,

    edge_attr_atom=edge_attr_atom,

    X_residue=X_residue,

    pos_residue=pos_residue,

    edge_index_residue=edge_index_residue,

    edge_attr_residue=edge_attr_residue,

    X_global=X_global,

    atom_node_ids=atom_node_ids,

    residue_node_ids=residue_node_ids,

    is_binding_site_atom=atom_binding_mask_24,

    is_direct_contact_atom=direct_contact_mask_24,

    is_binding_site_residue=residue_binding_mask_24,

    atom_feature_names=np.asarray(
        atom_feature_names_24,
        dtype=str
    ),

    residue_feature_names=np.asarray(
        residue_feature_names_24,
        dtype=str
    ),

    global_feature_names=np.asarray(
        global_feature_names_24,
        dtype=str
    ),

    atom_edge_feature_names=np.asarray(
        atom_edge_feature_names_24,
        dtype=str
    ),

    residue_edge_feature_names=np.asarray(
        residue_edge_feature_names_24,
        dtype=str
    )

)


# ============================================================
# 18. SAVE FEATURE TABLES
# ============================================================


ml_atom_csv_24 = (

    output_dir_24

    / f"{PDB_NAME}_ml_atom_features.csv"

)


ml_residue_csv_24 = (

    output_dir_24

    / f"{PDB_NAME}_ml_residue_features.csv"

)


atom_mapping_csv_24 = (

    output_dir_24

    / f"{PDB_NAME}_ml_atom_node_mapping.csv"

)


residue_mapping_csv_24 = (

    output_dir_24

    / f"{PDB_NAME}_ml_residue_node_mapping.csv"

)


df_ml_atom_features.to_csv(

    ml_atom_csv_24,

    index=False

)


df_ml_residue_features.to_csv(

    ml_residue_csv_24,

    index=False

)


df_atom_node_mapping_24.to_csv(

    atom_mapping_csv_24,

    index=False

)


df_residue_node_mapping_24.to_csv(

    residue_mapping_csv_24,

    index=False

)


# ------------------------------------------------------------
# 18.1. Global features
# ------------------------------------------------------------

global_csv_24 = (

    output_dir_24

    / f"{PDB_NAME}_ml_global_features.csv"

)


if len(X_global) > 0:

    pd.DataFrame(

        [X_global],

        columns=global_feature_names_24

    ).to_csv(

        global_csv_24,

        index=False

    )


# ============================================================
# 19. FEATURE MANIFEST
# ============================================================


ml_manifest_24 = {

    "pdb_id": PDB_NAME,

    "description": (
        "Unified protein structural representation "
        "for machine learning and graph neural networks"
    ),

    "source_cells": [22, 23, 24],

    "atom_graph": {

        "node_count": int(
            len(X_atom)
        ),

        "feature_count": int(
            X_atom.shape[1]
        ),

        "coordinate_dimensions": 3,

        "edge_count_directed": int(
            edge_index_atom.shape[1]
        ),

        "feature_names": atom_feature_names_24,

        "edge_feature_names": atom_edge_feature_names_24,

        "node_id_field": "atom_node_ids",

        "edge_index_convention": (
            "zero-based row positions in X_atom"
        )

    },

    "residue_graph": {

        "node_count": int(
            len(X_residue)
        ),

        "feature_count": int(
            X_residue.shape[1]
        ),

        "coordinate_dimensions": 3,

        "edge_count_directed": int(
            edge_index_residue.shape[1]
        ),

        "feature_names": residue_feature_names_24,

        "edge_feature_names": residue_edge_feature_names_24,

        "node_id_field": "residue_node_ids",

        "edge_index_convention": (
            "zero-based row positions in X_residue"
        )

    },

    "global_features": {

        "feature_count": int(
            len(X_global)
        ),

        "feature_names": global_feature_names_24,

        "available": bool(
            len(X_global) > 0
        )

    },

    "binding_site": {

        "source_available": bool(
            binding_available_24
        ),

        "binding_residue_count": int(
            len(binding_residue_keys_24)
        ),

        "matched_binding_residue_count": int(
            len(matched_residue_keys_24)
        ),

        "unmatched_binding_residue_count": int(
            len(unmatched_residue_keys_24)
        ),

        "binding_atom_count": int(
            atom_binding_mask_24.sum()
        ),

        "binding_residue_node_count": int(
            residue_binding_mask_24.sum()
        ),

        "direct_contact_atom_count": int(
            direct_contact_mask_24.sum()
        ),

        "direct_contact_feature_available": bool(
            direct_contact_available_24
        ),

        "direct_contact_mapping": (
            direct_contact_mapping_info_24
        )

    },

    "preprocessing": {

        "categorical_encoding": "one-hot",

        "missing_numeric_values": (
            "zero imputation"
        ),

        "missing_indicators": bool(
            ADD_MISSING_INDICATORS_24
        ),

        "feature_standardization": False,

        "coordinates_separate_from_node_features": True,

        "edge_representation": (
            "bidirectional, duplicate undirected edges removed"
        ),

        "edge_index_convention": (
            "zero-based row positions in node matrices"
        ),

        "node_identifiers_saved_separately": True,

        "training_performed": False

    },

    "files": {

        "ml_graph_data": ml_data_path_24.name,

        "atom_features": ml_atom_csv_24.name,

        "residue_features": ml_residue_csv_24.name,

        "atom_node_mapping": atom_mapping_csv_24.name,

        "residue_node_mapping": residue_mapping_csv_24.name,

        "global_features": (

            global_csv_24.name

            if len(X_global) > 0

            else None

        )

    }

}


manifest_path_24 = (

    output_dir_24

    / f"{PDB_NAME}_ml_manifest.json"

)


with open(

    manifest_path_24,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        ml_manifest_24,

        file,

        ensure_ascii=False,

        indent=4,

        allow_nan=False

    )


# ============================================================
# 20. FINAL REPORT
# ============================================================


print("\n" + "=" * 70)
print("UNIFIED PROTEIN REPRESENTATION SUMMARY")
print("=" * 70)


print("\nATOM GRAPH")
print("-" * 70)

print(
    f"Atom nodes:             {X_atom.shape[0]:,}"
)

print(
    f"Atom features:          {X_atom.shape[1]:,}"
)

print(
    f"Atom coordinates:       {pos_atom.shape}"
)

print(
    f"Atom edge index:        {edge_index_atom.shape}"
)

print(
    f"Atom edge features:     {edge_attr_atom.shape}"
)


print("\nRESIDUE GRAPH")
print("-" * 70)

print(
    f"Residue nodes:          {X_residue.shape[0]:,}"
)

print(
    f"Residue features:       {X_residue.shape[1]:,}"
)

print(
    f"Residue coordinates:    {pos_residue.shape}"
)

print(
    f"Residue edge index:     {edge_index_residue.shape}"
)

print(
    f"Residue edge features:  {edge_attr_residue.shape}"
)


print("\nGLOBAL FEATURES")
print("-" * 70)

print(
    f"Global feature count:   {len(X_global):,}"
)


print("\nBINDING SITE")
print("-" * 70)

print(
    "Binding-site residues:",
    len(binding_residue_keys_24)
)

print(
    "Matched binding residues:",
    len(matched_residue_keys_24)
)

print(
    "Binding-site atom count:",
    int(atom_binding_mask_24.sum())
)

print(
    "Direct-contact atom count:",
    int(direct_contact_mask_24.sum())
)

print(
    "Binding-site residue nodes:",
    int(residue_binding_mask_24.sum())
)


print("\nDIRECT CONTACT MAPPING")
print("-" * 70)

for key, value in direct_contact_mapping_info_24.items():

    print(
        f"{key:30s}: {value}"
    )


# ============================================================
# 21. SAVE VALIDATION
# ============================================================


output_files_24 = [

    ml_data_path_24,

    ml_atom_csv_24,

    ml_residue_csv_24,

    atom_mapping_csv_24,

    residue_mapping_csv_24,

    manifest_path_24

]


if len(X_global) > 0:

    output_files_24.append(
        global_csv_24
    )


print("\nSAVED FILES")
print("-" * 70)


for path in output_files_24:

    exists = path.exists()

    print(

        f"{path.name:48s}"

        f"{'OK' if exists else 'NOT FOUND'}"

    )

    assert exists, (
        f"Файл не создан: {path}"
    )


# Проверка возможности повторного чтения NPZ.

with np.load(
    ml_data_path_24,
    allow_pickle=False
) as npz_check_24:

    required_npz_keys_24 = [

        "X_atom",
        "pos_atom",
        "edge_index_atom",
        "edge_attr_atom",

        "X_residue",
        "pos_residue",
        "edge_index_residue",
        "edge_attr_residue",

        "X_global",

        "atom_node_ids",
        "residue_node_ids"

    ]

    missing_npz_keys_24 = [

        key

        for key in required_npz_keys_24

        if key not in npz_check_24.files

    ]

    if missing_npz_keys_24:

        raise RuntimeError(
            "В NPZ отсутствуют обязательные массивы: "
            f"{missing_npz_keys_24}"
        )


# Проверка JSON manifest.

with open(
    manifest_path_24,
    "r",
    encoding="utf-8"
) as file:

    manifest_check_24 = json.load(file)


if manifest_check_24.get("pdb_id") != PDB_NAME:

    raise RuntimeError(
        "PDB ID в manifest не совпадает с текущей структурой."
    )


# ============================================================
# 22. FINAL STATUS
# ============================================================


print("\n" + "=" * 70)
print("CELL 24 COMPLETED")
print("=" * 70)

print(
    "Unified protein representation successfully created."
)

print(
    f"\nOutput directory: {output_dir_24}"
)


print("\nMAIN ML OBJECTS")


main_arrays_24 = {

    "X_atom": X_atom,

    "pos_atom": pos_atom,

    "edge_index_atom": edge_index_atom,

    "edge_attr_atom": edge_attr_atom,

    "X_residue": X_residue,

    "pos_residue": pos_residue,

    "edge_index_residue": edge_index_residue,

    "edge_attr_residue": edge_attr_residue,

    "X_global": X_global,

    "atom_node_ids": atom_node_ids,

    "residue_node_ids": residue_node_ids,

    "is_binding_site_atom": atom_binding_mask_24,

    "is_direct_contact_atom": direct_contact_mask_24,

    "is_binding_site_residue": residue_binding_mask_24

}


for name, array in main_arrays_24.items():

    print(

        f"{name:30s}"

        f"shape={str(array.shape):18s}"

        f"dtype={array.dtype}"

    )


print("\nAdditional mapping tables:")

print(
    "df_atom_node_mapping_24:",
    df_atom_node_mapping_24.shape
)

print(
    "df_residue_node_mapping_24:",
    df_residue_node_mapping_24.shape
)


print("\nAll validation checks passed.")
```

В этом ноутбуке я скачал трёхмерную структуру белка из базы данных PDB и провёл её комплексный анализ. Сначала изучил состав белка: цепи, аминокислоты, атомы и дополнительные молекулы. Затем получил координаты атомов, проверил полноту и качество структурных данных и построил интерактивную 3D-модель с помощью NGLView.

На следующем этапе я реализовал поиск и идентификацию небелковых компонентов структуры. Выбрал один из компонентов, определил аминокислотные остатки в его ближайшем окружении и проанализировал возможные взаимодействия между ним и белком.

Все полученные результаты сохранил в таблицы и файлы JSON и CSV, а также сформировал итоговый отчёт в форматах DOCX и PDF.

Выполненная работа позволила мне разобраться в основных этапах анализа трёхмерных структур белков и подготовить основу для дальнейшего исследования взаимодействий белков с потенциальными лекарственными соединениями с использованием методов машинного обучения в области SBDD.
